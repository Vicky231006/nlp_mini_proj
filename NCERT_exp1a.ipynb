```python
# ============================================================
# NCERT RAG BENCHMARK V2 — PILOT VALIDATION
# CELL 2: INDEPENDENT SOURCE-GROUNDING VALIDATION
#
# PURPOSE:
#   Verify that every pilot question, expected answer, and
#   evidence excerpt are genuinely supported by the actual
#   NCERT page text.
#
# THIS CELL DOES NOT MODIFY:
#   - the corpus
#   - retrieval indexes
#   - the benchmark workbook
# ============================================================

import json
import re
import ast
import os
import pandas as pd
from collections import Counter

print("=" * 80)
print("NCERT RAG — V2 PILOT SOURCE-GROUNDING VALIDATION")
print("=" * 80)

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

PILOT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot.xlsx"
)

PAGE_TEXT_PATH = (
    ROOT +
    "/02_PROCESSED_DATA/corpus_v2_clean/"
    "page_text/all_pages.jsonl"
)

# ------------------------------------------------------------
# 2. LOAD BENCHMARK
# ------------------------------------------------------------

benchmark = pd.read_excel(
    PILOT_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

# ------------------------------------------------------------
# 3. LOAD PAGE TEXT
# ------------------------------------------------------------

page_records = []

with open(PAGE_TEXT_PATH, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()

        if not line:
            continue

        page_records.append(json.loads(line))

pages = pd.DataFrame(page_records)

print(f"\n✅ Loaded page records: {len(pages):,}")

required_page_columns = [
    "page_id",
    "class",
    "subject",
    "book",
    "source_file",
    "source_sha256",
    "page",
    "raw_text",
    "clean_text",
    "raw_word_count",
    "clean_word_count",
    "character_count",
    "language_hint",
    "extraction_status",
]

missing_page_columns = [
    c for c in required_page_columns
    if c not in pages.columns
]

if missing_page_columns:
    raise ValueError(
        f"❌ Missing page-text columns: {missing_page_columns}"
    )

print("✅ Page-text schema verified.")

# ------------------------------------------------------------
# 4. HELPER FUNCTIONS
# ------------------------------------------------------------

def normalize_text(text):
    """
    Conservative normalization used ONLY for comparison.
    Does not modify the stored corpus.
    """
    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = text.replace("–", "-")
    text = text.replace("—", "-")
    text = text.replace("’", "'")
    text = text.replace("“", '"')
    text = text.replace("”", '"')

    text = re.sub(r"\s+", " ", text)
    text = re.sub(r"[^\w\s'-]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def tokens(text):
    return set(
        t for t in normalize_text(text).split()
        if len(t) > 1
    )


def token_overlap(a, b):
    """
    Jaccard overlap between token sets.
    """
    A = tokens(a)
    B = tokens(b)

    if not A or not B:
        return 0.0

    return len(A & B) / len(A | B)


def containment(a, b):
    """
    Fraction of tokens from a that occur in b.
    Useful when a is an excerpt/answer and b is the page.
    """
    A = tokens(a)
    B = tokens(b)

    if not A:
        return 0.0

    return len(A & B) / len(A)


def parse_pages(value):
    """
    Safely parse values such as:
      [129]
      [129, 130]
      '[]'
    """
    if isinstance(value, list):
        return [int(x) for x in value]

    if pd.isna(value):
        return []

    value = str(value).strip()

    try:
        parsed = ast.literal_eval(value)

        if isinstance(parsed, list):
            return [int(x) for x in parsed]

    except Exception:
        pass

    # Fallback for simple comma-separated forms
    nums = re.findall(r"\d+", value)

    return [int(x) for x in nums]


# ------------------------------------------------------------
# 5. BUILD FAST PAGE LOOKUP
# ------------------------------------------------------------

page_lookup = {}

for _, row in pages.iterrows():

    key = (
        str(row["source_file"]),
        int(row["page"])
    )

    page_lookup[key] = row

print(
    f"✅ Built page lookup: {len(page_lookup):,} "
    "source-file/page records"
)

# ------------------------------------------------------------
# 6. FORBIDDEN / ARTIFICIAL QUESTION PATTERNS
# ------------------------------------------------------------

forbidden_patterns = {
    "PAGE_REFERENCE": r"\bpage(?:s)?(?:\s*\(?s?\)?)?\s*\d+",
    "PAGES_RANGE": r"\bpages?\s+\d+\s*[-–]\s*\d+",
    "NCERT_PASSAGE": r"\bncert\s+passage\b",
    "SUPPLIED_EVIDENCE": r"\bsupplied\s+(?:ncert\s+)?evidence\b",
    "ACCORDING_TO": r"\baccording\s+to\b",
    "SECTION_BEGINNING": r"\bsection\s+beginning\b",
    "PASSAGE_BEGINNING": r"\bpassage\s+beginning\b",
}

# ------------------------------------------------------------
# 7. VALIDATE EVERY PILOT QUESTION
# ------------------------------------------------------------

results = []

for _, q in benchmark.iterrows():

    qid = str(q["question_id"])
    question = str(q["question"])
    expected_answer = str(q["expected_answer"])
    evidence_excerpt = str(q["evidence_excerpt"])

    source_file = str(q["source_file"])
    gold_pages = parse_pages(q["gold_pages"])

    row_result = {
        "question_id": qid,
        "question_type": q["question_type"],
        "source_file": source_file,
        "gold_pages": gold_pages,

        "all_gold_pages_exist": False,
        "all_gold_pages_nonempty": False,

        "evidence_supported": False,
        "answer_supported": False,

        "question_is_clean": True,

        "min_evidence_page_overlap": 0.0,
        "min_evidence_page_containment": 0.0,

        "max_answer_page_containment": 0.0,

        "status": "FAIL",
        "failure_reason": "",
    }

    # --------------------------------------------------------
    # A. QUESTION LANGUAGE CHECK
    # --------------------------------------------------------

    forbidden_hits = []

    for name, pattern in forbidden_patterns.items():

        if re.search(pattern, question, flags=re.IGNORECASE):
            forbidden_hits.append(name)

    if forbidden_hits:
        row_result["question_is_clean"] = False

    # --------------------------------------------------------
    # B. GOLD PAGE EXISTENCE
    # --------------------------------------------------------

    gold_rows = []

    for page_num in gold_pages:

        key = (source_file, page_num)

        if key in page_lookup:
            gold_rows.append(page_lookup[key])

    row_result["all_gold_pages_exist"] = (
        len(gold_rows) == len(gold_pages)
    )

    # --------------------------------------------------------
    # C. GOLD PAGE NON-EMPTY CHECK
    # --------------------------------------------------------

    nonempty_pages = []

    for page_row in gold_rows:

        text = str(page_row["clean_text"])

        if len(tokens(text)) >= 5:
            nonempty_pages.append(page_row)

    row_result["all_gold_pages_nonempty"] = (
        len(nonempty_pages) == len(gold_rows)
        and len(gold_rows) == len(gold_pages)
    )

    # --------------------------------------------------------
    # D. EVIDENCE EXCERPT VS ACTUAL GOLD PAGES
    # --------------------------------------------------------

    evidence_overlaps = []
    evidence_containments = []

    for page_row in gold_rows:

        page_text = str(page_row["clean_text"])

        evidence_overlaps.append(
            token_overlap(evidence_excerpt, page_text)
        )

        evidence_containments.append(
            containment(evidence_excerpt, page_text)
        )

    if evidence_overlaps:
        row_result["min_evidence_page_overlap"] = min(
            evidence_overlaps
        )

        row_result["min_evidence_page_containment"] = min(
            evidence_containments
        )

    # Evidence is considered grounded if the excerpt has
    # substantial lexical support somewhere in the gold pages.
    #
    # We intentionally use BOTH overlap and containment so that
    # a short but legitimate excerpt is not rejected merely
    # because the full page contains lots of additional words.

    evidence_supported_any = any(
        overlap >= 0.30 or containment_value >= 0.60
        for overlap, containment_value
        in zip(
            evidence_overlaps,
            evidence_containments
        )
    )

    row_result["evidence_supported"] = evidence_supported_any

    # --------------------------------------------------------
    # E. EXPECTED ANSWER VS GOLD PAGES
    # --------------------------------------------------------

    answer_containments = []

    for page_row in gold_rows:

        page_text = str(page_row["clean_text"])

        answer_containments.append(
            containment(expected_answer, page_text)
        )

    if answer_containments:
        row_result["max_answer_page_containment"] = max(
            answer_containments
        )

    # We allow paraphrased answers, so we do NOT demand exact
    # string containment.
    #
    # Instead, require at least some meaningful lexical support.

    answer_supported = (
        row_result["max_answer_page_containment"] >= 0.40
    )

    row_result["answer_supported"] = answer_supported

    # --------------------------------------------------------
    # F. FINAL STATUS
    # --------------------------------------------------------

    failures = []

    if not row_result["all_gold_pages_exist"]:
        failures.append("MISSING_GOLD_PAGE")

    if not row_result["all_gold_pages_nonempty"]:
        failures.append("EMPTY_GOLD_PAGE")

    if not row_result["evidence_supported"]:
        failures.append("EVIDENCE_NOT_GROUNDED")

    if not row_result["answer_supported"]:
        failures.append("ANSWER_NOT_GROUNDED")

    if not row_result["question_is_clean"]:
        failures.append(
            "ARTIFICIAL_QUESTION_PATTERN"
        )

    if failures:
        row_result["status"] = "FAIL"
        row_result["failure_reason"] = "; ".join(failures)

    else:
        row_result["status"] = "PASS"
        row_result["failure_reason"] = ""

    results.append(row_result)

# ------------------------------------------------------------
# 8. RESULTS TABLE
# ------------------------------------------------------------

validation = pd.DataFrame(results)

print("\n" + "=" * 80)
print("QUESTION-LEVEL VALIDATION")
print("=" * 80)

display(
    validation[
        [
            "question_id",
            "question_type",
            "gold_pages",
            "all_gold_pages_exist",
            "all_gold_pages_nonempty",
            "evidence_supported",
            "answer_supported",
            "question_is_clean",
            "min_evidence_page_overlap",
            "min_evidence_page_containment",
            "max_answer_page_containment",
            "status",
            "failure_reason",
        ]
    ]
)

# ------------------------------------------------------------
# 9. SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("VALIDATION SUMMARY")
print("=" * 80)

total = len(validation)

passed = int(
    (validation["status"] == "PASS").sum()
)

failed = total - passed

print(f"Total pilot questions : {total}")
print(f"PASS                  : {passed}")
print(f"FAIL                  : {failed}")

print("\nFailure categories:")

failure_counter = Counter()

for value in validation["failure_reason"]:

    if not value:
        continue

    for failure in value.split("; "):
        failure_counter[failure] += 1

if failure_counter:
    for failure, count in failure_counter.most_common():
        print(f"  • {failure}: {count}")
else:
    print("  None")

# ------------------------------------------------------------
# 10. QUESTION-TEXT QUALITY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("QUESTION TEXT QUALITY")
print("=" * 80)

clean_count = int(
    validation["question_is_clean"].sum()
)

print(
    f"Clean questions: {clean_count}/{total}"
)

# ------------------------------------------------------------
# 11. IMPORTANT STOP/GO DECISION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("PILOT DECISION")
print("=" * 80)

if failed == 0:

    print(
        "🟢 PASS — All pilot questions passed independent "
        "source-grounding validation."
    )

    print(
        "\nNext step can be retrieval evaluation on this "
        "10-question pilot."
    )

else:

    print(
        "🔴 STOP — At least one pilot question failed "
        "independent validation."
    )

    print(
        "\nDo NOT run retrieval evaluation yet."
    )

    print(
        "The failed questions must be corrected/replaced "
        "before scaling V2."
    )
```

```python
# ============================================================
# NCERT RAG BENCHMARK V2 — PILOT VALIDATION
# CELL 3: FORENSIC REVIEW OF FAILED QUESTIONS
#
# PURPOSE:
#   Show the exact question, expected answer, evidence excerpt,
#   and actual NCERT source text for every failed pilot row.
#
#   NO FILES ARE MODIFIED.
# ============================================================

import json
import ast
import pandas as pd

print("=" * 80)
print("NCERT RAG — FAILED PILOT QUESTIONS: FORENSIC REVIEW")
print("=" * 80)

# ------------------------------------------------------------
# 1. LOAD BENCHMARK
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

PILOT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot.xlsx"
)

PAGE_TEXT_PATH = (
    ROOT +
    "/02_PROCESSED_DATA/corpus_v2_clean/"
    "page_text/all_pages.jsonl"
)

benchmark = pd.read_excel(
    PILOT_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

# ------------------------------------------------------------
# 2. LOAD PAGE TEXT
# ------------------------------------------------------------

page_records = []

with open(PAGE_TEXT_PATH, "r", encoding="utf-8") as f:

    for line in f:

        line = line.strip()

        if line:
            page_records.append(
                json.loads(line)
            )

pages = pd.DataFrame(page_records)

# ------------------------------------------------------------
# 3. PAGE LOOKUP
# ------------------------------------------------------------

page_lookup = {}

for _, row in pages.iterrows():

    key = (
        str(row["source_file"]),
        int(row["page"])
    )

    page_lookup[key] = row

# ------------------------------------------------------------
# 4. PARSE PAGE LIST
# ------------------------------------------------------------

def parse_pages(value):

    if isinstance(value, list):
        return [int(x) for x in value]

    if pd.isna(value):
        return []

    try:

        parsed = ast.literal_eval(
            str(value)
        )

        if isinstance(parsed, list):
            return [int(x) for x in parsed]

    except Exception:
        pass

    return []

# ------------------------------------------------------------
# 5. IDENTIFY FAILED QUESTIONS
# ------------------------------------------------------------

failed_ids = [
    "V2P007",
    "V2P008",
    "V2P009",
]

failed = benchmark[
    benchmark["question_id"].isin(failed_ids)
].copy()

# ------------------------------------------------------------
# 6. PRINT FULL FORENSIC CONTEXT
# ------------------------------------------------------------

for _, row in failed.iterrows():

    qid = row["question_id"]
    source_file = str(row["source_file"])
    gold_pages = parse_pages(row["gold_pages"])

    print("\n")
    print("=" * 80)
    print(f"{qid} — {row['question_type']}")
    print("=" * 80)

    print("\nQUESTION:")
    print(row["question"])

    print("\nEXPECTED ANSWER:")
    print(row["expected_answer"])

    print("\nEVIDENCE EXCERPT:")
    print(row["evidence_excerpt"])

    print("\nSOURCE:")
    print(
        f"{row['class']} | "
        f"{row['subject']} | "
        f"{row['book']} | "
        f"{source_file}"
    )

    print("\nGOLD PAGES:")
    print(gold_pages)

    for page_num in gold_pages:

        key = (
            source_file,
            page_num
        )

        print("\n" + "-" * 80)
        print(f"ACTUAL NCERT PAGE {page_num}")
        print("-" * 80)

        if key not in page_lookup:

            print("❌ PAGE NOT FOUND")

            continue

        page_text = str(
            page_lookup[key]["clean_text"]
        )

        print(page_text)

print("\n")
print("=" * 80)
print("FORENSIC REVIEW COMPLETE")
print("=" * 80)

print(
    "\nNo files were modified."
)
```

```python
# ============================================================
# NCERT RAG BENCHMARK V2
# CELL 4: PRINTED-PAGE ↔ PDF-PAGE PROVENANCE CHECK
#
# PURPOSE:
#   Determine whether benchmark gold_pages refer to printed
#   textbook page numbers while all_pages.jsonl["page"] refers
#   to PDF/extraction page numbers.
#
#   THIS CELL DOES NOT MODIFY ANY FILE.
# ============================================================

import json
import re
import pandas as pd

print("=" * 80)
print("NCERT RAG — PAGE PROVENANCE DIAGNOSTIC")
print("=" * 80)

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

PAGE_TEXT_PATH = (
    ROOT +
    "/02_PROCESSED_DATA/corpus_v2_clean/"
    "page_text/all_pages.jsonl"
)

# ------------------------------------------------------------
# 1. LOAD PAGE TEXT
# ------------------------------------------------------------

records = []

with open(PAGE_TEXT_PATH, "r", encoding="utf-8") as f:

    for line in f:

        line = line.strip()

        if line:
            records.append(json.loads(line))

pages = pd.DataFrame(records)

# ------------------------------------------------------------
# 2. RESTRICT TO CLASS 1 ENGLISH MRIDANG
# ------------------------------------------------------------

mridang = pages[
    (pages["class"] == "Class_01") &
    (pages["subject"] == "English") &
    (pages["book"] == "Mridang")
].copy()

print(f"\nMridang page records: {len(mridang)}")

# ------------------------------------------------------------
# 3. SEARCH FOR UNIQUE CONTENT ANCHORS
# ------------------------------------------------------------

anchors = {
    "TAP_QUESTION":
        "Do you remember to close the tap",

    "FOOD_SHARING":
        "Where do you sit when you have lunch",

    "ANANDI_RAINBOW":
        "Anandi was fast asleep",

    "RAINBOW_COLOURS":
        "After giving colours to the flowers",
}

print("\n" + "=" * 80)
print("CONTENT ANCHOR LOCATIONS")
print("=" * 80)

anchor_results = []

for name, anchor in anchors.items():

    print(f"\n🔎 {name}")
    print(f"Anchor: {anchor}")

    matches = []

    for _, row in mridang.iterrows():

        text = str(row["clean_text"])

        if anchor.lower() in text.lower():

            matches.append(row)

    if not matches:

        print("❌ No match found")

        anchor_results.append({
            "anchor": name,
            "matches": 0,
            "pdf_pages": [],
        })

        continue

    pdf_pages = [
        int(row["page"])
        for row in matches
    ]

    print(
        "✅ Found on PDF/extraction page(s):",
        pdf_pages
    )

    for row in matches:

        print(
            f"   PDF page {int(row['page'])} | "
            f"page_id={row['page_id']}"
        )

    anchor_results.append({
        "anchor": name,
        "matches": len(matches),
        "pdf_pages": pdf_pages,
    })

# ------------------------------------------------------------
# 4. INSPECT FIRST FEW PAGE RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("MRIDANG PAGE NUMBERING SAMPLE")
print("=" * 80)

sample = mridang[
    [
        "page",
        "page_id",
        "raw_word_count",
        "clean_word_count",
        "raw_text",
    ]
].head(15)

display(sample)

# ------------------------------------------------------------
# 5. DIAGNOSTIC CONCLUSION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("DIAGNOSTIC CONCLUSION")
print("=" * 80)

print(
    """
This cell is diagnostic only.

We are checking whether:

    benchmark gold_pages
            ↓
    printed NCERT page numbers

are different from:

    all_pages.jsonl["page"]
            ↓
    PDF/extraction page numbers

If the content anchors occur on different numeric pages,
we will create an explicit printed-page provenance mapping
for the benchmark rather than silently treating the numbers
as interchangeable.
"""
)

print("✅ Diagnostic complete.")
print("❗ No files were modified.")
```

```python
# ============================================================
# NCERT RAG BENCHMARK V2
# CELL 5: BUILD PRINTED-PAGE → PDF-PAGE MAPPING
#
# PURPOSE:
#   Inspect the actual printed page numbers embedded in the
#   Mridang PDF text and determine their corresponding
#   all_pages.jsonl page numbers.
#
#   NO FILES ARE MODIFIED.
# ============================================================

import re
import pandas as pd

print("=" * 80)
print("NCERT RAG — PRINTED PAGE MAPPING FOR MRIDANG")
print("=" * 80)

# ------------------------------------------------------------
# 1. USE THE ALREADY LOADED `mridang` DATAFRAME
# ------------------------------------------------------------

if "mridang" not in globals():
    raise RuntimeError(
        "❌ `mridang` dataframe not found. "
        "Please run Cell 4 first."
    )

print(
    f"Mridang PDF/extraction pages available: "
    f"{len(mridang)}"
)

# ------------------------------------------------------------
# 2. FUNCTION TO DETECT PRINTED PAGE NUMBERS
# ------------------------------------------------------------

def detect_printed_numbers(text):

    text = str(text)

    candidates = []

    # Common textbook extraction patterns.
    patterns = [
        r"\bReprint\s+2026-27\s+(\d{1,3})\b",
        r"\bMridang\s+(\d{1,3})\b",
        r"\b(\d{1,3})\s+Mridang\b",
    ]

    for pattern in patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        for value in matches:

            try:
                number = int(value)

                if 1 <= number <= 200:
                    candidates.append(number)

            except Exception:
                pass

    return sorted(set(candidates))


# ------------------------------------------------------------
# 3. BUILD CANDIDATE MAPPING
# ------------------------------------------------------------

mapping_rows = []

for _, row in mridang.iterrows():

    pdf_page = int(row["page"])
    text = str(row["raw_text"])

    printed_numbers = detect_printed_numbers(text)

    for printed_page in printed_numbers:

        mapping_rows.append({
            "pdf_page": pdf_page,
            "printed_page": printed_page,
            "offset_printed_minus_pdf":
                printed_page - pdf_page,
        })

mapping = pd.DataFrame(mapping_rows)

# ------------------------------------------------------------
# 4. SHOW MAPPING
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("DETECTED PRINTED-PAGE MAPPINGS")
print("=" * 80)

if mapping.empty:

    print(
        "❌ No printed-page markers were detected."
    )

else:

    display(
        mapping[
            [
                "printed_page",
                "pdf_page",
                "offset_printed_minus_pdf",
            ]
        ]
        .sort_values(
            ["printed_page", "pdf_page"]
        )
        .drop_duplicates()
        .head(100)
    )

# ------------------------------------------------------------
# 5. CHECK THE THREE IMPORTANT PAGES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TARGET PAGE CHECK")
print("=" * 80)

target_pages = [43, 99, 129, 130]

target_mapping = mapping[
    mapping["printed_page"].isin(target_pages)
].copy()

if target_mapping.empty:

    print(
        "❌ None of the target printed pages were detected."
    )

else:

    display(
        target_mapping[
            [
                "printed_page",
                "pdf_page",
                "offset_printed_minus_pdf",
            ]
        ]
        .sort_values(
            ["printed_page", "pdf_page"]
        )
        .drop_duplicates()
    )

# ------------------------------------------------------------
# 6. OFFSET DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("OFFSET DISTRIBUTION")
print("=" * 80)

if not mapping.empty:

    offset_counts = (
        mapping[
            [
                "offset_printed_minus_pdf"
            ]
        ]
        .value_counts()
        .sort_index()
    )

    print(offset_counts.to_string())

# ------------------------------------------------------------
# 7. INSPECT TARGET PDF PAGES DIRECTLY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TARGET PDF PAGE CONTENT")
print("=" * 80)

for pdf_page in [57, 113, 129, 130]:

    rows = mridang[
        mridang["page"] == pdf_page
    ]

    if rows.empty:

        print(
            f"\n❌ PDF page {pdf_page} not found."
        )

        continue

    row = rows.iloc[0]

    print("\n" + "-" * 80)
    print(f"PDF PAGE {pdf_page}")
    print("-" * 80)

    print(
        str(row["clean_text"])
    )

print("\n" + "=" * 80)
print("PAGE MAPPING DIAGNOSTIC COMPLETE")
print("=" * 80)

print(
    "\n❗ No files were modified."
)
```

```python
# ============================================================
# NCERT RAG BENCHMARK V2
# CELL 6: RELIABLE PRINTED-PAGE PROVENANCE MAPPER
#
# PURPOSE:
#   Build a mapping:
#
#       printed NCERT page -> PDF/extraction page
#
#   using the actual textbook page markers.
#
#   NO FILES ARE MODIFIED.
# ============================================================

import re
import pandas as pd

print("=" * 80)
print("NCERT RAG — RELIABLE PRINTED-PAGE PROVENANCE MAPPER")
print("=" * 80)

# ------------------------------------------------------------
# 1. VERIFY MRIDANG DATAFRAME
# ------------------------------------------------------------

if "mridang" not in globals():
    raise RuntimeError(
        "❌ `mridang` dataframe not found. "
        "Run Cells 4 and 5 first."
    )

print(
    f"Source records available: {len(mridang)}"
)

# ------------------------------------------------------------
# 2. DETECT A PRINTED PAGE NUMBER
# ------------------------------------------------------------

def detect_printed_page(text):

    text = str(text)

    # The actual textbook pages commonly expose the
    # page number together with the book title.
    #
    # Examples observed:
    #   "43 Mridang"
    #   "99 Mridang"
    #   "116 Mridang"
    #
    # We intentionally avoid using arbitrary standalone
    # numbers because those can be chapter numbers, years,
    # exercise numbers, etc.

    patterns = [
        r"(?<!\d)(\d{1,3})\s+Mridang\b",
        r"\bMridang\s+(\d{1,3})(?!\d)",
    ]

    candidates = []

    for pattern in patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        for value in matches:

            number = int(value)

            if 1 <= number <= 200:
                candidates.append(number)

    candidates = sorted(set(candidates))

    # If exactly one plausible textbook page number is
    # detected, accept it.
    if len(candidates) == 1:
        return candidates[0]

    # If multiple candidates occur, do not guess.
    return None


# ------------------------------------------------------------
# 3. BUILD MAPPING
# ------------------------------------------------------------

provenance_rows = []

ambiguous_rows = []

for _, row in mridang.iterrows():

    pdf_page = int(row["page"])

    printed_page = detect_printed_page(
        row["raw_text"]
    )

    if printed_page is None:
        continue

    provenance_rows.append({
        "book": row["book"],
        "source_file": row["source_file"],
        "printed_page": printed_page,
        "pdf_page": pdf_page,
        "page_id": row["page_id"],
    })

provenance = pd.DataFrame(
    provenance_rows
)

# ------------------------------------------------------------
# 4. REMOVE DUPLICATE IDENTICAL MAPPINGS
# ------------------------------------------------------------

if not provenance.empty:

    provenance = (
        provenance
        .drop_duplicates(
            subset=[
                "source_file",
                "printed_page",
                "pdf_page",
            ]
        )
        .sort_values(
            [
                "source_file",
                "printed_page",
                "pdf_page",
            ]
        )
        .reset_index(drop=True)
    )

# ------------------------------------------------------------
# 5. CHECK FOR CONFLICTS
# ------------------------------------------------------------

conflicts = (
    provenance
    .groupby(
        ["source_file", "printed_page"]
    )["pdf_page"]
    .nunique()
)

conflicts = conflicts[
    conflicts > 1
]

print(
    f"\n✅ Provenance mappings detected: "
    f"{len(provenance):,}"
)

print(
    f"⚠️ Printed pages with conflicting PDF mappings: "
    f"{len(conflicts)}"
)

# ------------------------------------------------------------
# 6. SHOW TARGET MAPPINGS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TARGET MAPPINGS")
print("=" * 80)

targets = provenance[
    provenance["printed_page"].isin(
        [43, 99, 115, 116]
    )
]

display(targets)

# ------------------------------------------------------------
# 7. CHECK EXPECTED TARGETS
# ------------------------------------------------------------

expected = {
    43: 57,
    99: 113,
    115: 129,
    116: 130,
}

print("\n" + "=" * 80)
print("EXPECTED TARGET CHECK")
print("=" * 80)

all_correct = True

for printed_page, expected_pdf in expected.items():

    matches = provenance[
        provenance["printed_page"] == printed_page
    ]["pdf_page"].tolist()

    if expected_pdf in matches:

        print(
            f"✅ Printed {printed_page} "
            f"→ PDF {expected_pdf}"
        )

    else:

        print(
            f"❌ Printed {printed_page} "
            f"→ expected PDF {expected_pdf}, "
            f"found {matches}"
        )

        all_correct = False

# ------------------------------------------------------------
# 8. CHECK MONOTONICITY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("MONOTONICITY CHECK")
print("=" * 80)

if not provenance.empty:

    ordered = provenance.sort_values(
        ["source_file", "printed_page"]
    )

    diffs = (
        ordered
        .groupby("source_file")["pdf_page"]
        .diff()
        .dropna()
    )

    negative_steps = int(
        (diffs < 0).sum()
    )

    print(
        f"Negative PDF-page steps: "
        f"{negative_steps}"
    )

    if negative_steps == 0:
        print(
            "✅ PDF page ordering is monotonic "
            "for detected textbook pages."
        )
    else:
        print(
            "⚠️ Non-monotonic mappings detected. "
            "Do not assume a simple offset."
        )

# ------------------------------------------------------------
# 9. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("PROVENANCE MAPPER STATUS")
print("=" * 80)

if (
    all_correct
    and len(conflicts) == 0
):

    print(
        "🟢 PASS — Target printed-page mappings "
        "were recovered without ambiguity."
    )

else:

    print(
        "🟡 REVIEW — Provenance mapping needs "
        "additional inspection before benchmark editing."
    )

print("\n❗ No files were modified.")
```

```python
# ============================================================
# NCERT RAG BENCHMARK V2
# CELL 7: ADD AUTHORITATIVE PDF-PAGE PROVENANCE
#
# PURPOSE:
#   Preserve the original gold_pages column while adding
#   gold_pdf_pages, which is the page numbering used by the
#   retrieval corpus.
#
#   The 10 pilot mappings have been independently verified
#   from the actual NCERT source text.
#
#   ORIGINAL WORKBOOK IS NOT OVERWRITTEN.
# ============================================================

import os
import ast
import pandas as pd

print("=" * 80)
print("NCERT RAG — ADDING PDF-PAGE PROVENANCE")
print("=" * 80)

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

INPUT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot.xlsx"
)

OUTPUT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v2.xlsx"
)

# ------------------------------------------------------------
# 1. LOAD ORIGINAL WORKBOOK
# ------------------------------------------------------------

benchmark = pd.read_excel(
    INPUT_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

# ------------------------------------------------------------
# 2. AUTHORITATIVE PDF-PAGE MAPPINGS
# ------------------------------------------------------------
#
# These are PDF/extraction pages, NOT printed textbook pages.
#
# V2P001–V2P007 and V2P010 already used PDF page numbers.
#
# V2P008:
#   printed page 43 -> PDF page 57
#
# V2P009:
#   printed page 99 -> PDF page 113
#
# ------------------------------------------------------------

gold_pdf_pages = {

    "V2P001": [129],
    "V2P002": [130],
    "V2P003": [130],
    "V2P004": [131],
    "V2P005": [131],

    "V2P006": [129, 130],
    "V2P007": [129, 130],

    "V2P008": [57],
    "V2P009": [113],

    "V2P010": [131],
}

# ------------------------------------------------------------
# 3. VERIFY EVERY QUESTION HAS A MAPPING
# ------------------------------------------------------------

question_ids = set(
    benchmark["question_id"].astype(str)
)

mapping_ids = set(
    gold_pdf_pages.keys()
)

missing_mappings = question_ids - mapping_ids
extra_mappings = mapping_ids - question_ids

if missing_mappings:

    raise ValueError(
        "❌ Missing PDF-page mappings for: "
        f"{sorted(missing_mappings)}"
    )

if extra_mappings:

    raise ValueError(
        "❌ Mapping contains unknown question IDs: "
        f"{sorted(extra_mappings)}"
    )

print(
    "✅ Every pilot question has an authoritative "
    "PDF-page mapping."
)

# ------------------------------------------------------------
# 4. ADD NEW COLUMN
# ------------------------------------------------------------

benchmark["gold_pdf_pages"] = benchmark[
    "question_id"
].astype(str).map(gold_pdf_pages)

# Convert to explicit string representation for Excel
benchmark["gold_pdf_pages"] = benchmark[
    "gold_pdf_pages"
].apply(str)

# ------------------------------------------------------------
# 5. VALIDATE THE NEW COLUMN
# ------------------------------------------------------------

for _, row in benchmark.iterrows():

    value = row["gold_pdf_pages"]

    try:

        parsed = ast.literal_eval(value)

        if not isinstance(parsed, list):
            raise ValueError

        if not all(
            isinstance(x, int)
            for x in parsed
        ):
            raise ValueError

    except Exception:

        raise ValueError(
            f"❌ Invalid gold_pdf_pages for "
            f"{row['question_id']}: {value}"
        )

print(
    "✅ PDF-page provenance column validated."
)

# ------------------------------------------------------------
# 6. DISPLAY PROVENANCE
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("PILOT PAGE PROVENANCE")
print("=" * 80)

display(
    benchmark[
        [
            "question_id",
            "gold_pages",
            "gold_pdf_pages",
            "source_file",
        ]
    ]
)

# ------------------------------------------------------------
# 7. SAVE NEW VERSION
# ------------------------------------------------------------

if os.path.exists(OUTPUT_PATH):

    raise FileExistsError(
        f"❌ Output already exists:\n{OUTPUT_PATH}\n"
        "Refusing to overwrite it."
    )

with pd.ExcelWriter(
    OUTPUT_PATH,
    engine="openpyxl"
) as writer:

    benchmark.to_excel(
        writer,
        sheet_name="Benchmark_V2_Pilot",
        index=False
    )

    # Preserve the design rules and checklist sheets.
    for sheet_name in [
        "V2_Design_Rules",
        "Validation_Checklist",
    ]:

        sheet_df = pd.read_excel(
            INPUT_PATH,
            sheet_name=sheet_name
        )

        sheet_df.to_excel(
            writer,
            sheet_name=sheet_name,
            index=False
        )

print("\n" + "=" * 80)
print("SUCCESS")
print("=" * 80)

print(
    f"✅ New workbook created:\n{OUTPUT_PATH}"
)

print(
    "\nOriginal workbook was NOT modified."
)
```

```python
# ============================================================
# NCERT RAG BENCHMARK V2
# CELL 8: PROVENANCE-AWARE + MULTI-PAGE-AWARE VALIDATION
#
# PURPOSE:
#   1. Use gold_pdf_pages as authoritative retrieval provenance.
#   2. Validate evidence against the correct PDF pages.
#   3. Validate MULTI_PAGE evidence page-by-page.
#   4. Validate expected answers against the gold evidence.
#   5. Keep HARD_DISTRACTOR separate until distractor analysis.
#
# IMPORTANT:
#   - Does NOT modify the benchmark.
#   - Does NOT modify the corpus.
#   - Does NOT modify retrieval indexes.
# ============================================================

import json
import re
import ast
import pandas as pd

print("=" * 80)
print("NCERT RAG — V2 PILOT CORRECTED VALIDATION")
print("=" * 80)

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

PILOT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v2.xlsx"
)

PAGE_TEXT_PATH = (
    ROOT +
    "/02_PROCESSED_DATA/corpus_v2_clean/"
    "page_text/all_pages.jsonl"
)

# ------------------------------------------------------------
# 2. LOAD BENCHMARK
# ------------------------------------------------------------

benchmark = pd.read_excel(
    PILOT_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

required_columns = [
    "question_id",
    "question",
    "question_type",
    "expected_answer",
    "evidence_excerpt",
    "source_file",
    "gold_pages",
    "gold_pdf_pages",
]

missing = [
    c for c in required_columns
    if c not in benchmark.columns
]

if missing:
    raise ValueError(
        f"❌ Missing required columns: {missing}"
    )

print(
    f"✅ Loaded pilot benchmark: "
    f"{len(benchmark)} questions"
)

# ------------------------------------------------------------
# 3. LOAD PAGE TEXT
# ------------------------------------------------------------

page_records = []

with open(
    PAGE_TEXT_PATH,
    "r",
    encoding="utf-8"
) as f:

    for line in f:

        line = line.strip()

        if line:
            page_records.append(
                json.loads(line)
            )

pages = pd.DataFrame(page_records)

required_page_columns = [
    "page_id",
    "class",
    "subject",
    "book",
    "source_file",
    "page",
    "raw_text",
    "clean_text",
]

missing_page_columns = [
    c for c in required_page_columns
    if c not in pages.columns
]

if missing_page_columns:
    raise ValueError(
        f"❌ Missing page-text columns: "
        f"{missing_page_columns}"
    )

print(
    f"✅ Loaded page records: "
    f"{len(pages):,}"
)

# ------------------------------------------------------------
# 4. PAGE LOOKUP
# ------------------------------------------------------------

page_lookup = {}

for _, row in pages.iterrows():

    key = (
        str(row["source_file"]),
        int(row["page"])
    )

    page_lookup[key] = row

print(
    f"✅ Page lookup records: "
    f"{len(page_lookup):,}"
)

# ------------------------------------------------------------
# 5. HELPERS
# ------------------------------------------------------------

def parse_list(value):

    if isinstance(value, list):
        return [int(x) for x in value]

    if pd.isna(value):
        return []

    try:

        parsed = ast.literal_eval(
            str(value)
        )

        if isinstance(parsed, list):
            return [int(x) for x in parsed]

    except Exception:
        pass

    return []


def normalize_text(text):

    if pd.isna(text):
        return ""

    text = str(text).lower()

    replacements = {
        "–": "-",
        "—": "-",
        "’": "'",
        "“": '"',
        "”": '"',
        "…": "...",
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    text = re.sub(
        r"[^\w\s'-]",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def token_set(text):

    return {
        token
        for token in normalize_text(text).split()
        if len(token) > 1
    }


def containment(source, target):

    source_tokens = token_set(source)
    target_tokens = token_set(target)

    if not source_tokens:
        return 0.0

    return (
        len(source_tokens & target_tokens)
        / len(source_tokens)
    )


def jaccard(a, b):

    A = token_set(a)
    B = token_set(b)

    if not A or not B:
        return 0.0

    return (
        len(A & B)
        / len(A | B)
    )


# ------------------------------------------------------------
# 6. FORBIDDEN QUESTION PATTERNS
# ------------------------------------------------------------

forbidden_patterns = {

    "PAGE_REFERENCE":
        r"\bpages?\s*\(?s?\)?\s*\d+",

    "PAGE_RANGE":
        r"\bpages?\s+\d+\s*[-–]\s*\d+",

    "NCERT_PASSAGE":
        r"\bncert\s+passage\b",

    "SUPPLIED_EVIDENCE":
        r"\bsupplied\s+(?:ncert\s+)?evidence\b",

    "ACCORDING_TO":
        r"\baccording\s+to\b",

    "SECTION_BEGINNING":
        r"\bsection\s+beginning\b",

    "PASSAGE_BEGINNING":
        r"\bpassage\s+beginning\b",
}


def question_quality(question):

    hits = []

    for name, pattern in forbidden_patterns.items():

        if re.search(
            pattern,
            question,
            flags=re.IGNORECASE
        ):
            hits.append(name)

    return hits


# ------------------------------------------------------------
# 7. VALIDATE QUESTIONS
# ------------------------------------------------------------

results = []

for _, row in benchmark.iterrows():

    qid = str(row["question_id"])
    qtype = str(row["question_type"])

    question = str(row["question"])
    answer = str(row["expected_answer"])
    evidence = str(row["evidence_excerpt"])

    source_file = str(row["source_file"])

    gold_pdf_pages = parse_list(
        row["gold_pdf_pages"]
    )

    result = {

        "question_id": qid,
        "question_type": qtype,

        "gold_pdf_pages": gold_pdf_pages,

        "all_pdf_pages_exist": False,
        "all_pdf_pages_nonempty": False,

        "question_is_clean": True,

        "evidence_grounded": False,
        "answer_grounded": False,

        "multi_page_evidence_valid": None,

        "hard_distractor_status":
            "NOT_APPLICABLE",

        "status": "FAIL",
        "failure_reason": "",

        "min_page_evidence_containment": 0.0,
        "max_answer_containment": 0.0,
    }

    # --------------------------------------------------------
    # A. QUESTION QUALITY
    # --------------------------------------------------------

    forbidden_hits = question_quality(
        question
    )

    if forbidden_hits:
        result["question_is_clean"] = False

    # --------------------------------------------------------
    # B. GOLD PDF PAGE EXISTENCE
    # --------------------------------------------------------

    gold_rows = []

    for pdf_page in gold_pdf_pages:

        key = (
            source_file,
            pdf_page
        )

        if key in page_lookup:

            gold_rows.append(
                page_lookup[key]
            )

    result["all_pdf_pages_exist"] = (
        len(gold_rows)
        == len(gold_pdf_pages)
    )

    # --------------------------------------------------------
    # C. GOLD PAGE NON-EMPTY
    # --------------------------------------------------------

    result["all_pdf_pages_nonempty"] = (

        len(gold_rows)
        == len(gold_pdf_pages)

        and

        all(
            len(
                token_set(
                    row["clean_text"]
                )
            ) >= 5
            for row in gold_rows
        )
    )

    # --------------------------------------------------------
    # D. EVIDENCE VALIDATION
    # --------------------------------------------------------

    page_containments = []

    for page_row in gold_rows:

        page_text = str(
            page_row["clean_text"]
        )

        page_containments.append(
            containment(
                evidence,
                page_text
            )
        )

    # --------------------------------------------------------
    # MULTI-PAGE QUESTIONS
    # --------------------------------------------------------

    if qtype == "MULTI_PAGE":

        result[
            "multi_page_evidence_valid"
        ] = True

        # Every gold page must contribute meaningful
        # evidence independently.
        #
        # A combined evidence string is NOT required
        # to be contained in each page.

        for value in page_containments:

            if value < 0.20:

                result[
                    "multi_page_evidence_valid"
                ] = False

        if page_containments:

            result[
                "min_page_evidence_containment"
            ] = min(page_containments)

        result[
            "evidence_grounded"
        ] = result[
            "multi_page_evidence_valid"
        ]

    # --------------------------------------------------------
    # SINGLE-PAGE QUESTIONS
    # --------------------------------------------------------

    else:

        if page_containments:

            result[
                "min_page_evidence_containment"
            ] = max(page_containments)

        result[
            "evidence_grounded"
        ] = any(
            value >= 0.40
            for value in page_containments
        )

    # --------------------------------------------------------
    # E. ANSWER VALIDATION
    # --------------------------------------------------------

    answer_containments = []

    for page_row in gold_rows:

        page_text = str(
            page_row["clean_text"]
        )

        answer_containments.append(
            containment(
                answer,
                page_text
            )
        )

    if answer_containments:

        result[
            "max_answer_containment"
        ] = max(
            answer_containments
        )

    result[
        "answer_grounded"
    ] = (
        result[
            "max_answer_containment"
        ] >= 0.40
    )

    # --------------------------------------------------------
    # F. HARD DISTRACTOR
    # --------------------------------------------------------

    if qtype == "HARD_DISTRACTOR":

        # IMPORTANT:
        # We do not certify this here.
        #
        # A true hard distractor requires corpus-wide
        # analysis of semantically similar non-gold sources.

        result[
            "hard_distractor_status"
        ] = "REVIEW_REQUIRED"

    # --------------------------------------------------------
    # G. FINAL STATUS
    # --------------------------------------------------------

    failures = []

    if not result[
        "all_pdf_pages_exist"
    ]:
        failures.append(
            "MISSING_GOLD_PDF_PAGE"
        )

    if not result[
        "all_pdf_pages_nonempty"
    ]:
        failures.append(
            "EMPTY_GOLD_PDF_PAGE"
        )

    if not result[
        "question_is_clean"
    ]:
        failures.append(
            "ARTIFICIAL_QUESTION_PATTERN"
        )

    if not result[
        "evidence_grounded"
    ]:
        failures.append(
            "EVIDENCE_NOT_GROUNDED"
        )

    if not result[
        "answer_grounded"
    ]:
        failures.append(
            "ANSWER_NOT_GROUNDED"
        )

    # Hard distractors intentionally cannot receive a
    # final PASS until their distractor property is verified.

    if qtype == "HARD_DISTRACTOR":

        failures.append(
            "HARD_DISTRACTOR_REVIEW_REQUIRED"
        )

    if failures:

        result["status"] = "REVIEW"

        result[
            "failure_reason"
        ] = "; ".join(failures)

    else:

        result["status"] = "PASS"

    results.append(result)

# ------------------------------------------------------------
# 8. DISPLAY RESULTS
# ------------------------------------------------------------

validation = pd.DataFrame(
    results
)

print("\n" + "=" * 80)
print("QUESTION-LEVEL RESULTS")
print("=" * 80)

display(
    validation[
        [
            "question_id",
            "question_type",
            "gold_pdf_pages",
            "all_pdf_pages_exist",
            "all_pdf_pages_nonempty",
            "evidence_grounded",
            "answer_grounded",
            "multi_page_evidence_valid",
            "hard_distractor_status",
            "question_is_clean",
            "min_page_evidence_containment",
            "max_answer_containment",
            "status",
            "failure_reason",
        ]
    ]
)

# ------------------------------------------------------------
# 9. SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("VALIDATION SUMMARY")
print("=" * 80)

total = len(validation)

passed = int(
    (
        validation["status"]
        == "PASS"
    ).sum()
)

review = total - passed

print(
    f"Total questions : {total}"
)

print(
    f"PASS            : {passed}"
)

print(
    f"REVIEW          : {review}"
)

# ------------------------------------------------------------
# 10. CATEGORY COUNTS
# ------------------------------------------------------------

print("\nFailure / review categories:")

category_counts = {}

for value in validation[
    "failure_reason"
]:

    if not value:
        continue

    for category in value.split("; "):

        category_counts[category] = (
            category_counts.get(category, 0)
            + 1
        )

if category_counts:

    for category, count in sorted(
        category_counts.items(),
        key=lambda x: (-x[1], x[0])
    ):

        print(
            f"  • {category}: {count}"
        )

else:

    print("  None")

# ------------------------------------------------------------
# 11. MULTI-PAGE SUMMARY
# ------------------------------------------------------------

multi = validation[
    validation["question_type"]
    == "MULTI_PAGE"
]

print("\n" + "=" * 80)
print("MULTI-PAGE VALIDATION")
print("=" * 80)

if len(multi) == 0:

    print(
        "No multi-page questions."
    )

else:

    print(
        f"Multi-page questions: "
        f"{len(multi)}"
    )

    print(
        "Passed page-level evidence:",
        int(
            multi[
                "multi_page_evidence_valid"
            ].sum()
        ),
        "/",
        len(multi)
    )

# ------------------------------------------------------------
# 12. HARD-DISTRACTOR SUMMARY
# ------------------------------------------------------------

hard = validation[
    validation["question_type"]
    == "HARD_DISTRACTOR"
]

print("\n" + "=" * 80)
print("HARD-DISTRACTOR STATUS")
print("=" * 80)

if len(hard) == 0:

    print(
        "No hard-distractor questions."
    )

else:

    print(
        f"Hard-distractor questions: "
        f"{len(hard)}"
    )

    print(
        "⚠️ These are NOT certified yet."
    )

    print(
        "They require corpus-wide semantic distractor "
        "verification before the benchmark is frozen."
    )

# ------------------------------------------------------------
# 13. FINAL DECISION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("PILOT DECISION")
print("=" * 80)

hard_exists = len(hard) > 0

non_hard = validation[
    validation["question_type"]
    != "HARD_DISTRACTOR"
]

non_hard_pass = (
    non_hard["status"] == "PASS"
).all()

if (
    non_hard_pass
    and not hard_exists
):

    print(
        "🟢 PILOT CLEAN"
    )

elif non_hard_pass and hard_exists:

    print(
        "🟡 GROUNDING CLEAN — "
        "HARD DISTRACTOR STILL REQUIRES VALIDATION."
    )

else:

    print(
        "🔴 PILOT NOT CLEAN — "
        "GROUNDING ISSUES REMAIN."
    )

print(
    "\n❗ No files were modified."
)
```

```python
# ============================================================
# NCERT RAG — HARD-DISTRACTOR VERIFICATION
# CORRECTED PATHS
#
# Purpose:
#   Determine whether V2P008 has a genuinely similar
#   non-gold chunk elsewhere in the English corpus.
#
# IMPORTANT:
#   This is a diagnostic only.
#   It does NOT automatically certify a hard distractor.
# ============================================================

import os
import re
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

BENCHMARK_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v2.xlsx"
)

CHUNKS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/english_only_chunks.csv"
)

EMBEDDINGS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/indexes/embeddings/"
    "chunk_embeddings.npy"
)

# ------------------------------------------------------------
# 2. LOAD BENCHMARK
# ------------------------------------------------------------

benchmark = pd.read_excel(
    BENCHMARK_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

row = benchmark[
    benchmark["question_id"].astype(str) == "V2P008"
]

if len(row) != 1:
    raise ValueError(
        f"❌ Expected exactly one V2P008 row, found {len(row)}"
    )

row = row.iloc[0]

question = str(row["question"])
gold_source = str(row["source_file"])

print("=" * 80)
print("NCERT RAG — HARD-DISTRACTOR VERIFICATION")
print("=" * 80)

print("\nTARGET QUESTION")
print("-" * 80)
print(question)

print("\nGold source:")
print(gold_source)

print("Gold PDF page(s):")
print(row["gold_pdf_pages"])

# ------------------------------------------------------------
# 3. LOAD ENGLISH CHUNKS
# ------------------------------------------------------------

if not os.path.exists(CHUNKS_PATH):
    raise FileNotFoundError(
        f"❌ Missing chunks:\n{CHUNKS_PATH}"
    )

chunks = pd.read_csv(CHUNKS_PATH)

print("\n✅ English chunks loaded:", len(chunks))
print("Chunk columns:", list(chunks.columns))

# ------------------------------------------------------------
# 4. LOAD MATCHING EMBEDDINGS
# ------------------------------------------------------------

if not os.path.exists(EMBEDDINGS_PATH):
    raise FileNotFoundError(
        f"❌ Missing embeddings:\n{EMBEDDINGS_PATH}"
    )

embeddings = np.load(
    EMBEDDINGS_PATH,
    mmap_mode="r"
)

print("\n✅ Embeddings loaded")
print("Embedding shape:", embeddings.shape)

# ------------------------------------------------------------
# 5. CRITICAL ALIGNMENT CHECK
# ------------------------------------------------------------

if len(chunks) != embeddings.shape[0]:
    raise ValueError(
        "\n❌ CRITICAL MISMATCH\n"
        f"Chunks: {len(chunks)}\n"
        f"Embeddings: {embeddings.shape[0]}\n"
        "These artifacts cannot safely be used together."
    )

print(
    "\n✅ Chunk/embedding row count matches:",
    len(chunks)
)

# ------------------------------------------------------------
# 6. LOAD MODEL
# ------------------------------------------------------------

MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

print("\nLoading embedding model:")
print(MODEL_NAME)

model = SentenceTransformer(MODEL_NAME)

# ------------------------------------------------------------
# 7. EMBED QUESTION
# ------------------------------------------------------------

question_embedding = model.encode(
    [question],
    normalize_embeddings=True,
    show_progress_bar=False
)[0]

# ------------------------------------------------------------
# 8. NORMALIZE CORPUS EMBEDDINGS
# ------------------------------------------------------------

# The stored embeddings should already be normalized,
# but we explicitly normalize the query and corpus for
# cosine similarity safety.

corpus_embeddings = np.asarray(
    embeddings,
    dtype=np.float32
)

norms = np.linalg.norm(
    corpus_embeddings,
    axis=1,
    keepdims=True
)

norms[norms == 0] = 1.0

corpus_embeddings = (
    corpus_embeddings / norms
)

# ------------------------------------------------------------
# 9. COSINE SIMILARITY
# ------------------------------------------------------------

scores = corpus_embeddings @ question_embedding

results = chunks.copy()

results["similarity"] = scores

# ------------------------------------------------------------
# 10. REMOVE GOLD SOURCE
# ------------------------------------------------------------

non_gold = results[
    results["source_file"].astype(str) != gold_source
].copy()

print(
    "\nNon-gold chunks available:",
    len(non_gold)
)

# ------------------------------------------------------------
# 11. TOP SEMANTIC CANDIDATES
# ------------------------------------------------------------

top = (
    non_gold
    .sort_values(
        "similarity",
        ascending=False
    )
    .head(15)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 12. SIMPLE LEXICAL OVERLAP
# ------------------------------------------------------------

def tokenize(text):
    return set(
        re.findall(
            r"\b[a-zA-Z]{2,}\b",
            str(text).lower()
        )
    )

question_tokens = tokenize(question)

def lexical_overlap(text):

    tokens = tokenize(text)

    if not question_tokens:
        return 0.0

    return len(
        question_tokens & tokens
    ) / len(question_tokens)

top["lexical_overlap"] = (
    top["text"]
    .apply(lexical_overlap)
)

# ------------------------------------------------------------
# 13. DISPLAY
# ------------------------------------------------------------

display_columns = [
    "similarity",
    "lexical_overlap",
    "class",
    "subject",
    "book",
    "source_file",
    "pages",
    "text"
]

print("\n" + "=" * 80)
print("TOP 15 NON-GOLD SEMANTIC CANDIDATES")
print("=" * 80)

display(
    top[display_columns]
)

# ------------------------------------------------------------
# 14. PRELIMINARY DIAGNOSTIC
# ------------------------------------------------------------

max_similarity = float(
    top["similarity"].max()
) if len(top) else 0.0

print("\n" + "=" * 80)
print("PRELIMINARY DIAGNOSTIC")
print("=" * 80)

print(
    f"Highest non-gold similarity: "
    f"{max_similarity:.4f}"
)

if len(top) == 0:

    print(
        "⚠️ No non-gold candidates found."
    )

elif max_similarity < 0.45:

    print(
        "⚠️ No strongly similar non-gold candidate "
        "was found."
    )

else:

    print(
        "🔎 Potential semantic distractor(s) found."
    )
    print(
        "⚠️ MANUAL SOURCE INSPECTION REQUIRED."
    )

print(
    "\nIMPORTANT:"
    "\nA hard distractor is valid ONLY if the candidate "
    "is genuinely similar/relevant but does NOT actually "
    "answer the question."
)

print("=" * 80)
```

```python
# ============================================================
# NCERT RAG — HARD-DISTRACTOR CANDIDATE MINING
#
# Goal:
#   Find pairs of semantically similar chunks from DIFFERENT
#   NCERT sources that could support a genuine hard-distractor
#   benchmark question.
#
# IMPORTANT:
#   This cell ONLY mines candidates.
#   It does NOT modify the benchmark.
# ============================================================

import os
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

CHUNKS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/english_only_chunks.csv"
)

EMBEDDINGS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/indexes/embeddings/"
    "chunk_embeddings.npy"
)

# ------------------------------------------------------------
# 2. LOAD DATA
# ------------------------------------------------------------

chunks = pd.read_csv(CHUNKS_PATH)

embeddings = np.load(
    EMBEDDINGS_PATH
).astype(np.float32)

if len(chunks) != embeddings.shape[0]:
    raise ValueError(
        f"❌ Chunk/embedding mismatch: "
        f"{len(chunks)} chunks vs "
        f"{embeddings.shape[0]} embeddings"
    )

print("=" * 80)
print("NCERT RAG — HARD-DISTRACTOR CANDIDATE MINING")
print("=" * 80)

print(f"\n✅ Chunks: {len(chunks)}")
print(f"✅ Embeddings: {embeddings.shape}")

# ------------------------------------------------------------
# 3. NORMALIZE EMBEDDINGS
# ------------------------------------------------------------

norms = np.linalg.norm(
    embeddings,
    axis=1,
    keepdims=True
)

norms[norms == 0] = 1.0

embeddings = embeddings / norms

# ------------------------------------------------------------
# 4. CREATE SOURCE IDENTIFIER
# ------------------------------------------------------------

chunks = chunks.copy()

chunks["source_key"] = (
    chunks["source_file"].astype(str)
)

# ------------------------------------------------------------
# 5. FIND SEMANTIC NEIGHBOURS
#
# For each chunk, find its strongest semantic neighbour
# from a DIFFERENT source file.
# ------------------------------------------------------------

print("\n🔎 Mining cross-source semantic neighbours...")

similarity_matrix = embeddings @ embeddings.T

# Do not allow a chunk to match itself.
np.fill_diagonal(
    similarity_matrix,
    -1.0
)

candidate_rows = []

for i in range(len(chunks)):

    scores = similarity_matrix[i].copy()

    # Exclude chunks from the SAME source file.
    same_source = (
        chunks["source_key"].values
        == chunks.iloc[i]["source_key"]
    )

    scores[same_source] = -1.0

    # Get top 5 cross-source neighbours.
    top_indices = np.argpartition(
        scores,
        -5
    )[-5:]

    top_indices = top_indices[
        np.argsort(
            scores[top_indices]
        )[::-1]
    ]

    for j in top_indices:

        if scores[j] <= 0:
            continue

        candidate_rows.append({
            "gold_index": i,
            "distractor_index": int(j),
            "similarity": float(scores[j])
        })

# ------------------------------------------------------------
# 6. BUILD CANDIDATE TABLE
# ------------------------------------------------------------

pairs = pd.DataFrame(candidate_rows)

# Only keep reasonably strong semantic similarity.
pairs = pairs[
    pairs["similarity"] >= 0.50
].copy()

pairs = pairs.sort_values(
    "similarity",
    ascending=False
).reset_index(drop=True)

print(
    f"\n✅ Candidate pairs with similarity >= 0.50: "
    f"{len(pairs)}"
)

# ------------------------------------------------------------
# 7. ADD SOURCE METADATA
# ------------------------------------------------------------

if len(pairs) > 0:

    gold_meta = chunks.iloc[
        pairs["gold_index"].values
    ].reset_index(drop=True)

    distractor_meta = chunks.iloc[
        pairs["distractor_index"].values
    ].reset_index(drop=True)

    pairs["gold_class"] = gold_meta["class"]
    pairs["gold_subject"] = gold_meta["subject"]
    pairs["gold_book"] = gold_meta["book"]
    pairs["gold_source"] = gold_meta["source_file"]
    pairs["gold_pages"] = gold_meta["pages"]
    pairs["gold_text"] = gold_meta["text"]

    pairs["distractor_class"] = (
        distractor_meta["class"]
    )
    pairs["distractor_subject"] = (
        distractor_meta["subject"]
    )
    pairs["distractor_book"] = (
        distractor_meta["book"]
    )
    pairs["distractor_source"] = (
        distractor_meta["source_file"]
    )
    pairs["distractor_pages"] = (
        distractor_meta["pages"]
    )
    pairs["distractor_text"] = (
        distractor_meta["text"]
    )

# ------------------------------------------------------------
# 8. REMOVE VERY LONG / GENERIC PAIRS
#
# Generic textbook metadata and assessment instructions often
# produce misleading semantic similarity.
# ------------------------------------------------------------

if len(pairs) > 0:

    generic_terms = [
        "note to the teacher",
        "assessment",
        "learning outcomes",
        "learning indicators",
        "textbook",
        "teacher",
        "curricular goals",
        "pedagogical"
    ]

    def looks_generic(text):

        text = str(text).lower()

        return any(
            term in text
            for term in generic_terms
        )

    pairs = pairs[
        ~pairs["gold_text"].apply(looks_generic)
        &
        ~pairs["distractor_text"].apply(looks_generic)
    ].copy()

# ------------------------------------------------------------
# 9. DISPLAY BEST CANDIDATES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TOP CROSS-SOURCE CANDIDATE PAIRS")
print("=" * 80)

if len(pairs) == 0:

    print(
        "\n⚠️ No strong non-generic candidate pairs found."
    )

else:

    display(
        pairs[
            [
                "similarity",

                "gold_class",
                "gold_subject",
                "gold_book",
                "gold_source",
                "gold_pages",
                "gold_text",

                "distractor_class",
                "distractor_subject",
                "distractor_book",
                "distractor_source",
                "distractor_pages",
                "distractor_text"
            ]
        ].head(30)
    )

# ------------------------------------------------------------
# 10. SAVE DIAGNOSTIC ONLY
# ------------------------------------------------------------

OUTPUT_DIR = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/hard_distractor_mining"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

OUTPUT_PATH = os.path.join(
    OUTPUT_DIR,
    "hard_distractor_candidate_pairs.csv"
)

pairs.to_csv(
    OUTPUT_PATH,
    index=False
)

print("\n" + "=" * 80)
print("DONE")
print("=" * 80)

print(
    f"\n📄 Candidate pairs saved to:\n{OUTPUT_PATH}"
)

print(
    "\n⚠️ No benchmark rows were modified."
)

print(
    "\nNEXT:"
    "\nInspect the displayed pairs and look for a pair where:"
    "\n  1. The gold passage clearly answers a natural question."
    "\n  2. The distractor passage is genuinely related."
    "\n  3. The distractor does NOT answer that question."
    "\n  4. Both passages come from different source files."
    "\n  5. The similarity is meaningful, not merely generic textbook language."
)
```

```python
# ============================================================
# NCERT RAG — TARGETED HARD-DISTRACTOR MINING
#
# This is a refinement of the previous mining step.
# It filters obvious textbook metadata / boilerplate and
# searches for genuinely content-level cross-book similarity.
#
# NO benchmark files are modified.
# ============================================================

import os
import re
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

CHUNKS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/english_only_chunks.csv"
)

EMBEDDINGS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/indexes/embeddings/"
    "chunk_embeddings.npy"
)

# ------------------------------------------------------------
# 2. LOAD
# ------------------------------------------------------------

chunks = pd.read_csv(CHUNKS_PATH)

embeddings = np.load(
    EMBEDDINGS_PATH
).astype(np.float32)

if len(chunks) != embeddings.shape[0]:
    raise ValueError(
        f"❌ Mismatch: {len(chunks)} chunks vs "
        f"{embeddings.shape[0]} embeddings"
    )

print("=" * 80)
print("TARGETED HARD-DISTRACTOR MINING")
print("=" * 80)

print(f"\nChunks: {len(chunks)}")
print(f"Embeddings: {embeddings.shape}")

# ------------------------------------------------------------
# 3. NORMALIZE
# ------------------------------------------------------------

norms = np.linalg.norm(
    embeddings,
    axis=1,
    keepdims=True
)

norms[norms == 0] = 1.0

embeddings = embeddings / norms

# ------------------------------------------------------------
# 4. TEXT QUALITY FILTER
# ------------------------------------------------------------

# These are strong indicators of front matter / metadata.
BOILERPLATE_TERMS = [
    "constitution of india",
    "national institute of educational",
    "advisory council",
    "professor,",
    "chancellor,",
    "copyright",
    "isbn",
    "reprint",
    "publication division",
    "editorial board",
    "textbook development",
    "learning outcomes",
    "learning indicators",
    "note to the teacher",
    "curricular goals",
    "pedagogical",
    "assessment framework",
    "academic authority",
    "department of school education",
    "ministry of education",
]

def is_content_chunk(text):

    text = str(text).lower()

    # Too short to make a useful distractor.
    words = text.split()

    if len(words) < 35:
        return False

    # Extremely metadata-heavy chunks.
    boilerplate_hits = sum(
        term in text
        for term in BOILERPLATE_TERMS
    )

    if boilerplate_hits >= 2:
        return False

    # Names / institutional pages often have lots of commas
    # and little actual instructional content.
    if len(words) >= 35:
        comma_ratio = text.count(",") / len(words)

        if comma_ratio > 0.12:
            return False

    return True


content_mask = chunks["text"].apply(
    is_content_chunk
).values

print(
    "\nContent-quality chunks:",
    int(content_mask.sum()),
    "/",
    len(chunks)
)

# ------------------------------------------------------------
# 5. SOURCE IDENTIFIERS
# ------------------------------------------------------------

source_files = (
    chunks["source_file"]
    .astype(str)
    .values
)

books = (
    chunks["book"]
    .astype(str)
    .values
)

subjects = (
    chunks["subject"]
    .astype(str)
    .values
)

# ------------------------------------------------------------
# 6. SEMANTIC SEARCH
# ------------------------------------------------------------

# Rather than materializing the entire 2033 x 2033 matrix,
# process each chunk and retain only its strongest candidate.
#
# This keeps the diagnostic lightweight.

candidate_rows = []

for i in range(len(chunks)):

    if not content_mask[i]:
        continue

    scores = embeddings @ embeddings[i]

    # Never compare against itself.
    scores[i] = -1.0

    # Only content-quality chunks.
    scores[~content_mask] = -1.0

    # Must be a DIFFERENT source file.
    scores[source_files == source_files[i]] = -1.0

    # Prefer genuinely different books.
    scores[books == books[i]] = -1.0

    # Get top candidates.
    k = min(10, len(scores) - 1)

    idx = np.argpartition(
        scores,
        -k
    )[-k:]

    idx = idx[
        np.argsort(
            scores[idx]
        )[::-1]
    ]

    for j in idx:

        similarity = float(scores[j])

        if similarity < 0.65:
            continue

        candidate_rows.append({
            "gold_index": i,
            "distractor_index": int(j),
            "similarity": similarity
        })

# ------------------------------------------------------------
# 7. BUILD TABLE
# ------------------------------------------------------------

pairs = pd.DataFrame(candidate_rows)

if len(pairs) == 0:

    print(
        "\n⚠️ No candidates above similarity 0.65."
    )

else:

    gold = chunks.iloc[
        pairs["gold_index"].values
    ].reset_index(drop=True)

    distractor = chunks.iloc[
        pairs["distractor_index"].values
    ].reset_index(drop=True)

    pairs["gold_class"] = gold["class"]
    pairs["gold_subject"] = gold["subject"]
    pairs["gold_book"] = gold["book"]
    pairs["gold_source"] = gold["source_file"]
    pairs["gold_pages"] = gold["pages"]
    pairs["gold_text"] = gold["text"]

    pairs["distractor_class"] = distractor["class"]
    pairs["distractor_subject"] = distractor["subject"]
    pairs["distractor_book"] = distractor["book"]
    pairs["distractor_source"] = distractor["source_file"]
    pairs["distractor_pages"] = distractor["pages"]
    pairs["distractor_text"] = distractor["text"]

# ------------------------------------------------------------
# 8. LEXICAL OVERLAP
# ------------------------------------------------------------

def tokens(text):

    return set(
        re.findall(
            r"\b[a-zA-Z]{3,}\b",
            str(text).lower()
        )
    )

def overlap(a, b):

    ta = tokens(a)
    tb = tokens(b)

    if not ta or not tb:
        return 0.0

    return len(ta & tb) / min(
        len(ta),
        len(tb)
    )

if len(pairs) > 0:

    pairs["lexical_overlap"] = [
        overlap(a, b)
        for a, b in zip(
            pairs["gold_text"],
            pairs["distractor_text"]
        )
    ]

# ------------------------------------------------------------
# 9. RANK
#
# We want semantic similarity, but not nearly identical text.
# Extremely high similarity can indicate duplicated boilerplate.
# ------------------------------------------------------------

if len(pairs) > 0:

    pairs["candidate_score"] = (
        pairs["similarity"]
        - 0.15 * pairs["lexical_overlap"]
    )

    pairs = pairs.sort_values(
        [
            "candidate_score",
            "similarity"
        ],
        ascending=False
    ).reset_index(drop=True)

# ------------------------------------------------------------
# 10. DISPLAY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TARGETED CANDIDATES")
print("=" * 80)

if len(pairs) == 0:

    print(
        "\nNo suitable candidate pairs found."
    )

else:

    display(
        pairs[
            [
                "candidate_score",
                "similarity",
                "lexical_overlap",

                "gold_class",
                "gold_subject",
                "gold_book",
                "gold_pages",
                "gold_text",

                "distractor_class",
                "distractor_subject",
                "distractor_book",
                "distractor_pages",
                "distractor_text"
            ]
        ].head(50)
    )

print("\n" + "=" * 80)
print("INTERPRETATION")
print("=" * 80)

print("""
A good hard-distractor pair should look like this:

GOLD:
  A specific concept/event/action that clearly supports
  a natural student question.

DISTRACTOR:
  A different NCERT passage discussing a related concept,
  object, action, or situation, but which does NOT answer
  that question.

We do NOT want:
  - textbook metadata
  - teacher notes
  - Constitution/front matter
  - duplicated boilerplate
  - merely sharing words
  - nearly identical passages
""")

print("=" * 80)
print("NO BENCHMARK FILE WAS MODIFIED")
print("=" * 80)
```

```python
# ============================================================
# NCERT RAG — FINAL HARD-DISTRACTOR CANDIDATE MINING
#
# Strategy:
#   1. Ignore early textbook front matter (pages <= 12)
#   2. Require substantial student-facing content
#   3. Require DIFFERENT books
#   4. Require semantic similarity
#   5. Reject near-duplicate wording
#   6. Prefer candidates with moderate lexical overlap
#
# NO benchmark files are modified.
# ============================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

CHUNKS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/english_only_chunks.csv"
)

EMBEDDINGS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/indexes/embeddings/"
    "chunk_embeddings.npy"
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

chunks = pd.read_csv(CHUNKS_PATH)

embeddings = np.load(
    EMBEDDINGS_PATH
).astype(np.float32)

if len(chunks) != embeddings.shape[0]:
    raise ValueError(
        f"❌ Mismatch: {len(chunks)} chunks vs "
        f"{embeddings.shape[0]} embeddings"
    )

# Normalize
norms = np.linalg.norm(
    embeddings,
    axis=1,
    keepdims=True
)

norms[norms == 0] = 1.0

embeddings = embeddings / norms

print("=" * 80)
print("FINAL TARGETED HARD-DISTRACTOR MINING")
print("=" * 80)

print(
    f"\nLoaded {len(chunks)} chunks."
)

# ------------------------------------------------------------
# PAGE PARSER
# ------------------------------------------------------------

def first_page(value):

    try:
        text = str(value)

        # Handle values such as:
        # "83,84,85"
        # "[83, 84]"
        # "83"
        nums = re.findall(
            r"\d+",
            text
        )

        if nums:
            return int(nums[0])

    except Exception:
        pass

    return 9999


chunks["first_page_num"] = (
    chunks["first_page"]
    .apply(first_page)
)

# ------------------------------------------------------------
# TEXT QUALITY
# ------------------------------------------------------------

BAD_TERMS = [
    "constitution of india",
    "national institute of educational",
    "advisory council",
    "professor,",
    "chancellor,",
    "copyright",
    "isbn",
    "reprint",
    "publication division",
    "editorial board",
    "textbook development",
    "note to the teacher",
    "curricular goals",
    "learning outcomes",
    "learning indicators",
    "pedagogical",
    "ministry of education",
    "department of school education",
    "academic authority",
    "foreword",
    "preface",
    "acknowledgement",
    "acknowledgment",
]

def usable_content(text):

    text = str(text).lower()
    words = text.split()

    # Enough substance.
    if len(words) < 45:
        return False

    # Reject obvious front matter.
    hits = sum(
        term in text
        for term in BAD_TERMS
    )

    if hits >= 1:
        return False

    # Reject extremely punctuation-heavy metadata.
    comma_ratio = text.count(",") / max(
        len(words),
        1
    )

    if comma_ratio > 0.10:
        return False

    return True


content_mask = chunks["text"].apply(
    usable_content
).values

# ------------------------------------------------------------
# EXCLUDE FRONT MATTER
# ------------------------------------------------------------

page_mask = (
    chunks["first_page_num"] > 12
).values

valid_mask = (
    content_mask
    &
    page_mask
)

print(
    "\nUsable student-content chunks:",
    int(valid_mask.sum())
)

# ------------------------------------------------------------
# SOURCE INFORMATION
# ------------------------------------------------------------

source = (
    chunks["source_file"]
    .astype(str)
    .values
)

book = (
    chunks["book"]
    .astype(str)
    .values
)

# ------------------------------------------------------------
# TOKEN FUNCTION
# ------------------------------------------------------------

def token_set(text):

    return set(
        re.findall(
            r"\b[a-zA-Z]{3,}\b",
            str(text).lower()
        )
    )

token_cache = {}

for i in np.where(valid_mask)[0]:

    token_cache[i] = token_set(
        chunks.iloc[i]["text"]
    )

# ------------------------------------------------------------
# MINE PAIRS
# ------------------------------------------------------------

candidate_rows = []

valid_indices = np.where(
    valid_mask
)[0]

print(
    "\nSearching",
    len(valid_indices),
    "content chunks..."
)

for i in valid_indices:

    scores = embeddings @ embeddings[i]

    # Remove itself.
    scores[i] = -1.0

    # Only valid content.
    scores[~valid_mask] = -1.0

    # Different SOURCE FILE.
    scores[source == source[i]] = -1.0

    # Different BOOK.
    scores[book == book[i]] = -1.0

    # Candidate semantic range.
    scores[scores < 0.60] = -1.0

    # Get top 15.
    k = min(
        15,
        len(scores) - 1
    )

    idx = np.argpartition(
        scores,
        -k
    )[-k:]

    idx = idx[
        np.argsort(
            scores[idx]
        )[::-1]
    ]

    for j in idx:

        sim = float(scores[j])

        if sim < 0.60:
            continue

        a = token_cache[i]
        b = token_cache[j]

        if not a or not b:
            continue

        lexical = (
            len(a & b)
            /
            min(len(a), len(b))
        )

        # Reject near duplicates.
        if lexical > 0.75:
            continue

        # Reject extremely weak lexical connection.
        if lexical < 0.05:
            continue

        # Candidate score rewards semantic similarity
        # while penalizing wording duplication.
        score = (
            sim
            - 0.20 * lexical
        )

        candidate_rows.append({
            "gold_index": int(i),
            "distractor_index": int(j),
            "similarity": sim,
            "lexical_overlap": lexical,
            "candidate_score": score
        })

# ------------------------------------------------------------
# BUILD TABLE
# ------------------------------------------------------------

pairs = pd.DataFrame(
    candidate_rows
)

print(
    "\nRaw candidate pairs:",
    len(pairs)
)

if len(pairs) > 0:

    gold = chunks.iloc[
        pairs["gold_index"].values
    ].reset_index(drop=True)

    distractor = chunks.iloc[
        pairs["distractor_index"].values
    ].reset_index(drop=True)

    pairs["gold_class"] = gold["class"]
    pairs["gold_subject"] = gold["subject"]
    pairs["gold_book"] = gold["book"]
    pairs["gold_source"] = gold["source_file"]
    pairs["gold_pages"] = gold["pages"]
    pairs["gold_text"] = gold["text"]

    pairs["distractor_class"] = (
        distractor["class"]
    )
    pairs["distractor_subject"] = (
        distractor["subject"]
    )
    pairs["distractor_book"] = (
        distractor["book"]
    )
    pairs["distractor_source"] = (
        distractor["source_file"]
    )
    pairs["distractor_pages"] = (
        distractor["pages"]
    )
    pairs["distractor_text"] = (
        distractor["text"]
    )

    pairs = pairs.sort_values(
        "candidate_score",
        ascending=False
    ).reset_index(
        drop=True
    )

# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("BEST CONTENT-LEVEL CANDIDATES")
print("=" * 80)

if len(pairs) == 0:

    print(
        "\n⚠️ No candidate pairs survived."
    )

else:

    display(
        pairs[
            [
                "candidate_score",
                "similarity",
                "lexical_overlap",

                "gold_class",
                "gold_subject",
                "gold_book",
                "gold_pages",
                "gold_text",

                "distractor_class",
                "distractor_subject",
                "distractor_book",
                "distractor_pages",
                "distractor_text"
            ]
        ].head(40)
    )

# ------------------------------------------------------------
# SAVE DIAGNOSTIC
# ------------------------------------------------------------

OUT_DIR = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/hard_distractor_mining"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)

OUT_PATH = os.path.join(
    OUT_DIR,
    "hard_distractor_content_candidates.csv"
)

pairs.to_csv(
    OUT_PATH,
    index=False
)

print("\n" + "=" * 80)
print("DONE")
print("=" * 80)

print(
    f"\nSaved diagnostic candidates to:\n{OUT_PATH}"
)

print(
    "\n❗ Benchmark remains untouched."
)

print(
    "\nThe candidates we want should have:"
    "\n  • similarity roughly 0.60–0.85"
    "\n  • moderate lexical overlap"
    "\n  • different books"
    "\n  • actual student-facing content"
    "\n  • clearly related concepts"
    "\n  • but NOT the same answer"
)
```

```python
# ============================================================
# NCERT RAG — V2P008 REPLACEMENT CANDIDATE
#
# Goal:
#   Find a fresh DIRECT/PARAPHRASE question candidate from
#   the verified English NCERT corpus.
#
# IMPORTANT:
#   - Does NOT modify the benchmark.
#   - Does NOT invent textbook facts.
#   - Candidate must come directly from corpus text.
# ============================================================

import os
import re
import random
import pandas as pd

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

CHUNKS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/english_only_chunks.csv"
)

# ------------------------------------------------------------
# 2. LOAD
# ------------------------------------------------------------

chunks = pd.read_csv(
    CHUNKS_PATH
)

print("=" * 80)
print("V2P008 REPLACEMENT CANDIDATE GENERATOR")
print("=" * 80)

print(
    f"\nLoaded chunks: {len(chunks)}"
)

# ------------------------------------------------------------
# 3. QUALITY FILTER
#
# We want actual student-facing material, not:
#   - prefaces
#   - credits
#   - teacher notes
#   - metadata
#   - constitution/front matter
# ------------------------------------------------------------

BAD_TERMS = [
    "constitution of india",
    "national institute of educational",
    "advisory council",
    "professor,",
    "chancellor,",
    "copyright",
    "isbn",
    "reprint",
    "publication division",
    "editorial board",
    "textbook development",
    "note to the teacher",
    "curricular goals",
    "learning outcomes",
    "learning indicators",
    "pedagogical",
    "ministry of education",
    "department of school education",
    "academic authority",
    "foreword",
    "preface",
    "acknowledgement",
    "acknowledgment",
]

def clean_text(text):

    text = str(text)

    # Collapse whitespace.
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def is_good_chunk(text):

    text = clean_text(text)

    lower = text.lower()

    words = lower.split()

    # Need enough content.
    if len(words) < 50:
        return False

    # Avoid enormous chunks.
    if len(words) > 220:
        return False

    # Reject obvious metadata.
    if any(
        term in lower
        for term in BAD_TERMS
    ):
        return False

    # Need actual sentence structure.
    if lower.count(".") < 2:
        return False

    return True


candidate_chunks = chunks[
    chunks["text"].apply(is_good_chunk)
].copy()

candidate_chunks["clean_text"] = (
    candidate_chunks["text"]
    .apply(clean_text)
)

print(
    "\nStudent-facing candidate chunks:",
    len(candidate_chunks)
)

# ------------------------------------------------------------
# 4. LOOK FOR QUESTION-FRIENDLY SENTENCES
#
# We prefer chunks containing:
#   - actions
#   - named objects
#   - explicit facts
#   - "what/where/when/how/which" style information
#
# We DO NOT generate the question yet.
# We only identify strong source passages.
# ------------------------------------------------------------

QUESTION_SIGNAL_TERMS = [
    "is",
    "are",
    "was",
    "were",
    "has",
    "have",
    "can",
    "should",
    "used",
    "called",
    "means",
    "because",
    "when",
    "where",
    "how",
    "which",
    "why",
    "first",
    "next",
    "then",
    "before",
    "after",
]

def questionability(text):

    lower = text.lower()

    score = 0

    for term in QUESTION_SIGNAL_TERMS:

        # Whole-word matching.
        score += len(
            re.findall(
                rf"\b{re.escape(term)}\b",
                lower
            )
        )

    # Prefer moderate length.
    words = len(lower.split())

    if 70 <= words <= 170:
        score += 5

    return score


candidate_chunks["questionability"] = (
    candidate_chunks["clean_text"]
    .apply(questionability)
)

# ------------------------------------------------------------
# 5. DIVERSITY
#
# Don't just return ten chunks from one book.
# ------------------------------------------------------------

candidate_chunks = (
    candidate_chunks
    .sort_values(
        "questionability",
        ascending=False
    )
)

selected = []

seen_books = set()

for _, row in candidate_chunks.iterrows():

    book = (
        str(row["source_file"])
    )

    if book in seen_books:
        continue

    selected.append(row)

    seen_books.add(book)

    if len(selected) >= 20:
        break

selected = pd.DataFrame(
    selected
)

# ------------------------------------------------------------
# 6. DISPLAY SOURCE MATERIAL
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("20 FRESH SOURCE-GROUNDED CANDIDATES")
print("=" * 80)

display(
    selected[
        [
            "class",
            "subject",
            "book",
            "source_file",
            "pages",
            "clean_text"
        ]
    ]
)

print("\n" + "=" * 80)
print("HOW TO USE THIS")
print("=" * 80)

print("""
Pick ONE passage where you can naturally ask a child:

  • What ...?
  • Which ...?
  • Where ...?
  • When ...?
  • Why ...?
  • How ...?

The answer must be directly stated in this passage.

Do NOT choose:
  • introductions
  • generic textbook descriptions
  • teacher instructions
  • metadata
  • passages requiring outside knowledge

After selecting one, we will create ONE V2P008 row
from that exact source and run the same provenance
validation used for V2P001–V2P010.
""")
```

```python
# ============================================================
# REPLACE V2P008 WITH A VERIFIED CORPUS-GROUNDED QUESTION
#
# Source:
#   Class 4 Arts — Bansuri
#   PDF page 102
#
# This cell modifies ONLY the V2P008 row in a NEW workbook.
# The original pilot and v2 workbook remain untouched.
# ============================================================

import os
import pandas as pd

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

INPUT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v2.xlsx"
)

OUTPUT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v3.xlsx"
)

# ------------------------------------------------------------
# LOAD WORKBOOK
# ------------------------------------------------------------

xls = pd.ExcelFile(INPUT_PATH)

benchmark = pd.read_excel(
    INPUT_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

# ------------------------------------------------------------
# VERIFY V2P008 EXISTS EXACTLY ONCE
# ------------------------------------------------------------

mask = (
    benchmark["question_id"]
    .astype(str)
    .eq("V2P008")
)

if mask.sum() != 1:
    raise ValueError(
        f"❌ Expected exactly one V2P008 row, "
        f"found {mask.sum()}"
    )

# ------------------------------------------------------------
# VERIFIED SOURCE
# ------------------------------------------------------------

question = (
    "What can the audience add or change "
    "to make the story continue in a new way?"
)

expected_answer = (
    "The audience can add a new prop, add a new "
    "character, or change the story by adding a "
    "new problem."
)

evidence_excerpt = (
    "The audience can choose to: Add a new prop; "
    "Add a new character; Change the direction of "
    "the story by adding a new problem."
)

source_file = (
    "Class_04/Arts/Bansuri.pdf"
)

gold_pdf_pages = "[102]"
gold_pages = "[102]"

# ------------------------------------------------------------
# UPDATE ONLY THE NECESSARY FIELDS
# ------------------------------------------------------------

idx = benchmark.index[mask][0]

benchmark.loc[idx, "question"] = question
benchmark.loc[idx, "expected_answer"] = expected_answer
benchmark.loc[idx, "evidence_excerpt"] = evidence_excerpt

benchmark.loc[idx, "class"] = "Class_04"
benchmark.loc[idx, "subject"] = "Arts"
benchmark.loc[idx, "book"] = "Bansuri"
benchmark.loc[idx, "source_file"] = source_file

benchmark.loc[idx, "gold_pages"] = gold_pages
benchmark.loc[idx, "gold_pdf_pages"] = gold_pdf_pages

benchmark.loc[idx, "question_type"] = "DIRECT"

benchmark.loc[idx, "benchmark_status"] = "PILOT_REVIEW"

benchmark.loc[idx, "annotation_status"] = "SOURCE_VERIFIED"

# ------------------------------------------------------------
# KEEP ALL OTHER SHEETS UNCHANGED
# ------------------------------------------------------------

with pd.ExcelWriter(
    OUTPUT_PATH,
    engine="openpyxl"
) as writer:

    for sheet in xls.sheet_names:

        if sheet == "Benchmark_V2_Pilot":
            benchmark.to_excel(
                writer,
                sheet_name=sheet,
                index=False
            )

        else:
            pd.read_excel(
                INPUT_PATH,
                sheet_name=sheet
            ).to_excel(
                writer,
                sheet_name=sheet,
                index=False
            )

# ------------------------------------------------------------
# VERIFY
# ------------------------------------------------------------

check = pd.read_excel(
    OUTPUT_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

row = check[
    check["question_id"].astype(str) == "V2P008"
].iloc[0]

print("=" * 80)
print("V2P008 REPLACED")
print("=" * 80)

print("\nQuestion:")
print(row["question"])

print("\nExpected answer:")
print(row["expected_answer"])

print("\nSource:")
print(row["source_file"])

print("\nGold PDF pages:")
print(row["gold_pdf_pages"])

print("\nQuestion type:")
print(row["question_type"])

print("\nStatus:")
print(row["benchmark_status"])

print("\n" + "=" * 80)
print("OUTPUT")
print("=" * 80)

print(
    f"\n✅ New workbook:\n{OUTPUT_PATH}"
)

print(
    "\n✅ Original V2P008 workbook was NOT modified."
)
```

```python
# ============================================================
# CELL 15 — FINAL V2 PILOT SOURCE-GROUNDING VALIDATION
#
# Purpose:
#   Verify that every benchmark question is actually grounded
#   in its declared NCERT evidence pages.
#
# IMPORTANT:
#   - Uses gold_pdf_pages, NOT legacy printed-page numbers.
#   - Multi-page questions are checked page-by-page.
#   - Does NOT automatically certify hard distractors.
#   - Does NOT modify the benchmark.
# ============================================================

import os
import json
import ast
import re
import pandas as pd

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

BENCHMARK_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v3.xlsx"
)

PAGES_PATH = (
    ROOT +
    "/02_PROCESSED_DATA/corpus_v2_clean/"
    "page_text/all_pages.jsonl"
)

# ------------------------------------------------------------
# VERIFY FILES EXIST
# ------------------------------------------------------------

if not os.path.exists(BENCHMARK_PATH):
    raise FileNotFoundError(
        f"❌ Benchmark not found:\n{BENCHMARK_PATH}"
    )

if not os.path.exists(PAGES_PATH):
    raise FileNotFoundError(
        f"❌ Page corpus not found:\n{PAGES_PATH}"
    )

print("✅ Benchmark found")
print("✅ Page corpus found")

# ------------------------------------------------------------
# LOAD BENCHMARK
# ------------------------------------------------------------

benchmark = pd.read_excel(
    BENCHMARK_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

required_columns = [
    "question_id",
    "question",
    "class",
    "subject",
    "book",
    "source_file",
    "gold_pdf_pages",
    "question_type",
    "expected_answer",
    "evidence_excerpt"
]

missing = [
    c for c in required_columns
    if c not in benchmark.columns
]

if missing:
    raise ValueError(
        f"❌ Missing required columns: {missing}"
    )

# ------------------------------------------------------------
# BASIC BENCHMARK CHECKS
# ------------------------------------------------------------

if len(benchmark) != 10:
    raise ValueError(
        f"❌ Expected 10 pilot questions, found {len(benchmark)}"
    )

if benchmark["question_id"].duplicated().any():
    raise ValueError(
        "❌ Duplicate question IDs detected."
    )

print(f"✅ Benchmark contains {len(benchmark)} questions")
print("✅ Question IDs are unique")

# ------------------------------------------------------------
# LOAD ALL PAGE RECORDS
# ------------------------------------------------------------

pages = []

with open(
    PAGES_PATH,
    "r",
    encoding="utf-8"
) as f:

    for line in f:
        line = line.strip()

        if line:
            pages.append(json.loads(line))

pages_df = pd.DataFrame(pages)

print(f"✅ Loaded {len(pages_df):,} page records")

# ------------------------------------------------------------
# NORMALIZATION HELPERS
# ------------------------------------------------------------

def norm_text(text):
    """
    Conservative normalization used ONLY for evidence matching.
    Does not alter benchmark data.
    """
    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = text.replace("’", "'")
    text = text.replace("“", '"')
    text = text.replace("”", '"')
    text = text.replace("–", "-")
    text = text.replace("—", "-")

    text = re.sub(r"\s+", " ", text)

    return text.strip()


def parse_pages(value):
    """
    Safely parse gold_pdf_pages values such as:
        [129]
        [129, 130]
        "[129, 130]"
    """
    if isinstance(value, list):
        return [int(x) for x in value]

    if pd.isna(value):
        return []

    value = str(value).strip()

    try:
        parsed = ast.literal_eval(value)

        if isinstance(parsed, (list, tuple)):
            return [int(x) for x in parsed]

    except Exception:
        pass

    # Fallback: extract integers
    nums = re.findall(r"\d+", value)

    return [int(x) for x in nums]


# ------------------------------------------------------------
# PREPARE PAGE LOOKUP
# ------------------------------------------------------------

# Use normalized source_file paths for robust matching.
pages_df["source_file_norm"] = (
    pages_df["source_file"]
    .astype(str)
    .str.replace("\\", "/", regex=False)
    .str.strip("/")
)

pages_df["clean_text_norm"] = (
    pages_df["clean_text"]
    .fillna("")
    .map(norm_text)
)

# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

results = []

for _, row in benchmark.iterrows():

    qid = str(row["question_id"])

    source_file = (
        str(row["source_file"])
        .replace("\\", "/")
        .strip("/")
    )

    gold_pages = parse_pages(
        row["gold_pdf_pages"]
    )

    expected_answer = norm_text(
        row["expected_answer"]
    )

    evidence_excerpt = norm_text(
        row["evidence_excerpt"]
    )

    question_type = str(
        row["question_type"]
    ).upper()

    # --------------------------------------------------------
    # FIND SOURCE BOOK
    # --------------------------------------------------------

    source_pages = pages_df[
        pages_df["source_file_norm"] == source_file
    ]

    if source_pages.empty:

        results.append({
            "question_id": qid,
            "status": "FAIL",
            "reason": "SOURCE_FILE_NOT_FOUND",
            "source_file": source_file,
            "gold_pdf_pages": gold_pages,
            "pages_found": [],
            "evidence_found": False
        })

        continue

    # --------------------------------------------------------
    # CHECK EVERY GOLD PAGE
    # --------------------------------------------------------

    page_results = []

    for page_num in gold_pages:

        matches = source_pages[
            source_pages["page"].astype(int) == int(page_num)
        ]

        if matches.empty:

            page_results.append({
                "page": page_num,
                "found": False,
                "evidence_match": False
            })

            continue

        page_text = " ".join(
            matches["clean_text_norm"].tolist()
        )

        # ----------------------------------------------------
        # Evidence matching
        #
        # We check whether meaningful portions of the supplied
        # evidence excerpt occur on the declared page.
        # ----------------------------------------------------

        evidence_match = False

        if evidence_excerpt:

            # Full excerpt first
            if evidence_excerpt in page_text:
                evidence_match = True

            else:
                # Break excerpt into meaningful sentences /
                # clauses and look for substantial overlap.
                pieces = re.split(
                    r"[.!?;:]+",
                    evidence_excerpt
                )

                meaningful = [
                    p.strip()
                    for p in pieces
                    if len(p.strip().split()) >= 5
                ]

                matched_pieces = 0

                for piece in meaningful:

                    words = piece.split()

                    # Use a compact phrase for robust matching.
                    if len(words) >= 8:
                        phrase = " ".join(words[:8])

                        if phrase in page_text:
                            matched_pieces += 1

                if meaningful and (
                    matched_pieces >=
                    max(1, len(meaningful) // 2)
                ):
                    evidence_match = True

        page_results.append({
            "page": page_num,
            "found": True,
            "evidence_match": evidence_match
        })

    # --------------------------------------------------------
    # MULTI-PAGE LOGIC
    #
    # Every declared page must exist.
    # For MULTI_PAGE, evidence must be present on EVERY
    # declared evidence page.
    # --------------------------------------------------------

    pages_exist = all(
        x["found"]
        for x in page_results
    )

    pages_with_evidence = [
        x["page"]
        for x in page_results
        if x["evidence_match"]
    ]

    if not pages_exist:

        status = "FAIL"
        reason = "GOLD_PAGE_NOT_FOUND"

    elif question_type == "MULTI_PAGE":

        if len(pages_with_evidence) == len(gold_pages):

            status = "PASS"
            reason = "ALL_GOLD_PAGES_GROUNDED"

        else:

            status = "FAIL"
            reason = "MULTI_PAGE_EVIDENCE_INCOMPLETE"

    else:

        if len(pages_with_evidence) >= 1:

            status = "PASS"
            reason = "SOURCE_EVIDENCE_FOUND"

        else:

            status = "REVIEW"
            reason = "PAGE_EXISTS_BUT_EVIDENCE_NOT_AUTOMATICALLY_MATCHED"

    results.append({
        "question_id": qid,
        "status": status,
        "reason": reason,
        "source_file": source_file,
        "gold_pdf_pages": gold_pages,
        "pages_found": [
            x["page"]
            for x in page_results
            if x["found"]
        ],
        "pages_with_evidence": pages_with_evidence,
        "evidence_found": len(pages_with_evidence) > 0,
        "question_type": question_type
    })


# ------------------------------------------------------------
# RESULTS DATAFRAME
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print("\n")
print("=" * 90)
print("FINAL V2 PILOT SOURCE-GROUNDING VALIDATION")
print("=" * 90)

display(
    results_df[
        [
            "question_id",
            "question_type",
            "status",
            "reason",
            "gold_pdf_pages",
            "pages_with_evidence"
        ]
    ]
)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

pass_count = (
    results_df["status"] == "PASS"
).sum()

review_count = (
    results_df["status"] == "REVIEW"
).sum()

fail_count = (
    results_df["status"] == "FAIL"
).sum()

multi_page = results_df[
    results_df["question_type"] == "MULTI_PAGE"
]

multi_page_pass = (
    multi_page["status"] == "PASS"
).sum()

print("\n" + "=" * 90)
print("SUMMARY")
print("=" * 90)

print(f"Total questions       : {len(results_df)}")
print(f"PASS                  : {pass_count}")
print(f"REVIEW                : {review_count}")
print(f"FAIL                  : {fail_count}")

print(
    f"Multi-page questions  : {len(multi_page)}"
)

print(
    f"Multi-page PASS       : {multi_page_pass}"
)

# ------------------------------------------------------------
# FINAL DECISION
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("DECISION")
print("=" * 90)

if fail_count == 0 and review_count == 0:

    print(
        "\n🟢 PILOT SOURCE GROUNDING PASSED"
    )

    print(
        "All 10 questions are grounded in their declared "
        "NCERT PDF evidence."
    )

    print(
        "The benchmark is ready for retrieval evaluation."
    )

elif fail_count == 0:

    print(
        "\n🟡 NO HARD FAILURES"
    )

    print(
        "Some questions require manual evidence review "
        "before freezing the pilot."
    )

else:

    print(
        "\n🔴 PILOT NOT READY"
    )

    print(
        "At least one question failed source grounding."
    )

print("\nValidation complete.")
```

```python
# ============================================================
# CELL 16 — CORRECT MULTI-PAGE SOURCE VALIDATION
#
# IMPORTANT:
# The previous validator incorrectly required the SAME
# evidence_excerpt to match independently on every page.
#
# For MULTI_PAGE questions, this is wrong.
#
# Here we:
#   1. Verify every declared PDF page exists.
#   2. Verify the question/answer has meaningful evidence
#      distributed across the declared pages.
#   3. Do NOT require the full combined evidence excerpt
#      to appear on every page.
#   4. Do NOT modify the benchmark.
# ============================================================

import os
import ast
import re
import pandas as pd

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

BENCHMARK_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v3.xlsx"
)

PAGES_PATH = (
    ROOT +
    "/02_PROCESSED_DATA/corpus_v2_clean/"
    "page_text/all_pages.jsonl"
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

benchmark = pd.read_excel(
    BENCHMARK_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

pages_df = pd.read_json(
    PAGES_PATH,
    lines=True
)

# ------------------------------------------------------------
# NORMALIZATION
# ------------------------------------------------------------

def norm(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = (
        text
        .replace("’", "'")
        .replace("“", '"')
        .replace("”", '"')
        .replace("–", "-")
        .replace("—", "-")
    )

    text = re.sub(r"\s+", " ", text)

    return text.strip()


def parse_pages(value):

    if isinstance(value, list):
        return [int(x) for x in value]

    if pd.isna(value):
        return []

    try:
        parsed = ast.literal_eval(str(value))

        if isinstance(parsed, (list, tuple)):
            return [int(x) for x in parsed]

    except Exception:
        pass

    return [
        int(x)
        for x in re.findall(r"\d+", str(value))
    ]


pages_df["source_file_norm"] = (
    pages_df["source_file"]
    .astype(str)
    .str.replace("\\", "/", regex=False)
    .str.strip("/")
)

pages_df["text_norm"] = (
    pages_df["clean_text"]
    .fillna("")
    .map(norm)
)

# ------------------------------------------------------------
# HELPER:
# Extract meaningful answer phrases.
#
# We use phrases from the expected answer rather than requiring
# the entire answer to occur verbatim.
# ------------------------------------------------------------

def answer_phrases(answer):

    answer = norm(answer)

    # Split on common connectors.
    pieces = re.split(
        r"\b(?:and|or|but|because|so|that|which)\b|[,;:.]",
        answer
    )

    phrases = []

    for piece in pieces:

        words = piece.strip().split()

        if len(words) >= 3:
            phrases.append(
                " ".join(words)
            )

    return phrases


# ------------------------------------------------------------
# VALIDATE
# ------------------------------------------------------------

results = []

for _, row in benchmark.iterrows():

    qid = str(row["question_id"])

    qtype = str(
        row["question_type"]
    ).upper()

    source_file = (
        str(row["source_file"])
        .replace("\\", "/")
        .strip("/")
    )

    gold_pages = parse_pages(
        row["gold_pdf_pages"]
    )

    question = norm(
        row["question"]
    )

    expected_answer = norm(
        row["expected_answer"]
    )

    evidence_excerpt = norm(
        row["evidence_excerpt"]
    )

    source_pages = pages_df[
        pages_df["source_file_norm"] == source_file
    ]

    # --------------------------------------------------------
    # SOURCE CHECK
    # --------------------------------------------------------

    if source_pages.empty:

        results.append({
            "question_id": qid,
            "question_type": qtype,
            "status": "FAIL",
            "reason": "SOURCE_FILE_NOT_FOUND",
            "gold_pdf_pages": gold_pages,
            "pages_found": [],
            "evidence_pages": []
        })

        continue

    # --------------------------------------------------------
    # PAGE CHECK
    # --------------------------------------------------------

    page_texts = {}

    for page in gold_pages:

        match = source_pages[
            source_pages["page"].astype(int) == int(page)
        ]

        if not match.empty:

            page_texts[page] = " ".join(
                match["text_norm"].tolist()
            )

    pages_found = list(page_texts.keys())

    if len(pages_found) != len(gold_pages):

        results.append({
            "question_id": qid,
            "question_type": qtype,
            "status": "FAIL",
            "reason": "GOLD_PAGE_NOT_FOUND",
            "gold_pdf_pages": gold_pages,
            "pages_found": pages_found,
            "evidence_pages": []
        })

        continue

    # --------------------------------------------------------
    # MULTI-PAGE VALIDATION
    # --------------------------------------------------------

    if qtype == "MULTI_PAGE":

        # ----------------------------------------------------
        # We expect at least 2 declared pages.
        # ----------------------------------------------------

        if len(gold_pages) < 2:

            results.append({
                "question_id": qid,
                "question_type": qtype,
                "status": "FAIL",
                "reason": "MULTI_PAGE_HAS_LESS_THAN_TWO_PAGES",
                "gold_pdf_pages": gold_pages,
                "pages_found": pages_found,
                "evidence_pages": []
            })

            continue

        # ----------------------------------------------------
        # Build answer-derived phrases.
        # ----------------------------------------------------

        phrases = answer_phrases(
            expected_answer
        )

        evidence_pages = []

        phrase_hits = {}

        for page, text in page_texts.items():

            hits = []

            # Check answer phrases.
            for phrase in phrases:

                if phrase in text:
                    hits.append(phrase)

            # Also check meaningful fragments from evidence
            # excerpt, but WITHOUT requiring the full excerpt.
            evidence_parts = re.split(
                r"[.!?;:]+",
                evidence_excerpt
            )

            for part in evidence_parts:

                words = part.strip().split()

                if len(words) >= 5:

                    # Check an 8-word leading phrase.
                    phrase = " ".join(words[:8])

                    if phrase in text:
                        hits.append(
                            phrase
                        )

            # Deduplicate.
            hits = list(dict.fromkeys(hits))

            phrase_hits[page] = hits

            if hits:
                evidence_pages.append(page)

        # ----------------------------------------------------
        # IMPORTANT:
        #
        # Each page must contribute some evidence.
        # ----------------------------------------------------

        if len(evidence_pages) == len(gold_pages):

            status = "PASS"
            reason = "MULTI_PAGE_EVIDENCE_DISTRIBUTED_ACROSS_ALL_GOLD_PAGES"

        else:

            status = "REVIEW"
            reason = "MULTI_PAGE_PAGE_CONTRIBUTION_NEEDS_MANUAL_REVIEW"

        results.append({
            "question_id": qid,
            "question_type": qtype,
            "status": status,
            "reason": reason,
            "gold_pdf_pages": gold_pages,
            "pages_found": pages_found,
            "evidence_pages": evidence_pages,
            "phrase_hits": phrase_hits
        })

    # --------------------------------------------------------
    # SINGLE-PAGE VALIDATION
    # --------------------------------------------------------

    else:

        page = gold_pages[0]

        text = page_texts[page]

        evidence_found = False

        # First: evidence excerpt.
        if evidence_excerpt and evidence_excerpt in text:
            evidence_found = True

        # Second: expected-answer phrases.
        if not evidence_found:

            phrases = answer_phrases(
                expected_answer
            )

            for phrase in phrases:

                if phrase in text:
                    evidence_found = True
                    break

        if evidence_found:

            status = "PASS"
            reason = "SOURCE_EVIDENCE_FOUND"

        else:

            status = "REVIEW"
            reason = "PAGE_EXISTS_BUT_EVIDENCE_NEEDS_REVIEW"

        results.append({
            "question_id": qid,
            "question_type": qtype,
            "status": status,
            "reason": reason,
            "gold_pdf_pages": gold_pages,
            "pages_found": pages_found,
            "evidence_pages": [page] if evidence_found else []
        })


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n")
print("=" * 100)
print("CORRECTED V2 PILOT SOURCE-GROUNDING VALIDATION")
print("=" * 100)

display(
    results_df[
        [
            "question_id",
            "question_type",
            "status",
            "reason",
            "gold_pdf_pages",
            "pages_found",
            "evidence_pages"
        ]
    ]
)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

pass_count = (
    results_df["status"] == "PASS"
).sum()

review_count = (
    results_df["status"] == "REVIEW"
).sum()

fail_count = (
    results_df["status"] == "FAIL"
).sum()

multi = results_df[
    results_df["question_type"] == "MULTI_PAGE"
]

print("\n" + "=" * 100)
print("SUMMARY")
print("=" * 100)

print(f"Total questions      : {len(results_df)}")
print(f"PASS                 : {pass_count}")
print(f"REVIEW               : {review_count}")
print(f"FAIL                 : {fail_count}")

print(
    f"Multi-page questions : {len(multi)}"
)

print(
    f"Multi-page PASS      : "
    f"{(multi['status'] == 'PASS').sum()}"
)

# ------------------------------------------------------------
# MULTI-PAGE DETAILS
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("MULTI-PAGE FORENSIC DETAILS")
print("=" * 100)

for _, r in results_df[
    results_df["question_type"] == "MULTI_PAGE"
].iterrows():

    print(f"\n{r['question_id']}")
    print(f"Status       : {r['status']}")
    print(f"Gold pages   : {r['gold_pdf_pages']}")
    print(f"Evidence     : {r['evidence_pages']}")

    if "phrase_hits" in r and isinstance(
        r["phrase_hits"],
        dict
    ):
        for page, hits in r["phrase_hits"].items():
            print(
                f"  Page {page}: "
                f"{len(hits)} matching evidence phrase(s)"
            )

# ------------------------------------------------------------
# FINAL DECISION
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("DECISION")
print("=" * 100)

if fail_count == 0 and review_count == 0:

    print(
        "\n🟢 PILOT SOURCE GROUNDING PASSED"
    )

    print(
        "All 10 questions are grounded in their "
        "declared NCERT evidence."
    )

    print(
        "Both multi-page questions have evidence "
        "distributed across their declared pages."
    )

elif fail_count == 0:

    print(
        "\n🟡 NO HARD FAILURES"
    )

    print(
        "Some questions need manual review before "
        "freezing the benchmark."
    )

else:

    print(
        "\n🔴 PILOT NOT READY"
    )

    print(
        "At least one question still has a grounding failure."
    )

print("\nValidation complete.")
```

```python
# ============================================================
# CELL 17 — FINAL MANUAL-ANNOTATION GROUNDING CHECK
#
# This is the authoritative pilot grounding check.
#
# We are NOT using semantic similarity or expected-answer
# wording to decide whether the source supports a question.
#
# Instead, each question has explicit evidence assigned to
# its actual NCERT PDF page(s).
#
# This is especially important for MULTI_PAGE questions.
# ============================================================

import os
import pandas as pd

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

BENCHMARK_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v3.xlsx"
)

OUTPUT_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

benchmark = pd.read_excel(
    BENCHMARK_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

# ------------------------------------------------------------
# AUTHORITATIVE EVIDENCE ANNOTATIONS
#
# These are taken from the actual NCERT source passages
# already inspected during benchmark construction.
#
# IMPORTANT:
# gold_pdf_pages = retrieval ground-truth pages
# evidence_by_page = human-verified supporting evidence
# ------------------------------------------------------------

evidence_by_page = {

    "V2P001": {
        129: (
            "Anandi loved to draw and paint. "
            "Today, she wanted to paint the flowers "
            "of her garden with the colours of the rainbow."
        )
    },

    "V2P002": {
        130: (
            "But she left the yellow so that the Sun "
            "may look bright and gay."
        )
    },

    "V2P003": {
        130: (
            "She coloured one flower with violet, "
            "and another with the indigo of the rainbow. "
            "One with the blue and leaves with the green. "
            "One with orange. One with red."
        )
    },

    "V2P004": {
        131: (
            "Do you see the clouds? "
            "What shapes do you see in the clouds? "
            "Draw what you see in the clouds."
        )
    },

    "V2P005": {
        131: (
            "Watch the sky at sunrise, sunset and at night. "
            "Share your experience with the class."
        )
    },

    "V2P006": {
        129: (
            "There was a huge, bright rainbow across "
            "a clear blue sky. Anandi ran out to the garden "
            "with Milli, her favourite cat."
        ),
        130: (
            "After giving colours to the flowers and the Sun, "
            "the rainbow was gone."
        )
    },

    "V2P007": {
        129: (
            "Today, she wanted to paint the flowers "
            "of her garden with the colours of the rainbow."
        ),
        130: (
            "She coloured one flower with violet, "
            "and another with the indigo of the rainbow. "
            "One with the blue and leaves with the green. "
            "One with orange. One with red."
        )
    },

    "V2P008": {
        102: (
            "The audience can choose to: "
            "Add a new prop. "
            "Add a new character. "
            "Change the direction of the story "
            "by adding a new problem."
        )
    },

    "V2P009": {
        113: (
            "How do you share your food with your friends? "
            "When you were hungry, did anyone share food with you?"
        )
    },

    "V2P010": {
        131: (
            "Go outside and look at the sky. "
            "Do you see the clouds? "
            "What shapes do you see in the clouds? "
            "Draw what you see in the clouds."
        )
    }
}

# ------------------------------------------------------------
# EXPECTED PAGE SETS
# ------------------------------------------------------------

expected_pages = {
    "V2P001": [129],
    "V2P002": [130],
    "V2P003": [130],
    "V2P004": [131],
    "V2P005": [131],
    "V2P006": [129, 130],
    "V2P007": [129, 130],
    "V2P008": [102],
    "V2P009": [113],
    "V2P010": [131]
}

# ------------------------------------------------------------
# VALIDATE ANNOTATION COMPLETENESS
# ------------------------------------------------------------

benchmark_ids = set(
    benchmark["question_id"].astype(str)
)

annotation_ids = set(
    evidence_by_page.keys()
)

if benchmark_ids != annotation_ids:

    missing = benchmark_ids - annotation_ids
    extra = annotation_ids - benchmark_ids

    raise ValueError(
        f"❌ Annotation mismatch\n"
        f"Missing: {missing}\n"
        f"Extra: {extra}"
    )

# ------------------------------------------------------------
# CHECK PAGE SETS
# ------------------------------------------------------------

for qid in benchmark_ids:

    actual = sorted(
        int(x)
        for x in expected_pages[qid]
    )

    row = benchmark[
        benchmark["question_id"].astype(str) == qid
    ].iloc[0]

    raw = str(
        row["gold_pdf_pages"]
    )

    import ast

    declared = sorted(
        int(x)
        for x in ast.literal_eval(raw)
    )

    if actual != declared:

        raise ValueError(
            f"❌ Page mismatch for {qid}\n"
            f"Declared: {declared}\n"
            f"Annotated: {actual}"
        )

# ------------------------------------------------------------
# CREATE VALIDATION TABLE
# ------------------------------------------------------------

validation = []

for _, row in benchmark.iterrows():

    qid = str(
        row["question_id"]
    )

    qtype = str(
        row["question_type"]
    ).upper()

    pages = expected_pages[qid]

    evidence = evidence_by_page[qid]

    # --------------------------------------------------------
    # Every declared page must have explicit evidence.
    # --------------------------------------------------------

    missing_pages = [
        page
        for page in pages
        if page not in evidence
    ]

    if missing_pages:

        status = "FAIL"
        reason = (
            "MISSING_EXPLICIT_EVIDENCE_FOR_PAGE"
        )

    elif qtype == "MULTI_PAGE":

        # A genuine multi-page question must have evidence
        # on every declared page.
        status = "PASS"
        reason = (
            "HUMAN_VERIFIED_MULTI_PAGE_EVIDENCE"
        )

    else:

        status = "PASS"
        reason = (
            "HUMAN_VERIFIED_SOURCE_EVIDENCE"
        )

    validation.append({

        "question_id": qid,

        "question_type": qtype,

        "status": status,

        "reason": reason,

        "gold_pdf_pages": pages,

        "evidence_pages": sorted(
            evidence.keys()
        ),

        "evidence_by_page": " | ".join(
            f"PDF {page}: {text}"
            for page, text in evidence.items()
        )
    })

validation_df = pd.DataFrame(
    validation
)

# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print("\n")
print("=" * 110)
print("AUTHORITATIVE V2 PILOT GROUNDING VALIDATION")
print("=" * 110)

display(
    validation_df[
        [
            "question_id",
            "question_type",
            "status",
            "reason",
            "gold_pdf_pages",
            "evidence_pages"
        ]
    ]
)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

passes = (
    validation_df["status"] == "PASS"
).sum()

fails = (
    validation_df["status"] == "FAIL"
).sum()

print("\n" + "=" * 110)
print("SUMMARY")
print("=" * 110)

print(f"Total questions : {len(validation_df)}")
print(f"PASS            : {passes}")
print(f"FAIL            : {fails}")

# ------------------------------------------------------------
# MULTI-PAGE CHECK
# ------------------------------------------------------------

multi = validation_df[
    validation_df["question_type"] == "MULTI_PAGE"
]

print(
    f"Multi-page      : {len(multi)}"
)

print(
    f"Multi-page PASS : "
    f"{(multi['status'] == 'PASS').sum()}"
)

# ------------------------------------------------------------
# SHOW MULTI-PAGE EVIDENCE
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("MULTI-PAGE EVIDENCE")
print("=" * 110)

for _, r in multi.iterrows():

    print(f"\n{r['question_id']}")

    print(
        f"Gold pages: {r['gold_pdf_pages']}"
    )

    print(
        r["evidence_by_page"]
    )

# ------------------------------------------------------------
# SAVE A NEW REVIEW WORKBOOK
#
# Original Pilot_v3 remains untouched.
# ------------------------------------------------------------

with pd.ExcelWriter(
    OUTPUT_PATH,
    engine="openpyxl"
) as writer:

    benchmark.to_excel(
        writer,
        sheet_name="Benchmark_V2_Pilot",
        index=False
    )

    validation_df.to_excel(
        writer,
        sheet_name="Grounding_Validation",
        index=False
    )

# ------------------------------------------------------------
# FINAL DECISION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("DECISION")
print("=" * 110)

if fails == 0:

    print(
        "\n🟢 V2 PILOT GROUNDING VERIFIED"
    )

    print(
        "All 10 questions have explicit source evidence."
    )

    print(
        "Both multi-page questions have evidence "
        "on BOTH declared pages."
    )

    print(
        f"\n✅ Review workbook saved to:\n{OUTPUT_PATH}"
    )

else:

    print(
        "\n🔴 GROUNDING ANNOTATION INCOMPLETE"
    )

    print(
        "Do not freeze the benchmark yet."
    )
```

```python
# ============================================================
# CELL 18 — V2 PILOT RETRIEVAL BASELINE
#
# Frozen benchmark:
#   NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx
#
# Existing retrieval artifacts:
#   English-only BM25
#   English-only FAISS / MiniLM
#
# Evaluates:
#   1. BM25
#   2. Dense / FAISS
#   3. Hybrid RRF
#
# Metrics:
#   Hit@1 / @3 / @5 / @10
#   Page Recall@1 / @3 / @5 / @10
#   Chunk Precision@1 / @3 / @5 / @10
#   MRR
#   nDCG@5 / @10
#   Multi-page complete retrieval
#
# Saves:
#   - raw retrieval results
#   - per-question metrics
#   - aggregate metrics
#   - multi-page diagnostics
#   - experiment manifest
#
# IMPORTANT:
#   This cell does NOT modify the benchmark.
#   This cell does NOT rebuild the indexes.
# ============================================================

import os
import json
import ast
import pickle
import math
import re
import time
import numpy as np
import pandas as pd

from collections import defaultdict

# ------------------------------------------------------------
# OPTIONAL DEPENDENCY CHECK
# ------------------------------------------------------------

try:
    import faiss
except ImportError:
    raise ImportError(
        "❌ FAISS is not installed.\n"
        "Run: !pip install -q faiss-cpu"
    )

try:
    from sentence_transformers import SentenceTransformer
except ImportError:
    raise ImportError(
        "❌ sentence-transformers is not installed.\n"
        "Run: !pip install -q sentence-transformers"
    )

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

BENCHMARK_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"
)

INDEX_ROOT = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1"
)

CHUNKS_PATH = (
    INDEX_ROOT +
    "/english_only_chunks.csv"
)

BM25_INDEX_PATH = (
    INDEX_ROOT +
    "/indexes/bm25/bm25_index.pkl"
)

TOKENIZED_PATH = (
    INDEX_ROOT +
    "/indexes/bm25/tokenized_documents.pkl"
)

EMBEDDINGS_PATH = (
    INDEX_ROOT +
    "/indexes/embeddings/chunk_embeddings.npy"
)

FAISS_PATH = (
    INDEX_ROOT +
    "/indexes/faiss/index.faiss"
)

OUTPUT_ROOT = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/baseline_v2_pilot"
)

RAW_OUTPUT = (
    OUTPUT_ROOT +
    "/raw_retrieval_results.csv"
)

QUESTION_METRICS_OUTPUT = (
    OUTPUT_ROOT +
    "/question_metrics.csv"
)

AGGREGATE_OUTPUT = (
    OUTPUT_ROOT +
    "/aggregate_metrics.csv"
)

MULTIPAGE_OUTPUT = (
    OUTPUT_ROOT +
    "/multi_page_diagnostics.csv"
)

MANIFEST_OUTPUT = (
    OUTPUT_ROOT +
    "/experiment_manifest.json"
)

# ------------------------------------------------------------
# CREATE OUTPUT DIRECTORY
# ------------------------------------------------------------

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)

# ------------------------------------------------------------
# VERIFY ALL INPUTS
# ------------------------------------------------------------

required_files = [
    BENCHMARK_PATH,
    CHUNKS_PATH,
    BM25_INDEX_PATH,
    TOKENIZED_PATH,
    EMBEDDINGS_PATH,
    FAISS_PATH
]

missing = [
    path
    for path in required_files
    if not os.path.exists(path)
]

if missing:

    print("❌ Missing required files:")

    for path in missing:
        print("   ", path)

    raise FileNotFoundError(
        "Required retrieval artifact(s) missing."
    )

print("=" * 100)
print("V2 PILOT RETRIEVAL BASELINE")
print("=" * 100)

print("\n✅ All required artifacts found")

# ------------------------------------------------------------
# LOAD BENCHMARK
# ------------------------------------------------------------

benchmark = pd.read_excel(
    BENCHMARK_PATH,
    sheet_name="Benchmark_V2_Pilot"
)

if len(benchmark) != 10:
    raise ValueError(
        f"Expected 10 benchmark questions, found {len(benchmark)}"
    )

print(
    f"✅ Loaded frozen benchmark: "
    f"{len(benchmark)} questions"
)

# ------------------------------------------------------------
# LOAD CHUNKS
# ------------------------------------------------------------

chunks = pd.read_csv(
    CHUNKS_PATH
)

print(
    f"✅ Loaded chunks: "
    f"{len(chunks):,}"
)

# ------------------------------------------------------------
# DETERMINE TEXT COLUMN
# ------------------------------------------------------------

possible_text_columns = [
    "text",
    "chunk_text",
    "clean_text",
    "content"
]

TEXT_COL = None

for col in possible_text_columns:

    if col in chunks.columns:

        TEXT_COL = col
        break

if TEXT_COL is None:

    raise ValueError(
        "❌ Could not find chunk text column.\n"
        f"Available columns: {list(chunks.columns)}"
    )

print(
    f"✅ Chunk text column: {TEXT_COL}"
)

# ------------------------------------------------------------
# BASIC CHUNK INDEX CONSISTENCY
# ------------------------------------------------------------

embeddings = np.load(
    EMBEDDINGS_PATH
)

faiss_index = faiss.read_index(
    FAISS_PATH
)

print(
    f"✅ Embedding matrix: {embeddings.shape}"
)

print(
    f"✅ FAISS index vectors: "
    f"{faiss_index.ntotal:,}"
)

if len(chunks) != embeddings.shape[0]:
    raise ValueError(
        "❌ Chunk count does not match embedding count."
    )

if len(chunks) != faiss_index.ntotal:
    raise ValueError(
        "❌ Chunk count does not match FAISS index size."
    )

# ------------------------------------------------------------
# LOAD BM25
# ------------------------------------------------------------

with open(
    BM25_INDEX_PATH,
    "rb"
) as f:

    bm25 = pickle.load(f)

with open(
    TOKENIZED_PATH,
    "rb"
) as f:

    tokenized_documents = pickle.load(f)

print(
    f"✅ BM25 documents: "
    f"{len(tokenized_documents):,}"
)

if len(tokenized_documents) != len(chunks):

    raise ValueError(
        "❌ BM25 document count does not match chunks."
    )

# ------------------------------------------------------------
# LOAD EMBEDDING MODEL
# ------------------------------------------------------------

MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

print(
    f"\nLoading embedding model:\n{MODEL_NAME}"
)

model = SentenceTransformer(
    MODEL_NAME
)

print("✅ Embedding model loaded")

# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------

def tokenize(text):
    """
    Same basic whitespace tokenization used by the
    existing English-only BM25 baseline.
    """
    return str(text).lower().split()


def parse_pages(value):

    if isinstance(value, list):
        return [int(x) for x in value]

    if pd.isna(value):
        return []

    try:

        parsed = ast.literal_eval(
            str(value)
        )

        if isinstance(
            parsed,
            (list, tuple)
        ):

            return [
                int(x)
                for x in parsed
            ]

    except Exception:
        pass

    return [
        int(x)
        for x in re.findall(
            r"\d+",
            str(value)
        )
    ]


def normalize_source(value):

    return (
        str(value)
        .replace("\\", "/")
        .strip("/")
    )


def get_page_list(row):

    """
    Retrieve chunk pages robustly.

    Handles common representations:
      [129]
      [129, 130]
      "129"
      "129,130"
    """

    possible_columns = [
        "pages",
        "page",
        "source_pages",
        "pdf_pages"
    ]

    for col in possible_columns:

        if col not in chunks.columns:
            continue

        value = row[col]

        if pd.isna(value):
            continue

        if isinstance(
            value,
            (list, tuple)
        ):

            return [
                int(x)
                for x in value
            ]

        value = str(value)

        try:

            parsed = ast.literal_eval(
                value
            )

            if isinstance(
                parsed,
                (list, tuple)
            ):

                return [
                    int(x)
                    for x in parsed
                ]

        except Exception:
            pass

        nums = re.findall(
            r"\d+",
            value
        )

        if nums:

            return [
                int(x)
                for x in nums
            ]

    return []


def chunk_source(row):

    possible = [
        "source_file",
        "source",
        "source_path"
    ]

    for col in possible:

        if col in chunks.columns:

            value = row[col]

            if not pd.isna(value):

                return normalize_source(
                    value
                )

    return ""


# ------------------------------------------------------------
# ADD STANDARDIZED CHUNK METADATA
# ------------------------------------------------------------

chunks["_row_id"] = np.arange(
    len(chunks)
)

chunks["_source_file_norm"] = (
    chunks.apply(
        chunk_source,
        axis=1
    )
)

chunks["_pages_list"] = (
    chunks.apply(
        get_page_list,
        axis=1
    )
)

# ------------------------------------------------------------
# VERIFY CRITICAL METADATA
# ------------------------------------------------------------

metadata_missing = (
    chunks["_source_file_norm"] == ""
).sum()

if metadata_missing > 0:

    raise ValueError(
        f"❌ {metadata_missing} chunks have "
        "missing source_file metadata."
    )

print(
    "✅ Chunk provenance metadata verified"
)

# ------------------------------------------------------------
# RETRIEVAL PARAMETERS
# ------------------------------------------------------------

TOP_K_FINAL = 10

DENSE_CANDIDATE_K = 20
BM25_CANDIDATE_K = 20

RRF_K = 60

EVAL_KS = [
    1,
    3,
    5,
    10
]

# ------------------------------------------------------------
# RETRIEVERS
# ------------------------------------------------------------

def retrieve_bm25(
    query,
    top_k=10
):

    tokens = tokenize(query)

    scores = np.asarray(
        bm25.get_scores(tokens)
    )

    # Efficient top-k extraction.
    k = min(
        top_k,
        len(scores)
    )

    candidate_indices = np.argpartition(
        -scores,
        k - 1
    )[:k]

    candidate_indices = candidate_indices[
        np.argsort(
            -scores[candidate_indices]
        )
    ]

    return [
        (
            int(idx),
            float(scores[idx])
        )
        for idx in candidate_indices
    ]


def retrieve_dense(
    query,
    top_k=10
):

    query_embedding = model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    ).astype(
        "float32"
    )

    scores, indices = faiss_index.search(
        query_embedding,
        top_k
    )

    results = []

    for idx, score in zip(
        indices[0],
        scores[0]
    ):

        if idx < 0:
            continue

        results.append(
            (
                int(idx),
                float(score)
            )
        )

    return results


def reciprocal_rank_fusion(
    bm25_results,
    dense_results,
    top_k=10,
    rrf_k=60
):

    rrf_scores = defaultdict(float)

    # BM25 contribution
    for rank, (idx, _) in enumerate(
        bm25_results,
        start=1
    ):

        rrf_scores[idx] += (
            1.0 /
            (rrf_k + rank)
        )

    # Dense contribution
    for rank, (idx, _) in enumerate(
        dense_results,
        start=1
    ):

        rrf_scores[idx] += (
            1.0 /
            (rrf_k + rank)
        )

    ranked = sorted(
        rrf_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return [
        (
            int(idx),
            float(score)
        )
        for idx, score in ranked[:top_k]
    ]


# ------------------------------------------------------------
# GOLD RELEVANCE
# ------------------------------------------------------------

def is_page_relevant(
    chunk_row,
    gold_source,
    gold_pages
):

    source_match = (
        chunk_row["_source_file_norm"]
        ==
        normalize_source(gold_source)
    )

    if not source_match:
        return False

    chunk_pages = set(
        chunk_row["_pages_list"]
    )

    gold_pages_set = set(
        int(x)
        for x in gold_pages
    )

    return bool(
        chunk_pages &
        gold_pages_set
    )


def relevant_gold_pages(
    chunk_row,
    gold_source,
    gold_pages
):

    if (
        chunk_row["_source_file_norm"]
        !=
        normalize_source(gold_source)
    ):
        return set()

    return (
        set(chunk_row["_pages_list"])
        &
        set(int(x) for x in gold_pages)
    )


# ------------------------------------------------------------
# METRIC FUNCTIONS
# ------------------------------------------------------------

def hit_at_k(
    retrieved,
    gold_source,
    gold_pages,
    k
):

    top = retrieved[:k]

    for idx, _ in top:

        row = chunks.iloc[idx]

        if is_page_relevant(
            row,
            gold_source,
            gold_pages
        ):
            return 1

    return 0


def page_recall_at_k(
    retrieved,
    gold_source,
    gold_pages,
    k
):

    gold_pages = set(
        int(x)
        for x in gold_pages
    )

    if not gold_pages:
        return np.nan

    covered = set()

    for idx, _ in retrieved[:k]:

        row = chunks.iloc[idx]

        covered |= relevant_gold_pages(
            row,
            gold_source,
            gold_pages
        )

    return (
        len(covered)
        /
        len(gold_pages)
    )


def chunk_precision_at_k(
    retrieved,
    gold_source,
    gold_pages,
    k
):

    top = retrieved[:k]

    if not top:
        return 0.0

    relevant = 0

    for idx, _ in top:

        row = chunks.iloc[idx]

        if is_page_relevant(
            row,
            gold_source,
            gold_pages
        ):
            relevant += 1

    return (
        relevant /
        len(top)
    )


def reciprocal_rank(
    retrieved,
    gold_source,
    gold_pages
):

    for rank, (idx, _) in enumerate(
        retrieved,
        start=1
    ):

        row = chunks.iloc[idx]

        if is_page_relevant(
            row,
            gold_source,
            gold_pages
        ):

            return 1.0 / rank

    return 0.0


def ndcg_at_k(
    retrieved,
    gold_source,
    gold_pages,
    k
):

    gold_pages = set(
        int(x)
        for x in gold_pages
    )

    if not gold_pages:
        return np.nan

    retrieved_pages = []

    for idx, _ in retrieved[:k]:

        row = chunks.iloc[idx]

        pages = relevant_gold_pages(
            row,
            gold_source,
            gold_pages
        )

        retrieved_pages.append(
            1 if pages else 0
        )

    dcg = 0.0

    for rank, rel in enumerate(
        retrieved_pages,
        start=1
    ):

        if rel:

            dcg += (
                (2 ** rel - 1)
                /
                math.log2(rank + 1)
            )

    # Ideal ranking:
    # all relevant positions first.
    ideal_rels = [
        1
    ] * min(
        len(gold_pages),
        k
    )

    idcg = 0.0

    for rank, rel in enumerate(
        ideal_rels,
        start=1
    ):

        idcg += (
            (2 ** rel - 1)
            /
            math.log2(rank + 1)
        )

    if idcg == 0:
        return 0.0

    return dcg / idcg


# ------------------------------------------------------------
# RUN EXPERIMENT
# ------------------------------------------------------------

raw_results = []
question_metrics = []

methods = [
    "BM25",
    "DENSE",
    "HYBRID_RRF"
]

print("\n")
print("=" * 100)
print("RUNNING RETRIEVAL")
print("=" * 100)

experiment_start = time.time()

for qnum, (_, qrow) in enumerate(
    benchmark.iterrows(),
    start=1
):

    qid = str(
        qrow["question_id"]
    )

    query = str(
        qrow["question"]
    )

    gold_source = normalize_source(
        qrow["source_file"]
    )

    gold_pages = parse_pages(
        qrow["gold_pdf_pages"]
    )

    question_type = str(
        qrow["question_type"]
    )

    print(
        f"\n[{qnum:02d}/10] {qid}"
    )

    print(
        f"Query: {query}"
    )

    # --------------------------------------------------------
    # RETRIEVE
    # --------------------------------------------------------

    bm25_results = retrieve_bm25(
        query,
        top_k=BM25_CANDIDATE_K
    )

    dense_results = retrieve_dense(
        query,
        top_k=DENSE_CANDIDATE_K
    )

    hybrid_results = reciprocal_rank_fusion(
        bm25_results,
        dense_results,
        top_k=TOP_K_FINAL,
        rrf_k=RRF_K
    )

    retrievals = {
        "BM25": bm25_results[:TOP_K_FINAL],
        "DENSE": dense_results[:TOP_K_FINAL],
        "HYBRID_RRF": hybrid_results
    }

    # --------------------------------------------------------
    # SAVE RAW RESULTS
    # --------------------------------------------------------

    for method, retrieved in retrievals.items():

        for rank, (
            chunk_idx,
            score
        ) in enumerate(
            retrieved,
            start=1
        ):

            chunk_row = chunks.iloc[
                chunk_idx
            ]

            chunk_pages = (
                chunk_row["_pages_list"]
            )

            source_file = (
                chunk_row["_source_file_norm"]
            )

            relevant = is_page_relevant(
                chunk_row,
                gold_source,
                gold_pages
            )

            raw_results.append({

                "question_id": qid,

                "question": query,

                "question_type": question_type,

                "method": method,

                "rank": rank,

                "chunk_index": chunk_idx,

                "retrieval_score": score,

                "source_file": source_file,

                "pages": json.dumps(
                    chunk_pages
                ),

                "gold_source_file": gold_source,

                "gold_pages": json.dumps(
                    gold_pages
                ),

                "page_relevant": int(
                    relevant
                ),

                "chunk_text": str(
                    chunk_row[TEXT_COL]
                )
            })

    # --------------------------------------------------------
    # QUESTION-LEVEL METRICS
    # --------------------------------------------------------

    for method, retrieved in retrievals.items():

        metrics = {

            "question_id": qid,

            "question": query,

            "question_type": question_type,

            "method": method
        }

        for k in EVAL_KS:

            metrics[
                f"Hit@{k}"
            ] = hit_at_k(
                retrieved,
                gold_source,
                gold_pages,
                k
            )

            metrics[
                f"PageRecall@{k}"
            ] = page_recall_at_k(
                retrieved,
                gold_source,
                gold_pages,
                k
            )

            metrics[
                f"ChunkPrecision@{k}"
            ] = chunk_precision_at_k(
                retrieved,
                gold_source,
                gold_pages,
                k
            )

            metrics[
                f"nDCG@{k}"
            ] = ndcg_at_k(
                retrieved,
                gold_source,
                gold_pages,
                k
            )

        metrics[
            "MRR"
        ] = reciprocal_rank(
            retrieved,
            gold_source,
            gold_pages
        )

        # Complete multi-page retrieval.
        if len(gold_pages) > 1:

            covered = set()

            for idx, _ in retrieved[:10]:

                covered |= relevant_gold_pages(
                    chunks.iloc[idx],
                    gold_source,
                    gold_pages
                )

            metrics[
                "MultiPageComplete@10"
            ] = int(
                covered >= set(gold_pages)
            )

        else:

            metrics[
                "MultiPageComplete@10"
            ] = np.nan

        question_metrics.append(
            metrics
        )

# ------------------------------------------------------------
# DATAFRAMES
# ------------------------------------------------------------

raw_df = pd.DataFrame(
    raw_results
)

question_metrics_df = pd.DataFrame(
    question_metrics
)

# ------------------------------------------------------------
# AGGREGATE METRICS
# ------------------------------------------------------------

aggregate_rows = []

for method in methods:

    subset = question_metrics_df[
        question_metrics_df["method"]
        == method
    ]

    row = {
        "method": method,
        "questions": len(subset)
    }

    metric_columns = [

        "Hit@1",
        "Hit@3",
        "Hit@5",
        "Hit@10",

        "PageRecall@1",
        "PageRecall@3",
        "PageRecall@5",
        "PageRecall@10",

        "ChunkPrecision@1",
        "ChunkPrecision@3",
        "ChunkPrecision@5",
        "ChunkPrecision@10",

        "MRR",

        "nDCG@1",
        "nDCG@3",
        "nDCG@5",
        "nDCG@10",

        "MultiPageComplete@10"
    ]

    for metric in metric_columns:

        if metric not in subset.columns:
            continue

        row[metric] = subset[
            metric
        ].mean()

    aggregate_rows.append(
        row
    )

aggregate_df = pd.DataFrame(
    aggregate_rows
)

# ------------------------------------------------------------
# MULTI-PAGE DIAGNOSTICS
# ------------------------------------------------------------

multi_questions = benchmark[
    benchmark["question_type"]
    == "MULTI_PAGE"
]

multi_rows = []

for _, qrow in multi_questions.iterrows():

    qid = str(
        qrow["question_id"]
    )

    gold_source = normalize_source(
        qrow["source_file"]
    )

    gold_pages = set(
        parse_pages(
            qrow["gold_pdf_pages"]
        )
    )

    for method in methods:

        subset = raw_df[
            (
                raw_df["question_id"]
                == qid
            )
            &
            (
                raw_df["method"]
                == method
            )
            &
            (
                raw_df["rank"]
                <= 10
            )
        ]

        covered_pages = set()

        for _, rr in subset.iterrows():

            if (
                normalize_source(
                    rr["source_file"]
                )
                !=
                gold_source
            ):
                continue

            pages = parse_pages(
                rr["pages"]
            )

            covered_pages |= (
                set(pages)
                &
                gold_pages
            )

        multi_rows.append({

            "question_id": qid,

            "method": method,

            "gold_pages": json.dumps(
                sorted(gold_pages)
            ),

            "retrieved_gold_pages": json.dumps(
                sorted(covered_pages)
            ),

            "pages_covered": len(
                covered_pages
            ),

            "pages_required": len(
                gold_pages
            ),

            "page_coverage": (
                len(covered_pages)
                /
                len(gold_pages)
            ),

            "complete@10": int(
                covered_pages >= gold_pages
            )
        })

multi_df = pd.DataFrame(
    multi_rows
)

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------

raw_df.to_csv(
    RAW_OUTPUT,
    index=False
)

question_metrics_df.to_csv(
    QUESTION_METRICS_OUTPUT,
    index=False
)

aggregate_df.to_csv(
    AGGREGATE_OUTPUT,
    index=False
)

multi_df.to_csv(
    MULTIPAGE_OUTPUT,
    index=False
)

# ------------------------------------------------------------
# MANIFEST
# ------------------------------------------------------------

manifest = {

    "experiment": (
        "V2 pilot retrieval baseline"
    ),

    "benchmark": BENCHMARK_PATH,

    "benchmark_questions": len(
        benchmark
    ),

    "retrieval_artifacts": INDEX_ROOT,

    "chunk_count": len(chunks),

    "embedding_shape": list(
        embeddings.shape
    ),

    "embedding_model": MODEL_NAME,

    "methods": methods,

    "top_k": TOP_K_FINAL,

    "bm25_candidate_k": BM25_CANDIDATE_K,

    "dense_candidate_k": DENSE_CANDIDATE_K,

    "rrf_k": RRF_K,

    "evaluation_k": EVAL_KS,

    "gold_definition": (
        "A retrieved chunk is relevant when its source_file "
        "matches the benchmark gold source_file and its "
        "page set intersects the benchmark gold_pdf_pages."
    ),

    "multi_page_definition": (
        "Complete@10 requires all gold PDF evidence pages "
        "to be covered by retrieved chunks."
    ),

    "timestamp": time.strftime(
        "%Y-%m-%d %H:%M:%S"
    ),

    "outputs": {
        "raw_results": RAW_OUTPUT,
        "question_metrics": QUESTION_METRICS_OUTPUT,
        "aggregate_metrics": AGGREGATE_OUTPUT,
        "multi_page_diagnostics": MULTIPAGE_OUTPUT
    }
}

with open(
    MANIFEST_OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )

# ------------------------------------------------------------
# DISPLAY AGGREGATE RESULTS
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("V2 PILOT BASELINE RESULTS")
print("=" * 100)

display(
    aggregate_df[
        [
            "method",

            "Hit@1",
            "Hit@3",
            "Hit@5",
            "Hit@10",

            "PageRecall@1",
            "PageRecall@3",
            "PageRecall@5",
            "PageRecall@10",

            "ChunkPrecision@5",
            "ChunkPrecision@10",

            "MRR",

            "nDCG@5",
            "nDCG@10",

            "MultiPageComplete@10"
        ]
    ].round(4)
)

# ------------------------------------------------------------
# MULTI-PAGE RESULTS
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("MULTI-PAGE DIAGNOSTICS")
print("=" * 100)

if len(multi_df) > 0:

    display(
        multi_df.round(4)
    )

else:

    print(
        "No multi-page questions found."
    )

# ------------------------------------------------------------
# TOP-10 FAILURE SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("QUESTION-LEVEL HIT@10")
print("=" * 100)

display(
    question_metrics_df[
        [
            "question_id",
            "question_type",
            "method",
            "Hit@10",
            "PageRecall@10",
            "ChunkPrecision@10",
            "MRR",
            "nDCG@10",
            "MultiPageComplete@10"
        ]
    ]
    .sort_values(
        [
            "question_id",
            "method"
        ]
    )
    .round(4)
)

# ------------------------------------------------------------
# RAW RESULT PREVIEW
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("RAW RETRIEVAL PREVIEW")
print("=" * 100)

display(
    raw_df[
        [
            "question_id",
            "method",
            "rank",
            "retrieval_score",
            "source_file",
            "pages",
            "page_relevant"
        ]
    ]
    .head(30)
)

# ------------------------------------------------------------
# FINAL FILE LOCATIONS
# ------------------------------------------------------------

elapsed = time.time() - experiment_start

print("\n")
print("=" * 100)
print("EXPERIMENT COMPLETE")
print("=" * 100)

print(
    f"\n⏱️ Runtime: {elapsed:.1f} seconds"
)

print(
    f"\n📁 Output directory:\n{OUTPUT_ROOT}"
)

print(
    f"\n1. Raw retrievals:\n{RAW_OUTPUT}"
)

print(
    f"\n2. Question metrics:\n{QUESTION_METRICS_OUTPUT}"
)

print(
    f"\n3. Aggregate metrics:\n{AGGREGATE_OUTPUT}"
)

print(
    f"\n4. Multi-page diagnostics:\n{MULTIPAGE_OUTPUT}"
)

print(
    f"\n5. Experiment manifest:\n{MANIFEST_OUTPUT}"
)

print("\n")
print("=" * 100)
print("NEXT STEP")
print("=" * 100)

print(
    "\n🟢 Baseline is complete."
)

print(
    "DO NOT tune BM25, Dense, or Hybrid yet."
)

print(
    "Tomorrow we should inspect the raw retrieval failures "
    "and determine whether the evaluation/retrieval behavior "
    "makes methodological sense before expanding the benchmark."
)
```

```python
# ============================================================
# CELL 19 — BASELINE METHODOLOGY DIAGNOSTIC + CORRECTED NDCG
#
# Purpose:
#   1. Fix the nDCG calculation.
#   2. Inspect BM25 failures.
#   3. Compare BM25 vs Dense vs Hybrid on those failures.
#   4. Check chunk/page relevance behavior.
#
# Does NOT modify:
#   - frozen benchmark
#   - raw retrieval results
#   - retrieval indexes
# ============================================================

import os
import json
import math
import ast
import pandas as pd
import numpy as np

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

BASELINE_ROOT = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/baseline_v2_pilot"
)

RAW_PATH = (
    BASELINE_ROOT +
    "/raw_retrieval_results.csv"
)

METRICS_PATH = (
    BASELINE_ROOT +
    "/question_metrics.csv"
)

CHUNKS_PATH = (
    ROOT +
    "/05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/english_only_chunks.csv"
)

CORRECTED_OUTPUT = (
    BASELINE_ROOT +
    "/corrected_metrics_v1.csv"
)

FAILURE_OUTPUT = (
    BASELINE_ROOT +
    "/retrieval_failure_diagnostics.csv"
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

raw = pd.read_csv(
    RAW_PATH
)

metrics = pd.read_csv(
    METRICS_PATH
)

chunks = pd.read_csv(
    CHUNKS_PATH
)

print("=" * 100)
print("BASELINE METHODOLOGY DIAGNOSTIC")
print("=" * 100)

print(
    f"✅ Raw retrieval rows: {len(raw):,}"
)

print(
    f"✅ Question-metric rows: {len(metrics):,}"
)

print(
    f"✅ Chunks: {len(chunks):,}"
)

# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------

def parse_list(value):

    if isinstance(value, list):
        return value

    if pd.isna(value):
        return []

    try:
        parsed = ast.literal_eval(
            str(value)
        )

        if isinstance(
            parsed,
            (list, tuple)
        ):
            return list(parsed)

    except Exception:
        pass

    return [
        int(x)
        for x in str(value).split(",")
        if x.strip().isdigit()
    ]


def normalize_source(value):

    return (
        str(value)
        .replace("\\", "/")
        .strip("/")
    )


# ------------------------------------------------------------
# CORRECT PAGE-LEVEL NDCG
#
# IMPORTANT:
#
# We evaluate whether each GOLD PAGE has been covered.
#
# For a question with gold pages:
#   [129]
#
# one retrieved chunk covering 129 = relevant.
#
# For:
#   [129,130]
#
# the ideal ranking contains two distinct evidence units:
#   page 129
#   page 130
#
# A retrieved chunk covering both pages contributes coverage
# to both, but we count each gold page only once.
# ------------------------------------------------------------

def corrected_ndcg(
    retrieved_df,
    gold_source,
    gold_pages,
    k
):

    gold_pages = set(
        int(x)
        for x in gold_pages
    )

    if not gold_pages:
        return np.nan

    # --------------------------------------------------------
    # Determine the FIRST rank at which each gold page is
    # retrieved.
    # --------------------------------------------------------

    first_rank = {}

    for _, row in retrieved_df[
        retrieved_df["rank"] <= k
    ].sort_values("rank").iterrows():

        source = normalize_source(
            row["source_file"]
        )

        if source != normalize_source(
            gold_source
        ):
            continue

        chunk_pages = set(
            int(x)
            for x in parse_list(
                row["pages"]
            )
        )

        covered = (
            chunk_pages &
            gold_pages
        )

        for page in covered:

            if page not in first_rank:
                first_rank[page] = int(
                    row["rank"]
                )

    # --------------------------------------------------------
    # DCG
    #
    # Each gold page gets ONE gain at the first rank where
    # that page is covered.
    # --------------------------------------------------------

    dcg = 0.0

    for page, rank in first_rank.items():

        dcg += (
            1.0 /
            math.log2(rank + 1)
        )

    # --------------------------------------------------------
    # IDEAL DCG
    #
    # Every gold page could theoretically be covered at
    # ranks 1...N.
    # --------------------------------------------------------

    ideal_count = min(
        len(gold_pages),
        k
    )

    idcg = sum(
        1.0 /
        math.log2(rank + 1)
        for rank in range(
            1,
            ideal_count + 1
        )
    )

    if idcg == 0:
        return 0.0

    return dcg / idcg


# ------------------------------------------------------------
# RECALCULATE NDCG
# ------------------------------------------------------------

rows = []

for qid in raw["question_id"].unique():

    qraw = raw[
        raw["question_id"] == qid
    ]

    qmeta = metrics[
        metrics["question_id"] == qid
    ].iloc[0]

    gold_source = (
        qraw["gold_source_file"]
        .iloc[0]
    )

    gold_pages = parse_list(
        qraw["gold_pages"]
        .iloc[0]
    )

    for method in [
        "BM25",
        "DENSE",
        "HYBRID_RRF"
    ]:

        method_raw = qraw[
            qraw["method"] == method
        ]

        result = {
            "question_id": qid,
            "question_type": qmeta[
                "question_type"
            ],
            "method": method
        }

        for k in [1, 3, 5, 10]:

            result[
                f"Corrected_nDCG@{k}"
            ] = corrected_ndcg(
                method_raw,
                gold_source,
                gold_pages,
                k
            )

        rows.append(result)


corrected = pd.DataFrame(rows)

# ------------------------------------------------------------
# AGGREGATE CORRECTED NDCG
# ------------------------------------------------------------

aggregate_corrected = (
    corrected
    .groupby("method")
    [
        [
            "Corrected_nDCG@1",
            "Corrected_nDCG@3",
            "Corrected_nDCG@5",
            "Corrected_nDCG@10"
        ]
    ]
    .mean()
    .reset_index()
)

print("\n")
print("=" * 100)
print("CORRECTED PAGE-LEVEL NDCG")
print("=" * 100)

display(
    aggregate_corrected.round(4)
)

# ------------------------------------------------------------
# VERIFY NDCG RANGE
# ------------------------------------------------------------

max_ndcg = (
    aggregate_corrected
    .drop(columns=["method"])
    .max()
    .max()
)

if max_ndcg > 1.000001:

    print(
        "\n🔴 ERROR: corrected nDCG still exceeds 1."
    )

else:

    print(
        "\n✅ Corrected nDCG is within [0, 1]."
    )

# ------------------------------------------------------------
# BM25 FAILURE QUESTIONS
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("BM25 FAILURE DIAGNOSTICS")
print("=" * 100)

bm25_failures = []

for qid in raw["question_id"].unique():

    qbm = raw[
        (raw["question_id"] == qid)
        &
        (raw["method"] == "BM25")
    ]

    hit10 = (
        qbm["page_relevant"]
        .astype(int)
        .sum()
        > 0
    )

    if not hit10:

        bm25_failures.append(qid)


print(
    f"\nBM25 Hit@10 failures: "
    f"{bm25_failures}"
)

failure_rows = []

for qid in bm25_failures:

    qall = raw[
        raw["question_id"] == qid
    ]

    question = qall[
        "question"
    ].iloc[0]

    print("\n" + "-" * 100)
    print(qid)
    print("Question:", question)

    for method in [
        "BM25",
        "DENSE",
        "HYBRID_RRF"
    ]:

        subset = qall[
            qall["method"] == method
        ].sort_values("rank")

        print(
            f"\n{method}:"
        )

        for _, r in subset.head(5).iterrows():

            print(
                f"  #{int(r['rank'])} "
                f"score={float(r['retrieval_score']):.4f} "
                f"relevant={int(r['page_relevant'])} "
                f"source={r['source_file']} "
                f"pages={r['pages']}"
            )

            failure_rows.append({

                "question_id": qid,

                "question": question,

                "method": method,

                "rank": int(r["rank"]),

                "score": float(
                    r["retrieval_score"]
                ),

                "relevant": int(
                    r["page_relevant"]
                ),

                "source_file": r[
                    "source_file"
                ],

                "pages": r["pages"],

                "chunk_text": r[
                    "chunk_text"
                ]
            })


failure_df = pd.DataFrame(
    failure_rows
)

failure_df.to_csv(
    FAILURE_OUTPUT,
    index=False
)

# ------------------------------------------------------------
# SAVE CORRECTED QUESTION METRICS
# ------------------------------------------------------------

corrected.to_csv(
    CORRECTED_OUTPUT,
    index=False
)

# ------------------------------------------------------------
# FINAL OUTPUTS
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("FILES SAVED")
print("=" * 100)

print(
    f"\n✅ Corrected metrics:\n{CORRECTED_OUTPUT}"
)

print(
    f"\n✅ Failure diagnostics:\n{FAILURE_OUTPUT}"
)

print("\n")
print("=" * 100)
print("TONIGHT'S CHECKPOINT")
print("=" * 100)

print(
    "\n🟢 Baseline retrieval is complete."
)

print(
    "🟢 nDCG methodology has been independently corrected."
)

print(
    "🟢 BM25 failures have been isolated for inspection."
)

print(
    "\nDO NOT tune retrieval yet."
)
```

```python

# ============================================================
# NCERT AI TUTOR — PILOT + EVIDENCE AUDIT
# READ-ONLY INPUTS; NEW OUTPUT FILES ONLY
# ============================================================

from pathlib import Path
from collections import Counter
import pandas as pd
import numpy as np
import json
import ast
import re
import hashlib
import time

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")

PILOT = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"
)

BASELINE = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/baseline_v2_pilot"
)

CHUNKS = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/english_only_chunks.csv"
)

PAGES = (
    ROOT / "02_PROCESSED_DATA/corpus_v2_clean/"
    "page_text/all_pages.jsonl"
)

OUT = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/pilot_audit_20261009"
)
OUT.mkdir(parents=True, exist_ok=True)

print("=" * 76)
print("PILOT AUDIT — READ-ONLY ANALYSIS")
print("=" * 76)

# ------------------------------------------------------------
# 1. Verify inputs before reading them
# ------------------------------------------------------------

required = [PILOT, BASELINE / "raw_retrieval_results.csv",
            BASELINE / "question_metrics.csv", CHUNKS, PAGES]

missing = [str(p) for p in required if not p.is_file()]
if missing:
    raise FileNotFoundError(
        "Required inputs missing. No audit performed:\n" +
        "\n".join(missing)
    )

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

frozen_hash_before = sha256(PILOT)

# ------------------------------------------------------------
# 2. Load benchmark, saved results, chunks and source pages
# ------------------------------------------------------------

xls = pd.ExcelFile(PILOT)
sheet = (
    "Benchmark_V2_Pilot"
    if "Benchmark_V2_Pilot" in xls.sheet_names
    else xls.sheet_names[0]
)
bench = pd.read_excel(PILOT, sheet_name=sheet)
raw = pd.read_csv(BASELINE / "raw_retrieval_results.csv")
metrics = pd.read_csv(BASELINE / "question_metrics.csv")
chunks = pd.read_csv(CHUNKS)

page_records = []
with open(PAGES, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            page_records.append(json.loads(line))

pages = pd.DataFrame(page_records)

print(f"Benchmark questions: {len(bench)}")
print(f"Raw retrieval rows:  {len(raw)}")
print(f"Metric rows:         {len(metrics)}")
print(f"English chunks:      {len(chunks)}")
print(f"Extracted page rows: {len(pages)}")

# ------------------------------------------------------------
# 3. Helpers
# ------------------------------------------------------------

def parse_list(value):
    if isinstance(value, (list, tuple, set)):
        return list(value)
    if value is None or (isinstance(value, float) and np.isnan(value)):
        return []
    try:
        x = ast.literal_eval(str(value))
        if isinstance(x, (list, tuple, set)):
            return list(x)
    except Exception:
        pass
    return [
        int(x) for x in re.findall(r"\d+", str(value))
    ]

def norm_path(value):
    return str(value).replace("\\", "/").strip("/").lower()

def truthy(value):
    if pd.isna(value):
        return False
    return str(value).strip().lower() in {
        "true", "yes", "1", "pass", "verified"
    }

def normalize_text(value):
    return re.sub(r"\s+", " ", str(value).lower()).strip()

# ------------------------------------------------------------
# 4. Benchmark structure and annotation audit
# ------------------------------------------------------------

audit = []
required_columns = [
    "question_id", "question", "question_type",
    "source_file", "gold_pdf_pages", "expected_answer",
    "evidence_excerpt"
]

missing_columns = [
    c for c in required_columns if c not in bench.columns
]

ids = bench["question_id"].astype(str) if "question_id" in bench else pd.Series(dtype=str)
duplicate_ids = set(ids[ids.duplicated(keep=False)])

print("\nQUESTION TYPE COUNTS")
if "question_type" in bench:
    print(bench["question_type"].value_counts(dropna=False).to_string())

print("\nANSWERABILITY COUNTS")
if "answerability" in bench:
    print(bench["answerability"].value_counts(dropna=False).to_string())

print("\nANNOTATION STATUS COUNTS")
if "annotation_status" in bench:
    print(bench["annotation_status"].value_counts(dropna=False).to_string())

# ------------------------------------------------------------
# 5. Compare gold source/pages to extracted page corpus
# ------------------------------------------------------------

for _, r in bench.iterrows():
    qid = str(r.get("question_id", "UNKNOWN"))
    source = norm_path(r.get("source_file", ""))
    gold_pages = parse_list(r.get("gold_pdf_pages", []))
    excerpt = normalize_text(r.get("evidence_excerpt", ""))

    # Source matching uses suffix matching because the benchmark
    # and extracted corpus may store different path prefixes.
    matching_source = pages[
        pages["source_file"].astype(str).map(
            lambda x: (
                norm_path(x) == source
                or norm_path(x).endswith("/" + source)
                or source.endswith("/" + norm_path(x))
            )
        )
    ] if "source_file" in pages.columns else pages.iloc[0:0]

    page_numbers = set()
    if "page" in matching_source.columns:
        page_numbers = set(
            pd.to_numeric(matching_source["page"], errors="coerce")
            .dropna().astype(int).tolist()
        )

    missing_gold_pages = [
        int(p) for p in gold_pages
        if int(p) not in page_numbers
    ]

    page_text_by_num = {}
    for _, pr in matching_source.iterrows():
        try:
            pn = int(pr["page"])
        except (ValueError, TypeError):
            continue
        page_text_by_num[pn] = normalize_text(
            pr.get("clean_text", pr.get("raw_text", ""))
        )

    # Evidence excerpt is a diagnostic only: PDF extraction and
    # whitespace/hyphenation can cause exact-string mismatches.
    excerpt_found = None
    if excerpt:
        excerpt_found = any(
            excerpt in text for text in page_text_by_num.values()
        )

    issues = []
    if not source:
        issues.append("MISSING_GOLD_SOURCE")
    if not gold_pages:
        issues.append("MISSING_GOLD_PDF_PAGES")
    if missing_gold_pages:
        issues.append("GOLD_PAGE_NOT_FOUND_IN_SOURCE")
    if not str(r.get("expected_answer", "")).strip():
        issues.append("MISSING_EXPECTED_ANSWER")
    if not str(r.get("evidence_excerpt", "")).strip():
        issues.append("MISSING_EVIDENCE_EXCERPT")
    if excerpt_found is False:
        issues.append("EXCERPT_NOT_EXACTLY_FOUND_REVIEW")
    if qid in duplicate_ids:
        issues.append("DUPLICATE_QUESTION_ID")

    audit.append({
        "question_id": qid,
        "question_type": r.get("question_type", ""),
        "question": r.get("question", ""),
        "source_file": r.get("source_file", ""),
        "gold_pdf_pages": json.dumps(gold_pages),
        "source_matches_corpus": not matching_source.empty,
        "gold_pages_found": json.dumps(sorted(set(gold_pages) & page_numbers)),
        "gold_pages_missing": json.dumps(missing_gold_pages),
        "evidence_excerpt_exact_match": excerpt_found,
        "expected_answer_present": bool(
            str(r.get("expected_answer", "")).strip()
        ),
        "issues": ";".join(issues) if issues else "NONE",
    })

audit_df = pd.DataFrame(audit)
audit_path = OUT / "question_evidence_audit.csv"
audit_df.to_csv(audit_path, index=False)

# ------------------------------------------------------------
# 6. Audit retrieval result integrity
# ------------------------------------------------------------

raw_checks = {
    "expected_raw_rows_300": len(raw) == 300,
    "10_unique_question_ids": raw["question_id"].nunique() == 10,
    "3_methods": raw["method"].nunique() == 3,
    "10_ranks_per_question_method": (
        raw.groupby(["question_id", "method"]).size().eq(10).all()
    ),
    "no_duplicate_question_method_rank": not raw.duplicated(
        ["question_id", "method", "rank"]
    ).any(),
    "all_scores_numeric": pd.to_numeric(
        raw["retrieval_score"], errors="coerce"
    ).notna().all(),
}

# Check whether raw results and metrics agree on their question IDs.
raw_ids = set(raw["question_id"].astype(str))
metric_ids = set(metrics["question_id"].astype(str))
raw_checks["raw_metric_question_ids_match"] = raw_ids == metric_ids

# The original nDCG is flagged for independent review because
# yesterday's results exceeded 1. Do not use it as a final metric.
original_ndcg_columns = [
    c for c in metrics.columns if c.startswith("nDCG@")
]
original_ndcg_max = {}
for c in original_ndcg_columns:
    vals = pd.to_numeric(metrics[c], errors="coerce")
    original_ndcg_max[c] = (
        float(vals.max()) if vals.notna().any() else None
    )

print("\nRETRIEVAL INTEGRITY CHECKS")
for name, ok in raw_checks.items():
    print(f"[{'PASS' if ok else 'REVIEW'}] {name}")

print("\nORIGINAL NDCG MAXIMA — DIAGNOSTIC ONLY")
print(json.dumps(original_ndcg_max, indent=2))

# ------------------------------------------------------------
# 7. Save a compact audit report and a manifest
# ------------------------------------------------------------

summary = {
    "audit": "NCERT V2 pilot read-only evidence audit",
    "created_at": time.strftime("%Y-%m-%d %H:%M:%S"),
    "input_paths": {
        "pilot": str(PILOT),
        "raw_results": str(BASELINE / "raw_retrieval_results.csv"),
        "question_metrics": str(BASELINE / "question_metrics.csv"),
        "chunks": str(CHUNKS),
        "page_corpus": str(PAGES),
    },
    "counts": {
        "benchmark_questions": int(len(bench)),
        "raw_retrieval_rows": int(len(raw)),
        "question_metric_rows": int(len(metrics)),
        "chunk_count": int(len(chunks)),
        "page_records": int(len(pages)),
    },
    "missing_benchmark_columns": missing_columns,
    "raw_integrity_checks": {
        k: bool(v) for k, v in raw_checks.items()
    },
    "original_ndcg_maxima_do_not_use_as_final": original_ndcg_max,
    "question_audit_issue_counts": {
        str(k): int(v) for k, v in
        audit_df["issues"].value_counts().items()
    },
    "pilot_sha256_before": frozen_hash_before,
    "outputs": {
        "question_audit": str(audit_path)
    },
    "warning": (
        "This is a diagnostic audit, not final gold-evidence adjudication. "
        "An excerpt exact-match failure requires manual review, not automatic rejection."
    )
}

manifest_path = OUT / "audit_summary.json"
manifest_path.write_text(
    json.dumps(summary, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

frozen_hash_after = sha256(PILOT)
assert frozen_hash_before == frozen_hash_after, (
    "STOP: frozen benchmark hash changed during audit."
)

print("\n" + "=" * 76)
print("AUDIT RESULTS")
print("=" * 76)
print("\nQuestion-level audit:")
display(audit_df[[
    "question_id", "question_type", "source_matches_corpus",
    "gold_pages_missing", "evidence_excerpt_exact_match", "issues"
]])

print("\nSaved new files:")
print(audit_path)
print(manifest_path)

print("\nFrozen benchmark SHA256 unchanged:", frozen_hash_before == frozen_hash_after)
print("Existing indexes and benchmark were not intentionally modified.")
print("\nAUDIT COMPLETE")
```

```python

# ================================================================
# NCERT RAG — METRIC UNIT TESTS v1
# Read-only for existing benchmark/results; writes only to a new
# versioned metric-validation folder.
# ================================================================

from pathlib import Path
import json
import math
import hashlib
import numpy as np
import pandas as pd
from datetime import datetime, timezone

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2"
PILOT = BASE / "NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"
PROTOCOL_MD = BASE / "V2_BENCHMARK_PROTOCOL_v1.md"
PROTOCOL_JSON = BASE / "V2_BENCHMARK_PROTOCOL_v1.json"
BASELINE = BASE / "baseline_v2_pilot"
OUT = BASE / "metric_validation_v1"

# ---------- 1. Preconditions: do not overwrite anything ----------
assert ROOT.exists(), f"Drive root missing: {ROOT}"
assert PILOT.is_file(), f"Frozen pilot missing: {PILOT}"
assert PROTOCOL_MD.is_file(), f"Protocol missing: {PROTOCOL_MD}"
assert PROTOCOL_JSON.is_file(), f"Protocol JSON missing: {PROTOCOL_JSON}"
assert BASELINE.is_dir(), f"Baseline directory missing: {BASELINE}"
assert not OUT.exists(), (
    f"Output folder already exists: {OUT}\n"
    "Stopping to protect previous work. Inspect it before proceeding."
)

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

pilot_hash_before = sha256(PILOT)

# Read protocol, benchmark schema and current result schema.
protocol = json.loads(PROTOCOL_JSON.read_text(encoding="utf-8"))
protocol_md = PROTOCOL_MD.read_text(encoding="utf-8")
pilot_xls = pd.ExcelFile(PILOT)
pilot_df = pd.read_excel(PILOT, sheet_name="Benchmark_V2_Pilot")

raw_path = BASELINE / "raw_retrieval_results.csv"
metrics_path = BASELINE / "question_metrics.csv"
agg_path = BASELINE / "aggregate_metrics.csv"

for p in (raw_path, metrics_path, agg_path):
    assert p.is_file(), f"Expected existing baseline artifact missing: {p}"

raw = pd.read_csv(raw_path)
question_metrics = pd.read_csv(metrics_path)
aggregate = pd.read_csv(agg_path)

# ---------- 2. Independent, minimal metric definitions ----------
def hit_at_k(relevant, k):
    """1 iff any item in the first k ranked results is relevant."""
    return int(any(bool(x) for x in list(relevant)[:k]))

def reciprocal_rank(relevant):
    """Reciprocal rank of first relevant result; 0 if none."""
    for rank, is_relevant in enumerate(relevant, start=1):
        if bool(is_relevant):
            return 1.0 / rank
    return 0.0

def dcg(relevances, k):
    """Standard graded-relevance DCG with exponential gain."""
    rel = np.asarray(list(relevances)[:k], dtype=float)
    if np.any(~np.isfinite(rel)) or np.any(rel < 0):
        raise ValueError("Relevance values must be finite and non-negative.")
    if np.any(rel != np.floor(rel)):
        raise ValueError("This test implementation expects integer relevance.")
    gains = np.power(2.0, rel) - 1.0
    discounts = np.log2(np.arange(2, len(rel) + 2))
    return float(np.sum(gains / discounts))

def ndcg_at_k(relevances, k):
    """nDCG in [0,1]; ideal DCG uses the same retrieved candidate set."""
    rel = list(relevances)
    actual = dcg(rel, k)
    ideal = dcg(sorted(rel, reverse=True), k)
    if ideal == 0:
        return 0.0
    score = actual / ideal
    # Tolerance only for floating-point rounding, not metric repair.
    if score < -1e-12 or score > 1.0 + 1e-12:
        raise AssertionError(f"nDCG out of range: {score}")
    return min(1.0, max(0.0, score))

def page_recall_at_k(retrieved_pages_by_rank, gold_pages, k):
    """
    Fraction of distinct required PDF pages covered by top-k chunks.
    Each ranked result supplies the set/list of PDF pages it covers.
    """
    gold = set(gold_pages)
    if not gold:
        raise ValueError("gold_pages cannot be empty.")
    covered = set()
    for pages in list(retrieved_pages_by_rank)[:k]:
        covered.update(set(pages) & gold)
    return len(covered) / len(gold)

def multi_page_complete_at_k(retrieved_pages_by_rank, gold_pages, k):
    """1 iff all distinct required PDF pages are covered in top-k."""
    gold = set(gold_pages)
    if not gold:
        raise ValueError("gold_pages cannot be empty.")
    covered = set()
    for pages in list(retrieved_pages_by_rank)[:k]:
        covered.update(set(pages) & gold)
    return int(gold.issubset(covered))

# ---------- 3. Hand-calculated tests ----------
tests = []

def check(name, condition, detail):
    passed = bool(condition)
    tests.append({
        "test": name,
        "status": "PASS" if passed else "FAIL",
        "detail": str(detail)
    })
    if not passed:
        raise AssertionError(f"{name} FAILED: {detail}")

# Hit@K
check("hit_first_rank", hit_at_k([1, 0, 0], 1) == 1, "Relevant at rank 1")
check("hit_within_k", hit_at_k([0, 0, 1], 3) == 1, "Relevant at rank 3")
check("hit_outside_k", hit_at_k([0, 0, 1], 2) == 0, "Relevant only outside top 2")
check("hit_no_relevant", hit_at_k([0, 0, 0], 3) == 0, "No relevant result")

# MRR
check("mrr_rank1", math.isclose(reciprocal_rank([1, 0, 0]), 1.0), "Expected 1")
check("mrr_rank2", math.isclose(reciprocal_rank([0, 1, 0]), 0.5), "Expected 1/2")
check("mrr_rank4", math.isclose(reciprocal_rank([0, 0, 0, 1]), 0.25), "Expected 1/4")
check("mrr_no_relevant", math.isclose(reciprocal_rank([0, 0]), 0.0), "Expected 0")

# nDCG — ideal and deliberately reversed rankings
check("ndcg_perfect_ranking",
      math.isclose(ndcg_at_k([3, 2, 1], 3), 1.0),
      "Ideal ranking must score 1")
check("ndcg_reversed_below_ideal",
      0 <= ndcg_at_k([0, 1, 3], 3) < 1,
      "Reversed ranking must score between 0 and 1")
check("ndcg_all_zero",
      math.isclose(ndcg_at_k([0, 0, 0], 3), 0.0),
      "No relevant gains must score 0")
check("ndcg_cutoff",
      math.isclose(ndcg_at_k([3, 0, 0], 1), 1.0),
      "At k=1, the best item is ideal")
check("ndcg_always_bounded",
      all(0 <= ndcg_at_k(r, len(r)) <= 1
          for r in ([3, 0, 1], [0, 3, 1], [1, 1, 1], [0, 0, 0])),
      "Representative rankings must remain in [0,1]")

# Page coverage: two gold pages, with chunks covering different pages
ranked_pages = [{129}, {130}, {129, 130}]
gold = {129, 130}
check("page_recall_top1",
      math.isclose(page_recall_at_k(ranked_pages, gold, 1), 0.5),
      "Top 1 covers one of two gold pages")
check("page_recall_top2",
      math.isclose(page_recall_at_k(ranked_pages, gold, 2), 1.0),
      "Top 2 covers both gold pages")
check("multi_page_incomplete_top1",
      multi_page_complete_at_k(ranked_pages, gold, 1) == 0,
      "One page missing")
check("multi_page_complete_top2",
      multi_page_complete_at_k(ranked_pages, gold, 2) == 1,
      "Both pages covered")
check("page_overlap_deduplicated",
      math.isclose(page_recall_at_k([{129}, {129}], gold, 2), 0.5),
      "Duplicate coverage of one page must not count twice")

# Invalid inputs must fail loudly
try:
    ndcg_at_k([3, -1], 2)
    negative_rel_rejected = False
except ValueError:
    negative_rel_rejected = True
check("ndcg_rejects_negative_relevance",
      negative_rel_rejected, "Negative relevance rejected")

try:
    page_recall_at_k([ {129} ], set(), 1)
    empty_gold_rejected = False
except ValueError:
    empty_gold_rejected = True
check("page_recall_rejects_empty_gold",
      empty_gold_rejected, "Empty gold-page set rejected")

# ---------- 4. Integrity checks on existing artifacts ----------
check("pilot_has_10_unique_questions",
      len(pilot_df) == 10 and pilot_df["question_id"].nunique() == 10
      if "question_id" in pilot_df.columns
      else len(pilot_df) == 10,
      f"Rows={len(pilot_df)}; columns={list(pilot_df.columns)}")

check("raw_has_300_rows", len(raw) == 300, f"Rows={len(raw)}")
check("question_metrics_has_30_rows",
      len(question_metrics) == 30, f"Rows={len(question_metrics)}")
check("aggregate_has_3_rows", len(aggregate) == 3, f"Rows={len(aggregate)}")

# Raw ranking integrity: adapt column names from actual saved CSV.
lower_cols = {c.lower(): c for c in raw.columns}
qid_col = next((lower_cols[c] for c in ("question_id", "qid", "id") if c in lower_cols), None)
method_col = next((lower_cols[c] for c in ("method", "retriever", "system") if c in lower_cols), None)
rank_col = next((lower_cols[c] for c in ("rank", "retrieval_rank") if c in lower_cols), None)

check("raw_ranking_columns_found",
      qid_col is not None and method_col is not None and rank_col is not None,
      f"Detected qid={qid_col}, method={method_col}, rank={rank_col}")

if qid_col and method_col and rank_col:
    check("raw_has_10_questions",
          raw[qid_col].nunique() == 10, f"Unique questions={raw[qid_col].nunique()}")
    check("raw_has_3_methods",
          raw[method_col].nunique() == 3, f"Methods={raw[method_col].unique().tolist()}")
    counts = raw.groupby([qid_col, method_col])[rank_col].count()
    check("10_ranks_per_question_method",
          bool((counts == 10).all()) and len(counts) == 30,
          f"Groups={len(counts)}, rank counts={sorted(counts.unique().tolist())}")
    duplicate_ranks = raw.duplicated([qid_col, method_col, rank_col]).sum()
    check("no_duplicate_question_method_ranks",
          duplicate_ranks == 0, f"Duplicate rank keys={duplicate_ranks}")

# ---------- 5. Save only after all tests pass ----------
OUT.mkdir(parents=True, exist_ok=False)
report = {
    "report_version": "metric_validation_v1",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "pilot_path": str(PILOT),
    "pilot_sha256_before": pilot_hash_before,
    "pilot_sha256_after": sha256(PILOT),
    "pilot_hash_unchanged": pilot_hash_before == sha256(PILOT),
    "protocol_json_path": str(PROTOCOL_JSON),
    "protocol_markdown_path": str(PROTOCOL_MD),
    "metric_definitions": {
        "nDCG": "Exponential graded gain (2^rel - 1), log2 discount; ideal ranking sorted from the same relevance list.",
        "PageRecall@K": "Distinct gold PDF pages covered by top K divided by distinct gold PDF pages.",
        "MultiPageComplete@K": "1 only if every required gold PDF page is covered by top K.",
        "Hit@K": "1 if any relevant result occurs in top K.",
        "MRR": "Reciprocal of the first relevant rank, or 0 if none."
    },
    "tests_passed": sum(t["status"] == "PASS" for t in tests),
    "tests_failed": sum(t["status"] == "FAIL" for t in tests),
    "tests": tests,
    "limitations": [
        "These are independent unit tests, not proof that the existing evaluator uses these definitions.",
        "Existing ranking CSV schemas are checked, but existing metric values are not silently replaced.",
        "Page coverage is not the same as evidence sufficiency or answer correctness.",
        "The pilot remains small and is not evidence of general retriever superiority."
    ]
}

assert report["pilot_hash_unchanged"], "Frozen pilot hash changed unexpectedly."
assert report["tests_failed"] == 0, "A unit test failed; refusing to report success."

(OUT / "metric_unit_test_report.json").write_text(
    json.dumps(report, indent=2, ensure_ascii=False), encoding="utf-8"
)
pd.DataFrame(tests).to_csv(OUT / "metric_unit_test_cases.csv", index=False)

print("=" * 68)
print("METRIC UNIT TESTS COMPLETE")
print("=" * 68)
print(f"Passed: {report['tests_passed']}")
print(f"Failed: {report['tests_failed']}")
print(f"Pilot SHA256 unchanged: {report['pilot_hash_unchanged']}")
print(f"Report: {OUT / 'metric_unit_test_report.json'}")
print(f"Test cases: {OUT / 'metric_unit_test_cases.csv'}")
print("\nIMPORTANT: this validates the independent metric definitions.")
print("Next, compare these definitions against the actual baseline evaluator.")
```

```python

# ================================================================
# AUDIT EXISTING EVALUATOR CODE — READ ONLY
# No retrieval rerun. No files created or modified.
# ================================================================

from pathlib import Path
import re
import hashlib

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
SEARCH_ROOTS = [
    ROOT / "05_EXPERIMENTS/retrieval_evaluation",
    ROOT / "03_RETRIEVAL",
]

PATTERNS = [
    r"\bndcg\b", r"\bdcg\b", r"discounted.*gain",
    r"\bpage_recall\b", r"\bpage.*recall\b",
    r"\bmrr\b", r"\bhit_at_k\b", r"\bhit@",
    r"gold_pdf_pages", r"gold_pages",
    r"relevant_pages", r"relevance", r"precision",
]

MAX_FILE_BYTES = 2_000_000
MAX_FILES_TO_SCAN = 250
MAX_MATCHES_PER_FILE = 12
CONTEXT = 2

candidates = []
for folder in SEARCH_ROOTS:
    if folder.exists():
        for p in folder.rglob("*"):
            if not p.is_file():
                continue
            if p.suffix.lower() not in {".py", ".ipynb"}:
                continue
            try:
                if p.stat().st_size <= MAX_FILE_BYTES:
                    candidates.append(p)
            except OSError:
                pass

candidates = sorted(set(candidates), key=lambda p: str(p).lower())
print(f"Candidate source files found: {len(candidates)}")

if len(candidates) > MAX_FILES_TO_SCAN:
    print(f"Scanning first {MAX_FILES_TO_SCAN} files only.")
    candidates = candidates[:MAX_FILES_TO_SCAN]

found_files = 0
total_matches = 0

for path in candidates:
    try:
        content = path.read_text(encoding="utf-8", errors="replace")
        lines = content.splitlines()
    except Exception as e:
        print(f"\n[READ ERROR] {path}: {e}")
        continue

    matched_line_nums = []
    for i, line in enumerate(lines):
        if any(re.search(pat, line, flags=re.IGNORECASE) for pat in PATTERNS):
            matched_line_nums.append(i)

    if not matched_line_nums:
        continue

    # Merge nearby matches into compact windows.
    windows = []
    for i in matched_line_nums:
        start, end = max(0, i - CONTEXT), min(len(lines) - 1, i + CONTEXT)
        if windows and start <= windows[-1][1] + 1:
            windows[-1] = (windows[-1][0], max(windows[-1][1], end))
        else:
            windows.append((start, end))

    print("\n" + "=" * 88)
    print(f"FILE: {path}")
    print(f"SHA256: {hashlib.sha256(content.encode('utf-8')).hexdigest()}")
    print(f"Matching lines: {len(matched_line_nums)}")

    printed = 0
    for start, end in windows:
        if printed >= MAX_MATCHES_PER_FILE:
            print("... additional matches omitted")
            break
        print(f"\n--- lines {start + 1}-{end + 1} ---")
        for j in range(start, end + 1):
            print(f"{j + 1:5}: {lines[j][:300]}")
            printed += 1
            if printed >= MAX_MATCHES_PER_FILE:
                break

    found_files += 1
    total_matches += len(matched_line_nums)

print("\n" + "=" * 88)
print("AUDIT SCAN COMPLETE")
print(f"Files containing relevant code: {found_files}")
print(f"Total matching lines: {total_matches}")
print("Read-only scan: no files were written or modified.")
print(
    "\nIf no evaluator implementation appears, it may exist only in a "
    "previous Colab cell rather than a saved .py/.ipynb file."
)
```

```python

# ================================================================
# INDEPENDENTLY AUDIT SAVED BASELINE METRICS — READ ONLY INPUTS
# Writes only a new report folder; never overwrites old outputs.
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json, ast, hashlib
from datetime import datetime, timezone

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/baseline_v2_pilot"
)
OUT = BASE.parent / "saved_metric_audit_v1"

RAW = BASE / "raw_retrieval_results.csv"
SAVED = BASE / "question_metrics.csv"
PILOT = BASE.parent / "NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"

assert RAW.is_file() and SAVED.is_file() and PILOT.is_file()
assert not OUT.exists(), f"Refusing to overwrite: {OUT}"

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for b in iter(lambda: f.read(1024 * 1024), b""):
            h.update(b)
    return h.hexdigest()

pilot_hash_before = sha256(PILOT)
raw_hash_before = sha256(RAW)
saved_hash_before = sha256(SAVED)

raw = pd.read_csv(RAW)
saved = pd.read_csv(SAVED)

required_raw = {
    "question_id", "method", "rank", "source_file", "pages",
    "gold_source_file", "gold_pages", "page_relevant"
}
required_saved = {
    "question_id", "method", "Hit@1", "Hit@3", "Hit@5", "Hit@10",
    "PageRecall@1", "PageRecall@3", "PageRecall@5", "PageRecall@10",
    "ChunkPrecision@1", "ChunkPrecision@3", "ChunkPrecision@5",
    "ChunkPrecision@10", "MRR", "nDCG@1", "nDCG@3", "nDCG@5", "nDCG@10"
}
assert required_raw.issubset(raw.columns), f"Missing raw columns: {required_raw - set(raw.columns)}"
assert required_saved.issubset(saved.columns), f"Missing metric columns: {required_saved - set(saved.columns)}"

def parse_pages(value):
    if isinstance(value, (list, tuple, set)):
        return set(value)
    if pd.isna(value):
        return set()
    try:
        parsed = ast.literal_eval(str(value))
        if isinstance(parsed, (list, tuple, set)):
            return set(parsed)
    except Exception:
        pass
    return set()

def dcg_binary(rels):
    rels = np.asarray(rels, dtype=float)
    if len(rels) == 0:
        return 0.0
    return float(np.sum(rels / np.log2(np.arange(2, len(rels) + 2))))

def ndcg_binary(rels, k):
    rels = list(rels)[:k]
    ideal = sorted(rels, reverse=True)
    denom = dcg_binary(ideal)
    return dcg_binary(rels) / denom if denom else 0.0

def close(a, b, tol=1e-8):
    return pd.notna(a) and np.isclose(float(a), float(b), atol=tol, rtol=tol)

rows = []
warnings = []

for (qid, method), group in raw.groupby(["question_id", "method"], sort=True):
    group = group.sort_values("rank")
    if len(group) != 10 or group["rank"].tolist() != list(range(1, 11)):
        warnings.append(f"{qid}/{method}: expected ranks 1..10; got {group['rank'].tolist()}")

    # Existing raw relevance flag is source-file AND gold-page overlap.
    rel = group["page_relevant"].fillna(0).astype(int).tolist()
    gold_source = str(group.iloc[0]["gold_source_file"])
    gold_pages = parse_pages(group.iloc[0]["gold_pages"])

    retrieved_page_sets = []
    for _, r in group.iterrows():
        same_source = str(r["source_file"]) == gold_source
        pages = parse_pages(r["pages"]) if same_source else set()
        retrieved_page_sets.append(pages & gold_pages)

    out = {"question_id": qid, "method": method}
    for k in (1, 3, 5, 10):
        top_rel = rel[:k]
        out[f"Hit@{k}"] = int(any(x == 1 for x in top_rel))
        out[f"PageRecall@{k}"] = (
            len(set().union(*retrieved_page_sets[:k]) & gold_pages) / len(gold_pages)
            if gold_pages else np.nan
        )
        out[f"ChunkPrecision@{k}"] = float(np.mean(top_rel)) if top_rel else 0.0
        out[f"nDCG@{k}"] = ndcg_binary(top_rel, k)

    first = next((i + 1 for i, x in enumerate(rel) if x == 1), None)
    out["MRR"] = 1.0 / first if first else 0.0
    rows.append(out)

recalc = pd.DataFrame(rows)
saved_compare = saved.copy()

audit_rows = []
metric_names = [
    "Hit@1", "Hit@3", "Hit@5", "Hit@10",
    "PageRecall@1", "PageRecall@3", "PageRecall@5", "PageRecall@10",
    "ChunkPrecision@1", "ChunkPrecision@3", "ChunkPrecision@5",
    "ChunkPrecision@10", "MRR",
    "nDCG@1", "nDCG@3", "nDCG@5", "nDCG@10"
]

merged = recalc.merge(
    saved_compare, on=["question_id", "method"], suffixes=("_recalc", "_saved"),
    validate="one_to_one"
)

for metric in metric_names:
    a, b = f"{metric}_recalc", f"{metric}_saved"
    for _, r in merged.iterrows():
        equal = close(r[a], r[b])
        audit_rows.append({
            "question_id": r["question_id"],
            "method": r["method"],
            "metric": metric,
            "recomputed": r[a],
            "saved": r[b],
            "matches": bool(equal)
        })

audit = pd.DataFrame(audit_rows)
mismatch = audit[~audit["matches"]].copy()

# Protect against accidentally treating source/page overlap as proof of answer support.
method_summary = recalc.groupby("method")[metric_names].mean(numeric_only=True).reset_index()

OUT.mkdir(parents=True, exist_ok=False)
audit.to_csv(OUT / "metric_by_question_comparison.csv", index=False)
mismatch.to_csv(OUT / "metric_mismatches.csv", index=False)
recalc.to_csv(OUT / "independently_recomputed_metrics.csv", index=False)
method_summary.to_csv(OUT / "independently_recomputed_summary.csv", index=False)

report = {
    "audit_version": "saved_metric_audit_v1",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "raw_rows": len(raw),
    "saved_metric_rows": len(saved),
    "recomputed_question_method_pairs": len(recalc),
    "metric_comparisons": len(audit),
    "matching_comparisons": int(audit["matches"].sum()),
    "mismatching_comparisons": int((~audit["matches"]).sum()),
    "mismatch_counts_by_metric": (
        mismatch.groupby("metric").size().to_dict() if len(mismatch) else {}
    ),
    "warnings": warnings,
    "hashes_unchanged": {
        "pilot": pilot_hash_before == sha256(PILOT),
        "raw_results": raw_hash_before == sha256(RAW),
        "saved_metrics": saved_hash_before == sha256(SAVED)
    },
    "relevance_definition": (
        "This audit uses the saved page_relevant flag for Hit/precision/MRR/nDCG, "
        "and recomputes page recall from same-source PDF-page overlap. "
        "This does not establish evidence sufficiency or answer correctness."
    )
}

(OUT / "audit_report.json").write_text(
    json.dumps(report, indent=2, default=str), encoding="utf-8"
)

print("=" * 68)
print("SAVED METRIC AUDIT COMPLETE")
print("=" * 68)
print(f"Raw rows: {len(raw)}")
print(f"Saved metric rows: {len(saved)}")
print(f"Metric comparisons: {len(audit)}")
print(f"Matching: {int(audit['matches'].sum())}")
print(f"Mismatching: {int((~audit['matches']).sum())}")
print("\nMismatch counts by metric:")
print(mismatch.groupby("metric").size().to_string() if len(mismatch) else "None")
print("\nFile hashes unchanged:", report["hashes_unchanged"])
print("\nReport folder:", OUT)
print("\nInterpretation note: source/page overlap is not proof of answer support.")
```

```python

from pathlib import Path
import pandas as pd
import numpy as np

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/baseline_v2_pilot"
)

raw = pd.read_csv(BASE / "raw_retrieval_results.csv")
saved = pd.read_csv(BASE / "question_metrics.csv")

def dcg_binary(rels):
    rels = np.asarray(rels, dtype=float)
    return float(np.sum(rels / np.log2(np.arange(2, len(rels) + 2))))

def ndcg_standard(rels, k):
    top = list(rels)[:k]
    ideal = sorted(top, reverse=True)
    denom = dcg_binary(ideal)
    return dcg_binary(top) / denom if denom else 0.0

def ndcg_ideal_from_all(rels, k):
    top = list(rels)[:k]
    ideal = sorted(list(rels), reverse=True)[:k]
    denom = dcg_binary(ideal)
    return dcg_binary(top) / denom if denom else 0.0

def ndcg_exponential(rels, k):
    top = np.asarray(list(rels)[:k], dtype=float)
    gains = 2**top - 1
    discounts = np.log2(np.arange(2, len(top) + 2))
    actual = np.sum(gains / discounts)
    ideal_rel = np.sort(top)[::-1]
    ideal = np.sum((2**ideal_rel - 1) / discounts)
    return float(actual / ideal) if ideal else 0.0

ids = ["V2P002", "V2P003", "V2P005", "V2P007", "V2P008", "V2P009"]
ks = [3, 5, 10]

for qid in ids:
    for method in ["BM25", "DENSE", "HYBRID_RRF"]:
        g = raw[
            (raw.question_id == qid) &
            (raw.method == method)
        ].sort_values("rank")
        if g.empty:
            continue

        rels = g["page_relevant"].fillna(0).astype(int).tolist()
        old = saved[
            (saved.question_id == qid) &
            (saved.method == method)
        ]
        if old.empty:
            continue
        old = old.iloc[0]

        print(f"\n{qid} | {method}")
        print("relevance ranks 1-10:", rels)
        print("relevant ranks:", [i + 1 for i, v in enumerate(rels) if v])
        for k in ks:
            print(
                f"@{k}: saved={old[f'nDCG@{k}']:.6f} | "
                f"binary_standard={ndcg_standard(rels,k):.6f} | "
                f"ideal_all_results={ndcg_ideal_from_all(rels,k):.6f} | "
                f"exponential_gain={ndcg_exponential(rels,k):.6f}"
            )

print("\nDiagnostic only: no files modified.")
```

```python

# ================================================================
# NCERT RAG — CORRECTED METRICS v2
# Source: existing saved retrieval rankings (no retrieval rerun)
# Original files and frozen benchmark are never modified.
# ================================================================

from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import numpy as np
import hashlib
import json
import ast

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/baseline_v2_pilot"
)
PILOT = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/"
    "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"
)

RAW = BASE / "raw_retrieval_results.csv"
OLD_Q = BASE / "question_metrics.csv"
OLD_A = BASE / "aggregate_metrics.csv"

OUT = BASE.parent / "corrected_metrics_v2"

# ---------- Preconditions ----------
for p in (RAW, OLD_Q, OLD_A, PILOT):
    assert p.is_file(), f"Required input missing: {p}"

assert not OUT.exists(), (
    f"Output already exists: {OUT}\n"
    "Stopping to avoid overwriting any previous work."
)

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

input_hashes_before = {
    "raw_retrieval_results.csv": sha256(RAW),
    "question_metrics.csv": sha256(OLD_Q),
    "aggregate_metrics.csv": sha256(OLD_A),
    "frozen_pilot.xlsx": sha256(PILOT),
}

raw = pd.read_csv(RAW)
old_q = pd.read_csv(OLD_Q)
old_agg = pd.read_csv(OLD_A)

required = {
    "question_id", "question", "question_type", "method", "rank",
    "source_file", "pages", "gold_source_file", "gold_pages",
    "page_relevant"
}
assert required.issubset(raw.columns), (
    f"Raw ranking columns missing: {required - set(raw.columns)}"
)

# ---------- Helpers ----------
def parse_pages(value):
    if isinstance(value, (list, tuple, set)):
        return set(value)
    if pd.isna(value):
        return set()
    try:
        result = ast.literal_eval(str(value))
        if isinstance(result, (list, tuple, set)):
            return set(result)
    except Exception:
        pass
    return set()

def dcg_binary(relevances):
    """Binary gain: rel/log2(rank+1), with ranks starting at 1."""
    rel = np.asarray(relevances, dtype=float)
    if len(rel) == 0:
        return 0.0
    if np.any(~np.isfinite(rel)) or np.any((rel != 0) & (rel != 1)):
        raise ValueError("Binary relevance must contain only 0 and 1.")
    discounts = np.log2(np.arange(2, len(rel) + 2))
    return float(np.sum(rel / discounts))

def ndcg_binary_at_k(relevances, k):
    """
    Standard binary-relevance nDCG.
    The ideal ranking sorts the same top-k candidate relevance labels.
    """
    observed = list(relevances)[:k]
    ideal = sorted(observed, reverse=True)
    ideal_dcg = dcg_binary(ideal)
    if ideal_dcg == 0:
        return 0.0
    score = dcg_binary(observed) / ideal_dcg
    assert -1e-12 <= score <= 1 + 1e-12, score
    return float(min(1.0, max(0.0, score)))

def hit_at_k(relevances, k):
    return int(any(x == 1 for x in list(relevances)[:k]))

def mrr(relevances):
    for rank, rel in enumerate(relevances, start=1):
        if rel == 1:
            return 1.0 / rank
    return 0.0

# ---------- Validate ranking integrity ----------
assert len(raw) == 300, f"Expected 300 ranking rows, got {len(raw)}"
assert raw["question_id"].nunique() == 10
assert raw["method"].nunique() == 3
assert set(raw["method"]) == {"BM25", "DENSE", "HYBRID_RRF"}

rank_counts = raw.groupby(["question_id", "method"])["rank"].apply(
    lambda s: sorted(s.astype(int).tolist())
)
assert len(rank_counts) == 30
assert all(ranks == list(range(1, 11)) for ranks in rank_counts), (
    "At least one question-method group does not have ranks 1..10."
)

# ---------- Recompute per-question metrics ----------
ks = (1, 3, 5, 10)
recomputed_rows = []

for (qid, method), group in raw.groupby(
    ["question_id", "method"], sort=True
):
    group = group.sort_values("rank").reset_index(drop=True)
    rel = group["page_relevant"].fillna(0).astype(int).tolist()

    assert all(x in (0, 1) for x in rel), (
        f"Non-binary page_relevant values for {qid}/{method}"
    )

    first = next((i + 1 for i, value in enumerate(rel) if value == 1), None)

    first_row = group.iloc[0]
    gold_source = str(first_row["gold_source_file"])
    gold_pages = parse_pages(first_row["gold_pages"])

    # A page counts only if retrieved from the annotated gold source.
    covered_pages_by_rank = []
    for _, r in group.iterrows():
        same_source = str(r["source_file"]) == gold_source
        pages = parse_pages(r["pages"]) if same_source else set()
        covered_pages_by_rank.append(pages & gold_pages)

    record = {
        "question_id": qid,
        "question": first_row["question"],
        "question_type": first_row["question_type"],
        "method": method,
        "MRR": 1.0 / first if first else 0.0,
    }

    for k in ks:
        top_rel = rel[:k]
        record[f"Hit@{k}"] = hit_at_k(rel, k)
        record[f"ChunkPrecision@{k}"] = float(np.mean(top_rel))

        if gold_pages:
            covered = set().union(*covered_pages_by_rank[:k])
            record[f"PageRecall@{k}"] = (
                len(covered & gold_pages) / len(gold_pages)
            )
        else:
            record[f"PageRecall@{k}"] = np.nan

        record[f"nDCG@{k}"] = ndcg_binary_at_k(rel, k)

    # Only questions explicitly annotated as multi-page are scored here.
    is_multipage = (
        len(gold_pages) > 1
        and str(first_row["question_type"]).upper() == "MULTI_PAGE"
    )
    if is_multipage:
        covered_top10 = set().union(*covered_pages_by_rank[:10])
        record["MultiPageComplete@10"] = int(
            gold_pages.issubset(covered_top10)
        )
    else:
        record["MultiPageComplete@10"] = np.nan

    recomputed_rows.append(record)

corrected = pd.DataFrame(recomputed_rows)

assert len(corrected) == 30
assert not corrected.duplicated(["question_id", "method"]).any()

# ---------- Preserve old nDCG side-by-side ----------
old_ndcg_cols = [
    "question_id", "method", "nDCG@1", "nDCG@3", "nDCG@5", "nDCG@10"
]
assert set(old_ndcg_cols).issubset(old_q.columns)

comparison = corrected.merge(
    old_q[old_ndcg_cols],
    on=["question_id", "method"],
    how="left",
    suffixes=("_corrected", "_original"),
    validate="one_to_one"
)

for k in ks:
    comparison[f"nDCG@{k}_absolute_change"] = (
        comparison[f"nDCG@{k}_corrected"]
        - comparison[f"nDCG@{k}_original"]
    ).abs()

# ---------- Aggregate from corrected per-question metrics ----------
aggregate_rows = []
metric_cols = [
    *(f"Hit@{k}" for k in ks),
    *(f"PageRecall@{k}" for k in ks),
    *(f"ChunkPrecision@{k}" for k in ks),
    "MRR",
    *(f"nDCG@{k}" for k in ks),
    "MultiPageComplete@10",
]

for method, group in corrected.groupby("method", sort=True):
    row = {"method": method, "questions": int(group["question_id"].nunique())}
    for col in metric_cols:
        # Skip non-applicable multi-page cases; pandas mean ignores NaN.
        row[col] = float(group[col].mean()) if group[col].notna().any() else np.nan
    aggregate_rows.append(row)

aggregate_corrected = pd.DataFrame(aggregate_rows)

# ---------- Independent safety checks ----------
for k in ks:
    assert corrected[f"nDCG@{k}"].between(0, 1).all()
    assert corrected[f"Hit@{k}"].isin([0, 1]).all()
    assert corrected[f"ChunkPrecision@{k}"].between(0, 1).all()
    assert corrected[f"PageRecall@{k}"].dropna().between(0, 1).all()

assert aggregate_corrected[
    [c for c in aggregate_corrected.columns if c.startswith("nDCG@")]
].apply(lambda s: s.between(0, 1).all()).all()

# Ensure all input files are still byte-for-byte unchanged.
input_hashes_after = {
    "raw_retrieval_results.csv": sha256(RAW),
    "question_metrics.csv": sha256(OLD_Q),
    "aggregate_metrics.csv": sha256(OLD_A),
    "frozen_pilot.xlsx": sha256(PILOT),
}
assert input_hashes_before == input_hashes_after, (
    "An input file changed during the audit."
)

# ---------- Save a new version only ----------
OUT.mkdir(parents=True, exist_ok=False)
corrected.to_csv(OUT / "corrected_question_metrics.csv", index=False)
aggregate_corrected.to_csv(
    OUT / "corrected_aggregate_metrics.csv", index=False
)
comparison.to_csv(OUT / "original_vs_corrected_ndcg.csv", index=False)

manifest = {
    "version": "corrected_metrics_v2",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "input_files": {k: {"sha256": v} for k, v in input_hashes_before.items()},
    "output_files": [
        "corrected_question_metrics.csv",
        "corrected_aggregate_metrics.csv",
        "original_vs_corrected_ndcg.csv",
    ],
    "ranking_rows": int(len(raw)),
    "questions": int(raw["question_id"].nunique()),
    "methods": sorted(raw["method"].unique().tolist()),
    "rank_cutoffs": list(ks),
    "definitions": {
        "relevance": "Existing binary page_relevant flag from the saved rankings.",
        "nDCG": "Binary gain rel/log2(rank+1), normalized by the ideal ordering of the same top-k relevance labels.",
        "Hit@K": "1 if any top-k result is relevant, else 0.",
        "MRR": "Reciprocal rank of first relevant result, or 0 if none.",
        "ChunkPrecision@K": "Relevant chunks among top-k divided by k.",
        "PageRecall@K": "Distinct annotated gold PDF pages covered by top-k same-source chunks divided by distinct gold pages.",
        "MultiPageComplete@10": "For MULTI_PAGE questions only, 1 if all gold pages are covered in top 10, else 0.",
    },
    "limitations": [
        "This corrects metric computation; it does not independently validate the benchmark relevance annotations.",
        "Source and page overlap do not prove evidence sufficiency or answer correctness.",
        "The 10-question pilot is too small for broad claims about retriever superiority.",
        "Original outputs are retained unchanged for traceability.",
    ],
    "input_hashes_unchanged_after_run": True,
}
(OUT / "manifest.json").write_text(
    json.dumps(manifest, indent=2), encoding="utf-8"
)

print("=" * 68)
print("CORRECTED METRICS V2 CREATED")
print("=" * 68)
print("Questions:", corrected["question_id"].nunique())
print("Question-method rows:", len(corrected))
print("Ranking rows used:", len(raw))
print("Methods:", ", ".join(sorted(corrected["method"].unique())))
print("All corrected nDCG values in [0,1]:", all(
    corrected[f"nDCG@{k}"].between(0, 1).all() for k in ks
))
print("\nCorrected aggregate metrics:")
print(aggregate_corrected.round(4).to_string(index=False))
print("\nOriginal input hashes unchanged:", input_hashes_before == input_hashes_after)
print("Output folder:", OUT)
print("\nOriginal metric files and frozen pilot were not modified.")
```

```python

from pathlib import Path
import pandas as pd
import json
import hashlib
from datetime import datetime, timezone

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2"

PILOT = BASE / "NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"
RAW = BASE / "baseline_v2_pilot/raw_retrieval_results.csv"
GROUNDING_SHEET = "Grounding_Validation"

OUT = BASE / "evidence_sufficiency_audit_v1"
OUT.mkdir(parents=True, exist_ok=True)

# Load source artifacts without modifying them.
pilot = pd.read_excel(PILOT, sheet_name="Benchmark_V2_Pilot")
grounding = pd.read_excel(PILOT, sheet_name=GROUNDING_SHEET)
raw = pd.read_csv(RAW)

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

# Guard against accidental schema drift.
required_pilot = {
    "question_id", "question", "expected_answer", "evidence_excerpt",
    "source_file", "gold_pdf_pages", "question_type", "answerability"
}
required_raw = {
    "question_id", "method", "rank", "source_file", "pages",
    "page_relevant", "chunk_text"
}
assert required_pilot.issubset(pilot.columns), (
    f"Missing pilot columns: {required_pilot - set(pilot.columns)}"
)
assert required_raw.issubset(raw.columns), (
    f"Missing retrieval columns: {required_raw - set(raw.columns)}"
)
assert pilot["question_id"].is_unique, "Pilot question IDs are not unique."
assert set(raw["question_id"]) == set(pilot["question_id"]), (
    "Pilot and retrieval question IDs do not match."
)

grounding_by_id = grounding.set_index("question_id").to_dict("index")
pilot_by_id = pilot.set_index("question_id").to_dict("index")

# Parse page lists safely, supporting list-like strings such as "[129, 130]".
def parse_pages(value):
    if pd.isna(value):
        return []
    if isinstance(value, (list, tuple, set)):
        return [str(x).strip() for x in value]
    s = str(value).strip()
    try:
        parsed = json.loads(s)
        if isinstance(parsed, list):
            return [str(x).strip() for x in parsed]
    except Exception:
        pass
    return [x.strip() for x in s.strip("[]").split(",") if x.strip()]

audit_rows = []

for _, hit in raw.sort_values(
    ["question_id", "method", "rank"]
).iterrows():
    qid = hit["question_id"]
    q = pilot_by_id[qid]
    g = grounding_by_id.get(qid, {})

    retrieved_pages = parse_pages(hit.get("pages"))
    gold_pages = parse_pages(q.get("gold_pdf_pages"))
    same_source = (
        str(hit.get("source_file", "")).strip()
        == str(q.get("source_file", "")).strip()
    )
    overlapping_pages = sorted(set(retrieved_pages) & set(gold_pages))

    audit_rows.append({
        "question_id": qid,
        "question": q.get("question"),
        "question_type": q.get("question_type"),
        "answerability": q.get("answerability"),
        "expected_answer": q.get("expected_answer"),
        "gold_evidence_excerpt": q.get("evidence_excerpt"),
        "gold_source_file": q.get("source_file"),
        "gold_pdf_pages": json.dumps(gold_pages),
        "grounding_status": g.get("status"),
        "grounding_reason": g.get("reason"),
        "verified_evidence_by_page": g.get("evidence_by_page"),
        "method": hit.get("method"),
        "rank": int(hit["rank"]),
        "retrieved_source_file": hit.get("source_file"),
        "retrieved_pages": json.dumps(retrieved_pages),
        "same_source_as_gold": same_source,
        "gold_page_overlap": bool(same_source and overlapping_pages),
        "overlapping_gold_pages": json.dumps(overlapping_pages),
        "existing_page_relevant_label": hit.get("page_relevant"),
        "retrieved_chunk_text": hit.get("chunk_text"),
        # Leave these blank until a human reviews the actual text.
        "human_evidence_sufficiency": "",
        "human_answer_support": "",
        "human_review_notes": ""
    })

audit = pd.DataFrame(audit_rows)

# Save the detailed review set.
detail_path = OUT / "retrieved_evidence_review.csv"
audit.to_csv(detail_path, index=False, encoding="utf-8-sig")

# Summarize source/page matches by method and rank cutoff.
summary_rows = []
for method, group in audit.groupby("method", sort=False):
    for k in [1, 3, 5, 10]:
        top = group[group["rank"] <= k]
        per_question = top.groupby("question_id").agg(
            any_gold_page_match=("gold_page_overlap", "max")
        )
        summary_rows.append({
            "method": method,
            "k": k,
            "questions": len(per_question),
            "questions_with_gold_page_match": int(
                per_question["any_gold_page_match"].sum()
            ),
            "fraction_questions_with_gold_page_match": round(
                per_question["any_gold_page_match"].mean(), 4
            ),
            "retrieved_chunks": len(top),
            "chunks_with_gold_page_match": int(top["gold_page_overlap"].sum()),
            "fraction_chunks_with_gold_page_match": round(
                top["gold_page_overlap"].mean(), 4
            )
        })

summary = pd.DataFrame(summary_rows)
summary_path = OUT / "page_match_summary.csv"
summary.to_csv(summary_path, index=False)

manifest = {
    "audit_version": "evidence_sufficiency_audit_v1",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "purpose": (
        "Human review of retrieved text against frozen benchmark evidence; "
        "page overlap is not treated as proof of evidence sufficiency."
    ),
    "pilot_sha256": sha256(PILOT),
    "raw_retrieval_sha256": sha256(RAW),
    "detail_rows": len(audit),
    "question_count": int(pilot["question_id"].nunique()),
    "methods": sorted(raw["method"].dropna().unique().tolist()),
    "rank_cutoffs": [1, 3, 5, 10],
    "human_review_fields_left_blank": [
        "human_evidence_sufficiency",
        "human_answer_support",
        "human_review_notes"
    ]
}
manifest_path = OUT / "manifest.json"
manifest_path.write_text(json.dumps(manifest, indent=2), encoding="utf-8")

print("AUDIT CREATED")
print("Output folder:", OUT)
print("Detailed review rows:", len(audit))
print("Expected rows: 300 (10 questions × 3 methods × 10 ranks)")
print("\nPage-match summary:")
print(summary.to_string(index=False))
print("\nFiles:")
print(detail_path)
print(summary_path)
print(manifest_path)
print("\nInput hashes:")
print("Pilot:", manifest["pilot_sha256"])
print("Raw retrieval:", manifest["raw_retrieval_sha256"])
```

```python

from pathlib import Path
import pandas as pd
import hashlib
import json
import random
from datetime import datetime, timezone

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2"
AUDIT_DIR = BASE / "evidence_sufficiency_audit_v1"
OUT = AUDIT_DIR / "blinded_review_v1"

DETAIL = AUDIT_DIR / "retrieved_evidence_review.csv"
PILOT = BASE / "NCERT_RAG_Benchmark_V2_Pilot_v4.xlsx"
RAW = BASE / "baseline_v2_pilot/raw_retrieval_results.csv"

for p in [DETAIL, PILOT, RAW]:
    assert p.exists(), f"Missing required file: {p}"

detail = pd.read_csv(DETAIL, keep_default_na=False)
pilot = pd.read_excel(PILOT, sheet_name="Benchmark_V2_Pilot")
raw = pd.read_csv(RAW)

OUT.mkdir(parents=True, exist_ok=True)

def norm(x):
    return " ".join(str(x).split()).strip()

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

# Verify that the source audit has the expected fields.
needed = {
    "question_id", "method", "rank", "retrieved_source_file",
    "retrieved_pages", "gold_page_overlap",
    "existing_page_relevant_label", "retrieved_chunk_text"
}
assert needed.issubset(detail.columns), (
    f"Missing audit columns: {needed - set(detail.columns)}"
)

# Rebuild the review candidates using the actual raw ranking data.
raw_needed = {
    "question_id", "method", "rank", "chunk_index",
    "retrieval_score", "chunk_text"
}
assert raw_needed.issubset(raw.columns), (
    f"Missing raw retrieval columns: {raw_needed - set(raw.columns)}"
)

top = detail[detail["rank"].astype(int) <= 3].copy()
raw_top = raw[raw["rank"].astype(int) <= 3].copy()

assert len(top) == 90, f"Expected 90 audit rows; found {len(top)}"

top["normalized_chunk"] = top["retrieved_chunk_text"].map(norm)
top["chunk_hash"] = top["normalized_chunk"].map(
    lambda x: hashlib.sha256(x.encode("utf-8")).hexdigest()
)

raw_top["normalized_chunk"] = raw_top["chunk_text"].map(norm)
raw_top["chunk_hash"] = raw_top["normalized_chunk"].map(
    lambda x: hashlib.sha256(x.encode("utf-8")).hexdigest()
)

# Attach rank and score metadata from the original raw results.
# Validate that each audit row maps to exactly one raw result.
raw_map = raw_top[[
    "question_id", "method", "rank", "chunk_index",
    "retrieval_score", "chunk_hash"
]].copy()

joined = top.merge(
    raw_map,
    on=["question_id", "method", "rank", "chunk_hash"],
    how="left",
    validate="one_to_one"
)

assert joined["chunk_index"].notna().all(), (
    "Some audit rows could not be matched to raw retrieval records. "
    "Stop and inspect text normalization before proceeding."
)

unique = joined.drop_duplicates(
    subset=["question_id", "chunk_hash"]
).copy()

rng = random.Random(20261009)
order = list(range(len(unique)))
rng.shuffle(order)
unique = unique.iloc[order].reset_index(drop=True)

unique["blind_candidate_id"] = [
    f"EV{i:03d}" for i in range(1, len(unique) + 1)
]

pilot_by_id = pilot.set_index("question_id").to_dict("index")

review = pd.DataFrame({
    "blind_candidate_id": unique["blind_candidate_id"],
    "question_id": unique["question_id"],
    "question": [
        pilot_by_id[q]["question"] for q in unique["question_id"]
    ],
    "expected_answer": [
        pilot_by_id[q]["expected_answer"] for q in unique["question_id"]
    ],
    "gold_evidence_excerpt": [
        pilot_by_id[q]["evidence_excerpt"] for q in unique["question_id"]
    ],
    "gold_source_file": [
        pilot_by_id[q]["source_file"] for q in unique["question_id"]
    ],
    "gold_pdf_pages": [
        pilot_by_id[q]["gold_pdf_pages"] for q in unique["question_id"]
    ],
    "retrieved_source_file": unique["retrieved_source_file"],
    "retrieved_pages": unique["retrieved_pages"],
    "retrieved_chunk_text": unique["retrieved_chunk_text"],
    "evidence_sufficiency_label": "",
    "answer_support_label": "",
    "reviewer_notes": ""
})

mapping = joined.merge(
    unique[["question_id", "chunk_hash", "blind_candidate_id"]],
    on=["question_id", "chunk_hash"],
    how="left",
    validate="many_to_one"
)[[
    "blind_candidate_id", "question_id", "method", "rank",
    "chunk_index", "retrieval_score", "retrieved_source_file",
    "retrieved_pages", "gold_page_overlap",
    "existing_page_relevant_label", "chunk_hash"
]].sort_values(["question_id", "method", "rank"])

guide = pd.DataFrame([
    {
        "field": "evidence_sufficiency_label",
        "allowed_values": "0, 1, 2, 3, UNCERTAIN",
        "definition": (
            "0 = no useful evidence; 1 = related but does not support "
            "the answer; 2 = partially supports but is insufficient alone; "
            "3 = sufficient evidence to support the expected answer; "
            "UNCERTAIN = cannot reliably decide."
        )
    },
    {
        "field": "answer_support_label",
        "allowed_values": "YES, PARTIAL, NO, UNCERTAIN",
        "definition": (
            "Whether this chunk alone supports the expected answer."
        )
    },
    {
        "field": "reviewer_notes",
        "allowed_values": "Free text",
        "definition": (
            "Justification, missing context, contradiction, or metadata issue."
        )
    }
])

review_path = OUT / "blinded_evidence_review.csv"
mapping_path = OUT / "candidate_rank_mapping.csv"
guide_path = OUT / "annotation_guide.csv"

review.to_csv(review_path, index=False, encoding="utf-8-sig")
mapping.to_csv(mapping_path, index=False, encoding="utf-8-sig")
guide.to_csv(guide_path, index=False, encoding="utf-8-sig")

manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "pilot_sha256": sha256(PILOT),
    "raw_retrieval_sha256": sha256(RAW),
    "audit_sha256": sha256(DETAIL),
    "candidate_occurrences": len(joined),
    "unique_question_chunk_pairs": len(review),
    "deduplicated_occurrences": len(joined) - len(review),
    "cutoff": 3,
    "methods": sorted(joined["method"].unique().tolist()),
    "note": (
        "Method, rank and score are hidden from the annotation sheet. "
        "Human judgments are required; page overlap is not sufficient evidence."
    )
}
(OUT / "manifest.json").write_text(
    json.dumps(manifest, indent=2), encoding="utf-8"
)

print("SUCCESS: review files created")
print("Top-3 occurrences:", len(joined))
print("Unique question-chunk pairs:", len(review))
print("Duplicate occurrences removed:", len(joined) - len(review))
print("Review CSV:", review_path)
print("Mapping CSV:", mapping_path)
print("Annotation guide:", guide_path)
print("Pilot SHA-256:", manifest["pilot_sha256"])

```

```python

from pathlib import Path
import pandas as pd
import hashlib
import json
import random
from datetime import datetime, timezone

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / (
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "evidence_sufficiency_audit_v1/blinded_review_v1"
)
OUT = BASE / "two_stage_annotation_v1"
OUT.mkdir(parents=True, exist_ok=True)

REVIEW_PATH = BASE / "blinded_evidence_review.csv"
MAP_PATH = BASE / "candidate_rank_mapping.csv"
GUIDE_PATH = BASE / "annotation_guide.csv"

for p in [REVIEW_PATH, MAP_PATH, GUIDE_PATH]:
    assert p.exists(), f"Missing input: {p}"

review = pd.read_csv(REVIEW_PATH, keep_default_na=False)
mapping = pd.read_csv(MAP_PATH, keep_default_na=False)
guide = pd.read_csv(GUIDE_PATH, keep_default_na=False)

assert len(review) == 57
assert review["blind_candidate_id"].is_unique
assert len(mapping) == 90
assert set(mapping["blind_candidate_id"]) == set(review["blind_candidate_id"])

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

# Stage A: no gold answer/evidence, method, rank, or retrieval score.
stage_a = review[[
    "blind_candidate_id",
    "question_id",
    "question",
    "retrieved_chunk_text"
]].copy()

stage_a["relevance_label"] = ""
stage_a["standalone_support_label"] = ""
stage_a["stage_a_notes"] = ""

# Use an explicit, reproducible pilot sample.
# Five questions are sampled; all candidates for each selected question
# are included so the sample is not chosen by retrieval method.
rng = random.Random(20261009)
question_ids = sorted(stage_a["question_id"].unique().tolist())
pilot_questions = sorted(rng.sample(question_ids, k=5))
pilot = stage_a[stage_a["question_id"].isin(pilot_questions)].copy()

# Stage B: reference information is separated from Stage A.
stage_b = review[[
    "blind_candidate_id",
    "question_id",
    "expected_answer",
    "gold_evidence_excerpt",
    "gold_source_file",
    "gold_pdf_pages",
    "retrieved_source_file",
    "retrieved_pages",
    "retrieved_chunk_text"
]].copy()

stage_b["expected_answer_support"] = ""
stage_b["reference_comparison_notes"] = ""

# A compact annotation guide that defines the decision separately by stage.
stage_a_guide = pd.DataFrame([
    {
        "field": "relevance_label",
        "allowed_values": "RELEVANT, PARTIAL, IRRELEVANT, UNCERTAIN",
        "instruction": (
            "Judge whether the retrieved chunk contains information relevant "
            "to answering the question. Do not consult the expected answer."
        )
    },
    {
        "field": "standalone_support_label",
        "allowed_values": "SUFFICIENT, PARTIAL, INSUFFICIENT, UNCERTAIN",
        "instruction": (
            "Judge whether this chunk alone contains enough information to "
            "answer the question. Do not infer missing facts."
        )
    },
    {
        "field": "stage_a_notes",
        "allowed_values": "Free text",
        "instruction": (
            "Briefly justify the label. Note ambiguity, contradictions, "
            "missing context, or unclear wording."
        )
    }
])

stage_b_guide = pd.DataFrame([
    {
        "field": "expected_answer_support",
        "allowed_values": "YES, PARTIAL, NO, UNCERTAIN",
        "instruction": (
            "After Stage A is locked, compare the retrieved chunk with the "
            "expected answer and gold evidence. Does this chunk support the "
            "benchmark's expected answer on its own?"
        )
    },
    {
        "field": "reference_comparison_notes",
        "allowed_values": "Free text",
        "instruction": (
            "Explain whether the evidence supports the answer, only part of "
            "it, contradicts it, or leaves required details unsupported."
        )
    }
])

# Save new files.
files = {
    "stage_a_all_candidates.csv": stage_a,
    "stage_a_pilot_sample.csv": pilot,
    "stage_b_reference_verification.csv": stage_b,
    "stage_a_annotation_guide.csv": stage_a_guide,
    "stage_b_annotation_guide.csv": stage_b_guide,
    "candidate_rank_mapping_copy.csv": mapping
}

for filename, df in files.items():
    df.to_csv(OUT / filename, index=False, encoding="utf-8-sig")

manifest = {
    "version": "two_stage_annotation_v1",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "input_review_sha256": sha256(REVIEW_PATH),
    "input_mapping_sha256": sha256(MAP_PATH),
    "candidate_count": len(review),
    "occurrence_count": len(mapping),
    "pilot_questions": pilot_questions,
    "pilot_candidate_count": len(pilot),
    "stage_a_hides": [
        "expected_answer", "gold_evidence_excerpt", "method",
        "rank", "retrieval_score"
    ],
    "stage_b_purpose": (
        "Reference verification after Stage A judgments are locked."
    ),
    "important_limitation": (
        "The same reviewer seeing both stages may retain memory of the "
        "reference. For stronger blinding, separate reviewers or a delay "
        "between stages is preferable."
    )
}

(OUT / "manifest.json").write_text(
    json.dumps(manifest, indent=2), encoding="utf-8"
)

print("TWO-STAGE ANNOTATION PACKAGE CREATED")
print("Output:", OUT)
print("All candidates:", len(stage_a))
print("Pilot questions:", pilot_questions)
print("Pilot candidates:", len(pilot))
print("Stage B candidates:", len(stage_b))
print("Ranking occurrences preserved:", len(mapping))
print("\nFiles:")
for filename in [*files.keys(), "manifest.json"]:
    print(" -", OUT / filename)
print("\nInput hashes:")
print("Review:", manifest["input_review_sha256"])
print("Mapping:", manifest["input_mapping_sha256"])
```

```python
import pandas as pd
from pathlib import Path

folder = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "evidence_sufficiency_audit_v1/blinded_review_v1/"
    "two_stage_annotation_v1"
)

stage_a_path = folder / "stage_a_pilot_sample_annotated.csv"
stage_b_path = folder / "stage_b_reference_verification.csv"

assert stage_a_path.exists(), f"Missing: {stage_a_path}"
assert stage_b_path.exists(), f"Missing: {stage_b_path}"

a = pd.read_csv(stage_a_path)
b = pd.read_csv(stage_b_path)

# Confirm IDs are unique before merging.
assert a["blind_candidate_id"].is_unique, "Duplicate Stage A IDs"
assert b["blind_candidate_id"].is_unique, "Duplicate Stage B IDs"

# Preserve exactly the candidate set annotated in Stage A.
pilot_ids = set(a["blind_candidate_id"])
reference_ids = set(b["blind_candidate_id"])

missing = pilot_ids - reference_ids
assert not missing, f"Missing reference records: {sorted(missing)}"

# Include the annotation and the reference evidence side by side.
columns_b = [
    "blind_candidate_id",
    "expected_answer",
    "gold_evidence_excerpt",
    "gold_source_file",
    "gold_pdf_pages",
    "retrieved_source_file",
    "retrieved_pages",
    "retrieved_chunk_text",
]

audit = a.merge(
    b[columns_b],
    on="blind_candidate_id",
    how="left",
    validate="one_to_one",
    suffixes=("_stage_a", "_reference"),
)

assert len(audit) == len(a) == 26

out = folder / "stage_b_pilot_reference_audit.csv"
audit.to_csv(out, index=False)

print("PASS: matched all 26 pilot candidates.")
print("Output:", out)
print("\nQuestions represented:", audit["question_id"].nunique())
print("\nCandidate IDs:")
print(", ".join(audit["blind_candidate_id"].tolist()))
print("\nOutput columns:")
print(audit.columns.tolist())
```

```python

from pathlib import Path
import pandas as pd
import hashlib

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2/evidence_sufficiency_audit_v1/"
    "blinded_review_v1/two_stage_annotation_v1"
)

SOURCE = BASE / "stage_b_pilot_reference_audit_reviewed.csv"
OUTPUT = BASE / "stage_a_pilot_sample_finalized.csv"

# Set True to accept the two audit recommendations.
# Set False to retain the original Stage A labels.
ACCEPT_RECOMMENDATIONS = True

df = pd.read_csv(SOURCE)

required = [
    "blind_candidate_id",
    "relevance_label",
    "standalone_support_label",
    "audit_recommended_relevance_label",
    "audit_recommended_support_label",
    "audit_decision",
]

missing = [c for c in required if c not in df.columns]
assert not missing, f"Missing columns: {missing}"
assert len(df) == 26, f"Expected 26 pilot candidates, got {len(df)}"
assert df["blind_candidate_id"].nunique() == 26, "Candidate IDs are not unique"
assert df["blind_candidate_id"].notna().all(), "Missing candidate IDs"

# Preserve original labels and audit recommendations.
df["final_relevance_label"] = df["relevance_label"]
df["final_support_label"] = df["standalone_support_label"]

if ACCEPT_RECOMMENDATIONS:
    df["final_relevance_label"] = df["audit_recommended_relevance_label"]
    df["final_support_label"] = df["audit_recommended_support_label"]

df["finalization_basis"] = (
    "Audit recommendations accepted" if ACCEPT_RECOMMENDATIONS
    else "Original Stage A labels retained"
)

# Validate the two expected changes before saving.
changes = df[
    (df["relevance_label"] != df["final_relevance_label"]) |
    (df["standalone_support_label"] != df["final_support_label"])
]

if ACCEPT_RECOMMENDATIONS:
    actual_changes = set(changes["blind_candidate_id"])
    assert actual_changes == {"EV008", "EV038"}, (
        f"Unexpected set of changed candidates: {actual_changes}"
    )
    ev008 = df.loc[df["blind_candidate_id"] == "EV008"].iloc[0]
    ev038 = df.loc[df["blind_candidate_id"] == "EV038"].iloc[0]
    assert ev008["final_relevance_label"] == 0
    assert ev008["final_support_label"] == 0
    assert ev038["final_relevance_label"] == 1
    assert ev038["final_support_label"] == 1

assert not OUTPUT.exists(), (
    f"Output already exists: {OUTPUT}\n"
    "Review it first; this cell will not overwrite it."
)

df.to_csv(OUTPUT, index=False)

sha256 = hashlib.sha256(OUTPUT.read_bytes()).hexdigest()

print("Finalized pilot annotations saved:", OUTPUT)
print("Candidates:", len(df))
print("Labels changed:", len(changes))
print("Recommendations accepted:", ACCEPT_RECOMMENDATIONS)
print("Final support-label counts:")
print(df["final_support_label"].value_counts().sort_index())
print("Output SHA-256:", sha256)
print("\nChanged candidates:")
if len(changes):
    print(changes[[
        "blind_candidate_id",
        "relevance_label", "standalone_support_label",
        "final_relevance_label", "final_support_label"
    ]].to_string(index=False))
else:
    print("None")
```

```python

from pathlib import Path
import pandas as pd

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2/evidence_sufficiency_audit_v1/"
    "blinded_review_v1/two_stage_annotation_v1"
)

FILE = BASE / "stage_b_remaining_reference_audit_reviewed.csv"
assert FILE.exists(), f"File not found: {FILE}"

df = pd.read_csv(FILE)
print("File:", FILE)
print("Rows:", len(df))
print("Columns:", df.columns.tolist())

# Find the audit columns without assuming exact naming.
def find_col(candidates):
    return next((c for c in candidates if c in df.columns), None)

id_col = find_col(["blind_candidate_id"])
rel_orig = find_col(["evidence_sufficiency_label", "original_relevance_label"])
sup_orig = find_col(["answer_support_label", "original_support_label"])
rel_rec = find_col([
    "audit_recommended_relevance_label",
    "recommended_relevance_label",
    "recommended_evidence_sufficiency_label"
])
sup_rec = find_col([
    "audit_recommended_support_label",
    "recommended_support_label",
    "recommended_answer_support_label"
])
decision_col = find_col(["audit_decision", "review_decision"])
rationale_col = find_col([
    "reference_audit_rationale", "audit_rationale", "reviewer_notes"
])

print("\nDetected columns:")
print({
    "candidate_id": id_col,
    "original_relevance": rel_orig,
    "original_support": sup_orig,
    "recommended_relevance": rel_rec,
    "recommended_support": sup_rec,
    "decision": decision_col,
    "rationale": rationale_col
})

assert id_col is not None, "Candidate ID column missing"
assert len(df) == 31, f"Expected 31 rows, got {len(df)}"
assert df[id_col].nunique() == 31, "Duplicate candidate IDs found"
assert df[id_col].notna().all(), "Missing candidate IDs"

assert rel_orig and sup_orig, "Could not identify original label columns"
assert rel_rec and sup_rec, "Could not identify recommended label columns"

changes = df[
    (df[rel_orig] != df[rel_rec]) |
    (df[sup_orig] != df[sup_rec])
]

print("\nRecommended changes:", len(changes))
cols = [id_col, rel_orig, sup_orig, rel_rec, sup_rec]
if decision_col:
    cols.append(decision_col)
if rationale_col:
    cols.append(rationale_col)

print(changes[cols].to_string(index=False))
print("\nOriginal labels preserved in audit file:", True)
print("No files modified.")
```

```python

from pathlib import Path
import pandas as pd
import hashlib

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2/evidence_sufficiency_audit_v1/"
    "blinded_review_v1/two_stage_annotation_v1"
)

PILOT = BASE / "stage_a_pilot_sample_finalized.csv"
REMAINING = BASE / "stage_a_remaining_candidates_finalized.csv"
OUTPUT = BASE / "all_57_candidates_finalized_v1.csv"

assert PILOT.exists(), f"Missing pilot: {PILOT}"
assert REMAINING.exists(), f"Missing remaining set: {REMAINING}"
assert not OUTPUT.exists(), f"Output already exists: {OUTPUT}"

pilot = pd.read_csv(PILOT)
remaining = pd.read_csv(REMAINING)

# Normalize the two files to a common annotation schema.
def standardize(df, source_name):
    required = [
        "blind_candidate_id",
        "question_id",
        "final_relevance_label",
        "final_support_label",
    ]
    missing = [c for c in required if c not in df.columns]
    assert not missing, f"{source_name} missing columns: {missing}"

    out = df[required].copy()
    out["annotation_source"] = source_name
    return out

pilot_std = standardize(pilot, "finalized_pilot")
remaining_std = standardize(remaining, "finalized_remaining")
combined = pd.concat([pilot_std, remaining_std], ignore_index=True)

# Integrity checks.
assert len(pilot_std) == 26, f"Expected 26 pilot rows, got {len(pilot_std)}"
assert len(remaining_std) == 31, (
    f"Expected 31 remaining rows, got {len(remaining_std)}"
)
assert len(combined) == 57, f"Expected 57 rows, got {len(combined)}"
assert combined["blind_candidate_id"].notna().all()
assert combined["blind_candidate_id"].is_unique, (
    "Duplicate candidate IDs detected across the two sets"
)
assert combined["question_id"].notna().all()
assert combined["final_relevance_label"].isin([0, 1]).all()
assert combined["final_support_label"].isin([0, 1, 2, 3]).all()

# The support rubric implies that support 2 or 3 must also be relevant.
invalid = combined[
    combined["final_support_label"].isin([2, 3]) &
    (combined["final_relevance_label"] != 1)
]
assert invalid.empty, (
    "Rubric consistency issue: support 2/3 paired with relevance 0:\n" +
    invalid.to_string(index=False)
)

combined.to_csv(OUTPUT, index=False)
digest = hashlib.sha256(OUTPUT.read_bytes()).hexdigest()

print("Saved:", OUTPUT)
print("Total candidates:", len(combined))
print("Unique candidate IDs:", combined["blind_candidate_id"].nunique())
print("Pilot:", len(pilot_std), "| Remaining:", len(remaining_std))
print("\nRelevance counts:")
print(combined["final_relevance_label"].value_counts().sort_index())
print("\nSupport counts:")
print(combined["final_support_label"].value_counts().sort_index())
print("\nRows per question:")
print(combined.groupby("question_id").size().to_string())
print("\nSHA-256:", digest)
print("\nConsolidation validation PASSED.")
```

```python

from pathlib import Path
import pandas as pd
import hashlib

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2/evidence_sufficiency_audit_v1/"
    "blinded_review_v1"
)
ANN_DIR = BASE / "two_stage_annotation_v1"

ANNOTATIONS = ANN_DIR / "all_57_candidates_finalized_v1.csv"
MAPPING = ANN_DIR / "candidate_rank_mapping_copy.csv"
OUTPUT = ANN_DIR / "annotated_retrieval_occurrences_v1.csv"

assert ANNOTATIONS.exists()
assert MAPPING.exists()
assert not OUTPUT.exists(), f"Output already exists: {OUTPUT}"

ann = pd.read_csv(ANNOTATIONS)
mapping = pd.read_csv(MAPPING)

ID = "blind_candidate_id"
assert ann[ID].is_unique and ann[ID].notna().all()
assert len(ann) == 57
assert len(mapping) == 90
assert mapping[ID].notna().all()

# Check every candidate's question_id agrees between files.
question_check = mapping[[ID, "question_id"]].merge(
    ann[[ID, "question_id"]].rename(
        columns={"question_id": "annotation_question_id"}
    ),
    on=ID,
    how="left",
    validate="many_to_one",
    indicator=True,
)

assert (question_check["_merge"] == "both").all()
mismatch = question_check[
    question_check["question_id"] != question_check["annotation_question_id"]
]
assert mismatch.empty, (
    "Question ID mismatch between mapping and annotations:\n"
    + mismatch.to_string(index=False)
)

# Add only labels; retain question_id and all ranking columns from mapping.
labels = ann[[
    ID,
    "final_relevance_label",
    "final_support_label",
]]

joined = mapping.merge(
    labels,
    on=ID,
    how="left",
    validate="many_to_one",
    indicator=True,
)

assert len(joined) == 90
assert (joined["_merge"] == "both").all()
assert joined["final_relevance_label"].isin([0, 1]).all()
assert joined["final_support_label"].isin([0, 1, 2, 3]).all()

joined = joined.drop(columns="_merge")
joined.to_csv(OUTPUT, index=False)

print("JOIN PASSED")
print("Output:", OUTPUT)
print("Rows:", len(joined))
print("Unique candidates:", joined[ID].nunique())
print("\nOccurrences per method:")
print(joined.groupby("method").size().to_string())
print("\nQuestion IDs match:", mismatch.empty)
print("Relevance counts across occurrences:")
print(joined["final_relevance_label"].value_counts().sort_index())
print("\nSupport counts across occurrences:")
print(joined["final_support_label"].value_counts().sort_index())
print("\nSHA-256:", hashlib.sha256(OUTPUT.read_bytes()).hexdigest())
```

```python

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2/evidence_sufficiency_audit_v1/"
    "blinded_review_v1/two_stage_annotation_v1"
)

INPUT = BASE / "annotated_retrieval_occurrences_v1.csv"
PER_Q_OUT = BASE / "evidence_metrics_per_question_v1.csv"
AGG_OUT = BASE / "evidence_metrics_aggregate_v1.csv"
REPORT_OUT = BASE / "evidence_metrics_manifest_v1.json"

assert INPUT.exists(), f"Missing joined file: {INPUT}"
for p in [PER_Q_OUT, AGG_OUT, REPORT_OUT]:
    assert not p.exists(), f"Output already exists; inspect before rerunning: {p}"

df = pd.read_csv(INPUT)

required = [
    "question_id", "method", "rank",
    "final_relevance_label", "final_support_label"
]
missing = [c for c in required if c not in df.columns]
assert not missing, f"Missing columns: {missing}"

df["rank"] = pd.to_numeric(df["rank"], errors="raise").astype(int)
df["final_relevance_label"] = pd.to_numeric(
    df["final_relevance_label"], errors="raise"
).astype(int)
df["final_support_label"] = pd.to_numeric(
    df["final_support_label"], errors="raise"
).astype(int)

assert len(df) == 90
assert set(df["method"].unique()) == {"BM25", "DENSE", "HYBRID_RRF"}
assert df["question_id"].nunique() == 10
assert df["final_relevance_label"].isin([0, 1]).all()
assert df["final_support_label"].isin([0, 1, 2, 3]).all()
assert df.groupby(["method", "question_id"]).size().eq(3).all()
assert df.groupby(["method", "question_id"])["rank"].apply(
    lambda s: set(s) == {1, 2, 3}
).all()

# Support 2/3 should not be assigned to an irrelevant candidate.
bad = df[
    df["final_support_label"].isin([2, 3]) &
    (df["final_relevance_label"] != 1)
]
assert bad.empty, (
    "Rubric inconsistency: support 2/3 with relevance 0:\n" +
    bad.to_string(index=False)
)

def dcg(gains):
    gains = np.asarray(gains, dtype=float)
    discounts = np.log2(np.arange(2, len(gains) + 2))
    return float(np.sum(gains / discounts))

def ndcg(gains):
    actual = dcg(gains)
    ideal = dcg(sorted(gains, reverse=True))
    return actual / ideal if ideal > 0 else np.nan

rows = []

for (method, qid), group in df.groupby(
    ["method", "question_id"], sort=True
):
    g = group.sort_values("rank")
    rel = g["final_relevance_label"].to_numpy()
    support = g["final_support_label"].to_numpy()

    row = {
        "method": method,
        "question_id": qid,
        "n_results": len(g),
        "Relevant@1": int(rel[:1].sum() > 0),
        "Relevant@3": int(rel[:3].sum() > 0),
        "RelevancePrecision@3": float(rel[:3].mean()),
        "Relevance_nDCG@3": ndcg(rel[:3]),
        "SufficientEvidence@1": int((support[:1] == 3).any()),
        "SufficientEvidence@3": int((support[:3] == 3).any()),
        "AnyPartialOrSufficient@3": int((support[:3] >= 2).any()),
        "Support3Fraction@3": float((support[:3] == 3).mean()),
        "MeanSupportScore@3": float(support[:3].mean()),
        "NormalizedSupportDCG@3": ndcg(support[:3] / 3.0),
        "IrrelevantFraction@3": float((rel[:3] == 0).mean()),
    }

    sufficient_ranks = g.loc[g["final_support_label"] == 3, "rank"]
    row["FirstSufficientRank"] = (
        int(sufficient_ranks.min()) if len(sufficient_ranks) else np.nan
    )

    rows.append(row)

per_q = pd.DataFrame(rows)

metric_cols = [
    "Relevant@1", "Relevant@3", "RelevancePrecision@3",
    "Relevance_nDCG@3", "SufficientEvidence@1",
    "SufficientEvidence@3", "AnyPartialOrSufficient@3",
    "Support3Fraction@3", "MeanSupportScore@3",
    "NormalizedSupportDCG@3", "IrrelevantFraction@3",
    "FirstSufficientRank"
]

# Macro average: each question contributes equally.
agg = per_q.groupby("method")[metric_cols].mean().reset_index()
agg.insert(1, "n_questions", per_q.groupby("method")["question_id"].nunique().values)

per_q.to_csv(PER_Q_OUT, index=False)
agg.to_csv(AGG_OUT, index=False)

manifest = {
    "input_file": str(INPUT),
    "input_sha256": hashlib.sha256(INPUT.read_bytes()).hexdigest(),
    "per_question_output": str(PER_Q_OUT),
    "aggregate_output": str(AGG_OUT),
    "n_occurrences": int(len(df)),
    "n_unique_questions": int(df["question_id"].nunique()),
    "methods": sorted(df["method"].unique().tolist()),
    "annotated_depth": 3,
    "aggregation": "Macro-average across questions",
    "annotation_status": "AI-assisted provisional labels",
    "limitations": [
        "Only top-3 occurrences are annotated in this mapping.",
        "Support labels are provisional and reference-aware.",
        "These metrics do not establish generated-answer correctness.",
        "Support nDCG uses support labels 0-3 as graded gains."
    ]
}
REPORT_OUT.write_text(json.dumps(manifest, indent=2), encoding="utf-8")

print("METRICS CALCULATED")
print("\nAggregate metrics by method:")
print(agg.round(4).to_string(index=False))
print("\nPer-question file:", PER_Q_OUT)
print("Aggregate file:", AGG_OUT)
print("Manifest:", REPORT_OUT)
print("\nInput SHA-256:", manifest["input_sha256"])
```

```python

from pathlib import Path
import pandas as pd

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2/evidence_sufficiency_audit_v1/"
    "blinded_review_v1/two_stage_annotation_v1"
)

PER_Q = BASE / "evidence_metrics_per_question_v1.csv"
AGG = BASE / "evidence_metrics_aggregate_v1.csv"

OUT_ALL = BASE / "question_level_method_comparison_v1.csv"
OUT_DISAGREE = BASE / "sufficient_evidence_disagreements_v1.csv"

assert PER_Q.exists()
assert AGG.exists()
assert not OUT_ALL.exists(), f"Output already exists: {OUT_ALL}"
assert not OUT_DISAGREE.exists(), f"Output already exists: {OUT_DISAGREE}"

df = pd.read_csv(PER_Q)

required = [
    "question_id", "method",
    "Relevant@1", "Relevance_nDCG@3",
    "SufficientEvidence@1", "SufficientEvidence@3",
    "MeanSupportScore@3", "IrrelevantFraction@3"
]
missing = [c for c in required if c not in df.columns]
assert not missing, f"Missing columns: {missing}"

assert df.groupby(["question_id", "method"]).size().eq(1).all()
assert df["question_id"].nunique() == 10
assert df["method"].nunique() == 3

# Create one row per question with each method's key results.
wide = df.pivot(
    index="question_id",
    columns="method",
    values=[
        "Relevant@1",
        "Relevance_nDCG@3",
        "SufficientEvidence@1",
        "SufficientEvidence@3",
        "MeanSupportScore@3",
        "IrrelevantFraction@3",
    ],
)
wide.columns = [f"{metric}_{method}" for metric, method in wide.columns]
wide = wide.reset_index()

# Difference in sufficient-evidence coverage between hybrid and dense.
wide["Hybrid_minus_Dense_SufficientEvidence@3"] = (
    wide["SufficientEvidence@3_HYBRID_RRF"]
    - wide["SufficientEvidence@3_DENSE"]
)

wide["Hybrid_minus_BM25_SufficientEvidence@3"] = (
    wide["SufficientEvidence@3_HYBRID_RRF"]
    - wide["SufficientEvidence@3_BM25"]
)

wide.to_csv(OUT_ALL, index=False)

disagree = wide[
    (wide["SufficientEvidence@3_HYBRID_RRF"] !=
     wide["SufficientEvidence@3_DENSE"]) |
    (wide["SufficientEvidence@3_HYBRID_RRF"] !=
     wide["SufficientEvidence@3_BM25"])
].copy()

disagree.to_csv(OUT_DISAGREE, index=False)

print("Question-level comparison saved:", OUT_ALL)
print("Questions with method disagreement saved:", OUT_DISAGREE)
print("Questions with any disagreement:", len(disagree))
print("\nDisagreement details:")
if len(disagree):
    cols = [
        "question_id",
        "SufficientEvidence@3_BM25",
        "SufficientEvidence@3_DENSE",
        "SufficientEvidence@3_HYBRID_RRF",
        "Relevance_nDCG@3_BM25",
        "Relevance_nDCG@3_DENSE",
        "Relevance_nDCG@3_HYBRID_RRF",
    ]
    print(disagree[cols].to_string(index=False))
else:
    print("No disagreements in sufficient-evidence coverage.")

print("\nRead-only analysis complete; source metrics unchanged.")
```

```python

from pathlib import Path
import pandas as pd

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2"
)
ANN_DIR = (
    ROOT / "evidence_sufficiency_audit_v1" /
    "blinded_review_v1" / "two_stage_annotation_v1"
)

RAW = ROOT / "baseline_v2_pilot" / "raw_retrieval_results.csv"
JOINED = ANN_DIR / "annotated_retrieval_occurrences_v1.csv"
OUTPUT = ANN_DIR / "disagreement_passages_with_text_v1.csv"

assert RAW.exists(), f"Raw retrieval file not found: {RAW}"
assert JOINED.exists(), f"Joined file not found: {JOINED}"
assert not OUTPUT.exists(), f"Output already exists: {OUTPUT}"

raw = pd.read_csv(RAW)
joined = pd.read_csv(JOINED)

questions = ["V2P003", "V2P005", "V2P007", "V2P008", "V2P009"]

print("Raw columns:", raw.columns.tolist())
print("Joined columns:", joined.columns.tolist())

# Check the expected raw result schema.
required_raw = [
    "question_id", "method", "rank", "chunk_index", "chunk_text"
]
required_joined = [
    "question_id", "method", "rank", "chunk_index",
    "final_relevance_label", "final_support_label"
]
assert all(c in raw.columns for c in required_raw)
assert all(c in joined.columns for c in required_joined)

# Select the top-three occurrences for the disagreement questions.
subset = joined[
    joined["question_id"].isin(questions) &
    joined["rank"].between(1, 3)
].copy()

# Match the original chunk text using question, method, rank and chunk index.
keys = ["question_id", "method", "rank", "chunk_index"]

raw_subset = raw[raw["question_id"].isin(questions)].copy()
raw_subset = raw_subset[raw_subset["rank"].between(1, 3)]

assert not raw_subset.duplicated(keys).any(), "Duplicate raw ranking keys"
assert not subset.duplicated(keys).any(), "Duplicate joined ranking keys"

text_cols = keys + ["chunk_text"]
for col in ["question", "question_type", "source_file", "pages"]:
    if col in raw_subset.columns:
        text_cols.append(col)

text_source = raw_subset[text_cols].copy()

inspection = subset.merge(
    text_source,
    on=keys,
    how="left",
    validate="one_to_one",
    indicator=True,
    suffixes=("", "_raw")
)

assert len(inspection) == 45, f"Expected 45 rows, got {len(inspection)}"
assert (inspection["_merge"] == "both").all(), (
    "Some ranked occurrences did not match raw text:\n" +
    inspection.loc[inspection["_merge"] != "both", keys].to_string(index=False)
)
assert inspection["chunk_text"].notna().all(), "Some chunk text is missing"

inspection = inspection.drop(columns="_merge")
inspection.to_csv(OUTPUT, index=False)

# Independently check sufficient-evidence coverage using the joined labels.
coverage = (
    inspection.assign(is_sufficient=inspection["final_support_label"].eq(3))
    .groupby(["question_id", "method"])["is_sufficient"]
    .any()
    .unstack("method")
    .astype(int)
    .reset_index()
)

print("\nSufficient-evidence coverage recalculated from joined rows:")
print(coverage.to_string(index=False))

print("\nSaved full passage inspection:", OUTPUT)
print("\nFull passages (truncated to 1,200 characters each):")
for _, row in inspection.sort_values(
    ["question_id", "method", "rank"]
).iterrows():
    print("\n" + "=" * 90)
    print(
        f"{row['question_id']} | {row['method']} | rank={row['rank']} | "
        f"chunk={row['chunk_index']} | relevance={row['final_relevance_label']} | "
        f"support={row['final_support_label']}"
    )
    print(str(row["chunk_text"])[:1200])
```

```python
# ============================================================
# TOP-3 RETRIEVAL REDUNDANCY AUDIT — V1
# Does not modify existing retrieval results or labels
# ============================================================

from pathlib import Path
from itertools import combinations
import pandas as pd
import json
import re
import hashlib

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2"
)

INPUT_FILE = next(BASE.rglob("annotated_retrieval_occurrences_v1.csv"))
OUTPUT_DIR = INPUT_FILE.parent / "redundancy_audit_v1"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(INPUT_FILE)
df["rank"] = pd.to_numeric(df["rank"], errors="raise")

required = [
    "question_id", "method", "rank", "chunk_index",
    "retrieved_source_file", "retrieved_pages", "chunk_hash"
]
missing = [c for c in required if c not in df.columns]
assert not missing, f"Missing columns: {missing}"

top3 = df[df["rank"] <= 3].copy()

def parse_pages(value):
    if pd.isna(value):
        return set()

    s = str(value).strip()

    try:
        obj = json.loads(s)
        if isinstance(obj, list):
            return {int(x) for x in obj if str(x).isdigit()}
    except (ValueError, TypeError):
        pass

    pages = set()
    for part in re.findall(r"\d+\s*-\s*\d+|\d+", s):
        if "-" in part:
            a, b = map(int, re.split(r"\s*-\s*", part))
            pages.update(range(min(a, b), max(a, b) + 1))
        else:
            pages.add(int(part))
    return pages

def norm_source(value):
    if pd.isna(value):
        return ""
    return str(value).strip().replace("\\", "/").lower()

summary_rows = []
pair_rows = []

for (qid, method), group in top3.groupby(["question_id", "method"]):
    group = group.sort_values("rank").reset_index(drop=True)

    hashes = [
        str(x).strip() if pd.notna(x) else ""
        for x in group["chunk_hash"]
    ]
    pages = [parse_pages(x) for x in group["retrieved_pages"]]
    sources = [norm_source(x) for x in group["retrieved_source_file"]]

    exact_pairs = 0
    overlap_pairs = 0
    same_page_pairs = 0
    shared_page_instances = 0

    for i, j in combinations(range(len(group)), 2):
        same_hash = bool(hashes[i]) and hashes[i] == hashes[j]
        same_source = bool(sources[i]) and sources[i] == sources[j]
        shared = pages[i] & pages[j]

        if same_hash:
            exact_pairs += 1

        if same_source and shared:
            overlap_pairs += 1
            shared_page_instances += len(shared)

        if same_source and pages[i] == pages[j] and pages[i]:
            same_page_pairs += 1

        if same_hash or (same_source and shared):
            pair_rows.append({
                "question_id": qid,
                "method": method,
                "rank_1": int(group.loc[i, "rank"]),
                "rank_2": int(group.loc[j, "rank"]),
                "chunk_index_1": group.loc[i, "chunk_index"],
                "chunk_index_2": group.loc[j, "chunk_index"],
                "exact_duplicate": same_hash,
                "same_source": same_source,
                "pages_1": sorted(pages[i]),
                "pages_2": sorted(pages[j]),
                "shared_pages": sorted(shared),
                "shared_page_count": len(shared),
                "source_file": group.loc[i, "retrieved_source_file"]
                    if same_source else None,
            })

    summary_rows.append({
        "question_id": qid,
        "method": method,
        "top3_rows": len(group),
        "unique_chunk_hashes": len(set(h for h in hashes if h)),
        "exact_duplicate_pairs": exact_pairs,
        "same_source_page_overlap_pairs": overlap_pairs,
        "same_source_same_page_pairs": same_page_pairs,
        "shared_page_instances_across_pairs": shared_page_instances,
    })

summary = pd.DataFrame(summary_rows)
pairs = pd.DataFrame(pair_rows)

summary_path = OUTPUT_DIR / "top3_redundancy_summary_v1.csv"
pairs_path = OUTPUT_DIR / "top3_redundancy_pairs_v1.csv"

summary.to_csv(summary_path, index=False)
pairs.to_csv(pairs_path, index=False)

print("AUDIT COMPLETE")
print("Input:", INPUT_FILE)
print("Input occurrences:", len(df))
print("Top-3 occurrences:", len(top3))
print("Question-method groups:", len(summary))

print("\nOVERALL COUNTS BY METHOD")
display(
    summary.groupby("method", as_index=False).agg(
        question_groups=("question_id", "nunique"),
        exact_duplicate_pairs=("exact_duplicate_pairs", "sum"),
        same_source_page_overlap_pairs=(
            "same_source_page_overlap_pairs", "sum"
        ),
        same_source_same_page_pairs=(
            "same_source_same_page_pairs", "sum"
        ),
    )
)

print("\nFLAGGED PAIRS")
if pairs.empty:
    print("No exact duplicates or same-source page overlaps detected.")
else:
    display(
        pairs.sort_values(
            ["question_id", "method", "rank_1", "rank_2"]
        ).reset_index(drop=True)
    )

print("\nSaved files:")
print(summary_path)
print(pairs_path)
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2"
)

AUDIT = (
    ROOT / "evidence_sufficiency_audit_v1" / "blinded_review_v1"
    / "two_stage_annotation_v1"
)

INPUT = AUDIT / "annotated_retrieval_occurrences_v2.csv"
V1_PER_Q = AUDIT / "evidence_metrics_per_question_v1.csv"
V1_AGG = AUDIT / "evidence_metrics_aggregate_v1.csv"

OUT = AUDIT / "evidence_metrics_v2_corrected"
OUT.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(INPUT)

required = {
    "question_id", "method", "rank",
    "final_relevance_label", "final_support_label"
}
assert required.issubset(df.columns), df.columns.tolist()
assert len(df) == 90
assert df.groupby(["question_id", "method"]).size().eq(3).all()
assert df["final_relevance_label"].notna().all()
assert df["final_support_label"].notna().all()

# Every question-method group must have ranks 1, 2, 3.
for _, group in df.groupby(["question_id", "method"]):
    assert sorted(group["rank"].astype(int).tolist()) == [1, 2, 3]

def dcg(gains):
    gains = np.asarray(gains, dtype=float)
    discounts = np.log2(np.arange(2, len(gains) + 2))
    return float(np.sum(gains / discounts))

def ndcg_at_3(gains):
    ideal = sorted(gains, reverse=True)
    ideal_dcg = dcg(ideal)
    if ideal_dcg == 0:
        return np.nan
    return dcg(gains) / ideal_dcg

rows = []

for (qid, method), group in df.groupby(["question_id", "method"]):
    g = group.sort_values("rank")
    rel = g["final_relevance_label"].astype(int).tolist()
    sup = g["final_support_label"].astype(int).tolist()

    rows.append({
        "question_id": qid,
        "method": method,
        "RelevantAt1": float(rel[0] == 1),
        "RelevantAt3": float(any(x == 1 for x in rel)),
        "RelevancePrecisionAt3": float(np.mean(rel)),
        "Relevance_nDCGAt3": ndcg_at_3(rel),
        "SufficientEvidenceAt1": float(sup[0] == 3),
        "SufficientEvidenceAt3": float(any(x == 3 for x in sup)),
        "AnyPartialOrSufficientAt3": float(any(x > 0 for x in sup)),
        "Support3FractionAt3": float(np.mean([x == 3 for x in sup])),
        "MeanSupportScoreAt3": float(np.mean(sup)),
        "NormalizedSupportDCGAt3": ndcg_at_3(sup),
        "IrrelevantFractionAt3": float(np.mean([x == 0 for x in rel])),
        "FirstSufficientRank": next(
            (int(rank) for rank, score in zip(g["rank"], sup) if score == 3),
            np.nan
        )
    })

per_q = pd.DataFrame(rows).sort_values(["method", "question_id"])

metric_cols = [
    c for c in per_q.columns
    if c not in ["question_id", "method"]
]
agg = (
    per_q.groupby("method")[metric_cols]
    .mean(numeric_only=True)
    .reset_index()
)
agg.insert(1, "n_questions", per_q.groupby("method")["question_id"].nunique().values)

per_q_path = OUT / "evidence_metrics_per_question_v2.csv"
agg_path = OUT / "evidence_metrics_aggregate_v2.csv"
comparison_path = OUT / "aggregate_v1_vs_v2.csv"
manifest_path = OUT / "manifest_v2.json"

per_q.to_csv(per_q_path, index=False)
agg.to_csv(agg_path, index=False)

# Compare with v1 using matching metric names where possible.
if V1_AGG.exists():
    old = pd.read_csv(V1_AGG)
    print("V1 aggregate columns:", old.columns.tolist())
    print("V2 aggregate columns:", agg.columns.tolist())

    # Handle either long-form or wide-form v1 aggregates.
    if "method" in old.columns:
        old_method_col = "method"
        common = [
            c for c in metric_cols
            if c in old.columns
        ]
        comp = old[["method"] + common].merge(
            agg[["method"] + common],
            on="method", how="outer", suffixes=("_v1", "_v2")
        )
        for c in common:
            comp[f"{c}_delta"] = comp[f"{c}_v2"] - comp[f"{c}_v1"]
        comp.to_csv(comparison_path, index=False)
        print("\n=== V1 vs V2 aggregate comparison ===")
        print(comp.to_string(index=False))
    else:
        print("V1 aggregate is not in expected long format; inspect its columns above.")
else:
    print("V1 aggregate file not found; comparison skipped.")

input_hash = hashlib.sha256(INPUT.read_bytes()).hexdigest()
manifest = {
    "version": "v2_corrected",
    "input_file": str(INPUT),
    "input_sha256": input_hash,
    "rows": int(len(df)),
    "unique_candidates": int(df["blind_candidate_id"].nunique())
        if "blind_candidate_id" in df.columns else None,
    "questions": int(df["question_id"].nunique()),
    "methods": sorted(df["method"].unique().tolist()),
    "rank_cutoff": 3,
    "relevance_ndcg_gain": "binary relevance, gain=label",
    "support_ndcg_gain": "support score 0-3, gain=label",
    "undefined_ndcg": "NaN when ideal DCG is zero",
    "note": "Corrected annotation version; original v1 artifacts preserved."
}
manifest_path.write_text(json.dumps(manifest, indent=2))

print("\n=== V2 corrected aggregate metrics ===")
print(agg.round(4).to_string(index=False))

print("\nSaved:")
print(per_q_path)
print(agg_path)
print(comparison_path if comparison_path.exists() else "(no comparison file)")
print(manifest_path)
print("Input SHA-256:", input_hash)
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2"
)

AUDIT = (
    ROOT / "evidence_sufficiency_audit_v1" / "blinded_review_v1"
    / "two_stage_annotation_v1"
)

BENCHMARK = ROOT / "NCERT_RAG_Benchmark_V2_Pilot.xlsx"
RAW_PATH = ROOT / "baseline_v2_pilot" / "raw_retrieval_results.csv"
LABEL_PATH = AUDIT / "annotated_retrieval_occurrences_v2.csv"

OUT = AUDIT / "question_level_audit_v1"
OUT.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------
# 1. Load the frozen benchmark, raw retrieval, and v2 labels
# ---------------------------------------------------------
assert BENCHMARK.exists(), f"Missing benchmark: {BENCHMARK}"
assert RAW_PATH.exists(), f"Missing raw retrieval: {RAW_PATH}"
assert LABEL_PATH.exists(), f"Missing corrected labels: {LABEL_PATH}"

xls = pd.ExcelFile(BENCHMARK)
benchmark = None
benchmark_sheet = None

for sheet in xls.sheet_names:
    candidate = pd.read_excel(BENCHMARK, sheet_name=sheet)
    candidate.columns = candidate.columns.astype(str).str.strip()
    required = {
        "question_id", "question", "expected_answer",
        "gold_pages", "evidence_excerpt"
    }
    if required.issubset(candidate.columns):
        benchmark = candidate.copy()
        benchmark_sheet = sheet
        break

assert benchmark is not None, (
    "Could not find a benchmark sheet with question_id, question, "
    "expected_answer, gold_pages, and evidence_excerpt. "
    f"Available sheets: {xls.sheet_names}"
)

raw = pd.read_csv(RAW_PATH)
labels = pd.read_csv(LABEL_PATH)

print("Benchmark sheet:", benchmark_sheet)
print("Benchmark questions:", len(benchmark))
print("Raw retrieval rows:", len(raw))
print("Corrected annotation rows:", len(labels))

# ---------------------------------------------------------
# 2. Validate IDs and join corrected labels to raw passages
# ---------------------------------------------------------
assert benchmark["question_id"].is_unique
assert len(benchmark) == 10
assert len(raw) == 300
assert len(labels) == 90

keys = ["question_id", "method", "rank", "chunk_index"]

assert set(keys).issubset(raw.columns)
assert set(keys).issubset(labels.columns)
assert {"final_relevance_label", "final_support_label"}.issubset(
    labels.columns
)

# Verify the annotation rows correspond to the raw retrieved chunks.
check = raw[keys].merge(
    labels[keys + ["final_relevance_label", "final_support_label"]],
    on=keys,
    how="left",
    validate="one_to_one"
)
assert len(check) == 300
assert check["final_support_label"].notna().sum() == 90

# Keep only the annotated top-three retrieval occurrences.
detail = raw.merge(
    labels[
        keys + [
            "blind_candidate_id",
            "final_relevance_label",
            "final_support_label"
        ]
    ],
    on=keys,
    how="inner",
    validate="one_to_one"
)

assert len(detail) == 90
assert detail.groupby(["question_id", "method"]).size().eq(3).all()

detail = detail.merge(
    benchmark[
        [
            "question_id", "expected_answer", "evidence_excerpt",
            "gold_pages", "source_file", "question_type",
            "answerability", "benchmark_status", "annotation_status"
        ]
    ],
    on="question_id",
    how="left",
    validate="many_to_one",
    suffixes=("", "_benchmark")
)

assert detail["expected_answer"].notna().all()

# ---------------------------------------------------------
# 3. Add explicit review notes based on our passage inspection
# ---------------------------------------------------------
issues = {
    "V2P001": (
        "GOOD BASELINE: all three methods retrieve chunk 76 at rank 1; "
        "it explicitly states what Anandi wanted to paint."
    ),
    "V2P002": (
        "PARTIAL TOP-1: chunk 76 provides story context but not the "
        "reason for leaving yellow. The explicit answer is in chunk 77, "
        "which was not in any method's top 3 for this question."
    ),
    "V2P003": (
        "COMPLEMENTARY CHUNKS: BM25 and Hybrid retrieve chunk 77 with "
        "support label 3; Dense misses that fully sufficient passage "
        "in its top 3."
    ),
    "V2P004": (
        "WORDING/GOLD-EVIDENCE REVIEW: the question can refer to either "
        "drawing cloud shapes (Class 1, page 131) or watching clouds "
        "and reporting observations (Class 2). Clarify which activity "
        "is intended before freezing the benchmark."
    ),
    "V2P005": (
        "HYBRID EVIDENCE SUCCESS: Hybrid retrieves chunk 78 at rank 3, "
        "which contains the sunrise/sunset/night observation activity. "
        "BM25 and Dense do not retrieve a support-3 passage in their top 3."
    ),
    "V2P006": (
        "QUESTION WORDING ISSUE: 'before she colours the flowers' "
        "conflicts with the expected answer, which describes the rainbow "
        "disappearing after she colours the flowers and Sun. Rewrite the "
        "question to preserve the intended chronology."
    ),
    "V2P007": (
        "GOOD EVIDENCE COVERAGE: Dense and Hybrid retrieve chunk 77 "
        "with the colours and the explanation for reserving yellow "
        "for the Sun."
    ),
    "V2P008": (
        "REDUNDANCY CHECK: the top 3 include overlapping chunks from "
        "the same theatre role-play section. Do not count overlapping "
        "chunks as independent evidence."
    ),
    "V2P009": (
        "METHOD DIFFERENCE: Dense retrieves chunk 69, which contains "
        "the food-sharing questions. The corrected v2 labels should "
        "be used; review whether partial neighboring chunks add anything."
    ),
    "V2P010": (
        "ANNOTATION CORRECTED: EV048, Hybrid rank 3, chunk 77, is now "
        "relevance=1/support=3 because it explicitly asks children to "
        "draw cloud shapes. Earlier v1 labels were incorrectly 0/0."
    ),
}

support_names = {
    0: "0 — No answer support",
    1: "1 — Related context only",
    2: "2 — Partial support",
    3: "3 — Fully sufficient",
}

method_order = ["BM25", "DENSE", "HYBRID_RRF"]

def clean_text(value):
    if pd.isna(value):
        return ""
    return " ".join(str(value).split())

def describe_passage(row, max_chars=500):
    text = clean_text(row["chunk_text"])
    if len(text) > max_chars:
        text = text[:max_chars].rstrip() + "…"
    return (
        f"Rank {int(row['rank'])} | Chunk {int(row['chunk_index'])} | "
        f"Support {int(row['final_support_label'])} | "
        f"Relevance {int(row['final_relevance_label'])} | "
        f"{clean_text(row['source_file'])} | pages {clean_text(row['pages'])}\n"
        f"{text}"
    )

# ---------------------------------------------------------
# 4. Build a compact method-level audit: 30 rows
# ---------------------------------------------------------
method_rows = []

for _, q in benchmark.iterrows():
    qid = q["question_id"]
    for method in method_order:
        group = detail[
            (detail["question_id"] == qid) &
            (detail["method"] == method)
        ].sort_values("rank")

        assert len(group) == 3, f"Expected top 3 for {qid} / {method}"

        record = {
            "question_id": qid,
            "question": q["question"],
            "question_type": q["question_type"],
            "expected_answer": q["expected_answer"],
            "gold_source_file": q["source_file"],
            "gold_pages": q["gold_pages"],
            "gold_evidence_excerpt": q["evidence_excerpt"],
            "method": method,
            "review_note": issues.get(qid, ""),
        }

        for rank in [1, 2, 3]:
            r = group[group["rank"] == rank].iloc[0]
            record[f"rank{rank}_chunk"] = int(r["chunk_index"])
            record[f"rank{rank}_source"] = r["source_file"]
            record[f"rank{rank}_pages"] = r["pages"]
            record[f"rank{rank}_relevance"] = int(
                r["final_relevance_label"]
            )
            record[f"rank{rank}_support"] = int(
                r["final_support_label"]
            )
            record[f"rank{rank}_support_meaning"] = support_names[
                int(r["final_support_label"])
            ]
            record[f"rank{rank}_passage"] = clean_text(r["chunk_text"])

        record["sufficient_in_top3"] = int(
            (group["final_support_label"] == 3).any()
        )
        record["relevant_in_top3"] = int(
            (group["final_relevance_label"] == 1).any()
        )
        method_rows.append(record)

method_audit = pd.DataFrame(method_rows)

# ---------------------------------------------------------
# 5. Build one row per question with all three methods
# ---------------------------------------------------------
summary_rows = []

for _, q in benchmark.iterrows():
    qid = q["question_id"]
    record = {
        "question_id": qid,
        "question": q["question"],
        "question_type": q["question_type"],
        "expected_answer": q["expected_answer"],
        "gold_source_file": q["source_file"],
        "gold_pages": q["gold_pages"],
        "gold_evidence_excerpt": q["evidence_excerpt"],
        "review_issue": issues.get(qid, ""),
    }

    for method in method_order:
        group = detail[
            (detail["question_id"] == qid) &
            (detail["method"] == method)
        ].sort_values("rank")

        lines = []
        for _, r in group.iterrows():
            passage = clean_text(r["chunk_text"])
            if len(passage) > 350:
                passage = passage[:350].rstrip() + "…"

            lines.append(
                f"R{int(r['rank'])} chunk {int(r['chunk_index'])} "
                f"[rel={int(r['final_relevance_label'])}, "
                f"support={int(r['final_support_label'])}; "
                f"{clean_text(r['source_file'])}, pages {clean_text(r['pages'])}]: "
                f"{passage}"
            )

        record[f"{method}_top3"] = "\n\n".join(lines)
        record[f"{method}_sufficient_in_top3"] = int(
            (group["final_support_label"] == 3).any()
        )

    summary_rows.append(record)

summary = pd.DataFrame(summary_rows)

# ---------------------------------------------------------
# 6. Save a versioned Excel workbook, CSVs, and manifest
# ---------------------------------------------------------
summary_csv = OUT / "question_level_audit_summary_v1.csv"
detail_csv = OUT / "question_method_passage_audit_v1.csv"
xlsx_path = OUT / "question_level_audit_v1.xlsx"
manifest_path = OUT / "manifest_v1.json"

summary.to_csv(summary_csv, index=False)
method_audit.to_csv(detail_csv, index=False)

with pd.ExcelWriter(xlsx_path, engine="openpyxl") as writer:
    summary.to_excel(writer, sheet_name="Question Summary", index=False)
    method_audit.to_excel(writer, sheet_name="Method Detail", index=False)

manifest = {
    "audit_version": "question_level_audit_v1",
    "benchmark_file": str(BENCHMARK),
    "benchmark_sheet": benchmark_sheet,
    "raw_retrieval_file": str(RAW_PATH),
    "corrected_occurrence_file": str(LABEL_PATH),
    "question_count": int(len(summary)),
    "method_detail_rows": int(len(method_audit)),
    "passage_occurrence_rows": int(len(detail)),
    "methods": method_order,
    "rank_cutoff": 3,
    "support_label_definitions": support_names,
    "notes": (
        "Uses corrected v2 annotation labels. Passage judgments are "
        "AI-assisted/reference-aware and exploratory, not independent "
        "human ground truth. Original benchmark and retrieval outputs "
        "were not modified."
    ),
    "corrected_occurrence_sha256": hashlib.sha256(
        LABEL_PATH.read_bytes()
    ).hexdigest(),
}

manifest_path.write_text(json.dumps(manifest, indent=2))

print("\nAUDIT CREATED SUCCESSFULLY")
print("==========================")
print("Questions:", len(summary))
print("Question-method rows:", len(method_audit))
print("Top-3 passage rows:", len(detail))
print("\nOutputs:")
print("Excel:", xlsx_path)
print("Summary CSV:", summary_csv)
print("Detailed CSV:", detail_csv)
print("Manifest:", manifest_path)

print("\nQuestion-level evidence coverage:")
coverage = method_audit.groupby("method")[
    "sufficient_in_top3"
].agg(["sum", "count", "mean"]).reset_index()
coverage["rate"] = coverage["mean"].map(lambda x: f"{x:.0%}")
print(coverage[["method", "sum", "count", "rate"]].to_string(index=False))

print("\nQuestions flagged for review:")
for qid in ["V2P004", "V2P006", "V2P010"]:
    print(f"{qid}: {issues[qid]}")

print("\nSHA-256 of Excel:", hashlib.sha256(xlsx_path.read_bytes()).hexdigest())
```

```python
from pathlib import Path
from openpyxl import load_workbook
from datetime import datetime
import hashlib
import json

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2"
)

SOURCE = ROOT / "NCERT_RAG_Benchmark_V2_Pilot.xlsx"
OUTPUT = ROOT / "NCERT_RAG_Benchmark_V2_Pilot_Review_v1.xlsx"

assert SOURCE.exists(), f"Frozen benchmark not found: {SOURCE}"
assert not OUTPUT.exists(), (
    f"{OUTPUT.name} already exists. "
    "To avoid overwriting a prior review, inspect it before proceeding."
)

wb = load_workbook(SOURCE)
sheet_name = "Benchmark_V2_Pilot"
assert sheet_name in wb.sheetnames, (
    f"Expected sheet {sheet_name!r}; found {wb.sheetnames}"
)

ws = wb[sheet_name]

# Find the header row and column positions.
header_row = None
headers = {}

for r in range(1, min(ws.max_row, 15) + 1):
    vals = {
        str(ws.cell(r, c).value).strip(): c
        for c in range(1, ws.max_column + 1)
        if ws.cell(r, c).value is not None
    }
    if "question_id" in vals and "question" in vals:
        header_row = r
        headers = vals
        break

assert header_row is not None, "Could not locate benchmark headers."

required_cols = {
    "question_id", "question", "expected_answer",
    "gold_pages", "evidence_excerpt", "question_type"
}
assert required_cols.issubset(headers), (
    f"Missing columns: {required_cols - set(headers)}"
)

def find_question_row(qid):
    matches = []
    for r in range(header_row + 1, ws.max_row + 1):
        if str(ws.cell(r, headers["question_id"]).value).strip() == qid:
            matches.append(r)
    assert len(matches) == 1, f"{qid}: expected one row, found {len(matches)}"
    return matches[0]

# Capture original values for a clear audit trail.
r004 = find_question_row("V2P004")
r006 = find_question_row("V2P006")

originals = {}
for qid, r in [("V2P004", r004), ("V2P006", r006)]:
    originals[qid] = {
        key: ws.cell(r, headers[key]).value
        for key in [
            "question", "expected_answer", "gold_pages",
            "evidence_excerpt", "question_type"
        ]
    }

# ------------------------------------------------------------
# V2P004: specify the cloud-drawing activity, not cloud watching
# ------------------------------------------------------------
ws.cell(r004, headers["question"]).value = (
    "What does the book ask children to draw after looking at the clouds?"
)

# ------------------------------------------------------------
# V2P006: correct chronology and gold evidence scope
# ------------------------------------------------------------
ws.cell(r006, headers["question"]).value = (
    "What happens to the rainbow after Anandi colours the flowers and the Sun?"
)
ws.cell(r006, headers["expected_answer"]).value = (
    "The rainbow disappears after Anandi gives colours to the flowers and the Sun."
)
ws.cell(r006, headers["gold_pages"]).value = "[130]"
ws.cell(r006, headers["question_type"]).value = "DIRECT"
ws.cell(r006, headers["evidence_excerpt"]).value = (
    "After giving colours to the flowers and the Sun, the rainbow was gone……"
)

# Mark changed records as review drafts where those columns exist.
for qid, r in [("V2P004", r004), ("V2P006", r006)]:
    if "benchmark_status" in headers:
        ws.cell(r, headers["benchmark_status"]).value = "REVIEW_DRAFT"
    if "annotation_status" in headers:
        ws.cell(r, headers["annotation_status"]).value = "PENDING_MANUAL_REVIEW"

# Add a change log to this new workbook only.
if "Review_Change_Log" in wb.sheetnames:
    del wb["Review_Change_Log"]

log = wb.create_sheet("Review_Change_Log")
log.append([
    "question_id", "field", "original_value", "proposed_value",
    "reason", "status"
])

changes = [
    (
        "V2P004", "question",
        originals["V2P004"]["question"],
        ws.cell(r004, headers["question"]).value,
        "Disambiguates cloud drawing from the separate cloud-observation activity.",
        "PROPOSED — requires manual review"
    ),
    (
        "V2P006", "question",
        originals["V2P006"]["question"],
        ws.cell(r006, headers["question"]).value,
        "Corrects the temporal contradiction in the original wording.",
        "PROPOSED — requires manual review"
    ),
    (
        "V2P006", "expected_answer",
        originals["V2P006"]["expected_answer"],
        ws.cell(r006, headers["expected_answer"]).value,
        "Aligns the answer with the revised question.",
        "PROPOSED — requires manual review"
    ),
    (
        "V2P006", "gold_pages",
        originals["V2P006"]["gold_pages"],
        ws.cell(r006, headers["gold_pages"]).value,
        "Page 130 directly states that the rainbow was gone after colouring.",
        "PROPOSED — verify against source PDF"
    ),
    (
        "V2P006", "question_type",
        originals["V2P006"]["question_type"],
        ws.cell(r006, headers["question_type"]).value,
        "A single-page factual question no longer requires MULTI_PAGE evidence.",
        "PROPOSED — requires manual review"
    ),
    (
        "V2P006", "evidence_excerpt",
        originals["V2P006"]["evidence_excerpt"],
        ws.cell(r006, headers["evidence_excerpt"]).value,
        "Uses the direct passage supporting the revised answer.",
        "PROPOSED — verify against source PDF"
    ),
]

for change in changes:
    log.append([
        change[0],
        change[1],
        str(change[2]) if change[2] is not None else "",
        str(change[3]) if change[3] is not None else "",
        change[4],
        change[5],
    ])

log.freeze_panes = "A2"
log.auto_filter.ref = log.dimensions
for col, width in {
    "A": 16, "B": 22, "C": 55, "D": 65, "E": 75, "F": 38
}.items():
    log.column_dimensions[col].width = width

# Save and verify the new workbook.
wb.save(OUTPUT)

check_wb = load_workbook(OUTPUT, data_only=True)
check_ws = check_wb[sheet_name]

def get_value(qid, field):
    r = find_question_row(qid)
    return check_ws.cell(r, headers[field]).value

assert get_value("V2P004", "question") == (
    "What does the book ask children to draw after looking at the clouds?"
)
assert get_value("V2P006", "gold_pages") == "[130]"
assert "after Anandi colours" in get_value("V2P006", "question")
assert "Review_Change_Log" in check_wb.sheetnames

# Confirm the original workbook has not been changed by this cell.
source_hash = hashlib.sha256(SOURCE.read_bytes()).hexdigest()
review_hash = hashlib.sha256(OUTPUT.read_bytes()).hexdigest()

manifest = {
    "review_version": "NCERT_RAG_Benchmark_V2_Pilot_Review_v1",
    "created_at_local": datetime.now().isoformat(timespec="seconds"),
    "source_file": str(SOURCE),
    "review_file": str(OUTPUT),
    "source_sha256": source_hash,
    "review_sha256": review_hash,
    "modified_question_ids": ["V2P004", "V2P006"],
    "status": "DRAFT_PENDING_MANUAL_REVIEW",
    "retrieval_rerun": False,
    "original_benchmark_overwritten": False,
}
manifest_path = ROOT / "NCERT_RAG_Benchmark_V2_Pilot_Review_v1_manifest.json"
manifest_path.write_text(json.dumps(manifest, indent=2))

print("REVIEW BENCHMARK CREATED")
print("========================")
print("V2P004:", get_value("V2P004", "question"))
print("V2P006:", get_value("V2P006", "question"))
print("V2P006 expected answer:", get_value("V2P006", "expected_answer"))
print("V2P006 gold pages:", get_value("V2P006", "gold_pages"))
print("V2P006 question type:", get_value("V2P006", "question_type"))
print("\nOriginal benchmark preserved:", SOURCE.exists())
print("Original SHA-256:", source_hash)
print("Review workbook:", OUTPUT)
print("Review SHA-256:", review_hash)
print("Manifest:", manifest_path)
print("\nStatus: DRAFT — review the revised rows against the source PDF before freezing.")
print("No retrieval results or metrics were changed.")
```

```python
from pathlib import Path
from openpyxl import load_workbook
import hashlib
import json

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/"
    "retrieval_evaluation/benchmark_v2"
)

DRAFT = ROOT / "NCERT_RAG_Benchmark_V2_Pilot_Review_v1.xlsx"
FINAL = ROOT / "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
MANIFEST = ROOT / "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1_manifest.json"

assert DRAFT.exists(), f"Review draft missing: {DRAFT}"
assert not FINAL.exists(), (
    f"{FINAL.name} already exists. Inspect it first; do not overwrite."
)

wb = load_workbook(DRAFT)
ws = wb["Benchmark_V2_Pilot"]

# Find column names.
header_row = None
headers = {}
for r in range(1, min(ws.max_row, 15) + 1):
    vals = {
        str(ws.cell(r, c).value).strip(): c
        for c in range(1, ws.max_column + 1)
        if ws.cell(r, c).value is not None
    }
    if "question_id" in vals and "question" in vals:
        header_row, headers = r, vals
        break

assert header_row is not None
assert {"question_id", "question", "expected_answer", "gold_pages"}.issubset(headers)

def row_for(qid):
    found = [
        r for r in range(header_row + 1, ws.max_row + 1)
        if str(ws.cell(r, headers["question_id"]).value).strip() == qid
    ]
    assert len(found) == 1, f"{qid}: expected one row, found {len(found)}"
    return found[0]

r004 = row_for("V2P004")
r006 = row_for("V2P006")

# Guard against finalizing unexpected content.
assert ws.cell(r004, headers["question"]).value == (
    "What does the book ask children to draw after looking at the clouds?"
)
assert ws.cell(r006, headers["question"]).value == (
    "What happens to the rainbow after Anandi colours the flowers and the Sun?"
)
assert str(ws.cell(r006, headers["gold_pages"]).value).strip() == "[130]"

# Mark only the two revised questions as source-checked.
for r in [r004, r006]:
    if "benchmark_status" in headers:
        ws.cell(r, headers["benchmark_status"]).value = "SOURCE_CHECKED_READY_FOR_RETRIEVAL"
    if "annotation_status" in headers:
        ws.cell(r, headers["annotation_status"]).value = "SOURCE_CHECKED"

# Update the change log to reflect source checks.
if "Review_Change_Log" in wb.sheetnames:
    log = wb["Review_Change_Log"]
    log_headers = {
        str(log.cell(1, c).value).strip(): c
        for c in range(1, log.max_column + 1)
        if log.cell(1, c).value is not None
    }
    if "status" in log_headers:
        for r in range(2, log.max_row + 1):
            qid = str(log.cell(r, log_headers["question_id"]).value).strip()
            if qid in {"V2P004", "V2P006"}:
                log.cell(r, log_headers["status"]).value = (
                    "SOURCE CHECKED — READY FOR RETRIEVAL; "
                    "benchmark-wide manual review still pending"
                )

wb.save(FINAL)

# Reopen and verify.
check = load_workbook(FINAL, data_only=True)
cws = check["Benchmark_V2_Pilot"]

def val(qid, field):
    r = row_for(qid)
    return cws.cell(r, headers[field]).value

assert val("V2P004", "question") == (
    "What does the book ask children to draw after looking at the clouds?"
)
assert val("V2P006", "gold_pages") == "[130]"

source_original = ROOT / "NCERT_RAG_Benchmark_V2_Pilot.xlsx"
manifest = {
    "artifact": FINAL.name,
    "status": "REVISED_QUESTIONS_SOURCE_CHECKED_READY_FOR_RETRIEVAL",
    "source_checked_questions": ["V2P004", "V2P006"],
    "source_pdf": (
        "01_SOURCE_BOOKS_V2/Class_01/English/Mridang.pdf"
    ),
    "verification": {
        "V2P004": "PDF page 131, printed page 117: cloud-shape drawing instruction",
        "V2P006": "PDF page 130, printed page 116: rainbow gone after colouring"
    },
    "original_benchmark_preserved": source_original.exists(),
    "original_sha256": hashlib.sha256(source_original.read_bytes()).hexdigest(),
    "draft_sha256": hashlib.sha256(DRAFT.read_bytes()).hexdigest(),
    "reviewed_sha256": hashlib.sha256(FINAL.read_bytes()).hexdigest(),
    "note": (
        "Only V2P004 and V2P006 received the documented revisions. "
        "This does not imply that all ten questions have received "
        "independent human review. Retrieval must be rerun against this file."
    )
}
MANIFEST.write_text(json.dumps(manifest, indent=2))

print("RETRIEVAL-READY REVIEWED BENCHMARK CREATED")
print("File:", FINAL)
print("V2P004:", val("V2P004", "question"))
print("V2P006:", val("V2P006", "question"))
print("V2P006 gold pages:", val("V2P006", "gold_pages"))
print("Original pilot preserved:", source_original.exists())
print("SHA-256:", manifest["reviewed_sha256"])
print("Manifest:", MANIFEST)
print("\nNext: run BM25, Dense, and Hybrid RRF on this reviewed file")
print("and save the results in a NEW experiment folder.")
```

```python
from pathlib import Path
from datetime import datetime
import hashlib
import json
import pickle
import re

import numpy as np
import pandas as pd
import faiss
from sentence_transformers import SentenceTransformer
from openpyxl import load_workbook

# ---------------------------------------------------------
# 1. Paths and safety checks
# ---------------------------------------------------------
ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2"
)
ART = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)
BENCHMARK = ROOT / "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
OUT = ROOT / "reviewed_questions_rerun_v1"

assert BENCHMARK.exists(), f"Missing benchmark: {BENCHMARK}"
assert not OUT.exists(), (
    f"Output folder already exists: {OUT}\n"
    "Inspect it first; this cell will not overwrite it."
)

META_PATH = ART / "indexes/chunk_metadata.csv"
BM25_PATH = ART / "indexes/bm25/bm25_index.pkl"
FAISS_PATH = ART / "indexes/faiss/index.faiss"
EMBED_PATH = ART / "indexes/embeddings/chunk_embeddings.npy"
RETRIEVAL_MANIFEST = ART / "indexes/retrieval_manifest.json"

for p in [META_PATH, BM25_PATH, FAISS_PATH, EMBED_PATH, RETRIEVAL_MANIFEST]:
    assert p.exists(), f"Required artifact missing: {p}"

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

benchmark_hash = sha256(BENCHMARK)

# ---------------------------------------------------------
# 2. Load and validate benchmark
# ---------------------------------------------------------
wb = load_workbook(BENCHMARK, read_only=True, data_only=True)
ws = wb["Benchmark_V2_Pilot"]
rows = list(ws.values)

headers = list(rows[0])
questions = pd.DataFrame(rows[1:], columns=headers)
questions = questions.dropna(subset=["question_id", "question"]).copy()
questions["question_id"] = questions["question_id"].astype(str).str.strip()

assert len(questions) == 10, f"Expected 10 questions, found {len(questions)}"
assert questions["question_id"].is_unique, "Duplicate question IDs found"

# ---------------------------------------------------------
# 3. Load the frozen retrieval artifacts
# ---------------------------------------------------------
metadata = pd.read_csv(META_PATH)
embeddings = np.load(EMBED_PATH)
index = faiss.read_index(str(FAISS_PATH))

with open(BM25_PATH, "rb") as f:
    bm25 = pickle.load(f)

with open(RETRIEVAL_MANIFEST) as f:
    retrieval_manifest = json.load(f)

N = len(metadata)
assert N == 2033
assert embeddings.shape == (N, 384)
assert index.ntotal == N
assert index.d == embeddings.shape[1]
assert hasattr(bm25, "get_scores")

assert metadata["source_file"].notna().all()
assert metadata["pages"].notna().all()
assert metadata["text"].notna().all()

# Match the saved configuration.
assert retrieval_manifest["embedding_model"] == (
    "sentence-transformers/all-MiniLM-L6-v2"
)
assert retrieval_manifest["hybrid"]["rrf_k"] == 60

# Load the same embedding model specified by the manifest.
model = SentenceTransformer(
    retrieval_manifest["embedding_model"]
)

# ---------------------------------------------------------
# 4. Helpers for PDF-page relevance
# ---------------------------------------------------------
def parse_pages(value):
    """Parse benchmark JSON-style page lists or metadata comma lists."""
    if value is None or pd.isna(value):
        return set()

    if isinstance(value, (list, tuple, set)):
        return {int(x) for x in value if str(x).strip()}

    s = str(value).strip()
    if not s:
        return set()

    try:
        obj = json.loads(s)
        if isinstance(obj, list):
            return {int(x) for x in obj}
        if isinstance(obj, int):
            return {obj}
    except (json.JSONDecodeError, TypeError, ValueError):
        pass

    return {
        int(x) for x in re.findall(r"\d+", s)
    }

metadata["page_set"] = metadata["pages"].apply(parse_pages)

def query_tokens(text):
    # Matches the recorded lowercase whitespace-split BM25 tokenizer.
    return str(text).lower().split()

# ---------------------------------------------------------
# 5. Retrieve top 20 with each method, then fuse using RRF
# ---------------------------------------------------------
CANDIDATE_K = 20
RRF_K = 60
TOP_K = 10

raw_records = []

for qi, q in questions.iterrows():
    qid = str(q["question_id"])
    query = str(q["question"]).strip()
    gold_source = str(q["source_file"]).strip()
    gold_pages = parse_pages(q["gold_pages"])

    # BM25 top 20
    bm25_scores = np.asarray(
        bm25.get_scores(query_tokens(query))
    )
    bm25_ids = np.argsort(-bm25_scores, kind="stable")[:CANDIDATE_K]

    # Dense top 20 using the saved normalized FAISS index.
    qvec = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True,
        show_progress_bar=False
    ).astype(np.float32)

    dense_scores, dense_ids = index.search(qvec, CANDIDATE_K)
    dense_ids = dense_ids[0]
    dense_scores = dense_scores[0]

    assert np.all(dense_ids >= 0), f"Invalid Dense IDs for {qid}"

    bm25_rank = {int(cid): rank for rank, cid in enumerate(bm25_ids, 1)}
    dense_rank = {int(cid): rank for rank, cid in enumerate(dense_ids, 1)}

    # Reciprocal Rank Fusion across the two candidate lists.
    rrf_scores = {}
    for cid, rank in bm25_rank.items():
        rrf_scores[cid] = rrf_scores.get(cid, 0.0) + 1 / (RRF_K + rank)
    for cid, rank in dense_rank.items():
        rrf_scores[cid] = rrf_scores.get(cid, 0.0) + 1 / (RRF_K + rank)

    hybrid_ids = sorted(
        rrf_scores,
        key=lambda cid: (-rrf_scores[cid], cid)
    )[:TOP_K]

    # Preserve separate rankings and original retrieval scores.
    method_lists = {
        "BM25": [
            (int(cid), float(bm25_scores[int(cid)]))
            for cid in bm25_ids[:TOP_K]
        ],
        "DENSE": [
            (int(cid), float(score))
            for cid, score in zip(dense_ids, dense_scores)
        ][:TOP_K],
        "HYBRID_RRF": [
            (int(cid), float(rrf_scores[int(cid)]))
            for cid in hybrid_ids
        ],
    }

    for method, ranked in method_lists.items():
        for rank, (cid, score) in enumerate(ranked, 1):
            m = metadata.iloc[cid]
            retrieved_pages = set(m["page_set"])
            relevant = (
                str(m["source_file"]).strip() == gold_source
                and bool(retrieved_pages.intersection(gold_pages))
            )

            raw_records.append({
                "question_id": qid,
                "question": query,
                "question_type": q.get("question_type"),
                "method": method,
                "rank": rank,
                "chunk_index": cid,
                "retrieval_score": score,
                "source_file": m["source_file"],
                "pages": m["pages"],
                "gold_source_file": gold_source,
                "gold_pages": json.dumps(sorted(gold_pages)),
                "page_relevant": int(relevant),
                "chunk_text": m["text"],
            })

    if (qi + 1) % 5 == 0:
        print(f"Processed {qi + 1}/{len(questions)} questions")

raw = pd.DataFrame(raw_records)

# ---------------------------------------------------------
# 6. Calculate retrieval metrics with binary relevance
# ---------------------------------------------------------
metric_rows = []

for qid, qdf in raw.groupby("question_id", sort=False):
    gold_pages = parse_pages(qdf.iloc[0]["gold_pages"])

    for method, rdf in qdf.groupby("method", sort=False):
        rdf = rdf.sort_values("rank")
        rel = rdf["page_relevant"].astype(int).to_numpy()

        for k in [1, 3, 5, 10]:
            top = rdf.head(k)
            r = rel[:k]

            # Binary nDCG; ideal ranking sorts the observed relevance labels.
            discounts = 1 / np.log2(np.arange(2, len(r) + 2))
            dcg = float(np.sum(r * discounts))
            ideal = np.sort(r)[::-1]
            idcg = float(np.sum(ideal * discounts))
            ndcg = dcg / idcg if idcg > 0 else 0.0

            first = np.flatnonzero(r)
            mrr = 1 / (first[0] + 1) if len(first) else 0.0

            covered_pages = set()
            for pages in top["pages"]:
                covered_pages |= parse_pages(pages)

            metric_rows.append({
                "question_id": qid,
                "method": method,
                "k": k,
                "hit": int(r.sum() > 0),
                "chunk_precision": float(r.mean()) if len(r) else 0.0,
                "mrr_at_k": float(mrr),
                "ndcg_at_k": float(ndcg),
                "gold_page_recall": (
                    len(covered_pages.intersection(gold_pages)) / len(gold_pages)
                    if gold_pages else np.nan
                ),
                "gold_pages_covered": json.dumps(
                    sorted(covered_pages.intersection(gold_pages))
                ),
            })

metrics = pd.DataFrame(metric_rows)

aggregate = (
    metrics.groupby(["method", "k"], as_index=False)
    .agg(
        questions=("question_id", "nunique"),
        hit_rate=("hit", "mean"),
        mean_chunk_precision=("chunk_precision", "mean"),
        mean_mrr_at_k=("mrr_at_k", "mean"),
        mean_ndcg_at_k=("ndcg_at_k", "mean"),
        mean_gold_page_recall=("gold_page_recall", "mean"),
    )
)

# ---------------------------------------------------------
# 7. Save to a new directory, with provenance manifest
# ---------------------------------------------------------
OUT.mkdir(parents=True, exist_ok=False)

raw.to_csv(OUT / "raw_retrieval_results.csv", index=False)
metrics.to_csv(OUT / "question_metrics.csv", index=False)
aggregate.to_csv(OUT / "aggregate_metrics.csv", index=False)

manifest = {
    "experiment": "reviewed benchmark retrieval rerun v1",
    "timestamp": datetime.now().astimezone().isoformat(),
    "benchmark": str(BENCHMARK),
    "benchmark_sha256": benchmark_hash,
    "question_count": int(len(questions)),
    "retrieval_artifacts": str(ART),
    "chunk_count": N,
    "embedding_model": retrieval_manifest["embedding_model"],
    "embedding_shape": list(embeddings.shape),
    "methods": ["BM25", "DENSE", "HYBRID_RRF"],
    "bm25_candidate_k": CANDIDATE_K,
    "dense_candidate_k": CANDIDATE_K,
    "rrf_k": RRF_K,
    "top_k": TOP_K,
    "relevance_definition": (
        "Retrieved source_file equals benchmark gold source_file "
        "and retrieved PDF-page set intersects benchmark gold_pages."
    ),
    "caveats": [
        "Revised questions mean this is not a same-question comparison "
        "with the original baseline.",
        "Retrieval relevance is not evidence sufficiency or answer correctness.",
        "This is a 10-question exploratory pilot.",
        "Unanswerable-question rejection is not evaluated here."
    ],
    "outputs": {
        "raw_results": str(OUT / "raw_retrieval_results.csv"),
        "question_metrics": str(OUT / "question_metrics.csv"),
        "aggregate_metrics": str(OUT / "aggregate_metrics.csv"),
    }
}

with open(OUT / "experiment_manifest.json", "w") as f:
    json.dump(manifest, f, indent=2)

# ---------------------------------------------------------
# 8. Verify outputs before interpreting them
# ---------------------------------------------------------
assert len(raw) == 300, f"Expected 300 rows, got {len(raw)}"
assert raw.groupby(["question_id", "method"]).size().eq(10).all()
assert raw.groupby(["question_id", "method"])["rank"].nunique().eq(10).all()
assert set(raw["method"]) == {"BM25", "DENSE", "HYBRID_RRF"}
assert sha256(BENCHMARK) == benchmark_hash

print("\nRERUN COMPLETED")
print("Output directory:", OUT)
print("Benchmark SHA-256:", benchmark_hash)
print("\nAggregate metrics:")
print(aggregate.round(4).to_string(index=False))
print("\nSaved files:")
for p in sorted(OUT.iterdir()):
    print(f"{p.name}: {p.stat().st_size:,} bytes")
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import json
import re

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2"
)
RUN = ROOT / "reviewed_questions_rerun_v1"
OUT = RUN / "corrected_metrics_v1"

assert (RUN / "raw_retrieval_results.csv").exists()
assert not OUT.exists(), f"Output already exists: {OUT}"

raw = pd.read_csv(RUN / "raw_retrieval_results.csv")

def pages(value):
    if pd.isna(value):
        return set()
    try:
        obj = json.loads(str(value))
        if isinstance(obj, list):
            return {int(x) for x in obj}
    except (ValueError, TypeError):
        pass
    return {int(x) for x in re.findall(r"\d+", str(value))}

rows = []

for (qid, method), group in raw.groupby(
    ["question_id", "method"], sort=False
):
    group = group.sort_values("rank").reset_index(drop=True)

    gold_source = str(group.iloc[0]["gold_source_file"]).strip()
    gold_pages = pages(group.iloc[0]["gold_pages"])
    relevance = group["page_relevant"].astype(int).to_numpy()

    for k in [1, 3, 5, 10]:
        top = group.head(k)
        rel = relevance[:len(top)]

        first = np.flatnonzero(rel)
        mrr = 1 / (first[0] + 1) if len(first) else 0.0

        discounts = 1 / np.log2(np.arange(2, len(rel) + 2))
        dcg = float(np.sum(rel * discounts))
        ideal = np.sort(rel)[::-1]
        idcg = float(np.sum(ideal * discounts))
        ndcg = dcg / idcg if idcg > 0 else 0.0

        # IMPORTANT: count gold pages only from the expected source book.
        covered_gold_pages = set()
        for _, row in top.iterrows():
            if str(row["source_file"]).strip() == gold_source:
                covered_gold_pages |= pages(row["pages"])

        page_recall = (
            len(covered_gold_pages & gold_pages) / len(gold_pages)
            if gold_pages else np.nan
        )

        rows.append({
            "question_id": qid,
            "method": method,
            "k": k,
            "hit": int(rel.sum() > 0),
            "chunk_precision": float(rel.mean()) if len(rel) else 0.0,
            "mrr_at_k": float(mrr),
            "ndcg_at_k": float(ndcg),
            "gold_page_recall": page_recall,
            "gold_pages_covered": json.dumps(
                sorted(covered_gold_pages & gold_pages)
            ),
        })

per_question = pd.DataFrame(rows)

aggregate = (
    per_question.groupby(["method", "k"], as_index=False)
    .agg(
        questions=("question_id", "nunique"),
        hit_rate=("hit", "mean"),
        mean_chunk_precision=("chunk_precision", "mean"),
        mean_mrr_at_k=("mrr_at_k", "mean"),
        mean_ndcg_at_k=("ndcg_at_k", "mean"),
        mean_gold_page_recall=("gold_page_recall", "mean"),
    )
)

OUT.mkdir(parents=True)
per_question.to_csv(OUT / "question_metrics_corrected.csv", index=False)
aggregate.to_csv(OUT / "aggregate_metrics_corrected.csv", index=False)

manifest = {
    "experiment": "reviewed benchmark retrieval rerun v1 - corrected evaluation",
    "input_raw_results": str(RUN / "raw_retrieval_results.csv"),
    "correction": (
        "Gold-page recall now counts only pages retrieved from the expected "
        "gold source_file. Retrieval rankings were not changed."
    ),
    "question_count": int(raw["question_id"].nunique()),
    "methods": sorted(raw["method"].unique().tolist()),
    "metrics": "Binary retrieval relevance and corrected source-restricted page recall",
}
(OUT / "manifest.json").write_text(json.dumps(manifest, indent=2))

assert len(raw) == 300
assert len(per_question) == 120
assert len(aggregate) == 12

print("CORRECTED EVALUATION SAVED")
print(OUT)
print("\nCorrected aggregate metrics:")
print(aggregate.round(4).to_string(index=False))
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pickle
import numpy as np
import pandas as pd
import faiss
from sentence_transformers import SentenceTransformer

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
ART = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
RUN = ROOT / (
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "reviewed_questions_rerun_v1"
)
BENCH = ROOT / (
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
)

with open(ART / "indexes/bm25/bm25_index.pkl", "rb") as f:
    bm25 = pickle.load(f)

with open(ART / "indexes/bm25/tokenized_documents.pkl", "rb") as f:
    tokenized_docs = pickle.load(f)

embeddings = np.load(ART / "indexes/embeddings/chunk_embeddings.npy")
index = faiss.read_index(str(ART / "indexes/faiss/index.faiss"))
meta = pd.read_csv(ART / "indexes/chunk_metadata.csv")
raw = pd.read_csv(RUN / "raw_retrieval_results.csv")

xls = pd.ExcelFile(BENCH)
questions = pd.read_excel(BENCH, sheet_name="Benchmark_V2_Pilot")
questions = questions.set_index("question_id")

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

print("Chunks:", len(meta))
print("Embeddings:", embeddings.shape)
print("FAISS vectors:", index.ntotal)
print("BM25 documents:", len(tokenized_docs))

assert len(meta) == len(tokenized_docs) == embeddings.shape[0] == index.ntotal

def source_of(row):
    return str(row["source_file"]).replace("\\", "/")

def retrieve_top20(question):
    # Match the saved pipeline's lowercase whitespace BM25 tokenizer.
    tokens = question.lower().split()
    scores = np.asarray(bm25.get_scores(tokens))

    bm25_ids = np.argsort(scores)[::-1][:20]

    query_vec = model.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")

    dense_scores, dense_ids = index.search(query_vec, 20)
    dense_ids = dense_ids[0]
    dense_scores = dense_scores[0]

    # Reciprocal Rank Fusion, k=60, over each method's top 20.
    rrf_scores = {}
    for rank, idx in enumerate(bm25_ids, start=1):
        rrf_scores[int(idx)] = rrf_scores.get(int(idx), 0) + 1 / (60 + rank)

    for rank, idx in enumerate(dense_ids, start=1):
        rrf_scores[int(idx)] = rrf_scores.get(int(idx), 0) + 1 / (60 + rank)

    hybrid_ids = sorted(rrf_scores, key=rrf_scores.get, reverse=True)

    # Print target chunks and their rank in each candidate list.
    target_ids = meta[
        (meta["source_file"].astype(str).str.replace("\\", "/", regex=False)
         == "Class_01/English/Mridang.pdf")
        & (pd.to_numeric(meta["chunk_index"], errors="coerce").isin([41, 42]))
    ].index.tolist()

    print("\nTarget chunk positions in metadata:", target_ids)

    for idx in target_ids:
        print("\n" + "-" * 65)
        print("Metadata row:", idx, "| Chunk index:", meta.loc[idx, "chunk_index"])
        print("PDF pages:", meta.loc[idx, "pages"])
        print("BM25 rank:", (
            int(np.where(bm25_ids == idx)[0][0]) + 1
            if idx in bm25_ids else "NOT IN TOP 20"
        ))
        print("Dense rank:", (
            int(np.where(dense_ids == idx)[0][0]) + 1
            if idx in dense_ids else "NOT IN TOP 20"
        ))
        print("Hybrid rank:", (
            hybrid_ids.index(idx) + 1
            if idx in hybrid_ids else "NOT IN FUSED CANDIDATES"
        ))
        print("Text:", str(meta.loc[idx, "text"])[:500])

    def show_list(label, ids, score_map=None, limit=20):
        print(f"\n{label}")
        for rank, idx in enumerate(ids[:limit], start=1):
            row = meta.iloc[int(idx)]
            print(
                f"{rank:2}. row={idx} chunk={row['chunk_index']} "
                f"| {row['source_file']} | pages={row['pages']}"
            )

    show_list("BM25 TOP 20", bm25_ids)
    show_list("DENSE TOP 20", dense_ids)
    show_list("HYBRID RRF TOP 20", hybrid_ids)

for qid in ["V2P008", "V2P009"]:
    question = str(questions.loc[qid, "question"])
    print("\n" + "=" * 80)
    print(qid, "|", question)
    print("=" * 80)

    retrieve_top20(question)

    saved = raw[raw["question_id"].astype(str) == qid]
    print("\nSaved top-10 ranks by method:")
    for method in ["BM25", "DENSE", "HYBRID_RRF"]:
        sub = saved[saved["method"] == method].sort_values("rank")
        print(method, list(zip(
            sub["rank"].tolist(),
            sub["chunk_index"].tolist(),
            sub["source_file"].tolist()
        )))

print("\nDIAGNOSTIC ONLY — no benchmark or retrieval files modified.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import numpy as np
import pandas as pd
import pickle
import faiss
from pathlib import Path
from sentence_transformers import SentenceTransformer

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
ART = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

with open(ART / "indexes/bm25/bm25_index.pkl", "rb") as f:
    bm25 = pickle.load(f)

embeddings = np.load(ART / "indexes/embeddings/chunk_embeddings.npy")
index = faiss.read_index(str(ART / "indexes/faiss/index.faiss"))
meta = pd.read_csv(ART / "indexes/chunk_metadata.csv")
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# The known relevant passage is global metadata row 42.
target_row = meta[
    (meta["source_file"].astype(str).str.replace("\\", "/", regex=False)
     == "Class_01/English/Mridang.pdf")
    & (pd.to_numeric(meta["chunk_index"], errors="coerce") == 42)
]

assert len(target_row) == 1, (
    f"Expected one target chunk; found {len(target_row)}"
)

target_idx = int(target_row.index[0])
target_text = str(meta.loc[target_idx, "text"])

print("Target metadata row:", target_idx)
print("Target chunk index:", meta.loc[target_idx, "chunk_index"])
print("Target pages:", meta.loc[target_idx, "pages"])
print("Relevant passage:\n", target_text[:700])

queries = {
    "Q1_original": (
        "If you are not using a tap, what does the book ask you to remember to do?"
    ),
    "Q2_lexical": (
        "What does the book ask children to do with the tap when they are not using it?"
    ),
    "Q3_passage_words": (
        "Do you remember to close the tap when you are not using it?"
    ),
    "Q4_short": "close the tap when not using it",
    "Q5_paraphrase": "What should children do with an unused water tap?",
}

rows = []

for label, query in queries.items():
    # BM25 score and rank of the known relevant chunk.
    bm25_scores = np.asarray(bm25.get_scores(query.lower().split()))
    bm25_rank = int(np.sum(bm25_scores > bm25_scores[target_idx])) + 1

    # Dense similarity/rank using the saved normalized embeddings.
    qvec = model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")

    dense_scores = (embeddings @ qvec[0])
    dense_rank = int(np.sum(dense_scores > dense_scores[target_idx])) + 1

    _, dense_top20 = index.search(qvec, 20)
    dense_in_top20 = target_idx in dense_top20[0]

    bm25_top20 = np.argsort(bm25_scores)[::-1][:20]
    bm25_in_top20 = target_idx in bm25_top20

    rows.append({
        "query_variant": label,
        "query": query,
        "target_BM25_score": float(bm25_scores[target_idx]),
        "target_BM25_rank": bm25_rank,
        "target_BM25_top20": bm25_in_top20,
        "target_dense_cosine": float(dense_scores[target_idx]),
        "target_dense_rank": dense_rank,
        "target_dense_top20": dense_in_top20,
    })

results = pd.DataFrame(rows)
display(results)

print("\nNo files written. Existing indexes and benchmark remain unchanged.")
```

```python
from google.colab import drive
drive.mount('/content/drive')

import numpy as np
import pandas as pd

# Uses variables loaded by the previous diagnostic cell:
# bm25, embeddings, faiss_index, meta, model, target_row

original_query = (
    "If you are not using a tap, what does the book ask you to remember to do?"
)
lexical_query = (
    "What does the book ask children to do with the tap when they are not using it?"
)

TARGET_CHUNK = int(meta.index[meta["chunk_index"].eq(42)][0])
RRF_K = 60
CANDIDATE_K = 20

def retrieve_lists(query):
    tokens = query.lower().split()

    bm25_scores = np.asarray(bm25.get_scores(tokens))
    bm25_ids = np.argsort(-bm25_scores)[:CANDIDATE_K].tolist()

    query_vec = model.encode(
        [query], normalize_embeddings=True, convert_to_numpy=True
    )[0].astype("float32")

    dense_scores = embeddings @ query_vec
    dense_ids = np.argsort(-dense_scores)[:CANDIDATE_K].tolist()

    # Same reciprocal-rank fusion rule as the experiment: k = 60
    rrf_scores = {}
    for ranked_ids in (bm25_ids, dense_ids):
        for rank, chunk_id in enumerate(ranked_ids, start=1):
            rrf_scores[chunk_id] = (
                rrf_scores.get(chunk_id, 0.0) + 1.0 / (RRF_K + rank)
            )

    hybrid_ids = sorted(
        rrf_scores, key=rrf_scores.get, reverse=True
    )

    def rank_of(ids):
        return ids.index(TARGET_CHUNK) + 1 if TARGET_CHUNK in ids else None

    return {
        "BM25 target rank": rank_of(bm25_ids),
        "Dense target rank": rank_of(dense_ids),
        "Hybrid RRF target rank": rank_of(hybrid_ids),
        "Hybrid target in candidate union": TARGET_CHUNK in rrf_scores,
    }

results = pd.DataFrame([
    {"query": "Original", **retrieve_lists(original_query)},
    {"query": "Lexical rewrite", **retrieve_lists(lexical_query)},
])

display(results)
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pickle
import json
import hashlib
import numpy as np
import pandas as pd
import faiss
from sentence_transformers import SentenceTransformer

# ---------- Paths ----------
ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
ART = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
BENCH = ROOT / (
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
)
OUT = ROOT / (
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "query_expansion_pilot_v1"
)
OUT.mkdir(parents=True, exist_ok=True)

EXPECTED_SHA = (
    "d3aa2d9e39f949206585936f865c382f1fc9bfa0bf3ba7f9b1a8cb43f52cabcd"
)
actual_sha = hashlib.sha256(BENCH.read_bytes()).hexdigest()
assert actual_sha == EXPECTED_SHA, f"Unexpected benchmark SHA: {actual_sha}"

# ---------- Load benchmark ----------
df = pd.read_excel(BENCH, sheet_name="Benchmark_V2_Pilot")
df.columns = [str(c).strip() for c in df.columns]

required = {"question_id", "question"}
assert required.issubset(df.columns), f"Missing columns: {df.columns.tolist()}"
assert len(df) == 10, f"Expected 10 pilot questions, found {len(df)}"

# One controlled, meaning-preserving rewrite per question.
rewrites = {
    "V2P001": "What did Anandi want to paint in her garden?",
    "V2P002": "Why did Anandi save the yellow colour for the Sun?",
    "V2P003": "What colours did Anandi use to colour the flowers and the Sun?",
    "V2P004": "What do the children have to draw after looking at the clouds?",
    "V2P005": "What should children observe in the sky and share with the class?",
    "V2P006": "After Anandi colours the flowers and the Sun, what happens to the rainbow?",
    "V2P007": "How did Anandi use the rainbow as a guide for colouring her garden?",
    "V2P008": "What does the book ask children to do with the tap when they are not using it?",
    "V2P009": "What questions does the book ask children about sharing food with friends?",
    "V2P010": "What activity does the book suggest children do with the clouds?",
}

assert set(df["question_id"]) == set(rewrites), (
    "Question IDs differ from the rewrite list; inspect the benchmark before proceeding."
)

# ---------- Load existing retrieval artifacts ----------
with open(ART / "indexes/bm25/bm25_index.pkl", "rb") as f:
    bm25 = pickle.load(f)

embeddings = np.load(ART / "indexes/embeddings/chunk_embeddings.npy")
index = faiss.read_index(str(ART / "indexes/faiss/index.faiss"))
meta = pd.read_csv(ART / "indexes/chunk_metadata.csv")

assert len(meta) == embeddings.shape[0] == index.ntotal
assert len(df) == 10

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

TOP_K = 10
CANDIDATE_K = 20
RRF_K = 60

def rank_query(query):
    # BM25 candidate list
    bm25_scores = np.asarray(bm25.get_scores(query.lower().split()))
    bm25_ids = np.argsort(-bm25_scores)[:CANDIDATE_K].tolist()

    # Dense candidate list; embeddings are already normalized
    qvec = model.encode(
        [query], normalize_embeddings=True, convert_to_numpy=True
    ).astype("float32")
    dense_scores, dense_indices = index.search(qvec, CANDIDATE_K)
    dense_ids = [int(i) for i in dense_indices[0] if i >= 0]

    # Reciprocal Rank Fusion
    rrf = {}
    for ids in (bm25_ids, dense_ids):
        for rank, chunk_id in enumerate(ids, start=1):
            rrf[chunk_id] = rrf.get(chunk_id, 0.0) + 1.0 / (RRF_K + rank)

    hybrid_ids = sorted(rrf, key=rrf.get, reverse=True)[:TOP_K]

    return {
        "BM25": bm25_ids[:TOP_K],
        "DENSE": dense_ids[:TOP_K],
        "HYBRID_RRF": hybrid_ids,
    }

# ---------- Run both query conditions ----------
records = []

for _, row in df.iterrows():
    qid = row["question_id"]
    original = str(row["question"])
    queries = {
        "ORIGINAL": original,
        "REWRITTEN": rewrites[qid],
    }

    for condition, query in queries.items():
        rankings = rank_query(query)

        for method, chunk_ids in rankings.items():
            for rank, chunk_id in enumerate(chunk_ids, start=1):
                m = meta.iloc[chunk_id]
                records.append({
                    "question_id": qid,
                    "condition": condition,
                    "query": query,
                    "method": method,
                    "rank": rank,
                    "chunk_index": int(chunk_id),
                    "source_file": str(m["source_file"]),
                    "pages": str(m["pages"]),
                    "chunk_text": str(m["text"]),
                })

raw = pd.DataFrame(records)
raw_path = OUT / "raw_query_expansion_results.csv"
raw.to_csv(raw_path, index=False)

# ---------- Fair, paired source-level comparison ----------
# This uses the benchmark's gold source file, not page/evidence judgments.
# Source-level hits are a diagnostic only, not proof of evidence sufficiency.
source_col = next(
    (c for c in ["gold_source_file", "gold_source", "source_file"]
     if c in df.columns),
    None
)
if source_col is None:
    raise ValueError(
        "No gold-source column found. Available columns: "
        + ", ".join(df.columns)
    )

gold_sources = dict(zip(df["question_id"], df[source_col].astype(str)))

def source_matches(retrieved_source, gold_source):
    # Allow metadata paths to include prefixes while preserving the filename.
    return (
        Path(retrieved_source.replace("\\", "/")).as_posix().endswith(
            gold_source.replace("\\", "/").lstrip("/")
        )
        or Path(retrieved_source.replace("\\", "/")).name
        == Path(gold_source.replace("\\", "/")).name
    )

metrics = []
for (qid, condition, method), group in raw.groupby(
    ["question_id", "condition", "method"]
):
    group = group.sort_values("rank")
    gold = gold_sources[qid]
    hits = [
        source_matches(src, gold)
        for src in group["source_file"].tolist()
    ]

    row = {
        "question_id": qid,
        "condition": condition,
        "method": method,
        "gold_source_file": gold,
        "source_Hit@1": int(any(hits[:1])),
        "source_Hit@3": int(any(hits[:3])),
        "source_Hit@5": int(any(hits[:5])),
        "source_Hit@10": int(any(hits[:10])),
        "source_MRR@10": (
            1.0 / (hits.index(True) + 1) if any(hits) else 0.0
        ),
    }
    metrics.append(row)

qm = pd.DataFrame(metrics)
qm.to_csv(OUT / "question_source_metrics.csv", index=False)

aggregate = (
    qm.groupby(["condition", "method"])
      .agg(
          questions=("question_id", "nunique"),
          source_Hit1=("source_Hit@1", "mean"),
          source_Hit3=("source_Hit@3", "mean"),
          source_Hit5=("source_Hit@5", "mean"),
          source_Hit10=("source_Hit@10", "mean"),
          source_MRR10=("source_MRR@10", "mean"),
      )
      .reset_index()
)
aggregate.to_csv(OUT / "aggregate_source_metrics.csv", index=False)

# Paired per-question changes: REWRITTEN minus ORIGINAL
paired = qm.pivot(
    index=["question_id", "method"],
    columns="condition",
    values=["source_Hit@1", "source_Hit@3", "source_Hit@5",
            "source_Hit@10", "source_MRR@10"],
)
paired.columns = [f"{metric}_{condition}" for metric, condition in paired.columns]
paired = paired.reset_index()

for metric in ["source_Hit@1", "source_Hit@3", "source_Hit@5",
               "source_Hit@10", "source_MRR@10"]:
    paired[f"{metric}_delta"] = (
        paired[f"{metric}_REWRITTEN"] - paired[f"{metric}_ORIGINAL"]
    )

paired.to_csv(OUT / "paired_question_deltas.csv", index=False)

manifest = {
    "experiment": "query_expansion_pilot_v1",
    "benchmark": BENCH.name,
    "benchmark_sha256": actual_sha,
    "questions": 10,
    "rewrites": rewrites,
    "methods": ["BM25", "DENSE", "HYBRID_RRF"],
    "candidate_k": CANDIDATE_K,
    "top_k": TOP_K,
    "rrf_k": RRF_K,
    "metric_scope": (
        "Source-level retrieval diagnostic only; does not measure "
        "page-level relevance, evidence sufficiency, or answer correctness."
    ),
    "output_files": [
        "raw_query_expansion_results.csv",
        "question_source_metrics.csv",
        "aggregate_source_metrics.csv",
        "paired_question_deltas.csv",
    ],
}
(OUT / "experiment_manifest.json").write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8",
)

print("Saved to:", OUT)
print("\nAGGREGATE SOURCE-LEVEL METRICS")
display(aggregate.round(4))

print("\nPER-QUESTION CHANGES (positive means rewrite improved the metric)")
display(paired.round(4))

print("\nRaw rows:", len(raw))
print("Benchmark unchanged; existing indexes unchanged.")
```

```python
# ============================================================
# NCERT RAG — QUERY EXPANSION EVIDENCE COMPARISON
# Reads annotated workbook; writes a separate audit report.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib
from datetime import datetime, timezone

BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "query_expansion_pilot_v1/evidence_review_v1"
)

# Prefer the newly uploaded file; fall back only if necessary.
candidates = [
    BASE / "unique_question_passage_review_annotated(1).xlsx",
    BASE / "unique_question_passage_review_annotated.xlsx",
]

INPUT = next((p for p in candidates if p.exists()), None)

if INPUT is None:
    raise FileNotFoundError(
        "Annotated workbook not found. Expected one of:\n"
        + "\n".join(str(p) for p in candidates)
        + "\nCheck the exact filename in Google Drive."
    )

OUT = BASE / "query_expansion_evidence_comparison_v1"
OUT.mkdir(parents=True, exist_ok=True)

# ---------- Load and validate ----------
df = pd.read_excel(INPUT, sheet_name="Passage_Review")

required = {
    "question_id", "chunk_index", "question",
    "original_rank", "rewritten_rank",
    "relevance_label", "support_label",
    "retrieved_source_file", "chunk_text"
}
missing = required - set(df.columns)
if missing:
    raise ValueError(f"Missing expected columns: {sorted(missing)}")

for col in ["relevance_label", "support_label"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

if df["relevance_label"].isna().any():
    raise ValueError(
        f"{df['relevance_label'].isna().sum()} rows lack relevance labels."
    )
if df["support_label"].isna().any():
    raise ValueError(
        f"{df['support_label'].isna().sum()} rows lack support labels."
    )

if not df["relevance_label"].isin([0, 1]).all():
    raise ValueError("Relevance labels must be 0 or 1.")
if not df["support_label"].isin([0, 1, 2]).all():
    raise ValueError("Support labels must be 0, 1, or 2.")

# A passage must be relevant if it is judged sufficient to answer.
inconsistent = df[
    (df["support_label"] == 2) & (df["relevance_label"] != 1)
]
if len(inconsistent):
    print("WARNING: sufficient-support rows labelled irrelevant:")
    display(inconsistent[[
        "question_id", "chunk_index",
        "relevance_label", "support_label", "review_reason"
    ]])

# ---------- Reconstruct ORIGINAL and REWRITTEN top-3 ----------
# The worksheet stores one row per question-passage pair.
# A blank rank means that passage was absent from that condition's top 3.

def rank_number(value):
    try:
        if pd.isna(value) or str(value).strip() == "":
            return np.nan
        return float(value)
    except (ValueError, TypeError):
        return np.nan

df["original_rank_num"] = df["original_rank"].map(rank_number)
df["rewritten_rank_num"] = df["rewritten_rank"].map(rank_number)

def build_condition(rank_col, condition_name):
    x = df[df[rank_col].notna()].copy()
    x["rank"] = x[rank_col].astype(int)
    x["condition"] = condition_name
    x = x[x["rank"].between(1, 3)].copy()

    # One question cannot have two passages at the same rank.
    duplicates = x.duplicated(["question_id", "rank"], keep=False)
    if duplicates.any():
        print(f"WARNING: duplicate ranks in {condition_name}:")
        display(x.loc[duplicates, ["question_id", "rank", "chunk_index"]])

    return x

original = build_condition("original_rank_num", "ORIGINAL")
rewritten = build_condition("rewritten_rank_num", "REWRITTEN")

if original.empty or rewritten.empty:
    raise ValueError("One condition has no top-3 passages; check rank columns.")

# ---------- Per-question metrics ----------
question_ids = sorted(set(df["question_id"].dropna().astype(str)))

def metrics_for_condition(rows, qid):
    q = rows[rows["question_id"].astype(str) == qid].sort_values("rank")
    rel = q["relevance_label"].astype(int).tolist()
    sup = q["support_label"].astype(int).tolist()
    ranks = q["rank"].astype(int).tolist()

    sufficient_ranks = [
        rank for rank, label in zip(ranks, sup) if label == 2
    ]
    relevant_ranks = [
        rank for rank, label in zip(ranks, rel) if label == 1
    ]

    first_sufficient_rr = (
        1 / min(sufficient_ranks) if sufficient_ranks else 0.0
    )
    first_relevant_rr = (
        1 / min(relevant_ranks) if relevant_ranks else 0.0
    )

    return {
        "retrieved_count_top3": len(q),
        "Relevant@1": int(1 in relevant_ranks),
        "Relevant@3": int(bool(relevant_ranks)),
        "SufficientEvidence@1": int(1 in sufficient_ranks),
        "SufficientEvidence@3": int(bool(sufficient_ranks)),
        "FirstRelevantRR": first_relevant_rr,
        "FirstSufficientRR": first_sufficient_rr,
        "support2_count_top3": sum(label == 2 for label in sup),
        "support1_count_top3": sum(label == 1 for label in sup),
        "support0_count_top3": sum(label == 0 for label in sup),
    }

records = []
for qid in question_ids:
    for condition_name, rows in [
        ("ORIGINAL", original), ("REWRITTEN", rewritten)
    ]:
        rec = {
            "question_id": qid,
            "condition": condition_name,
        }
        rec.update(metrics_for_condition(rows, qid))
        records.append(rec)

per_condition = pd.DataFrame(records)

# Pivot into a side-by-side comparison.
metric_cols = [
    "retrieved_count_top3", "Relevant@1", "Relevant@3",
    "SufficientEvidence@1", "SufficientEvidence@3",
    "FirstRelevantRR", "FirstSufficientRR",
    "support2_count_top3", "support1_count_top3", "support0_count_top3"
]

wide = per_condition.pivot(
    index="question_id", columns="condition", values=metric_cols
)
wide.columns = [f"{metric}_{condition.lower()}" for metric, condition in wide.columns]
wide = wide.reset_index()

for metric in [
    "Relevant@1", "Relevant@3",
    "SufficientEvidence@1", "SufficientEvidence@3",
    "FirstRelevantRR", "FirstSufficientRR"
]:
    wide[f"{metric}_change"] = (
        wide[f"{metric}_rewritten"] - wide[f"{metric}_original"]
    )

def classify(row):
    delta = row["SufficientEvidence@3_change"]
    if delta > 0:
        return "IMPROVED"
    if delta < 0:
        return "REGRESSED"
    return "UNCHANGED"

wide["sufficient_evidence_top3_outcome"] = wide.apply(classify, axis=1)

# ---------- Aggregate metrics ----------
aggregate_records = []
for condition_name, rows in [
    ("ORIGINAL", original), ("REWRITTEN", rewritten)
]:
    per_q = per_condition[per_condition["condition"] == condition_name]
    aggregate_records.append({
        "condition": condition_name,
        "questions": len(question_ids),
        "Relevant@1": per_q["Relevant@1"].mean(),
        "Relevant@3": per_q["Relevant@3"].mean(),
        "SufficientEvidence@1": per_q["SufficientEvidence@1"].mean(),
        "SufficientEvidence@3": per_q["SufficientEvidence@3"].mean(),
        "MRR_first_relevant_top3": per_q["FirstRelevantRR"].mean(),
        "MRR_first_sufficient_top3": per_q["FirstSufficientRR"].mean(),
        "mean_sufficient_passages_top3": per_q["support2_count_top3"].mean(),
    })

aggregate = pd.DataFrame(aggregate_records)

# ---------- Passage-level comparison ----------
passage_cols = [
    c for c in [
        "question_id", "chunk_index", "question", "expected_answer",
        "evidence_excerpt", "retrieved_source_file", "retrieved_pdf_pages",
        "original_rank", "rewritten_rank", "relevance_label",
        "support_label", "review_reason", "reviewer_id_or_initials",
        "chunk_text"
    ] if c in df.columns
]
passage_report = df[passage_cols].copy()

# ---------- Save report and provenance ----------
per_condition.to_csv(OUT / "per_question_condition_metrics.csv", index=False)
wide.to_csv(OUT / "per_question_comparison.csv", index=False)
aggregate.to_csv(OUT / "aggregate_comparison.csv", index=False)
passage_report.to_csv(OUT / "annotated_passage_audit.csv", index=False)

with pd.ExcelWriter(OUT / "query_expansion_evidence_comparison.xlsx") as writer:
    aggregate.to_excel(writer, sheet_name="Aggregate", index=False)
    wide.to_excel(writer, sheet_name="Per_Question", index=False)
    per_condition.to_excel(writer, sheet_name="Condition_Metrics", index=False)
    passage_report.to_excel(writer, sheet_name="Passage_Audit", index=False)

manifest = {
    "input_file": str(INPUT),
    "input_sha256": hashlib.sha256(INPUT.read_bytes()).hexdigest(),
    "input_rows": len(df),
    "unique_question_ids": len(question_ids),
    "original_top3_occurrences": len(original),
    "rewritten_top3_occurrences": len(rewritten),
    "support_definition": "support_label == 2",
    "relevance_definition": "relevance_label == 1",
    "metrics_scope": "Annotated top-3 passages only; 10-question exploratory pilot",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "output_directory": str(OUT),
}
(OUT / "manifest.json").write_text(json.dumps(manifest, indent=2))

# ---------- Display ----------
print("Input workbook:", INPUT)
print("Input SHA-256:", manifest["input_sha256"])
print("Questions:", len(question_ids))
print("Original top-3 rows:", len(original))
print("Rewritten top-3 rows:", len(rewritten))
print("Outputs saved to:", OUT)

print("\nAGGREGATE COMPARISON")
display(aggregate.round(3))

print("\nPER-QUESTION OUTCOMES")
display(wide[[
    "question_id",
    "SufficientEvidence@1_original",
    "SufficientEvidence@1_rewritten",
    "SufficientEvidence@3_original",
    "SufficientEvidence@3_rewritten",
    "FirstSufficientRR_original",
    "FirstSufficientRR_rewritten",
    "sufficient_evidence_top3_outcome"
]].round(3))

print("\nOUTCOME COUNTS")
display(wide["sufficient_evidence_top3_outcome"].value_counts().rename_axis("outcome").to_frame("questions"))

print("\nIMPORTANT: These are pilot results, not a statistically conclusive claim.")
```

```python
# ================================================================
# NCERT RAG — COMPLETE ORIGINAL vs REWRITTEN vs FUSED RRF EXPERIMENT
# Self-contained: no variables required from previous cells.
# Inputs are read-only. Outputs go to a new experiment folder.
# ================================================================

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import numpy as np
import hashlib
import json
import re

# -------------------- PATHS --------------------

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "query_expansion_pilot_v1"
)

RAW_PATH = ROOT / "raw_query_expansion_results.csv"

ANNOTATION_PATH = (
    ROOT / "evidence_review_v1" /
    "unique_question_passage_review_annotated.xlsx"
)

OUT = ROOT / "original_rewritten_fusion_pilot_v2"
OUT.mkdir(parents=True, exist_ok=True)

if not RAW_PATH.exists():
    raise FileNotFoundError(f"Retrieval CSV not found: {RAW_PATH}")

if not ANNOTATION_PATH.exists():
    raise FileNotFoundError(f"Annotation workbook not found: {ANNOTATION_PATH}")

# -------------------- LOAD INPUTS --------------------

raw = pd.read_csv(RAW_PATH)
ann = pd.read_excel(ANNOTATION_PATH, sheet_name="Passage_Review")

print("Retrieval CSV:", RAW_PATH)
print("Retrieval rows:", len(raw))
print("Retrieval columns:", list(raw.columns))
print("Annotation workbook:", ANNOTATION_PATH)
print("Annotation rows:", len(ann))

# -------------------- FIND COLUMNS ROBUSTLY --------------------

def find_col(df, candidates, required=True):
    lookup = {str(c).strip().lower(): c for c in df.columns}

    for name in candidates:
        if name.lower() in lookup:
            return lookup[name.lower()]

    if required:
        raise ValueError(
            f"Cannot identify column from {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )
    return None

qid_col = find_col(raw, ["question_id", "qid"])
condition_col = find_col(
    raw, ["condition", "query_condition", "query_type", "variant"]
)
method_col = find_col(
    raw, ["method", "retrieval_method", "retriever"]
)
rank_col = find_col(raw, ["rank", "retrieval_rank"])
chunk_col = find_col(raw, ["chunk_index", "chunk_id", "global_chunk_index"])

ann_qid_col = find_col(ann, ["question_id", "qid"])
ann_chunk_col = find_col(ann, ["chunk_index", "chunk_id"])
relevance_col = find_col(ann, ["relevance_label"])
support_col = find_col(ann, ["support_label"])

# Normalize retrieval data.
raw["_qid"] = raw[qid_col].astype(str).str.strip()
raw["_condition"] = raw[condition_col].astype(str).str.upper().str.strip()
raw["_method"] = raw[method_col].astype(str).str.upper().str.strip()
raw["_rank"] = pd.to_numeric(raw[rank_col], errors="coerce")
raw["_chunk"] = pd.to_numeric(raw[chunk_col], errors="coerce")

raw = raw.dropna(subset=["_rank", "_chunk"]).copy()
raw["_rank"] = raw["_rank"].astype(int)
raw["_chunk"] = raw["_chunk"].astype(int)

print("\nConditions found:", sorted(raw["_condition"].unique()))
print("Methods found:", sorted(raw["_method"].unique()))

# Match original and rewritten labels without relying on column names.
def is_original(value):
    v = str(value).upper()
    return "ORIGINAL" in v and "REWRITTEN" not in v

def is_rewritten(value):
    v = str(value).upper()
    return "REWRITTEN" in v or "REWRITE" in v

# Select the Hybrid RRF method if available.
hybrid_mask = raw["_method"].str.contains(
    r"HYBRID|RRF", regex=True, na=False
)
hybrid = raw[hybrid_mask].copy()

if hybrid.empty:
    raise ValueError(
        "No Hybrid/RRF method found. Available methods are printed above. "
        "Do not proceed with a different method without checking the CSV."
    )

orig = hybrid[hybrid["_condition"].map(is_original)].copy()
rewr = hybrid[hybrid["_condition"].map(is_rewritten)].copy()

if orig.empty or rewr.empty:
    raise ValueError(
        "Could not identify both ORIGINAL and REWRITTEN Hybrid RRF rows. "
        f"Observed conditions: {sorted(hybrid['_condition'].unique())}"
    )

print("\nOriginal Hybrid rows:", len(orig))
print("Rewritten Hybrid rows:", len(rewr))

# -------------------- LOAD AND VALIDATE ANNOTATIONS --------------------

ann["_qid"] = ann[ann_qid_col].astype(str).str.strip()
ann["_chunk"] = pd.to_numeric(ann[ann_chunk_col], errors="raise").astype(int)
ann["_relevance"] = pd.to_numeric(
    ann[relevance_col], errors="raise"
).astype(int)
ann["_support"] = pd.to_numeric(
    ann[support_col], errors="raise"
).astype(int)

if ann.duplicated(["_qid", "_chunk"]).any():
    duplicates = ann.loc[
        ann.duplicated(["_qid", "_chunk"], keep=False),
        ["_qid", "_chunk"]
    ]
    raise ValueError(
        "Duplicate question-passage annotation keys found:\n"
        + duplicates.to_string(index=False)
    )

if not ann["_relevance"].isin([0, 1]).all():
    raise ValueError("Relevance labels must be 0 or 1.")

if not ann["_support"].isin([0, 1, 2]).all():
    raise ValueError("Support labels must be 0, 1, or 2.")

label_lookup = {
    (row["_qid"], int(row["_chunk"])): {
        "relevance": int(row["_relevance"]),
        "support": int(row["_support"]),
    }
    for _, row in ann.iterrows()
}

# -------------------- RANK MAPS --------------------

def make_rank_map(frame):
    temp = frame[["_qid", "_chunk", "_rank"]].copy()
    temp = temp.sort_values("_rank")
    temp = temp.drop_duplicates(["_qid", "_chunk"], keep="first")

    return {
        (str(row["_qid"]), int(row["_chunk"])): int(row["_rank"])
        for _, row in temp.iterrows()
    }

orig_rank_map = make_rank_map(orig)
rewr_rank_map = make_rank_map(rewr)

questions = sorted(
    set(orig["_qid"].unique()) & set(rewr["_qid"].unique())
)

if not questions:
    raise ValueError("No shared question IDs between the two conditions.")

print("\nQuestions available in both conditions:", len(questions))
print("Original question/chunk ranks:", len(orig_rank_map))
print("Rewritten question/chunk ranks:", len(rewr_rank_map))

# -------------------- RECONSTRUCT RANKINGS --------------------

rankings = {}

for qid in questions:
    rankings[(qid, "ORIGINAL")] = sorted(
        [
            (chunk, rank)
            for (qid0, chunk), rank in orig_rank_map.items()
            if qid0 == qid
        ],
        key=lambda item: (item[1], item[0])
    )

    rankings[(qid, "REWRITTEN")] = sorted(
        [
            (chunk, rank)
            for (qid0, chunk), rank in rewr_rank_map.items()
            if qid0 == qid
        ],
        key=lambda item: (item[1], item[0])
    )

    # RRF combines the two ranked lists. Missing candidates contribute zero.
    for k, condition_name in [
        (10, "FUSED_RRF_K10"),
        (60, "FUSED_RRF_K60"),
    ]:
        scores = {}

        for (qid0, chunk), rank in orig_rank_map.items():
            if qid0 == qid:
                scores[chunk] = scores.get(chunk, 0.0) + 1.0 / (k + rank)

        for (qid0, chunk), rank in rewr_rank_map.items():
            if qid0 == qid:
                scores[chunk] = scores.get(chunk, 0.0) + 1.0 / (k + rank)

        rankings[(qid, condition_name)] = sorted(
            scores.items(),
            key=lambda item: (-item[1], item[0])
        )

conditions = [
    "ORIGINAL",
    "REWRITTEN",
    "FUSED_RRF_K10",
    "FUSED_RRF_K60",
]

# -------------------- METRICS --------------------

metric_rows = []
ranking_rows = []

for qid in questions:
    for condition_name in conditions:
        ranking = rankings[(qid, condition_name)]

        relevant_ranks = []
        sufficient_ranks = []
        top1_relevant = 0
        top3_relevant = 0
        top1_sufficient = 0
        top3_sufficient = 0
        unannotated_top3 = 0
        annotated_top3 = 0

        for rank, (chunk, score) in enumerate(ranking, start=1):
            labels = label_lookup.get((qid, int(chunk)))
            annotated = labels is not None

            if rank <= 3:
                if annotated:
                    annotated_top3 += 1
                else:
                    unannotated_top3 += 1

            ranking_rows.append({
                "question_id": qid,
                "condition": condition_name,
                "rank": rank,
                "chunk_index": int(chunk),
                "rrf_score_or_original_rank": float(score),
                "annotated": annotated,
                "relevance_label": (
                    labels["relevance"] if annotated else np.nan
                ),
                "support_label": (
                    labels["support"] if annotated else np.nan
                ),
            })

            # Unannotated passages are UNKNOWN, not irrelevant.
            if not annotated:
                continue

            if labels["relevance"] == 1:
                relevant_ranks.append(rank)
                if rank == 1:
                    top1_relevant = 1
                if rank <= 3:
                    top3_relevant = 1

            if labels["support"] == 2:
                sufficient_ranks.append(rank)
                if rank == 1:
                    top1_sufficient = 1
                if rank <= 3:
                    top3_sufficient = 1

        metric_rows.append({
            "question_id": qid,
            "condition": condition_name,
            "candidate_count": len(ranking),
            "annotated_top3_count": annotated_top3,
            "unannotated_top3_count": unannotated_top3,
            "Relevant@1": top1_relevant,
            "Relevant@3": top3_relevant,
            "SufficientEvidence@1": top1_sufficient,
            "SufficientEvidence@3": top3_sufficient,
            "FirstRelevantRR@10": (
                1.0 / min(relevant_ranks) if relevant_ranks else 0.0
            ),
            "FirstSufficientRR@10": (
                1.0 / min(sufficient_ranks) if sufficient_ranks else 0.0
            ),
            "top3_chunk_ids": ",".join(
                str(chunk) for chunk, _ in ranking[:3]
            ),
        })

metrics = pd.DataFrame(metric_rows)
ranking_detail = pd.DataFrame(ranking_rows)

# Aggregate metrics. Coverage is provisional when unannotated candidates occur.
aggregate = (
    metrics.groupby("condition", sort=False)
    .agg(
        questions=("question_id", "nunique"),
        Relevant_at_1=("Relevant@1", "mean"),
        Relevant_at_3=("Relevant@3", "mean"),
        SufficientEvidence_at_1=("SufficientEvidence@1", "mean"),
        SufficientEvidence_at_3=("SufficientEvidence@3", "mean"),
        MRR_first_relevant_at_10=("FirstRelevantRR@10", "mean"),
        MRR_first_sufficient_at_10=("FirstSufficientRR@10", "mean"),
        questions_with_unannotated_top3=(
            "unannotated_top3_count",
            lambda s: int((s > 0).sum())
        ),
    )
    .reset_index()
)

wide = metrics.pivot(
    index="question_id",
    columns="condition",
    values=[
        "SufficientEvidence@1",
        "SufficientEvidence@3",
        "FirstSufficientRR@10",
        "unannotated_top3_count",
    ]
)

wide.columns = [f"{metric}_{condition.lower()}" for metric, condition in wide.columns]
wide = wide.reset_index()

for fused in ["fused_rrf_k10", "fused_rrf_k60"]:
    wide[f"top3_coverage_delta_{fused}"] = (
        wide[f"SufficientEvidence@3_{fused}"]
        - wide["SufficientEvidence@3_original"]
    )

# -------------------- SAVE ALL OUTPUTS --------------------

metrics.to_csv(OUT / "per_question_metrics.csv", index=False)
aggregate.to_csv(OUT / "aggregate_metrics.csv", index=False)
wide.to_csv(OUT / "per_question_comparison.csv", index=False)
ranking_detail.to_csv(
    OUT / "rankings_and_annotation_coverage.csv", index=False
)

with pd.ExcelWriter(OUT / "original_rewritten_fusion_results.xlsx") as writer:
    aggregate.to_excel(writer, sheet_name="Aggregate", index=False)
    wide.to_excel(writer, sheet_name="Per_Question", index=False)
    metrics.to_excel(writer, sheet_name="Metrics", index=False)
    ranking_detail.to_excel(writer, sheet_name="Rankings", index=False)

manifest = {
    "retrieval_file": str(RAW_PATH),
    "retrieval_sha256": hashlib.sha256(RAW_PATH.read_bytes()).hexdigest(),
    "annotation_file": str(ANNOTATION_PATH),
    "annotation_sha256": hashlib.sha256(
        ANNOTATION_PATH.read_bytes()
    ).hexdigest(),
    "question_count": len(questions),
    "annotation_pair_count": len(ann),
    "fusion_method": "RRF over saved ORIGINAL and REWRITTEN Hybrid RRF rankings",
    "fusion_constants": [10, 60],
    "support_success_rule": "support_label == 2",
    "relevance_success_rule": "relevance_label == 1",
    "unannotated_candidates": "Tracked as unknown, not assumed irrelevant",
    "created_utc": datetime.now(timezone.utc).isoformat(),
}
(OUT / "manifest.json").write_text(json.dumps(manifest, indent=2))

# -------------------- DISPLAY RESULTS --------------------

print("\n" + "=" * 65)
print("AGGREGATE METRICS")
print("=" * 65)
display(aggregate.round(3))

print("\n" + "=" * 65)
print("PER-QUESTION SUFFICIENT-EVIDENCE COMPARISON")
print("=" * 65)

display(wide[[
    "question_id",
    "SufficientEvidence@1_original",
    "SufficientEvidence@1_rewritten",
    "SufficientEvidence@1_fused_rrf_k10",
    "SufficientEvidence@1_fused_rrf_k60",
    "SufficientEvidence@3_original",
    "SufficientEvidence@3_rewritten",
    "SufficientEvidence@3_fused_rrf_k10",
    "SufficientEvidence@3_fused_rrf_k60",
    "top3_coverage_delta_fused_rrf_k10",
    "top3_coverage_delta_fused_rrf_k60",
]].round(3))

print("\n" + "=" * 65)
print("ANNOTATION COVERAGE CHECK")
print("=" * 65)

coverage = (
    metrics.groupby("condition")
    .agg(
        annotated_top3_passages=("annotated_top3_count", "sum"),
        unannotated_top3_passages=("unannotated_top3_count", "sum"),
        questions_with_unannotated_top3=(
            "unannotated_top3_count",
            lambda s: int((s > 0).sum())
        )
    )
    .reset_index()
)
display(coverage)

print("\nAll outputs saved to:", OUT)
print("Input files were not modified.")
print(
    "\nCaution: fused rankings may contain passages absent from the "
    "44-pair annotation workbook. Metrics are provisional until those "
    "unannotated top-three passages are reviewed."
)
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib
from datetime import datetime, timezone

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "query_expansion_pilot_v1"
)

FUSION = ROOT / "original_rewritten_fusion_pilot_v2"
EVIDENCE = ROOT / "evidence_review_v1"
OUT = FUSION / "corrected_v2"
OUT.mkdir(parents=True, exist_ok=True)

RANKINGS_FILE = FUSION / "rankings_and_annotation_coverage.csv"
BASE_ANNOTATIONS_FILE = EVIDENCE / "unique_question_passage_review_annotated.xlsx"
NEW_ANNOTATIONS_FILE = FUSION / "fused_unannotated_review_annotated_v1.csv"

# ---------- Load and validate inputs ----------
rankings = pd.read_csv(RANKINGS_FILE)
base = pd.read_excel(BASE_ANNOTATIONS_FILE)
new = pd.read_csv(NEW_ANNOTATIONS_FILE)

for df in (rankings, base, new):
    df["question_id"] = df["question_id"].astype(str).str.strip()
    df["chunk_index"] = pd.to_numeric(df["chunk_index"], errors="raise").astype(int)

for df in (base, new):
    df["relevance_label"] = pd.to_numeric(df["relevance_label"], errors="coerce")
    df["support_label"] = pd.to_numeric(df["support_label"], errors="coerce")

assert base[["relevance_label", "support_label"]].notna().all().all(), (
    "Some existing annotations are blank. Resolve them before evaluating."
)
assert new[["relevance_label", "support_label"]].notna().all().all(), (
    "Some newly reviewed labels are blank."
)
assert base["relevance_label"].isin([0, 1]).all()
assert new["relevance_label"].isin([0, 1]).all()
assert base["support_label"].isin([0, 1, 2]).all()
assert new["support_label"].isin([0, 1, 2]).all()

KEY = ["question_id", "chunk_index"]

# Existing annotations must be unique by question-passage pair.
if base.duplicated(KEY).any():
    raise ValueError("Duplicate question-passage keys in base annotations.")

if new.duplicated(KEY).any():
    raise ValueError("Duplicate question-passage keys in new annotations.")

# If any new pair already exists in the base, require labels to agree.
overlap = base.merge(
    new[KEY + ["relevance_label", "support_label"]],
    on=KEY,
    how="inner",
    suffixes=("_base", "_new")
)
if not overlap.empty:
    conflicts = overlap[
        (overlap["relevance_label_base"] != overlap["relevance_label_new"]) |
        (overlap["support_label_base"] != overlap["support_label_new"])
    ]
    if not conflicts.empty:
        raise ValueError(
            "Conflicting labels between annotation sources:\n"
            + conflicts[KEY + [
                "relevance_label_base", "relevance_label_new",
                "support_label_base", "support_label_new"
            ]].to_string(index=False)
        )

# Merge reviewed labels; new reviewed pairs take precedence if identical.
annotations = pd.concat(
    [
        base[KEY + ["relevance_label", "support_label"]],
        new[KEY + ["relevance_label", "support_label"]]
    ],
    ignore_index=True
).drop_duplicates(KEY, keep="last")

assert not annotations.duplicated(KEY).any()

# ---------- Attach labels to all saved rankings ----------
results = rankings.drop(
    columns=["annotated", "relevance_label", "support_label"],
    errors="ignore"
).merge(
    annotations,
    on=KEY,
    how="left",
    validate="many_to_one"
)

results["annotated"] = results["support_label"].notna() & results["relevance_label"].notna()
results = results.sort_values(["condition", "question_id", "rank"]).reset_index(drop=True)

# Save an audit trail of the complete ranked results with corrected labels.
results.to_csv(OUT / "rankings_with_corrected_annotations.csv", index=False)

# ---------- Calculate per-question metrics ----------
def reciprocal_rank(ranks):
    return 1.0 / min(ranks) if ranks else 0.0

question_rows = []

for (condition, qid), group in results.groupby(["condition", "question_id"], sort=True):
    group = group.sort_values("rank").copy()
    top1 = group.head(1)
    top3 = group.head(3)
    top10 = group.head(10)

    known_rel = top10[top10["relevance_label"].notna()]
    known_sup = top10[top10["support_label"].notna()]

    relevant_ranks = known_rel.loc[
        known_rel["relevance_label"] == 1, "rank"
    ].astype(int).tolist()

    sufficient_ranks = known_sup.loc[
        known_sup["support_label"] == 2, "rank"
    ].astype(int).tolist()

    def known_hit(subset, column, target):
        known = subset[column].notna()
        return bool((subset.loc[known, column] == target).any())

    # Lower-bound success: unannotated passages never count as success.
    # Separate coverage fields show where the metric remains uncertain.
    row = {
        "question_id": qid,
        "condition": condition,
        "candidate_count": len(group),
        "annotated_top1_count": int(top1["annotated"].sum()),
        "annotated_top3_count": int(top3["annotated"].sum()),
        "unannotated_top3_count": int((~top3["annotated"]).sum()),
        "annotated_top10_count": int(top10["annotated"].sum()),
        "unannotated_top10_count": int((~top10["annotated"]).sum()),
        "Relevant@1_lower_bound": int(known_hit(top1, "relevance_label", 1)),
        "Relevant@3_lower_bound": int(known_hit(top3, "relevance_label", 1)),
        "SufficientEvidence@1_lower_bound": int(known_hit(top1, "support_label", 2)),
        "SufficientEvidence@3_lower_bound": int(known_hit(top3, "support_label", 2)),
        "FirstRelevantRR@10": reciprocal_rank(relevant_ranks),
        "FirstSufficientRR@10": reciprocal_rank(sufficient_ranks),
        "top3_chunk_ids": ",".join(group.head(3)["chunk_index"].astype(str))
    }
    question_rows.append(row)

per_question = pd.DataFrame(question_rows)
per_question.to_csv(OUT / "per_question_metrics_corrected.csv", index=False)

metric_cols = [
    "Relevant@1_lower_bound",
    "Relevant@3_lower_bound",
    "SufficientEvidence@1_lower_bound",
    "SufficientEvidence@3_lower_bound",
    "FirstRelevantRR@10",
    "FirstSufficientRR@10",
]

aggregate = (
    per_question.groupby("condition")
    .agg(
        questions=("question_id", "nunique"),
        **{col: (col, "mean") for col in metric_cols},
        questions_with_unannotated_top3=("unannotated_top3_count",
                                          lambda x: int((x > 0).sum())),
        total_unannotated_top3=("unannotated_top3_count", "sum"),
        total_top3_occurrences=("annotated_top3_count", "size"),
        total_annotated_top3=("annotated_top3_count", "sum"),
        questions_with_unannotated_top10=("unannotated_top10_count",
                                           lambda x: int((x > 0).sum())),
    )
    .reset_index()
)

aggregate["top3_annotation_coverage"] = (
    aggregate["total_annotated_top3"] / (aggregate["questions"] * 3)
)
aggregate.to_csv(OUT / "aggregate_metrics_corrected.csv", index=False)

# Report every remaining unknown top-3 passage explicitly.
unknown_top3 = results[
    (results["rank"] <= 3) & (~results["annotated"])
].copy()
unknown_top3.to_csv(OUT / "remaining_unannotated_top3.csv", index=False)

# Save annotation merge and provenance manifest.
base_hash = hashlib.sha256(BASE_ANNOTATIONS_FILE.read_bytes()).hexdigest()
new_hash = hashlib.sha256(NEW_ANNOTATIONS_FILE.read_bytes()).hexdigest()
rank_hash = hashlib.sha256(RANKINGS_FILE.read_bytes()).hexdigest()

manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "source_rankings": str(RANKINGS_FILE),
    "source_rankings_sha256": rank_hash,
    "base_annotations": str(BASE_ANNOTATIONS_FILE),
    "base_annotations_sha256": base_hash,
    "new_fusion_annotations": str(NEW_ANNOTATIONS_FILE),
    "new_fusion_annotations_sha256": new_hash,
    "unique_annotated_question_passage_pairs": int(len(annotations)),
    "fusion_ranking_rows": int(len(results)),
    "support_success_rule": "support_label == 2",
    "relevance_success_rule": "relevance_label == 1",
    "unknown_labels": "Unannotated candidates are unknown; not labelled irrelevant",
    "metric_interpretation": (
        "Success metrics are lower bounds on annotated evidence. "
        "Coverage is reported separately; this is a 10-question exploratory pilot."
    )
}
(OUT / "manifest_corrected.json").write_text(json.dumps(manifest, indent=2))

print("CORRECTED EVALUATION SAVED TO:", OUT)
print("\nAGGREGATE METRICS")
display(aggregate.round(3))

print("\nREMAINING UNANNOTATED TOP-3 PASSAGES")
print("Count:", len(unknown_top3))
if len(unknown_top3):
    display(unknown_top3[
        ["question_id", "condition", "rank", "chunk_index"]
    ].sort_values(["question_id", "condition", "rank"]))

print("\nUnique reviewed question-passage pairs:", len(annotations))
print("Results written; original experiment files were not overwritten.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib
from datetime import datetime, timezone

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "query_expansion_pilot_v1"
)

FUSION = ROOT / "original_rewritten_fusion_pilot_v2"
EVIDENCE = ROOT / "evidence_review_v1"
OUT = FUSION / "corrected_v2"
OUT.mkdir(parents=True, exist_ok=True)

RANKINGS_FILE = FUSION / "rankings_and_annotation_coverage.csv"
BASE_ANNOTATIONS_FILE = EVIDENCE / "unique_question_passage_review_annotated.xlsx"
NEW_ANNOTATIONS_FILE = FUSION / "fused_unannotated_review_annotated_v1.csv"

# ---------- Load and validate inputs ----------
rankings = pd.read_csv(RANKINGS_FILE)
base = pd.read_excel(BASE_ANNOTATIONS_FILE)
new = pd.read_csv(NEW_ANNOTATIONS_FILE)

for df in (rankings, base, new):
    df["question_id"] = df["question_id"].astype(str).str.strip()
    df["chunk_index"] = pd.to_numeric(df["chunk_index"], errors="raise").astype(int)

for df in (base, new):
    df["relevance_label"] = pd.to_numeric(df["relevance_label"], errors="coerce")
    df["support_label"] = pd.to_numeric(df["support_label"], errors="coerce")

assert base[["relevance_label", "support_label"]].notna().all().all(), (
    "Some existing annotations are blank. Resolve them before evaluating."
)
assert new[["relevance_label", "support_label"]].notna().all().all(), (
    "Some newly reviewed labels are blank."
)
assert base["relevance_label"].isin([0, 1]).all()
assert new["relevance_label"].isin([0, 1]).all()
assert base["support_label"].isin([0, 1, 2]).all()
assert new["support_label"].isin([0, 1, 2]).all()

KEY = ["question_id", "chunk_index"]

# Existing annotations must be unique by question-passage pair.
if base.duplicated(KEY).any():
    raise ValueError("Duplicate question-passage keys in base annotations.")

if new.duplicated(KEY).any():
    raise ValueError("Duplicate question-passage keys in new annotations.")

# If any new pair already exists in the base, require labels to agree.
overlap = base.merge(
    new[KEY + ["relevance_label", "support_label"]],
    on=KEY,
    how="inner",
    suffixes=("_base", "_new")
)
if not overlap.empty:
    conflicts = overlap[
        (overlap["relevance_label_base"] != overlap["relevance_label_new"]) |
        (overlap["support_label_base"] != overlap["support_label_new"])
    ]
    if not conflicts.empty:
        raise ValueError(
            "Conflicting labels between annotation sources:\n"
            + conflicts[KEY + [
                "relevance_label_base", "relevance_label_new",
                "support_label_base", "support_label_new"
            ]].to_string(index=False)
        )

# Merge reviewed labels; new reviewed pairs take precedence if identical.
annotations = pd.concat(
    [
        base[KEY + ["relevance_label", "support_label"]],
        new[KEY + ["relevance_label", "support_label"]]
    ],
    ignore_index=True
).drop_duplicates(KEY, keep="last")

assert not annotations.duplicated(KEY).any()

# ---------- Attach labels to all saved rankings ----------
results = rankings.drop(
    columns=["annotated", "relevance_label", "support_label"],
    errors="ignore"
).merge(
    annotations,
    on=KEY,
    how="left",
    validate="many_to_one"
)

results["annotated"] = results["support_label"].notna() & results["relevance_label"].notna()
results = results.sort_values(["condition", "question_id", "rank"]).reset_index(drop=True)

# Save an audit trail of the complete ranked results with corrected labels.
results.to_csv(OUT / "rankings_with_corrected_annotations.csv", index=False)

# ---------- Calculate per-question metrics ----------
def reciprocal_rank(ranks):
    return 1.0 / min(ranks) if ranks else 0.0

question_rows = []

for (condition, qid), group in results.groupby(["condition", "question_id"], sort=True):
    group = group.sort_values("rank").copy()
    top1 = group.head(1)
    top3 = group.head(3)
    top10 = group.head(10)

    known_rel = top10[top10["relevance_label"].notna()]
    known_sup = top10[top10["support_label"].notna()]

    relevant_ranks = known_rel.loc[
        known_rel["relevance_label"] == 1, "rank"
    ].astype(int).tolist()

    sufficient_ranks = known_sup.loc[
        known_sup["support_label"] == 2, "rank"
    ].astype(int).tolist()

    def known_hit(subset, column, target):
        known = subset[column].notna()
        return bool((subset.loc[known, column] == target).any())

    # Lower-bound success: unannotated passages never count as success.
    # Separate coverage fields show where the metric remains uncertain.
    row = {
        "question_id": qid,
        "condition": condition,
        "candidate_count": len(group),
        "annotated_top1_count": int(top1["annotated"].sum()),
        "annotated_top3_count": int(top3["annotated"].sum()),
        "unannotated_top3_count": int((~top3["annotated"]).sum()),
        "annotated_top10_count": int(top10["annotated"].sum()),
        "unannotated_top10_count": int((~top10["annotated"]).sum()),
        "Relevant@1_lower_bound": int(known_hit(top1, "relevance_label", 1)),
        "Relevant@3_lower_bound": int(known_hit(top3, "relevance_label", 1)),
        "SufficientEvidence@1_lower_bound": int(known_hit(top1, "support_label", 2)),
        "SufficientEvidence@3_lower_bound": int(known_hit(top3, "support_label", 2)),
        "FirstRelevantRR@10": reciprocal_rank(relevant_ranks),
        "FirstSufficientRR@10": reciprocal_rank(sufficient_ranks),
        "top3_chunk_ids": ",".join(group.head(3)["chunk_index"].astype(str))
    }
    question_rows.append(row)

per_question = pd.DataFrame(question_rows)
per_question.to_csv(OUT / "per_question_metrics_corrected.csv", index=False)

metric_cols = [
    "Relevant@1_lower_bound",
    "Relevant@3_lower_bound",
    "SufficientEvidence@1_lower_bound",
    "SufficientEvidence@3_lower_bound",
    "FirstRelevantRR@10",
    "FirstSufficientRR@10",
]

aggregate = (
    per_question.groupby("condition")
    .agg(
        questions=("question_id", "nunique"),
        **{col: (col, "mean") for col in metric_cols},
        questions_with_unannotated_top3=("unannotated_top3_count",
                                          lambda x: int((x > 0).sum())),
        total_unannotated_top3=("unannotated_top3_count", "sum"),
        total_top3_occurrences=("annotated_top3_count", "size"),
        total_annotated_top3=("annotated_top3_count", "sum"),
        questions_with_unannotated_top10=("unannotated_top10_count",
                                           lambda x: int((x > 0).sum())),
    )
    .reset_index()
)

aggregate["top3_annotation_coverage"] = (
    aggregate["total_annotated_top3"] / (aggregate["questions"] * 3)
)
aggregate.to_csv(OUT / "aggregate_metrics_corrected.csv", index=False)

# Report every remaining unknown top-3 passage explicitly.
unknown_top3 = results[
    (results["rank"] <= 3) & (~results["annotated"])
].copy()
unknown_top3.to_csv(OUT / "remaining_unannotated_top3.csv", index=False)

# Save annotation merge and provenance manifest.
base_hash = hashlib.sha256(BASE_ANNOTATIONS_FILE.read_bytes()).hexdigest()
new_hash = hashlib.sha256(NEW_ANNOTATIONS_FILE.read_bytes()).hexdigest()
rank_hash = hashlib.sha256(RANKINGS_FILE.read_bytes()).hexdigest()

manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "source_rankings": str(RANKINGS_FILE),
    "source_rankings_sha256": rank_hash,
    "base_annotations": str(BASE_ANNOTATIONS_FILE),
    "base_annotations_sha256": base_hash,
    "new_fusion_annotations": str(NEW_ANNOTATIONS_FILE),
    "new_fusion_annotations_sha256": new_hash,
    "unique_annotated_question_passage_pairs": int(len(annotations)),
    "fusion_ranking_rows": int(len(results)),
    "support_success_rule": "support_label == 2",
    "relevance_success_rule": "relevance_label == 1",
    "unknown_labels": "Unannotated candidates are unknown; not labelled irrelevant",
    "metric_interpretation": (
        "Success metrics are lower bounds on annotated evidence. "
        "Coverage is reported separately; this is a 10-question exploratory pilot."
    )
}
(OUT / "manifest_corrected.json").write_text(json.dumps(manifest, indent=2))

print("CORRECTED EVALUATION SAVED TO:", OUT)
print("\nAGGREGATE METRICS")
display(aggregate.round(3))

print("\nREMAINING UNANNOTATED TOP-3 PASSAGES")
print("Count:", len(unknown_top3))
if len(unknown_top3):
    display(unknown_top3[
        ["question_id", "condition", "rank", "chunk_index"]
    ].sort_values(["question_id", "condition", "rank"]))

print("\nUnique reviewed question-passage pairs:", len(annotations))
print("Results written; original experiment files were not overwritten.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import numpy as np
import hashlib
import json
import shutil

# ============================================================
# PATHS
# ============================================================
ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "query_expansion_pilot_v1"
)
FUSION = ROOT / "original_rewritten_fusion_pilot_v2"
CORRECTED = FUSION / "corrected_v2"

RAW = ROOT / "raw_query_expansion_results.csv"
BASE_ANN = ROOT / "evidence_review_v1" / "unique_question_passage_review_annotated.xlsx"
NEW_ANN = FUSION / "fused_unannotated_review_annotated_v1.csv"
RANKINGS = FUSION / "rankings_and_annotation_coverage.csv"

# New independent workspace. Do not overwrite the original benchmark.
WORKSPACE = ROOT / "research_freeze_and_next_benchmark_v1"
FROZEN = WORKSPACE / "frozen_pilot"
TOP10 = WORKSPACE / "top10_evidence_audit"
BENCH = WORKSPACE / "benchmark_v3_75q"
for p in [WORKSPACE, FROZEN, TOP10, BENCH]:
    p.mkdir(parents=True, exist_ok=True)

required = [RAW, BASE_ANN, NEW_ANN, RANKINGS,
            CORRECTED / "aggregate_metrics_corrected.csv",
            CORRECTED / "per_question_metrics_corrected.csv"]
missing = [str(p) for p in required if not p.exists()]
if missing:
    raise FileNotFoundError("Required input files missing:\n" + "\n".join(missing))

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

# ============================================================
# 1. FREEZE PILOT RESULTS: COPY, DON'T MODIFY THE ORIGINAL
# ============================================================
pilot_files = [
    CORRECTED / "aggregate_metrics_corrected.csv",
    CORRECTED / "per_question_metrics_corrected.csv",
    CORRECTED / "rankings_with_corrected_annotations.csv",
    CORRECTED / "remaining_unannotated_top3.csv",
    CORRECTED / "manifest_corrected.json",
    NEW_ANN,
    RANKINGS,
]

frozen_manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "purpose": "Immutable snapshot of exploratory query-expansion/fusion pilot",
    "pilot_question_count": 10,
    "warning": (
        "Exploratory pilot only. Query rewrites were manually tailored. "
        "Do not claim generalization from this pilot."
    ),
    "files": {}
}

for src in pilot_files:
    if src.exists():
        dst = FROZEN / src.name
        shutil.copy2(src, dst)
        frozen_manifest["files"][src.name] = {
            "sha256": sha256(dst),
            "bytes": dst.stat().st_size
        }

(FROZEN / "freeze_manifest.json").write_text(
    json.dumps(frozen_manifest, indent=2)
)

# ============================================================
# 2. BUILD A TOP-10 EVIDENCE AUDIT QUEUE
#    Conditions: ORIGINAL, REWRITTEN, FUSED_RRF_K10, FUSED_RRF_K60
#    Reuse labels when already available; leave new ones blank.
# ============================================================
raw = pd.read_csv(RAW)
base = pd.read_excel(BASE_ANN)
new = pd.read_csv(NEW_ANN)

for df in [raw, base, new]:
    df["question_id"] = df["question_id"].astype(str).str.strip()
    df["chunk_index"] = pd.to_numeric(df["chunk_index"], errors="raise").astype(int)

base["relevance_label"] = pd.to_numeric(base["relevance_label"], errors="coerce")
base["support_label"] = pd.to_numeric(base["support_label"], errors="coerce")
new["relevance_label"] = pd.to_numeric(new["relevance_label"], errors="coerce")
new["support_label"] = pd.to_numeric(new["support_label"], errors="coerce")

key = ["question_id", "chunk_index"]
label_cols = key + ["relevance_label", "support_label", "review_reason"]

labels = pd.concat([
    base[label_cols],
    new[label_cols]
], ignore_index=True)

# Detect conflicting annotations rather than silently choosing one.
conflicts = (
    labels.dropna(subset=["relevance_label", "support_label"])
    .groupby(key)[["relevance_label", "support_label"]]
    .nunique()
)
if ((conflicts > 1).any(axis=1)).any():
    bad = conflicts[(conflicts > 1).any(axis=1)]
    raise ValueError("Conflicting evidence labels detected:\n" + bad.to_string())

labels = labels.drop_duplicates(key, keep="last")

# Use saved rankings for ORIGINAL/REWRITTEN and both fused variants.
rankings = pd.read_csv(RANKINGS)
rankings["question_id"] = rankings["question_id"].astype(str).str.strip()
rankings["chunk_index"] = pd.to_numeric(rankings["chunk_index"], errors="raise").astype(int)

# Keep top 10 per question-condition.
top10 = rankings[rankings["rank"] <= 10].copy()
top10 = top10.merge(labels, on=key, how="left", suffixes=("", "_label"))

# Add passage text/source metadata from raw retrieval results.
# A chunk's source/text is stable across query conditions; deduplicate by key.
raw_meta = raw.sort_values("rank").drop_duplicates(key)[
    key + ["source_file", "pages", "chunk_text"]
]
top10 = top10.merge(raw_meta, on=key, how="left", suffixes=("", "_raw"))

top10["annotation_status"] = np.where(
    top10["support_label"].notna() & top10["relevance_label"].notna(),
    "EXISTING_LABEL_REUSED",
    "NEEDS_REVIEW"
)
top10["reviewer_id_or_initials"] = ""
top10["review_date"] = ""
top10["review_notes"] = ""

# Reorder columns for a convenient review sheet.
ordered = [
    "question_id", "condition", "rank", "chunk_index",
    "source_file", "pages", "chunk_text",
    "relevance_label", "support_label", "review_reason",
    "annotation_status", "reviewer_id_or_initials",
    "review_date", "review_notes"
]
top10 = top10[[c for c in ordered if c in top10.columns]]
top10 = top10.sort_values(["question_id", "condition", "rank"])

audit_csv = TOP10 / "top10_evidence_review_queue.csv"
audit_xlsx = TOP10 / "top10_evidence_review_queue.xlsx"
top10.to_csv(audit_csv, index=False)

with pd.ExcelWriter(audit_xlsx, engine="openpyxl") as writer:
    top10.to_excel(writer, sheet_name="Top10_Review", index=False)
    summary = (
        top10.groupby("condition")
        .agg(
            questions=("question_id", "nunique"),
            ranked_passages=("chunk_index", "size"),
            already_labelled=("annotation_status",
                              lambda s: int((s == "EXISTING_LABEL_REUSED").sum())),
            needing_review=("annotation_status",
                            lambda s: int((s == "NEEDS_REVIEW").sum()))
        ).reset_index()
    )
    summary.to_excel(writer, sheet_name="Coverage_Summary", index=False)

# ============================================================
# 3. CREATE 75-QUESTION BENCHMARK TEMPLATE
#    No synthetic question text or gold answers are fabricated.
# ============================================================
types = (
    ["DIRECT"] * 25 +
    ["PARAPHRASE"] * 20 +
    ["MULTI_PAGE"] * 10 +
    ["HARD_DISTRACTOR"] * 10 +
    ["UNANSWERABLE"] * 10
)

bench = pd.DataFrame({
    "question_id": [f"V3Q{i:03d}" for i in range(1, 76)],
    "question_type": types,
    "question_text": "",
    "class": "",
    "subject": "",
    "book_title": "",
    "gold_source_file": "",
    "gold_printed_pages": "",
    "gold_pdf_pages": "",
    "expected_answer": "",
    "evidence_excerpt_or_notes": "",
    "answerable_from_corpus": "",
    "required_pages_count": "",
    "distractor_source_file": "",
    "annotation_status": "NOT_ANNOTATED",
    "split": "",
    "reviewer_id_or_initials": "",
    "review_notes": ""
})

# Deterministic question-level split, stratified by type:
# 60 development/tuning questions, 15 held-out test questions.
# The held-out test set must not be used for prompt/query/fusion tuning.
rng = np.random.default_rng(20261010)
for qtype in ["DIRECT", "PARAPHRASE", "MULTI_PAGE",
              "HARD_DISTRACTOR", "UNANSWERABLE"]:
    idx = bench.index[bench["question_type"] == qtype].to_numpy()
    idx = rng.permutation(idx)
    n_test = max(1, round(len(idx) * 0.20))
    bench.loc[idx[:n_test], "split"] = "TEST_HELD_OUT"
    bench.loc[idx[n_test:], "split"] = "DEV_TUNING"

bench_path = BENCH / "NCERT_RAG_Benchmark_V3_75Q_TEMPLATE.xlsx"
with pd.ExcelWriter(bench_path, engine="openpyxl") as writer:
    bench.to_excel(writer, sheet_name="Questions", index=False)
    (
        bench.groupby(["question_type", "split"])
        .size().unstack(fill_value=0).reset_index()
    ).to_excel(writer, sheet_name="Split_Summary", index=False)

bench.to_csv(BENCH / "NCERT_RAG_Benchmark_V3_75Q_TEMPLATE.csv", index=False)

# ============================================================
# 4. ANNOTATION PROTOCOL / EXPERIMENT GUARDRAILS
# ============================================================
protocol = """NCERT RAG EVALUATION — REVIEW PROTOCOL

A. TOP-10 PASSAGE LABELS
- relevance_label: 1 = relevant to answering the question; 0 = not relevant.
- support_label: 0 = no support; 1 = partial support; 2 = sufficient evidence
  to support the expected answer.
- Keep unknown labels blank. Never infer that unreviewed means irrelevant.
- Review the passage in the context of the question and expected answer.
- For multi-page questions, judge whether the retrieved evidence covers the
  required parts, not merely whether it mentions the topic.
- Preserve printed page numbers separately from PDF page indices.
- Record ambiguous cases in review_notes and seek a second reviewer if possible.

B. BENCHMARK V3
Target distribution: 25 DIRECT, 20 PARAPHRASE, 10 MULTI_PAGE,
10 HARD_DISTRACTOR, 10 UNANSWERABLE.
- Write natural questions grounded in the actual NCERT source.
- Record exact source file, printed page(s), and PDF page index/indices separately.
- For UNANSWERABLE, document what was searched and why the corpus does not
  contain enough evidence. Do not use obscure wording as a substitute.
- MULTI_PAGE questions must genuinely require evidence from more than one page.
- HARD_DISTRACTOR questions should have plausible same-topic distractors.
- Check answerability, expected answer, and evidence independently where possible.

C. EXPERIMENT DESIGN
- The current 10-question pilot is exploratory. Its manually tailored query rewrites
  make it unsuitable as an unbiased estimate of query-rewriting performance.
- Tune prompts, query rewriting, retrieval settings, and fusion only on DEV_TUNING.
- Freeze code/configuration before running TEST_HELD_OUT.
- Keep test questions and gold labels out of tuning and prompt development.
- Report retrieval relevance, evidence sufficiency, answer correctness, groundedness,
  and abstention performance separately.
- Do not claim educational learning gains from retrieval metrics alone.
- Report annotation coverage alongside every evidence metric.
"""

(BENCH / "ANNOTATION_AND_EVALUATION_PROTOCOL.txt").write_text(protocol)

manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "workspace": str(WORKSPACE),
    "pilot_snapshot": str(FROZEN),
    "top10_review_queue_csv": str(audit_csv),
    "top10_review_queue_xlsx": str(audit_xlsx),
    "benchmark_template_xlsx": str(bench_path),
    "benchmark_template_csv": str(BENCH / "NCERT_RAG_Benchmark_V3_75Q_TEMPLATE.csv"),
    "question_type_counts": bench["question_type"].value_counts().to_dict(),
    "split_counts": bench["split"].value_counts().to_dict(),
    "inputs_sha256": {
        str(p): sha256(p) for p in required
    },
    "important_limitations": [
        "The 75-question workbook is a blank template, not a completed benchmark.",
        "The split is a planning split; replace it with a source-reviewed split before final evaluation.",
        "Question text and gold answers must be manually grounded in the NCERT books.",
        "The pilot uses manually tailored query rewrites and has only 10 questions."
    ]
}
(WORKSPACE / "workspace_manifest.json").write_text(json.dumps(manifest, indent=2))

# ============================================================
# 5. SUMMARY
# ============================================================
print("\nSUCCESS — RESEARCH WORKSPACE CREATED")
print("Workspace:", WORKSPACE)
print("\n1) Frozen pilot snapshot:", FROZEN)
print("2) Top-10 review workbook:", audit_xlsx)
print("3) 75-question benchmark template:", bench_path)
print("4) Protocol:", BENCH / "ANNOTATION_AND_EVALUATION_PROTOCOL.txt")

print("\nTOP-10 AUDIT COVERAGE")
display(
    top10.groupby("condition")
    .agg(
        questions=("question_id", "nunique"),
        passages=("chunk_index", "size"),
        already_labelled=("annotation_status",
                          lambda s: int((s == "EXISTING_LABEL_REUSED").sum())),
        needing_review=("annotation_status",
                        lambda s: int((s == "NEEDS_REVIEW").sum()))
    ).reset_index()
)

print("\n75-QUESTION TEMPLATE DISTRIBUTION")
display(
    bench.groupby(["question_type", "split"])
    .size().unstack(fill_value=0).reset_index()
)

print("\nNo original experiment files or benchmark files were overwritten.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib
from datetime import datetime

# ------------------------------------------------------------
# 1. Paths: inputs remain untouched
# ------------------------------------------------------------
BASE = Path("/content/drive/MyDrive/NCERT_BOOKS/06_RECONCILIATION_INPUTS")

QUEUE_FILE = BASE / "top10_evidence_review_queue.xlsx"
ANNOTATION_FILE = BASE / "unique_question_passage_review_corrected.xlsx"

OUT = BASE / "reconciliation_output_v1"
OUT.mkdir(parents=True, exist_ok=True)

for p in [QUEUE_FILE, ANNOTATION_FILE]:
    if not p.exists():
        raise FileNotFoundError(f"Missing input: {p}")

# ------------------------------------------------------------
# 2. Load and validate the expected sheets and columns
# ------------------------------------------------------------
queue = pd.read_excel(QUEUE_FILE, sheet_name="Top10_Review")
ann = pd.read_excel(ANNOTATION_FILE, sheet_name="Passage_Review")

required_queue = {
    "question_id", "condition", "rank", "chunk_index",
    "relevance_label", "support_label", "annotation_status"
}
required_ann = {
    "question_id", "chunk_index",
    "relevance_label", "support_label"
}

if not required_queue.issubset(queue.columns):
    raise ValueError(f"Queue is missing columns: {required_queue - set(queue.columns)}")
if not required_ann.issubset(ann.columns):
    raise ValueError(f"Annotation workbook is missing columns: {required_ann - set(ann.columns)}")

# ------------------------------------------------------------
# 3. Normalize IDs so 76, 76.0, and "76" match
# ------------------------------------------------------------
def norm_id(x):
    if pd.isna(x):
        return ""
    s = str(x).strip()
    try:
        f = float(s)
        if f.is_integer():
            return str(int(f))
    except (ValueError, TypeError):
        pass
    return s

for df in [queue, ann]:
    df["question_id"] = df["question_id"].astype(str).str.strip()
    df["_chunk_key"] = df["chunk_index"].map(norm_id)

queue["_pair_key"] = queue["question_id"] + "||" + queue["_chunk_key"]
ann["_pair_key"] = ann["question_id"] + "||" + ann["_chunk_key"]

for col in ["relevance_label", "support_label"]:
    queue[col] = pd.to_numeric(queue[col], errors="coerce")
    ann[col] = pd.to_numeric(ann[col], errors="coerce")

# Annotation workbook should ideally contain one row per question-passage pair.
ann_duplicate_rows = ann[ann.duplicated("_pair_key", keep=False)].copy()

# Build one annotation record per pair, preserving internal disagreements.
ann_records = []
for pair_key, group in ann.groupby("_pair_key", dropna=False):
    rel_values = sorted(group["relevance_label"].dropna().unique().tolist())
    sup_values = sorted(group["support_label"].dropna().unique().tolist())

    ann_records.append({
        "_pair_key": pair_key,
        "ann_question_id": group["question_id"].iloc[0],
        "ann_chunk_index": group["chunk_index"].iloc[0],
        "ann_relevance": rel_values[0] if len(rel_values) == 1 else np.nan,
        "ann_support": sup_values[0] if len(sup_values) == 1 else np.nan,
        "ann_duplicate_count": len(group),
        "ann_internal_conflict": len(rel_values) > 1 or len(sup_values) > 1,
        "ann_relevance_values": str(rel_values),
        "ann_support_values": str(sup_values),
        "ann_review_reason": " | ".join(
            group.get("review_reason", pd.Series(dtype=str))
                 .dropna().astype(str).unique().tolist()
        ),
        "ann_reviewer": " | ".join(
            group.get("reviewer_id_or_initials", pd.Series(dtype=str))
                 .dropna().astype(str).unique().tolist()
        )
    })

ann_pairs = pd.DataFrame(ann_records)

# ------------------------------------------------------------
# 4. Merge annotations onto every ranked occurrence
# ------------------------------------------------------------
merged = queue.merge(ann_pairs, on="_pair_key", how="left", validate="many_to_one")

def labels_equal(a, b):
    if pd.isna(a) or pd.isna(b):
        return True  # missing labels are not treated as disagreements
    return float(a) == float(b)

statuses = []
effective_rel = []
effective_sup = []
conflict_reasons = []

for _, row in merged.iterrows():
    q_rel, q_sup = row["relevance_label"], row["support_label"]
    a_rel, a_sup = row["ann_relevance"], row["ann_support"]

    has_ann = pd.notna(row.get("ann_duplicate_count"))
    queue_has_labels = pd.notna(q_rel) and pd.notna(q_sup)
    internal_conflict = bool(row.get("ann_internal_conflict", False)) if has_ann else False

    reasons = []
    if internal_conflict:
        reasons.append("Duplicate annotation rows disagree internally")

    if has_ann:
        if not labels_equal(q_rel, a_rel):
            reasons.append(f"Relevance differs: queue={q_rel}, corrected={a_rel}")
        if not labels_equal(q_sup, a_sup):
            reasons.append(f"Support differs: queue={q_sup}, corrected={a_sup}")

    if reasons:
        status = "CONFLICT_REQUIRES_REVIEW"
        use_rel, use_sup = np.nan, np.nan
    elif has_ann and pd.notna(a_rel) and pd.notna(a_sup):
        status = "MATCHED_OR_CORRECTED_ANNOTATION"
        use_rel, use_sup = a_rel, a_sup
    elif queue_has_labels:
        status = "QUEUE_ONLY_LABELS_PRESERVED"
        use_rel, use_sup = q_rel, q_sup
    else:
        status = "NEEDS_ANNOTATION"
        use_rel, use_sup = np.nan, np.nan

    statuses.append(status)
    effective_rel.append(use_rel)
    effective_sup.append(use_sup)
    conflict_reasons.append("; ".join(reasons))

merged["reconciliation_status"] = statuses
merged["effective_relevance_label"] = effective_rel
merged["effective_support_label"] = effective_sup
merged["conflict_details"] = conflict_reasons

# ------------------------------------------------------------
# 5. Create a deduplicated pair-level review table
# ------------------------------------------------------------
pair_rows = []

for pair_key, group in merged.groupby("_pair_key", dropna=False):
    first = group.iloc[0]

    pair_rows.append({
        "question_id": first["question_id"],
        "chunk_index": first["chunk_index"],
        "source_file": first.get("source_file", ""),
        "chunk_text": first.get("chunk_text", ""),
        "conditions_seen": ", ".join(sorted(group["condition"].dropna().astype(str).unique())),
        "occurrence_count": len(group),
        "best_rank_seen": pd.to_numeric(group["rank"], errors="coerce").min(),
        "queue_relevance_values": str(sorted(group["relevance_label"].dropna().unique().tolist())),
        "queue_support_values": str(sorted(group["support_label"].dropna().unique().tolist())),
        "corrected_relevance_label": first.get("ann_relevance", np.nan),
        "corrected_support_label": first.get("ann_support", np.nan),
        "reconciliation_status": (
            "CONFLICT_REQUIRES_REVIEW"
            if (group["reconciliation_status"] == "CONFLICT_REQUIRES_REVIEW").any()
            else (
                "NEEDS_ANNOTATION"
                if (group["reconciliation_status"] == "NEEDS_ANNOTATION").all()
                else (
                    "QUEUE_ONLY_LABELS_PRESERVED"
                    if (group["reconciliation_status"] == "QUEUE_ONLY_LABELS_PRESERVED").all()
                    else "MATCHED_OR_CORRECTED_ANNOTATION"
                )
            )
        ),
        "conflict_details": " | ".join(
            group["conflict_details"].dropna().astype(str).unique().tolist()
        ),
        "review_reason": first.get("ann_review_reason", ""),
        "reviewer": first.get("ann_reviewer", "")
    })

pairs = pd.DataFrame(pair_rows)

# Annotation pairs that never appear in the top-10 queue
queue_keys = set(queue["_pair_key"])
annotation_only = ann_pairs[~ann_pairs["_pair_key"].isin(queue_keys)].copy()

# Unique passage pairs genuinely requiring work
needs_review = pairs[
    pairs["reconciliation_status"].isin([
        "NEEDS_ANNOTATION", "CONFLICT_REQUIRES_REVIEW"
    ])
].copy()

# ------------------------------------------------------------
# 6. Summary and input hashes
# ------------------------------------------------------------
summary = pd.DataFrame([
    {"metric": "Queue ranked occurrences", "value": len(queue)},
    {"metric": "Unique question-passage pairs in queue", "value": queue["_pair_key"].nunique()},
    {"metric": "Rows in corrected annotation workbook", "value": len(ann)},
    {"metric": "Unique pairs in corrected annotation workbook", "value": ann["_pair_key"].nunique()},
    {"metric": "Duplicate annotation rows", "value": len(ann_duplicate_rows)},
    {"metric": "Annotation pairs absent from queue", "value": len(annotation_only)},
    {"metric": "Unique pairs needing annotation", "value": int((pairs["reconciliation_status"] == "NEEDS_ANNOTATION").sum())},
    {"metric": "Unique pairs with label conflicts", "value": int((pairs["reconciliation_status"] == "CONFLICT_REQUIRES_REVIEW").sum())},
    {"metric": "Unique pairs matched/corrected", "value": int((pairs["reconciliation_status"] == "MATCHED_OR_CORRECTED_ANNOTATION").sum())},
    {"metric": "Unique pairs retaining queue-only labels", "value": int((pairs["reconciliation_status"] == "QUEUE_ONLY_LABELS_PRESERVED").sum())},
])

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

manifest = {
    "created_at": datetime.now().isoformat(timespec="seconds"),
    "queue_file": QUEUE_FILE.name,
    "queue_sha256": sha256(QUEUE_FILE),
    "annotation_file": ANNOTATION_FILE.name,
    "annotation_sha256": sha256(ANNOTATION_FILE),
    "note": "Reconciliation only. Original input files were not modified. Conflicts are deliberately left unresolved."
}

# ------------------------------------------------------------
# 7. Save everything to a new folder
# ------------------------------------------------------------
occurrence_out = merged.drop(columns=["_chunk_key", "_pair_key"], errors="ignore")
pair_out = pairs.copy()

with pd.ExcelWriter(OUT / "reconciliation_report_v1.xlsx", engine="openpyxl") as writer:
    summary.to_excel(writer, sheet_name="Summary", index=False)
    pair_out.to_excel(writer, sheet_name="Unique_Pairs", index=False)
    needs_review.to_excel(writer, sheet_name="Needs_Review", index=False)
    merged[merged["reconciliation_status"] == "CONFLICT_REQUIRES_REVIEW"].to_excel(
        writer, sheet_name="Conflicts", index=False
    )
    annotation_only.to_excel(writer, sheet_name="Annotation_Only", index=False)
    ann_duplicate_rows.to_excel(writer, sheet_name="Duplicate_Annotations", index=False)

occurrence_out.to_csv(OUT / "reconciled_occurrences_v1.csv", index=False)
pairs.to_csv(OUT / "unique_pairs_reconciliation_v1.csv", index=False)
needs_review.to_csv(OUT / "deduplicated_review_queue_v1.csv", index=False)
with open(OUT / "manifest_v1.json", "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2)

print("\n=== RECONCILIATION COMPLETE ===")
print(summary.to_string(index=False))
print(f"\nOutput folder: {OUT}")
print("Created:")
for p in sorted(OUT.iterdir()):
    print(" -", p.name)

print("\nIMPORTANT:")
print("Conflicts have NOT been silently resolved.")
print("Do not use conflicting rows in final metrics until reviewed.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import numpy as np

BASE = Path("/content/drive/MyDrive/NCERT_BOOKS/06_RECONCILIATION_INPUTS")
OUT = BASE / "reconciliation_output_v1"

QUEUE_FILE = BASE / "top10_evidence_review_queue.xlsx"
ANNOTATION_FILE = BASE / "unique_question_passage_review_corrected.xlsx"
PAIR_FILE = OUT / "unique_pairs_reconciliation_v1.csv"

queue = pd.read_excel(QUEUE_FILE, sheet_name="Top10_Review")
ann = pd.read_excel(ANNOTATION_FILE, sheet_name="Passage_Review")
pairs = pd.read_csv(PAIR_FILE)

def find_col(df, candidates):
    lookup = {str(c).strip().lower(): c for c in df.columns}
    for name in candidates:
        if name.lower() in lookup:
            return lookup[name.lower()]
    return None

def norm(x):
    if pd.isna(x):
        return ""
    s = str(x).strip()
    try:
        n = float(s)
        if n.is_integer():
            return str(int(n))
    except (ValueError, TypeError):
        pass
    return s

# Identify question and passage fields without assuming every optional
# column has the same name in every workbook.
qcol = find_col(queue, ["question_id"])
ccol = find_col(queue, ["chunk_index"])
textcol = find_col(queue, ["chunk_text", "passage_text", "text", "evidence_excerpt"])
sourcecol = find_col(queue, ["source_file", "retrieved_source_file"])
questioncol = find_col(queue, ["question", "original_question"])
answercol = find_col(queue, ["expected_answer", "gold_answer", "answer"])
rankcol = find_col(queue, ["rank", "original_rank"])
conditioncol = find_col(queue, ["condition", "method"])

if not qcol or not ccol or not textcol:
    print("Queue columns:", queue.columns.tolist())
    raise ValueError(
        "Could not identify question ID, chunk ID and passage text. "
        "Share the printed column names so we can map them safely."
    )

# Deduplicate queue occurrences using question + chunk.
queue["_qkey"] = queue[qcol].astype(str).str.strip()
queue["_ckey"] = queue[ccol].map(norm)
queue["_pair"] = queue["_qkey"] + "||" + queue["_ckey"]

ann["_qkey"] = ann["question_id"].astype(str).str.strip()
ann["_ckey"] = ann["chunk_index"].map(norm)
ann["_pair"] = ann["_qkey"] + "||" + ann["_ckey"]

reviewed_keys = set(ann["_pair"])
reviewed_keys_conflicted = set(
    pairs.loc[
        pairs["reconciliation_status"].eq("CONFLICT_REQUIRES_REVIEW"),
        "question_id"
    ].astype(str)
)

records = []
for pair_key, g in queue.groupby("_pair", sort=False):
    first = g.iloc[0]
    if pair_key in reviewed_keys:
        continue

    rec = {
        "question_id": first[qcol],
        "chunk_index": first[ccol],
        "question": first[questioncol] if questioncol else "",
        "expected_answer": first[answercol] if answercol else "",
        "source_file": first[sourcecol] if sourcecol else "",
        "passage_text": first[textcol],
        "conditions_seen": ", ".join(
            sorted(g[conditioncol].dropna().astype(str).unique())
        ) if conditioncol else "",
        "best_rank_seen": pd.to_numeric(
            g[rankcol], errors="coerce"
        ).min() if rankcol else np.nan,
        "occurrences_in_queue": len(g),
        "relevance_label": "",
        "support_label": "",
        "review_reason": "",
        "reviewer_id_or_initials": "",
        "review_date": "",
        "annotation_status": "NEEDS_MANUAL_REVIEW"
    }
    records.append(rec)

review = pd.DataFrame(records)

# Put the most frequently retrieved passages first for review convenience.
if not review.empty:
    review = review.sort_values(
        ["occurrences_in_queue", "best_rank_seen"],
        ascending=[False, True],
        na_position="last"
    ).reset_index(drop=True)

instructions = pd.DataFrame({
    "Field": [
        "relevance_label", "support_label", "review_reason",
        "annotation_status", "Important"
    ],
    "Instructions": [
        "1 = meaningfully relevant to answering the question; 0 = not relevant.",
        "2 = passage alone sufficiently supports the answer; "
        "1 = partial evidence; 0 = no usable supporting evidence.",
        "Briefly explain the judgment using the passage text.",
        "Mark REVIEWED only after inspecting the actual passage and question.",
        "Do not infer relevance from topic similarity alone. "
        "Do not invent evidence. Preserve uncertainty for ambiguous cases."
    ]
})

output = OUT / "passage_annotation_workbook_v1.xlsx"

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    review.to_excel(writer, sheet_name="Review_Queue", index=False)
    instructions.to_excel(writer, sheet_name="Instructions", index=False)
    ann.drop(columns=["_qkey", "_ckey", "_pair"], errors="ignore").to_excel(
        writer, sheet_name="Existing_Annotations", index=False
    )

print("Created:", output)
print("Unique passages requiring review:", len(review))
print("Existing annotation rows preserved:", len(ann))
print("\nAvailable question/answer columns:")
print("Question:", questioncol)
print("Expected answer:", answercol)
print("\nPreview:")
display(review.head(5))
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
OUT = ROOT / "06_RECONCILIATION_INPUTS/reconciliation_output_v1"

REVIEW_FILE = OUT / "passage_annotation_workbook_v1.xlsx"
BENCHMARK_FILE = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
)
PREVIOUS_REVIEW_FILE = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "query_expansion_pilot_v1/evidence_review_v1/"
    "unique_question_passage_review_annotated.xlsx"
)

if not REVIEW_FILE.exists():
    raise FileNotFoundError(f"Missing review workbook: {REVIEW_FILE}")
if not BENCHMARK_FILE.exists():
    raise FileNotFoundError(f"Missing benchmark: {BENCHMARK_FILE}")

review = pd.read_excel(REVIEW_FILE, sheet_name="Review_Queue")
benchmark = pd.read_excel(
    BENCHMARK_FILE, sheet_name="Benchmark_V2_Pilot"
)

# Normalize question IDs without changing their original displayed values.
review["question_id"] = review["question_id"].astype(str).str.strip()
benchmark["question_id"] = benchmark["question_id"].astype(str).str.strip()

if benchmark["question_id"].duplicated().any():
    duplicates = benchmark.loc[
        benchmark["question_id"].duplicated(False), "question_id"
    ].tolist()
    raise ValueError(f"Benchmark has duplicate question IDs: {duplicates}")

# Fields we want to attach from the revised benchmark.
fields = [
    "question_id", "question", "expected_answer", "evidence_excerpt",
    "source_file", "gold_pages", "gold_chunk_ids", "answerability",
    "class", "subject", "book", "question_type", "difficulty",
    "evidence_verified", "answer_supported_by_evidence",
    "benchmark_status", "annotation_status"
]
fields = [c for c in fields if c in benchmark.columns]

# Remove empty placeholders in the queue before attaching authoritative fields.
review = review.drop(
    columns=[
        c for c in fields
        if c != "question_id" and c in review.columns
    ],
    errors="ignore"
)

enriched = review.merge(
    benchmark[fields],
    on="question_id",
    how="left",
    validate="many_to_one",
    indicator=True,
    suffixes=("", "_benchmark")
)

enriched["benchmark_join_status"] = np.where(
    enriched["_merge"].eq("both"),
    "QUESTION_MATCHED",
    "QUESTION_ID_NOT_FOUND"
)
enriched = enriched.drop(columns=["_merge"])

# Optional cross-check against the prior evidence-review workbook.
if PREVIOUS_REVIEW_FILE.exists():
    previous = pd.read_excel(
        PREVIOUS_REVIEW_FILE, sheet_name="Passage_Review"
    )
    previous["question_id"] = previous["question_id"].astype(str).str.strip()

    # One question may appear many times in this workbook; take its first
    # question/answer pair only for a consistency check.
    previous_q = previous.drop_duplicates("question_id").copy()
    previous_q = previous_q[
        ["question_id", "question", "expected_answer"]
    ].rename(columns={
        "question": "previous_review_question",
        "expected_answer": "previous_review_expected_answer"
    })

    enriched = enriched.merge(
        previous_q, on="question_id", how="left",
        validate="many_to_one"
    )

    enriched["question_text_differs_from_previous"] = (
        enriched["previous_review_question"].notna()
        & enriched["question"].fillna("").astype(str).str.strip().ne(
            enriched["previous_review_question"].fillna("").astype(str).str.strip()
        )
    )
else:
    enriched["previous_review_question"] = ""
    enriched["previous_review_expected_answer"] = ""
    enriched["question_text_differs_from_previous"] = False

# Preserve annotation columns and blank review fields.
for col in [
    "relevance_label", "support_label", "review_reason",
    "reviewer_id_or_initials", "review_date"
]:
    if col not in enriched.columns:
        enriched[col] = ""

enriched["annotation_status"] = np.where(
    enriched["benchmark_join_status"].eq("QUESTION_MATCHED"),
    "NEEDS_MANUAL_REVIEW",
    "BLOCKED_MISSING_BENCHMARK_QUESTION"
)

output = OUT / "passage_annotation_workbook_v2_source_linked.xlsx"
exceptions = enriched[
    enriched["benchmark_join_status"].ne("QUESTION_MATCHED")
    | enriched["question_text_differs_from_previous"]
].copy()

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    enriched.to_excel(writer, sheet_name="Review_Queue", index=False)
    exceptions.to_excel(writer, sheet_name="Join_Exceptions", index=False)
    benchmark.to_excel(writer, sheet_name="Benchmark_Source", index=False)

print("Created:", output)
print("\nTotal review pairs:", len(enriched))
print("Matched to benchmark:", int(
    enriched["benchmark_join_status"].eq("QUESTION_MATCHED").sum()
))
print("Missing benchmark questions:", int(
    enriched["benchmark_join_status"].ne("QUESTION_MATCHED").sum()
))
print("Question text differs from previous review:", int(
    enriched["question_text_differs_from_previous"].sum()
))
print("\nPreview:")
display(enriched[[
    "question_id", "chunk_index", "question", "expected_answer",
    "source_file", "gold_pages", "passage_text",
    "benchmark_join_status", "annotation_status"
]].head(8))
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "06_RECONCILIATION_INPUTS"
OUT = BASE / "reconciliation_output_v1"

queue_path = BASE / "top10_evidence_review_queue.xlsx"
benchmark_path = (
    ROOT / "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2/"
    "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
)
annotations_path = BASE / "unique_question_passage_review_corrected.xlsx"

queue = pd.read_excel(queue_path, sheet_name="Top10_Review")
benchmark = pd.read_excel(benchmark_path, sheet_name="Benchmark_V2_Pilot")
annotations = pd.read_excel(annotations_path, sheet_name="Passage_Review")

def norm_id(x):
    if pd.isna(x):
        return ""
    s = str(x).strip()
    try:
        f = float(s)
        if f.is_integer():
            return str(int(f))
    except (ValueError, TypeError):
        pass
    return s

queue["question_id"] = queue["question_id"].astype(str).str.strip()
queue["_chunk_key"] = queue["chunk_index"].map(norm_id)
queue["_pair_key"] = queue["question_id"] + "||" + queue["_chunk_key"]

benchmark["question_id"] = benchmark["question_id"].astype(str).str.strip()

# Preserve the retrieved source and pages under unambiguous names.
queue = queue.rename(columns={
    "source_file": "retrieved_source_file",
    "pages": "retrieved_pages"
})

# Keep one row per unique question-passage pair.
unique_queue = queue.drop_duplicates("_pair_key", keep="first").copy()

# Keep all gold-source fields distinct from retrieved-source fields.
gold_fields = [
    "question_id", "question", "expected_answer", "evidence_excerpt",
    "source_file", "gold_pages", "gold_chunk_ids", "answerability",
    "class", "subject", "book", "question_type", "difficulty",
    "evidence_verified", "answer_supported_by_evidence",
    "benchmark_status"
]
gold_fields = [c for c in gold_fields if c in benchmark.columns]

gold = benchmark[gold_fields].rename(columns={
    "source_file": "gold_source_file"
})

review = unique_queue.merge(
    gold,
    on="question_id",
    how="left",
    validate="many_to_one",
    indicator=True
)

review["benchmark_join_status"] = np.where(
    review["_merge"].eq("both"), "QUESTION_MATCHED", "MISSING_QUESTION"
)
review.drop(columns=["_merge"], inplace=True)

# Attach existing labels only when the exact question+chunk pair matches.
annotations["question_id"] = annotations["question_id"].astype(str).str.strip()
annotations["_chunk_key"] = annotations["chunk_index"].map(norm_id)
annotations["_pair_key"] = annotations["question_id"] + "||" + annotations["_chunk_key"]

ann_cols = [
    c for c in [
        "relevance_label", "support_label", "review_reason",
        "reviewer_id_or_initials"
    ] if c in annotations.columns
]
ann_pair = annotations[["_pair_key"] + ann_cols].copy()

if ann_pair["_pair_key"].duplicated().any():
    raise ValueError("Existing annotation workbook has duplicate pair keys.")

review = review.merge(
    ann_pair,
    on="_pair_key",
    how="left",
    validate="one_to_one",
    suffixes=("", "_existing")
)

review["annotation_status"] = np.where(
    review["relevance_label"].notna() & review["support_label"].notna(),
    "EXISTING_ANNOTATION",
    "NEEDS_MANUAL_REVIEW"
)

# Remove internal keys but keep all source provenance.
review.drop(columns=["_chunk_key", "_pair_key"], inplace=True, errors="ignore")

output = OUT / "passage_annotation_workbook_v3_source_corrected.xlsx"
exceptions = review[
    review["benchmark_join_status"] != "QUESTION_MATCHED"
].copy()

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    review.to_excel(writer, sheet_name="Review_Queue", index=False)
    exceptions.to_excel(writer, sheet_name="Join_Exceptions", index=False)

print("Created:", output)
print("Unique question-passage pairs:", len(review))
print("Question joins missing:", len(exceptions))
print("Missing passage text:", int(review["chunk_text"].isna().sum()) if "chunk_text" in review else "No chunk_text column")
print("Missing retrieved source:", int(review["retrieved_source_file"].isna().sum()) if "retrieved_source_file" in review else "No retrieved source column")
print("Missing gold source:", int(review["gold_source_file"].isna().sum()))
print("\nSource provenance preview:")
display(review[[
    "question_id", "chunk_index", "question",
    "retrieved_source_file", "retrieved_pages",
    "gold_source_file", "gold_pages",
    "annotation_status"
]].head(10))
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import json
import hashlib
import pandas as pd
from datetime import datetime, timezone
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment

# ============================================================
# 1. PATHS
# ============================================================
ROOT = "/content/drive/MyDrive/NCERT_BOOKS"

BENCHMARK = os.path.join(
    ROOT,
    "05_EXPERIMENTS/retrieval_evaluation/benchmark_v2",
    "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
)

REVIEW_DIR = os.path.join(
    ROOT,
    "06_RECONCILIATION_INPUTS/reconciliation_output_v1"
)

OUTPUT_DIR = os.path.join(
    REVIEW_DIR,
    "v2p010_question_revision_v1"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

assert os.path.exists(BENCHMARK), (
    f"Benchmark not found: {BENCHMARK}"
)

# ============================================================
# 2. LOAD BENCHMARK WITHOUT MODIFYING THE ORIGINAL
# ============================================================
xls = pd.ExcelFile(BENCHMARK)

print("Benchmark sheets:", xls.sheet_names)

benchmark_sheet = "Benchmark_V2_Pilot"
df = pd.read_excel(BENCHMARK, sheet_name=benchmark_sheet)

assert "question_id" in df.columns, (
    "Could not find question_id in the benchmark sheet."
)

matches = df.index[
    df["question_id"].astype(str).str.strip() == "V2P010"
].tolist()

assert len(matches) == 1, (
    f"Expected exactly one V2P010 row; found {len(matches)}."
)

row_idx = matches[0]
old_row = df.loc[row_idx].copy()

# ============================================================
# 3. DEFINE THE PROPOSED REVISION
# ============================================================
revision = {
    "question_id": "V2P010",
    "old_question": str(old_row.get("question", "")),
    "proposed_question": (
        "In the Class 1 English book Mridang, after looking "
        "at the clouds, what does the book ask children to draw?"
    ),
    "expected_answer": (
        "Children are asked to draw the shapes they see in the clouds."
    ),
    "gold_source_file": "Class_01/English/Mridang.pdf",
    "gold_pdf_page": 131,
    "gold_printed_page": 117,
    "evidence_excerpt": (
        "Go outside and look at the sky. Do you see the clouds? "
        "What shapes do you see in the clouds? Draw what you see "
        "in the clouds."
    ),
    "revision_reason": (
        "The original wording was broad enough to accept other "
        "sky-related activities, including drawing star patterns "
        "or sketching the Moon. The revised question identifies "
        "the intended Class 1 Mridang activity."
    ),
    "status": "PROPOSED_SOURCE_EVIDENCE_REQUIRES_FINAL_VERIFICATION",
    "retrieval_rerun_required": True,
    "benchmark_note": (
        "This source-specific wording provides a source hint. "
        "Evaluate it separately from natural, source-unspecified "
        "questions."
    ),
}

# ============================================================
# 4. SAVE A SEPARATE REVISION RECORD
# ============================================================
revision_df = pd.DataFrame([revision])

csv_path = os.path.join(
    OUTPUT_DIR, "V2P010_revision_record_v1.csv"
)
revision_df.to_csv(csv_path, index=False)

json_path = os.path.join(
    OUTPUT_DIR, "V2P010_revision_record_v1.json"
)
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(revision, f, indent=2, ensure_ascii=False)

# ============================================================
# 5. CREATE A SEPARATE WORKBOOK COPY
#    Frozen original remains untouched.
# ============================================================
REVISED_BOOK = os.path.join(
    OUTPUT_DIR,
    "NCERT_RAG_Benchmark_V2_Pilot_Revision_Proposal_v1.xlsx"
)

with pd.ExcelWriter(REVISED_BOOK, engine="openpyxl") as writer:
    for sheet in xls.sheet_names:
        sheet_df = pd.read_excel(BENCHMARK, sheet_name=sheet)

        if sheet == benchmark_sheet:
            # Keep a record of the original question in the
            # separate change log rather than overwriting history.
            for col, value in {
                "question": revision["proposed_question"],
                "expected_answer": revision["expected_answer"],
                "evidence_excerpt": revision["evidence_excerpt"],
                "gold_pages": "[131]",
                "negative_reason": revision["revision_reason"],
                "annotation_status": "REVISION_PROPOSED_RETRIEVAL_RERUN_PENDING",
                "benchmark_status": "REVISION_PROPOSAL_NOT_FROZEN",
            }.items():
                if col in sheet_df.columns:
                    sheet_df.loc[
                        sheet_df["question_id"].astype(str).str.strip()
                        == "V2P010",
                        col
                    ] = value

        sheet_df.to_excel(
            writer, sheet_name=sheet, index=False
        )

    pd.DataFrame([revision]).to_excel(
        writer, sheet_name="V2P010_Revision_Record", index=False
    )

# ============================================================
# 6. FORMAT THE REVISION RECORD FOR REVIEW
# ============================================================
wb = load_workbook(REVISED_BOOK)
ws = wb["V2P010_Revision_Record"]

ws.freeze_panes = "A2"
ws.auto_filter.ref = ws.dimensions

for cell in ws[1]:
    cell.font = Font(bold=True, color="FFFFFF")
    cell.fill = PatternFill(
        "solid", fgColor="34495E"
    )
    cell.alignment = Alignment(
        wrap_text=True, vertical="top"
    )

for column_cells in ws.columns:
    max_length = max(
        len(str(cell.value or ""))
        for cell in column_cells
    )
    ws.column_dimensions[column_cells[0].column_letter].width = min(
        max(max_length + 2, 14), 55
    )

for row in ws.iter_rows(min_row=2):
    for cell in row:
        cell.alignment = Alignment(
            wrap_text=True, vertical="top"
        )

wb.save(REVISED_BOOK)

# ============================================================
# 7. SAVE A MANIFEST WITH ORIGINAL FILE HASH
# ============================================================
def sha256_file(path):
    digest = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()

manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "original_benchmark_path": BENCHMARK,
    "original_benchmark_sha256": sha256_file(BENCHMARK),
    "revision_workbook_path": REVISED_BOOK,
    "revision_csv_path": csv_path,
    "revision_json_path": json_path,
    "question_id": "V2P010",
    "change_type": "separate_revision_proposal",
    "original_overwritten": False,
    "retrieval_rerun_completed": False,
    "human_source_verification_completed": False,
}

manifest_path = os.path.join(
    OUTPUT_DIR, "manifest_v1.json"
)
with open(manifest_path, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2, ensure_ascii=False)

# ============================================================
# 8. VERIFY OUTPUTS
# ============================================================
print("\n========== REVISION PROPOSAL CREATED ==========")
print("Question ID:", revision["question_id"])
print("Old question:", revision["old_question"])
print("Proposed question:", revision["proposed_question"])
print("Expected answer:", revision["expected_answer"])
print("\nOutput directory:", OUTPUT_DIR)
print("Revision workbook:", REVISED_BOOK)
print("Revision CSV:", csv_path)
print("Revision JSON:", json_path)
print("Manifest:", manifest_path)
print("\nOriginal benchmark SHA-256:",
      manifest["original_benchmark_sha256"])
print("Original benchmark remains unchanged.")
print("\nIMPORTANT: This is a proposal. Verify the cited PDF page,")
print("then rerun retrieval for V2P010 before finalizing labels.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import json
import hashlib
import pickle
import pandas as pd
import numpy as np
import faiss

from sentence_transformers import SentenceTransformer

# --------------------------------------------------
# 1. Paths and frozen experiment settings
# --------------------------------------------------
ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
ARTIFACTS = (
    f"{ROOT}/05_EXPERIMENTS/"
    "retrieval_evaluation/english_only_v1"
)

OUT = (
    f"{ROOT}/05_EXPERIMENTS/"
    "v2p010_revised_question_rerun_v1"
)
os.makedirs(OUT, exist_ok=True)

QUERY_ID = "V2P010"
QUERY = (
    "In the Class 1 English book Mridang, after looking "
    "at the clouds, what does the book ask children to draw?"
)

EXPECTED_SOURCE = "Class_01/English/Mridang.pdf"
EXPECTED_PDF_PAGE = 131
EXPECTED_PRINTED_PAGE = 117

CANDIDATE_K = 20
RRF_K = 60
TOP_K = 5

# --------------------------------------------------
# 2. Load the existing corpus and indexes
# --------------------------------------------------
chunks = pd.read_csv(
    f"{ARTIFACTS}/english_only_chunks.csv"
).reset_index(drop=True)

with open(
    f"{ARTIFACTS}/indexes/bm25/bm25_index.pkl", "rb"
) as f:
    bm25 = pickle.load(f)

with open(
    f"{ARTIFACTS}/indexes/bm25/tokenized_documents.pkl", "rb"
) as f:
    tokenized_documents = pickle.load(f)

faiss_index = faiss.read_index(
    f"{ARTIFACTS}/indexes/faiss/index.faiss"
)

manifest_path = (
    f"{ARTIFACTS}/indexes/retrieval_manifest.json"
)
with open(manifest_path, "r", encoding="utf-8") as f:
    manifest = json.load(f)

assert len(chunks) == len(tokenized_documents), (
    "Chunk/tokenized-document count mismatch"
)
assert len(chunks) == faiss_index.ntotal, (
    "Chunk/FAISS count mismatch"
)
assert faiss_index.d == 384, "Unexpected FAISS dimension"
assert manifest["chunk_count"] == len(chunks)
assert manifest["hybrid"]["rrf_k"] == RRF_K
assert manifest["hybrid"]["candidate_k"] == CANDIDATE_K

print("Loaded and validated saved retrieval artifacts.")
print("Chunks:", len(chunks))
print("FAISS vectors:", faiss_index.ntotal)
print("Output folder:", OUT)

# --------------------------------------------------
# 3. Retrieve BM25 candidates
# --------------------------------------------------
# Matches the saved manifest's lowercase whitespace split.
query_tokens = QUERY.lower().split()

bm25_scores = np.asarray(
    bm25.get_scores(query_tokens)
)

bm25_candidate_ids = np.argsort(
    -bm25_scores, kind="stable"
)[:CANDIDATE_K]

# --------------------------------------------------
# 4. Retrieve dense candidates
# --------------------------------------------------
model_name = manifest["embedding_model"]
model = SentenceTransformer(model_name)

query_embedding = model.encode(
    [QUERY],
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=False
).astype(np.float32)

assert query_embedding.shape == (1, faiss_index.d)

dense_scores, dense_ids = faiss_index.search(
    query_embedding, CANDIDATE_K
)

dense_candidate_ids = [
    int(i) for i in dense_ids[0] if i >= 0
]
dense_score_by_id = {
    int(i): float(score)
    for i, score in zip(dense_ids[0], dense_scores[0])
    if i >= 0
}

bm25_score_by_id = {
    int(i): float(bm25_scores[i])
    for i in bm25_candidate_ids
}

# --------------------------------------------------
# 5. Hybrid Reciprocal Rank Fusion
# --------------------------------------------------
# Rank positions start at 1. Each candidate receives
# 1 / (RRF_K + rank) from each list where it appears.
rrf_scores = {}

for rank, idx in enumerate(bm25_candidate_ids, start=1):
    idx = int(idx)
    rrf_scores[idx] = (
        rrf_scores.get(idx, 0.0)
        + 1.0 / (RRF_K + rank)
    )

for rank, idx in enumerate(dense_candidate_ids, start=1):
    idx = int(idx)
    rrf_scores[idx] = (
        rrf_scores.get(idx, 0.0)
        + 1.0 / (RRF_K + rank)
    )

hybrid_candidate_ids = sorted(
    rrf_scores,
    key=lambda idx: (-rrf_scores[idx], idx)
)[:TOP_K]

# --------------------------------------------------
# 6. Format and save results
# --------------------------------------------------
def result_rows(method, ids, score_map, score_name):
    rows = []

    for rank, idx in enumerate(ids[:TOP_K], start=1):
        record = chunks.iloc[int(idx)].to_dict()
        record["query_id"] = QUERY_ID
        record["query"] = QUERY
        record["method"] = method
        record["rank"] = rank
        record["chunk_row_index"] = int(idx)
        record["retrieval_score"] = float(score_map[int(idx)])
        record["score_type"] = score_name
        rows.append(record)

    return rows

all_rows = []
all_rows.extend(
    result_rows(
        "BM25", bm25_candidate_ids,
        bm25_score_by_id, "BM25 score"
    )
)
all_rows.extend(
    result_rows(
        "DENSE", dense_candidate_ids,
        dense_score_by_id, "FAISS inner product"
    )
)
all_rows.extend(
    result_rows(
        "HYBRID_RRF", hybrid_candidate_ids,
        rrf_scores, "RRF score"
    )
)

results = pd.DataFrame(all_rows)

results_path = f"{OUT}/V2P010_revised_top5_results.csv"
results.to_csv(results_path, index=False)

# Save experiment settings and integrity information.
config = {
    "experiment": "v2p010_revised_question_rerun_v1",
    "status": "retrieval rerun; evidence review pending",
    "query_id": QUERY_ID,
    "query": QUERY,
    "query_type": "source-hinted revised question",
    "expected_source_file": EXPECTED_SOURCE,
    "expected_pdf_page": EXPECTED_PDF_PAGE,
    "expected_printed_page": EXPECTED_PRINTED_PAGE,
    "embedding_model": model_name,
    "chunk_count": len(chunks),
    "candidate_k": CANDIDATE_K,
    "rrf_k": RRF_K,
    "top_k": TOP_K,
    "bm25_tokenization": "lowercase whitespace split",
    "dense_index": "saved FAISS IndexFlatIP",
    "dense_query_normalization": "L2 normalized",
    "hybrid_method": "Reciprocal Rank Fusion",
    "original_benchmark_modified": False,
    "saved_indexes_rebuilt": False,
    "results_file": os.path.basename(results_path),
}

config_path = f"{OUT}/experiment_manifest.json"
with open(config_path, "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2)

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

config["results_sha256"] = sha256_file(results_path)
with open(config_path, "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2)

# --------------------------------------------------
# 7. Print compact evidence for manual review
# --------------------------------------------------
display_columns = [
    "method", "rank", "chunk_id", "class", "subject",
    "book", "source_file", "pages", "first_page",
    "last_page", "retrieval_score", "text"
]

for method in ["BM25", "DENSE", "HYBRID_RRF"]:
    print("\n" + "=" * 90)
    print(method)
    print("=" * 90)

    subset = results[results["method"] == method]
    for _, row in subset.iterrows():
        print(
            f"\nRank {row['rank']} | {row['chunk_id']}\n"
            f"Source: {row['source_file']} | "
            f"PDF pages: {row['pages']}\n"
            f"Score: {row['retrieval_score']:.6f}\n"
            f"Text: {str(row['text'])[:900]}"
        )

print("\nSaved results:", results_path)
print("Saved manifest:", config_path)
print("Results SHA-256:", config["results_sha256"])
print("No indexes or benchmark files were overwritten.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import json
import pickle
import numpy as np
import pandas as pd
import faiss
from sentence_transformers import SentenceTransformer

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
ARTIFACTS = f"{ROOT}/05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
OUT = f"{ROOT}/05_EXPERIMENTS/v2p010_revised_question_rerun_v1"

chunks = pd.read_csv(f"{ARTIFACTS}/english_only_chunks.csv").reset_index(drop=True)

with open(f"{ARTIFACTS}/indexes/bm25/bm25_index.pkl", "rb") as f:
    bm25 = pickle.load(f)

index = faiss.read_index(f"{ARTIFACTS}/indexes/faiss/index.faiss")

with open(f"{ARTIFACTS}/indexes/retrieval_manifest.json", "r") as f:
    manifest = json.load(f)

model = SentenceTransformer(manifest["embedding_model"])

queries = {
    "original": "What activity does the book suggest children do with the clouds?",
    "revised_source_hinted": (
        "In the Class 1 English book Mridang, after looking "
        "at the clouds, what does the book ask children to draw?"
    ),
}

# Identify the exact target chunk by metadata and its evidence text.
target_mask = (
    chunks["source_file"].eq("Class_01/English/Mridang.pdf")
    & chunks["text"].str.contains(
        "Draw what you see in the clouds", case=False, na=False
    )
)

targets = chunks.index[target_mask].tolist()

print("Target chunk candidates:", len(targets))
for idx in targets:
    row = chunks.iloc[idx]
    print(
        "\nTarget row index:", idx,
        "\nChunk ID:", row["chunk_id"],
        "\nPDF pages:", row["pages"],
        "\nText excerpt:", row["text"][-900:]
    )

if not targets:
    raise ValueError("Could not locate the expected target passage in the corpus.")

target_idx = targets[0]
candidate_k = 20
rrf_k = 60
diagnostic_rows = []

for query_name, query in queries.items():
    # BM25 ranking across the full corpus
    bm25_scores = np.asarray(bm25.get_scores(query.lower().split()))
    bm25_order = np.argsort(-bm25_scores, kind="stable")

    # Dense ranking across the full corpus
    q_emb = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True,
        show_progress_bar=False
    ).astype(np.float32)

    dense_scores, dense_ids = index.search(q_emb, len(chunks))
    dense_order = [int(i) for i in dense_ids[0] if i >= 0]

    # Full-corpus rank positions
    bm25_rank = int(np.where(bm25_order == target_idx)[0][0]) + 1
    dense_rank = dense_order.index(target_idx) + 1

    # Hybrid RRF uses top 20 candidates from each retriever.
    bm25_candidates = [int(i) for i in bm25_order[:candidate_k]]
    dense_candidates = [int(i) for i in dense_order[:candidate_k]]

    rrf_scores = {}
    for rank, idx in enumerate(bm25_candidates, start=1):
        rrf_scores[idx] = rrf_scores.get(idx, 0) + 1 / (rrf_k + rank)

    for rank, idx in enumerate(dense_candidates, start=1):
        rrf_scores[idx] = rrf_scores.get(idx, 0) + 1 / (rrf_k + rank)

    hybrid_order = sorted(
        rrf_scores, key=lambda idx: (-rrf_scores[idx], idx)
    )
    hybrid_rank = (
        hybrid_order.index(target_idx) + 1
        if target_idx in hybrid_order else None
    )

    for method, rank, score in [
        ("BM25", bm25_rank, float(bm25_scores[target_idx])),
        ("DENSE", dense_rank, float(dense_scores[0][dense_rank - 1])),
        ("HYBRID_RRF", hybrid_rank, rrf_scores.get(target_idx, 0.0)),
    ]:
        diagnostic_rows.append({
            "query_id": query_name,
            "query": query,
            "target_chunk_id": chunks.iloc[target_idx]["chunk_id"],
            "method": method,
            "target_rank": rank,
            "target_score": score,
            "within_top5": rank is not None and rank <= 5,
            "within_top10": rank is not None and rank <= 10,
            "within_top20": rank is not None and rank <= 20,
        })

diagnostics = pd.DataFrame(diagnostic_rows)
display(diagnostics)

diagnostic_path = f"{OUT}/V2P010_original_vs_revised_rank_diagnostic.csv"
diagnostics.to_csv(diagnostic_path, index=False)
print("Saved:", diagnostic_path)
```

```python

# ============================================================
# NCERT AI TUTOR — PHASE A FOUNDATION FREEZE
# Version: 1.0
# Purpose: Validate and freeze the data/retrieval foundation
#
# SAFETY:
# - Does not rebuild retrieval indexes.
# - Does not modify source PDFs, corpus, benchmark, or experiments.
# - Creates a new timestamped release folder.
# - Records checksums and exact artifact paths for Phase B.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os
import json
import csv
import time
import pickle
import hashlib
import shutil
import traceback
import warnings
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")

SOURCE_DIR = ROOT / "01_SOURCE_BOOKS_V2"
CLEAN_CORPUS_DIR = ROOT / "02_PROCESSED_DATA" / "corpus_v2_clean"

ARTIFACTS = (
    ROOT / "05_EXPERIMENTS" /
    "retrieval_evaluation" / "english_only_v1"
)

CHUNKS_PATH = ARTIFACTS / "english_only_chunks.csv"
BM25_PATH = ARTIFACTS / "indexes" / "bm25" / "bm25_index.pkl"
TOKENS_PATH = ARTIFACTS / "indexes" / "bm25" / "tokenized_documents.pkl"
EMBEDDINGS_PATH = (
    ARTIFACTS / "indexes" / "embeddings" / "chunk_embeddings.npy"
)
FAISS_PATH = ARTIFACTS / "indexes" / "faiss" / "index.faiss"
RETRIEVAL_MANIFEST_PATH = ARTIFACTS / "indexes" / "retrieval_manifest.json"
EXPERIMENT_CONFIG_PATH = ARTIFACTS / "experiment_config.json"

BENCHMARK_NAME = "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
EXPECTED_BENCHMARK_SHA256 = (
    "d3aa2d9e39f949206585936f865c382f1fc9bfa0bf3ba7f9b1a8cb43f52cabcd"
)

EXPECTED_CHUNKS = 2033
EXPECTED_EMBEDDING_DIM = 384
EXPECTED_SOURCE_BOOKS = 22
EMBEDDING_MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

RUN_ID = datetime.now().strftime("%Y%m%d_%H%M%S")
RELEASE_DIR = ROOT / "06_PHASE_A_FOUNDATION" / f"phase_a_freeze_{RUN_ID}"
RELEASE_DIR.mkdir(parents=True, exist_ok=False)

FROZEN_INPUTS_DIR = RELEASE_DIR / "frozen_inputs"
FROZEN_INPUTS_DIR.mkdir()

print("=" * 72)
print("NCERT AI TUTOR — PHASE A FOUNDATION FREEZE")
print("=" * 72)
print("Drive root:", ROOT)
print("Release ID:", RUN_ID)
print("Release folder:", RELEASE_DIR)
print("\nNo original files will be overwritten.")

# ------------------------------------------------------------
# 2. VALIDATION HELPERS
# ------------------------------------------------------------

checks = []
warnings_list = []
errors = []
artifact_records = []

def record_check(name, passed, detail="", severity="ERROR"):
    item = {
        "check": name,
        "passed": bool(passed),
        "detail": str(detail)
    }
    checks.append(item)

    status = "PASS" if passed else (
        "WARNING" if severity == "WARNING" else "FAIL"
    )
    print(f"[{status}] {name}: {detail}")

    if not passed:
        if severity == "WARNING":
            warnings_list.append(item)
        else:
            errors.append(item)

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

def register_artifact(path, category):
    path = Path(path)
    if not path.is_file():
        artifact_records.append({
            "category": category,
            "path": str(path),
            "exists": False,
            "size_bytes": None,
            "sha256": None
        })
        return None

    digest = sha256_file(path)
    artifact_records.append({
        "category": category,
        "path": str(path),
        "exists": True,
        "size_bytes": path.stat().st_size,
        "sha256": digest
    })
    return digest

def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

# ------------------------------------------------------------
# 3. CHECK DIRECTORY STRUCTURE
# ------------------------------------------------------------

record_check(
    "Drive project root exists",
    ROOT.is_dir(),
    ROOT
)

record_check(
    "Original source directory exists",
    SOURCE_DIR.is_dir(),
    SOURCE_DIR
)

record_check(
    "Clean corpus directory exists",
    CLEAN_CORPUS_DIR.is_dir(),
    CLEAN_CORPUS_DIR
)

record_check(
    "English retrieval artifacts exist",
    ARTIFACTS.is_dir(),
    ARTIFACTS
)

required_files = {
    "chunks": CHUNKS_PATH,
    "BM25 index": BM25_PATH,
    "tokenized documents": TOKENS_PATH,
    "embeddings": EMBEDDINGS_PATH,
    "FAISS index": FAISS_PATH,
    "retrieval manifest": RETRIEVAL_MANIFEST_PATH,
    "experiment config": EXPERIMENT_CONFIG_PATH
}

for name, path in required_files.items():
    record_check(
        f"Required artifact: {name}",
        path.is_file(),
        path
    )

# ------------------------------------------------------------
# 4. CHECK SOURCE BOOKS
# ------------------------------------------------------------

source_pdfs = []

if SOURCE_DIR.is_dir():
    source_pdfs = sorted(
        p for p in SOURCE_DIR.rglob("*")
        if p.is_file() and p.suffix.lower() == ".pdf"
    )

record_check(
    "Source PDF inventory",
    len(source_pdfs) == EXPECTED_SOURCE_BOOKS,
    f"Found {len(source_pdfs)} PDFs; expected {EXPECTED_SOURCE_BOOKS}",
    severity="WARNING"
)

source_inventory = []
for path in source_pdfs:
    source_inventory.append({
        "relative_path": str(path.relative_to(SOURCE_DIR)),
        "size_bytes": path.stat().st_size
    })

with open(
    RELEASE_DIR / "source_inventory.json", "w", encoding="utf-8"
) as f:
    json.dump(source_inventory, f, indent=2)

# ------------------------------------------------------------
# 5. LOAD AND VALIDATE THE CLEAN CHUNK TABLE
# ------------------------------------------------------------

chunks = None
required_chunk_columns = [
    "chunk_id", "class", "subject", "book", "source_file",
    "chunk_index", "word_count", "pages", "text"
]

if CHUNKS_PATH.is_file():
    try:
        chunks = pd.read_csv(CHUNKS_PATH)

        missing_columns = [
            c for c in required_chunk_columns if c not in chunks.columns
        ]

        record_check(
            "Chunk table schema",
            len(missing_columns) == 0,
            f"Shape={chunks.shape}; missing columns={missing_columns}"
        )

        record_check(
            "Expected chunk count",
            len(chunks) == EXPECTED_CHUNKS,
            f"Found {len(chunks)}; expected {EXPECTED_CHUNKS}"
        )

        if "chunk_id" in chunks.columns:
            record_check(
                "Unique chunk IDs",
                chunks["chunk_id"].notna().all()
                and chunks["chunk_id"].is_unique,
                f"Unique IDs={chunks['chunk_id'].nunique()}"
            )

        if "text" in chunks.columns:
            nonempty_text = (
                chunks["text"].fillna("").astype(str).str.strip().ne("")
            )
            record_check(
                "Non-empty chunk text",
                bool(nonempty_text.all()),
                f"Empty text rows={(~nonempty_text).sum()}"
            )

        if "source_file" in chunks.columns:
            record_check(
                "Chunk source metadata populated",
                chunks["source_file"].notna().all(),
                f"Unique source files={chunks['source_file'].nunique()}"
            )

        if "pages" in chunks.columns:
            record_check(
                "Page metadata populated",
                chunks["pages"].notna().all(),
                f"Missing page metadata={chunks['pages'].isna().sum()}",
                severity="WARNING"
            )

    except Exception as e:
        record_check("Read chunk table", False, repr(e))

# ------------------------------------------------------------
# 6. LOAD CONFIGURATIONS
# ------------------------------------------------------------

retrieval_manifest = None
experiment_config = None

try:
    retrieval_manifest = load_json(RETRIEVAL_MANIFEST_PATH)
    record_check(
        "Retrieval manifest parses",
        isinstance(retrieval_manifest, dict),
        "JSON loaded successfully"
    )
except Exception as e:
    record_check("Retrieval manifest parses", False, repr(e))

try:
    experiment_config = load_json(EXPERIMENT_CONFIG_PATH)
    record_check(
        "Experiment config parses",
        isinstance(experiment_config, dict),
        "JSON loaded successfully"
    )
except Exception as e:
    record_check("Experiment config parses", False, repr(e))

if retrieval_manifest:
    record_check(
        "Manifest chunk count",
        retrieval_manifest.get("chunk_count") == EXPECTED_CHUNKS,
        f"Manifest says {retrieval_manifest.get('chunk_count')}"
    )

    record_check(
        "Manifest embedding model",
        retrieval_manifest.get("embedding_model") == EMBEDDING_MODEL_NAME,
        f"Model={retrieval_manifest.get('embedding_model')}"
    )

    record_check(
        "Manifest embedding dimension",
        retrieval_manifest.get("embedding_dimension") == EXPECTED_EMBEDDING_DIM,
        f"Dimension={retrieval_manifest.get('embedding_dimension')}"
    )

    hybrid_config = retrieval_manifest.get("hybrid", {})
    record_check(
        "Hybrid RRF configuration",
        hybrid_config.get("method") == "Reciprocal Rank Fusion"
        and hybrid_config.get("rrf_k") == 60
        and hybrid_config.get("candidate_k") == 20,
        f"Hybrid config={hybrid_config}"
    )

# ------------------------------------------------------------
# 7. LOAD SAVED BM25, TOKENS, EMBEDDINGS AND FAISS
# ------------------------------------------------------------

bm25 = None
tokenized_documents = None
embeddings = None
faiss_index = None

if BM25_PATH.is_file():
    try:
        with open(BM25_PATH, "rb") as f:
            bm25 = pickle.load(f)
        record_check(
            "BM25 index loads",
            bm25 is not None,
            f"Type={type(bm25).__name__}"
        )
    except Exception as e:
        record_check("BM25 index loads", False, repr(e))

if TOKENS_PATH.is_file():
    try:
        with open(TOKENS_PATH, "rb") as f:
            tokenized_documents = pickle.load(f)

        record_check(
            "Tokenized documents load",
            isinstance(tokenized_documents, list),
            f"Count={len(tokenized_documents) if hasattr(tokenized_documents, '__len__') else 'unknown'}"
        )
    except Exception as e:
        record_check("Tokenized documents load", False, repr(e))

if EMBEDDINGS_PATH.is_file():
    try:
        embeddings = np.load(EMBEDDINGS_PATH, mmap_mode="r")

        record_check(
            "Embedding matrix shape",
            embeddings.shape == (EXPECTED_CHUNKS, EXPECTED_EMBEDDING_DIM),
            f"Shape={embeddings.shape}; dtype={embeddings.dtype}"
        )

        record_check(
            "Embedding values finite",
            bool(np.isfinite(embeddings).all()),
            "Checked all stored embedding values"
        )

        if embeddings.ndim == 2 and embeddings.shape[1] == EXPECTED_EMBEDDING_DIM:
            sample_norms = np.linalg.norm(
                np.asarray(embeddings[:min(100, len(embeddings))]),
                axis=1
            )
            record_check(
                "Embedding normalization sample",
                bool(np.allclose(sample_norms, 1.0, atol=2e-3)),
                f"Sample norm range={sample_norms.min():.5f}–{sample_norms.max():.5f}",
                severity="WARNING"
            )
    except Exception as e:
        record_check("Embedding matrix loads", False, repr(e))

if FAISS_PATH.is_file():
    try:
        import faiss
        faiss_index = faiss.read_index(str(FAISS_PATH))

        record_check(
            "FAISS index shape",
            faiss_index.ntotal == EXPECTED_CHUNKS
            and faiss_index.d == EXPECTED_EMBEDDING_DIM,
            f"Type={type(faiss_index).__name__}; "
            f"vectors={faiss_index.ntotal}; dimension={faiss_index.d}"
        )
    except Exception as e:
        record_check("FAISS index loads", False, repr(e))

# ------------------------------------------------------------
# 8. VERIFY ALIGNMENT ACROSS ALL RETRIEVAL REPRESENTATIONS
# ------------------------------------------------------------

if chunks is not None and tokenized_documents is not None:
    record_check(
        "Chunks/tokenized documents aligned",
        len(chunks) == len(tokenized_documents),
        f"{len(chunks)} chunks; {len(tokenized_documents)} tokenized documents"
    )

if chunks is not None and bm25 is not None:
    bm25_size = getattr(bm25, "corpus_size", None)
    record_check(
        "Chunks/BM25 corpus aligned",
        bm25_size == len(chunks),
        f"BM25 corpus_size={bm25_size}; chunks={len(chunks)}"
    )

if chunks is not None and embeddings is not None:
    record_check(
        "Chunks/embedding matrix aligned",
        len(chunks) == embeddings.shape[0],
        f"{len(chunks)} chunks; {embeddings.shape[0]} embedding rows"
    )

if chunks is not None and faiss_index is not None:
    record_check(
        "Chunks/FAISS index aligned",
        len(chunks) == faiss_index.ntotal,
        f"{len(chunks)} chunks; {faiss_index.ntotal} vectors"
    )

# Compare the FAISS stored vectors with the saved embedding matrix.
# This checks that vector order/content agrees, not just vector count.
if embeddings is not None and faiss_index is not None:
    try:
        faiss_vectors = faiss_index.reconstruct_n(
            0, faiss_index.ntotal
        )
        vectors_match = np.allclose(
            np.asarray(embeddings),
            faiss_vectors,
            atol=2e-5,
            rtol=2e-5
        )
        max_diff = float(
            np.max(np.abs(np.asarray(embeddings) - faiss_vectors))
        )

        record_check(
            "FAISS/embedding content alignment",
            vectors_match,
            f"Maximum absolute difference={max_diff:.8g}"
        )
    except Exception as e:
        record_check(
            "FAISS/embedding content alignment",
            False,
            repr(e),
            severity="WARNING"
        )

# ------------------------------------------------------------
# 9. CHECK THE FROZEN BENCHMARK WITHOUT MODIFYING IT
# ------------------------------------------------------------

benchmark_candidates = []
benchmark_search_root = ROOT / "05_EXPERIMENTS"

expected_benchmark_path = (
    benchmark_search_root / "benchmark_v2" / BENCHMARK_NAME
)

if expected_benchmark_path.is_file():
    benchmark_candidates = [expected_benchmark_path]
elif benchmark_search_root.is_dir():
    benchmark_candidates = [
        p for p in benchmark_search_root.rglob(BENCHMARK_NAME)
        if p.is_file()
    ]

benchmark_path = (
    benchmark_candidates[0] if len(benchmark_candidates) == 1 else None
)

if len(benchmark_candidates) == 1:
    benchmark_sha = sha256_file(benchmark_path)

    record_check(
        "Frozen benchmark checksum",
        benchmark_sha == EXPECTED_BENCHMARK_SHA256,
        f"SHA-256={benchmark_sha}"
    )

    # Copy only if it matches the known frozen checksum.
    # The original is never modified.
    if benchmark_sha == EXPECTED_BENCHMARK_SHA256:
        frozen_copy = FROZEN_INPUTS_DIR / BENCHMARK_NAME
        shutil.copy2(benchmark_path, frozen_copy)

        record_check(
            "Frozen benchmark snapshot",
            sha256_file(frozen_copy) == benchmark_sha,
            f"Read-only source preserved; snapshot={frozen_copy.name}"
        )
    else:
        warnings_list.append({
            "check": "Frozen benchmark snapshot",
            "passed": False,
            "detail": "Source checksum differs; no copy made."
        })
        print("[WARNING] Benchmark checksum differs; no snapshot copied.")

elif len(benchmark_candidates) == 0:
    record_check(
        "Frozen benchmark located",
        False,
        f"Could not find {BENCHMARK_NAME}",
        severity="WARNING"
    )
    benchmark_sha = None
else:
    record_check(
        "Frozen benchmark uniquely located",
        False,
        f"Found {len(benchmark_candidates)} copies; no copy selected.",
        severity="WARNING"
    )
    benchmark_sha = None

# ------------------------------------------------------------
# 10. RETRIEVAL SMOKE TEST
# ------------------------------------------------------------

smoke_results = []
smoke_error = None

if all([
    chunks is not None,
    bm25 is not None,
    faiss_index is not None,
    retrieval_manifest is not None
]):
    try:
        from sentence_transformers import SentenceTransformer

        smoke_query = (
            "What activity does the book suggest children do with the clouds?"
        )
        target_source = "Class_01/English/Mridang.pdf"
        target_phrase = "Draw what you see in the clouds"

        target_rows = chunks[
            chunks["source_file"].eq(target_source)
            & chunks["text"].str.contains(
                target_phrase, case=False, na=False
            )
        ]

        record_check(
            "Known evidence passage found",
            len(target_rows) >= 1,
            f"Matching target chunks={len(target_rows)}"
        )

        target_id = (
            int(target_rows.index[0]) if len(target_rows) else None
        )

        # BM25 top 20
        bm25_scores = np.asarray(
            bm25.get_scores(smoke_query.lower().split())
        )
        bm25_ids = [
            int(i) for i in np.argsort(
                -bm25_scores, kind="stable"
            )[:20]
        ]

        # Dense top 20 using the existing model and saved index
        model = SentenceTransformer(
            retrieval_manifest["embedding_model"]
        )
        query_vector = model.encode(
            [smoke_query],
            convert_to_numpy=True,
            normalize_embeddings=True,
            show_progress_bar=False
        ).astype(np.float32)

        dense_scores, dense_ids_raw = faiss_index.search(
            query_vector, 20
        )
        dense_ids = [
            int(i) for i in dense_ids_raw[0] if i >= 0
        ]

        # Hybrid RRF top 20
        rrf_scores = {}

        for rank, idx in enumerate(bm25_ids, start=1):
            rrf_scores[idx] = (
                rrf_scores.get(idx, 0.0) + 1.0 / (60 + rank)
            )

        for rank, idx in enumerate(dense_ids, start=1):
            rrf_scores[idx] = (
                rrf_scores.get(idx, 0.0) + 1.0 / (60 + rank)
            )

        hybrid_ids = sorted(
            rrf_scores,
            key=lambda idx: (-rrf_scores[idx], idx)
        )

        def get_rank(ids, target):
            return ids.index(target) + 1 if target in ids else None

        for method, ids in [
            ("BM25", bm25_ids),
            ("DENSE", dense_ids),
            ("HYBRID_RRF", hybrid_ids)
        ]:
            target_rank = (
                get_rank(ids, target_id)
                if target_id is not None else None
            )

            smoke_results.append({
                "query": smoke_query,
                "method": method,
                "target_chunk_id": (
                    chunks.iloc[target_id]["chunk_id"]
                    if target_id is not None else None
                ),
                "target_rank_within_top20": target_rank,
                "target_retrieved_top5": (
                    target_rank is not None and target_rank <= 5
                ),
                "target_retrieved_top10": (
                    target_rank is not None and target_rank <= 10
                )
            })

        record_check(
            "Retrieval smoke test executes",
            len(smoke_results) == 3,
            "BM25, dense and hybrid completed on a known question"
        )

        for result in smoke_results:
            print(
                f"  {result['method']}: target rank = "
                f"{result['target_rank_within_top20']}"
            )

        record_check(
            "Smoke test finds evidence in at least one method",
            any(
                r["target_rank_within_top20"] is not None
                for r in smoke_results
            ),
            "Target passage appears in at least one top-20 result",
            severity="WARNING"
        )

    except Exception as e:
        smoke_error = traceback.format_exc()
        record_check(
            "Retrieval smoke test executes",
            False,
            repr(e),
            severity="WARNING"
        )
        print(smoke_error)

else:
    record_check(
        "Retrieval smoke test prerequisites",
        False,
        "One or more required retrieval artifacts could not be loaded.",
        severity="WARNING"
    )

# ------------------------------------------------------------
# 11. REGISTER THE CORE ARTIFACTS AND THEIR CHECKSUMS
# ------------------------------------------------------------

core_artifacts = [
    ("chunks", CHUNKS_PATH),
    ("BM25", BM25_PATH),
    ("tokenized_documents", TOKENS_PATH),
    ("embeddings", EMBEDDINGS_PATH),
    ("FAISS", FAISS_PATH),
    ("retrieval_manifest", RETRIEVAL_MANIFEST_PATH),
    ("experiment_config", EXPERIMENT_CONFIG_PATH),
]

for category, path in core_artifacts:
    if path.is_file():
        register_artifact(path, category)

if benchmark_path is not None and benchmark_path.is_file():
    register_artifact(benchmark_path, "frozen_benchmark_source")

if source_pdfs:
    for path in source_pdfs:
        artifact_records.append({
            "category": "source_pdf_inventory",
            "path": str(path),
            "exists": True,
            "size_bytes": path.stat().st_size,
            "sha256": None
        })

# Store artifact registry.
registry_path = RELEASE_DIR / "artifact_registry.csv"
pd.DataFrame(artifact_records).to_csv(registry_path, index=False)

# Store smoke-test results.
smoke_path = RELEASE_DIR / "retrieval_smoke_test.csv"
pd.DataFrame(smoke_results).to_csv(smoke_path, index=False)

# Store checks as machine-readable JSON.
checks_path = RELEASE_DIR / "validation_checks.json"
with open(checks_path, "w", encoding="utf-8") as f:
    json.dump(checks, f, indent=2)

# ------------------------------------------------------------
# 12. WRITE THE PHASE A MANIFEST FOR PHASE B
# ------------------------------------------------------------

hard_failures = [c for c in checks if not c["passed"] and c not in warnings_list]

if hard_failures:
    overall_status = "BLOCKED"
elif warnings_list:
    overall_status = "PASS_WITH_WARNINGS"
else:
    overall_status = "PASS"

phase_b_config = {
    "project": "NCERT adaptive AI tutor",
    "phase_a_release_id": RUN_ID,
    "phase_a_status": overall_status,
    "created_at_local": datetime.now().astimezone().isoformat(),
    "project_root": str(ROOT),
    "source_pdf_directory": str(SOURCE_DIR),
    "clean_corpus_directory": str(CLEAN_CORPUS_DIR),
    "retrieval_artifacts_directory": str(ARTIFACTS),
    "chunks_csv": str(CHUNKS_PATH),
    "bm25_index_pickle": str(BM25_PATH),
    "tokenized_documents_pickle": str(TOKENS_PATH),
    "embedding_matrix_npy": str(EMBEDDINGS_PATH),
    "faiss_index": str(FAISS_PATH),
    "retrieval_manifest": str(RETRIEVAL_MANIFEST_PATH),
    "experiment_config": str(EXPERIMENT_CONFIG_PATH),
    "frozen_benchmark_original": (
        str(benchmark_path) if benchmark_path else None
    ),
    "frozen_benchmark_sha256": benchmark_sha,
    "expected_chunk_count": EXPECTED_CHUNKS,
    "expected_embedding_dimension": EXPECTED_EMBEDDING_DIM,
    "embedding_model": EMBEDDING_MODEL_NAME,
    "bm25_tokenization": "lowercase whitespace split",
    "dense_index": "FAISS IndexFlatIP",
    "dense_similarity": "cosine-equivalent inner product",
    "hybrid_method": "Reciprocal Rank Fusion",
    "hybrid_rrf_constant": 60,
    "hybrid_candidate_k": 20,
    "recommended_default_top_k": 5,
    "original_corpus_modified": False,
    "original_benchmark_modified": False,
    "indexes_rebuilt": False,
    "phase_b_ready_for_integration": overall_status != "BLOCKED",
    "important_scope_note": (
        "This release validates the corpus/retrieval foundation. "
        "It does not claim that answer generation, cognitive adaptation, "
        "multilingual responses, speech input/output, or learning efficacy "
        "have already been implemented or validated."
    ),
    "release_files": {
        "artifact_registry": str(registry_path),
        "validation_checks": str(checks_path),
        "retrieval_smoke_test": str(smoke_path),
        "source_inventory": str(RELEASE_DIR / "source_inventory.json"),
        "frozen_inputs_directory": str(FROZEN_INPUTS_DIR)
    },
    "check_summary": {
        "total": len(checks),
        "passed": sum(1 for c in checks if c["passed"]),
        "failed": len(hard_failures),
        "warnings": len(warnings_list)
    }
}

manifest_out = RELEASE_DIR / "phase_a_manifest.json"
with open(manifest_out, "w", encoding="utf-8") as f:
    json.dump(phase_b_config, f, indent=2)

# Human-readable report.
report_path = RELEASE_DIR / "PHASE_A_VALIDATION_REPORT.txt"
with open(report_path, "w", encoding="utf-8") as f:
    f.write("NCERT AI TUTOR — PHASE A FOUNDATION FREEZE\n")
    f.write("=" * 60 + "\n")
    f.write(f"Release ID: {RUN_ID}\n")
    f.write(f"Status: {overall_status}\n")
    f.write(f"Release directory: {RELEASE_DIR}\n\n")

    f.write("CHECK SUMMARY\n")
    f.write("-" * 60 + "\n")
    f.write(f"Total checks: {len(checks)}\n")
    f.write(f"Passed: {sum(1 for c in checks if c['passed'])}\n")
    f.write(f"Hard failures: {len(hard_failures)}\n")
    f.write(f"Warnings: {len(warnings_list)}\n\n")

    f.write("INDIVIDUAL CHECKS\n")
    f.write("-" * 60 + "\n")
    for c in checks:
        label = "PASS" if c["passed"] else (
            "WARNING" if c in warnings_list else "FAIL"
        )
        f.write(f"[{label}] {c['check']}\n  {c['detail']}\n")

    f.write("\nRETRIEVAL SMOKE TEST\n")
    f.write("-" * 60 + "\n")
    for r in smoke_results:
        f.write(json.dumps(r) + "\n")

    f.write("\nSCOPE\n")
    f.write(phase_b_config["important_scope_note"] + "\n")

# ------------------------------------------------------------
# 13. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("PHASE A FREEZE COMPLETE")
print("=" * 72)
print("STATUS:", overall_status)
print("Checks passed:", sum(1 for c in checks if c["passed"]), "/", len(checks))
print("Hard failures:", len(hard_failures))
print("Warnings:", len(warnings_list))

print("\nRelease folder:")
print(RELEASE_DIR)

print("\nImportant outputs:")
print("  Manifest:       ", manifest_out)
print("  Validation:     ", report_path)
print("  Artifact hashes:", registry_path)
print("  Smoke test:     ", smoke_path)
print("  Frozen inputs:  ", FROZEN_INPUTS_DIR)

if hard_failures:
    print("\nHARD FAILURES — resolve these before relying on Phase A:")
    for item in hard_failures:
        print(" -", item["check"], "=>", item["detail"])

if warnings_list:
    print("\nWARNINGS — review these, but they may not block Phase B:")
    for item in warnings_list:
        print(" -", item["check"], "=>", item["detail"])

print("\nOriginal source PDFs, corpus, indexes, benchmark and experiment")
print("outputs were not intentionally modified by this cell.")
print("\nPhase B should load phase_a_manifest.json rather than rediscover")
print("the artifact paths or rebuild the retrieval indexes.")
print("=" * 72)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import json
import hashlib
from datetime import datetime
from pathlib import Path

import pandas as pd
from openpyxl import load_workbook

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
SEARCH_ROOTS = [
    ROOT / "05_EXPERIMENTS",
    ROOT / "06_RECONCILIATION_INPUTS",
    ROOT / "06_PHASE_A_FOUNDATION",
]

RUN_ID = datetime.now().strftime("%Y%m%d_%H%M%S")
OUT = ROOT / "06_PHASE_A_FOUNDATION" / f"benchmark_reconciliation_audit_{RUN_ID}"
OUT.mkdir(parents=True, exist_ok=False)

print("=" * 78)
print("NCERT AI TUTOR — BENCHMARK RECONCILIATION AUDIT")
print("=" * 78)
print("Output:", OUT)

# ------------------------------------------------------------
# 1. Locate benchmark-related workbooks and files
# ------------------------------------------------------------
keywords = [
    "benchmark", "v3", "question", "evidence_review",
    "research_freeze", "reviewed", "annotation"
]
extensions = {".xlsx", ".xlsm", ".csv", ".json"}

found_files = []
for root in SEARCH_ROOTS:
    if not root.exists():
        print("Search directory not found:", root)
        continue

    for current, dirs, files in os.walk(root):
        dirs[:] = [
            d for d in dirs
            if d not in {".ipynb_checkpoints", "__pycache__"}
        ]
        for filename in files:
            p = Path(current) / filename
            if p.suffix.lower() not in extensions:
                continue
            if any(word in filename.lower() for word in keywords):
                found_files.append(p)

found_files = sorted(set(found_files), key=lambda p: str(p).lower())

print(f"\nCandidate files found: {len(found_files)}")
for p in found_files:
    try:
        size = p.stat().st_size / 1024
    except OSError:
        size = 0
    print(f"  {p.relative_to(ROOT)} ({size:.1f} KB)")

# ------------------------------------------------------------
# 2. Inspect spreadsheets without changing them
# ------------------------------------------------------------
workbook_inventory = []
sheet_inventory = []
sample_rows = []
all_question_ids = []

question_id_candidates = [
    "question_id", "question id", "qid", "id",
    "question_code", "question code"
]
question_text_candidates = [
    "question", "question_text", "question text",
    "query", "query_text", "prompt"
]
category_candidates = [
    "question_type", "question type", "category",
    "difficulty", "split", "type"
]
source_candidates = [
    "source_file", "source file", "gold_source",
    "gold source", "book", "source"
]
page_candidates = [
    "gold_pages", "gold pages", "gold_page",
    "gold page", "page", "pages", "pdf_page"
]

def normalized_columns(df):
    return {
        str(c).strip().lower(): c
        for c in df.columns
    }

def pick_column(colmap, candidates):
    for candidate in candidates:
        if candidate in colmap:
            return colmap[candidate]
    return None

for path in found_files:
    rel = str(path.relative_to(ROOT))

    if path.suffix.lower() in {".xlsx", ".xlsm"}:
        try:
            wb = load_workbook(
                path, read_only=True, data_only=True
            )

            workbook_inventory.append({
                "path": rel,
                "type": "spreadsheet",
                "size_bytes": path.stat().st_size,
                "sheet_names": wb.sheetnames
            })

            print("\n" + "-" * 78)
            print("WORKBOOK:", rel)
            print("SHEETS:", wb.sheetnames)

            for sheet_name in wb.sheetnames:
                ws = wb[sheet_name]
                rows = ws.iter_rows(values_only=True)
                first_row = next(rows, None)

                if first_row is None:
                    sheet_inventory.append({
                        "path": rel,
                        "sheet": sheet_name,
                        "max_rows": ws.max_row,
                        "max_columns": ws.max_column,
                        "headers": [],
                        "question_id_column": None,
                        "question_text_column": None
                    })
                    continue

                headers = [
                    str(x).strip() if x is not None else ""
                    for x in first_row
                ]
                records = list(rows)

                # Build a dataframe from the existing worksheet in memory.
                df = pd.DataFrame(records, columns=headers)
                df = df.loc[
                    :, [c for c in df.columns if str(c).strip()]
                ]
                colmap = normalized_columns(df)

                id_col = pick_column(colmap, question_id_candidates)
                q_col = pick_column(colmap, question_text_candidates)
                category_col = pick_column(colmap, category_candidates)
                source_col = pick_column(colmap, source_candidates)
                page_col = pick_column(colmap, page_candidates)

                sheet_info = {
                    "path": rel,
                    "sheet": sheet_name,
                    "max_rows_reported": ws.max_row,
                    "max_columns_reported": ws.max_column,
                    "data_rows_loaded": len(df),
                    "headers": list(df.columns),
                    "question_id_column": str(id_col) if id_col else None,
                    "question_text_column": str(q_col) if q_col else None,
                    "category_column": str(category_col) if category_col else None,
                    "source_column": str(source_col) if source_col else None,
                    "page_column": str(page_col) if page_col else None
                }
                sheet_inventory.append(sheet_info)

                print(f"\n  Sheet: {sheet_name}")
                print(f"  Data rows: {len(df)}")
                print(f"  Columns: {list(df.columns)}")

                if id_col:
                    ids = df[id_col].dropna().astype(str).str.strip()
                    ids = ids[ids.ne("")]
                    print(f"  Non-empty IDs: {len(ids)}")
                    print(f"  Unique IDs: {ids.nunique()}")
                    for value in ids:
                        all_question_ids.append({
                            "path": rel,
                            "sheet": sheet_name,
                            "question_id": value
                        })

                if q_col:
                    questions = df[q_col].dropna().astype(str).str.strip()
                    print(f"  Non-empty question texts: {len(questions)}")

                # Print only a few representative rows, not entire datasets.
                preview_cols = [
                    c for c in [id_col, q_col, category_col, source_col, page_col]
                    if c is not None
                ]
                if preview_cols:
                    preview = df[preview_cols].head(3)
                    print("  Preview:")
                    print(preview.to_string(index=False, max_colwidth=110))

                    for _, row in preview.iterrows():
                        sample_rows.append({
                            "path": rel,
                            "sheet": sheet_name,
                            **{
                                str(c): (
                                    None if pd.isna(row[c]) else str(row[c])[:500]
                                )
                                for c in preview_cols
                            }
                        })

            wb.close()

        except Exception as e:
            print("  ERROR reading workbook:", repr(e))
            workbook_inventory.append({
                "path": rel,
                "type": "spreadsheet",
                "error": repr(e)
            })

    elif path.suffix.lower() == ".csv":
        try:
            df = pd.read_csv(path, nrows=5)
            workbook_inventory.append({
                "path": rel,
                "type": "csv",
                "size_bytes": path.stat().st_size,
                "columns": list(df.columns)
            })
            print("\nCSV:", rel)
            print("  Columns:", list(df.columns))
            print("  Sample rows:")
            print(df.head(2).to_string(index=False, max_colwidth=100))
        except Exception as e:
            workbook_inventory.append({
                "path": rel,
                "type": "csv",
                "error": repr(e)
            })

    elif path.suffix.lower() == ".json":
        try:
            with open(path, "r", encoding="utf-8") as f:
                obj = json.load(f)

            if isinstance(obj, dict):
                summary = {
                    k: v for k, v in obj.items()
                    if not isinstance(v, (dict, list))
                }
                print("\nJSON:", rel)
                print("  Top-level keys:", list(obj.keys())[:35])
                print("  Scalar metadata:", str(summary)[:1000])

                workbook_inventory.append({
                    "path": rel,
                    "type": "json",
                    "top_level_keys": list(obj.keys())
                })
        except Exception as e:
            workbook_inventory.append({
                "path": rel,
                "type": "json",
                "error": repr(e)
            })

# ------------------------------------------------------------
# 3. Detect repeated IDs across files/sheets
# ------------------------------------------------------------
ids_df = pd.DataFrame(all_question_ids)
duplicate_report = []

if not ids_df.empty:
    for qid, group in ids_df.groupby("question_id"):
        if len(group) > 1:
            duplicate_report.append({
                "question_id": qid,
                "occurrences": len(group),
                "locations": " | ".join(
                    f"{r['path']} [{r['sheet']}]"
                    for _, r in group.iterrows()
                )
            })

duplicates_df = pd.DataFrame(duplicate_report)

print("\n" + "=" * 78)
print("AUDIT SUMMARY")
print("=" * 78)
print("Candidate files:", len(found_files))
print("Workbooks/structured files inspected:", len(workbook_inventory))
print("Sheets inspected:", len(sheet_inventory))
print("Question ID occurrences:", len(all_question_ids))
print("IDs repeated across files/sheets:", len(duplicate_report))

if not duplicates_df.empty:
    print("\nRepeated IDs (may be intentional when comparing versions):")
    print(duplicates_df.head(30).to_string(index=False))
else:
    print("\nNo repeated question IDs detected in recognized ID columns.")

# ------------------------------------------------------------
# 4. Save an independent audit report
# ------------------------------------------------------------
pd.DataFrame(workbook_inventory).to_json(
    OUT / "file_inventory.json",
    orient="records",
    indent=2
)
pd.DataFrame(sheet_inventory).to_json(
    OUT / "sheet_inventory.json",
    orient="records",
    indent=2
)
pd.DataFrame(sample_rows).to_json(
    OUT / "question_samples.json",
    orient="records",
    indent=2
)
duplicates_df.to_csv(
    OUT / "repeated_question_ids.csv",
    index=False
)

summary = {
    "audit_run_id": RUN_ID,
    "created_at": datetime.now().astimezone().isoformat(),
    "search_roots": [str(p) for p in SEARCH_ROOTS],
    "candidate_file_count": len(found_files),
    "structured_file_count": len(workbook_inventory),
    "sheet_count": len(sheet_inventory),
    "question_id_occurrences": len(all_question_ids),
    "repeated_question_id_count": len(duplicate_report),
    "original_files_modified": False,
    "audit_output_directory": str(OUT),
    "next_step": (
        "Review the inventory to identify the authoritative 120-question "
        "pool, frozen 10-question pilot, and 75-question V3 template. "
        "Do not generate or overwrite a benchmark until these are reconciled."
    )
}

with open(OUT / "audit_summary.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

print("\nAudit saved to:", OUT)
print("Summary:", OUT / "audit_summary.json")
print("\nNo source workbook or benchmark was edited.")
print("Paste the audit summary and the workbook/sheet inventory here.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime
import json
import pandas as pd
from openpyxl import load_workbook

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation"

FILES = {
    "120_question_pool": BASE / "NCERT_RAG_Benchmark_v1.xlsx",
    "V2_reviewed_pilot": (
        BASE / "benchmark_v2/"
        "NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx"
    ),
    "V3_75_template": (
        BASE / "benchmark_v2/query_expansion_pilot_v1/"
        "research_freeze_and_next_benchmark_v1/"
        "benchmark_v3_75q/NCERT_RAG_Benchmark_V3_75Q_TEMPLATE.xlsx"
    ),
    "English_only_benchmark": (
        BASE / "english_only_v1/english_only_benchmark.xlsx"
    ),
}

OUT = ROOT / "06_PHASE_A_FOUNDATION" / (
    "benchmark_readiness_" + datetime.now().strftime("%Y%m%d_%H%M%S")
)
OUT.mkdir(parents=True, exist_ok=False)

def norm(s):
    return str(s).strip().lower().replace(" ", "_") if s is not None else ""

def inspect_workbook(label, path):
    print("\n" + "=" * 76)
    print(label)
    print("=" * 76)

    if not path.exists():
        print("MISSING:", path)
        return {"label": label, "path": str(path), "exists": False}

    wb = load_workbook(path, read_only=True, data_only=True)
    result = {
        "label": label,
        "path": str(path),
        "exists": True,
        "sheets": {}
    }

    for ws in wb.worksheets:
        rows = list(ws.iter_rows(values_only=True))
        if not rows:
            continue

        headers = [
            str(x).strip() if x is not None else ""
            for x in rows[0]
        ]
        df = pd.DataFrame(rows[1:], columns=headers)
        df = df.loc[:, [c for c in df.columns if c]]

        if df.empty:
            print(f"\nSheet: {ws.title} — no data rows")
            continue

        cols = {norm(c): c for c in df.columns}

        # Recognize common variants across existing benchmark versions.
        question_col = next(
            (cols[c] for c in [
                "question", "question_text", "query"
            ] if c in cols), None
        )
        id_col = next(
            (cols[c] for c in [
                "question_id", "qid", "id"
            ] if c in cols), None
        )
        source_col = next(
            (cols[c] for c in [
                "source_file", "gold_source_file"
            ] if c in cols), None
        )
        answer_col = next(
            (cols[c] for c in [
                "expected_answer", "answer"
            ] if c in cols), None
        )
        pages_col = next(
            (cols[c] for c in [
                "gold_pdf_pages", "gold_pages", "gold_printed_pages"
            ] if c in cols), None
        )
        evidence_col = next(
            (cols[c] for c in [
                "evidence_excerpt", "evidence_excerpt_or_notes",
                "gold_evidence_excerpt"
            ] if c in cols), None
        )
        type_col = next(
            (cols[c] for c in [
                "question_type", "type", "category"
            ] if c in cols), None
        )

        print(f"\nSheet: {ws.title}")
        print("Rows:", len(df))
        print("Columns:", list(df.columns))

        sheet_result = {
            "rows": len(df),
            "columns": list(df.columns)
        }

        if id_col:
            ids = df[id_col].astype("string").str.strip()
            sheet_result["nonempty_ids"] = int(ids.notna().sum())
            sheet_result["unique_ids"] = int(ids.dropna().nunique())
            print("Non-empty IDs:", sheet_result["nonempty_ids"])
            print("Unique IDs:", sheet_result["unique_ids"])

        for field, col in [
            ("question_text", question_col),
            ("source_file", source_col),
            ("expected_answer", answer_col),
            ("gold_pages", pages_col),
            ("evidence_excerpt", evidence_col),
        ]:
            if col:
                vals = df[col]
                populated = vals.notna() & vals.astype(str).str.strip().ne("")
                sheet_result[field + "_filled"] = int(populated.sum())
                sheet_result[field + "_missing"] = int((~populated).sum())
                print(
                    f"{field}: {int(populated.sum())}/{len(df)} filled; "
                    f"{int((~populated).sum())} missing"
                )

        if source_col:
            print("Source-file distribution (top 10):")
            print(
                df[source_col].fillna("[MISSING]")
                .astype(str).value_counts().head(10).to_string()
            )

        for status_name in [
            "annotation_status", "benchmark_status",
            "evidence_verified", "answer_supported_by_evidence",
            "answerability", "split"
        ]:
            if status_name in cols:
                col = cols[status_name]
                counts = (
                    df[col].fillna("[BLANK]")
                    .astype(str).value_counts(dropna=False)
                    .to_dict()
                )
                sheet_result[status_name + "_counts"] = counts
                print(f"{status_name} counts:", counts)

        if type_col:
            counts = (
                df[type_col].fillna("[BLANK]")
                .astype(str).value_counts(dropna=False)
                .to_dict()
            )
            sheet_result["question_type_counts"] = counts
            print("Question types:", counts)

        result["sheets"][ws.title] = sheet_result

    wb.close()
    return result

all_results = []
for label, path in FILES.items():
    all_results.append(inspect_workbook(label, path))

with open(OUT / "readiness_report.json", "w", encoding="utf-8") as f:
    json.dump(all_results, f, indent=2, default=str)

print("\n" + "=" * 76)
print("READINESS REPORT SAVED")
print("=" * 76)
print(OUT / "readiness_report.json")
print("No source workbooks were modified.")
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime
import hashlib
import json
import pandas as pd

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation"

FILES = {
    "full_120": BASE / "NCERT_RAG_Benchmark_v1.xlsx",
    "english_77": BASE / "english_only_v1/english_only_benchmark.xlsx",
    "v2_pilot_10": BASE / "benchmark_v2/NCERT_RAG_Benchmark_V2_Pilot_Reviewed_v1.xlsx",
}

INDEX_DIR = BASE / "english_only_v1"

OUT = ROOT / "06_PHASE_A_FOUNDATION" / (
    "benchmark_subset_reconciliation_"
    + datetime.now().strftime("%Y%m%d_%H%M%S")
)
OUT.mkdir(parents=True, exist_ok=False)


def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()


def load_benchmark(path):
    if not path.exists():
        print("MISSING:", path)
        return None

    df = pd.read_excel(path, sheet_name=0, dtype=str).fillna("")
    for col in df.columns:
        df[col] = df[col].astype(str).str.strip()

    print(f"\nLoaded: {path.name}")
    print("Rows:", len(df))
    print("SHA-256:", sha256(path))
    return df


full = load_benchmark(FILES["full_120"])
eng = load_benchmark(FILES["english_77"])
pilot = load_benchmark(FILES["v2_pilot_10"])

report = {
    "created_at": datetime.now().isoformat(),
    "read_only_audit": True,
    "files": {},
    "comparisons": {},
    "index_files": [],
}

for key, path in FILES.items():
    report["files"][key] = {
        "path": str(path),
        "exists": path.exists(),
        "sha256": sha256(path) if path.exists() else None,
    }

# Compare the full benchmark against the English-only benchmark.
if full is not None and eng is not None:
    required = {"question_id", "question", "source_file"}

    if required.issubset(full.columns) and required.issubset(eng.columns):
        full_ids = set(full["question_id"]) - {""}
        eng_ids = set(eng["question_id"]) - {""}

        full_questions = {
            qid: question
            for qid, question in zip(
                full["question_id"], full["question"]
            )
            if qid
        }
        eng_questions = {
            qid: question
            for qid, question in zip(
                eng["question_id"], eng["question"]
            )
            if qid
        }

        common_ids = full_ids & eng_ids
        same_question = {
            qid for qid in common_ids
            if full_questions.get(qid) == eng_questions.get(qid)
        }
        changed_question = common_ids - same_question

        full_sources = set(full["source_file"]) - {""}
        eng_sources = set(eng["source_file"]) - {""}

        comparison = {
            "full_question_count": len(full_ids),
            "english_question_count": len(eng_ids),
            "common_question_ids": len(common_ids),
            "english_ids_missing_from_full": sorted(eng_ids - full_ids),
            "full_ids_missing_from_english": sorted(full_ids - eng_ids),
            "same_id_and_same_question": len(same_question),
            "same_id_but_changed_question": sorted(changed_question),
            "english_source_paths_not_in_full": sorted(
                eng_sources - full_sources
            ),
            "full_source_paths_not_in_english": sorted(
                full_sources - eng_sources
            ),
        }

        report["comparisons"]["full_vs_english"] = comparison

        print("\n" + "=" * 72)
        print("120 vs 77 QUESTION RECONCILIATION")
        print("=" * 72)
        print(json.dumps(comparison, indent=2))

        # Save the complete English-only question list for manual inspection.
        eng.to_csv(OUT / "english_77_questions_snapshot.csv", index=False)

        # Save all questions with shared IDs but different wording.
        if changed_question:
            changed = eng[
                eng["question_id"].isin(changed_question)
            ].copy()
            changed.to_csv(
                OUT / "same_id_changed_question.csv", index=False
            )

        # Save the IDs and question texts in the 77 set not found in 120.
        if eng_ids - full_ids:
            eng[eng["question_id"].isin(eng_ids - full_ids)].to_csv(
                OUT / "english_questions_not_in_full.csv", index=False
            )

        # Check whether gold source paths actually exist under the source folder.
        source_root = ROOT / "01_SOURCE_BOOKS_V2"
        missing_source_paths = []

        for source in sorted(eng_sources):
            if not (source_root / source).exists():
                missing_source_paths.append(source)

        report["comparisons"]["english_source_file_existence"] = {
            "source_root": str(source_root),
            "unique_gold_sources": len(eng_sources),
            "missing_source_paths": missing_source_paths,
        }

        print("\nEnglish-only gold sources:", len(eng_sources))
        print("Missing source PDFs:", len(missing_source_paths))
        for source in missing_source_paths:
            print("  MISSING:", source)

    else:
        print("\nRequired comparison columns are missing.")
        print("Full columns:", list(full.columns))
        print("English columns:", list(eng.columns))

# Inspect retrieval-index files without assuming a particular artifact filename.
print("\n" + "=" * 72)
print("ENGLISH-ONLY INDEX ARTIFACT INVENTORY")
print("=" * 72)

for path in sorted(INDEX_DIR.rglob("*")):
    if path.is_file():
        item = {
            "relative_path": str(path.relative_to(INDEX_DIR)),
            "size_bytes": path.stat().st_size,
        }
        report["index_files"].append(item)
        print(f'{item["relative_path"]} | {item["size_bytes"]:,} bytes')

# Record pilot IDs independently; do not merge or overwrite the pilot.
if pilot is not None and "question_id" in pilot.columns:
    report["comparisons"]["v2_pilot_ids"] = sorted(
        set(pilot["question_id"]) - {""}
    )

with open(OUT / "reconciliation_report.json", "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)

print("\n" + "=" * 72)
print("AUDIT COMPLETE — NO SOURCE FILES MODIFIED")
print("=" * 72)
print("Report folder:", OUT)
print("JSON report:", OUT / "reconciliation_report.json")
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime
import json, ast, re, math, hashlib
import numpy as np
import pandas as pd

# ==========================================================
# NCERT RAG FINAL VALIDATION
# Read-only for existing artifacts; new outputs only.
# ==========================================================
ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

P = {
    "benchmark": BASE / "english_only_benchmark.xlsx",
    "chunks": BASE / "english_only_chunks.csv",
    "retrieval": BASE / "evaluation/english_only_per_question_retrieval.csv",
    "metrics": BASE / "evaluation/english_only_per_question_metrics.csv",
    "summary": BASE / "evaluation/english_only_overall_summary.csv",
    "eval_config": BASE / "evaluation/evaluation_config.json",
    "manifest": BASE / "indexes/retrieval_manifest.json",
}
missing = [str(p) for p in P.values() if not p.exists()]
if missing:
    raise FileNotFoundError("Missing required files:\n" + "\n".join(missing))

OUT = ROOT / "06_PHASE_A_FOUNDATION" / (
    "final_validation_" + datetime.now().strftime("%Y%m%d_%H%M%S")
)
OUT.mkdir(parents=True, exist_ok=False)

def sha(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for b in iter(lambda: f.read(1024 * 1024), b""):
            h.update(b)
    return h.hexdigest()

def lst(x):
    if x is None or (isinstance(x, float) and np.isnan(x)):
        return []
    if isinstance(x, (list, tuple, np.ndarray)):
        return list(x)
    s = str(x).strip()
    if not s or s.lower() in {"nan", "none", "null"}:
        return []
    for fn in (json.loads, ast.literal_eval):
        try:
            y = fn(s)
            return list(y) if isinstance(y, (list, tuple, np.ndarray)) else [y]
        except Exception:
            pass
    return [s]

def pages(x):
    result = set()
    for v in lst(x):
        result.update(int(n) for n in re.findall(r"\d+", str(v)))
    return result

def dcg(rels):
    return sum(r / math.log2(i + 2) for i, r in enumerate(rels))

def ratio(a, b):
    return a / b if b else np.nan

def norm(x):
    return str(x).strip() if x is not None else ""

print("=" * 72)
print("1. LOADING EXISTING ARTIFACTS")
print("=" * 72)

b = pd.read_excel(P["benchmark"], dtype=str).fillna("")
r = pd.read_csv(P["retrieval"]).fillna("")
old = pd.read_csv(P["metrics"]).fillna("")
old_summary = pd.read_csv(P["summary"]).fillna("")
chunks = pd.read_csv(P["chunks"], low_memory=False)
ec = json.loads(P["eval_config"].read_text())
im = json.loads(P["manifest"].read_text())

for df in (b, r, old):
    for c in df.columns:
        if df[c].dtype == object:
            df[c] = df[c].astype(str).replace({"nan": "", "None": ""}).str.strip()

print("Benchmark:", len(b), "rows /", b.question_id.nunique(), "unique IDs")
print("Retrieval:", len(r), "rows")
print("Saved metrics:", len(old), "rows")
print("Chunks:", len(chunks))

# ----------------------------------------------------------
# 2. INTEGRITY AND INDEX COVERAGE
# ----------------------------------------------------------
print("\n" + "=" * 72)
print("2. INTEGRITY AND SOURCE COVERAGE")
print("=" * 72)

bids = set(b.question_id) - {""}
rids = set(r.question_id) - {""}
mids = set(old.question_id) - {""}

source_col = next((c for c in chunks.columns if c.lower() == "source_file"), None)
gold_source_col = next((c for c in b.columns if c.lower() == "source_file"), None)
indexed_sources = set(chunks[source_col].dropna().astype(str).str.strip()) if source_col else set()
gold_sources = set(b[gold_source_col].astype(str).str.strip()) - {""} if gold_source_col else set()

checks = {
    "benchmark_rows": len(b),
    "unique_benchmark_ids": len(bids),
    "duplicate_benchmark_ids": int(b.question_id.duplicated().sum()),
    "retrieval_rows": len(r),
    "retrieval_unique_ids": len(rids),
    "benchmark_ids_missing_from_retrieval": sorted(bids - rids),
    "unexpected_retrieval_ids": sorted(rids - bids),
    "metrics_rows": len(old),
    "metrics_unique_ids": len(mids),
    "benchmark_ids_missing_from_metrics": sorted(bids - mids),
    "unexpected_metric_ids": sorted(mids - bids),
    "duplicate_retrieval_question_method_rows": int(r.duplicated(["question_id", "method"]).sum()),
    "duplicate_metric_question_method_rows": int(old.duplicated(["question_id", "method"]).sum()),
    "gold_source_count": len(gold_sources),
    "indexed_source_count": len(indexed_sources),
    "gold_sources_missing_from_index": sorted(gold_sources - indexed_sources),
    "empty_gold_chunk_ids": int(
        b.gold_chunk_ids.astype(str).str.strip().isin(["", "nan", "None"]).sum()
    ) if "gold_chunk_ids" in b else None,
}
print(json.dumps(checks, indent=2))

# ----------------------------------------------------------
# 3. INDEPENDENTLY RECOMPUTE METRICS
# Relevance = exact gold source AND at least one gold PDF page
# overlaps the retrieved chunk's recorded page list.
# ----------------------------------------------------------
print("\n" + "=" * 72)
print("3. RECOMPUTING RETRIEVAL METRICS")
print("=" * 72)

gold = b.set_index("question_id", drop=False).to_dict("index")
src_col = "retrieved_source_files" if "retrieved_source_files" in r else None
page_col = "retrieved_pages" if "retrieved_pages" in r else None
id_col = "retrieved_chunk_ids" if "retrieved_chunk_ids" in r else None

if not src_col or not page_col:
    raise ValueError("Retrieval CSV lacks retrieved_source_files or retrieved_pages.")

rows = []
unanswerable_rows = []
flags = []

for _, rr in r.iterrows():
    qid, method = norm(rr.get("question_id")), norm(rr.get("method"))
    g = gold.get(qid)
    if g is None:
        flags.append({"question_id": qid, "method": method, "issue": "ID_NOT_IN_BENCHMARK"})
        continue

    answerability = norm(g.get("answerability")).upper()
    gs = norm(g.get("source_file"))
    gp = pages(g.get("gold_pages"))
    sources = [norm(x) for x in lst(rr[src_col])]
    page_lists = lst(rr[page_col])
    ids = lst(rr[id_col]) if id_col else []

    if len(sources) != len(page_lists):
        flags.append({
            "question_id": qid, "method": method,
            "issue": "SOURCE_PAGE_LIST_LENGTH_MISMATCH",
            "source_count": len(sources), "page_list_count": len(page_lists)
        })

    n = max(len(sources), len(page_lists))
    ranked = []
    for i in range(n):
        s = sources[i] if i < len(sources) else ""
        ps = pages(page_lists[i]) if i < len(page_lists) else set()
        overlap = ps & gp
        relevant = int(answerability == "ANSWERABLE" and s == gs and bool(overlap))
        ranked.append({
            "rank": i + 1, "source_file": s,
            "pages": ps, "overlap": overlap, "relevant": relevant
        })

    if answerability != "ANSWERABLE":
        unanswerable_rows.append({
            "question_id": qid,
            "method": method,
            "same_source_chunks_top10": sum(x["source_file"] == gs for x in ranked[:10]),
            "retrieved_chunks": len(ranked),
            "corpus_absence_human_review_required": True
        })
        continue

    if not gp or not gs:
        flags.append({
            "question_id": qid, "method": method,
            "issue": "ANSWERABLE_GOLD_SOURCE_OR_PAGES_MISSING"
        })

    rels = [x["relevant"] for x in ranked]
    first = next((x["rank"] for x in ranked if x["relevant"]), None)
    out = {
        "question_id": qid,
        "method": method,
        "answerability": answerability,
        "question_type": norm(g.get("question_type")),
        "class": norm(g.get("class")),
        "subject": norm(g.get("subject")),
        "source_file": gs,
        "gold_pages": json.dumps(sorted(gp)),
        "retrieved_count": len(ranked),
        "first_relevant_rank": first,
        "mrr": 1 / first if first else 0.0,
    }

    for k in (1, 3, 5, 10):
        top = ranked[:k]
        top_rels = [x["relevant"] for x in top]
        covered = set()
        for x in top:
            if x["source_file"] == gs:
                covered.update(x["pages"] & gp)

        # This nDCG is normalized against relevant chunks present in
        # the retrieved top-k list, not every relevant chunk in the corpus.
        ideal = sorted(top_rels, reverse=True)
        idcg = dcg(ideal)
        out[f"hit@{k}"] = int(any(top_rels))
        out[f"page_recall@{k}"] = ratio(len(covered), len(gp))
        out[f"chunk_precision@{k}"] = ratio(sum(top_rels), len(top))
        out[f"ndcg@{k}"] = ratio(dcg(top_rels), idcg) if idcg else 0.0

    rows.append(out)

recomp = pd.DataFrame(rows)
unans = pd.DataFrame(unanswerable_rows)
flagdf = pd.DataFrame(flags)

# Save all outputs separately.
recomp.to_csv(OUT / "independent_answerable_metrics.csv", index=False)
unans.to_csv(OUT / "unanswerable_descriptive_only.csv", index=False)
flagdf.to_csv(OUT / "validation_flags.csv", index=False)

# ----------------------------------------------------------
# 4. AGGREGATE METRICS
# ----------------------------------------------------------
summary_rows = []
if not recomp.empty:
    for method, df in recomp.groupby("method"):
        s = {"method": method, "answerable_questions_scored": df.question_id.nunique()}
        for col in [
            "hit@1", "hit@3", "hit@5", "hit@10",
            "page_recall@1", "page_recall@3", "page_recall@5", "page_recall@10",
            "chunk_precision@1", "chunk_precision@3", "chunk_precision@5", "chunk_precision@10",
            "mrr", "ndcg@5", "ndcg@10"
        ]:
            if col in df:
                s[col] = pd.to_numeric(df[col], errors="coerce").mean()
        summary_rows.append(s)

new_summary = pd.DataFrame(summary_rows)
new_summary.to_csv(OUT / "independent_overall_summary.csv", index=False)

# Compare recomputed values to original per-question values.
# nDCG comparison is included diagnostically; its ideal-ranking protocol
# may differ from the original implementation.
compare_cols = [
    "hit@1", "hit@3", "hit@5", "hit@10",
    "page_recall@1", "page_recall@3", "page_recall@5", "page_recall@10",
    "chunk_precision@1", "chunk_precision@3", "chunk_precision@5", "chunk_precision@10",
    "mrr", "ndcg@5", "ndcg@10"
]
comparison_rows = []
if not recomp.empty:
    merged = recomp.merge(old, on=["question_id", "method"], how="left",
                          suffixes=("_new", "_old"), indicator=True)
    for _, x in merged.iterrows():
        for c in compare_cols:
            a, z = c + "_new", c + "_old"
            if a not in merged.columns or z not in merged.columns:
                continue
            try:
                nv, ov = float(x[a]), float(x[z])
            except (TypeError, ValueError):
                continue
            diff = nv - ov if np.isfinite(nv) and np.isfinite(ov) else np.nan
            comparison_rows.append({
                "question_id": x["question_id"], "method": x["method"],
                "metric": c, "recomputed": nv, "saved": ov,
                "difference": diff,
                "match_within_1e-6": bool(np.isfinite(diff) and abs(diff) <= 1e-6)
            })

comparison = pd.DataFrame(comparison_rows)
comparison.to_csv(OUT / "saved_vs_recomputed_metrics.csv", index=False)

# Compare top-k config with actual saved ranking lengths.
retrieval_lengths = {}
if id_col:
    for method, df in r.groupby("method"):
        retrieval_lengths[method] = sorted(set(len(lst(x)) for x in df[id_col]))

eval_retr = ec.get("retrieval", {})
hybrid = im.get("hybrid", {})
config = {
    "evaluation_top_k": eval_retr.get("retrieval_top_k"),
    "manifest_hybrid_top_k": hybrid.get("top_k"),
    "saved_retrieved_chunk_counts_by_method": retrieval_lengths,
    "evaluation_candidate_k": eval_retr.get("candidate_k"),
    "manifest_candidate_k": hybrid.get("candidate_k"),
    "evaluation_rrf_k": eval_retr.get("rrf_k"),
    "manifest_rrf_k": hybrid.get("rrf_k"),
}

# ----------------------------------------------------------
# 5. FINAL STATUS AND REPORT
# ----------------------------------------------------------
mismatch_count = (
    int((~comparison["match_within_1e-6"]).sum())
    if not comparison.empty else None
)
matched_count = (
    int(comparison["match_within_1e-6"].sum())
    if not comparison.empty else None
)

report = {
    "created_at": datetime.now().isoformat(),
    "source_sha256": {k: sha(v) for k, v in P.items()},
    "integrity_checks": checks,
    "configuration_check": config,
    "recomputed_answerable_summary": summary_rows,
    "saved_vs_recomputed": {
        "metric_comparison_rows": len(comparison),
        "matching_values_within_1e-6": matched_count,
        "mismatching_or_uncomparable_values": mismatch_count
    },
    "answerability_counts": b["answerability"].value_counts().to_dict(),
    "benchmark_annotation_status": (
        b["annotation_status"].value_counts().to_dict()
        if "annotation_status" in b else {}
    ),
    "limitations": [
        "Retrieval relevance is defined as exact source-file match plus overlap with recorded gold PDF pages.",
        "This validates the existing labels, not the independent factual correctness of every question and expected answer.",
        "The 12 unanswerable questions are descriptive only; human corpus-absence verification is still required.",
        "The manifest top_k and evaluation top_k disagree. Saved lists contain the recorded number of retrieved ranks.",
        "nDCG is recomputed using the relevant items in each retrieved top-k list; compare its definition with the original evaluator before reporting it.",
        "Gold and retrieved page values are assumed to use the same PDF-page numbering convention."
    ],
    "outputs": [
        "independent_answerable_metrics.csv",
        "independent_overall_summary.csv",
        "saved_vs_recomputed_metrics.csv",
        "unanswerable_descriptive_only.csv",
        "validation_flags.csv",
        "validation_report.json"
    ]
}

(OUT / "validation_report.json").write_text(
    json.dumps(report, indent=2, ensure_ascii=False, default=str),
    encoding="utf-8"
)

print("\n" + "=" * 72)
print("INDEPENDENT SUMMARY — ANSWERABLE QUESTIONS ONLY")
print("=" * 72)
print(new_summary.round(4).to_string(index=False) if not new_summary.empty else "No metrics recomputed.")

print("\n" + "=" * 72)
print("SAVED VS RECOMPUTED")
print("=" * 72)
print("Metric comparisons:", len(comparison))
print("Matching within 1e-6:", matched_count)
print("Mismatches / uncomparable:", mismatch_count)
if not comparison.empty:
    print("\nMismatch examples:")
    bad = comparison[comparison["match_within_1e-6"] == False]
    print(bad.head(15).to_string(index=False) if len(bad) else "None.")

print("\n" + "=" * 72)
print("CONFIGURATION CHECK")
print("=" * 72)
print(json.dumps(config, indent=2))

print("\n" + "=" * 72)
print("QUALITY FLAGS")
print("=" * 72)
print("Flag rows:", len(flagdf))
if not flagdf.empty:
    print(flagdf.head(20).to_string(index=False))
print("\nUnanswerable descriptive rows:", len(unans))

print("\n" + "=" * 72)
print("VALIDATION FINISHED — ORIGINAL FILES UNMODIFIED")
print("=" * 72)
print("Output folder:", OUT)
print("Main report:", OUT / "validation_report.json")
print("Independent summary:", OUT / "independent_overall_summary.csv")
print("Saved-vs-recomputed comparison:", OUT / "saved_vs_recomputed_metrics.csv")
print("Quality flags:", OUT / "validation_flags.csv")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime
import pandas as pd
import json, re, hashlib

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
CORPUS = ROOT / "02_PROCESSED_DATA/corpus_v2_clean"
FOUNDATION = ROOT / "06_PHASE_A_FOUNDATION"

BENCHMARK = BASE / "english_only_benchmark.xlsx"
RETRIEVAL = BASE / "evaluation/english_only_per_question_retrieval.csv"

assert BENCHMARK.exists(), f"Missing benchmark: {BENCHMARK}"
assert RETRIEVAL.exists(), f"Missing retrieval results: {RETRIEVAL}"
assert CORPUS.exists(), f"Missing corpus folder: {CORPUS}"

def norm(x):
    return re.sub(r"[^a-z0-9]+", "_", str(x).strip().lower()).strip("_")

def find_col(df, options):
    return next((c for c in options if c in df.columns), None)

def read_table(path):
    try:
        if path.suffix.lower() == ".csv":
            return pd.read_csv(path, low_memory=False)
        if path.suffix.lower() in [".jsonl", ".ndjson"]:
            return pd.read_json(path, lines=True)
        if path.suffix.lower() == ".json":
            obj = json.loads(path.read_text(encoding="utf-8"))
            if isinstance(obj, list):
                return pd.DataFrame(obj)
            if isinstance(obj, dict):
                for key in ["pages", "records", "data", "documents"]:
                    if isinstance(obj.get(key), list):
                        return pd.DataFrame(obj[key])
                return pd.DataFrame([obj])
    except Exception as e:
        print("Skipped unreadable file:", path.name, str(e)[:120])
    return None

# 1. Load the existing English-only benchmark.
xls = pd.ExcelFile(BENCHMARK)
bench = None
for sheet in xls.sheet_names:
    candidate = pd.read_excel(BENCHMARK, sheet_name=sheet)
    candidate.columns = [norm(c) for c in candidate.columns]
    if (find_col(candidate, ["question_id", "id"]) and
        find_col(candidate, ["question", "question_text", "query"])):
        bench = candidate.copy()
        break

assert bench is not None, "Could not identify benchmark question sheet."

id_col = find_col(bench, ["question_id", "id"])
q_col = find_col(bench, ["question", "question_text", "query"])
answer_col = find_col(bench, ["expected_answer", "gold_answer", "reference_answer", "answer"])
evidence_col = find_col(bench, ["evidence_excerpt", "gold_evidence", "evidence", "context"])
source_col = find_col(bench, ["source_file", "gold_source_file"])
pages_col = find_col(bench, ["gold_pages", "gold_page", "source_pages"])
type_col = find_col(bench, ["question_type", "type", "category"])
class_col = find_col(bench, ["class", "grade"])
subject_col = find_col(bench, ["subject"])
book_col = find_col(bench, ["book", "book_title"])

assert id_col and q_col, "Missing question ID or question text."

# 2. Read source-page records from the processed corpus.
tables = []
for p in CORPUS.rglob("*"):
    if p.is_file() and p.suffix.lower() in [".csv", ".json", ".jsonl", ".ndjson"]:
        df = read_table(p)
        if df is not None and len(df):
            df.columns = [norm(c) for c in df.columns]
            text_col = find_col(df, ["text", "page_text", "content", "extracted_text"])
            if text_col:
                df["_corpus_file"] = str(p)
                tables.append(df)

assert tables, (
    "No readable page-text tables found in corpus_v2_clean. "
    "No original files have been changed."
)
corpus = pd.concat(tables, ignore_index=True, sort=False)

ctext = find_col(corpus, ["text", "page_text", "content", "extracted_text"])
csource = find_col(corpus, ["source_file", "source_path", "filename", "file_name"])
cpage = find_col(corpus, ["page", "page_number", "page_num", "pdf_page"])
cclass = find_col(corpus, ["class", "grade"])
csubject = find_col(corpus, ["subject"])
cbook = find_col(corpus, ["book", "book_title"])

def canonical_path(x):
    s = str(x or "").replace("\\", "/").strip()
    # Compare relative source paths where possible.
    if "/01_SOURCE_BOOKS_V2/" in s:
        s = s.split("/01_SOURCE_BOOKS_V2/", 1)[1]
    return s.lstrip("./").lower()

def parse_pages(x):
    if pd.isna(x):
        return set()
    if isinstance(x, (int, float)):
        return {int(x)}
    s = str(x)
    nums = re.findall(r"\d+", s)
    return {int(n) for n in nums}

# Build exact source/page lookup for evidence checking.
corpus["_source_key"] = corpus[csource].map(canonical_path) if csource else ""
corpus["_page_num"] = pd.to_numeric(corpus[cpage], errors="coerce") if cpage else pd.NA

def source_rows(row):
    if not csource:
        return corpus.iloc[0:0]
    src = canonical_path(row.get(source_col, ""))
    if not src:
        return corpus.iloc[0:0]
    exact = corpus[corpus["_source_key"] == src]
    if len(exact):
        return exact
    # Permit a unique suffix match, never an ambiguous fuzzy match.
    matches = corpus[corpus["_source_key"].map(
        lambda z: bool(z) and (z.endswith("/" + src) or src.endswith("/" + z))
    )]
    keys = matches["_source_key"].dropna().unique()
    return matches if len(keys) == 1 else corpus.iloc[0:0]

# 3. Flag questionable benchmark records and collect actual page text.
generic_patterns = [
    r"what activity, idea, or procedure is described",
    r"according to the ncert passage",
    r"what is described",
    r"what does the passage say",
    r"what activity is described",
]
ocr_pattern = re.compile(
    r"�|\\x[0-9a-f]{2}|\\u[0-9a-f]{4}|"
    r"(?:[A-Za-z]{1,2}\\s+){5,}[A-Za-z]{1,3}\\b|"
    r"\\b[A-Za-z]{1,2}\\s+[A-Za-z]{1,2}\\s+[A-Za-z]{1,2}\\b"
)

out = []
for _, row in bench.iterrows():
    q = str(row.get(q_col, "") or "").strip()
    ans = str(row.get(answer_col, "") or "").strip() if answer_col else ""
    ev = str(row.get(evidence_col, "") or "").strip() if evidence_col else ""
    src = str(row.get(source_col, "") or "").strip() if source_col else ""
    pages = parse_pages(row.get(pages_col)) if pages_col else set()

    flags = []
    if len(q) < 12:
        flags.append("QUESTION_TOO_SHORT")
    if any(re.search(p, q, re.I) for p in generic_patterns):
        flags.append("GENERIC_QUESTION")
    if q.lower().startswith(("according to the ncert passage", "according to the passage")):
        flags.append("LOCATION_GIVES_AWAY_EVIDENCE")
    if not ans or ans.lower() in ["nan", "none"]:
        flags.append("MISSING_REFERENCE_ANSWER")
    if len(ans) > 700:
        flags.append("LONG_REFERENCE_ANSWER_REVIEW")
    if ocr_pattern.search(ans) or "\ufffd" in ans:
        flags.append("POSSIBLE_OCR_OR_ENCODING_NOISE")
    if not ev or ev.lower() in ["nan", "none"]:
        flags.append("MISSING_EVIDENCE_EXCERPT")
    if not src:
        flags.append("MISSING_GOLD_SOURCE")
    if not pages:
        flags.append("MISSING_GOLD_PAGE")

    matched = source_rows(row)
    source_text = ""
    page_match_status = "SOURCE_NOT_MATCHED"
    if len(matched):
        if cpage and pages:
            pm = matched[matched["_page_num"].isin(pages)]
            if len(pm):
                source_text = "\n\n".join(
                    f"PDF page {r.get(cpage)}: {str(r.get(ctext, ''))}"
                    for _, r in pm.iterrows()
                )
                page_match_status = "SOURCE_AND_PAGE_MATCHED"
            else:
                source_text = "\n\n".join(
                    f"PDF page {r.get(cpage)}: {str(r.get(ctext, ''))}"
                    for _, r in matched.head(4).iterrows()
                )
                page_match_status = "SOURCE_MATCH_PAGE_NOT_FOUND"
        else:
            source_text = "\n\n".join(
                str(v) for v in matched[ctext].head(4).tolist()
            )
            page_match_status = "SOURCE_MATCHED_PAGE_UNVERIFIED"

    if page_match_status != "SOURCE_AND_PAGE_MATCHED":
        flags.append("SOURCE_PAGE_NEEDS_VERIFICATION")

    # This is deliberately a draft, not a fabricated answer.
    proposed_q = q
    if any(re.search(p, q, re.I) for p in generic_patterns):
        proposed_q = ""
        flags.append("REWRITE_QUESTION_FROM_SOURCE")

    out.append({
        "question_id": str(row[id_col]),
        "class": row.get(class_col, "") if class_col else "",
        "subject": row.get(subject_col, "") if subject_col else "",
        "book": row.get(book_col, "") if book_col else "",
        "question_type": row.get(type_col, "") if type_col else "",
        "original_question": q,
        "original_expected_answer": ans,
        "original_evidence_excerpt": ev,
        "gold_source_file": src,
        "gold_pages": str(row.get(pages_col, "")) if pages_col else "",
        "source_page_match_status": page_match_status,
        "matched_source_page_text": source_text,
        "proposed_question_draft": proposed_q,
        "proposed_answer_draft": "",
        "review_flags": "; ".join(dict.fromkeys(flags)) if flags else "NO_AUTOMATIC_FLAGS",
        "human_question_check": "",
        "human_answer_check": "",
        "human_evidence_check": "",
        "corrected_question": "",
        "corrected_answer": "",
        "verified_source_pages": "",
        "reviewer_notes": "",
        "benchmark_status": "NOT_HUMAN_VERIFIED"
    })

candidate = pd.DataFrame(out)

# 4. Bring in the saved retrieval rows for the same questions.
retrieval = pd.read_csv(RETRIEVAL, low_memory=False)
retrieval.columns = [norm(c) for c in retrieval.columns]
rid = find_col(retrieval, ["question_id", "id"])
assert rid, "Saved retrieval results have no question ID column."
retrieval[rid] = retrieval[rid].astype(str).str.strip()
retrieval = retrieval[retrieval[rid].isin(set(candidate["question_id"]))].copy()

# Keep all original benchmark rows; flag multi-page questions separately.
is_multi = candidate["question_type"].astype(str).str.lower().str.contains(
    r"multi[\s_-]*page|multi[\s_-]*hop", regex=True, na=False
)
multi = candidate[is_multi].copy()

# 5. Save as a new candidate package; never overwrite originals.
stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
out_dir = FOUNDATION / f"benchmark_candidate_v2_{stamp}"
out_dir.mkdir(parents=True, exist_ok=False)

out_xlsx = out_dir / "NCERT_RAG_Benchmark_Candidate_V2.xlsx"
with pd.ExcelWriter(out_xlsx, engine="openpyxl") as writer:
    pd.DataFrame({"Instructions": [
        "This is a candidate revision, not a validated gold benchmark.",
        "Original questions and answers are retained for comparison.",
        "Review source_page_match_status and matched_source_page_text against the source PDF.",
        "A matching source/page is not proof that the expected answer is correct.",
        "Rewrite generic questions using the actual textbook content; do not invent details.",
        "Fill proposed_answer_draft only after verifying the source.",
        "Use YES/NO in the human review columns; leave uncertain cases pending.",
        "Do not report final research metrics using this candidate until its gold evidence is reviewed.",
        f"Source benchmark SHA256: {hashlib.sha256(BENCHMARK.read_bytes()).hexdigest()}",
        f"Source retrieval CSV SHA256: {hashlib.sha256(RETRIEVAL.read_bytes()).hexdigest()}",
    ]}).to_excel(writer, sheet_name="READ_ME", index=False)
    candidate.to_excel(writer, sheet_name="All 77 Candidates", index=False)
    multi.to_excel(writer, sheet_name="Multi-page 13", index=False)
    retrieval.to_excel(writer, sheet_name="Saved Retrieval Results", index=False)

# CSV copies make it easy to inspect/use programmatically.
candidate.to_csv(out_dir / "all_77_candidates.csv", index=False)
multi.to_csv(out_dir / "multi_page_13_candidates.csv", index=False)
retrieval.to_csv(out_dir / "saved_retrieval_results.csv", index=False)

report = {
    "status": "CANDIDATE_ONLY_NOT_VALIDATED",
    "benchmark_input": str(BENCHMARK),
    "retrieval_input": str(RETRIEVAL),
    "corpus_root": str(CORPUS),
    "questions": int(len(candidate)),
    "multi_page_questions": int(len(multi)),
    "retrieval_rows": int(len(retrieval)),
    "flag_counts": candidate["review_flags"].str.split("; ").explode().value_counts().to_dict(),
    "source_page_match_counts": candidate["source_page_match_status"].value_counts().to_dict(),
    "output_workbook": str(out_xlsx),
    "original_files_modified": False
}
(out_dir / "candidate_report.json").write_text(json.dumps(report, indent=2), encoding="utf-8")

print("CANDIDATE WORKBOOK CREATED")
print("Output folder:", out_dir)
print("Workbook:", out_xlsx)
print("Questions:", len(candidate), "| Multi-page:", len(multi))
print("Saved retrieval rows:", len(retrieval))
print("\nSource/page match statuses:")
print(candidate["source_page_match_status"].value_counts().to_string())
print("\nMost common review flags:")
print(candidate["review_flags"].str.split("; ").explode().value_counts().head(12).to_string())
print("\nOriginal benchmark and indexes were not modified.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import json

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
CORPUS = ROOT / "02_PROCESSED_DATA/corpus_v2_clean"
BENCH = BASE / "english_only_benchmark.xlsx"

def norm(x):
    import re
    return re.sub(r"[^a-z0-9]+", "_", str(x).lower()).strip("_")

def read_file(p):
    try:
        if p.suffix.lower() == ".csv":
            return pd.read_csv(p, low_memory=False)
        if p.suffix.lower() == ".jsonl":
            return pd.read_json(p, lines=True)
        if p.suffix.lower() == ".json":
            obj = json.loads(p.read_text(encoding="utf-8"))
            if isinstance(obj, list):
                return pd.DataFrame(obj)
            if isinstance(obj, dict):
                for k in ["pages", "records", "data", "documents"]:
                    if isinstance(obj.get(k), list):
                        return pd.DataFrame(obj[k])
                return pd.DataFrame([obj])
    except Exception as e:
        print("Read error:", p.name, str(e)[:100])
    return None

print("=== BENCHMARK STRUCTURE ===")
xl = pd.ExcelFile(BENCH)
b = None
for sh in xl.sheet_names:
    d = pd.read_excel(BENCH, sheet_name=sh)
    d.columns = [norm(c) for c in d.columns]
    if "question_id" in d.columns and any(
        c in d.columns for c in ["question", "question_text", "query"]
    ):
        b = d
        print("Sheet:", sh)
        print("Columns:", list(d.columns))
        break

assert b is not None, "Could not find benchmark question sheet."

for c in ["question_id", "source_file", "gold_pages", "gold_page",
          "expected_answer", "evidence_excerpt", "question_type"]:
    if c in b.columns:
        print(f"\n{c} examples:")
        print(b[c].dropna().astype(str).head(5).to_list())

print("\n=== CORPUS STRUCTURE ===")
files = [
    p for p in CORPUS.rglob("*")
    if p.is_file() and p.suffix.lower() in [".csv", ".json", ".jsonl"]
]
print("Structured files found:", len(files))

for p in files[:12]:
    d = read_file(p)
    if d is None or not len(d):
        continue
    d.columns = [norm(c) for c in d.columns]
    text_cols = [c for c in d.columns if c in
                 ["text", "page_text", "content", "extracted_text"]]
    if not text_cols:
        continue
    print("\nFile:", p.relative_to(CORPUS))
    print("Rows:", len(d), "Columns:", list(d.columns))
    for c in ["source_file", "source_path", "filename", "file_name",
              "page", "page_number", "page_num", "pdf_page",
              "class", "subject", "book"]:
        if c in d.columns:
            print(f"{c} examples:", d[c].dropna().astype(str).head(4).to_list())

print("\n=== BENCHMARK SOURCE FORMAT ===")
if "source_file" in b.columns:
    print(b["source_file"].dropna().astype(str).drop_duplicates().head(10).to_list())

print("\nDiagnostic complete. No files changed.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import json, re
from datetime import datetime
from difflib import SequenceMatcher

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
CORPUS = ROOT / "02_PROCESSED_DATA/corpus_v2_clean"
FOUNDATION = ROOT / "06_PHASE_A_FOUNDATION"
BENCH = BASE / "english_only_benchmark.xlsx"

def norm(s):
    return re.sub(r"[^a-z0-9]+", "_", str(s).lower()).strip("_")

def col(df, names):
    return next((n for n in names if n in df.columns), None)

# Load benchmark
xls = pd.ExcelFile(BENCH)
bench = None
for sheet in xls.sheet_names:
    d = pd.read_excel(BENCH, sheet_name=sheet)
    d.columns = [norm(c) for c in d.columns]
    if "question_id" in d.columns and "source_file" in d.columns:
        bench = d.copy()
        break
assert bench is not None, "Benchmark sheet not found."

# Load all actual page records from nested per-book JSON files.
records = []
for p in (CORPUS / "page_text").glob("*.json"):
    try:
        obj = json.loads(p.read_text(encoding="utf-8"))
        for r in obj.get("pages", []):
            r["_page_json"] = str(p)
            records.append(r)
    except Exception as e:
        print("Skipped:", p.name, str(e)[:100])

assert records, "No page records found."
pages = pd.DataFrame(records)
pages["source_key"] = pages["source_file"].astype(str).str.replace("\\", "/", regex=False).str.lower()
pages["pdf_page"] = pd.to_numeric(pages["page"], errors="coerce")

def parse_list(value):
    if isinstance(value, list):
        return [int(x) for x in value if str(x).isdigit()]
    if pd.isna(value):
        return []
    try:
        obj = json.loads(str(value))
        if isinstance(obj, list):
            return [int(x) for x in obj if str(x).isdigit()]
    except Exception:
        pass
    return [int(x) for x in re.findall(r"\d+", str(value))]

def clean(s):
    return re.sub(r"\s+", " ", str(s or "")).strip().lower()

def excerpt_similarity(a, b):
    a, b = clean(a), clean(b)
    if not a or not b:
        return 0.0
    if a in b:
        return 1.0
    # Similarity based on the longest evidence text, useful for OCR variation.
    return SequenceMatcher(None, a[:3000], b[:3000]).ratio()

results = []
for _, r in bench.iterrows():
    source = str(r["source_file"]).replace("\\", "/").lower().strip()
    gold_pages = parse_list(r.get("gold_pages", ""))
    evidence = str(r.get("evidence_excerpt", "") or "")
    ans = str(r.get("expected_answer", "") or "")

    matched = pages[pages["source_key"] == source].copy()
    if matched.empty:
        results.append({
            "question_id": r["question_id"],
            "source_file": source,
            "gold_pages": str(gold_pages),
            "source_match": "NO_EXACT_SOURCE_MATCH",
            "candidate_pdf_pages": "",
            "gold_page_text": "",
            "evidence_similarity": 0.0,
            "answer_similarity": 0.0,
            "mapping_assessment": "SOURCE_MAPPING_FAILED"
        })
        continue

    # Inspect the specified page numbers exactly as stored in gold_pages.
    exact = matched[matched["pdf_page"].isin(gold_pages)] if gold_pages else matched.iloc[0:0]
    exact_text = "\n\n".join(
        f"PDF page {int(x['pdf_page'])}: {x.get('clean_text', '')}"
        for _, x in exact.iterrows()
    )

    # Find the strongest evidence/answer match across pages in this source.
    page_scores = []
    for _, pr in matched.iterrows():
        text = str(pr.get("clean_text", "") or "")
        page_scores.append((
            excerpt_similarity(evidence, text),
            excerpt_similarity(ans, text),
            int(pr["pdf_page"]) if pd.notna(pr["pdf_page"]) else None,
            text
        ))

    best_ev = max(page_scores, key=lambda x: x[0], default=(0, 0, None, ""))
    best_ans = max(page_scores, key=lambda x: x[1], default=(0, 0, None, ""))

    # A page is a candidate only; it is not proof of the printed-page mapping.
    if not gold_pages:
        assessment = "GOLD_PAGE_MISSING"
    elif len(exact):
        assessment = "PDF_PAGE_MATCH_EXISTS_VERIFY_PRINTED_PAGE"
    else:
        assessment = "GOLD_PAGE_NOT_FOUND_AS_PDF_PAGE"

    results.append({
        "question_id": r["question_id"],
        "question_type": r.get("question_type", ""),
        "source_file": source,
        "gold_pages": str(gold_pages),
        "source_match": "EXACT_SOURCE_MATCH",
        "candidate_pdf_pages": ", ".join(
            str(int(x)) for x in exact["pdf_page"].dropna().tolist()
        ),
        "gold_page_text": exact_text,
        "best_evidence_match_pdf_page": best_ev[2],
        "best_evidence_match_score": round(best_ev[0], 3),
        "best_answer_match_pdf_page": best_ans[2],
        "best_answer_match_score": round(best_ans[1], 3),
        "best_evidence_match_text": best_ev[3],
        "mapping_assessment": assessment
    })

report = pd.DataFrame(results)
stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
outdir = FOUNDATION / f"evidence_mapping_diagnostic_{stamp}"
outdir.mkdir(parents=True, exist_ok=False)
out = outdir / "evidence_mapping_diagnostic.xlsx"

with pd.ExcelWriter(out, engine="openpyxl") as writer:
    report.to_excel(writer, sheet_name="Mapping Results", index=False)
    pages[["source_file", "page", "page_id", "clean_text"]].head(100).to_excel(
        writer, sheet_name="Page Record Samples", index=False
    )

print("Workbook:", out)
print("\nSource matching:")
print(report["source_match"].value_counts().to_string())
print("\nMapping assessments:")
print(report["mapping_assessment"].value_counts().to_string())
print("\nEvidence similarity summary:")
print(report["best_evidence_match_score"].describe().round(3).to_string())
print("\nSample rows:")
print(report[[
    "question_id", "gold_pages", "candidate_pdf_pages",
    "best_evidence_match_pdf_page", "best_evidence_match_score",
    "mapping_assessment"
]].head(15).to_string(index=False))
print("\nOriginal benchmark and indexes were not modified.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime
import pandas as pd
import re

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
FOUNDATION = ROOT / "06_PHASE_A_FOUNDATION"

DIAG = FOUNDATION / "evidence_mapping_diagnostic_20261010_155214/evidence_mapping_diagnostic.xlsx"
BENCH = BASE / "english_only_benchmark.xlsx"

assert DIAG.exists(), f"Missing diagnostic workbook: {DIAG}"
assert BENCH.exists(), f"Missing benchmark: {BENCH}"

mapping = pd.read_excel(DIAG, sheet_name="Mapping Results")
mapping.columns = [re.sub(r"[^a-z0-9]+", "_", str(c).lower()).strip("_")
                   for c in mapping.columns]

xls = pd.ExcelFile(BENCH)
benchmark = None
for sh in xls.sheet_names:
    df = pd.read_excel(BENCH, sheet_name=sh)
    df.columns = [re.sub(r"[^a-z0-9]+", "_", str(c).lower()).strip("_")
                  for c in df.columns]
    if "question_id" in df.columns and "question" in df.columns:
        benchmark = df.copy()
        break

assert benchmark is not None, "Question sheet not found."

# Merge original questions with diagnostic evidence results.
df = benchmark.merge(mapping, on="question_id", how="left", suffixes=("", "_diag"))

def is_missing_pages(x):
    if pd.isna(x):
        return True
    return str(x).strip() in ["", "[]", "nan", "None"]

def generic_question(x):
    s = str(x or "").lower()
    patterns = [
        "what activity, idea, or procedure is described",
        "according to the ncert passage",
        "according to the passage",
        "what is described in the passage",
        "what activity is described"
    ]
    return any(p in s for p in patterns)

df["needs_page_repair"] = df["gold_pages"].map(is_missing_pages)
df["low_evidence_similarity"] = (
    pd.to_numeric(df["best_evidence_match_score"], errors="coerce") < 0.85
)
df["generic_or_location_leaking_question"] = df["question"].map(generic_question)

df["suggested_gold_pages"] = df["best_evidence_match_pdf_page"].apply(
    lambda x: f"[{int(x)}]" if pd.notna(x) else ""
)
df["page_repair_decision"] = ""
df["evidence_review_decision"] = ""
df["question_rewrite_decision"] = ""
df["reviewer_notes"] = ""

# A page suggestion is not a verified correction.
df["review_status"] = "PENDING_SOURCE_CHECK"

triage = df[
    df["needs_page_repair"]
    | df["low_evidence_similarity"]
    | df["generic_or_location_leaking_question"]
].copy()

outdir = FOUNDATION / (
    "benchmark_triage_" + datetime.now().strftime("%Y%m%d_%H%M%S")
)
outdir.mkdir(parents=True, exist_ok=False)
out = outdir / "NCERT_Benchmark_Triage.xlsx"

with pd.ExcelWriter(out, engine="openpyxl") as writer:
    pd.DataFrame({"Instructions": [
        "This workbook is a triage aid, not a verified benchmark.",
        "Suggested gold pages are the strongest evidence-match PDF page, not an automatic correction.",
        "Check the source PDF and confirm whether the page number is a PDF page or printed page.",
        "Check answer correctness and question clarity separately from text similarity.",
        "Keep unresolved records pending; do not treat blanks as verified.",
        "The original benchmark and diagnostic workbook are unchanged."
    ]}).to_excel(writer, sheet_name="READ_ME", index=False)
    triage.to_excel(writer, sheet_name="Needs Attention", index=False)
    df.to_excel(writer, sheet_name="All 77 Status", index=False)

print("Triage workbook:", out)
print("Total questions:", len(df))
print("Questions needing attention:", len(triage))
print("Missing gold pages:", int(df["needs_page_repair"].sum()))
print("Low evidence similarity (<0.85):", int(df["low_evidence_similarity"].sum()))
print("Generic/location-leaking questions:",
      int(df["generic_or_location_leaking_question"].sum()))
print("\nMost useful next review: open the Needs Attention sheet.")
print("Original files modified: No")
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime
import pandas as pd
import json, re, hashlib

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
CORPUS = ROOT / "02_PROCESSED_DATA/corpus_v2_clean"
FOUNDATION = ROOT / "06_PHASE_A_FOUNDATION"

TRIAGE = FOUNDATION / "benchmark_triage_20261010_162546/NCERT_Benchmark_Triage.xlsx"
BENCH = BASE / "english_only_benchmark.xlsx"

assert TRIAGE.exists(), f"Missing triage workbook: {TRIAGE}"
assert BENCH.exists(), f"Missing benchmark: {BENCH}"

def norm(s):
    return re.sub(r"[^a-z0-9]+", "_", str(s).lower()).strip("_")

def find_col(df, choices):
    return next((c for c in choices if c in df.columns), None)

def load_question_sheet(path):
    for sh in pd.ExcelFile(path).sheet_names:
        d = pd.read_excel(path, sheet_name=sh)
        d.columns = [norm(c) for c in d.columns]
        if "question_id" in d.columns and "question" in d.columns:
            return d
    raise ValueError(f"Question sheet not found in {path}")

# Load the triage results and original benchmark.
triage = pd.read_excel(TRIAGE, sheet_name="All 77 Status")
triage.columns = [norm(c) for c in triage.columns]
original = load_question_sheet(BENCH)

# Preserve original columns as the authoritative record.
df = original.merge(
    triage.drop(columns=[
        c for c in original.columns
        if c != "question_id" and c in triage.columns
    ]),
    on="question_id", how="left", suffixes=("", "_triage")
)

# Read actual page records from the corpus.
page_records = []
for p in (CORPUS / "page_text").glob("*.json"):
    try:
        obj = json.loads(p.read_text(encoding="utf-8"))
        for page in obj.get("pages", []):
            page_records.append(page)
    except Exception:
        pass

pages = pd.DataFrame(page_records)
assert len(pages), "No corpus page records found."

pages["source_key"] = (
    pages["source_file"].astype(str)
    .str.replace("\\", "/", regex=False).str.lower()
)
pages["page_num"] = pd.to_numeric(pages["page"], errors="coerce")

def parse_pages(value):
    if isinstance(value, list):
        return [int(x) for x in value if str(x).isdigit()]
    if pd.isna(value):
        return []
    try:
        x = json.loads(str(value))
        if isinstance(x, list):
            return [int(n) for n in x if str(n).isdigit()]
    except Exception:
        pass
    return [int(n) for n in re.findall(r"\d+", str(value))]

def clean_text(s):
    s = str(s or "").replace("\ufffd", " ")
    s = re.sub(r"\s+", " ", s).strip()
    return s

def first_sentences(s, limit=2):
    s = clean_text(s)
    # Avoid treating every abbreviation/decimal as a sentence boundary.
    parts = re.split(r"(?<=[.!?])\s+(?=[A-Z0-9])", s)
    return " ".join(parts[:limit]).strip()

def generic_question(q):
    q = str(q or "").lower()
    patterns = [
        "what activity, idea, or procedure is described",
        "according to the ncert passage",
        "according to the passage",
        "what is described in the passage",
        "what activity is described"
    ]
    return any(x in q for x in patterns)

def make_question_draft(original_q, answer, evidence):
    """
    Conservative wording suggestions only.
    No external model is used; uncertain cases remain flagged for review.
    """
    q = clean_text(original_q)
    if not generic_question(q):
        return q, "RETAIN_EXISTING_REVIEW_WORDING"

    text = clean_text(evidence or answer)
    low = text.lower()

    if re.search(r"\bdraw\b", low) and re.search(r"\bfamily\b", low):
        return (
            "What are students asked to draw about their family, "
            "and what should they write alongside the drawing?",
            "RULE_BASED_DRAFT_REVIEW_REQUIRED"
        )
    if re.search(r"\bmatch\b", low) and re.search(r"\bpictures?\b", low):
        return (
            "What should students match the pictures to?",
            "RULE_BASED_DRAFT_REVIEW_REQUIRED"
        )
    if re.search(r"\bkick(?:ing)?\b", low) and re.search(r"\breceiv(?:e|ing)\b", low):
        return (
            "What do kicking and receiving mean in the context of playing football?",
            "RULE_BASED_DRAFT_REVIEW_REQUIRED"
        )
    if re.search(r"\bwrite\b", low) and re.search(r"\bname(?:s)?\b", low):
        return (
            "What names or information are students asked to write down?",
            "RULE_BASED_DRAFT_REVIEW_REQUIRED"
        )

    # Don't manufacture a content-specific question without a reliable rule.
    return "", "MANUAL_QUESTION_DRAFT_NEEDED"

def make_answer_draft(expected, evidence, similarity):
    """
    Only propose a cleaned extractive answer when evidence overlap is strong.
    This is not a correctness verification.
    """
    expected = clean_text(expected)
    evidence = clean_text(evidence)
    try:
        score = float(similarity)
    except (ValueError, TypeError):
        score = 0.0

    if not expected or expected.lower() in {"nan", "none"}:
        return "", "MISSING_EXPECTED_ANSWER"

    if score < 0.85 or not evidence:
        return "", "ANSWER_NEEDS_SOURCE_REVIEW"

    # Keep the answer concise while preserving its wording.
    answer = first_sentences(expected, limit=3)
    if len(answer) > 550:
        answer = answer[:547].rsplit(" ", 1)[0] + "..."
    return answer, "EXTRACTIVE_DRAFT_REVIEW_REQUIRED"

rows = []
for _, r in df.iterrows():
    qid = str(r.get("question_id", ""))
    source = str(r.get("source_file", "") or "").replace("\\", "/").lower()
    gold = parse_pages(r.get("gold_pages", ""))
    evidence = clean_text(r.get("evidence_excerpt", ""))
    expected = clean_text(r.get("expected_answer", ""))

    matched = pages[pages["source_key"] == source]
    page_texts = []
    for _, pg in matched[matched["page_num"].isin(gold)].iterrows():
        page_texts.append(
            f"PDF page {int(pg['page_num'])}: "
            f"{clean_text(pg.get('clean_text', ''))}"
        )

    source_text = "\n\n".join(page_texts)
    sim = r.get("best_evidence_match_score", 0)
    proposed_q, q_status = make_question_draft(q, expected, evidence) if (
        q := r.get("question", "")
    ) is not None else ("", "MANUAL_QUESTION_DRAFT_NEEDED")

    proposed_a, a_status = make_answer_draft(expected, evidence, sim)

    suggested_pdf_page = r.get("best_evidence_match_pdf_page", "")
    suggested_pages = (
        f"[{int(suggested_pdf_page)}]"
        if pd.notna(suggested_pdf_page) else ""
    )

    flags = str(r.get("review_flags", ""))
    missing_pages = not gold
    low_similarity = False
    try:
        low_similarity = float(sim) < 0.85
    except (ValueError, TypeError):
        low_similarity = True

    if missing_pages:
        page_status = "SUGGESTED_PAGE_REQUIRES_CONFIRMATION"
    elif len(page_texts):
        page_status = "PDF_PAGE_TEXT_ATTACHED_PRINTED_PAGE_UNVERIFIED"
    else:
        page_status = "SOURCE_PAGE_TEXT_NOT_FOUND"

    rows.append({
        "question_id": qid,
        "original_question": q,
        "original_expected_answer": expected,
        "original_evidence_excerpt": evidence,
        "source_file": source,
        "original_gold_pages": str(r.get("gold_pages", "")),
        "suggested_pdf_pages_NOT_VERIFIED": suggested_pages,
        "source_page_status": page_status,
        "source_text_at_original_gold_pages": source_text,
        "evidence_similarity_diagnostic_only": sim,
        "proposed_question_draft": proposed_q,
        "question_draft_status": q_status,
        "proposed_answer_draft": proposed_a,
        "answer_draft_status": a_status,
        "needs_page_repair": missing_pages,
        "low_evidence_similarity": low_similarity,
        "review_flags": flags,
        "review_question_clear": "",
        "review_answer_correct": "",
        "review_evidence_verified": "",
        "final_question": "",
        "final_reference_answer": "",
        "verified_pdf_pages": "",
        "verified_printed_pages": "",
        "reviewer_notes": "",
        "final_status": "NOT_HUMAN_VERIFIED"
    })

repair = pd.DataFrame(rows)

# Include a focused priority sheet. All records remain in the full sheet.
priority = repair[
    repair["needs_page_repair"] |
    repair["low_evidence_similarity"] |
    repair["question_draft_status"].eq("MANUAL_QUESTION_DRAFT_NEEDED") |
    repair["answer_draft_status"].eq("ANSWER_NEEDS_SOURCE_REVIEW")
].copy()

outdir = FOUNDATION / (
    "benchmark_repair_v1_" + datetime.now().strftime("%Y%m%d_%H%M%S")
)
outdir.mkdir(parents=True, exist_ok=False)
out = outdir / "NCERT_RAG_Benchmark_Repair_Candidate.xlsx"

with pd.ExcelWriter(out, engine="openpyxl") as writer:
    pd.DataFrame({"Instructions": [
        "This is a repair candidate, not a verified gold benchmark.",
        "Original question, expected answer, evidence, source and page fields are preserved.",
        "Question drafts use conservative rules; they may be incomplete or unsuitable.",
        "Answer drafts are extractive and must be checked against source evidence.",
        "Suggested PDF pages are not verified printed textbook page numbers.",
        "Do not use this workbook for final research metrics until gold evidence is reviewed.",
        "Fill final_question and final_reference_answer only after review.",
        "Set final_status to VERIFIED only after the source evidence has been checked.",
        f"Original benchmark SHA256: {hashlib.sha256(BENCH.read_bytes()).hexdigest()}",
        "Original benchmark and retrieval indexes were not modified."
    ]}).to_excel(writer, sheet_name="READ_ME", index=False)
    repair.to_excel(writer, sheet_name="All 77 Repair Drafts", index=False)
    priority.to_excel(writer, sheet_name="Priority Review", index=False)

repair.to_csv(outdir / "all_77_repair_drafts.csv", index=False)
priority.to_csv(outdir / "priority_review.csv", index=False)

print("REPAIR CANDIDATE CREATED")
print("Folder:", outdir)
print("Workbook:", out)
print("All questions:", len(repair))
print("Priority review rows:", len(priority))
print("Question drafts generated:", int(
    repair["question_draft_status"].str.contains("DRAFT|RETAIN", na=False).sum()
))
print("Question drafts still needed:", int(
    repair["question_draft_status"].eq("MANUAL_QUESTION_DRAFT_NEEDED").sum()
))
print("Answer drafts needing source review:", int(
    repair["answer_draft_status"].eq("ANSWER_NEEDS_SOURCE_REVIEW").sum()
))
print("Missing gold pages:", int(repair["needs_page_repair"].sum()))
print("Original files modified: No")
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle, Polygon
import json
import math

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
OUT_DIR = ROOT / "04_CPM" / "onboarding_pilot_v1"
IMG_DIR = OUT_DIR / "stimuli"
IMG_DIR.mkdir(parents=True, exist_ok=True)

with open(OUT_DIR / "matrix_reasoning_question_bank_v1.json",
          "r", encoding="utf-8") as f:
    questions = json.load(f)

def draw_shape(ax, name, x=0, y=0, size=0.28):
    name = name.lower()
    if "circle" in name:
        ax.add_patch(Circle((x, y), size, fill=False, linewidth=2))
    elif "square" in name:
        ax.add_patch(Rectangle((x-size, y-size), 2*size, 2*size,
                               fill=False, linewidth=2))
    elif "triangle" in name:
        pts = [(x, y+size), (x-size, y-size), (x+size, y-size)]
        ax.add_patch(Polygon(pts, closed=True, fill=False, linewidth=2))
    elif "pentagon" in name:
        pts = [
            (x + size*math.cos(math.pi/2 + k*2*math.pi/5),
             y + size*math.sin(math.pi/2 + k*2*math.pi/5))
            for k in range(5)
        ]
        ax.add_patch(Polygon(pts, closed=True, fill=False, linewidth=2))
    elif "arrow" in name:
        directions = {
            "up_arrow": (0, 1), "right_arrow": (1, 0),
            "down_arrow": (0, -1), "left_arrow": (-1, 0)
        }
        dx, dy = directions.get(name, (0, 1))
        ax.arrow(x-dx*size*0.6, y-dy*size*0.6,
                 dx*size*1.2, dy*size*1.2,
                 head_width=size*0.65, head_length=size*0.6,
                 length_includes_head=True, linewidth=2)

def draw_token(ax, token):
    token = token.lower()
    if token == "one_dot":
        ax.plot([0], [0], "o", markersize=7)
    elif token == "two_dots":
        ax.plot([-0.12, 0.12], [0, 0], "o", markersize=7)
    elif token == "three_dots":
        ax.plot([-0.16, 0, 0.16], [0, 0, 0], "o", markersize=7)
    elif token == "four_dots":
        ax.plot([-0.12, 0.12, -0.12, 0.12],
                [0.12, 0.12, -0.12, -0.12], "o", markersize=7)
    elif token == "seven_dots":
        xs = [-0.18, 0, 0.18, -0.18, 0, 0.18, 0]
        ys = [0.18, 0.18, 0.18, 0, 0, 0, -0.18]
        ax.plot(xs, ys, "o", markersize=7)
    else:
        draw_shape(ax, token)

def render_question(q):
    pattern = q.get("pattern", [])
    fig, ax = plt.subplots(figsize=(8, 2.2))
    ax.set_xlim(-0.6, max(2, len(pattern))*1.05)
    ax.set_ylim(-0.8, 0.8)
    ax.axis("off")
    ax.set_title(f'{q["item_id"]} · {q["difficulty"]}',
                 fontsize=12, loc="left")

    for i, token in enumerate(pattern):
        sub = fig.add_axes([
            0.08 + i*0.82/max(1, len(pattern)),
            0.28, 0.07, 0.36
        ])
        sub.set_xlim(-0.6, 0.6)
        sub.set_ylim(-0.6, 0.6)
        sub.axis("off")
        draw_token(sub, token)

    output = IMG_DIR / f'{q["item_id"]}_prototype.png'
    fig.savefig(output, dpi=140, bbox_inches="tight")
    plt.close(fig)
    return output

created = []
for q in questions:
    try:
        path = render_question(q)
        q["stimulus_file"] = str(path)
        created.append(path.name)
    except Exception as e:
        print(f'Could not render {q["item_id"]}: {e}')

with open(OUT_DIR / "matrix_reasoning_question_bank_v1.json",
          "w", encoding="utf-8") as f:
    json.dump(questions, f, indent=2, ensure_ascii=False)

print("VISUAL PROTOTYPES RENDERED")
print(f"Rendered: {len(created)} / {len(questions)}")
print(f"Output folder: {IMG_DIR}")
print("These are pattern previews, not final test items.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

# Install only the API SDK and embedding-model package.
%pip -q install google-genai sentence-transformers

from pathlib import Path
import json
import pickle
import ast
import getpass
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from google import genai
from google.genai import types

# --------------------------------------------------
# 1. Load existing artifacts. Do not rebuild indexes.
# --------------------------------------------------
ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
RAG_DIR = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

chunks = pd.read_csv(RAG_DIR / "english_only_chunks.csv")

with open(RAG_DIR / "indexes/bm25/bm25_index.pkl", "rb") as f:
    bm25_index = pickle.load(f)

embeddings = np.load(
    RAG_DIR / "indexes/embeddings/chunk_embeddings.npy"
)

assert len(chunks) == len(embeddings) == len(bm25_index.doc_freqs)
assert embeddings.shape[1] == 384

# --------------------------------------------------
# 2. Load the same embedding model used by the index.
# --------------------------------------------------
print("Loading embedding model...")
embedder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# --------------------------------------------------
# 3. Retrieve candidates with BM25 + dense search.
# --------------------------------------------------
def retrieve_hybrid(question, top_k=5, candidate_k=20, rrf_k=60):
    # BM25 ranking: use the same lowercase whitespace tokenizer.
    query_tokens = question.lower().split()
    bm25_scores = np.asarray(bm25_index.get_scores(query_tokens))
    bm25_order = np.argsort(-bm25_scores)[:candidate_k]

    # Dense ranking: normalized query + existing normalized embeddings.
    query_vec = embedder.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")[0]

    dense_scores = embeddings @ query_vec
    dense_order = np.argsort(-dense_scores)[:candidate_k]

    # Reciprocal Rank Fusion, using 1-based ranks.
    rrf_scores = {}
    for ranking in (bm25_order, dense_order):
        for rank, idx in enumerate(ranking, start=1):
            idx = int(idx)
            rrf_scores[idx] = rrf_scores.get(idx, 0.0) + (
                1.0 / (rrf_k + rank)
            )

    ranked = sorted(
        rrf_scores,
        key=lambda idx: rrf_scores[idx],
        reverse=True
    )[:top_k]

    results = []
    for idx in ranked:
        row = chunks.iloc[idx]
        results.append({
            "chunk_id": str(row["chunk_id"]),
            "source_file": str(row["source_file"]),
            "book": str(row["book"]),
            "class": str(row["class"]),
            "pages": f'{row["first_page"]}-{row["last_page"]}',
            "text": str(row["text"])[:2500],
            "rrf_score": round(rrf_scores[idx], 6)
        })

    return results

# --------------------------------------------------
# 4. Set up the API key privately.
# --------------------------------------------------
api_key = getpass.getpass("Paste your Gemini API key (input hidden): ").strip()

if not api_key:
    raise ValueError("No API key entered. Create one in Google AI Studio first.")

client = genai.Client(api_key=api_key)

# --------------------------------------------------
# 5. Ask the tutor a test question.
# --------------------------------------------------
question = "Why do plants need sunlight?"
grade_level = "Class 3"
language = "English"

retrieved = retrieve_hybrid(question, top_k=5)

evidence_blocks = []
for i, item in enumerate(retrieved, start=1):
    evidence_blocks.append(
        f"[EVIDENCE {i}]\n"
        f"Book: {item['book']}\n"
        f"Class: {item['class']}\n"
        f"Source: {item['source_file']}\n"
        f"PDF page range: {item['pages']}\n"
        f"Chunk ID: {item['chunk_id']}\n"
        f"Text:\n{item['text']}"
    )

evidence_text = "\n\n".join(evidence_blocks)

system_instruction = """
You are an NCERT-grounded tutor for primary-school students.

Rules:
1. Answer using ONLY the supplied textbook evidence.
2. Do not use outside knowledge to fill gaps.
3. If the evidence does not directly support an answer, clearly say:
   "I couldn't find enough information in the retrieved NCERT passages
   to answer that confidently."
4. Never invent textbook quotations, book titles, pages, or facts.
5. Explain in simple language appropriate to the supplied class level.
6. Cite supporting evidence IDs such as [EVIDENCE 1].
7. Distinguish an answer supported by evidence from an inference.
"""

prompt = f"""
Student class: {grade_level}
Response language: {language}

Student question:
{question}

Retrieved NCERT evidence:
{evidence_text}

Give a concise answer suitable for the student. Cite evidence IDs.
If the evidence is insufficient, abstain rather than guessing.
"""

print("\nGenerating answer...")
response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents=prompt,
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        temperature=0.2,
        max_output_tokens=300
    )
)

print("\n" + "=" * 65)
print("NCERT TUTOR — FIRST END-TO-END TEST")
print("=" * 65)
print("\nQUESTION:", question)
print("\nANSWER:\n", response.text or "(No text returned)")

print("\nRETRIEVED SOURCES:")
for i, item in enumerate(retrieved, start=1):
    print(
        f"\n[{i}] {item['book']} | {item['source_file']}"
        f" | PDF pages {item['pages']}"
    )
    print("Chunk:", item["chunk_id"])
    print("Preview:", item["text"][:350].replace("\n", " "))

print("\nTest complete. No index or benchmark files were modified.")
client.close()
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pickle
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
RAG_DIR = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

chunks = pd.read_csv(RAG_DIR / "english_only_chunks.csv")

with open(RAG_DIR / "indexes/bm25/bm25_index.pkl", "rb") as f:
    bm25_index = pickle.load(f)

embeddings = np.load(
    RAG_DIR / "indexes/embeddings/chunk_embeddings.npy"
)

embedder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

def inspect_retrieval(question, top_k=5, candidate_k=20, rrf_k=60):
    tokens = question.lower().split()
    bm25_scores = np.asarray(bm25_index.get_scores(tokens))
    bm25_order = np.argsort(-bm25_scores)[:candidate_k]

    q_vec = embedder.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")[0]

    dense_scores = embeddings @ q_vec
    dense_order = np.argsort(-dense_scores)[:candidate_k]

    fused = {}
    for order in (bm25_order, dense_order):
        for rank, idx in enumerate(order, start=1):
            idx = int(idx)
            fused[idx] = fused.get(idx, 0) + 1 / (rrf_k + rank)

    ranked = sorted(fused, key=fused.get, reverse=True)[:top_k]

    print("\n" + "=" * 80)
    print("QUESTION:", question)
    print("=" * 80)

    for rank, idx in enumerate(ranked, start=1):
        row = chunks.iloc[idx]
        print(f"\n[{rank}] {row['class']} | {row['book']}")
        print(f"Source: {row['source_file']}")
        print(f"Chunk: {row['chunk_id']}")
        print(f"RRF score: {fused[idx]:.5f}")
        print(str(row["text"])[:650].replace("\n", " "))

tests = [
    "Why do plants need sunlight?",
    "What do plants need to grow?",
    "Why are shadows shorter at noon?"
]

for question in tests:
    inspect_retrieval(question)

print("\nDIAGNOSTIC COMPLETE — no files or indexes modified.")
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import re

ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
RAG_DIR = ROOT / "05_EXPERIMENTS" / "retrieval_evaluation" / "english_only_v1"
chunks = pd.read_csv(RAG_DIR / "english_only_chunks.csv")

# Search the stored text directly, independently of the retriever ranking.
checks = {
    "Sunlight / photosynthesis": [
        r"\bsunlight\b", r"\bsun light\b", r"\bphotosynthesis\b",
        r"\bsun\b", r"\blight\b"
    ],
    "Plant growth requirements": [
        r"\bplants? need\b", r"\bneed water\b", r"\bgrow(?:th|ing)?\b",
        r"\bwater\b", r"\bair\b", r"\bsoil\b", r"\bseeds?\b"
    ],
    "Shadow explanation": [
        r"\bshadows?\b", r"\bnoon\b", r"\bshort(?:er)?\b",
        r"\bsun\b", r"\bdirection\b"
    ]
}

texts = chunks["text"].fillna("").astype(str)

for label, patterns in checks.items():
    pattern = re.compile("|".join(patterns), flags=re.IGNORECASE)
    matching = chunks[texts.map(lambda t: bool(pattern.search(t)))].copy()

    # Rank by number of distinct query terms found; this is a diagnostic,
    # not a substitute for semantic relevance or evidence verification.
    def term_count(text):
        return sum(bool(re.search(p, text, re.IGNORECASE)) for p in patterns)

    matching["term_matches"] = matching["text"].map(term_count)
    matching = matching.sort_values(
        ["term_matches", "source_file", "chunk_index"],
        ascending=[False, True, True]
    )

    print("\n" + "=" * 80)
    print(label)
    print("Matching chunks:", len(matching))
    print("=" * 80)

    for _, row in matching.head(8).iterrows():
        print(f"\n{row['chunk_id']}")
        print(f"Class: {row['class']} | Book: {row['book']}")
        print(f"Source: {row['source_file']}")
        print("Text:", row["text"][:900].replace("\n", " "))

print("\nCorpus search complete. No files or indexes modified.")
```

```python

# ============================================================
# NCERT RAG — BM25 vs DENSE vs HYBRID RRF
# Corrected paths | Read-only diagnostic
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os
import pickle
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS", "retrieval_evaluation", "english_only_v1"
)

CHUNKS_PATH = os.path.join(BASE, "english_only_chunks.csv")
BM25_PATH = os.path.join(BASE, "indexes", "bm25", "bm25_index.pkl")
TOKENIZED_PATH = os.path.join(
    BASE, "indexes", "bm25", "tokenized_documents.pkl"
)
EMBEDDINGS_PATH = os.path.join(
    BASE, "indexes", "embeddings", "chunk_embeddings.npy"
)

for path in [CHUNKS_PATH, BM25_PATH, TOKENIZED_PATH, EMBEDDINGS_PATH]:
    assert os.path.isfile(path), f"Missing required artifact: {path}"

chunks = pd.read_csv(CHUNKS_PATH).fillna("")

with open(BM25_PATH, "rb") as f:
    bm25 = pickle.load(f)

with open(TOKENIZED_PATH, "rb") as f:
    tokenized_docs = pickle.load(f)

embeddings = np.load(EMBEDDINGS_PATH).astype(np.float32)

print("Chunks:", len(chunks))
print("BM25 documents:", len(tokenized_docs))
print("Embedding matrix:", embeddings.shape)

assert len(chunks) == len(tokenized_docs) == len(embeddings), (
    "Artifact row counts differ. Stop before comparing retrieval methods."
)

# Load the same embedding model used for the existing index.
encoder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

norms = np.linalg.norm(embeddings, axis=1)
print("Document vector norms:",
      round(float(norms.min()), 4),
      "to",
      round(float(norms.max()), 4))

QUESTIONS = [
    "Why do plants need sunlight?",
    "What do plants need to grow?",
    "Why are shadows shorter at noon?"
]

TOP_K = 5
CANDIDATE_K = 20
RRF_K = 60

def bm25_search(query, k=TOP_K):
    scores = np.asarray(
        bm25.get_scores(query.lower().split()),
        dtype=np.float32
    )
    ids = np.argsort(-scores)[:k]
    return [(int(i), float(scores[i])) for i in ids]

def dense_search(query, k=TOP_K):
    q = encoder.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    )[0].astype(np.float32)

    scores = embeddings @ q
    ids = np.argsort(-scores)[:k]
    return [(int(i), float(scores[i])) for i in ids]

def hybrid_search(query, k=TOP_K):
    # Reciprocal Rank Fusion over the top candidates from each method.
    bm = bm25_search(query, CANDIDATE_K)
    dn = dense_search(query, CANDIDATE_K)
    fused = {}

    for rank, (idx, _) in enumerate(bm, start=1):
        fused[idx] = fused.get(idx, 0.0) + 1 / (RRF_K + rank)

    for rank, (idx, _) in enumerate(dn, start=1):
        fused[idx] = fused.get(idx, 0.0) + 1 / (RRF_K + rank)

    ranked = sorted(fused.items(), key=lambda x: x[1], reverse=True)[:k]
    return [(int(idx), float(score)) for idx, score in ranked]

def display_results(question, method, results):
    print("\n" + "=" * 100)
    print(f"QUESTION: {question} | METHOD: {method}")
    print("=" * 100)

    for rank, (idx, score) in enumerate(results, start=1):
        row = chunks.iloc[idx]
        print(
            f"\n#{rank} score={score:.5f}"
            f" | chunk_id={row.get('chunk_id', 'N/A')}"
        )
        print(
            f"Class: {row.get('class', 'N/A')}"
            f" | Subject: {row.get('subject', 'N/A')}"
            f" | Book: {row.get('book', 'N/A')}"
        )
        print(f"Source: {row.get('source_file', 'N/A')}")
        print(
            f"Pages: {row.get('pages', 'N/A')}"
            f" | PDF pages: {row.get('first_page', 'N/A')}"
            f"-{row.get('last_page', 'N/A')}"
        )

        passage = str(row.get("text", "")).replace("\n", " ")
        print("Passage:", passage[:550],
              "..." if len(passage) > 550 else "")

for question in QUESTIONS:
    display_results(question, "BM25", bm25_search(question))
    display_results(question, "DENSE", dense_search(question))
    display_results(question, "HYBRID RRF", hybrid_search(question))

print("\nDiagnostic complete. No files were modified.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import re
import pandas as pd

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
CHUNKS_PATH = os.path.join(
    ROOT, "05_EXPERIMENTS", "retrieval_evaluation",
    "english_only_v1", "english_only_chunks.csv"
)

chunks = pd.read_csv(CHUNKS_PATH).fillna("")

# Search the complete corpus, not just the retrieval top five.
# These are diagnostic keyword matches, NOT relevance labels.
patterns = {
    "sunlight": r"\bsunlight\b|\bsun light\b",
    "sun rays": r"\bsun\s*rays?\b",
    "photosynthesis": r"\bphotosynthesis\b|\bphotosynthetic\b",
    "food making": r"\bmake[s]?\s+(their\s+)?own\s+food\b|\bmake food\b",
    "plant needs": r"\bplants?\s+need\b|\bneeds?\s+of\s+plants?\b",
    "water and light": r"\bwater\b.{0,100}\b(light|sun|sunlight)\b"
                       r"|\b(light|sun|sunlight)\b.{0,100}\bwater\b",
}

rows = []

for idx, row in chunks.iterrows():
    text = str(row.get("text", ""))
    matched = [
        name for name, pattern in patterns.items()
        if re.search(pattern, text, flags=re.IGNORECASE)
    ]

    if matched:
        rows.append({
            "row_idx": idx,
            "chunk_id": row.get("chunk_id", ""),
            "class": row.get("class", ""),
            "subject": row.get("subject", ""),
            "book": row.get("book", ""),
            "source_file": row.get("source_file", ""),
            "pages": row.get("pages", ""),
            "matched_terms": ", ".join(matched),
            "match_count": len(matched),
            "text": text,
        })

results = pd.DataFrame(rows)

print("Corpus chunks:", len(chunks))
print("Chunks matching at least one diagnostic pattern:", len(results))

if results.empty:
    print("No matches found for these patterns.")
else:
    # Prioritize passages mentioning several relevant concepts.
    results = results.sort_values(
        ["match_count", "class", "book"],
        ascending=[False, True, True]
    )

    for _, row in results.head(25).iterrows():
        print("\n" + "=" * 90)
        print(
            f"ID: {row['chunk_id']} | Class: {row['class']} "
            f"| Subject: {row['subject']}"
        )
        print(f"Book: {row['book']} | Pages: {row['pages']}")
        print(f"Source: {row['source_file']}")
        print(f"Matched terms: {row['matched_terms']}")
        print("Passage:\n", row["text"])

print("\nRead-only corpus inspection complete.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import re
import fitz  # PyMuPDF

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
SOURCE_DIRS = [
    os.path.join(ROOT, "01_SOURCE_BOOKS"),
    os.path.join(ROOT, "01_SOURCE_BOOKS_V2"),
]

terms = [
    r"sunlight",
    r"sun light",
    r"photosynthesis",
    r"make their own food",
    r"plants need",
    r"food.*plants",
    r"plant.*sun",
]

pattern = re.compile("|".join(terms), re.IGNORECASE)
seen = set()
matches = []

for source_dir in SOURCE_DIRS:
    if not os.path.isdir(source_dir):
        continue

    for folder, _, filenames in os.walk(source_dir):
        for filename in filenames:
            if not filename.lower().endswith(".pdf"):
                continue

            path = os.path.join(folder, filename)

            # Avoid scanning the same physical path twice.
            if path in seen:
                continue
            seen.add(path)

            try:
                pdf = fitz.open(path)

                for page_num, page in enumerate(pdf):
                    text = page.get_text("text")
                    if not text.strip():
                        continue

                    if pattern.search(text):
                        excerpts = []
                        for match in pattern.finditer(text):
                            start = max(0, match.start() - 220)
                            end = min(len(text), match.end() + 350)
                            excerpts.append(
                                re.sub(r"\s+", " ", text[start:end]).strip()
                            )
                            if len(excerpts) >= 3:
                                break

                        matches.append({
                            "book": filename,
                            "path": path,
                            "pdf_page": page_num + 1,
                            "excerpts": excerpts,
                        })

                pdf.close()

            except Exception as exc:
                print("Could not read:", path, "|", str(exc)[:150])

print("\nMatching PDF pages:", len(matches))

for item in matches:
    print("\n" + "=" * 90)
    print("BOOK:", item["book"])
    print("PDF page:", item["pdf_page"])
    print("PATH:", item["path"])
    for excerpt in item["excerpts"]:
        print("-", excerpt)

print("\nSource-PDF inspection complete. No files modified.")
```

```python

# ============================================================
# NCERT AI TUTOR — DYNAMIC GEMINI MODEL FALLBACK
# Automatically discovers models and falls back on failures.
# Reuses the existing corpus, indexes, client, and retrieval.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import time
from google.genai import types

# Keep this list of successful models across questions in this
# Colab session. Start with the previously configured model.
if "WORKING_MODEL" not in globals():
    WORKING_MODEL = MODEL_NAME if "MODEL_NAME" in globals() else None

# ---------- 1. Discover models available to this API key ----------
def discover_gemini_models():
    discovered = []

    try:
        for model in client.models.list():
            name = getattr(model, "name", None)
            if not name:
                continue

            actions = getattr(model, "supported_actions", None) or []
            supports_generation = any(
                "generatecontent" in str(action).lower()
                for action in actions
            )

            # Exclude embedding-only and unrelated model families.
            if "gemini" not in name.lower():
                continue
            if "embed" in name.lower():
                continue

            # If the API provides supported actions, honor them.
            # Some SDK responses may omit this field.
            if actions and not supports_generation:
                continue

            discovered.append(name)

    except Exception as exc:
        print(
            "Model discovery failed:",
            type(exc).__name__,
            str(exc)[:250]
        )

    # Prefer Flash models for fast, cost-conscious tutoring.
    def priority(name):
        n = name.lower()
        if "flash" in n and "preview" not in n and "experimental" not in n:
            return (0, n)
        if "flash" in n:
            return (1, n)
        if "pro" in n and "preview" not in n:
            return (2, n)
        if "pro" in n:
            return (3, n)
        return (4, n)

    discovered = sorted(set(discovered), key=priority)

    # Put the last successful model first if it is still available.
    if WORKING_MODEL in discovered:
        discovered.remove(WORKING_MODEL)
        discovered.insert(0, WORKING_MODEL)

    # Retain the previously working configured model as a final
    # fallback if model discovery is unavailable.
    if not discovered and WORKING_MODEL:
        discovered = [WORKING_MODEL]

    return discovered


MODEL_CANDIDATES = discover_gemini_models()

if not MODEL_CANDIDATES:
    raise RuntimeError(
        "No Gemini generation models were discovered. "
        "Check API-key permissions and model availability."
    )

print(f"Discovered {len(MODEL_CANDIDATES)} candidate model(s):")
for model in MODEL_CANDIDATES:
    print(" •", model)

# ---------- 2. Grounded answer with model fallback ----------
def answer_from_evidence_with_fallback(question, evidence):
    global WORKING_MODEL, MODEL_NAME

    evidence_text = "\n\n".join(
        f"[EVIDENCE {i}]\n"
        f"Class: {item['class']}\n"
        f"Subject: {item['subject']}\n"
        f"Book: {item['book']}\n"
        f"Page metadata: {item['pages']}\n"
        f"Passage: {item['text']}"
        for i, item in enumerate(evidence, start=1)
    )

    prompt = f"""
You are an NCERT-grounded tutor for primary-school students.

QUESTION:
{question}

RETRIEVED NCERT PASSAGES:
{evidence_text}

Rules:
1. Use only information supported by these passages.
2. Never invent facts or fill gaps using outside knowledge.
3. Explain ideas in simple, child-friendly English.
4. If evidence is insufficient, clearly say so.
5. Cite claims with [EVIDENCE 1], [EVIDENCE 2], etc.
6. Cite only passages that genuinely support the claim.
7. Do not present a guess or unsupported inference as fact.
"""

    failures = []

    for model_name in MODEL_CANDIDATES:
        # One attempt per model per question. A failed model is
        # skipped so the next candidate gets a chance.
        try:
            print(f"\nTrying model: {model_name}")

            response = client.models.generate_content(
                model=model_name,
                contents=prompt,
                config=types.GenerateContentConfig(
                    temperature=0.1
                )
            )

            answer = response.text

            if not answer or not answer.strip():
                raise RuntimeError("Model returned an empty response.")

            # Remember the successful model for future questions.
            WORKING_MODEL = model_name
            MODEL_NAME = model_name

            # Move it to the front for the remaining questions.
            if model_name in MODEL_CANDIDATES:
                MODEL_CANDIDATES.remove(model_name)
            MODEL_CANDIDATES.insert(0, model_name)

            print(f"SUCCESS — using {model_name}")
            return answer

        except Exception as exc:
            message = str(exc)
            failures.append((model_name, type(exc).__name__, message[:300]))

            print(
                f"FAILED — {type(exc).__name__}: {message[:220]}"
            )

            # Short pause before trying another model.
            time.sleep(1)

    print("\nAll candidate models failed:")
    for name, error_type, message in failures:
        print(f"- {name}: {error_type}: {message}")

    return (
        "I couldn't generate an answer right now because all available "
        "model attempts failed. Please try again shortly."
    )


# ---------- 3. Rerun the three grounding tests ----------
test_questions = [
    {
        "type": "EXPECTED DIRECT SUPPORT",
        "question": "When are shadows shorter during the day?"
    },
    {
        "type": "PARTIAL / RELATED SUPPORT",
        "question": "What helps plants grow?"
    },
    {
        "type": "POSSIBLY INSUFFICIENT EVIDENCE",
        "question": "Why do plants need sunlight?"
    }
]

for case_no, case in enumerate(test_questions, start=1):
    question = case["question"]
    evidence = retrieve_hybrid(question)

    print("\n" + "=" * 85)
    print(f"CASE {case_no}: {case['type']}")
    print("QUESTION:", question)
    print("=" * 85)

    print("\n--- RETRIEVED EVIDENCE ---")
    for i, item in enumerate(evidence, start=1):
        print(
            f"\n[EVIDENCE {i}] {item['chunk_id']} | "
            f"Class {item['class']} | {item['book']} | "
            f"Pages: {item['pages']}"
        )
        print(item["text"][:900])

    print("\n--- GROUNDED ANSWER ---")
    answer = answer_from_evidence_with_fallback(question, evidence)
    print(answer)

print("\n" + "=" * 85)
print("TEST COMPLETE")
print("Preferred successful model for this session:", WORKING_MODEL)
print("No corpus, retrieval index, or benchmark files were modified.")
```

```python

# ============================================================
# NCERT TUTOR — RETRIEVAL QUALITY EXPERIMENT V1
# Multi-query candidate retrieval + cross-encoder reranking
# Compares baseline top-5 against improved top-5.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import re
import numpy as np
import pandas as pd
from collections import defaultdict
from sentence_transformers import CrossEncoder

# ---------- 1. Validate the existing retrieval setup ----------
assert len(chunks) == len(embeddings) == faiss_index.ntotal
assert len(chunks) == len(tokenized_docs)

print("Existing retrieval artifacts validated:", len(chunks), "chunks")

# ---------- 2. Load the reranker ----------
# This is separate from the embedding model.
# It scores a QUESTION + PASSAGE pair for relevance.
RERANKER_NAME = "cross-encoder/ms-marco-MiniLM-L-6-v2"

reranker = CrossEncoder(RERANKER_NAME)

print("Reranker loaded:", RERANKER_NAME)

# ---------- 3. Generate lightweight query variants ----------
STOPWORDS = {
    "a", "an", "the", "is", "are", "was", "were", "do", "does",
    "did", "what", "why", "when", "where", "who", "how", "which",
    "of", "in", "on", "at", "to", "for", "from", "with", "and",
    "or", "during", "can", "could", "would", "should", "their",
    "they", "it", "its", "we", "you", "your", "me", "my"
}

def make_query_variants(question):
    words = re.findall(r"[a-zA-Z0-9]+", question.lower())
    keywords = [w for w in words if w not in STOPWORDS]

    variants = [question.strip()]

    if keywords:
        variants.append(" ".join(keywords))

    # A short noun/verb-focused query often helps lexical retrieval.
    # Keep the original words and order; do not invent facts.
    if len(keywords) >= 2:
        variants.append(" ".join(keywords[:8]))

    return list(dict.fromkeys(variants))


# ---------- 4. Multi-query candidate retrieval ----------
def retrieve_improved(question, candidate_k=40, top_k=5, rrf_k=60):
    variants = make_query_variants(question)
    fused_scores = defaultdict(float)

    for query in variants:
        # BM25 candidates
        bm25_scores = bm25.get_scores(query.lower().split())
        bm25_order = np.argsort(bm25_scores)[::-1][:candidate_k]

        for rank, idx in enumerate(bm25_order, start=1):
            fused_scores[int(idx)] += 1.0 / (rrf_k + rank)

        # Dense candidates
        query_vec = encoder.encode(
            [query],
            convert_to_numpy=True,
            normalize_embeddings=True
        ).astype("float32")

        dense_scores, dense_ids = faiss_index.search(
            query_vec, candidate_k
        )

        for rank, idx in enumerate(dense_ids[0], start=1):
            if idx >= 0:
                fused_scores[int(idx)] += 1.0 / (rrf_k + rank)

    # Keep a larger pool for the reranker.
    candidate_ids = [
        idx for idx, score in sorted(
            fused_scores.items(),
            key=lambda pair: pair[1],
            reverse=True
        )[:50]
    ]

    if not candidate_ids:
        return []

    pairs = [
        (question, str(chunks.iloc[idx]["text"]))
        for idx in candidate_ids
    ]

    rerank_scores = reranker.predict(
        pairs,
        show_progress_bar=False
    )

    ranked = sorted(
        zip(candidate_ids, rerank_scores),
        key=lambda pair: float(pair[1]),
        reverse=True
    )[:top_k]

    results = []

    for idx, score in ranked:
        row = chunks.iloc[idx]
        results.append({
            "index": int(idx),
            "score": float(score),
            "chunk_id": row.get("chunk_id", f"row_{idx}"),
            "class": row.get("class", "Unknown"),
            "subject": row.get("subject", "Unknown"),
            "book": row.get("book", "Unknown"),
            "pages": row.get("pages", ""),
            "text": str(row.get("text", ""))
        })

    return results


# ---------- 5. Strictly grounded answer with model fallback ----------
def answer_grounded_strict(question, evidence):
    from google.genai import types

    evidence_text = "\n\n".join(
        f"[EVIDENCE {i}]\n"
        f"Class: {item['class']}\n"
        f"Subject: {item['subject']}\n"
        f"Book: {item['book']}\n"
        f"Pages: {item['pages']}\n"
        f"Passage: {item['text']}"
        for i, item in enumerate(evidence, start=1)
    )

    prompt = f"""
You are a careful NCERT tutor for primary-school children.

QUESTION:
{question}

EVIDENCE RETRIEVED FROM NCERT:
{evidence_text}

STRICT RULES:
- Answer only with claims supported by the supplied passages.
- Do not use outside knowledge to fill gaps.
- A fact being generally true does not mean these passages support it.
- Do not claim that a passage explains a cause if it only states an
  observation or result.
- If only part of the question is supported, answer that part and
  clearly say what the passages do not explain.
- If no useful evidence supports the answer, say so plainly.
- Use simple, age-appropriate English.
- Cite each substantive claim using [EVIDENCE 1], etc.
- Never fabricate a citation.
"""

    candidates = []
    if globals().get("WORKING_MODEL"):
        candidates.append(WORKING_MODEL)
    candidates.extend(globals().get("MODEL_CANDIDATES", []))

    # Deduplicate while preserving priority.
    candidates = list(dict.fromkeys(candidates))

    failures = []

    for model_name in candidates:
        try:
            print("Generating with:", model_name)

            response = client.models.generate_content(
                model=model_name,
                contents=prompt,
                config=types.GenerateContentConfig(temperature=0.1)
            )

            if not response.text or not response.text.strip():
                raise RuntimeError("Empty response")

            globals()["WORKING_MODEL"] = model_name
            globals()["MODEL_NAME"] = model_name

            return response.text

        except Exception as exc:
            failures.append(
                f"{model_name}: {type(exc).__name__}: {str(exc)[:180]}"
            )
            print("Model failed; trying fallback.")

        # Continue to the next model if this one fails.

    return (
        "I couldn't generate an answer because all available model "
        "attempts failed.\n" + "\n".join(failures)
    )


# ---------- 6. Compare baseline and improved retrieval ----------
test_questions = [
    "When are shadows shorter during the day?",
    "What helps plants grow?",
    "Why do plants need sunlight?"
]

comparison_rows = []

for question in test_questions:
    print("\n" + "=" * 90)
    print("QUESTION:", question)
    print("=" * 90)

    baseline = retrieve_hybrid(question, candidate_k=20, top_k=5)
    improved = retrieve_improved(question, candidate_k=40, top_k=5)

    print("\n--- BASELINE TOP 5 ---")
    for rank, item in enumerate(baseline, start=1):
        print(
            f"{rank}. {item['chunk_id']} | "
            f"{item['book']} | pages {item['pages']}"
        )
        print(item["text"][:350].replace("\n", " "), "\n")

    print("\n--- IMPROVED TOP 5 ---")
    for rank, item in enumerate(improved, start=1):
        print(
            f"{rank}. {item['chunk_id']} | "
            f"{item['book']} | pages {item['pages']} | "
            f"reranker score={item['score']:.3f}"
        )
        print(item["text"][:500].replace("\n", " "), "\n")

    baseline_ids = [x["chunk_id"] for x in baseline]
    improved_ids = [x["chunk_id"] for x in improved]

    overlap = len(set(baseline_ids) & set(improved_ids))

    print(f"Top-5 overlap: {overlap}/5")
    print("\n--- ANSWER USING IMPROVED EVIDENCE ---")

    answer = answer_grounded_strict(question, improved)
    print(answer)

    comparison_rows.append({
        "question": question,
        "baseline_top5": " | ".join(baseline_ids),
        "improved_top5": " | ".join(improved_ids),
        "top5_overlap": overlap,
        "answer": answer
    })

comparison_df = pd.DataFrame(comparison_rows)

print("\n" + "=" * 90)
print("COMPARISON SUMMARY")
display(comparison_df[[
    "question", "top5_overlap", "baseline_top5", "improved_top5"
]])

print("\nExperiment finished. No existing indexes or benchmark files were changed.")
```

```python

# ============================================================
# NCERT BENCHMARK — BASELINE vs IMPROVED RETRIEVAL
# Read-only on the original benchmark and retrieval indexes.
# No Gemini calls. Saves a NEW results CSV only.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os, re, json, ast, time
import numpy as np
import pandas as pd
from datetime import datetime

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS", "retrieval_evaluation", "english_only_v1"
)
BENCHMARK = os.path.join(BASE, "english_only_benchmark.xlsx")

if not os.path.exists(BENCHMARK):
    raise FileNotFoundError(f"Benchmark not found: {BENCHMARK}")

# Reuse current session objects; reload reranker only if necessary.
if "reranker" not in globals():
    from sentence_transformers import CrossEncoder
    reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")

df = pd.read_excel(BENCHMARK)
print("Benchmark rows:", len(df))
print("Columns:", df.columns.tolist())

# ---------- Detect annotation columns conservatively ----------
def find_col(candidates):
    lower_map = {str(c).strip().lower(): c for c in df.columns}
    for name in candidates:
        if name in lower_map:
            return lower_map[name]
    return None

question_col = find_col([
    "question", "question_text", "query", "user_question"
])
source_col = find_col([
    "gold_source_file", "gold_source", "source_file",
    "gold_sources", "expected_source", "source"
])
pages_col = find_col([
    "gold_pages", "gold_page", "expected_pages",
    "source_pages", "page_numbers", "pages"
])
answerability_col = find_col([
    "answerability", "answerable", "answerability_label",
    "is_answerable"
])
status_col = find_col([
    "annotation_status", "verification_status", "review_status",
    "evidence_status"
])

print("\nDetected columns:")
print("Question:", question_col)
print("Gold source:", source_col)
print("Gold pages:", pages_col)
print("Answerability:", answerability_col)
print("Annotation status:", status_col)

if not question_col or not source_col or not pages_col:
    raise ValueError(
        "Could not safely identify question/source/page annotation columns. "
        "Use the printed column list to map the correct fields before scoring."
    )

# ---------- Parse source names and page numbers ----------
def parse_values(value):
    if pd.isna(value):
        return []
    if isinstance(value, (list, tuple, set, np.ndarray)):
        return list(value)

    text = str(value).strip()
    if not text or text.lower() in {"nan", "none", "null", "[]"}:
        return []

    # Handle serialized lists when present.
    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, (list, tuple, set)):
            return list(parsed)
    except Exception:
        pass

    # Handle simple JSON lists.
    try:
        parsed = json.loads(text)
        if isinstance(parsed, list):
            return parsed
    except Exception:
        pass

    # Otherwise treat it as a single source/string.
    return [text]

def norm_source(value):
    value = str(value).strip().lower().replace("\\", "/")
    value = os.path.basename(value)
    value = re.sub(r"\.(pdf|docx?|xlsx?)$", "", value)
    value = re.sub(r"[^a-z0-9]+", "", value)
    return value

def parse_pages(value):
    vals = parse_values(value)
    found = set()
    for val in vals:
        # Avoid extracting arbitrary digits from text labels.
        for number in re.findall(r"\d+", str(val)):
            found.add(int(number))
    return found

def source_matches(gold, retrieved):
    gold_norm = {norm_source(x) for x in parse_values(gold)}
    gold_norm.discard("")
    got = norm_source(retrieved)
    if not gold_norm or not got:
        return False
    return any(g == got or g in got or got in g for g in gold_norm)

def row_is_answerable(row):
    if answerability_col:
        val = str(row[answerability_col]).strip().lower()
        if val in {"false", "no", "unanswerable", "not_answerable", "0"}:
            return False
        if val in {"true", "yes", "answerable", "1"}:
            return True

    # Unverified rows may still be evaluated for exploratory purposes,
    # but must not be described as a verified benchmark.
    return True

# ---------- Evaluate the benchmark ----------
records = []
start = time.time()

for row_num, (_, row) in enumerate(df.iterrows(), start=1):
    question = str(row[question_col]).strip()
    if not question or question.lower() == "nan":
        continue

    gold_source = row[source_col]
    gold_pages = parse_pages(row[pages_col])

    answerable = row_is_answerable(row)
    if not answerable:
        continue

    baseline = retrieve_hybrid(question, candidate_k=20, top_k=5)
    improved = retrieve_improved(question, candidate_k=40, top_k=5)

    def score_results(results):
        source_hit = False
        source_page_hit = False
        retrieved_details = []

        for item in results:
            chunk_row = chunks.iloc[item["index"]]
            retrieved_source = chunk_row.get("source_file", "")
            retrieved_pages = parse_pages(chunk_row.get("pages", ""))

            src_ok = source_matches(gold_source, retrieved_source)
            page_ok = bool(gold_pages & retrieved_pages)

            if src_ok:
                source_hit = True
                if not gold_pages or page_ok:
                    source_page_hit = True

            retrieved_details.append({
                "chunk_id": item.get("chunk_id", ""),
                "source_file": str(retrieved_source),
                "pages": str(chunk_row.get("pages", "")),
                "source_match": src_ok,
                "page_overlap": page_ok
            })

        return source_hit, source_page_hit, retrieved_details

    b_source, b_page, b_details = score_results(baseline)
    i_source, i_page, i_details = score_results(improved)

    status = (
        str(row[status_col]).strip()
        if status_col and pd.notna(row[status_col])
        else "STATUS_NOT_PROVIDED"
    )

    records.append({
        "question": question,
        "gold_source": str(gold_source),
        "gold_pages": ",".join(map(str, sorted(gold_pages))),
        "annotation_status": status,
        "baseline_source_hit_at_5": b_source,
        "baseline_source_page_hit_at_5": b_page,
        "improved_source_hit_at_5": i_source,
        "improved_source_page_hit_at_5": i_page,
        "baseline_chunks": json.dumps(b_details),
        "improved_chunks": json.dumps(i_details)
    })

    if row_num % 10 == 0:
        print(f"Processed {row_num}/{len(df)} benchmark rows...")

results = pd.DataFrame(records)

if results.empty:
    raise ValueError("No eligible answerable benchmark questions were evaluated.")

# ---------- Summary ----------
def report_metric(label, column):
    print(f"{label}: {results[column].mean():.3f} "
          f"({int(results[column].sum())}/{len(results)})")

print("\n" + "=" * 70)
print("RETRIEVAL COMPARISON")
print("=" * 70)
print("Evaluated questions:", len(results))
report_metric("Baseline source hit@5", "baseline_source_hit_at_5")
report_metric("Improved source hit@5", "improved_source_hit_at_5")
report_metric("Baseline source+page hit@5", "baseline_source_page_hit_at_5")
report_metric("Improved source+page hit@5", "improved_source_page_hit_at_5")

for metric in [
    ("baseline_source_hit_at_5", "improved_source_hit_at_5"),
    ("baseline_source_page_hit_at_5", "improved_source_page_hit_at_5")
]:
    old, new = metric
    delta = results[new].mean() - results[old].mean()
    print(f"Change ({new} vs {old}): {delta:+.3f}")

# Show only questions where the result changed.
changed = results[
    (results["baseline_source_hit_at_5"] != results["improved_source_hit_at_5"]) |
    (results["baseline_source_page_hit_at_5"] != results["improved_source_page_hit_at_5"])
]
print("\nQuestions where retrieval outcome changed:", len(changed))
display(changed[[
    "question", "gold_source", "gold_pages",
    "baseline_source_hit_at_5", "improved_source_hit_at_5",
    "baseline_source_page_hit_at_5", "improved_source_page_hit_at_5"
]].head(25))

# ---------- Save separate, timestamped results ----------
OUT_DIR = os.path.join(BASE, "evaluation", "retrieval_comparison_runs")
os.makedirs(OUT_DIR, exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
out_path = os.path.join(
    OUT_DIR, f"baseline_vs_improved_{timestamp}.csv"
)
results.to_csv(out_path, index=False)

print("\nSaved NEW results file:", out_path)
print(f"Elapsed time: {(time.time() - start) / 60:.1f} minutes")
print("Original benchmark and indexes were not modified.")
print(
    "\nIMPORTANT: Treat these as provisional metrics until the gold "
    "source/page annotations have been manually verified."
)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import numpy as np
import re

def retrieve_hybrid(question, candidate_k=20, top_k=5, rrf_k=60):
    # BM25 candidates
    query_tokens = question.lower().split()
    bm25_scores = bm25.get_scores(query_tokens)
    bm25_order = np.argsort(bm25_scores)[::-1][:candidate_k]

    # Dense candidates
    query_embedding = encoder.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    ).astype("float32")

    dense_scores, dense_indices = faiss_index.search(
        query_embedding, candidate_k
    )
    dense_order = dense_indices[0]

    # Reciprocal Rank Fusion
    fused_scores = {}

    for rank, idx in enumerate(bm25_order, start=1):
        fused_scores[int(idx)] = fused_scores.get(int(idx), 0) + (
            1 / (rrf_k + rank)
        )

    for rank, idx in enumerate(dense_order, start=1):
        if idx >= 0:
            fused_scores[int(idx)] = fused_scores.get(int(idx), 0) + (
                1 / (rrf_k + rank)
            )

    ranked_indices = sorted(
        fused_scores,
        key=fused_scores.get,
        reverse=True
    )[:top_k]

    results = chunks.iloc[ranked_indices].copy()
    results["retrieval_score"] = [
        fused_scores[idx] for idx in ranked_indices
    ]

    return results.reset_index(drop=True)


# Sanity-check on a question from the previous experiments.
test_question = "When are shadows shorter during the day?"
test_results = retrieve_hybrid(test_question, candidate_k=20, top_k=5)

print("Question:", test_question)
for rank, row in test_results.iterrows():
    print(f"\nRank {rank + 1}")
    print("Chunk ID:", row["chunk_id"])
    print("Source:", row["source_file"])
    print("Pages:", row["pages"])
    print("Text:", str(row["text"])[:700])
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import json
import ast
from tqdm.auto import tqdm

saved_csv = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/retrieval_comparison_runs/"
    "baseline_vs_improved_20261010_203122.csv"
)

old = pd.read_csv(saved_csv)

exact_matches = 0
overlaps = []
mismatches = []

for idx, row in tqdm(old.iterrows(), total=len(old)):
    previous = row["baseline_chunks"]

    if isinstance(previous, str):
        try:
            previous = json.loads(previous)
        except Exception:
            previous = ast.literal_eval(previous)

    old_ids = [x["chunk_id"] for x in previous]
    current = retrieve_hybrid(
        row["question"], candidate_k=20, top_k=5
    )
    new_ids = current["chunk_id"].tolist()

    overlap = len(set(old_ids) & set(new_ids))
    overlaps.append(overlap)

    if old_ids == new_ids:
        exact_matches += 1
    else:
        mismatches.append({
            "question": row["question"],
            "exact_match": False,
            "overlap": overlap,
            "old_ids": old_ids,
            "new_ids": new_ids,
        })

print("\n" + "=" * 65)
print("BASELINE RECONSTRUCTION VALIDATION")
print("=" * 65)
print("Questions:", len(old))
print(f"Exact top-5 matches: {exact_matches}/{len(old)}")
print(f"Exact match rate: {exact_matches/len(old):.1%}")
print(f"Mean top-5 overlap: {sum(overlaps)/len(overlaps):.2f}/5")
print("Questions with ranking differences:", len(mismatches))

if mismatches:
    print("\nFirst 3 mismatches:")
    for item in mismatches[:3]:
        print("\nQuestion:", item["question"])
        print("Overlap:", item["overlap"], "/ 5")
        print("Saved:", item["old_ids"])
        print("Reconstructed:", item["new_ids"])
```

```python

# ================================================================
# NCERT AI TUTOR — IMPROVED RETRIEVAL RECONSTRUCTION + EVALUATION
# Baseline: BM25 + dense retrieval + RRF
# Improved: multi-query BM25/dense + RRF + cross-encoder reranking
# Read-only inputs; all new results are timestamped.
# ================================================================

from google.colab import drive
drive.mount('/content/drive')

import os
import re
import json
import ast
import time
import pickle
import subprocess
import sys
import importlib.util
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import torch
import faiss

# -------------------- 1. Install missing dependency --------------------
if importlib.util.find_spec("sentence_transformers") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q",
        "sentence-transformers"
    ])

if importlib.util.find_spec("rank_bm25") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q",
        "rank-bm25"
    ])

from sentence_transformers import SentenceTransformer, CrossEncoder

# -------------------- 2. Paths and configuration --------------------
ROOT = Path("/content/drive/MyDrive/NCERT_BOOKS")
BASE = ROOT / "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

CHUNKS_PATH = BASE / "english_only_chunks.csv"
BM25_PATH = BASE / "indexes/bm25/bm25_index.pkl"
TOKENS_PATH = BASE / "indexes/bm25/tokenized_documents.pkl"
EMBEDDINGS_PATH = BASE / "indexes/embeddings/chunk_embeddings.npy"
FAISS_PATH = BASE / "indexes/faiss/index.faiss"

PREVIOUS_RESULTS = (
    BASE / "evaluation/retrieval_comparison_runs/"
    "baseline_vs_improved_20261010_203122.csv"
)

OUTPUT_DIR = BASE / "evaluation/retrieval_comparison_runs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

ENCODER_NAME = "sentence-transformers/all-MiniLM-L6-v2"
RERANKER_NAME = "cross-encoder/ms-marco-MiniLM-L-6-v2"

RRF_K = 60
BASELINE_CANDIDATES = 20
IMPROVED_PER_RETRIEVER = 40
IMPROVED_POOL_SIZE = 50
TOP_K = 5
BATCH_SIZE = 32

device = "cuda" if torch.cuda.is_available() else "cpu"
start_time = time.time()

print("=" * 72)
print("NCERT IMPROVED RETRIEVAL EXPERIMENT")
print("=" * 72)
print("Device:", device)
print("Loading saved artifacts...")

# -------------------- 3. Load artifacts if not already in memory --------------------
if "chunks" not in globals():
    chunks = pd.read_csv(CHUNKS_PATH)

if "bm25" not in globals():
    with open(BM25_PATH, "rb") as f:
        bm25 = pickle.load(f)

if "tokenized_docs" not in globals():
    with open(TOKENS_PATH, "rb") as f:
        tokenized_docs = pickle.load(f)

if "embeddings" not in globals():
    embeddings = np.load(EMBEDDINGS_PATH)

if "faiss_index" not in globals():
    faiss_index = faiss.read_index(str(FAISS_PATH))

if "encoder" not in globals():
    encoder = SentenceTransformer(ENCODER_NAME, device=device)

assert len(chunks) == len(tokenized_docs) == len(embeddings)
assert faiss_index.ntotal == len(chunks)
assert faiss_index.d == embeddings.shape[1] == 384
assert "text" in chunks.columns
assert "source_file" in chunks.columns
assert "pages" in chunks.columns

print(f"Artifacts validated: {len(chunks)} chunks")

# -------------------- 4. Reconstruct the verified baseline --------------------
def retrieve_hybrid(question, candidate_k=20, top_k=5, rrf_k=60):
    """Reconstructed baseline; previously verified against all 65 saved rankings."""
    bm25_scores = bm25.get_scores(question.lower().split())
    bm25_order = np.argsort(bm25_scores)[::-1][:candidate_k]

    qvec = encoder.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    ).astype("float32")

    _, dense_indices = faiss_index.search(qvec, candidate_k)
    dense_order = dense_indices[0]

    fused = {}

    for rank, idx in enumerate(bm25_order, start=1):
        idx = int(idx)
        fused[idx] = fused.get(idx, 0.0) + 1.0 / (rrf_k + rank)

    for rank, idx in enumerate(dense_order, start=1):
        if idx >= 0:
            idx = int(idx)
            fused[idx] = fused.get(idx, 0.0) + 1.0 / (rrf_k + rank)

    ranked = sorted(fused, key=fused.get, reverse=True)[:top_k]
    result = chunks.iloc[ranked].copy().reset_index(drop=True)
    result["retrieval_score"] = [fused[i] for i in ranked]
    return result


# -------------------- 5. Multi-query generation --------------------
STOPWORDS = {
    "a", "an", "the", "is", "are", "was", "were", "be", "been",
    "being", "do", "does", "did", "what", "when", "where", "why",
    "how", "which", "who", "whom", "whose", "according", "to",
    "from", "in", "on", "at", "of", "for", "with", "by", "and",
    "or", "but", "as", "it", "its", "this", "that", "these",
    "those", "their", "there", "they", "you", "your", "we", "our",
    "have", "has", "had", "can", "could", "would", "should",
    "explain", "describe", "state", "stated", "passage", "section",
    "page", "pages", "ncert", "activity", "idea", "procedure"
}

def make_query_variants(question):
    """Original question plus a lightweight keyword-focused variant."""
    original = re.sub(r"\s+", " ", str(question)).strip()
    words = re.findall(r"[a-zA-Z0-9]+", original.lower())
    keywords = [w for w in words if w not in STOPWORDS and len(w) > 1]

    keyword_query = " ".join(keywords).strip()
    variants = [original]

    if keyword_query and keyword_query.lower() != original.lower():
        variants.append(keyword_query)

    # Stable deduplication
    return list(dict.fromkeys(variants))


# -------------------- 6. Reconstructed improved retrieval --------------------
print("Loading cross-encoder:", RERANKER_NAME)
reranker = CrossEncoder(RERANKER_NAME, device=device)

def retrieve_improved(
    question,
    candidate_k=40,
    pool_size=50,
    top_k=5,
    rrf_k=60
):
    """
    Reconstructed improved pipeline:
    original + keyword query -> BM25/dense retrieval -> RRF pool
    -> cross-encoder reranking -> final top-k.
    """
    variants = make_query_variants(question)
    fused = {}

    for query in variants:
        # BM25 results
        scores = bm25.get_scores(query.lower().split())
        bm25_order = np.argsort(scores)[::-1][:candidate_k]

        for rank, idx in enumerate(bm25_order, start=1):
            idx = int(idx)
            fused[idx] = fused.get(idx, 0.0) + 1.0 / (rrf_k + rank)

        # Dense results
        qvec = encoder.encode(
            [query],
            normalize_embeddings=True,
            convert_to_numpy=True,
            show_progress_bar=False
        ).astype("float32")

        _, dense_indices = faiss_index.search(qvec, candidate_k)

        for rank, idx in enumerate(dense_indices[0], start=1):
            if idx >= 0:
                idx = int(idx)
                fused[idx] = fused.get(idx, 0.0) + 1.0 / (rrf_k + rank)

    # Limit the pool before the more expensive cross-encoder stage.
    candidate_ids = sorted(
        fused, key=fused.get, reverse=True
    )[:pool_size]

    if not candidate_ids:
        return chunks.iloc[[]].copy()

    candidate_texts = chunks.iloc[candidate_ids]["text"].fillna("").astype(str).tolist()
    pairs = [[str(question), text] for text in candidate_texts]

    cross_scores = reranker.predict(
        pairs,
        batch_size=BATCH_SIZE,
        show_progress_bar=False
    )

    reranked = sorted(
        zip(candidate_ids, cross_scores),
        key=lambda item: float(item[1]),
        reverse=True
    )[:top_k]

    final_ids = [idx for idx, score in reranked]
    final_scores = [float(score) for idx, score in reranked]

    result = chunks.iloc[final_ids].copy().reset_index(drop=True)
    result["reranker_score"] = final_scores
    result["rrf_score"] = [fused[idx] for idx in final_ids]
    return result


# -------------------- 7. Load the saved benchmark --------------------
if not PREVIOUS_RESULTS.exists():
    raise FileNotFoundError(
        f"Previous results CSV not found: {PREVIOUS_RESULTS}"
    )

benchmark = pd.read_csv(PREVIOUS_RESULTS)

required = {
    "question", "gold_source", "gold_pages",
    "baseline_chunks", "annotation_status"
}
missing = required - set(benchmark.columns)
if missing:
    raise ValueError(f"Benchmark is missing required columns: {sorted(missing)}")

print("\nBenchmark rows:", len(benchmark))
print("Annotation statuses:")
print(benchmark["annotation_status"].value_counts(dropna=False).to_dict())

# -------------------- 8. Source/page comparison helpers --------------------
def parse_list(value):
    if isinstance(value, list):
        return value
    if not isinstance(value, str) or not value.strip():
        return []
    for parser in (json.loads, ast.literal_eval):
        try:
            parsed = parser(value)
            if isinstance(parsed, list):
                return parsed
        except Exception:
            pass
    return []

def normalize_source(value):
    if pd.isna(value):
        return ""
    value = str(value).replace("\\", "/").strip().lower()
    value = re.sub(r"/+", "/", value)
    return value

def source_matches(gold_source, candidate_source):
    gold = normalize_source(gold_source)
    candidate = normalize_source(candidate_source)

    if not gold or not candidate:
        return False
    if gold == candidate:
        return True

    # Filename fallback only if the filenames match after normalization.
    gold_name = re.sub(r"[^a-z0-9]", "", gold.split("/")[-1])
    candidate_name = re.sub(r"[^a-z0-9]", "", candidate.split("/")[-1])
    return bool(gold_name and gold_name == candidate_name)

def parse_page_numbers(value):
    if pd.isna(value):
        return set()
    return {int(x) for x in re.findall(r"\d+", str(value))}

def page_overlap(gold_pages, candidate_pages):
    gold = parse_page_numbers(gold_pages)
    candidate = parse_page_numbers(candidate_pages)
    return bool(gold & candidate)

def evaluate_top5(gold_source, gold_pages, candidates):
    source_hit = False
    source_page_hit = False

    for _, candidate in candidates.iterrows():
        sm = source_matches(gold_source, candidate.get("source_file", ""))
        po = page_overlap(gold_pages, candidate.get("pages", ""))

        if sm:
            source_hit = True
        if sm and po:
            source_page_hit = True

    return source_hit, source_page_hit

# -------------------- 9. Run controlled comparison --------------------
rows = []
total = len(benchmark)

print("\nRunning baseline and reconstructed improved retrieval...")
print("The cross-encoder may take several minutes on CPU.")

for i, (_, row) in enumerate(benchmark.iterrows(), start=1):
    question = str(row["question"])

    # Recompute baseline with the now-verified implementation.
    baseline_results = retrieve_hybrid(
        question,
        candidate_k=BASELINE_CANDIDATES,
        top_k=TOP_K,
        rrf_k=RRF_K
    )

    # Run the reconstructed multi-query reranker.
    improved_results = retrieve_improved(
        question,
        candidate_k=IMPROVED_PER_RETRIEVER,
        pool_size=IMPROVED_POOL_SIZE,
        top_k=TOP_K,
        rrf_k=RRF_K
    )

    b_source, b_source_page = evaluate_top5(
        row["gold_source"], row["gold_pages"], baseline_results
    )
    i_source, i_source_page = evaluate_top5(
        row["gold_source"], row["gold_pages"], improved_results
    )

    # Compare reconstructed baseline against saved top-5 IDs.
    old_baseline = parse_list(row["baseline_chunks"])
    old_ids = [x.get("chunk_id") for x in old_baseline if isinstance(x, dict)]
    new_ids = baseline_results["chunk_id"].tolist()

    rows.append({
        "question": question,
        "gold_source": row["gold_source"],
        "gold_pages": row["gold_pages"],
        "annotation_status": row["annotation_status"],

        "baseline_source_hit_at_5": b_source,
        "baseline_source_page_hit_at_5": b_source_page,
        "improved_source_hit_at_5": i_source,
        "improved_source_page_hit_at_5": i_source_page,

        "baseline_exact_match_saved_top5": old_ids == new_ids,
        "baseline_top5": json.dumps([
            {
                "chunk_id": r["chunk_id"],
                "source_file": r["source_file"],
                "pages": r["pages"]
            }
            for _, r in baseline_results.iterrows()
        ], ensure_ascii=False),

        "improved_top5": json.dumps([
            {
                "chunk_id": r["chunk_id"],
                "source_file": r["source_file"],
                "pages": r["pages"],
                "reranker_score": float(r["reranker_score"]),
                "rrf_score": float(r["rrf_score"])
            }
            for _, r in improved_results.iterrows()
        ], ensure_ascii=False)
    })

    if i % 5 == 0 or i == total:
        print(f"Processed {i}/{total} questions...")

# -------------------- 10. Save new results --------------------
results = pd.DataFrame(rows)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_path = OUTPUT_DIR / f"reconstructed_improved_comparison_{timestamp}.csv"
results.to_csv(output_path, index=False)

# -------------------- 11. Report metrics --------------------
def metric_summary(column):
    values = results[column].astype(bool)
    return int(values.sum()), float(values.mean())

b_s_count, b_s_rate = metric_summary("baseline_source_hit_at_5")
i_s_count, i_s_rate = metric_summary("improved_source_hit_at_5")
b_p_count, b_p_rate = metric_summary("baseline_source_page_hit_at_5")
i_p_count, i_p_rate = metric_summary("improved_source_page_hit_at_5")
exact_count = int(results["baseline_exact_match_saved_top5"].sum())

print("\n" + "=" * 72)
print("RECONSTRUCTED RETRIEVAL COMPARISON")
print("=" * 72)
print(f"Questions evaluated: {len(results)}")
print(
    f"Baseline exact match with saved top-5: "
    f"{exact_count}/{len(results)} ({exact_count/len(results):.1%})"
)
print()
print(f"Baseline source hit@5:     {b_s_rate:.3f} ({b_s_count}/{len(results)})")
print(f"Improved source hit@5:     {i_s_rate:.3f} ({i_s_count}/{len(results)})")
print(f"Source hit@5 delta:        {i_s_rate-b_s_rate:+.3f}")
print()
print(f"Baseline source+page hit@5: {b_p_rate:.3f} ({b_p_count}/{len(results)})")
print(f"Improved source+page hit@5: {i_p_rate:.3f} ({i_p_count}/{len(results)})")
print(f"Source+page hit@5 delta:    {i_p_rate-b_p_rate:+.3f}")

changed = results[
    (results["baseline_source_hit_at_5"] != results["improved_source_hit_at_5"])
    | (
        results["baseline_source_page_hit_at_5"]
        != results["improved_source_page_hit_at_5"]
    )
]

print("\nQuestions with changed hit outcomes:", len(changed))
if len(changed):
    display(changed[[
        "question", "gold_source", "gold_pages",
        "baseline_source_hit_at_5", "improved_source_hit_at_5",
        "baseline_source_page_hit_at_5", "improved_source_page_hit_at_5"
    ]].head(20))

print("\nNew results saved to:")
print(output_path)
print(f"\nElapsed time: {(time.time()-start_time)/60:.1f} minutes")

print(
    "\nIMPORTANT: This is a reconstructed improved pipeline, not the recovered "
    "original function. Treat metrics as provisional until gold source/page "
    "annotations are manually verified."
)
```

```python

# ================================================================
# FULL 65-QUESTION RETRIEVAL COMPARISON
# Requires retrieve_hybrid and retrieve_improved to be loaded.
# Saves a new CSV; does not modify corpus, indexes, or benchmark.
# ================================================================

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import json
import re
import time
from pathlib import Path
from datetime import datetime
from tqdm.auto import tqdm

# ---- 1. Verify functions are available ----
if "retrieve_hybrid" not in globals():
    raise RuntimeError(
        "retrieve_hybrid is not loaded. Run the retrieval reconstruction cell first."
    )

if "retrieve_improved" not in globals():
    raise RuntimeError(
        "retrieve_improved is not loaded. Run the improved retrieval reconstruction cell first."
    )

# ---- 2. Load the frozen previous comparison as question/gold-label input ----
BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)

INPUT_CSV = (
    BASE / "evaluation/retrieval_comparison_runs/"
    "baseline_vs_improved_20261010_203122.csv"
)

OUTPUT_DIR = BASE / "evaluation/retrieval_comparison_runs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

benchmark = pd.read_csv(INPUT_CSV)

required_cols = ["question", "gold_source", "gold_pages"]
missing = [c for c in required_cols if c not in benchmark.columns]
if missing:
    raise ValueError(f"Missing required benchmark columns: {missing}")

print("Questions:", len(benchmark))
print("Annotation status counts:")
if "annotation_status" in benchmark.columns:
    print(benchmark["annotation_status"].value_counts(dropna=False).to_dict())

# ---- 3. Evaluation helpers ----
def norm_source(x):
    if pd.isna(x):
        return ""
    x = str(x).replace("\\", "/").strip().lower()
    x = re.sub(r"/+", "/", x)
    return x

def same_source(gold, candidate):
    g = norm_source(gold)
    c = norm_source(candidate)
    if not g or not c:
        return False
    if g == c:
        return True

    # Fallback: compare normalized filenames.
    g_name = re.sub(r"[^a-z0-9]", "", g.split("/")[-1])
    c_name = re.sub(r"[^a-z0-9]", "", c.split("/")[-1])
    return bool(g_name and g_name == c_name)

def page_set(value):
    if pd.isna(value):
        return set()
    return {int(n) for n in re.findall(r"\d+", str(value))}

def score_top5(gold_source, gold_pages, result_df):
    source_hit = False
    source_page_hit = False

    for _, candidate in result_df.iterrows():
        src_hit = same_source(
            gold_source, candidate.get("source_file", "")
        )
        pg_hit = bool(
            page_set(gold_pages)
            & page_set(candidate.get("pages", ""))
        )

        if src_hit:
            source_hit = True
        if src_hit and pg_hit:
            source_page_hit = True

    return source_hit, source_page_hit

# ---- 4. Run both pipelines on every question ----
records = []
start = time.time()

for _, row in tqdm(
    benchmark.iterrows(),
    total=len(benchmark),
    desc="Evaluating"
):
    question = str(row["question"])

    baseline = retrieve_hybrid(
        question, candidate_k=20, top_k=5
    )
    improved = retrieve_improved(
        question, candidate_k=40, pool_size=50, top_k=5
    )

    b_src, b_page = score_top5(
        row["gold_source"], row["gold_pages"], baseline
    )
    i_src, i_page = score_top5(
        row["gold_source"], row["gold_pages"], improved
    )

    records.append({
        "question": question,
        "gold_source": row["gold_source"],
        "gold_pages": row["gold_pages"],
        "annotation_status": row.get("annotation_status", "UNKNOWN"),

        "baseline_source_hit_at_5": b_src,
        "improved_source_hit_at_5": i_src,
        "baseline_source_page_hit_at_5": b_page,
        "improved_source_page_hit_at_5": i_page,

        "baseline_top5": json.dumps([
            {
                "chunk_id": r["chunk_id"],
                "source_file": r["source_file"],
                "pages": r["pages"]
            }
            for _, r in baseline.iterrows()
        ], ensure_ascii=False),

        "improved_top5": json.dumps([
            {
                "chunk_id": r["chunk_id"],
                "source_file": r["source_file"],
                "pages": r["pages"],
                "reranker_score": (
                    float(r["reranker_score"])
                    if "reranker_score" in r else None
                )
            }
            for _, r in improved.iterrows()
        ], ensure_ascii=False)
    })

results = pd.DataFrame(records)

# ---- 5. Save a NEW timestamped result ----
stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_path = OUTPUT_DIR / f"full_reconstructed_comparison_{stamp}.csv"
results.to_csv(output_path, index=False)

# ---- 6. Report metrics ----
def report(col):
    vals = results[col].astype(bool)
    return int(vals.sum()), float(vals.mean())

b_s_n, b_s = report("baseline_source_hit_at_5")
i_s_n, i_s = report("improved_source_hit_at_5")
b_p_n, b_p = report("baseline_source_page_hit_at_5")
i_p_n, i_p = report("improved_source_page_hit_at_5")

changed = results[
    (results["baseline_source_hit_at_5"] != results["improved_source_hit_at_5"])
    | (
        results["baseline_source_page_hit_at_5"]
        != results["improved_source_page_hit_at_5"]
    )
]

print("\n" + "=" * 68)
print("FULL RETRIEVAL COMPARISON")
print("=" * 68)
print(f"Evaluated questions: {len(results)}")
print(f"Baseline source hit@5:      {b_s:.3f} ({b_s_n}/{len(results)})")
print(f"Improved source hit@5:      {i_s:.3f} ({i_s_n}/{len(results)})")
print(f"Source hit@5 delta:         {i_s-b_s:+.3f}")
print()
print(f"Baseline source+page hit@5: {b_p:.3f} ({b_p_n}/{len(results)})")
print(f"Improved source+page hit@5: {i_p:.3f} ({i_p_n}/{len(results)})")
print(f"Source+page hit@5 delta:    {i_p-b_p:+.3f}")
print()
print("Questions with changed hit outcomes:", len(changed))

if len(changed):
    display(changed[[
        "question",
        "baseline_source_hit_at_5",
        "improved_source_hit_at_5",
        "baseline_source_page_hit_at_5",
        "improved_source_page_hit_at_5"
    ]].head(25))

print("\nSaved new results:", output_path)
print(f"Elapsed: {(time.time()-start)/60:.1f} minutes")
print(
    "\nReminder: source/page annotations are not fully verified. "
    "Treat these metrics as provisional."
)
```

```python

# ================================================================
# FULL 65-QUESTION RETRIEVAL COMPARISON
# Requires retrieve_hybrid and retrieve_improved to be loaded.
# Saves a new CSV; does not modify corpus, indexes, or benchmark.
# ================================================================

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import json
import re
import time
from pathlib import Path
from datetime import datetime
from tqdm.auto import tqdm

# ---- 1. Verify functions are available ----
if "retrieve_hybrid" not in globals():
    raise RuntimeError(
        "retrieve_hybrid is not loaded. Run the retrieval reconstruction cell first."
    )

if "retrieve_improved" not in globals():
    raise RuntimeError(
        "retrieve_improved is not loaded. Run the improved retrieval reconstruction cell first."
    )

# ---- 2. Load the frozen previous comparison as question/gold-label input ----
BASE = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)

INPUT_CSV = (
    BASE / "evaluation/retrieval_comparison_runs/"
    "baseline_vs_improved_20261010_203122.csv"
)

OUTPUT_DIR = BASE / "evaluation/retrieval_comparison_runs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

benchmark = pd.read_csv(INPUT_CSV)

required_cols = ["question", "gold_source", "gold_pages"]
missing = [c for c in required_cols if c not in benchmark.columns]
if missing:
    raise ValueError(f"Missing required benchmark columns: {missing}")

print("Questions:", len(benchmark))
print("Annotation status counts:")
if "annotation_status" in benchmark.columns:
    print(benchmark["annotation_status"].value_counts(dropna=False).to_dict())

# ---- 3. Evaluation helpers ----
def norm_source(x):
    if pd.isna(x):
        return ""
    x = str(x).replace("\\", "/").strip().lower()
    x = re.sub(r"/+", "/", x)
    return x

def same_source(gold, candidate):
    g = norm_source(gold)
    c = norm_source(candidate)
    if not g or not c:
        return False
    if g == c:
        return True

    # Fallback: compare normalized filenames.
    g_name = re.sub(r"[^a-z0-9]", "", g.split("/")[-1])
    c_name = re.sub(r"[^a-z0-9]", "", c.split("/")[-1])
    return bool(g_name and g_name == c_name)

def page_set(value):
    if pd.isna(value):
        return set()
    return {int(n) for n in re.findall(r"\d+", str(value))}

def score_top5(gold_source, gold_pages, result_df):
    source_hit = False
    source_page_hit = False

    for _, candidate in result_df.iterrows():
        src_hit = same_source(
            gold_source, candidate.get("source_file", "")
        )
        pg_hit = bool(
            page_set(gold_pages)
            & page_set(candidate.get("pages", ""))
        )

        if src_hit:
            source_hit = True
        if src_hit and pg_hit:
            source_page_hit = True

    return source_hit, source_page_hit

# ---- 4. Run both pipelines on every question ----
records = []
start = time.time()

for _, row in tqdm(
    benchmark.iterrows(),
    total=len(benchmark),
    desc="Evaluating"
):
    question = str(row["question"])

    baseline = retrieve_hybrid(
        question, candidate_k=20, top_k=5
    )
    improved = retrieve_improved(
        question, candidate_k=40, pool_size=50, top_k=5
    )

    b_src, b_page = score_top5(
        row["gold_source"], row["gold_pages"], baseline
    )
    i_src, i_page = score_top5(
        row["gold_source"], row["gold_pages"], improved
    )

    records.append({
        "question": question,
        "gold_source": row["gold_source"],
        "gold_pages": row["gold_pages"],
        "annotation_status": row.get("annotation_status", "UNKNOWN"),

        "baseline_source_hit_at_5": b_src,
        "improved_source_hit_at_5": i_src,
        "baseline_source_page_hit_at_5": b_page,
        "improved_source_page_hit_at_5": i_page,

        "baseline_top5": json.dumps([
            {
                "chunk_id": r["chunk_id"],
                "source_file": r["source_file"],
                "pages": r["pages"]
            }
            for _, r in baseline.iterrows()
        ], ensure_ascii=False),

        "improved_top5": json.dumps([
            {
                "chunk_id": r["chunk_id"],
                "source_file": r["source_file"],
                "pages": r["pages"],
                "reranker_score": (
                    float(r["reranker_score"])
                    if "reranker_score" in r else None
                )
            }
            for _, r in improved.iterrows()
        ], ensure_ascii=False)
    })

results = pd.DataFrame(records)

# ---- 5. Save a NEW timestamped result ----
stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_path = OUTPUT_DIR / f"full_reconstructed_comparison_{stamp}.csv"
results.to_csv(output_path, index=False)

# ---- 6. Report metrics ----
def report(col):
    vals = results[col].astype(bool)
    return int(vals.sum()), float(vals.mean())

b_s_n, b_s = report("baseline_source_hit_at_5")
i_s_n, i_s = report("improved_source_hit_at_5")
b_p_n, b_p = report("baseline_source_page_hit_at_5")
i_p_n, i_p = report("improved_source_page_hit_at_5")

changed = results[
    (results["baseline_source_hit_at_5"] != results["improved_source_hit_at_5"])
    | (
        results["baseline_source_page_hit_at_5"]
        != results["improved_source_page_hit_at_5"]
    )
]

print("\n" + "=" * 68)
print("FULL RETRIEVAL COMPARISON")
print("=" * 68)
print(f"Evaluated questions: {len(results)}")
print(f"Baseline source hit@5:      {b_s:.3f} ({b_s_n}/{len(results)})")
print(f"Improved source hit@5:      {i_s:.3f} ({i_s_n}/{len(results)})")
print(f"Source hit@5 delta:         {i_s-b_s:+.3f}")
print()
print(f"Baseline source+page hit@5: {b_p:.3f} ({b_p_n}/{len(results)})")
print(f"Improved source+page hit@5: {i_p:.3f} ({i_p_n}/{len(results)})")
print(f"Source+page hit@5 delta:    {i_p-b_p:+.3f}")
print()
print("Questions with changed hit outcomes:", len(changed))

if len(changed):
    display(changed[[
        "question",
        "baseline_source_hit_at_5",
        "improved_source_hit_at_5",
        "baseline_source_page_hit_at_5",
        "improved_source_page_hit_at_5"
    ]].head(25))

print("\nSaved new results:", output_path)
print(f"Elapsed: {(time.time()-start)/60:.1f} minutes")
print(
    "\nReminder: source/page annotations are not fully verified. "
    "Treat these metrics as provisional."
)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import json
import pandas as pd

ROOT = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)

RESULTS = os.path.join(
    ROOT, "evaluation/retrieval_comparison_runs",
    "full_reconstructed_comparison_20261011_120838.csv"
)
CHUNKS = os.path.join(ROOT, "english_only_chunks.csv")

results = pd.read_csv(RESULTS)
chunks = pd.read_csv(CHUNKS).fillna("")

chunk_lookup = chunks.set_index("chunk_id", drop=False)

def parse_records(value):
    try:
        parsed = json.loads(value) if isinstance(value, str) else value
        return parsed if isinstance(parsed, list) else []
    except Exception:
        return []

def normalize_path(value):
    return str(value).replace("\\", "/").strip().lower()

def parse_gold_pages(value):
    import re
    return set(re.findall(r"\d+", str(value)))

def page_match(record, gold_pages):
    raw = str(record.get("pages", ""))
    retrieved_pages = set(__import__("re").findall(r"\d+", raw))
    return bool(gold_pages & retrieved_pages)

# Only inspect cases where page-level outcome changed.
regressions = results[
    results["baseline_source_page_hit_at_5"].astype(str).str.lower().eq("true")
    & results["improved_source_page_hit_at_5"].astype(str).str.lower().eq("false")
]

gains = results[
    results["baseline_source_page_hit_at_5"].astype(str).str.lower().eq("false")
    & results["improved_source_page_hit_at_5"].astype(str).str.lower().eq("true")
]

selected = pd.concat([
    regressions.assign(case_type="PAGE_MATCH_LOST"),
    gains.assign(case_type="PAGE_MATCH_GAINED")
])

print("Page regressions:", len(regressions))
print("Page gains:", len(gains))
print("Chunk metadata columns:", chunks.columns.tolist())

# Inspect up to 9 changed cases, 5 retrieved chunks per pipeline.
for _, row in selected.iterrows():
    gold_pages = parse_gold_pages(row["gold_pages"])
    gold_source = normalize_path(row["gold_source"])

    print("\n" + "=" * 100)
    print("CASE:", row["case_type"])
    print("QUESTION:", row["question"])
    print("GOLD SOURCE:", row["gold_source"])
    print("GOLD PAGES:", row["gold_pages"])

    for pipeline, column in [
        ("BASELINE", "baseline_top5"),
        ("RECONSTRUCTED IMPROVED", "improved_top5"),
    ]:
        records = parse_records(row[column])
        print(f"\n--- {pipeline} ---")

        for rank, record in enumerate(records, start=1):
            cid = record.get("chunk_id", "")
            score = record.get("score", record.get("rerank_score", ""))

            if cid not in chunk_lookup.index:
                print(f"Rank {rank}: chunk not found in CSV: {cid}")
                print("Record:", str(record)[:500])
                continue

            meta = chunk_lookup.loc[cid]
            if isinstance(meta, pd.DataFrame):
                meta = meta.iloc[0]

            source = str(meta.get("source_file", ""))
            pages = str(meta.get("pages", ""))
            text = str(meta.get("text", ""))

            source_norm = normalize_path(source)
            source_match = (
                source_norm == gold_source
                or source_norm.endswith(gold_source)
                or gold_source.endswith(source_norm)
            )
            retrieved_pages = set(
                __import__("re").findall(r"\d+", pages)
            )
            page_hit = bool(gold_pages & retrieved_pages)

            print(
                f"\nRank {rank} | ID={cid} | score={score}"
                f"\nSource={source}"
                f"\nPages={pages}"
                f"\nSource match (simple path check)={source_match}"
                f"\nGold page overlap={page_hit}"
                f"\nText={text[:650]}"
            )

print("\nInspection complete. No files modified.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import re
import json
import pandas as pd
from datetime import datetime

ROOT = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)

RESULTS = os.path.join(
    ROOT,
    "evaluation/retrieval_comparison_runs",
    "full_reconstructed_comparison_20261011_120838.csv"
)

CHUNKS_PATH = os.path.join(ROOT, "english_only_chunks.csv")

OUT_DIR = os.path.join(
    ROOT, "evaluation", "source_aware_experiments"
)
os.makedirs(OUT_DIR, exist_ok=True)

df = pd.read_csv(RESULTS).fillna("")
chunks = pd.read_csv(CHUNKS_PATH).fillna("")

required = [
    "question", "gold_source", "gold_pages",
    "baseline_top5", "improved_top5"
]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

# ---------- Normalization helpers ----------

def normalize_path(value):
    s = str(value).replace("\\", "/").strip().lower()
    s = re.sub(r"/+", "/", s)
    s = re.sub(r"\.pdf$", "", s)
    return s.strip("/")

def source_matches(candidate_source, gold_source):
    c = normalize_path(candidate_source)
    g = normalize_path(gold_source)

    # Require exact normalized source identity.
    # Avoid loose substring matching between different books.
    return c == g

def parse_json_list(value):
    try:
        obj = json.loads(value) if isinstance(value, str) else value
        return obj if isinstance(obj, list) else []
    except Exception:
        return []

def page_numbers(value):
    # Page metadata is a comma-separated list in this corpus.
    return set(re.findall(r"\d+", str(value)))

def gold_page_set(value):
    return set(re.findall(r"\d+", str(value)))

def bool_value(v):
    return str(v).strip().lower() in ("true", "1", "yes")

# Lookup actual chunk metadata by ID.
chunks["chunk_id"] = chunks["chunk_id"].astype(str)
chunk_lookup = chunks.drop_duplicates("chunk_id").set_index("chunk_id")

def enrich_candidates(raw_candidates):
    enriched = []

    for rank, rec in enumerate(parse_json_list(raw_candidates), start=1):
        cid = str(rec.get("chunk_id", ""))

        if cid not in chunk_lookup.index:
            enriched.append({
                "rank_original": rank,
                "chunk_id": cid,
                "source_file": "",
                "pages": "",
                "text": str(rec.get("text", "")),
                "missing_metadata": True,
            })
            continue

        meta = chunk_lookup.loc[cid]
        enriched.append({
            "rank_original": rank,
            "chunk_id": cid,
            "source_file": str(meta.get("source_file", "")),
            "pages": str(meta.get("pages", "")),
            "text": str(meta.get("text", "")),
            "missing_metadata": False,
        })

    return enriched

# Source-aware score:
# 1) exact source match first
# 2) page overlap within that source
# 3) original ranking as tie-breaker
#
# This deliberately does not reward page overlap from another book.
def rerank_source_aware(candidates, gold_source, gold_pages):
    for c in candidates:
        same_source = source_matches(c["source_file"], gold_source)
        overlaps_page = bool(
            gold_page_set(gold_pages) & page_numbers(c["pages"])
        )

        c["source_match"] = same_source
        c["page_overlap"] = overlaps_page
        c["source_page_match"] = same_source and overlaps_page

    return sorted(
        candidates,
        key=lambda c: (
            int(c["source_page_match"]),
            int(c["source_match"]),
            -c["rank_original"],
        ),
        reverse=True,
    )

def top5_metrics(candidates, gold_source, gold_pages):
    top5 = candidates[:5]
    source_hit = any(
        source_matches(c["source_file"], gold_source)
        for c in top5
    )
    source_page_hit = any(
        source_matches(c["source_file"], gold_source)
        and bool(gold_page_set(gold_pages) & page_numbers(c["pages"]))
        for c in top5
    )
    return source_hit, source_page_hit

# ---------- Evaluate saved candidate lists ----------

records = []

for _, row in df.iterrows():
    gold_source = row["gold_source"]
    gold_pages = row["gold_pages"]

    baseline = enrich_candidates(row["baseline_top5"])
    improved = enrich_candidates(row["improved_top5"])

    # Source-aware rerank the improved pipeline's saved candidates.
    aware = rerank_source_aware(
        [dict(c) for c in improved],
        gold_source,
        gold_pages,
    )

    b_source, b_page = top5_metrics(
        baseline, gold_source, gold_pages
    )
    i_source, i_page = top5_metrics(
        improved, gold_source, gold_pages
    )
    a_source, a_page = top5_metrics(
        aware, gold_source, gold_pages
    )

    records.append({
        "question": row["question"],
        "gold_source": gold_source,
        "gold_pages": gold_pages,
        "annotation_status": row.get(
            "annotation_status", ""
        ),
        "baseline_source_hit_at_5": b_source,
        "baseline_source_page_hit_at_5": b_page,
        "improved_source_hit_at_5": i_source,
        "improved_source_page_hit_at_5": i_page,
        "source_aware_source_hit_at_5": a_source,
        "source_aware_source_page_hit_at_5": a_page,
        "baseline_top5": json.dumps(baseline, ensure_ascii=False),
        "improved_top5": json.dumps(improved, ensure_ascii=False),
        "source_aware_top5": json.dumps(aware, ensure_ascii=False),
    })

out_df = pd.DataFrame(records)

def report(label, source_col, page_col):
    n = len(out_df)
    source_hits = int(out_df[source_col].sum())
    page_hits = int(out_df[page_col].sum())

    print(f"\n{label}")
    print(f"  Source hit@5:      {source_hits}/{n} = {source_hits/n:.3f}")
    print(f"  Source+page hit@5: {page_hits}/{n} = {page_hits/n:.3f}")

print("=" * 72)
print("SOURCE-AWARE RERANKING EXPERIMENT")
print("Questions:", len(out_df))
print("Candidate pool: saved top five only")
print("Gold labels are still unverified; metrics are provisional.")

report(
    "BASELINE",
    "baseline_source_hit_at_5",
    "baseline_source_page_hit_at_5",
)
report(
    "RECONSTRUCTED IMPROVED",
    "improved_source_hit_at_5",
    "improved_source_page_hit_at_5",
)
report(
    "SOURCE-AWARE RERANKED",
    "source_aware_source_hit_at_5",
    "source_aware_source_page_hit_at_5",
)

print("\nDelta vs reconstructed improved:")
print(
    "  Source hit@5:",
    int(out_df["source_aware_source_hit_at_5"].sum())
    - int(out_df["improved_source_hit_at_5"].sum())
)
print(
    "  Source+page hit@5:",
    int(out_df["source_aware_source_page_hit_at_5"].sum())
    - int(out_df["improved_source_page_hit_at_5"].sum())
)

changed = out_df[
    (out_df["source_aware_source_page_hit_at_5"]
     != out_df["improved_source_page_hit_at_5"])
    | (out_df["source_aware_source_hit_at_5"]
       != out_df["improved_source_hit_at_5"])
]

print("\nQuestions whose outcomes changed:", len(changed))
display(changed[[
    "question",
    "gold_source",
    "gold_pages",
    "improved_source_hit_at_5",
    "source_aware_source_hit_at_5",
    "improved_source_page_hit_at_5",
    "source_aware_source_page_hit_at_5",
]].head(20))

out_path = os.path.join(
    OUT_DIR,
    "source_aware_comparison_"
    + datetime.now().strftime("%Y%m%d_%H%M%S")
    + ".csv"
)
out_df.to_csv(out_path, index=False)

print("\nSaved:", out_path)
print(
    "\nIMPORTANT: this experiment only reorders the saved top-five "
    "candidates. It cannot recover chunks absent from those lists."
)
```

```python

# ============================================================
# NCERT RETRIEVAL FAILURE ANALYSIS
# Inspect source misses, top-5 passages, and target-text evidence
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os
import re
import json
import ast
import pandas as pd
from datetime import datetime
from IPython.display import display

ROOT = "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

COMPARISON_PATH = os.path.join(
    ROOT,
    "evaluation",
    "retrieval_comparison_runs",
    "full_reconstructed_comparison_20261011_120838.csv"
)

CHUNKS_PATH = os.path.join(ROOT, "english_only_chunks.csv")

OUTPUT_DIR = os.path.join(
    ROOT, "evaluation", "failure_analysis"
)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ---------- Load data ----------
assert os.path.exists(COMPARISON_PATH), (
    f"Comparison file not found: {COMPARISON_PATH}"
)
assert os.path.exists(CHUNKS_PATH), (
    f"Chunks file not found: {CHUNKS_PATH}"
)

results = pd.read_csv(COMPARISON_PATH)
chunks = pd.read_csv(CHUNKS_PATH).fillna("")

print("=" * 80)
print("NCERT RETRIEVAL FAILURE ANALYSIS")
print("=" * 80)
print("Benchmark rows:", len(results))
print("Corpus chunks:", len(chunks))
print("Annotation statuses:")
if "annotation_status" in results.columns:
    print(results["annotation_status"].value_counts(dropna=False).to_string())

# ---------- Normalization / parsing ----------
def norm_source(value):
    s = str(value or "").strip().replace("\\", "/").lower()
    s = re.sub(r"/+", "/", s)
    s = re.sub(r"\.pdf$", "", s)
    return s

def parse_list(value):
    """Parse JSON/Python-list strings used in saved top-five columns."""
    if isinstance(value, list):
        return value
    if value is None or (isinstance(value, float) and pd.isna(value)):
        return []
    s = str(value).strip()
    if not s or s.lower() in ("nan", "none", "[]"):
        return []
    for parser in (json.loads, ast.literal_eval):
        try:
            parsed = parser(s)
            if isinstance(parsed, list):
                return parsed
        except Exception:
            pass
    return []

def get_value(item, *keys, default=""):
    if not isinstance(item, dict):
        return default
    for key in keys:
        if key in item and item[key] is not None:
            return item[key]
    return default

# Make a lookup for full chunk text and metadata.
chunk_lookup = {}
for _, row in chunks.iterrows():
    cid = str(row.get("chunk_id", ""))
    if cid:
        chunk_lookup[cid] = row.to_dict()

def get_candidates(row, pipeline):
    """
    Read saved top-five candidate records.
    Supports both the JSON candidate columns and the older *_chunks columns.
    """
    possible_columns = (
        [f"{pipeline}_top5", f"{pipeline}_chunks"]
    )

    raw = []
    for col in possible_columns:
        if col in row.index:
            raw = parse_list(row[col])
            if raw:
                break

    candidates = []
    for rank, item in enumerate(raw, start=1):
        if isinstance(item, str):
            cid = item
            item = {"chunk_id": cid}
        elif not isinstance(item, dict):
            continue

        cid = str(get_value(item, "chunk_id", "id", default=""))
        meta = chunk_lookup.get(cid, {})

        source = get_value(
            item, "source_file", "source", "book_path",
            default=meta.get("source_file", "")
        )
        page = get_value(
            item, "pages", "page",
            default=meta.get("pages", "")
        )
        text = get_value(
            item, "text", "chunk_text", "content",
            default=meta.get("text", "")
        )
        cls = get_value(
            item, "class",
            default=meta.get("class", "")
        )
        subject = get_value(
            item, "subject",
            default=meta.get("subject", "")
        )
        book = get_value(
            item, "book",
            default=meta.get("book", "")
        )

        candidates.append({
            "pipeline": pipeline,
            "rank": rank,
            "chunk_id": cid,
            "source_file": str(source),
            "pages": str(page),
            "class": str(cls),
            "subject": str(subject),
            "book": str(book),
            "text": str(text),
            "score": get_value(
                item, "score", "rerank_score", "similarity",
                default=""
            )
        })

    return candidates

# ---------- Find source misses ----------
miss_rows = []

for idx, row in results.iterrows():
    gold_source = str(row.get("gold_source", ""))
    gold_norm = norm_source(gold_source)

    # Prefer saved source-hit outcomes when available.
    baseline_hit = row.get("baseline_source_hit_at_5", None)
    improved_hit = row.get("improved_source_hit_at_5", None)

    baseline_candidates = get_candidates(row, "baseline")
    improved_candidates = get_candidates(row, "improved")

    def candidates_contain_source(candidates):
        return any(
            norm_source(c["source_file"]) == gold_norm
            for c in candidates
        )

    baseline_actual_hit = (
        candidates_contain_source(baseline_candidates)
        if baseline_candidates else
        (bool(baseline_hit) if pd.notna(baseline_hit) else False)
    )
    improved_actual_hit = (
        candidates_contain_source(improved_candidates)
        if improved_candidates else
        (bool(improved_hit) if pd.notna(improved_hit) else False)
    )

    if not baseline_actual_hit or not improved_actual_hit:
        miss_rows.append({
            "row_index": idx,
            "question": str(row.get("question", "")),
            "gold_source": gold_source,
            "gold_pages": str(row.get("gold_pages", "")),
            "annotation_status": str(row.get("annotation_status", "")),
            "baseline_hit": baseline_actual_hit,
            "improved_hit": improved_actual_hit,
            "baseline_candidates": baseline_candidates,
            "improved_candidates": improved_candidates
        })

print("\nSource-miss cases found:", len(miss_rows))
print("Cases where baseline misses:", sum(not x["baseline_hit"] for x in miss_rows))
print("Cases where improved misses:", sum(not x["improved_hit"] for x in miss_rows))

# ---------- Search for distinctive query terms in candidate text ----------
STOPWORDS = set("""
a an the is are was were be been being do does did have has had
what when where why how which who whom whose and or but if then
to of in on at for from by with about into over under during as
it its this that these those we you your our their they them he she
i let us can could should would may might will shall not no yes
class chapter page book teacher students student following among
""".split())

def tokens(text):
    return [
        t for t in re.findall(r"[a-z0-9]+", str(text).lower())
        if len(t) >= 3 and t not in STOPWORDS
    ]

def lexical_overlap(question, candidate_text):
    q = set(tokens(question))
    t = set(tokens(candidate_text))
    if not q:
        return 0, []
    common = sorted(q & t, key=lambda x: (-len(x), x))
    return len(common) / len(q), common[:20]

def source_page_match(candidate, gold_source, gold_pages):
    source_ok = (
        norm_source(candidate["source_file"]) == norm_source(gold_source)
    )
    # Extract digit groups from the gold and candidate page labels.
    gold_page_nums = set(re.findall(r"\d+", str(gold_pages)))
    candidate_page_nums = set(re.findall(r"\d+", str(candidate["pages"])))
    page_ok = bool(gold_page_nums & candidate_page_nums)
    return source_ok, (source_ok and page_ok)

# ---------- Print cases ----------
report_rows = []

for case_no, case in enumerate(miss_rows, start=1):
    print("\n" + "=" * 100)
    print(f"CASE {case_no} | benchmark row {case['row_index']}")
    print("Question:", case["question"])
    print("Gold source:", case["gold_source"])
    print("Gold pages:", case["gold_pages"])
    print("Annotation status:", case["annotation_status"])
    print(
        "Outcome:",
        f"baseline={'HIT' if case['baseline_hit'] else 'MISS'} |",
        f"improved={'HIT' if case['improved_hit'] else 'MISS'}"
    )

    # Highlight whether quoted text in the question appears in any candidate.
    q = case["question"]
    quoted = re.findall(r'["“](.{8,}?)[”"]', q)
    quoted = [x.strip() for x in quoted if x.strip()]

    for pipeline, candidates in [
        ("BASELINE", case["baseline_candidates"]),
        ("IMPROVED", case["improved_candidates"])
    ]:
        print("\n---", pipeline, "TOP CANDIDATES ---")

        if not candidates:
            print("No candidate records could be parsed from the saved CSV.")
            continue

        for c in candidates:
            source_ok, page_ok = source_page_match(
                c, case["gold_source"], case["gold_pages"]
            )
            overlap, common = lexical_overlap(q, c["text"])
            quoted_hit = any(
                phrase.lower() in c["text"].lower()
                for phrase in quoted
            )

            print(
                f"\nRank {c['rank']} | chunk={c['chunk_id']} | "
                f"source_match={source_ok} | source+page_match={page_ok}"
            )
            print(
                f"Class={c['class']} | Subject={c['subject']} | "
                f"Book={c['book']} | Pages={c['pages']}"
            )
            if c["score"] != "":
                print("Saved score:", c["score"])
            print(
                f"Lexical overlap={overlap:.3f} | "
                f"quoted_phrase_found={quoted_hit} | "
                f"shared_terms={common}"
            )
            print("TEXT:", c["text"][:1000].replace("\n", " "))

            report_rows.append({
                "benchmark_row": case["row_index"],
                "question": q,
                "gold_source": case["gold_source"],
                "gold_pages": case["gold_pages"],
                "annotation_status": case["annotation_status"],
                "pipeline": pipeline.lower(),
                "rank": c["rank"],
                "chunk_id": c["chunk_id"],
                "candidate_source": c["source_file"],
                "candidate_pages": c["pages"],
                "source_match": source_ok,
                "source_page_match": page_ok,
                "lexical_overlap": overlap,
                "quoted_phrase_found": quoted_hit,
                "shared_query_terms": ", ".join(common),
                "candidate_text": c["text"]
            })

# ---------- Save report ----------
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
out_csv = os.path.join(
    OUTPUT_DIR, f"source_miss_candidate_audit_{timestamp}.csv"
)

report_df = pd.DataFrame(report_rows)
report_df.to_csv(out_csv, index=False, encoding="utf-8-sig")

print("\n" + "=" * 100)
print("AUDIT COMPLETE")
print("Miss cases:", len(miss_rows))
print("Candidate rows saved:", len(report_df))
print("Full report:", out_csv)

if len(report_df):
    print("\nCandidate-level summary:")
    summary = report_df.groupby("pipeline").agg(
        candidates=("chunk_id", "count"),
        correct_source_candidates=("source_match", "sum"),
        correct_source_page_candidates=("source_page_match", "sum"),
        quoted_phrase_matches=("quoted_phrase_found", "sum"),
        mean_lexical_overlap=("lexical_overlap", "mean")
    ).round(3)
    display(summary)

    print("\nCases to inspect first (baseline or improved has no correct source):")
    display(pd.DataFrame([{
        "row": x["row_index"],
        "question": x["question"][:160],
        "gold_source": x["gold_source"],
        "gold_pages": x["gold_pages"],
        "baseline_hit": x["baseline_hit"],
        "improved_hit": x["improved_hit"]
    } for x in miss_rows]).head(20))
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os, re, json, ast
import pandas as pd
from datetime import datetime
from IPython.display import display

ROOT = "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
CORPUS_PATH = os.path.join(ROOT, "english_only_chunks.csv")
BENCH_PATH = os.path.join(ROOT, "english_only_benchmark.xlsx")
RESULT_PATH = os.path.join(
    ROOT, "evaluation", "retrieval_comparison_runs",
    "full_reconstructed_comparison_20261011_120838.csv"
)
OUT_DIR = os.path.join(ROOT, "evaluation", "benchmark_audit")
os.makedirs(OUT_DIR, exist_ok=True)

for p in [CORPUS_PATH, BENCH_PATH, RESULT_PATH]:
    if not os.path.exists(p):
        raise FileNotFoundError(f"Required file not found: {p}")

corpus = pd.read_csv(CORPUS_PATH).fillna("")
benchmark = pd.read_excel(BENCH_PATH).fillna("")
results = pd.read_csv(RESULT_PATH).fillna("")

print("=" * 75)
print("BENCHMARK INTEGRITY AUDIT — READ ONLY")
print("=" * 75)
print("Corpus chunks:", len(corpus))
print("Benchmark rows:", len(benchmark))
print("Evaluation rows:", len(results))
print("No source files or existing benchmark rows will be modified.")

# ---------- Helpers ----------
def norm_path(s):
    s = str(s).strip().replace("\\", "/").lower()
    s = re.sub(r"/+", "/", s)
    s = re.sub(r"\.pdf$", "", s)
    return s

def page_numbers(s):
    return set(re.findall(r"\d+", str(s)))

def parse_candidates(value):
    if isinstance(value, list):
        return value
    s = str(value).strip()
    if not s or s.lower() in {"nan", "none"}:
        return []
    for parser in (json.loads, ast.literal_eval):
        try:
            x = parser(s)
            if isinstance(x, list):
                return x
        except Exception:
            pass
    return []

def get_col(df, candidates):
    lower_map = {str(c).lower().strip(): c for c in df.columns}
    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[candidate.lower()]
    return None

# ---------- Inspect actual benchmark columns ----------
print("\nBenchmark columns:")
print(list(benchmark.columns))
print("\nEvaluation columns:")
print(list(results.columns))

# Map likely column names; don't assume a single schema.
qcol = get_col(benchmark, ["question", "query", "user_question"])
sourcecol = get_col(benchmark, [
    "gold_source", "source_file", "source", "expected_source",
    "gold_source_file"
])
pagecol = get_col(benchmark, [
    "gold_pages", "pages", "page", "expected_pages", "gold_page"
])
statuscol = get_col(benchmark, ["annotation_status", "status"])

if qcol is None or sourcecol is None:
    raise ValueError(
        "Could not identify question and gold-source columns. "
        "Check the printed benchmark columns and update qcol/sourcecol."
    )

# ---------- Corpus source inventory ----------
source_col = get_col(corpus, ["source_file"])
pages_col = get_col(corpus, ["pages"])
text_col = get_col(corpus, ["text"])
chunk_col = get_col(corpus, ["chunk_id"])

if not all([source_col, pages_col, text_col, chunk_col]):
    raise ValueError("Corpus is missing source_file/pages/text/chunk_id columns.")

corpus["_norm_source"] = corpus[source_col].map(norm_path)
source_inventory = (
    corpus.groupby("_norm_source")[source_col]
    .agg(lambda x: str(x.iloc[0]))
    .to_dict()
)
source_groups = {
    key: group.copy()
    for key, group in corpus.groupby("_norm_source")
}

# ---------- Match benchmark rows to evaluation rows ----------
eval_qcol = get_col(results, ["question", "query", "user_question"])
if eval_qcol is None:
    raise ValueError("No question column found in evaluation CSV.")

results["_q_norm"] = results[eval_qcol].astype(str).str.strip()
eval_by_question = {
    q: row for q, (_, row) in zip(
        results["_q_norm"], results.iterrows()
    )
}

# Vague generic phrasing patterns; these are flags for review, not verdicts.
vague_patterns = [
    r"according to the ncert passage",
    r"what activity, idea, or procedure is described",
    r"what is stated in the section beginning",
    r"what does the passage say",
    r"what is mentioned in the passage",
    r"according to the passage",
]

audit = []
candidate_audit = []

for i, row in benchmark.iterrows():
    question = str(row[qcol]).strip()
    gold_source = str(row[sourcecol]).strip()
    gold_pages = str(row[pagecol]).strip() if pagecol else ""
    status = str(row[statuscol]).strip() if statuscol else "UNKNOWN"

    normalized_source = norm_path(gold_source)
    source_exists = normalized_source in source_groups

    # Suggest possible source matches if exact path doesn't exist.
    possible_sources = []
    if not source_exists:
        base = os.path.basename(normalized_source)
        for norm, display_name in source_inventory.items():
            if base and (
                os.path.basename(norm) == base
                or base.replace("_", "") in os.path.basename(norm).replace("_", "")
            ):
                possible_sources.append(display_name)

    matched_chunks = source_groups.get(
        normalized_source, pd.DataFrame()
    )
    requested_pages = page_numbers(gold_pages)

    if len(matched_chunks) and requested_pages:
        corpus_pages = set()
        for p in matched_chunks[pages_col].astype(str):
            corpus_pages |= page_numbers(p)
        missing_pages = sorted(requested_pages - corpus_pages, key=int)
        page_present = len(missing_pages) == 0
    else:
        missing_pages = sorted(requested_pages, key=int)
        page_present = False if requested_pages else None

    vague_hits = [
        pat for pat in vague_patterns if re.search(pat, question, re.I)
    ]
    generic_question = bool(vague_hits)

    # Search the correct source's corpus text for useful snippets.
    source_snippets = []
    if len(matched_chunks):
        qwords = set(re.findall(r"[a-z0-9]{3,}", question.lower()))
        for _, cr in matched_chunks.iterrows():
            text = str(cr[text_col])
            text_words = set(re.findall(r"[a-z0-9]{3,}", text.lower()))
            overlap = len(qwords & text_words) / max(1, len(qwords))
            source_snippets.append({
                "chunk_id": str(cr[chunk_col]),
                "pages": str(cr[pages_col]),
                "overlap": overlap,
                "text": text[:600]
            })
        source_snippets.sort(key=lambda x: x["overlap"], reverse=True)

    # Check existing saved top-five candidates where available.
    erow = eval_by_question.get(question)
    candidate_source_hit = None
    candidate_page_hit = None
    top5_col = get_col(results, ["improved_top5", "improved_chunks"])
    if erow is not None and top5_col:
        candidates = parse_candidates(erow[top5_col])
        source_hit = False
        source_page_hit = False
        for cand in candidates:
            if not isinstance(cand, dict):
                continue
            cid = str(cand.get("chunk_id", cand.get("id", "")))
            meta = corpus[corpus[chunk_col].astype(str) == cid]
            if len(meta):
                csource = str(meta.iloc[0][source_col])
                cpages = str(meta.iloc[0][pages_col])
            else:
                csource = str(cand.get("source_file", cand.get("source", "")))
                cpages = str(cand.get("pages", cand.get("page", "")))
            smatch = norm_path(csource) == normalized_source
            pmatch = bool(page_numbers(cpages) & requested_pages)
            source_hit |= smatch
            source_page_hit |= (smatch and pmatch)
        candidate_source_hit = source_hit
        candidate_page_hit = source_page_hit

    flags = []
    if not source_exists:
        flags.append("GOLD_SOURCE_NOT_FOUND")
    if source_exists and requested_pages and not page_present:
        flags.append("GOLD_PAGE_NOT_FOUND_IN_SOURCE_CHUNKS")
    if generic_question:
        flags.append("GENERIC_TEMPLATE_QUESTION")
    if status != "VERIFIED":
        flags.append("ANNOTATION_NOT_VERIFIED")
    if source_exists and not len(matched_chunks):
        flags.append("NO_CHUNKS_FOR_SOURCE")

    audit.append({
        "benchmark_row": i,
        "question": question,
        "gold_source": gold_source,
        "gold_pages": gold_pages,
        "annotation_status": status,
        "source_exists_exactly": source_exists,
        "possible_source_filename_matches": " | ".join(possible_sources),
        "gold_pages_present_in_source_chunks": page_present,
        "gold_pages_not_found": ", ".join(missing_pages),
        "generic_template_flag": generic_question,
        "generic_patterns_matched": " | ".join(vague_hits),
        "improved_top5_source_hit": candidate_source_hit,
        "improved_top5_source_page_hit": candidate_page_hit,
        "review_flags": " | ".join(flags),
        "best_source_chunk_ids": " | ".join(
            x["chunk_id"] for x in source_snippets[:3]
        ),
        "best_source_chunk_pages": " | ".join(
            x["pages"] for x in source_snippets[:3]
        ),
        "best_source_text_snippets": " || ".join(
            x["text"] for x in source_snippets[:3]
        )
    })

    for snip in source_snippets[:5]:
        candidate_audit.append({
            "benchmark_row": i,
            "question": question,
            "gold_source": gold_source,
            "gold_pages": gold_pages,
            **snip
        })

audit_df = pd.DataFrame(audit)
snippets_df = pd.DataFrame(candidate_audit)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
audit_path = os.path.join(OUT_DIR, f"benchmark_integrity_{timestamp}.csv")
snippets_path = os.path.join(OUT_DIR, f"gold_source_snippets_{timestamp}.csv")

audit_df.to_csv(audit_path, index=False, encoding="utf-8-sig")
snippets_df.to_csv(snippets_path, index=False, encoding="utf-8-sig")

# ---------- Summary ----------
print("\n" + "=" * 75)
print("AUDIT SUMMARY")
print("=" * 75)
print("Exact gold source exists:",
      int(audit_df["source_exists_exactly"].sum()), "/", len(audit_df))
print("Gold page(s) present in source chunks:",
      int(audit_df["gold_pages_present_in_source_chunks"].fillna(False).sum()),
      "/", len(audit_df))
print("Generic/template questions flagged:",
      int(audit_df["generic_template_flag"].sum()), "/", len(audit_df))
print("Rows with at least one review flag:",
      int((audit_df["review_flags"].str.len() > 0).sum()), "/", len(audit_df))
print("Improved top-five source hits:",
      int(audit_df["improved_top5_source_hit"].fillna(False).sum()), "/", len(audit_df))
print("Improved top-five source+page hits:",
      int(audit_df["improved_top5_source_page_hit"].fillna(False).sum()), "/", len(audit_df))

print("\nReview flags:")
display(
    audit_df["review_flags"].replace("", "(no automated flag)")
    .value_counts().rename_axis("flag_combination").reset_index(name="rows")
)

print("\nHighest-priority rows:")
priority = audit_df[
    (~audit_df["source_exists_exactly"])
    | (audit_df["generic_template_flag"])
    | (audit_df["gold_pages_present_in_source_chunks"].eq(False))
].copy()

display(priority[[
    "benchmark_row", "question", "gold_source", "gold_pages",
    "source_exists_exactly", "possible_source_filename_matches",
    "gold_pages_present_in_source_chunks", "generic_template_flag",
    "improved_top5_source_hit", "review_flags"
]].head(30))

print("\nSaved audit:", audit_path)
print("Saved source snippets:", snippets_path)
print("\nImportant: automated flags are review aids, not corrected gold labels.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
from datetime import datetime
from IPython.display import display

ROOT = "/content/drive/MyDrive/NCERT_BOOKS/05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

BENCH_PATH = os.path.join(ROOT, "english_only_benchmark.xlsx")
AUDIT_PATH = os.path.join(
    ROOT, "evaluation", "benchmark_audit",
    "benchmark_integrity_20261011_132943.csv"
)
OUT_DIR = os.path.join(ROOT, "evaluation", "benchmark_audit")
os.makedirs(OUT_DIR, exist_ok=True)

bench = pd.read_excel(BENCH_PATH).fillna("")
audit = pd.read_csv(AUDIT_PATH).fillna("")

# Preserve the benchmark row order; the audit uses zero-based row indices.
bench = bench.reset_index(drop=True)
audit = audit.sort_values("benchmark_row").reset_index(drop=True)

if len(bench) != len(audit):
    raise ValueError(
        f"Row count mismatch: benchmark={len(bench)}, audit={len(audit)}"
    )

# Bring across useful diagnostic fields.
diagnostic_cols = [
    "source_exists_exactly",
    "gold_pages_present_in_source_chunks",
    "gold_pages_not_found",
    "generic_template_flag",
    "improved_top5_source_hit",
    "improved_top5_source_page_hit",
    "review_flags",
    "best_source_chunk_ids",
    "best_source_chunk_pages",
    "best_source_text_snippets",
]

for col in diagnostic_cols:
    if col in audit.columns:
        bench[col] = audit[col].values

def nonempty(value):
    return bool(str(value).strip()) and str(value).strip().lower() not in {
        "nan", "none"
    }

def truthy(value):
    return str(value).strip().lower() in {"true", "1", "yes"}

def review_priority(row):
    reasons = []

    if not truthy(row.get("evidence_verified", "")):
        reasons.append("EVIDENCE_NOT_VERIFIED")

    if not truthy(row.get("answer_supported_by_evidence", "")):
        reasons.append("ANSWER_SUPPORT_NOT_VERIFIED")

    if not nonempty(row.get("evidence_excerpt", "")):
        reasons.append("MISSING_EVIDENCE_EXCERPT")

    if not nonempty(row.get("expected_answer", "")):
        reasons.append("MISSING_EXPECTED_ANSWER")

    if truthy(row.get("generic_template_flag", False)):
        reasons.append("GENERIC_QUESTION_REVIEW")

    if str(row.get("gold_pages_present_in_source_chunks", "")).lower() == "false":
        reasons.append("PAGE_MAPPING_REVIEW")

    if not truthy(row.get("source_exists_exactly", True)):
        reasons.append("SOURCE_PATH_REVIEW")

    return " | ".join(reasons) if reasons else "LOWER_PRIORITY_REVIEW"

bench["manual_review_reasons"] = bench.apply(review_priority, axis=1)

# Order review fields first, followed by all original benchmark fields.
preferred = [
    "question_id", "split", "question", "expected_answer",
    "evidence_excerpt", "gold_chunk_ids", "source_file", "gold_pages",
    "evidence_verified", "answer_supported_by_evidence",
    "negative_verification", "distractor_verification",
    "benchmark_status", "annotation_status",
    "manual_review_reasons", "review_flags",
    "source_exists_exactly", "gold_pages_present_in_source_chunks",
    "gold_pages_not_found", "generic_template_flag",
    "improved_top5_source_hit", "improved_top5_source_page_hit",
    "best_source_chunk_ids", "best_source_chunk_pages",
    "best_source_text_snippets"
]
ordered = [c for c in preferred if c in bench.columns]
ordered += [c for c in bench.columns if c not in ordered]
worksheet = bench[ordered].copy()

# Prioritise rows with missing verification/evidence first.
worksheet["_priority"] = worksheet["manual_review_reasons"].apply(
    lambda x: 0 if x != "LOWER_PRIORITY_REVIEW" else 1
)
worksheet = worksheet.sort_values(
    ["_priority", "question_id"], kind="stable"
).drop(columns="_priority")

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
csv_path = os.path.join(OUT_DIR, f"human_review_worksheet_{timestamp}.csv")
xlsx_path = os.path.join(OUT_DIR, f"human_review_worksheet_{timestamp}.xlsx")

worksheet.to_csv(csv_path, index=False, encoding="utf-8-sig")
worksheet.to_excel(xlsx_path, index=False)

print("=" * 70)
print("HUMAN REVIEW WORKSHEET CREATED")
print("=" * 70)
print("Total questions:", len(worksheet))
print("Questions requiring review:",
      (worksheet["manual_review_reasons"] != "LOWER_PRIORITY_REVIEW").sum())
print("\nReview reason counts:")
display(
    worksheet["manual_review_reasons"]
    .value_counts()
    .rename_axis("reason")
    .reset_index(name="count")
)

print("\nFirst 15 questions to review:")
display(worksheet[[
    c for c in [
        "question_id", "question", "expected_answer", "evidence_excerpt",
        "gold_chunk_ids", "gold_pages", "manual_review_reasons"
    ] if c in worksheet.columns
]].head(15))

print("\nSaved CSV:", csv_path)
print("Saved Excel:", xlsx_path)
```

```python

# ============================================================
# STEP 2 — AUDIT BENCHMARK EVIDENCE AGAINST THE NCERT CORPUS
# Does the cited source exist? Does the evidence resemble a
# real corpus chunk? Are class/subject labels consistent?
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os
import re
import pandas as pd
from datetime import datetime
from difflib import SequenceMatcher
from collections import Counter

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)
BENCHMARK_PATH = os.path.join(
    BASE, "child_perspective_ncert_benchmark_v1.xlsx"
)
CORPUS_PATH = os.path.join(BASE, "english_only_chunks.csv")
OUT_DIR = os.path.join(BASE, "evaluation/child_benchmark_v1")
os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(BENCHMARK_PATH):
    raise FileNotFoundError(f"Missing benchmark: {BENCHMARK_PATH}")
if not os.path.exists(CORPUS_PATH):
    raise FileNotFoundError(f"Missing corpus: {CORPUS_PATH}")

bench = pd.read_excel(BENCHMARK_PATH).fillna("")
corpus = pd.read_csv(CORPUS_PATH, low_memory=False).fillna("")

def norm_source(x):
    """Normalize filenames for comparison, not for source identity."""
    x = os.path.basename(str(x).strip()).lower()
    x = re.sub(r'\.(pdf|txt|docx?)$', '', x)
    return re.sub(r'[^a-z0-9]', '', x)

def norm_text(x):
    x = str(x).lower()
    x = re.sub(r'[^a-z0-9]+', ' ', x)
    return re.sub(r'\s+', ' ', x).strip()

def tokens(x):
    return set(norm_text(x).split())

# Prepare corpus metadata
corpus["_source_norm"] = corpus["source_file"].map(norm_source)
corpus["_text_norm"] = corpus["text"].map(norm_text)
corpus["_tokens"] = corpus["text"].map(tokens)

source_map = {}
for source in corpus["source_file"].astype(str).unique():
    source_map.setdefault(norm_source(source), source)

audit_rows = []

for _, q in bench.iterrows():
    qid = str(q.get("question_id", ""))
    expected_source = str(q.get("source_file", "")).strip()
    excerpt = str(q.get("evidence_excerpt", "")).strip()
    question = str(q.get("question", "")).strip()
    answer = str(q.get("expected_answer", "")).strip()
    qclass = str(q.get("class", "")).strip()
    qsubject = str(q.get("subject", "")).strip()
    answerability = str(q.get("answerability", "")).strip().upper()

    source_norm = norm_source(expected_source)
    source_chunks = corpus[corpus["_source_norm"] == source_norm].copy()
    exact_source_match = len(source_chunks) > 0

    # Compare the provided excerpt against chunks from its cited source.
    best_source_score = 0.0
    best_source_chunk = None
    excerpt_tokens = tokens(excerpt)

    if excerpt_tokens and len(source_chunks):
        for _, c in source_chunks.iterrows():
            ct = c["_tokens"]
            overlap = len(excerpt_tokens & ct) / max(1, len(excerpt_tokens))
            # Token coverage is useful for OCR/spacing variation,
            # but is not proof that the evidence is correct.
            if overlap > best_source_score:
                best_source_score = overlap
                best_source_chunk = c

    # Also find the best evidence match anywhere in the corpus.
    best_any_score = 0.0
    best_any_chunk = None
    if excerpt_tokens:
        for _, c in corpus.iterrows():
            overlap = len(excerpt_tokens & c["_tokens"]) / max(1, len(excerpt_tokens))
            if overlap > best_any_score:
                best_any_score = overlap
                best_any_chunk = c

    # Class/subject consistency checks where metadata is available.
    class_match = ""
    subject_match = ""
    if best_source_chunk is not None:
        class_match = (
            str(best_source_chunk["class"]).strip().lower()
            == str(qclass).strip().lower()
        ) if qclass else ""
        subject_match = (
            norm_text(best_source_chunk["subject"])
            == norm_text(qsubject)
        ) if qsubject else ""

    flags = []
    if not expected_source:
        flags.append("MISSING_GOLD_SOURCE")
    elif not exact_source_match:
        flags.append("GOLD_SOURCE_NOT_FOUND")

    if not excerpt:
        flags.append("MISSING_EVIDENCE_EXCERPT")
    elif best_source_score < 0.30:
        flags.append("WEAK_EVIDENCE_MATCH_IN_CITED_SOURCE")

    if excerpt and best_any_score >= 0.30 and not exact_source_match:
        flags.append("POSSIBLE_SOURCE_METADATA_MISMATCH")

    if class_match is False:
        flags.append("CLASS_MISMATCH")
    if subject_match is False:
        flags.append("SUBJECT_MISMATCH")

    if answerability == "ANSWERABLE" and not answer:
        flags.append("MISSING_EXPECTED_ANSWER")
    if answerability == "UNANSWERABLE" and answer:
        flags.append("CHECK_UNANSWERABLE_ROW_HAS_ANSWER")

    audit_rows.append({
        "question_id": qid,
        "question": question,
        "answerability": answerability,
        "expected_answer": answer,
        "gold_source_file": expected_source,
        "gold_source_exists": exact_source_match,
        "chunks_in_gold_source": len(source_chunks),
        "evidence_excerpt_present": bool(excerpt),
        "evidence_token_coverage_in_gold_source": round(best_source_score, 3),
        "best_gold_source_chunk_id": (
            best_source_chunk["chunk_id"] if best_source_chunk is not None else ""
        ),
        "best_gold_source_chunk_pages": (
            best_source_chunk["pages"] if best_source_chunk is not None else ""
        ),
        "best_any_source_token_coverage": round(best_any_score, 3),
        "best_any_source_file": (
            best_any_chunk["source_file"] if best_any_chunk is not None else ""
        ),
        "best_any_chunk_id": (
            best_any_chunk["chunk_id"] if best_any_chunk is not None else ""
        ),
        "class_match_to_best_source_chunk": class_match,
        "subject_match_to_best_source_chunk": subject_match,
        "audit_flags": " | ".join(flags) if flags else "NO_AUTOMATIC_FLAGS"
    })

audit = pd.DataFrame(audit_rows)

# A low overlap is a review flag, not conclusive proof of bad evidence.
def evidence_band(score):
    if score >= 0.70:
        return "STRONG_TEXT_OVERLAP_REVIEW"
    if score >= 0.30:
        return "PARTIAL_TEXT_OVERLAP_REVIEW"
    return "WEAK_OR_MISSING_TEXT_OVERLAP"

audit["evidence_overlap_band"] = audit[
    "evidence_token_coverage_in_gold_source"
].map(evidence_band)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
csv_path = os.path.join(OUT_DIR, f"evidence_audit_{timestamp}.csv")
xlsx_path = os.path.join(OUT_DIR, f"evidence_audit_{timestamp}.xlsx")

audit.to_csv(csv_path, index=False)
with pd.ExcelWriter(xlsx_path, engine="openpyxl") as writer:
    audit.to_excel(writer, index=False, sheet_name="Evidence Audit")
    audit["audit_flags"].value_counts().rename_axis(
        "audit_flags"
    ).reset_index(name="count").to_excel(
        writer, index=False, sheet_name="Flag Summary"
    )

print("=" * 65)
print("EVIDENCE AUDIT COMPLETE")
print("=" * 65)
print("Questions audited:", len(audit))
print("Exact normalized gold sources found:",
      int(audit["gold_source_exists"].sum()), "/", len(audit))
print("Missing evidence excerpts:",
      int((~audit["evidence_excerpt_present"]).sum()))
print("Strong text overlap (>= 0.70):",
      int((audit["evidence_token_coverage_in_gold_source"] >= 0.70).sum()))
print("Partial overlap (0.30–0.69):",
      int(((audit["evidence_token_coverage_in_gold_source"] >= 0.30) &
           (audit["evidence_token_coverage_in_gold_source"] < 0.70)).sum()))
print("Weak overlap (< 0.30):",
      int((audit["evidence_token_coverage_in_gold_source"] < 0.30).sum()))

print("\nFlag counts:")
print(
    audit["audit_flags"].str.split(" | ", regex=False)
    .explode().value_counts().to_string()
)

print("\nReview these rows first:")
display(audit[
    (audit["audit_flags"] != "NO_AUTOMATIC_FLAGS")
].head(20))

print("\nSaved files:")
print(csv_path)
print(xlsx_path)
print("\nNOTE: Text overlap is only an automated screening signal.")
print("It does NOT establish that an answer is factually supported.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
from datetime import datetime

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)
BENCH_PATH = os.path.join(BASE, "child_perspective_ncert_benchmark_v1.xlsx")
CORPUS_PATH = os.path.join(BASE, "english_only_chunks.csv")
OUT_DIR = os.path.join(BASE, "evaluation/child_benchmark_v1")
os.makedirs(OUT_DIR, exist_ok=True)

bench = pd.read_excel(BENCH_PATH).fillna("")
corpus = pd.read_csv(CORPUS_PATH, low_memory=False).fillna("")

def norm_source(x):
    x = str(x).strip().replace("\\", "/").lower()
    return "".join(ch for ch in x if ch.isalnum())

def norm_label(x):
    return "".join(ch for ch in str(x).lower() if ch.isalnum())

corpus["_src"] = corpus["source_file"].map(norm_source)
bench["_src"] = bench["source_file"].map(norm_source)

# Normalize class labels such as "1", "Class 01", and "Class_01".
def class_num(x):
    import re
    s = str(x).lower()
    m = re.search(r'(\d+)', s)
    return int(m.group(1)) if m else None

corpus["_class_num"] = corpus["class"].map(class_num)
bench["_class_num"] = bench["class"].map(class_num)

rows = []
for _, q in bench.iterrows():
    source = q["source_file"]
    matched = corpus[corpus["_src"] == q["_src"]]
    flags = []

    if matched.empty:
        flags.append("SOURCE_NOT_FOUND")
        source_classes = ""
        source_subjects = ""
        class_ok = False
        subject_ok = False
    else:
        source_classes = ", ".join(
            sorted(matched["class"].astype(str).unique())
        )
        source_subjects = ", ".join(
            sorted(matched["subject"].astype(str).unique())
        )

        class_values = set(matched["_class_num"].dropna().astype(int))
        expected_class = q["_class_num"]
        class_ok = (
            expected_class in class_values if expected_class is not None else None
        )

        subject_values = {
            norm_label(v) for v in matched["subject"].astype(str).unique()
        }
        expected_subject = norm_label(q["subject"])
        subject_ok = (
            expected_subject in subject_values if expected_subject else None
        )

        if class_ok is False:
            flags.append("CLASS_NOT_IN_CITED_SOURCE")
        if subject_ok is False:
            flags.append("SUBJECT_NOT_IN_CITED_SOURCE")

    answerability = str(q["answerability"]).strip().upper()
    answer = str(q["expected_answer"]).strip()

    if answerability == "UNANSWERABLE":
        flags.append("REVIEW_ABSTENTION_EXPECTATION")
        if answer:
            flags.append("UNANSWERABLE_HAS_REFERENCE_ANSWER")

    rows.append({
        "question_id": q["question_id"],
        "question": q["question"],
        "class": q["class"],
        "subject": q["subject"],
        "answerability": answerability,
        "expected_answer": answer,
        "source_file": source,
        "source_found": not matched.empty,
        "classes_in_source": source_classes,
        "subjects_in_source": source_subjects,
        "class_consistent": class_ok,
        "subject_consistent": subject_ok,
        "review_flags": " | ".join(flags) if flags else "NO_AUTOMATIC_FLAGS"
    })

check = pd.DataFrame(rows)
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
out_path = os.path.join(OUT_DIR, f"source_consistency_{timestamp}.xlsx")
check.to_excel(out_path, index=False)

print("=" * 60)
print("SOURCE CONSISTENCY CHECK")
print("=" * 60)
print("Questions:", len(check))
print("Cited sources found:", int(check["source_found"].sum()), "/", len(check))
print("Class consistent with cited source:",
      int((check["class_consistent"] == True).sum()))
print("Subject consistent with cited source:",
      int((check["subject_consistent"] == True).sum()))

print("\n--- MISSING SOURCE REFERENCES ---")
display(check[~check["source_found"]][
    ["question_id", "question", "source_file", "review_flags"]
])

print("\n--- UNANSWERABLE QUESTIONS ---")
display(check[check["answerability"] == "UNANSWERABLE"][
    ["question_id", "question", "expected_answer", "source_file", "review_flags"]
])

print("\n--- CLASS / SUBJECT ISSUES ---")
display(check[
    (check["class_consistent"] == False) |
    (check["subject_consistent"] == False)
][[
    "question_id", "question", "class", "subject",
    "source_file", "classes_in_source", "subjects_in_source",
    "review_flags"
]])

print("\nSaved:", out_path)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os, re
import pandas as pd

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)
BENCH = os.path.join(
    BASE, "evaluation/child_benchmark_v1",
    "benchmark_review_v20261011_152503.xlsx"
)
CHUNKS = os.path.join(BASE, "english_only_chunks.csv")

benchmark = pd.read_excel(BENCH)
chunks = pd.read_csv(CHUNKS).fillna("")

ids_to_inspect = ["CHILDQ036", "CHILDQ037", "CHILDQ038",
                  "CHILDQ039", "CHILDQ048"]

def norm(x):
    return re.sub(r"[^a-z0-9]+", "", str(x).lower())

for qid in ids_to_inspect:
    rows = benchmark[benchmark["question_id"].astype(str) == qid]
    if rows.empty:
        print(f"\n{'='*90}\n{qid}: NOT FOUND")
        continue

    r = rows.iloc[0]
    print(f"\n{'='*100}")
    print(f"{qid} | Class {r.get('class')} | {r.get('subject')}")
    print("QUESTION:", r.get("question"))
    print("ANSWERABILITY:", r.get("answerability"))
    print("EXPECTED ANSWER:", r.get("expected_answer"))
    print("SOURCE:", r.get("source_file"))
    print("CITED EVIDENCE:", r.get("evidence_excerpt"))
    print("REVIEW NOTE:", r.get("review_notes_new", ""))

    source = str(r.get("source_file", "")).strip()
    if not source:
        print("No source cited; intentional for some abstention tests.")
        continue

    source_matches = chunks[
        chunks["source_file"].astype(str).map(norm) == norm(source)
    ]

    if source_matches.empty:
        # Fallback: match the source basename if metadata differs slightly.
        basename = os.path.basename(source)
        source_matches = chunks[
            chunks["source_file"].astype(str).str.contains(
                re.escape(basename), case=False, na=False
            )
        ]

    if source_matches.empty:
        print("No exact source match in chunks. Check source metadata.")
        continue

    print(f"\nMatching source chunks: {len(source_matches)}")

    # Show chunks most relevant to the question and evidence excerpt.
    query_terms = set(re.findall(
        r"[a-z]+", str(r.get("question", "")).lower()
    ))
    evidence_terms = set(re.findall(
        r"[a-z]+", str(r.get("evidence_excerpt", "")).lower()
    ))
    stop = {"what", "why", "how", "the", "and", "are", "does", "do",
            "can", "i", "my", "a", "an", "is", "in", "of", "to", "with"}
    query_terms -= stop
    evidence_terms -= stop

    scored = []
    for _, c in source_matches.iterrows():
        text = str(c.get("text", ""))
        terms = set(re.findall(r"[a-z]+", text.lower()))
        score = len(query_terms & terms) + 0.5 * len(evidence_terms & terms)
        scored.append((score, c))

    for score, c in sorted(scored, key=lambda x: x[0], reverse=True)[:3]:
        print(f"\n--- chunk score {score:.1f} | page {c.get('page', '?')} ---")
        print(str(c.get("text", ""))[:1800])

print("\nInspection complete. No files were modified.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os, re, pickle, json
import numpy as np
import pandas as pd
from collections import defaultdict
from datetime import datetime

# ---------- Paths ----------
ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)
DIR = os.path.join(BASE, "evaluation/child_benchmark_v1")

BENCH = os.path.join(
    DIR, "child_benchmark_candidate_20261011_153141.xlsx"
)
CHUNKS = os.path.join(BASE, "english_only_chunks.csv")
BM25_PATH = os.path.join(
    BASE, "indexes/bm25/bm25_index.pkl"
)
TOKENS_PATH = os.path.join(
    BASE, "indexes/bm25/tokenized_documents.pkl"
)
EMB_PATH = os.path.join(
    BASE, "indexes/embeddings/chunk_embeddings.npy"
)
FAISS_PATH = os.path.join(
    BASE, "indexes/faiss/index.faiss"
)

for p in [BENCH, CHUNKS, BM25_PATH, TOKENS_PATH, EMB_PATH, FAISS_PATH]:
    if not os.path.exists(p):
        raise FileNotFoundError(f"Required artifact missing: {p}")

# ---------- Load artifacts ----------
benchmark = pd.read_excel(BENCH).fillna("")
chunks = pd.read_csv(CHUNKS).fillna("")
with open(BM25_PATH, "rb") as f:
    bm25 = pickle.load(f)
with open(TOKENS_PATH, "rb") as f:
    tokenized_docs = pickle.load(f)
embeddings = np.load(EMB_PATH)

import faiss
faiss_index = faiss.read_index(FAISS_PATH)

print("Questions:", len(benchmark))
print("Chunk rows:", len(chunks))
print("BM25 documents:", len(tokenized_docs))
print("Embedding matrix:", embeddings.shape)
print("FAISS vectors:", faiss_index.ntotal)

# ---------- Validate index alignment ----------
if not (len(chunks) == len(tokenized_docs) == embeddings.shape[0] == faiss_index.ntotal):
    raise ValueError(
        "Index alignment mismatch. Stop rather than evaluating misaligned artifacts."
    )

# ---------- Load embedding model ----------
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

def tokenize(text):
    return re.findall(r"\b\w+\b", str(text).lower())

def hybrid_rrf(bm25_ids, dense_ids, k=60):
    scores = defaultdict(float)
    for rank, idx in enumerate(bm25_ids, start=1):
        scores[int(idx)] += 1.0 / (k + rank)
    for rank, idx in enumerate(dense_ids, start=1):
        scores[int(idx)] += 1.0 / (k + rank)
    return [idx for idx, _ in sorted(
        scores.items(), key=lambda item: item[1], reverse=True
    )]

# ---------- Retrieve ----------
rows = []
top_k = 5

for i, r in benchmark.iterrows():
    q = str(r["question"]).strip()
    if not q:
        continue

    # BM25 top 20
    bm_scores = bm25.get_scores(tokenize(q))
    bm_ids = np.argsort(bm_scores)[::-1][:20].tolist()

    # Dense top 20; stored embeddings are normalized
    q_emb = model.encode(
        [q], normalize_embeddings=True, convert_to_numpy=True
    ).astype("float32")
    dense_scores, dense_indices = faiss_index.search(q_emb, 20)
    dense_ids = [int(x) for x in dense_indices[0] if int(x) >= 0]

    hybrid_ids = hybrid_rrf(bm_ids, dense_ids)[:20]

    for method, ids in [
        ("bm25", bm_ids[:top_k]),
        ("dense", dense_ids[:top_k]),
        ("hybrid_rrf", hybrid_ids[:top_k])
    ]:
        for rank, idx in enumerate(ids, start=1):
            c = chunks.iloc[idx]
            rows.append({
                "question_id": r.get("question_id", ""),
                "question": q,
                "answerability": r.get("answerability", ""),
                "annotation_status": r.get("review_status", ""),
                "method": method,
                "rank": rank,
                "chunk_index": idx,
                "source_file": c.get("source_file", ""),
                "page": c.get("page", ""),
                "text": c.get("text", "")
            })

results = pd.DataFrame(rows)
stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
out_csv = os.path.join(DIR, f"child_retrieval_smoketest_{stamp}.csv")
results.to_csv(out_csv, index=False)

# ---------- Summarize source coverage, not gold accuracy ----------
print("\nSaved retrieval results:", out_csv)
print("\nTop-5 source diversity per method:")
for method in ["bm25", "dense", "hybrid_rrf"]:
    sub = results[(results["method"] == method) & (results["rank"] <= 5)]
    unique_sources = sub.groupby("question_id")["source_file"].nunique()
    print(
        f"{method:12s} | questions={sub['question_id'].nunique():2d} "
        f"| mean unique sources in top-5={unique_sources.mean():.2f}"
    )

print("\nSample retrieved passages:")
for qid in benchmark["question_id"].astype(str).head(5):
    qtext = benchmark.loc[
        benchmark["question_id"].astype(str).eq(qid), "question"
    ].iloc[0]
    print(f"\n{qid}: {qtext}")
    sub = results[
        (results["question_id"].astype(str) == qid)
        & (results["method"] == "hybrid_rrf")
    ].sort_values("rank")
    for _, x in sub.head(3).iterrows():
        print(f"  #{x['rank']} | {x['source_file']} | page={x['page']}")
        print("   ", str(x["text"])[:350].replace("\n", " "))

print("\nNote: preliminary retrieval smoke test only.")
print("Gold evidence labels are not yet fully validated; do not report accuracy/recall as final.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import re
import pickle
import numpy as np
import pandas as pd
from datetime import datetime
from sentence_transformers import SentenceTransformer, CrossEncoder

# ============================================================
# 1. PATHS — READ EXISTING ARTIFACTS, WRITE ONLY NEW OUTPUTS
# ============================================================
ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)
DIR = os.path.join(BASE, "evaluation/child_benchmark_v1")

CHUNKS_FILE = os.path.join(BASE, "english_only_chunks.csv")
BM25_FILE = os.path.join(BASE, "indexes/bm25/bm25_index.pkl")
EMBEDDINGS_FILE = os.path.join(
    BASE, "indexes/embeddings/chunk_embeddings.npy"
)
FAILURE_FILE = os.path.join(
    DIR, "retrieval_failure_analysis_20261011_160325.xlsx"
)

for path in [CHUNKS_FILE, BM25_FILE, EMBEDDINGS_FILE, FAILURE_FILE]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing required file: {path}")

chunks = pd.read_csv(CHUNKS_FILE).fillna("")
failures = pd.read_excel(FAILURE_FILE).fillna("")
embeddings = np.load(EMBEDDINGS_FILE)

with open(BM25_FILE, "rb") as f:
    bm25 = pickle.load(f)

TEXT_COL = "text"
if TEXT_COL not in chunks.columns:
    raise ValueError(f"No text column. Found: {chunks.columns.tolist()}")

if len(chunks) != len(embeddings):
    raise ValueError(
        f"Alignment error: {len(chunks)} chunks vs "
        f"{len(embeddings)} embeddings"
    )

if len(chunks) != len(bm25.doc_len):
    raise ValueError(
        f"Alignment error: {len(chunks)} chunks vs "
        f"{len(bm25.doc_len)} BM25 documents"
    )

print("Corpus chunks:", len(chunks))
print("Embedding shape:", embeddings.shape)
print("BM25 documents:", len(bm25.doc_len))

# ============================================================
# 2. LOAD MODELS
# ============================================================
print("\nLoading embedding model...")
embedder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

print("Loading cross-encoder...")
reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")

# ============================================================
# 3. TARGET QUESTIONS
# ============================================================
targets = failures[
    failures["answerability"].astype(str).str.upper().eq("ANSWERABLE")
    | failures["question_id"].astype(str).str.contains("Q048")
].copy()

# Use 50 candidates per retriever, then RRF; rerank the merged pool.
CANDIDATE_K = 50
RRF_K = 60
FINAL_K = 5

def tokenize(text):
    # Match the existing whitespace-tokenized BM25 setup.
    return str(text).lower().split()

def safe_text(idx):
    return str(chunks.iloc[int(idx)][TEXT_COL])

def metadata(idx):
    row = chunks.iloc[int(idx)]
    return {
        "source": row.get("source_file", ""),
        "class": row.get("class", ""),
        "subject": row.get("subject", ""),
        "book": row.get("book", ""),
        "pages": row.get("pages", ""),
        "chunk_id": row.get("chunk_id", ""),
        "text": safe_text(idx),
    }

all_rows = []
summary_rows = []

# ============================================================
# 4. RUN THE CONTROLLED COMPARISON
# ============================================================
for _, qrow in targets.iterrows():
    qid = str(qrow["question_id"])
    question = str(qrow["question"])

    print(f"\nProcessing {qid}: {question}")

    # ---- BM25 ----
    bm25_scores = np.asarray(bm25.get_scores(tokenize(question)))
    bm25_order = np.argsort(bm25_scores)[::-1][:CANDIDATE_K]

    # ---- Dense retrieval ----
    qvec = embedder.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True
    )[0]

    dense_scores = embeddings @ qvec
    dense_order = np.argsort(dense_scores)[::-1][:CANDIDATE_K]

    # ---- Reciprocal Rank Fusion (RRF) ----
    rrf_scores = {}

    for rank, idx in enumerate(bm25_order, start=1):
        idx = int(idx)
        rrf_scores[idx] = rrf_scores.get(idx, 0.0) + 1 / (RRF_K + rank)

    for rank, idx in enumerate(dense_order, start=1):
        idx = int(idx)
        rrf_scores[idx] = rrf_scores.get(idx, 0.0) + 1 / (RRF_K + rank)

    rrf_order = sorted(
        rrf_scores.keys(),
        key=lambda idx: rrf_scores[idx],
        reverse=True
    )

    # ---- Cross-encoder reranking of RRF candidates ----
    pairs = [(question, safe_text(idx)) for idx in rrf_order]
    ce_scores = np.asarray(reranker.predict(pairs, show_progress_bar=False))
    ce_order_positions = np.argsort(ce_scores)[::-1]
    ce_order = [rrf_order[int(pos)] for pos in ce_order_positions]

    # Each strategy's final top five
    strategies = [
        ("BM25", [int(x) for x in bm25_order[:FINAL_K]],
         {int(x): float(bm25_scores[int(x)]) for x in bm25_order[:FINAL_K]}),
        ("Dense", [int(x) for x in dense_order[:FINAL_K]],
         {int(x): float(dense_scores[int(x)]) for x in dense_order[:FINAL_K]}),
        ("RRF", [int(x) for x in rrf_order[:FINAL_K]],
         {int(x): float(rrf_scores[int(x)]) for x in rrf_order[:FINAL_K]}),
        ("RRF+CrossEncoder", ce_order[:FINAL_K],
         {int(rrf_order[int(pos)]): float(ce_scores[int(pos)])
          for pos in ce_order_positions[:FINAL_K]}),
    ]

    rank_maps = {
        "BM25": {int(x): i for i, x in enumerate(bm25_order, start=1)},
        "Dense": {int(x): i for i, x in enumerate(dense_order, start=1)},
        "RRF": {int(x): i for i, x in enumerate(rrf_order, start=1)},
        "RRF+CrossEncoder": {int(x): i for i, x in enumerate(ce_order, start=1)},
    }

    for method, indices, scores in strategies:
        for rank, idx in enumerate(indices, start=1):
            m = metadata(idx)
            all_rows.append({
                "question_id": qid,
                "question": question,
                "answerability": qrow.get("answerability", ""),
                "method": method,
                "rank": rank,
                "score": scores.get(idx, np.nan),
                "chunk_id": m["chunk_id"],
                "class": m["class"],
                "subject": m["subject"],
                "book": m["book"],
                "source_file": m["source"],
                "pages": m["pages"],
                "text": m["text"],
            })

    summary_rows.append({
        "question_id": qid,
        "question": question,
        "bm25_top5": " | ".join(
            str(metadata(x)["source"]) for x in strategies[0][1]
        ),
        "dense_top5": " | ".join(
            str(metadata(x)["source"]) for x in strategies[1][1]
        ),
        "rrf_top5": " | ".join(
            str(metadata(x)["source"]) for x in strategies[2][1]
        ),
        "reranked_top5": " | ".join(
            str(metadata(x)["source"]) for x in strategies[3][1]
        ),
        "rrf_candidates": len(rrf_order),
    })

# ============================================================
# 5. SAVE NEW DIAGNOSTIC FILES
# ============================================================
results = pd.DataFrame(all_rows)
summary = pd.DataFrame(summary_rows)

stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
out_csv = os.path.join(DIR, f"retrieval_strategy_comparison_{stamp}.csv")
out_xlsx = os.path.join(DIR, f"retrieval_strategy_comparison_{stamp}.xlsx")

with pd.ExcelWriter(out_xlsx, engine="openpyxl") as writer:
    results.to_excel(writer, sheet_name="Top5_Passages", index=False)
    summary.to_excel(writer, sheet_name="Strategy_Summary", index=False)

results.to_csv(out_csv, index=False)

print("\n" + "=" * 80)
print("DONE — existing index and benchmark files were not modified.")
print("Excel:", out_xlsx)
print("CSV:", out_csv)

for qid in ["CHILDQ003", "CHILDQ011", "CHILDQ013",
            "CHILDQ025", "CHILDQ033", "CHILDQ035",
            "CHILDQ045", "CHILDQ048"]:
    subset = results[results["question_id"].eq(qid)]
    if subset.empty:
        continue

    print("\n" + "-" * 80)
    print(qid, "|", subset.iloc[0]["question"])

    for method in ["BM25", "Dense", "RRF", "RRF+CrossEncoder"]:
        m = subset[subset["method"].eq(method)].sort_values("rank")
        print(f"\n{method}:")
        for _, r in m.iterrows():
            print(
                f"  #{int(r['rank'])} | {r['source_file']} | "
                f"pages={r['pages']}\n"
                f"     {str(r['text'])[:250].replace(chr(10), ' ')}"
            )
```

```python

# ============================================================
# NCERT AI TUTOR — FULL 50-QUESTION RETRIEVAL COMPARISON
# BM25 vs Dense vs RRF vs RRF + Cross-Encoder
# Read-only existing indexes; outputs saved separately.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os, re, time, pickle, warnings
import numpy as np
import pandas as pd
import faiss
from datetime import datetime
from collections import defaultdict
from sentence_transformers import SentenceTransformer, CrossEncoder

warnings.filterwarnings("ignore")

# -------------------- PATHS --------------------

ROOT = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)

EVAL_DIR = os.path.join(
    ROOT, "evaluation", "child_benchmark_v1"
)

CHUNKS_PATH = os.path.join(ROOT, "english_only_chunks.csv")
BM25_PATH = os.path.join(
    ROOT, "indexes", "bm25", "bm25_index.pkl"
)
TOKENS_PATH = os.path.join(
    ROOT, "indexes", "bm25", "tokenized_documents.pkl"
)
EMBED_PATH = os.path.join(
    ROOT, "indexes", "embeddings", "chunk_embeddings.npy"
)
FAISS_PATH = os.path.join(
    ROOT, "indexes", "faiss", "index.faiss"
)

# Prefer the corrected candidate workbook if present.
CANDIDATE_PATH = os.path.join(
    EVAL_DIR, "child_benchmark_candidate_20261011_153141.xlsx"
)
ORIGINAL_PATH = os.path.join(
    ROOT, "child_perspective_ncert_benchmark_v1.xlsx"
)

if os.path.exists(CANDIDATE_PATH):
    BENCH_PATH = CANDIDATE_PATH
elif os.path.exists(ORIGINAL_PATH):
    BENCH_PATH = ORIGINAL_PATH
else:
    raise FileNotFoundError(
        "Benchmark workbook not found. Check BENCH_PATH."
    )

for p in [CHUNKS_PATH, BM25_PATH, EMBED_PATH, FAISS_PATH, BENCH_PATH]:
    if not os.path.exists(p):
        raise FileNotFoundError(f"Missing required file: {p}")

os.makedirs(EVAL_DIR, exist_ok=True)

# -------------------- LOAD FILES --------------------

chunks = pd.read_csv(CHUNKS_PATH).fillna("")
benchmark = pd.read_excel(BENCH_PATH)
embeddings = np.load(EMBED_PATH)

with open(BM25_PATH, "rb") as f:
    bm25 = pickle.load(f)

with open(TOKENS_PATH, "rb") as f:
    tokenized_docs = pickle.load(f)

faiss_index = faiss.read_index(FAISS_PATH)

# Normalize column names and strings.
chunks.columns = [str(c).strip() for c in chunks.columns]
benchmark.columns = [str(c).strip() for c in benchmark.columns]

required_chunk_cols = {"text", "source_file"}
missing = required_chunk_cols - set(chunks.columns)
if missing:
    raise ValueError(f"Missing chunk columns: {missing}")

if len(chunks) != len(embeddings):
    raise ValueError("Chunk and embedding counts do not match.")

if len(chunks) != len(tokenized_docs):
    raise ValueError("Chunk and BM25 token counts do not match.")

if faiss_index.ntotal != len(chunks):
    raise ValueError("Chunk and FAISS vector counts do not match.")

# -------------------- FIND BENCHMARK COLUMNS --------------------

def find_col(df, candidates, required=True):
    lookup = {str(c).strip().lower(): c for c in df.columns}
    for name in candidates:
        if name.lower() in lookup:
            return lookup[name.lower()]
    if required:
        raise ValueError(
            f"Could not find any of {candidates}. "
            f"Available columns: {list(df.columns)}"
        )
    return None

QCOL = find_col(benchmark, ["question"])
IDCOL = find_col(
    benchmark, ["question_id", "id"], required=False
)
ANS_COL = find_col(
    benchmark, ["answerability"], required=False
)
STATUS_COL = find_col(
    benchmark, ["annotation_status", "review_status"],
    required=False
)

if IDCOL is None:
    benchmark["question_id"] = [
        f"Q{i+1:03d}" for i in range(len(benchmark))
    ]
    IDCOL = "question_id"

if ANS_COL is None:
    benchmark["answerability"] = "UNKNOWN"
    ANS_COL = "answerability"

# Keep every question, including abstention tests.
benchmark = benchmark[
    benchmark[QCOL].astype(str).str.strip().ne("")
].copy().reset_index(drop=True)

print("Questions:", len(benchmark))
print("Corpus chunks:", len(chunks))
print("Embedding shape:", embeddings.shape)
print("FAISS vectors:", faiss_index.ntotal)
print("Benchmark:", BENCH_PATH)

# -------------------- LOAD MODELS --------------------

print("\nLoading dense encoder...")
encoder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

print("Loading cross-encoder...")
cross_encoder = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")

# -------------------- RETRIEVAL SETTINGS --------------------

DENSE_CANDIDATES = 50
BM25_CANDIDATES = 50
RRF_K = 60
FINAL_K = 5

def tokenize(text):
    # Matches the existing simple lowercase whitespace tokenization.
    return str(text).lower().split()

def get_source(idx):
    return str(chunks.iloc[int(idx)]["source_file"])

def get_text(idx):
    return str(chunks.iloc[int(idx)]["text"])

def get_pages(idx):
    row = chunks.iloc[int(idx)]
    if "pages" in chunks.columns:
        return str(row["pages"])
    if "first_page" in chunks.columns and "last_page" in chunks.columns:
        return f'{row["first_page"]}-{row["last_page"]}'
    return ""

def bm25_search(query, k=BM25_CANDIDATES):
    scores = np.asarray(bm25.get_scores(tokenize(query)))
    k = min(k, len(scores))
    ids = np.argsort(scores)[::-1][:k]
    return [(int(i), float(scores[i])) for i in ids]

def dense_search(query, k=DENSE_CANDIDATES):
    qvec = encoder.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True,
        show_progress_bar=False
    ).astype("float32")

    scores, ids = faiss_index.search(
        qvec, min(k, len(chunks))
    )
    return [
        (int(i), float(s))
        for i, s in zip(ids[0], scores[0])
        if int(i) >= 0
    ]

def rrf_fuse(bm25_results, dense_results, k=RRF_K):
    scores = defaultdict(float)

    for rank, (idx, _) in enumerate(bm25_results, start=1):
        scores[idx] += 1.0 / (k + rank)

    for rank, (idx, _) in enumerate(dense_results, start=1):
        scores[idx] += 1.0 / (k + rank)

    return sorted(
        scores.items(), key=lambda x: x[1], reverse=True
    )

def rerank_with_cross_encoder(query, candidate_ids):
    if not candidate_ids:
        return []

    pairs = [
        (query, get_text(idx)[:2500])
        for idx in candidate_ids
    ]

    scores = cross_encoder.predict(
        pairs, show_progress_bar=False
    )
    scores = np.asarray(scores).reshape(-1)

    ranked = sorted(
        zip(candidate_ids, scores),
        key=lambda x: float(x[1]),
        reverse=True
    )
    return [(int(idx), float(score)) for idx, score in ranked]

# -------------------- RUN COMPARISON --------------------

all_results = []
summary_rows = []
start_time = time.time()

for qi, row in benchmark.iterrows():
    qid = str(row[IDCOL])
    query = str(row[QCOL]).strip()
    answerability = str(row[ANS_COL]).strip()

    print(f"[{qi+1}/{len(benchmark)}] {qid}: {query[:85]}")

    bm25_results = bm25_search(query)
    dense_results = dense_search(query)
    fused_results = rrf_fuse(bm25_results, dense_results)

    # Each strategy produces a ranked top five.
    strategy_results = {
        "BM25": bm25_results[:FINAL_K],
        "Dense": dense_results[:FINAL_K],
        "RRF": fused_results[:FINAL_K],
    }

    # Rerank the union of both candidate lists using the cross-encoder.
    candidate_ids = list(dict.fromkeys(
        [idx for idx, _ in bm25_results] +
        [idx for idx, _ in dense_results]
    ))

    ce_results = rerank_with_cross_encoder(query, candidate_ids)
    strategy_results["RRF + CrossEncoder"] = ce_results[:FINAL_K]

    for strategy, ranked in strategy_results.items():
        for rank, (chunk_idx, score) in enumerate(ranked, start=1):
            chunk_row = chunks.iloc[chunk_idx]

            all_results.append({
                "question_id": qid,
                "question": query,
                "answerability": answerability,
                "strategy": strategy,
                "rank": rank,
                "chunk_index_in_dataframe": int(chunk_idx),
                "chunk_id": (
                    str(chunk_row["chunk_id"])
                    if "chunk_id" in chunks.columns else str(chunk_idx)
                ),
                "source_file": get_source(chunk_idx),
                "pages": get_pages(chunk_idx),
                "score": float(score),
                "passage": get_text(chunk_idx)
            })

# -------------------- BUILD OUTPUTS --------------------

results_df = pd.DataFrame(all_results)

# Human-reviewed relevance workbook from the earlier experiment.
# We use it only to bring forward judgements for the same question IDs.
REVIEW_PATH = os.path.join(
    EVAL_DIR,
    "hybrid_top5_human_review_20261011_153825.xlsx"
)

# Note: that file is the original blank review template.
# If your completed review is available in Drive, set REVIEW_PATH to it.
# Example: /content/drive/MyDrive/NCERT_BOOKS/.../hybrid_top5_human_review_REVIEWED_vicky.xlsx

# Save retrieval output regardless of whether review labels are available.
stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
OUT_CSV = os.path.join(
    EVAL_DIR, f"all50_retrieval_comparison_{stamp}.csv"
)
OUT_XLSX = os.path.join(
    EVAL_DIR, f"all50_retrieval_comparison_{stamp}.xlsx"
)

# -------------------- METRICS FROM HUMAN LABELS --------------------
# If a completed human review workbook is available, enter its path below.
# Labels must be explicitly marked Relevant / Partially relevant / Irrelevant.
# We do not infer relevance from similarity scores.

REVIEWED_PATH = ""  # Optional: paste full path to completed review workbook

metrics_rows = []

if REVIEWED_PATH and os.path.exists(REVIEWED_PATH):
    review = pd.read_excel(REVIEWED_PATH)
    review.columns = [str(c).strip() for c in review.columns]

    review_id_col = find_col(
        review, ["question_id", "id"]
    )

    # Convert the reviewed top-five labels into question-specific chunk relevance.
    relevance_by_question_source = defaultdict(dict)

    for _, rr in review.iterrows():
        rid = str(rr[review_id_col]).strip()
        for rank in range(1, 6):
            src_col = f"top{rank}_source"
            rel_col = f"top{rank}_relevance"

            if src_col in review.columns and rel_col in review.columns:
                source = str(rr.get(src_col, "")).strip()
                label = str(rr.get(rel_col, "")).strip().lower()

                if source:
                    relevance_by_question_source[rid][source] = label

    for (qid, strategy), group in results_df.groupby(
        ["question_id", "strategy"], sort=False
    ):
        group = group.sort_values("rank")
        labels = relevance_by_question_source.get(str(qid), {})

        # Source-level matching is approximate when multiple passages
        # come from the same book. Use it only as a preliminary metric.
        rels = []
        for _, passage in group.iterrows():
            label = labels.get(str(passage["source_file"]), "")
            rels.append(label == "relevant")

        relevant_ranks = [
            i + 1 for i, is_rel in enumerate(rels) if is_rel
        ]
        metrics_rows.append({
            "question_id": qid,
            "strategy": strategy,
            "relevant_at_5": int(any(rels)),
            "first_relevant_rank": (
                min(relevant_ranks) if relevant_ranks else None
            ),
            "reciprocal_rank": (
                1 / min(relevant_ranks) if relevant_ranks else 0
            ),
            "label_match_note": "Approximate source-level mapping"
        })

metrics_df = pd.DataFrame(metrics_rows)

with pd.ExcelWriter(OUT_XLSX, engine="openpyxl") as writer:
    results_df.to_excel(writer, sheet_name="All Top5 Results", index=False)
    if not metrics_df.empty:
        metrics_df.to_excel(writer, sheet_name="Preliminary Metrics", index=False)
    benchmark.to_excel(writer, sheet_name="Benchmark Copy", index=False)

results_df.to_csv(OUT_CSV, index=False)

# -------------------- PRINT SUMMARY --------------------

print("\n" + "=" * 72)
print("COMPARISON COMPLETE")
print("=" * 72)
print("Questions evaluated:", benchmark[QCOL].nunique())
print("Passages saved:", len(results_df))
print("Strategies:", results_df["strategy"].nunique())
print("Elapsed minutes:", round((time.time() - start_time) / 60, 2))
print("\nSaved Excel:", OUT_XLSX)
print("Saved CSV:", OUT_CSV)

print("\nTop-5 source diversity (diagnostic only):")
diversity = (
    results_df.groupby(["question_id", "strategy"])["source_file"]
    .nunique()
    .groupby("strategy")
    .mean()
    .sort_values(ascending=False)
)
print(diversity.round(2).to_string())

if not metrics_df.empty:
    print("\nPRELIMINARY HUMAN-LABEL METRICS")
    print("Warning: source-level label mapping can overestimate relevance.")
    print(
        metrics_df.groupby("strategy")
        .agg(
            relevant_at_5=("relevant_at_5", "mean"),
            MRR=("reciprocal_rank", "mean"),
            questions=("question_id", "nunique")
        )
        .round(3)
        .to_string()
    )
else:
    print(
        "\nNo relevance metrics calculated automatically because the "
        "completed review workbook path was not supplied."
    )
    print(
        "Use the saved top-five results for human annotation; do not "
        "treat source diversity as retrieval accuracy."
    )

print("\nIMPORTANT:")
print("- This is a fresh comparison, not a byte-for-byte replay of old runs.")
print("- Existing corpus, indexes, and benchmark were not overwritten.")
print("- Review the same 50 questions across all four methods.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
import numpy as np

ROOT = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

REVIEW_PATH = os.path.join(
    ROOT, "all50_human_relevance_review_corrected.xlsx"
)

if not os.path.exists(REVIEW_PATH):
    raise FileNotFoundError(
        f"Can't find the workbook:\n{REVIEW_PATH}\n"
        "Check the exact filename in Google Drive."
    )

df = pd.read_excel(REVIEW_PATH, sheet_name="Review")
df.columns = df.columns.str.strip()

required = {
    "question_id", "question", "answerability",
    "strategy", "rank", "relevance"
}
missing = required - set(df.columns)
if missing:
    raise ValueError(f"Missing columns: {missing}")

df["relevance"] = df["relevance"].astype(str).str.strip().str.lower()
df["strategy"] = df["strategy"].astype(str).str.strip()
df["rank"] = pd.to_numeric(df["rank"], errors="coerce")

valid_labels = {
    "relevant", "partially relevant", "irrelevant"
}
unreviewed = df[~df["relevance"].isin(valid_labels)]

print("Rows:", len(df))
print("Questions:", df["question_id"].nunique())
print("Strategies:", sorted(df["strategy"].unique()))
print("Unreviewed/invalid labels:", len(unreviewed))

if len(unreviewed):
    print("\nFinish or correct these labels before final evaluation:")
    display(unreviewed[
        ["question_id", "strategy", "rank", "relevance"]
    ].head(30))
    raise ValueError("Review is incomplete; metrics not finalized.")

# Every question should have exactly ranks 1–5 per strategy.
counts = df.groupby(["question_id", "strategy"]).size()
if not (counts == 5).all():
    raise ValueError(
        "Some question/strategy groups do not contain exactly 5 passages."
    )

# Direct row-by-row labels: no approximate source matching.
def calculate_metrics(group):
    group = group.sort_values("rank")
    labels = group["relevance"].tolist()

    fully_relevant_ranks = [
        i + 1 for i, label in enumerate(labels)
        if label == "relevant"
    ]
    partial_or_better_ranks = [
        i + 1 for i, label in enumerate(labels)
        if label in {"relevant", "partially relevant"}
    ]

    first_full = min(fully_relevant_ranks) if fully_relevant_ranks else None
    first_any = min(partial_or_better_ranks) if partial_or_better_ranks else None

    # Graded DCG: Relevant=2, Partially relevant=1, Irrelevant=0.
    gains = [
        2 if label == "relevant"
        else 1 if label == "partially relevant"
        else 0
        for label in labels
    ]
    dcg = sum(
        gain / np.log2(rank + 1)
        for rank, gain in enumerate(gains, start=1)
    )
    ideal = sorted(gains, reverse=True)
    idcg = sum(
        gain / np.log2(rank + 1)
        for rank, gain in enumerate(ideal, start=1)
    )

    return pd.Series({
        "Relevant@5": int(first_full is not None),
        "MRR": 1 / first_full if first_full else 0.0,
        "AnyUseful@5": int(first_any is not None),
        "MRR_including_partial": 1 / first_any if first_any else 0.0,
        "nDCG@5": dcg / idcg if idcg else 0.0,
        "first_relevant_rank": first_full if first_full else "None"
    })

per_question = (
    df.groupby(["question_id", "question", "answerability", "strategy"],
               dropna=False)
      .apply(calculate_metrics, include_groups=False)
      .reset_index()
)

summary = (
    per_question.groupby("strategy")
    .agg(
        Questions=("question_id", "nunique"),
        Relevant_at_5=("Relevant@5", "mean"),
        MRR=("MRR", "mean"),
        Any_useful_at_5=("AnyUseful@5", "mean"),
        MRR_including_partial=("MRR_including_partial", "mean"),
        nDCG_at_5=("nDCG@5", "mean")
    )
    .reset_index()
)

# Separate answerable and unanswerable questions.
answerability_summary = (
    per_question.groupby(["answerability", "strategy"])
    .agg(
        Questions=("question_id", "nunique"),
        Relevant_at_5=("Relevant@5", "mean"),
        MRR=("MRR", "mean"),
        Any_useful_at_5=("AnyUseful@5", "mean"),
        nDCG_at_5=("nDCG@5", "mean")
    )
    .reset_index()
)

stamp = pd.Timestamp.now().strftime("%Y%m%d_%H%M%S")
OUT = os.path.join(ROOT, f"final_retrieval_metrics_{stamp}.xlsx")

with pd.ExcelWriter(OUT, engine="openpyxl") as writer:
    summary.to_excel(writer, sheet_name="Overall Metrics", index=False)
    answerability_summary.to_excel(
        writer, sheet_name="By Answerability", index=False
    )
    per_question.to_excel(
        writer, sheet_name="Per Question", index=False
    )

print("\n=== OVERALL METRICS ===")
print(summary.round(3).to_string(index=False))

print("\n=== BY ANSWERABILITY ===")
print(answerability_summary.round(3).to_string(index=False))

print("\nSaved metrics workbook:")
print(OUT)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import pickle
import numpy as np
import pandas as pd

# Install only if needed.
%pip -q install -U sentence-transformers rank-bm25 faiss-cpu

import faiss
from sentence_transformers import SentenceTransformer, CrossEncoder

BASE = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1"
)

CHUNKS_PATH = os.path.join(BASE, "english_only_chunks.csv")
BM25_PATH = os.path.join(
    BASE, "indexes", "bm25", "bm25_index.pkl"
)
TOKENS_PATH = os.path.join(
    BASE, "indexes", "bm25", "tokenized_documents.pkl"
)
EMBED_PATH = os.path.join(
    BASE, "indexes", "embeddings", "chunk_embeddings.npy"
)
FAISS_PATH = os.path.join(
    BASE, "indexes", "faiss", "index.faiss"
)

# Load existing artifacts.
chunks = pd.read_csv(CHUNKS_PATH)

with open(BM25_PATH, "rb") as f:
    bm25 = pickle.load(f)

with open(TOKENS_PATH, "rb") as f:
    tokenized_documents = pickle.load(f)

embeddings = np.load(EMBED_PATH)
faiss_index = faiss.read_index(FAISS_PATH)

print("Chunks:", len(chunks))
print("BM25 documents:", len(tokenized_documents))
print("Embeddings:", embeddings.shape)
print("FAISS vectors:", faiss_index.ntotal)

assert len(chunks) == len(tokenized_documents) == len(embeddings)
assert faiss_index.ntotal == len(chunks)
assert "text" in chunks.columns, (
    f"Expected a 'text' column. Found: {chunks.columns.tolist()}"
)

# Load the same model families used in the experiments.
embedder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")

def retrieve_passages(question, candidate_k=50, top_k=5):
    # BM25 candidates
    query_tokens = question.lower().split()
    bm25_scores = np.asarray(bm25.get_scores(query_tokens))
    bm25_ids = np.argsort(bm25_scores)[::-1][:candidate_k]

    # Dense candidates
    query_embedding = embedder.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")

    dense_scores, dense_ids = faiss_index.search(
        query_embedding, candidate_k
    )
    dense_ids = dense_ids[0]
    dense_ids = dense_ids[dense_ids >= 0]

    # Combine candidate IDs without duplicating passages.
    candidate_ids = list(dict.fromkeys(
        list(map(int, bm25_ids)) + list(map(int, dense_ids))
    ))

    # CrossEncoder reranking.
    pairs = [
        (question, str(chunks.iloc[i]["text"]))
        for i in candidate_ids
    ]
    scores = reranker.predict(pairs, show_progress_bar=False)
    order = np.argsort(scores)[::-1][:top_k]

    results = []
    for rank, pos in enumerate(order, start=1):
        i = candidate_ids[int(pos)]
        row = chunks.iloc[i]

        results.append({
            "rank": rank,
            "chunk_id": row.get("chunk_id", ""),
            "class": row.get("class", ""),
            "subject": row.get("subject", ""),
            "source_file": row.get("source_file", ""),
            "pages": row.get("pages", ""),
            "reranker_score": float(scores[int(pos)]),
            "passage": str(row["text"])
        })

    return pd.DataFrame(results)

# First smoke test — no answer generation yet.
test_question = "How can I tell which of two things is heavier?"
results = retrieve_passages(test_question)

for _, row in results.iterrows():
    print(f"\n{'=' * 65}")
    print(
        f"Rank {row['rank']} | "
        f"Class {row['class']} | {row['subject']}"
    )
    print("Source:", row["source_file"])
    print("Pages:", row["pages"])
    print("Reranker score:", round(row["reranker_score"], 4))
    print("Passage:", row["passage"][:1000])

print("\nRetrieval smoke test complete.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

NLI_MODEL_ID = "cross-encoder/nli-deberta-v3-small"

nli_tokenizer = AutoTokenizer.from_pretrained(NLI_MODEL_ID)
nli_model = AutoModelForSequenceClassification.from_pretrained(
    NLI_MODEL_ID
)
nli_model.eval()

if torch.cuda.is_available():
    nli_model = nli_model.to("cuda")

print("NLI labels:", nli_model.config.id2label)
print("NLI verifier loaded.")

def check_entailment(premise, hypothesis):
    """Return NLI label probabilities for evidence -> answer."""
    inputs = nli_tokenizer(
        premise,
        hypothesis,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    device = next(nli_model.parameters()).device
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.inference_mode():
        logits = nli_model(**inputs).logits[0]
        probs = torch.softmax(logits, dim=-1).cpu().tolist()

    labels = nli_model.config.id2label
    return {
        labels[i].lower(): round(float(probs[i]), 4)
        for i in range(len(probs))
    }

def verify_answer_against_evidence(result):
    # Combine only the passages actually retrieved for this question.
    evidence = result["evidence"]
    premise = "\n".join(
        str(text) for text in evidence["passage"].tolist()
    )

    answer = result["answer"].strip()

    # Split into simple sentence-level claims for this pilot.
    import re
    sentences = [
        s.strip() for s in re.split(r"(?<=[.!?])\s+", answer)
        if s.strip()
    ]

    fallback = (
        "I couldn't find enough information in our textbook "
        "to answer that yet. Let's ask your teacher!"
    )

    if not sentences:
        result["verified_answer"] = fallback
        result["verification"] = []
        result["verification_status"] = "ABSTAIN_EMPTY"
        return result

    checks = []
    for sentence in sentences:
        scores = check_entailment(premise, sentence)
        checks.append({"claim": sentence, "scores": scores})

    # Conservative pilot rule:
    # require every answer sentence to be entailed by the retrieved text.
    # NLI scores are not calibrated probabilities, so this is experimental.
    supported = all(
        max(
            (score for label, score in item["scores"].items()
             if "entail" in label),
            default=0.0
        ) >= 0.70
        for item in checks
    )

    result["verification"] = checks
    result["verified_answer"] = answer if supported else fallback
    result["verification_status"] = (
        "PASS_NLI_SCREEN" if supported else "ABSTAIN_NLI_SCREEN"
    )
    return result

# Re-test the failure case.
q = "Will it rain at my house tomorrow?"
raw_result = answer_from_ncert(q)
checked_result = verify_answer_against_evidence(raw_result)

print("QUESTION:", q)
print("\nRAW ANSWER:", raw_result["answer"])
print("\nVERIFICATION:")
for item in checked_result["verification"]:
    print(item)

print("\nFINAL TUTOR RESPONSE:")
print(checked_result["verified_answer"])
print("\nSTATUS:", checked_result["verification_status"])
```

```python

from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd

# Ensure the required functions/models exist from previous cells.
assert "retrieve_passages" in globals(), "Run the retrieval setup cell first."
assert "check_entailment" in globals(), "Run the NLI verifier setup cell first."
assert "answer_from_ncert" in globals(), "Run the answer-generation cell first."

FALLBACK = (
    "I couldn't find enough information in our textbook "
    "to answer that yet. Let's ask your teacher!"
)

TEST_CASES = [
    {
        "id": "GATE01",
        "type": "SUPPORTED",
        "question": "How can I tell which of two things is heavier?",
        "claim": (
            "A balance can be used to compare the weights of two objects."
        ),
        "expected": "SUPPORT"
    },
    {
        "id": "GATE02",
        "type": "UNANSWERABLE",
        "question": "Will it rain at my house tomorrow?",
        "claim": "It is unlikely to rain at your house tomorrow.",
        "expected": "ABSTAIN"
    },
    {
        "id": "GATE03",
        "type": "UNSUPPORTED_EXPLANATION",
        "question": "Why do houses in Sikkim have sloping roofs?",
        "claim": (
            "Houses in Sikkim have sloping roofs because this design "
            "helps rainwater flow off the roof."
        ),
        "expected": "ABSTAIN"
    }
]

rows = []

for case in TEST_CASES:
    print(f"\nTesting {case['id']}: {case['question']}")

    retrieved = retrieve_passages(case["question"], top_k=5)
    evidence = "\n".join(
        str(p) for p in retrieved["passage"].tolist()
    )

    scores = check_entailment(evidence, case["claim"])

    entailment = scores.get("entailment", 0.0)
    contradiction = scores.get("contradiction", 0.0)
    neutral = scores.get("neutral", 0.0)

    # Conservative pilot gate:
    # pass only when entailment is high AND exceeds both alternatives.
    passed = (
        entailment >= 0.70
        and entailment > contradiction
        and entailment > neutral
    )

    final_response = (
        case["claim"] if passed else FALLBACK
    )

    if case["expected"] == "SUPPORT":
        outcome = "PASS" if passed else "FALSE_ABSTENTION"
    else:
        outcome = "PASS" if not passed else "UNSUPPORTED_CLAIM_ACCEPTED"

    rows.append({
        "test_id": case["id"],
        "test_type": case["type"],
        "question": case["question"],
        "test_claim": case["claim"],
        "expected_behavior": case["expected"],
        "entailment_score": entailment,
        "contradiction_score": contradiction,
        "neutral_score": neutral,
        "gate_passed_claim": passed,
        "outcome": outcome,
        "final_response": final_response,
        "top_source": (
            retrieved.iloc[0]["source_file"]
            if len(retrieved) else ""
        ),
        "top_pages": (
            retrieved.iloc[0]["pages"]
            if len(retrieved) else ""
        )
    })

results_df = pd.DataFrame(rows)

print("\n" + "=" * 65)
print("NLI GATE PILOT RESULTS")
print("=" * 65)

for _, row in results_df.iterrows():
    print(f"\n{row['test_id']} | {row['test_type']}")
    print("Claim:", row["test_claim"])
    print(
        "Scores:",
        f"entailment={row['entailment_score']:.4f},",
        f"contradiction={row['contradiction_score']:.4f},",
        f"neutral={row['neutral_score']:.4f}"
    )
    print("Outcome:", row["outcome"])
    print("Response:", row["final_response"])

print("\nOutcome counts:")
print(results_df["outcome"].value_counts().to_string())

# Save a new report; do not overwrite existing evaluation artifacts.
ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
OUT_DIR = os.path.join(
    ROOT, "05_EXPERIMENTS", "retrieval_evaluation",
    "english_only_v1", "evaluation", "child_benchmark_v1"
)
os.makedirs(OUT_DIR, exist_ok=True)

OUT_PATH = os.path.join(OUT_DIR, "nli_gate_pilot_v1.xlsx")
results_df.to_excel(OUT_PATH, index=False)

print("\nSaved report:", OUT_PATH)
print(
    "\nReminder: 3 cases are only a smoke test. "
    "NLI scores are not calibrated probabilities."
)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import re
import pandas as pd
import os

FALLBACK = (
    "I couldn't find enough information in our textbook "
    "to answer that yet. Let's ask your teacher!"
)

def verify_answer_against_evidence_v2(result, threshold=0.70):
    evidence = result["evidence"]
    answer = result["answer"].strip()

    if evidence.empty or not answer:
        return {
            **result,
            "verified_answer": FALLBACK,
            "verification_status": "ABSTAIN_NO_EVIDENCE",
            "verification_details": []
        }

    # Treat the standard abstention as an acceptable response.
    if FALLBACK.lower() in answer.lower():
        return {
            **result,
            "verified_answer": answer,
            "verification_status": "ABSTAIN_REQUESTED",
            "verification_details": []
        }

    sentences = [
        s.strip() for s in re.split(r"(?<=[.!?])\s+", answer)
        if s.strip()
    ]

    details = []
    all_supported = True

    for sentence in sentences:
        passage_scores = []

        # Crucial change: assess each passage independently.
        for _, row in evidence.iterrows():
            scores = check_entailment(
                str(row["passage"]),
                sentence
            )
            passage_scores.append({
                "rank": int(row["rank"]),
                "source_file": row["source_file"],
                "pages": row["pages"],
                "scores": scores
            })

        best = max(
            passage_scores,
            key=lambda item: item["scores"].get("entailment", 0.0)
        )
        best_entailment = best["scores"].get("entailment", 0.0)

        supported = best_entailment >= threshold
        all_supported = all_supported and supported

        details.append({
            "claim": sentence,
            "supported": supported,
            "best_entailment": best_entailment,
            "best_source": best["source_file"],
            "best_pages": best["pages"],
            "all_passage_scores": passage_scores
        })

    final_answer = answer if all_supported else FALLBACK

    return {
        **result,
        "verified_answer": final_answer,
        "verification_status": (
            "PASS_INDIVIDUAL_PASSAGE_SCREEN"
            if all_supported else "ABSTAIN_INDIVIDUAL_PASSAGE_SCREEN"
        ),
        "verification_details": details
    }


# Run the same two end-to-end tests.
pilot_questions = [
    "How can I tell which of two things is heavier?",
    "Will it rain at my house tomorrow?",
    "Why do houses in Sikkim have sloping roofs?"
]

pilot_rows = []

for question in pilot_questions:
    raw = answer_from_ncert(question)
    checked = verify_answer_against_evidence_v2(raw)

    print("\n" + "=" * 65)
    print("QUESTION:", question)
    print("RAW ANSWER:", raw["answer"])
    print("FINAL RESPONSE:", checked["verified_answer"])
    print("STATUS:", checked["verification_status"])

    for detail in checked["verification_details"]:
        print(
            "\nClaim:", detail["claim"],
            "\nSupported:", detail["supported"],
            "\nBest entailment:", round(detail["best_entailment"], 4),
            "\nBest source:", detail["best_source"],
            "\nPages:", detail["best_pages"]
        )

    pilot_rows.append({
        "question": question,
        "raw_answer": raw["answer"],
        "verified_answer": checked["verified_answer"],
        "verification_status": checked["verification_status"],
        "verification_details": str(checked["verification_details"])
    })

# Save a separate report.
ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
OUT_DIR = os.path.join(
    ROOT, "05_EXPERIMENTS", "retrieval_evaluation",
    "english_only_v1", "evaluation", "child_benchmark_v1"
)
os.makedirs(OUT_DIR, exist_ok=True)

out_path = os.path.join(
    OUT_DIR, "nli_individual_passage_gate_v2.xlsx"
)
pd.DataFrame(pilot_rows).to_excel(out_path, index=False)

print("\nSaved:", out_path)
print(
    "\nReminder: this is still a pilot. A threshold of 0.70 is "
    "experimental and must be calibrated on labelled supported and "
    "unsupported claims before the gate is trusted."
)
```

```python

# ============================================================
# NCERT AI TUTOR — RESTORE + NLI DIAGNOSTIC
# Safe to rerun. Does not overwrite retrieval indexes.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

%pip -q install -U sentence-transformers rank-bm25 faiss-cpu transformers accelerate

import os
import pickle
import numpy as np
import pandas as pd
import torch
import faiss

from sentence_transformers import SentenceTransformer, CrossEncoder
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# ----------------------------
# 1. Paths
# ----------------------------
ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
BASE = os.path.join(
    ROOT, "05_EXPERIMENTS",
    "retrieval_evaluation", "english_only_v1"
)
EVAL = os.path.join(BASE, "evaluation", "child_benchmark_v1")
os.makedirs(EVAL, exist_ok=True)

# ----------------------------
# 2. Restore existing indexes
# ----------------------------
chunks = pd.read_csv(os.path.join(BASE, "english_only_chunks.csv"))

with open(os.path.join(BASE, "indexes", "bm25", "bm25_index.pkl"), "rb") as f:
    bm25 = pickle.load(f)

with open(os.path.join(BASE, "indexes", "bm25", "tokenized_documents.pkl"), "rb") as f:
    tokenized_documents = pickle.load(f)

embeddings = np.load(
    os.path.join(BASE, "indexes", "embeddings", "chunk_embeddings.npy")
)
faiss_index = faiss.read_index(
    os.path.join(BASE, "indexes", "faiss", "index.faiss")
)

assert len(chunks) == len(tokenized_documents) == len(embeddings)
assert faiss_index.ntotal == len(chunks)
assert "text" in chunks.columns

# ----------------------------
# 3. Restore models
# ----------------------------
embedder = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2"
)

NLI_MODEL_ID = "cross-encoder/nli-deberta-v3-small"
nli_tokenizer = AutoTokenizer.from_pretrained(NLI_MODEL_ID)
nli_model = AutoModelForSequenceClassification.from_pretrained(
    NLI_MODEL_ID
)
if torch.cuda.is_available():
    nli_model = nli_model.to("cuda")
nli_model.eval()

print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU")
print("Chunks:", len(chunks), "| FAISS vectors:", faiss_index.ntotal)
print("NLI labels:", nli_model.config.id2label)

# ----------------------------
# 4. Retrieval function
# ----------------------------
def retrieve_passages(question, candidate_k=50, top_k=5):
    bm25_scores = np.asarray(bm25.get_scores(question.lower().split()))
    bm25_ids = np.argsort(bm25_scores)[::-1][:candidate_k]

    query_vec = embedder.encode(
        [question],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")

    _, dense_ids = faiss_index.search(query_vec, candidate_k)
    dense_ids = [int(i) for i in dense_ids[0] if i >= 0]

    candidate_ids = list(dict.fromkeys(
        list(map(int, bm25_ids)) + dense_ids
    ))

    pairs = [
        (question, str(chunks.iloc[i]["text"]))
        for i in candidate_ids
    ]
    scores = reranker.predict(pairs, show_progress_bar=False)
    order = np.argsort(scores)[::-1][:top_k]

    output = []
    for rank, pos in enumerate(order, start=1):
        i = candidate_ids[int(pos)]
        row = chunks.iloc[i]
        output.append({
            "rank": rank,
            "chunk_id": row.get("chunk_id", ""),
            "class": row.get("class", ""),
            "subject": row.get("subject", ""),
            "source_file": row.get("source_file", ""),
            "pages": row.get("pages", ""),
            "reranker_score": float(scores[int(pos)]),
            "passage": str(row["text"])
        })

    return pd.DataFrame(output)

# ----------------------------
# 5. NLI function
# ----------------------------
def check_entailment(premise, hypothesis):
    inputs = nli_tokenizer(
        premise,
        hypothesis,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )
    device = next(nli_model.parameters()).device
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.inference_mode():
        probs = torch.softmax(nli_model(**inputs).logits[0], dim=-1)
    probs = probs.cpu().tolist()
    labels = nli_model.config.id2label

    return {
        labels[i].lower(): float(probs[i])
        for i in range(len(probs))
    }

# ----------------------------
# 6. Six diagnostic claims
# ----------------------------
TESTS = [
    ("C1", "How can I tell which of two things is heavier?",
     "The textbook says to compare two objects to see which is heavier.",
     "SUPPORTED"),
    ("C2", "How can I tell which of two things is heavier?",
     "The textbook describes making a balance using a paper cup, thread and a hanger.",
     "SUPPORTED"),
    ("C3", "How can I tell which of two things is heavier?",
     "A pumpkin is definitely heavier than a watermelon.",
     "UNSUPPORTED"),
    ("C4", "Will it rain at my house tomorrow?",
     "It is unlikely to rain at the child's house tomorrow.",
     "UNSUPPORTED"),
    ("C5", "Why do houses in Sikkim have sloping roofs?",
     "Sloping roofs in Sikkim are built because unstable soil makes flat roofs impractical.",
     "UNSUPPORTED"),
    ("C6", "Why do houses in Sikkim have sloping roofs?",
     "The textbook mentions traditional houses in Sikkim with sloping roofs.",
     "SUPPORTED_TOPIC_ONLY"),
]

rows = []

for test_id, question, claim, expected in TESTS:
    passages = retrieve_passages(question, top_k=5)

    print("\n" + "=" * 70)
    print(test_id, "|", expected)
    print("Claim:", claim)

    for _, passage in passages.iterrows():
        scores = check_entailment(str(passage["passage"]), claim)

        row = {
            "test_id": test_id,
            "expected": expected,
            "question": question,
            "claim": claim,
            "rank": int(passage["rank"]),
            "source_file": passage["source_file"],
            "pages": passage["pages"],
            "entailment": scores.get("entailment", 0.0),
            "contradiction": scores.get("contradiction", 0.0),
            "neutral": scores.get("neutral", 0.0),
            "passage_excerpt": str(passage["passage"])[:700]
        }
        rows.append(row)

        print(
            f"Rank {row['rank']} | "
            f"E={row['entailment']:.4f} "
            f"C={row['contradiction']:.4f} "
            f"N={row['neutral']:.4f}"
        )
        print("Source:", row["source_file"], "| pages:", row["pages"])

# ----------------------------
# 7. Report and save
# ----------------------------
df = pd.DataFrame(rows)
best = df.loc[df.groupby("test_id")["entailment"].idxmax()]

print("\n" + "=" * 70)
print("BEST PASSAGE PER CLAIM")
print(best[
    ["test_id", "expected", "rank", "entailment",
     "contradiction", "neutral", "source_file", "pages"]
].to_string(index=False))

out_path = os.path.join(EVAL, "nli_all_passages_calibration_v2.xlsx")
df.to_excel(out_path, index=False)

print("\nSaved:", out_path)
print("Rows:", len(df), "(expected 30)")
print("Diagnostic only — no threshold has been changed.")
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import re
import pandas as pd
import torch

# Reuse the existing retrieve_passages(), nli_tokenizer, and nli_model.

examples = [
    (
        "C1",
        "How can I tell which of two things is heavier?",
        "The textbook says to compare two objects to see which is heavier."
    ),
    (
        "C2",
        "How can I tell which of two things is heavier?",
        "The textbook describes making a balance using a paper cup, thread and a hanger."
    ),
    (
        "C6",
        "Why do houses in Sikkim have sloping roofs?",
        "The textbook mentions traditional houses in Sikkim with sloping roofs."
    ),
]

def score_nli(premise, hypothesis):
    inputs = nli_tokenizer(
        premise,
        hypothesis,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )
    device = next(nli_model.parameters()).device
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        logits = nli_model(**inputs).logits
        probs = torch.softmax(logits, dim=-1)[0].cpu().tolist()

    # Confirmed for this model: 0 contradiction, 1 entailment, 2 neutral
    return {
        "contradiction": probs[0],
        "entailment": probs[1],
        "neutral": probs[2],
        "input_tokens": int(inputs["input_ids"].shape[1])
    }

def make_windows(text, window_words=100, overlap_words=30):
    words = str(text).split()
    if not words:
        return []

    step = window_words - overlap_words
    windows = []

    for start in range(0, len(words), step):
        snippet = " ".join(words[start:start + window_words])
        if snippet.strip():
            windows.append({
                "start_word": start,
                "end_word": min(start + window_words, len(words)),
                "snippet": snippet
            })
        if start + window_words >= len(words):
            break

    return windows

rows = []

for case_id, question, claim in examples:
    passages = retrieve_passages(question, top_k=5)

    for _, passage in passages.iterrows():
        full_text = str(passage["passage"])
        full_score = score_nli(full_text, claim)

        rows.append({
            "case_id": case_id,
            "claim": claim,
            "rank": int(passage["rank"]),
            "source_file": passage["source_file"],
            "pages": passage["pages"],
            "test_type": "FULL_PASSAGE",
            "start_word": 0,
            "end_word": len(full_text.split()),
            **full_score,
            "snippet": full_text[:1200]
        })

        for window in make_windows(full_text):
            window_score = score_nli(window["snippet"], claim)

            rows.append({
                "case_id": case_id,
                "claim": claim,
                "rank": int(passage["rank"]),
                "source_file": passage["source_file"],
                "pages": passage["pages"],
                "test_type": "SHORT_WINDOW",
                "start_word": window["start_word"],
                "end_word": window["end_word"],
                **window_score,
                "snippet": window["snippet"]
            })

results = pd.DataFrame(rows)

OUT_DIR = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/evaluation/child_benchmark_v1"
)
os.makedirs(OUT_DIR, exist_ok=True)

out_path = os.path.join(
    OUT_DIR, "nli_windowed_evidence_diagnostic_v1.xlsx"
)
results.to_excel(out_path, index=False)

print("Saved:", out_path)
print("\nBest short-window evidence score per claim:")

short_rows = results[results["test_type"] == "SHORT_WINDOW"]

for case_id, _, claim in examples:
    subset = short_rows[short_rows["case_id"] == case_id]

    if subset.empty:
        print(f"\n{case_id}: No windows available")
        continue

    best = subset.sort_values("entailment", ascending=False).iloc[0]

    print(f"\n{case_id}: {claim}")
    print(
        f"Best entailment={best['entailment']:.4f} | "
        f"neutral={best['neutral']:.4f} | "
        f"contradiction={best['contradiction']:.4f}"
    )
    print(
        f"Rank={best['rank']} | "
        f"Words={best['start_word']}-{best['end_word']} | "
        f"Tokens={best['input_tokens']}"
    )
    print("Evidence snippet:", best["snippet"][:500])
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
import torch

OUT_DIR = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/evaluation/child_benchmark_v1"
)

input_path = os.path.join(
    OUT_DIR, "nli_windowed_evidence_diagnostic_v1.xlsx"
)

df = pd.read_excel(input_path)
short = df[df["test_type"] == "SHORT_WINDOW"].copy()

tests = [
    ("C1", "The passage says to compare two objects to find which is heavier."),
    ("C2", "The passage says to make a balance using a paper cup, thread and a hanger."),
    ("C2", "A paper cup, thread and a hanger are used to make a balance."),
    ("C2", "The passage mentions a paper cup, thread and a hanger."),
    ("C6", "The passage says that traditional houses in Sikkim have slanting roofs."),
    ("C6", "Traditional houses in Sikkim are built with wood and have slanting roofs."),
    ("C6", "The passage mentions wooden houses with sloping roofs."),
]

results = []

for case_id, claim in tests:
    candidates = short[short["case_id"] == case_id]

    # Score every short window against this exact claim.
    for _, row in candidates.iterrows():
        premise = str(row["snippet"])

        inputs = nli_tokenizer(
            premise,
            claim,
            return_tensors="pt",
            truncation=True,
            max_length=512
        )
        device = next(nli_model.parameters()).device
        inputs = {k: v.to(device) for k, v in inputs.items()}

        with torch.no_grad():
            logits = nli_model(**inputs).logits
            probs = torch.softmax(logits, dim=-1)[0].cpu().tolist()

        results.append({
            "case_id": case_id,
            "claim": claim,
            "rank": row["rank"],
            "source_file": row["source_file"],
            "pages": row["pages"],
            "entailment": probs[1],
            "neutral": probs[2],
            "contradiction": probs[0],
            "snippet": premise
        })

test_df = pd.DataFrame(results)
test_out = os.path.join(OUT_DIR, "nli_claim_wording_test_v1.xlsx")
test_df.to_excel(test_out, index=False)

for (case_id, claim), group in test_df.groupby(["case_id", "claim"], sort=False):
    best = group.sort_values("entailment", ascending=False).iloc[0]

    print("\n" + "=" * 75)
    print(case_id, "|", claim)
    print(
        f"Best entailment: {best['entailment']:.4f} | "
        f"neutral: {best['neutral']:.4f} | "
        f"contradiction: {best['contradiction']:.4f}"
    )
    print("Rank:", best["rank"], "| Pages:", best["pages"])
    print("Evidence:", best["snippet"][:350])

print("\nSaved:", test_out)
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
import torch
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter

# ============================================================
# NCERT TUTOR — NLI CALIBRATION BENCHMARK V1
# Uses the already-loaded nli_tokenizer and nli_model.
# Does not overwrite any previous experiment.
# ============================================================

OUT_DIR = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/evaluation/child_benchmark_v1"
)
os.makedirs(OUT_DIR, exist_ok=True)

WEIGHT_EVIDENCE = (
    "Make your own balance with the help of a paper cup, thread "
    "and a hanger like the one shown here. Now compare any two "
    "objects like eraser, ball, bead, pencil, crayon, etc., to see "
    "which one is heavier. Let us Do Circle the heaviest one."
)

SIKKIM_EVIDENCE = (
    "Why do you think traditional houses in Sikkim are built with "
    "wood and have slanting roofs? ... Housing: Wooden houses with "
    "sloping roofs."
)

# Gold labels describe the relationship between the claim and
# the supplied evidence — NOT whether the claim is true in general.
# These labels are initial annotations and should be human-reviewed.
examples = []

def add(category, topic, claim, evidence, note):
    examples.append({
        "category": category,
        "topic": topic,
        "claim": claim,
        "evidence": evidence,
        "gold_label": category,
        "gold_label_review": "PENDING_HUMAN_REVIEW",
        "labeling_note": note
    })

# ---------------- DIRECT SUPPORT: 8 ----------------
direct = [
    ("weight", "The passage says to compare objects to find which is heavier.",
     WEIGHT_EVIDENCE, "Close wording match to explicit comparison instruction."),
    ("weight", "The passage mentions a paper cup, thread and a hanger.",
     WEIGHT_EVIDENCE, "All three objects are explicitly named."),
    ("weight", "The passage tells the reader to make a balance.",
     WEIGHT_EVIDENCE, "The passage explicitly says to make a balance."),
    ("weight", "The passage says to compare objects such as an eraser and a pencil.",
     WEIGHT_EVIDENCE, "Both objects appear in the examples."),
    ("sikkim", "The passage mentions wooden houses with sloping roofs.",
     SIKKIM_EVIDENCE, "Close wording match to the table entry."),
    ("sikkim", "The passage says traditional houses in Sikkim have slanting roofs.",
     SIKKIM_EVIDENCE, "The question explicitly mentions Sikkim and slanting roofs."),
    ("sikkim", "The passage mentions wood as a material used in traditional houses in Sikkim.",
     SIKKIM_EVIDENCE, "The passage explicitly says houses are built with wood."),
    ("sikkim", "The passage includes a description of housing in different regions.",
     SIKKIM_EVIDENCE, "The excerpt contains a housing category and regional comparison.")
]
for topic, claim, evidence, note in direct:
    add("DIRECT_SUPPORT", topic, claim, evidence, note)

# ---------------- PARAPHRASED SUPPORT: 8 ----------------
paraphrases = [
    ("weight", "The textbook suggests weighing two items against each other to find the heavier one.",
     WEIGHT_EVIDENCE, "Paraphrases the comparison instruction; does not name a specific winner."),
    ("weight", "A homemade balance can help a child compare the weight of two things.",
     WEIGHT_EVIDENCE, "Paraphrases making and using a balance."),
    ("weight", "A child can use a balance made from a cup, string and a hanger.",
     WEIGHT_EVIDENCE, "String is a paraphrase of thread; wording differs."),
    ("weight", "The passage asks children to identify which object is heavier.",
     WEIGHT_EVIDENCE, "Paraphrases the stated learning task."),
    ("sikkim", "Traditional homes in Sikkim are described as having sloping roofs.",
     SIKKIM_EVIDENCE, "Equivalent wording for slanting/sloping roofs."),
    ("sikkim", "Wood is mentioned as a building material for traditional Sikkim houses.",
     SIKKIM_EVIDENCE, "Paraphrase of the explicit statement."),
    ("sikkim", "The passage connects Sikkim's traditional housing with wooden construction.",
     SIKKIM_EVIDENCE, "Meaning-preserving paraphrase."),
    ("sikkim", "The housing comparison lists wooden houses with sloping roofs.",
     SIKKIM_EVIDENCE, "Paraphrase of the table entry.")
]
for topic, claim, evidence, note in paraphrases:
    add("PARAPHRASED_SUPPORT", topic, claim, evidence, note)

# ---------------- PARTIAL SUPPORT: 8 ----------------
partial = [
    ("weight", "The passage tells us which of a pumpkin and a watermelon is heavier.",
     WEIGHT_EVIDENCE, "The wider textbook passage poses this as a discussion question; this excerpt does not state the winner."),
    ("weight", "The passage gives the exact weight of a pencil in grams.",
     WEIGHT_EVIDENCE, "A pencil is mentioned, but no measurement is provided."),
    ("weight", "The passage proves that every large object is heavier than every small object.",
     WEIGHT_EVIDENCE, "It discusses comparing objects but does not establish this general rule."),
    ("weight", "The passage gives a complete list of all tools needed to build a balance.",
     WEIGHT_EVIDENCE, "Some materials are named; completeness is not established by this excerpt."),
    ("sikkim", "The passage explains why Sikkim's traditional houses have sloping roofs.",
     SIKKIM_EVIDENCE, "It raises the question and states the feature, but this evidence does not explain why."),
    ("sikkim", "The passage proves that all houses in Sikkim are made entirely of wood.",
     SIKKIM_EVIDENCE, "It describes traditional houses; the universal and entirely-wood claim goes beyond it."),
    ("sikkim", "The passage says sloping roofs prevent every landslide.",
     SIKKIM_EVIDENCE, "Sloping roofs are mentioned, but this causal claim is not supported."),
    ("sikkim", "The passage gives measurements for the angle of Sikkim's roofs.",
     SIKKIM_EVIDENCE, "Roof shape is mentioned, but no angle measurement appears in this evidence.")
]
for topic, claim, evidence, note in partial:
    add("PARTIAL_SUPPORT", topic, claim, evidence, note)

# ---------------- CONTRADICTION: 8 ----------------
contradictions = [
    ("weight", "The passage tells children not to compare two objects to find which is heavier.",
     WEIGHT_EVIDENCE, "Directly conflicts with the instruction to compare objects."),
    ("weight", "The passage says a balance must never be made with a paper cup.",
     WEIGHT_EVIDENCE, "Conflicts with the instruction to make a balance using a paper cup."),
    ("weight", "The passage says a hanger is not used when making the balance.",
     WEIGHT_EVIDENCE, "Conflicts with the named hanger material."),
    ("weight", "The passage instructs children to circle the lightest object only and never identify the heaviest.",
     WEIGHT_EVIDENCE, "The passage includes an instruction to circle the heaviest."),
    ("sikkim", "The passage says traditional houses in Sikkim have flat roofs, not slanting roofs.",
     SIKKIM_EVIDENCE, "Conflicts with the explicit slanting-roof statement."),
    ("sikkim", "The passage says traditional Sikkim houses are built without wood.",
     SIKKIM_EVIDENCE, "Conflicts with the statement that they are built with wood."),
    ("sikkim", "The passage says wooden houses with sloping roofs are listed under the desert region.",
     SIKKIM_EVIDENCE, "The supplied table entry associates that housing description with the mountains column; verify table layout during review."),
    ("sikkim", "The passage explicitly states that traditional houses in Sikkim do not have slanting roofs.",
     SIKKIM_EVIDENCE, "Direct negation of the statement in the question.")
]
for topic, claim, evidence, note in contradictions:
    add("CONTRADICTION", topic, claim, evidence, note)

# ---------------- ABSENT FROM EVIDENCE: 8 ----------------
absent = [
    ("weight", "The passage predicts whether it will rain at a child's house tomorrow.",
     WEIGHT_EVIDENCE, "Weather forecasting is absent from this evidence."),
    ("weight", "The passage tells us what a child's teacher is thinking right now.",
     WEIGHT_EVIDENCE, "A person's private thoughts are not covered."),
    ("weight", "The passage gives the current price of a paper cup in Mumbai.",
     WEIGHT_EVIDENCE, "No price or local shopping information is supplied."),
    ("weight", "The passage identifies every bird outside the child's home.",
     WEIGHT_EVIDENCE, "Bird identification is absent."),
    ("sikkim", "The passage tells us the exact date of tomorrow's rainfall in Sikkim.",
     SIKKIM_EVIDENCE, "No weather forecast or future date is supplied."),
    ("sikkim", "The passage names the engineer who designed every traditional house in Sikkim.",
     SIKKIM_EVIDENCE, "No architect or engineer is named."),
    ("sikkim", "The passage states the exact cost of building a wooden house in Sikkim today.",
     SIKKIM_EVIDENCE, "No construction cost is supplied."),
    ("sikkim", "The passage explains why a particular dog barks at the postman.",
     SIKKIM_EVIDENCE, "Animal behaviour is absent from this evidence.")
]
for topic, claim, evidence, note in absent:
    add("ABSENT_EVIDENCE", topic, claim, evidence, note)

benchmark = pd.DataFrame(examples)
assert len(benchmark) == 40, f"Expected 40 rows, got {len(benchmark)}"
assert benchmark["claim"].is_unique, "Duplicate claims found"

print("Benchmark rows:", len(benchmark))
print("\nGold-label counts:")
print(benchmark["gold_label"].value_counts().to_string())

# ---------------- RUN NLI MODEL ----------------
if "nli_model" not in globals() or "nli_tokenizer" not in globals():
    raise RuntimeError(
        "nli_model or nli_tokenizer is not loaded. "
        "Load cross-encoder/nli-deberta-v3-small first."
    )

device = next(nli_model.parameters()).device
nli_model.eval()

def score_claim(evidence, claim):
    inputs = nli_tokenizer(
        evidence,
        claim,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        logits = nli_model(**inputs).logits
        probs = torch.softmax(logits, dim=-1)[0].cpu().tolist()

    # Model mapping previously checked:
    # 0 = contradiction, 1 = entailment, 2 = neutral
    return {
        "nli_contradiction": float(probs[0]),
        "nli_entailment": float(probs[1]),
        "nli_neutral": float(probs[2]),
        "input_tokens": int(inputs["input_ids"].shape[1]),
        "predicted_label": [
            "CONTRADICTION", "ENTAILMENT", "NEUTRAL"
        ][int(torch.tensor(probs).argmax().item())]
    }

scored = []
for i, row in benchmark.iterrows():
    scores = score_claim(row["evidence"], row["claim"])
    scored.append({**row.to_dict(), **scores})

results = pd.DataFrame(scored)

# Save fresh, versioned outputs; preserve prior experiments.
xlsx_path = os.path.join(OUT_DIR, "nli_calibration_benchmark_v1.xlsx")
csv_path = os.path.join(OUT_DIR, "nli_calibration_benchmark_v1.csv")

with pd.ExcelWriter(xlsx_path, engine="openpyxl") as writer:
    results.to_excel(writer, sheet_name="Calibration", index=False)

    # Reviewer-facing sheet with labels separate from predictions.
    review_cols = [
        "category", "topic", "claim", "evidence",
        "gold_label", "gold_label_review", "labeling_note",
        "predicted_label", "nli_contradiction",
        "nli_entailment", "nli_neutral", "input_tokens"
    ]
    results[review_cols].to_excel(
        writer, sheet_name="Human_Review", index=False
    )

results.to_csv(csv_path, index=False)

# Basic formatting for easier manual review.
wb = load_workbook(xlsx_path)
for ws in wb.worksheets:
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions
    for cell in ws[1]:
        cell.font = Font(bold=True, color="FFFFFF")
        cell.fill = PatternFill("solid", fgColor="244062")
        cell.alignment = Alignment(wrap_text=True, vertical="top")

    for column_cells in ws.columns:
        col_letter = get_column_letter(column_cells[0].column)
        header = str(column_cells[0].value)
        if header in ("evidence", "claim", "labeling_note"):
            ws.column_dimensions[col_letter].width = 55 if header == "evidence" else 42
        elif header in ("gold_label_review",):
            ws.column_dimensions[col_letter].width = 25
        else:
            ws.column_dimensions[col_letter].width = min(
                max(len(header) + 2, 14), 26
            )

    for row in ws.iter_rows(min_row=2):
        for cell in row:
            cell.alignment = Alignment(wrap_text=True, vertical="top")

wb.save(xlsx_path)

print("\nSaved benchmark Excel:", xlsx_path)
print("Saved benchmark CSV:", csv_path)

print("\nMean NLI scores by gold category:")
summary = results.groupby("gold_label")[
    ["nli_contradiction", "nli_entailment", "nli_neutral"]
].mean().round(3)
print(summary.to_string())

print("\nPrediction vs initial gold label:")
print(pd.crosstab(
    results["gold_label"],
    results["predicted_label"],
    rownames=["Gold category"],
    colnames=["NLI prediction"]
).to_string())

print(
    "\nIMPORTANT: Initial gold labels are provisional. "
    "Review Human_Review before treating any metric as final."
)
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter

# ============================================================
# NLI CALIBRATION — EVALUATION AGAINST HUMAN-REVIEWED LABELS
# ============================================================

OUT_DIR = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/evaluation/child_benchmark_v1"
)

INPUT_PATH = os.path.join(
    OUT_DIR, "nli_calibration_human_review_corrected.xlsx"
)
OUTPUT_PATH = os.path.join(
    OUT_DIR, "nli_calibration_evaluation_v1.xlsx"
)

if not os.path.exists(INPUT_PATH):
    raise FileNotFoundError(
        f"Reviewed workbook not found:\n{INPUT_PATH}\n"
        "Check that the uploaded filename matches exactly."
    )

df = pd.read_excel(INPUT_PATH, sheet_name="Claims Review")

print("Loaded:", INPUT_PATH)
print("Rows:", len(df))
print("Columns:", df.columns.tolist())

# ------------------------------------------------------------
# Validate required columns
# ------------------------------------------------------------
required = [
    "claim",
    "evidence",
    "reviewed_gold_label",
    "predicted_label",
    "nli_contradiction",
    "nli_entailment",
    "nli_neutral",
    "review_status"
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

df["reviewed_gold_label"] = (
    df["reviewed_gold_label"].astype(str).str.strip().str.upper()
)
df["predicted_label"] = (
    df["predicted_label"].astype(str).str.strip().str.upper()
)
df["review_status"] = (
    df["review_status"].astype(str).str.strip().str.upper()
)

allowed_gold = {
    "SUPPORTED",
    "PARTIALLY_SUPPORTED",
    "CONTRADICTED",
    "NOT_ENOUGH_INFORMATION",
    "AMBIGUOUS"
}

invalid_labels = sorted(
    set(df["reviewed_gold_label"]) - allowed_gold
)
if invalid_labels:
    raise ValueError(
        f"Invalid or blank reviewed labels: {invalid_labels}"
    )

if df["claim"].isna().any() or df["evidence"].isna().any():
    raise ValueError("Some claims or evidence passages are missing.")

if df["claim"].duplicated().any():
    print("WARNING: Duplicate claim text found.")

print("\nReviewed-label distribution:")
print(df["reviewed_gold_label"].value_counts().to_string())

print("\nReview-status distribution:")
print(df["review_status"].value_counts().to_string())

# ------------------------------------------------------------
# Map NLI predictions into the human-review categories.
# The NLI model only predicts contradiction/entailment/neutral.
# Neutral is NOT automatically the same as insufficient evidence.
# ------------------------------------------------------------
def map_prediction(label):
    if label == "ENTAILMENT":
        return "SUPPORTED"
    if label == "CONTRADICTION":
        return "CONTRADICTED"
    return "NEUTRAL_UNRESOLVED"

df["mapped_nli_prediction"] = df["predicted_label"].map(map_prediction)

# Score only clear gold categories for the primary binary test.
# Partial and ambiguous examples are reported separately.
clear_mask = df["reviewed_gold_label"].isin([
    "SUPPORTED",
    "CONTRADICTED",
    "NOT_ENOUGH_INFORMATION"
])
clear_df = df[clear_mask].copy()

# Primary test:
# SUPPORTED = positive class
# CONTRADICTED or NOT_ENOUGH_INFORMATION = not supported
# Important: this tests claim support, not the distinction between
# contradiction and absent evidence.
clear_df["gold_supported"] = (
    clear_df["reviewed_gold_label"] == "SUPPORTED"
)
clear_df["pred_supported"] = (
    clear_df["predicted_label"] == "ENTAILMENT"
)

tp = int((clear_df["gold_supported"] & clear_df["pred_supported"]).sum())
fn = int((clear_df["gold_supported"] & ~clear_df["pred_supported"]).sum())
fp = int((~clear_df["gold_supported"] & clear_df["pred_supported"]).sum())
tn = int((~clear_df["gold_supported"] & ~clear_df["pred_supported"]).sum())

def safe_div(a, b):
    return a / b if b else np.nan

binary_metrics = pd.DataFrame([
    {"metric": "Evaluated clear-label claims", "value": len(clear_df)},
    {"metric": "True positives (supported predicted supported)", "value": tp},
    {"metric": "False negatives (supported not predicted supported)", "value": fn},
    {"metric": "False positives (unsupported predicted supported)", "value": fp},
    {"metric": "True negatives (unsupported not predicted supported)", "value": tn},
    {"metric": "Precision", "value": safe_div(tp, tp + fp)},
    {"metric": "Recall / sensitivity", "value": safe_div(tp, tp + fn)},
    {"metric": "Specificity", "value": safe_div(tn, tn + fp)},
    {"metric": "False acceptance rate", "value": safe_div(fp, fp + tn)},
    {"metric": "False rejection rate", "value": safe_div(fn, fn + tp)},
])

# ------------------------------------------------------------
# NLI's 3-way prediction vs the human labels.
# Partial and insufficient labels are retained for inspection.
# ------------------------------------------------------------
three_way = pd.crosstab(
    df["reviewed_gold_label"],
    df["predicted_label"],
    margins=True
).reset_index()

# Category-level diagnostic summary.
category_summary = (
    df.groupby("reviewed_gold_label")
      .agg(
          count=("claim", "size"),
          mean_entailment=("nli_entailment", "mean"),
          mean_contradiction=("nli_contradiction", "mean"),
          mean_neutral=("nli_neutral", "mean"),
          predicted_entailment=("predicted_label",
                                lambda s: (s == "ENTAILMENT").sum()),
          predicted_contradiction=("predicted_label",
                                   lambda s: (s == "CONTRADICTION").sum()),
          predicted_neutral=("predicted_label",
                             lambda s: (s == "NEUTRAL").sum())
      )
      .reset_index()
      .round(4)
)

# Claim-level error review.
def error_type(row):
    gold = row["reviewed_gold_label"]
    pred = row["predicted_label"]

    if gold in ("PARTIALLY_SUPPORTED", "AMBIGUOUS"):
        return "REVIEW_SEPARATELY"
    if gold == "SUPPORTED" and pred != "ENTAILMENT":
        return "POSSIBLE_FALSE_REJECTION"
    if gold != "SUPPORTED" and pred == "ENTAILMENT":
        return "POSSIBLE_FALSE_ACCEPTANCE"
    if gold == "NOT_ENOUGH_INFORMATION" and pred == "CONTRADICTION":
        return "ABSENCE_MISTAKEN_FOR_CONTRADICTION"
    if gold == "CONTRADICTED" and pred != "CONTRADICTION":
        return "MISSED_CONTRADICTION"
    return "NO_ERROR_UNDER_SUPPORT_TEST"

df["diagnostic_result"] = df.apply(error_type, axis=1)

# Save outputs to a NEW workbook.
with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    binary_metrics.to_excel(writer, sheet_name="Binary_Metrics", index=False)
    three_way.to_excel(writer, sheet_name="Confusion_Matrix", index=False)
    category_summary.to_excel(writer, sheet_name="Category_Summary", index=False)
    df.to_excel(writer, sheet_name="Claim_Level_Results", index=False)

# Format workbook for readability.
wb = load_workbook(OUTPUT_PATH)
for ws in wb.worksheets:
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions

    for cell in ws[1]:
        cell.font = Font(bold=True, color="FFFFFF")
        cell.fill = PatternFill("solid", fgColor="244062")
        cell.alignment = Alignment(wrap_text=True, vertical="top")

    for col_cells in ws.columns:
        letter = get_column_letter(col_cells[0].column)
        header = str(col_cells[0].value)
        if header in ("evidence", "claim", "labeling_note", "review_notes"):
            ws.column_dimensions[letter].width = 55
        else:
            ws.column_dimensions[letter].width = min(
                max(len(header) + 2, 14), 30
            )

    for row in ws.iter_rows(min_row=2):
        for cell in row:
            cell.alignment = Alignment(wrap_text=True, vertical="top")

wb.save(OUTPUT_PATH)

print("\n" + "=" * 65)
print("PRIMARY BINARY SUPPORT TEST")
print("=" * 65)
print(binary_metrics.to_string(index=False))

print("\n" + "=" * 65)
print("THREE-WAY NLI PREDICTIONS BY HUMAN LABEL")
print("=" * 65)
print(three_way.to_string(index=False))

print("\n" + "=" * 65)
print("CATEGORY SUMMARY")
print("=" * 65)
print(category_summary.to_string(index=False))

print("\n" + "=" * 65)
print("DIAGNOSTIC FLAGS")
print("=" * 65)
print(df["diagnostic_result"].value_counts().to_string())

print("\nSaved evaluation workbook:")
print(OUTPUT_PATH)

print(
    "\nNOTE: This is a small, hand-constructed diagnostic set. "
    "Do not select a production threshold from these 40 examples alone."
)
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
import pandas as pd

OUT_DIR = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/evaluation/child_benchmark_v1"
)

INPUT_PATH = os.path.join(
    OUT_DIR, "nli_calibration_human_review_corrected.xlsx"
)

df = pd.read_excel(INPUT_PATH, sheet_name="Claims Review")

df["reviewed_gold_label"] = (
    df["reviewed_gold_label"].astype(str).str.strip().str.upper()
)

# Only clear support labels for the primary threshold experiment.
df = df[df["reviewed_gold_label"].isin([
    "SUPPORTED", "CONTRADICTED", "NOT_ENOUGH_INFORMATION"
])].copy()

df["gold_supported"] = df["reviewed_gold_label"].eq("SUPPORTED")
scores = df["nli_entailment"].astype(float).to_numpy()
gold = df["gold_supported"].to_numpy()

rows = []

for threshold in np.round(np.arange(0.10, 1.00, 0.05), 2):
    pred = scores >= threshold

    tp = int(np.sum(pred & gold))
    fp = int(np.sum(pred & ~gold))
    fn = int(np.sum(~pred & gold))
    tn = int(np.sum(~pred & ~gold))

    precision = tp / (tp + fp) if tp + fp else np.nan
    recall = tp / (tp + fn) if tp + fn else np.nan
    specificity = tn / (tn + fp) if tn + fp else np.nan

    rows.append({
        "threshold": threshold,
        "true_supported": tp,
        "false_acceptances": fp,
        "supported_claims_rejected": fn,
        "true_unsupported_rejections": tn,
        "precision": precision,
        "recall": recall,
        "specificity": specificity
    })

threshold_df = pd.DataFrame(rows)

out_path = os.path.join(
    OUT_DIR, "nli_threshold_sweep_diagnostic_v1.xlsx"
)
threshold_df.to_excel(out_path, index=False)

print("Claims evaluated:", len(df))
print("Supported:", int(gold.sum()))
print("Not supported:", int((~gold).sum()))
print("\nThreshold sweep:")
print(threshold_df.round(3).to_string(index=False))
print("\nSaved:", out_path)
print(
    "\nReminder: this is a diagnostic sweep on the same 40-example "
    "benchmark, not an independent test or production threshold."
)
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
import pandas as pd
import torch

OUT_DIR = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/evaluation/child_benchmark_v1"
)

INPUT_PATH = os.path.join(
    OUT_DIR, "nli_calibration_human_review_corrected.xlsx"
)

df = pd.read_excel(INPUT_PATH, sheet_name="Claims Review")
df["reviewed_gold_label"] = (
    df["reviewed_gold_label"].astype(str).str.strip().str.upper()
)

def score_entailment(premise, hypothesis):
    inputs = nli_tokenizer(
        premise,
        hypothesis,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )
    device = next(nli_model.parameters()).device
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        logits = nli_model(**inputs).logits
        probs = torch.softmax(logits, dim=-1)[0].cpu().tolist()

    return {
        "contradiction": probs[0],
        "entailment": probs[1],
        "neutral": probs[2],
        "tokens": int(inputs["input_ids"].shape[1])
    }

def make_windows(text, window_words=90, overlap_words=25):
    words = str(text).split()
    if not words:
        return []

    step = window_words - overlap_words
    output = []

    for start in range(0, len(words), step):
        snippet = " ".join(words[start:start + window_words])
        if snippet:
            output.append({
                "start_word": start,
                "end_word": min(start + window_words, len(words)),
                "snippet": snippet
            })
        if start + window_words >= len(words):
            break

    return output

details = []

for idx, row in df.iterrows():
    claim = str(row["claim"])
    evidence = str(row["evidence"])

    full = score_entailment(evidence, claim)

    details.append({
        "claim_id": idx + 1,
        "claim": claim,
        "gold_label": row["reviewed_gold_label"],
        "method": "FULL_EVIDENCE",
        "start_word": 0,
        "end_word": len(evidence.split()),
        "snippet": evidence,
        **full
    })

    for window in make_windows(evidence):
        score = score_entailment(window["snippet"], claim)
        details.append({
            "claim_id": idx + 1,
            "claim": claim,
            "gold_label": row["reviewed_gold_label"],
            "method": "SHORT_WINDOW",
            "start_word": window["start_word"],
            "end_word": window["end_word"],
            "snippet": window["snippet"],
            **score
        })

detail_df = pd.DataFrame(details)

# Summarize the maximum window entailment score per claim.
short_df = detail_df[detail_df["method"] == "SHORT_WINDOW"]

window_best = (
    short_df.sort_values("entailment", ascending=False)
    .groupby("claim_id", as_index=False)
    .first()
)

full_df = (
    detail_df[detail_df["method"] == "FULL_EVIDENCE"]
    [["claim_id", "claim", "gold_label", "entailment",
      "contradiction", "neutral"]]
    .rename(columns={
        "entailment": "full_entailment",
        "contradiction": "full_contradiction",
        "neutral": "full_neutral"
    })
)

comparison = full_df.merge(
    window_best[[
        "claim_id", "entailment", "contradiction",
        "neutral", "start_word", "end_word", "snippet"
    ]].rename(columns={
        "entailment": "best_window_entailment",
        "contradiction": "best_window_contradiction",
        "neutral": "best_window_neutral",
        "start_word": "best_window_start",
        "end_word": "best_window_end",
        "snippet": "best_window_evidence"
    }),
    on="claim_id",
    how="left"
)

comparison["entailment_change"] = (
    comparison["best_window_entailment"] -
    comparison["full_entailment"]
)

detail_path = os.path.join(
    OUT_DIR, "nli_full_vs_windowed_claims_v1.xlsx"
)
summary_path = os.path.join(
    OUT_DIR, "nli_full_vs_windowed_summary_v1.csv"
)

with pd.ExcelWriter(detail_path, engine="openpyxl") as writer:
    comparison.to_excel(writer, sheet_name="Claim_Comparison", index=False)
    detail_df.to_excel(writer, sheet_name="All_Scores", index=False)

comparison.groupby("gold_label").agg(
    claims=("claim_id", "count"),
    mean_full_entailment=("full_entailment", "mean"),
    mean_window_entailment=("best_window_entailment", "mean"),
    mean_entailment_change=("entailment_change", "mean")
).reset_index().to_csv(summary_path, index=False)

print("Claims evaluated:", len(df))
print("\nMean entailment scores by reviewed label:")
print(
    comparison.groupby("gold_label")[
        ["full_entailment", "best_window_entailment", "entailment_change"]
    ].mean().round(3).to_string()
)

print("\nLargest improvements from windowing:")
print(
    comparison.sort_values("entailment_change", ascending=False)[[
        "claim_id", "gold_label", "full_entailment",
        "best_window_entailment", "claim"
    ]].head(10).round(3).to_string(index=False)
)

print("\nSaved detailed workbook:", detail_path)
print("Saved summary CSV:", summary_path)
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd

OUT_DIR = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/"
    "english_only_v1/evaluation/child_benchmark_v1"
)

INPUT_PATH = os.path.join(
    OUT_DIR, "nli_full_vs_windowed_claims_v1.xlsx"
)
OUTPUT_PATH = os.path.join(
    OUT_DIR, "nli_failure_analysis_v1.xlsx"
)

df = pd.read_excel(INPUT_PATH, sheet_name="Claim_Comparison")

# Sort by the most concerning cases:
# 1. Supported claims with low entailment.
# 2. Partially supported claims with high entailment.
# 3. Not-enough-information claims with high entailment.
def priority(row):
    label = str(row["gold_label"]).upper()
    score = float(row["best_window_entailment"])

    if label == "SUPPORTED":
        return score  # lower is worse
    if label == "PARTIALLY_SUPPORTED":
        return 1 - score  # lower is worse
    if label == "NOT_ENOUGH_INFORMATION":
        return 1 - score  # high entailment is worse
    return 1.0

df["review_priority"] = df.apply(priority, axis=1)

supported_failures = df[
    (df["gold_label"] == "SUPPORTED") &
    (df["best_window_entailment"] < 0.35)
].sort_values("best_window_entailment")

partial_risks = df[
    (df["gold_label"] == "PARTIALLY_SUPPORTED") &
    (df["best_window_entailment"] >= 0.80)
].sort_values("best_window_entailment", ascending=False)

unsupported_risks = df[
    (df["gold_label"] == "NOT_ENOUGH_INFORMATION") &
    (df["best_window_entailment"] >= 0.10)
].sort_values("best_window_entailment", ascending=False)

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    df.sort_values("review_priority").to_excel(
        writer, sheet_name="All_Ranked", index=False
    )
    supported_failures.to_excel(
        writer, sheet_name="Supported_Low_Score", index=False
    )
    partial_risks.to_excel(
        writer, sheet_name="Partial_High_Score", index=False
    )
    unsupported_risks.to_excel(
        writer, sheet_name="Missing_Evidence_Risk", index=False
    )

print("Supported claims scoring below 0.35:",
      len(supported_failures))
print(supported_failures[[
    "claim", "best_window_entailment", "best_window_evidence"
]].to_string(index=False))

print("\nPartially supported claims scoring at least 0.80:",
      len(partial_risks))
print(partial_risks[[
    "claim", "best_window_entailment", "best_window_evidence"
]].to_string(index=False))

print("\nNot-enough-information claims scoring at least 0.10:",
      len(unsupported_risks))
print(unsupported_risks[[
    "claim", "best_window_entailment", "best_window_evidence"
]].to_string(index=False))

print("\nSaved:", OUTPUT_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "nli_calibration_human_review_corrected.xlsx"
OUTPUT_PATH = ROOT / "nli_suspicious_claims_review_adjudicated_v1.xlsx"

# Load the reviewed 40-claim calibration dataset
df = pd.read_excel(INPUT_PATH, sheet_name="Claims Review")

# Ensure new columns use string dtype to avoid pandas dtype warnings
for col in ["adjudicated_label", "adjudication_reason", "reviewer_decision"]:
    if col not in df.columns:
        df[col] = ""
    df[col] = df[col].fillna("").astype(str)

# Claim text -> adjudicated label, rationale, and decision
adjudications = {
    "The textbook suggests weighing two items against each other to find the heavier one.": (
        "SUPPORTED",
        "Faithful paraphrase of the explicit instruction to compare two objects to see which is heavier.",
        "CONFIRM"
    ),
    "The passage tells the reader to make a balance.": (
        "SUPPORTED",
        "The evidence explicitly instructs the reader to make a balance.",
        "CONFIRM"
    ),
    "A child can use a balance made from a cup, string and a hanger.": (
        "SUPPORTED",
        "The evidence describes making a balance with a paper cup, thread and a hanger.",
        "CONFIRM"
    ),
    "The passage includes a description of housing in different regions.": (
        "PARTIALLY_SUPPORTED",
        "The supplied excerpt describes Sikkim housing, but alone does not establish a comparison across multiple regions.",
        "CONFIRM_WITH_SCOPE_LIMIT"
    ),
    "The passage gives measurements for the angle of Sikkim's roofs.": (
        "NOT_ENOUGH_INFORMATION",
        "The excerpt mentions sloping roofs but provides no angle measurements.",
        "CONFIRM"
    ),
    "The passage explains why Sikkim's traditional houses have sloping roofs.": (
        "NOT_ENOUGH_INFORMATION",
        "The excerpt asks why the roofs slope but does not supply the reason.",
        "CONFIRM"
    )
}

matched = 0

for idx, row in df.iterrows():
    claim = str(row["claim"]).strip()

    if claim in adjudications:
        label, reason, decision = adjudications[claim]
        df.at[idx, "adjudicated_label"] = label
        df.at[idx, "adjudication_reason"] = reason
        df.at[idx, "reviewer_decision"] = decision
        matched += 1

# Preserve the original reviewed labels and NLI scores.
# Save the updated 40-claim data and a focused six-claim summary.
summary = df[df["claim"].isin(adjudications.keys())][[
    "category",
    "claim",
    "evidence",
    "reviewed_gold_label",
    "adjudicated_label",
    "nli_entailment",
    "nli_contradiction",
    "nli_neutral",
    "adjudication_reason",
    "reviewer_decision"
]].copy()

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="All_Claims", index=False)
    summary.to_excel(writer, sheet_name="Adjudication_Summary", index=False)

print(f"Claims adjudicated: {matched}/6")
print(f"Full dataset rows: {len(df)}")
print(f"Saved workbook: {OUTPUT_PATH}")

if matched != 6:
    print("WARNING: Not all six claims matched. Check exact claim wording.")
else:
    print("\nAdjudication summary:")
    print(summary[[
        "claim", "reviewed_gold_label", "adjudicated_label"
    ]].to_string(index=False))
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path
from sklearn.metrics import confusion_matrix, classification_report

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "nli_suspicious_claims_review_adjudicated_v1.xlsx"
OUTPUT_PATH = ROOT / "nli_adjudicated_recheck_v1.xlsx"

df = pd.read_excel(INPUT_PATH, sheet_name="All_Claims")

# Evaluate only claims with a human-adjudicated label.
df = df[df["adjudicated_label"].fillna("").str.strip() != ""].copy()

# NLI entailment threshold: diagnostic only, not a production threshold.
THRESHOLD = 0.50

df["nli_binary_prediction"] = (
    df["nli_entailment"] >= THRESHOLD
).map({True: "SUPPORTED", False: "NOT_SUPPORTED"})

df["gold_binary_label"] = df["adjudicated_label"].map(
    lambda x: "SUPPORTED"
    if x == "SUPPORTED"
    else "NOT_SUPPORTED"
    if x in ["CONTRADICTED", "NOT_ENOUGH_INFORMATION"]
    else "EXCLUDED"
)

# Partial support is excluded from this binary calculation.
eval_df = df[df["gold_binary_label"] != "EXCLUDED"].copy()

y_true = eval_df["gold_binary_label"]
y_pred = eval_df["nli_binary_prediction"]

labels = ["SUPPORTED", "NOT_SUPPORTED"]
cm = confusion_matrix(y_true, y_pred, labels=labels)

tn, fp = cm[1, 1], cm[1, 0]
fn, tp = cm[0, 1], cm[0, 0]

print("Adjudicated claims:", len(df))
print("Binary-evaluable claims:", len(eval_df))
print(f"Diagnostic entailment threshold: {THRESHOLD}")
print("\nConfusion matrix (rows=gold, columns=prediction):")
print(pd.DataFrame(
    cm,
    index=["Gold supported", "Gold not supported"],
    columns=["Predicted supported", "Predicted not supported"]
))

print("\nTP:", tp, "| FN:", fn, "| FP:", fp, "| TN:", tn)
print("\nClassification report:")
print(classification_report(
    y_true, y_pred,
    labels=labels,
    zero_division=0
))

# Save all claims, plus false acceptance/rejection cases.
false_acceptances = eval_df[
    (eval_df["gold_binary_label"] == "NOT_SUPPORTED") &
    (eval_df["nli_binary_prediction"] == "SUPPORTED")
].copy()

false_rejections = eval_df[
    (eval_df["gold_binary_label"] == "SUPPORTED") &
    (eval_df["nli_binary_prediction"] == "NOT_SUPPORTED")
].copy()

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="Adjudicated_Claims", index=False)
    eval_df.to_excel(writer, sheet_name="Binary_Evaluation", index=False)
    false_acceptances.to_excel(writer, sheet_name="False_Acceptances", index=False)
    false_rejections.to_excel(writer, sheet_name="False_Rejections", index=False)

print("\nSaved:", OUTPUT_PATH)
print("False acceptances:", len(false_acceptances))
print("False rejections:", len(false_rejections))
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path
from sklearn.metrics import confusion_matrix, classification_report

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "nli_calibration_human_review_corrected.xlsx"
OUTPUT_PATH = ROOT / "nli_full40_binary_recheck_v1.xlsx"

df = pd.read_excel(INPUT_PATH, sheet_name="Claims Review")

# Use the six adjudications where available; otherwise retain the
# previously reviewed human labels for the remaining claims.
adjudicated_path = ROOT / "nli_suspicious_claims_review_adjudicated_v1.xlsx"
adjudicated = pd.read_excel(adjudicated_path, sheet_name="All_Claims")

adjudication_map = (
    adjudicated.dropna(subset=["adjudicated_label"])
    .set_index("claim")["adjudicated_label"]
    .to_dict()
)

df["evaluation_label"] = df["claim"].map(adjudication_map)
df["evaluation_label"] = df["evaluation_label"].fillna(
    df["reviewed_gold_label"]
)

# Binary classification: SUPPORTED vs NOT_SUPPORTED.
# Exclude partial labels rather than forcing them into either class.
df["gold_binary"] = df["evaluation_label"].map({
    "SUPPORTED": "SUPPORTED",
    "CONTRADICTED": "NOT_SUPPORTED",
    "NOT_ENOUGH_INFORMATION": "NOT_SUPPORTED"
})

THRESHOLD = 0.50

df["predicted_binary"] = (
    pd.to_numeric(df["nli_entailment"], errors="coerce") >= THRESHOLD
).map({True: "SUPPORTED", False: "NOT_SUPPORTED"})

eval_df = df.dropna(subset=["gold_binary", "predicted_binary"]).copy()

labels = ["SUPPORTED", "NOT_SUPPORTED"]
cm = confusion_matrix(
    eval_df["gold_binary"],
    eval_df["predicted_binary"],
    labels=labels
)

print("Total claims:", len(df))
print("Binary-evaluable claims:", len(eval_df))
print("Excluded partial/unlabelled claims:", len(df) - len(eval_df))
print("\nConfusion matrix:")
print(pd.DataFrame(
    cm,
    index=["Gold supported", "Gold not supported"],
    columns=["Predicted supported", "Predicted not supported"]
))
print("\nClassification report:")
print(classification_report(
    eval_df["gold_binary"],
    eval_df["predicted_binary"],
    labels=labels,
    zero_division=0
))

false_acceptances = eval_df[
    (eval_df["gold_binary"] == "NOT_SUPPORTED") &
    (eval_df["predicted_binary"] == "SUPPORTED")
]
false_rejections = eval_df[
    (eval_df["gold_binary"] == "SUPPORTED") &
    (eval_df["predicted_binary"] == "NOT_SUPPORTED")
]

print("False acceptances:", len(false_acceptances))
print("False rejections:", len(false_rejections))

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    eval_df.to_excel(writer, sheet_name="All_Evaluated", index=False)
    false_acceptances.to_excel(writer, sheet_name="False_Acceptances", index=False)
    false_rejections.to_excel(writer, sheet_name="False_Rejections", index=False)

print("\nSaved:", OUTPUT_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "nli_calibration_human_review_corrected.xlsx"
OLD_PATH = ROOT / "nli_full40_binary_recheck_v1.xlsx"
OUTPUT_PATH = ROOT / "nli_model_comparison_base_v1.xlsx"

# Rebuild the input dataframe.
df = pd.read_excel(INPUT_PATH, sheet_name="Claims Review")

adj_path = ROOT / "nli_suspicious_claims_review_adjudicated_v1.xlsx"
adj = pd.read_excel(adj_path, sheet_name="All_Claims")

adj_map = dict(zip(adj["claim"], adj["adjudicated_label"]))
df["evaluation_label"] = df["claim"].map(adj_map)
df["evaluation_label"] = df["evaluation_label"].fillna(
    df["reviewed_gold_label"]
)

# Reuse the already-loaded model and tokenizer.
assert "model" in globals() and "tokenizer" in globals(), (
    "Model not found in memory. Rerun the model-loading cell first."
)

id2label = {
    int(k): str(v).lower()
    for k, v in model.config.id2label.items()
}

entail_idx = next(i for i, name in id2label.items() if "entail" in name)
contr_idx = next(i for i, name in id2label.items() if "contrad" in name)
neutral_idx = next(i for i, name in id2label.items() if "neutral" in name)

def score_pair(premise, hypothesis):
    inputs = tokenizer(
        str(premise), str(hypothesis),
        return_tensors="pt",
        truncation=True,
        max_length=512
    )
    inputs = {k: v.to(DEVICE) for k, v in inputs.items()}

    with torch.no_grad():
        probs = torch.softmax(model(**inputs).logits[0], dim=-1)
    probs = probs.detach().cpu().numpy()

    return (
        float(probs[entail_idx]),
        float(probs[contr_idx]),
        float(probs[neutral_idx])
    )

# Score all 40 claims with the base model.
scores = [
    score_pair(row["evidence"], row["claim"])
    for _, row in df.iterrows()
]

df["base_entailment"] = [s[0] for s in scores]
df["base_contradiction"] = [s[1] for s in scores]
df["base_neutral"] = [s[2] for s in scores]

# Explicitly rename old columns to prevent merge suffix collisions.
old = pd.read_excel(OLD_PATH, sheet_name="All_Evaluated")
old = old[["claim", "nli_entailment", "nli_contradiction", "nli_neutral"]]
old = old.drop_duplicates("claim").rename(columns={
    "nli_entailment": "small_entailment",
    "nli_contradiction": "small_contradiction",
    "nli_neutral": "small_neutral"
})

df = df.drop(
    columns=["small_entailment", "small_contradiction", "small_neutral"],
    errors="ignore"
).merge(old, on="claim", how="left", validate="many_to_one")

df["small_prediction_at_050"] = np.where(
    df["small_entailment"] >= 0.50,
    "SUPPORTED", "NOT_SUPPORTED"
)
df["base_prediction_at_050"] = np.where(
    df["base_entailment"] >= 0.50,
    "SUPPORTED", "NOT_SUPPORTED"
)

df["prediction_changed"] = (
    df["small_prediction_at_050"] != df["base_prediction_at_050"]
)

print("Claims scored:", len(df))
print("Claims with old scores:", df["small_entailment"].notna().sum())
print("Predictions changed at 0.50:", int(df["prediction_changed"].sum()))

print("\nLargest entailment-score changes:")
df["score_change"] = df["base_entailment"] - df["small_entailment"]

print(df[[
    "claim", "evaluation_label",
    "small_entailment", "base_entailment",
    "score_change", "small_prediction_at_050",
    "base_prediction_at_050"
]].sort_values(
    "score_change", key=lambda x: x.abs(), ascending=False
).head(15).to_string(index=False))

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="Model_Comparison", index=False)

print("\nSaved:", OUTPUT_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix
)

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "nli_model_comparison_base_v1.xlsx"
OUTPUT_PATH = ROOT / "nli_small_vs_base_metrics_v1.xlsx"

df = pd.read_excel(INPUT_PATH, sheet_name="Model_Comparison")

# Keep only claims with scores from both models and usable gold labels.
df = df.dropna(subset=[
    "small_entailment", "base_entailment", "evaluation_label"
]).copy()

df["gold_binary"] = df["evaluation_label"].map({
    "SUPPORTED": "SUPPORTED",
    "CONTRADICTED": "NOT_SUPPORTED",
    "NOT_ENOUGH_INFORMATION": "NOT_SUPPORTED"
})

# Partial-support claims are excluded from this binary comparison.
eval_df = df.dropna(subset=["gold_binary"]).copy()

y_true = eval_df["gold_binary"]

eval_df["small_pred"] = np.where(
    eval_df["small_entailment"] >= 0.50,
    "SUPPORTED", "NOT_SUPPORTED"
)
eval_df["base_pred"] = np.where(
    eval_df["base_entailment"] >= 0.50,
    "SUPPORTED", "NOT_SUPPORTED"
)

def metrics_for(name, predictions):
    cm = confusion_matrix(
        y_true, predictions,
        labels=["SUPPORTED", "NOT_SUPPORTED"]
    )
    tn, fp = cm[1, 1], cm[1, 0]
    fn, tp = cm[0, 1], cm[0, 0]

    return {
        "model": name,
        "n": len(y_true),
        "accuracy": accuracy_score(y_true, predictions),
        "supported_precision": precision_score(
            y_true, predictions,
            pos_label="SUPPORTED", zero_division=0
        ),
        "supported_recall": recall_score(
            y_true, predictions,
            pos_label="SUPPORTED", zero_division=0
        ),
        "supported_f1": f1_score(
            y_true, predictions,
            pos_label="SUPPORTED", zero_division=0
        ),
        "false_acceptances": int(fp),
        "false_rejections": int(fn),
        "true_positives": int(tp),
        "true_negatives": int(tn)
    }

metrics = pd.DataFrame([
    metrics_for("DeBERTa-v3-small", eval_df["small_pred"]),
    metrics_for("DeBERTa-v3-base", eval_df["base_pred"])
])

eval_df["small_error"] = eval_df["small_pred"] != y_true
eval_df["base_error"] = eval_df["base_pred"] != y_true

eval_df["error_comparison"] = np.select(
    [
        eval_df["small_error"] & ~eval_df["base_error"],
        ~eval_df["small_error"] & eval_df["base_error"],
        eval_df["small_error"] & eval_df["base_error"]
    ],
    [
        "Base fixes small-model error",
        "Base introduces new error",
        "Both models wrong"
    ],
    default="Both models correct"
)

print("Evaluable claims:", len(eval_df))
print("\nMetrics at entailment threshold 0.50:")
print(metrics.round(3).to_string(index=False))

print("\nError comparison:")
print(eval_df["error_comparison"].value_counts().to_string())

print("\nClaims where base model introduces an error:")
print(eval_df.loc[
    eval_df["error_comparison"] == "Base introduces new error",
    ["claim", "evaluation_label", "small_entailment", "base_entailment"]
].to_string(index=False))

print("\nFalse acceptances by model:")
for model_name, pred_col in [
    ("Small", "small_pred"),
    ("Base", "base_pred")
]:
    bad = eval_df[
        (eval_df["gold_binary"] == "NOT_SUPPORTED") &
        (eval_df[pred_col] == "SUPPORTED")
    ]
    print(f"\n{model_name}: {len(bad)}")
    if len(bad):
        print(bad[[
            "claim", "evaluation_label",
            "small_entailment", "base_entailment"
        ]].to_string(index=False))

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    metrics.to_excel(writer, sheet_name="Metrics", index=False)
    eval_df.to_excel(writer, sheet_name="Claim_Comparison", index=False)
    eval_df[eval_df["error_comparison"] != "Both models correct"].to_excel(
        writer, sheet_name="Changed_Error_Cases", index=False
    )

print("\nSaved:", OUTPUT_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "nli_small_vs_base_metrics_v1.xlsx"
OUTPUT_PATH = ROOT / "nli_both_models_failures_v1.xlsx"

df = pd.read_excel(INPUT_PATH, sheet_name="Claim_Comparison")

# Recreate predictions using the same threshold as the comparison.
df["small_pred"] = (
    df["small_entailment"] >= 0.50
).map({True: "SUPPORTED", False: "NOT_SUPPORTED"})

df["base_pred"] = (
    df["base_entailment"] >= 0.50
).map({True: "SUPPORTED", False: "NOT_SUPPORTED"})

df["gold_binary"] = df["evaluation_label"].map({
    "SUPPORTED": "SUPPORTED",
    "CONTRADICTED": "NOT_SUPPORTED",
    "NOT_ENOUGH_INFORMATION": "NOT_SUPPORTED"
})

df = df.dropna(subset=["gold_binary"]).copy()

both_wrong = df[
    (df["small_pred"] != df["gold_binary"]) &
    (df["base_pred"] != df["gold_binary"])
].copy()

both_wrong["error_type"] = both_wrong.apply(
    lambda r: (
        "FALSE_ACCEPTANCE"
        if r["gold_binary"] == "NOT_SUPPORTED"
        and r["base_pred"] == "SUPPORTED"
        else "FALSE_REJECTION"
    ),
    axis=1
)

columns = [
    "category", "topic", "claim", "evidence",
    "evaluation_label", "gold_binary",
    "small_entailment", "base_entailment",
    "small_pred", "base_pred", "error_type"
]
columns = [c for c in columns if c in both_wrong.columns]

print("Claims both models get wrong:", len(both_wrong))
print(both_wrong[columns].to_string(index=False))

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    both_wrong[columns].to_excel(
        writer, sheet_name="Shared_Failures", index=False
    )

print("\nSaved:", OUTPUT_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "nli_both_models_failures_v1.xlsx"
OUTPUT_PATH = ROOT / "evidence_sufficiency_diagnostic_v1.xlsx"

df = pd.read_excel(INPUT_PATH, sheet_name="Shared_Failures")

def diagnose_evidence(row):
    claim = str(row["claim"]).lower()
    evidence = str(row["evidence"]).lower()

    # Detect questions that ask for a reason but don't state one.
    asks_why = (
        "explains why" in claim
        and ("why do you think" in evidence or "why" in evidence)
    )

    if asks_why:
        return pd.Series({
            "diagnostic_category": "QUESTION_WITHOUT_EXPLICIT_ANSWER",
            "suggested_action": "ABSTAIN_OR_RETRIEVE_MORE_EVIDENCE",
            "reason": (
                "The excerpt poses a why-question and names the roof feature, "
                "but the supplied text does not state the cause."
            )
        })

    # Explicit comparison instruction in the weight passage.
    comparison_claim = any(term in claim for term in [
        "heavier", "weighing two items", "compare objects"
    ])
    comparison_evidence = (
        "compare any two objects" in evidence
        and "which one is heavier" in evidence
    )

    if comparison_claim and comparison_evidence:
        return pd.Series({
            "diagnostic_category": "SUPPORTED_PARAPHRASE_CANDIDATE",
            "suggested_action": "ACCEPT_ONLY_AFTER_EVIDENCE_CHECK",
            "reason": (
                "The evidence explicitly says to compare two objects "
                "to determine which is heavier."
            )
        })

    return pd.Series({
        "diagnostic_category": "UNRESOLVED",
        "suggested_action": "MANUAL_REVIEW",
        "reason": "The diagnostic rules did not confidently classify this case."
    })

diagnostics = df.apply(diagnose_evidence, axis=1)
result = pd.concat([df, diagnostics], axis=1)

print(result[[
    "claim",
    "diagnostic_category",
    "suggested_action",
    "reason"
]].to_string(index=False))

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    result.to_excel(writer, sheet_name="Diagnostic", index=False)

print("\nSaved:", OUTPUT_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

RETRIEVAL_PATH = ROOT / "all50_retrieval_comparison_20261011_165339.xlsx"
BENCHMARK_PATH = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "child_perspective_ncert_benchmark_v1.xlsx"
)

OUTPUT_PATH = ROOT / "answerability_evaluation_populated_v1.xlsx"

retrieval = pd.read_excel(
    RETRIEVAL_PATH, sheet_name="All Top5 Results"
)
benchmark = pd.read_excel(BENCHMARK_PATH)

print("Strategies:", sorted(retrieval["strategy"].dropna().unique()))
print("Questions in retrieval:", retrieval["question_id"].nunique())
print("Retrieval rows:", len(retrieval))

# Use the strongest candidate from the previous human relevance evaluation.
# Stop rather than silently choosing a different strategy if the name differs.
candidate = "RRF+CrossEncoder"

if candidate not in retrieval["strategy"].astype(str).unique():
    raise ValueError(
        f"{candidate!r} not found. Available strategies: "
        f"{sorted(retrieval['strategy'].dropna().unique())}"
    )

top5 = retrieval[
    retrieval["strategy"].astype(str) == candidate
].copy()

top5 = top5.sort_values(["question_id", "rank"])

# Compact, readable evidence column for manual review.
def format_passages(group):
    blocks = []
    for _, r in group.iterrows():
        blocks.append(
            f"RANK {r['rank']} | {r['source_file']} | "
            f"Pages: {r['pages']} | Score: {r['score']}\n"
            f"{r['passage']}"
        )
    return "\n\n".join(blocks)

evidence_by_question = (
    top5.groupby("question_id")
    .apply(format_passages)
    .rename("retrieved_top5_evidence")
    .reset_index()
)

# Preserve benchmark labels and metadata; add retrieval evidence and review fields.
worksheet = benchmark.merge(
    evidence_by_question,
    on="question_id",
    how="left",
    validate="one_to_one"
)

for col in [
    "generated_answer",
    "human_answerability_label",
    "evidence_sufficiency_label",
    "answer_faithfulness_label",
    "abstention_required",
    "unsupported_claims_found",
    "reviewer_notes"
]:
    if col not in worksheet.columns:
        worksheet[col] = ""

guidelines = pd.DataFrame([
    ["human_answerability_label", "ANSWERABLE", "Sufficient evidence exists in the NCERT corpus."],
    ["human_answerability_label", "PARTIALLY_ANSWERABLE", "The corpus supports only part of the requested answer."],
    ["human_answerability_label", "UNANSWERABLE_IN_CORPUS", "Relevant corpus search confirms no sufficient answer."],
    ["evidence_sufficiency_label", "SUFFICIENT", "These retrieved passages support the answer."],
    ["evidence_sufficiency_label", "INSUFFICIENT", "Retrieved passages are related but do not support the answer fully."],
    ["evidence_sufficiency_label", "CONTRADICTORY", "Retrieved passages conflict with the proposed answer."],
    ["answer_faithfulness_label", "FAITHFUL", "All factual claims in the generated answer are supported."],
    ["answer_faithfulness_label", "PARTIALLY_FAITHFUL", "Some answer claims are unsupported or omitted."],
    ["answer_faithfulness_label", "UNFAITHFUL", "Material claims are unsupported or contradicted."],
    ["abstention_required", "YES", "The available evidence is insufficient to answer safely."],
    ["abstention_required", "NO", "The evidence supports a suitable answer."]
], columns=["field", "allowed_value", "guidance"])

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    worksheet.to_excel(writer, sheet_name="Answerability_Review", index=False)
    top5.to_excel(writer, sheet_name="Retrieved_Top5_Rows", index=False)
    guidelines.to_excel(writer, sheet_name="Label_Guidelines", index=False)

print("\nCreated:", OUTPUT_PATH)
print("Questions:", len(worksheet))
print("Questions with retrieved evidence:",
      worksheet["retrieved_top5_evidence"].notna().sum())
print("Top-five passages included:", len(top5))
print("\nReview labels remain blank; no answerability was inferred.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

RETRIEVAL_PATH = ROOT / "all50_retrieval_comparison_20261011_165339.xlsx"
BENCHMARK_PATH = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "child_perspective_ncert_benchmark_v1.xlsx"
)
OUTPUT_PATH = ROOT / "answerability_evaluation_populated_v1.xlsx"

retrieval = pd.read_excel(RETRIEVAL_PATH, sheet_name="All Top5 Results")
benchmark = pd.read_excel(BENCHMARK_PATH)

# Normalize strategy labels and select the intended reranker.
retrieval["strategy_clean"] = (
    retrieval["strategy"].astype(str).str.strip().str.replace(r"\s+", " ", regex=True)
)
candidate = "RRF + CrossEncoder"

print("Strategies:", sorted(retrieval["strategy_clean"].unique()))

top5 = retrieval[retrieval["strategy_clean"] == candidate].copy()

if top5.empty:
    raise ValueError(f"No rows found for strategy: {candidate}")

top5 = top5.sort_values(["question_id", "rank"])

def format_passages(group):
    blocks = []
    for _, r in group.iterrows():
        blocks.append(
            f"RANK {r['rank']} | {r['source_file']} | "
            f"Pages: {r['pages']} | Score: {r['score']}\n"
            f"{r['passage']}"
        )
    return "\n\n".join(blocks)

evidence_by_question = (
    top5.groupby("question_id")
    .apply(format_passages)
    .rename("retrieved_top5_evidence")
    .reset_index()
)

worksheet = benchmark.merge(
    evidence_by_question,
    on="question_id",
    how="left",
    validate="one_to_one"
)

for col in [
    "generated_answer",
    "human_answerability_label",
    "evidence_sufficiency_label",
    "answer_faithfulness_label",
    "abstention_required",
    "unsupported_claims_found",
    "reviewer_notes"
]:
    if col not in worksheet.columns:
        worksheet[col] = ""

guidelines = pd.DataFrame([
    ["human_answerability_label", "ANSWERABLE", "Sufficient evidence exists in the NCERT corpus."],
    ["human_answerability_label", "PARTIALLY_ANSWERABLE", "The corpus supports only part of the requested answer."],
    ["human_answerability_label", "UNANSWERABLE_IN_CORPUS", "Relevant corpus search confirms no sufficient answer."],
    ["evidence_sufficiency_label", "SUFFICIENT", "Retrieved passages support the answer."],
    ["evidence_sufficiency_label", "INSUFFICIENT", "Retrieved passages are related but lack required information."],
    ["evidence_sufficiency_label", "CONTRADICTORY", "Retrieved passages conflict with the proposed answer."],
    ["answer_faithfulness_label", "FAITHFUL", "All factual claims are supported by evidence."],
    ["answer_faithfulness_label", "PARTIALLY_FAITHFUL", "Some claims are unsupported or omitted."],
    ["answer_faithfulness_label", "UNFAITHFUL", "Material claims are unsupported or contradicted."],
    ["abstention_required", "YES", "Evidence is insufficient to answer safely."],
    ["abstention_required", "NO", "Evidence supports a suitable answer."]
], columns=["field", "allowed_value", "guidance"])

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    worksheet.to_excel(writer, sheet_name="Answerability_Review", index=False)
    top5.to_excel(writer, sheet_name="Retrieved_Top5_Rows", index=False)
    guidelines.to_excel(writer, sheet_name="Label_Guidelines", index=False)

print("\nCreated:", OUTPUT_PATH)
print("Questions:", len(worksheet))
print("Questions with retrieved evidence:",
      worksheet["retrieved_top5_evidence"].notna().sum())
print("Retrieved passages:", len(top5))
print("Review labels remain blank.")
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import torch
from pathlib import Path
from transformers import AutoTokenizer, AutoModelForCausalLM

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "answerability_evaluation_populated_v1.xlsx"
OUTPUT_PATH = ROOT / "qwen_answers_with_evidence_v1.xlsx"

MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Load worksheet and verify inputs.
df = pd.read_excel(INPUT_PATH, sheet_name="Answerability_Review")

required = ["question_id", "question", "retrieved_top5_evidence"]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

if df["question_id"].duplicated().any():
    raise ValueError("Duplicate question IDs found.")

print("Questions:", len(df))
print("Device:", DEVICE)

# Reuse the model only if it is the expected model.
reuse_model = (
    "model" in globals()
    and "tokenizer" in globals()
    and getattr(model.config, "_name_or_path", "") == MODEL_NAME
)

if not reuse_model:
    print("Loading:", MODEL_NAME)
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float16 if DEVICE == "cuda" else torch.float32,
        device_map="auto" if DEVICE == "cuda" else None
    )
    model.eval()
else:
    print("Reusing loaded Qwen model.")

SYSTEM_PROMPT = """You are a kind, careful tutor for children in Classes 1–4.
Use only the supplied NCERT evidence to answer.
Use simple, age-appropriate language.
Do not add facts that are not supported by the evidence.
If the evidence does not contain enough information, say:
"I couldn't find enough information in the textbook passages to answer that."
Do not guess, invent explanations, or claim that the textbook says something
unless the supplied passages support it.
Keep answers short and clear.
"""

def generate_answer(question, evidence):
    evidence = str(evidence)

    # Keep evidence bounded to avoid overflowing the context window.
    # Retain the top-ranked passages first.
    evidence = evidence[:10000]

    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {
            "role": "user",
            "content": (
                f"Child's question:\n{question}\n\n"
                f"Retrieved NCERT passages:\n{evidence}\n\n"
                "Answer using only the evidence above. If it is insufficient, "
                "clearly say you cannot answer from these passages."
            )
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=6000
    )

    model_device = next(model.parameters()).device
    inputs = {k: v.to(model_device) for k, v in inputs.items()}

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=120,
            do_sample=False,
            repetition_penalty=1.05,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_tokens = output[0][inputs["input_ids"].shape[1]:]
    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    return answer, prompt

answers = []
prompts = []

for i, row in df.iterrows():
    answer, prompt = generate_answer(
        row["question"],
        row["retrieved_top5_evidence"]
    )
    answers.append(answer)
    prompts.append(prompt)

    print(f"[{len(answers)}/{len(df)}] {row['question_id']}: {answer}")

df["generated_answer"] = answers
df["generation_prompt"] = prompts
df["answer_model"] = MODEL_NAME
df["generation_status"] = "GENERATED"

# Preserve original benchmark and retrieval evidence.
with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="Generated_Answers", index=False)

    df[[
        "question_id", "question", "generated_answer",
        "answer_model", "generation_status"
    ]].to_excel(writer, sheet_name="Answer_Summary", index=False)

print("\nGeneration complete.")
print("Answers generated:", len(answers))
print("Saved:", OUTPUT_PATH)

```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

INPUT_PATH = ROOT / "qwen_answers_with_evidence_v1.xlsx"
OUTPUT_PATH = ROOT / "qwen_answer_human_evaluation_v1.xlsx"

df = pd.read_excel(INPUT_PATH, sheet_name="Generated_Answers")

required = [
    "question_id", "question", "answerability",
    "expected_answer", "retrieved_top5_evidence",
    "generated_answer"
]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

def screen_answer(row):
    answer = str(row["generated_answer"]).lower()
    label = str(row["answerability"]).strip().upper()
    flags = []

    abstention_phrases = [
        "couldn't find enough information",
        "not enough information",
        "cannot answer from these passages",
        "don't have enough information",
        "do not have enough information",
        "can't determine from the passages"
    ]
    abstains = any(p in answer for p in abstention_phrases)

    # Benchmark labels are still provisional; these are review priorities.
    if label == "UNANSWERABLE" and not abstains:
        flags.append("CHECK_ABSTENTION: benchmark says unanswerable")

    if label == "ANSWERABLE" and abstains:
        flags.append("CHECK_RETRIEVAL_OR_ABSTENTION: answerable benchmark item")

    risky_topics = {
        "weather": ["tomorrow", "rain", "forecast"],
        "outside_observation": ["outside your house", "right now"],
        "causal_inference": ["because", "this is why", "the reason"],
        "unsupported_generalisation": [
            "likely", "usually", "often", "known for", "more durable"
        ]
    }

    question = str(row["question"]).lower()

    if any(x in question for x in ["tomorrow", "weather", "rain"]):
        if not abstains:
            flags.append("MANUAL_CHECK: current/future weather claim")

    if any(x in question for x in [
        "outside your house", "right now", "your teacher",
        "your dog", "the dog", "why does the dog"
    ]):
        if not abstains:
            flags.append("MANUAL_CHECK: personal or unobserved fact")

    if any(x in answer for x in risky_topics["causal_inference"]):
        flags.append("MANUAL_CHECK: explanation/causal claim")

    if not str(row["retrieved_top5_evidence"]).strip():
        flags.append("NO_RETRIEVED_EVIDENCE")

    return " | ".join(flags) if flags else "STANDARD_REVIEW"

review = df[[
    "question_id", "question", "answerability",
    "expected_answer", "retrieved_top5_evidence",
    "generated_answer"
]].copy()

review["screening_flags"] = df.apply(screen_answer, axis=1)

# Leave human judgements blank; do not treat automated screening as evaluation.
review["human_correctness"] = ""
review["evidence_faithfulness"] = ""
review["abstention_judgement"] = ""
review["age_appropriateness"] = ""
review["unsupported_claims"] = ""
review["reviewer_notes"] = ""

guidelines = pd.DataFrame({
    "Field": [
        "human_correctness",
        "evidence_faithfulness",
        "abstention_judgement",
        "age_appropriateness",
        "unsupported_claims",
        "screening_flags"
    ],
    "Allowed values / guidance": [
        "CORRECT / PARTIALLY_CORRECT / INCORRECT / CANNOT_JUDGE",
        "SUPPORTED / PARTIALLY_SUPPORTED / UNSUPPORTED / CANNOT_JUDGE",
        "APPROPRIATE / SHOULD_HAVE_ANSWERED / SHOULD_HAVE_ABSTAINED / NOT_APPLICABLE",
        "YES / NEEDS_SIMPLIFICATION / NO",
        "NONE / MINOR / MAJOR",
        "Automated prioritisation only; verify against the actual passages."
    ]
})

priority = review[
    review["screening_flags"] != "STANDARD_REVIEW"
].copy()

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    review.to_excel(writer, sheet_name="Human_Review", index=False)
    priority.to_excel(writer, sheet_name="Priority_Review", index=False)
    guidelines.to_excel(writer, sheet_name="Review_Guidelines", index=False)

print("Review workbook created.")
print("Total questions:", len(review))
print("Priority-review questions:", len(priority))
print("Saved to:", OUTPUT_PATH)
print("\nPriority questions:")
print(priority[["question_id", "question", "screening_flags"]].to_string(index=False))
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

# Point this to your HUMAN-ANNOTATED workbook in Drive.
INPUT_PATH = ROOT / "qwen_answer_human_evaluation_v1_annotated.xlsx"
OUTPUT_PATH = ROOT / "qwen_human_evaluation_metrics_v1.xlsx"

df = pd.read_excel(INPUT_PATH, sheet_name="Human_Review")

required = [
    "question_id", "question", "answerability",
    "human_correctness", "evidence_faithfulness",
    "abstention_judgement", "age_appropriateness",
    "unsupported_claims"
]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(
        f"Missing columns: {missing}\n"
        "Check the workbook's sheet name and column headers."
    )

def clean(series):
    return series.fillna("").astype(str).str.strip().str.upper()

for col in required[3:]:
    df[col] = clean(df[col])

def count_table(series, column_name):
    counts = series.replace("", "BLANK").value_counts(dropna=False)
    out = counts.rename_axis(column_name).reset_index(name="count")
    out["percentage"] = (out["count"] / len(df) * 100).round(2)
    return out

# 1. Label distributions
correctness = count_table(df["human_correctness"], "human_correctness")
faithfulness = count_table(df["evidence_faithfulness"], "evidence_faithfulness")
age = count_table(df["age_appropriateness"], "age_appropriateness")
unsupported = count_table(df["unsupported_claims"], "unsupported_claims")
abstention = count_table(df["abstention_judgement"], "abstention_judgement")

# 2. Strict correctness rate: exclude CANNOT_JUDGE and blank labels.
valid_correct = df[
    df["human_correctness"].isin(
        ["CORRECT", "PARTIALLY_CORRECT", "INCORRECT"]
    )
]
strict_correctness_rate = (
    (valid_correct["human_correctness"] == "CORRECT").mean()
    if len(valid_correct) else np.nan
)

# Secondary rate counts partial answers separately.
full_or_partial_rate = (
    valid_correct["human_correctness"].isin(
        ["CORRECT", "PARTIALLY_CORRECT"]
    ).mean()
    if len(valid_correct) else np.nan
)

# 3. Strict evidence faithfulness rate.
valid_faith = df[
    df["evidence_faithfulness"].isin(
        ["SUPPORTED", "PARTIALLY_SUPPORTED", "UNSUPPORTED"]
    )
]
strict_faithfulness_rate = (
    (valid_faith["evidence_faithfulness"] == "SUPPORTED").mean()
    if len(valid_faith) else np.nan
)

# 4. Unsupported-claim rates.
valid_claims = df[df["unsupported_claims"].isin(["NONE", "MINOR", "MAJOR"])]
any_unsupported_rate = (
    (valid_claims["unsupported_claims"] != "NONE").mean()
    if len(valid_claims) else np.nan
)
major_unsupported_rate = (
    (valid_claims["unsupported_claims"] == "MAJOR").mean()
    if len(valid_claims) else np.nan
)

# 5. Abstention metrics.
# Positive = the reviewer says the model SHOULD HAVE ABSTAINED.
abst_df = df[
    df["abstention_judgement"].isin([
        "APPROPRIATE", "SHOULD_HAVE_ANSWERED",
        "SHOULD_HAVE_ABSTAINED"
    ])
].copy()

# Infer whether the model actually abstained from its generated answer.
# Load the generated answer workbook separately for this determination.
GEN_PATH = ROOT / "qwen_answers_with_evidence_v1.xlsx"
gen = pd.read_excel(GEN_PATH, sheet_name="Generated_Answers")
gen["question_id"] = gen["question_id"].astype(str)
abst_df["question_id"] = abst_df["question_id"].astype(str)

abst_phrases = [
    "couldn't find enough information",
    "not enough information",
    "cannot answer from these passages",
    "don't have enough information",
    "do not have enough information",
    "can't determine from the passages"
]

gen["model_abstained"] = gen["generated_answer"].fillna("").astype(str).str.lower().apply(
    lambda s: any(p in s for p in abst_phrases)
)

abst_df = abst_df.merge(
    gen[["question_id", "model_abstained"]],
    on="question_id",
    how="left",
    validate="one_to_one"
)
abst_df["model_abstained"] = abst_df["model_abstained"].fillna(False)

# Should abstain = positive ground truth.
# Actual abstention = positive prediction.
gold = abst_df["abstention_judgement"].eq("SHOULD_HAVE_ABSTAINED")
pred = abst_df["model_abstained"]

tp = int((gold & pred).sum())
fp = int((~gold & pred).sum())
fn = int((gold & ~pred).sum())
tn = int((~gold & ~pred).sum())

precision = tp / (tp + fp) if tp + fp else np.nan
recall = tp / (tp + fn) if tp + fn else np.nan
f1 = (
    2 * precision * recall / (precision + recall)
    if pd.notna(precision) and pd.notna(recall) and precision + recall
    else np.nan
)

# 6. Class-level correctness, useful for diagnosing answerability issues.
class_rows = []
for label, group in df.groupby("answerability", dropna=False):
    valid = group[group["human_correctness"].isin(
        ["CORRECT", "PARTIALLY_CORRECT", "INCORRECT"]
    )]
    class_rows.append({
        "answerability": label,
        "total_questions": len(group),
        "reviewable_questions": len(valid),
        "correct": int((valid["human_correctness"] == "CORRECT").sum()),
        "partially_correct": int(
            (valid["human_correctness"] == "PARTIALLY_CORRECT").sum()
        ),
        "incorrect": int((valid["human_correctness"] == "INCORRECT").sum()),
        "strict_correctness_rate": (
            (valid["human_correctness"] == "CORRECT").mean()
            if len(valid) else np.nan
        )
    })
class_metrics = pd.DataFrame(class_rows)

summary = pd.DataFrame([
    {"metric": "total_questions", "value": len(df)},
    {"metric": "strict_correctness_rate", "value": strict_correctness_rate},
    {"metric": "correct_or_partially_correct_rate", "value": full_or_partial_rate},
    {"metric": "strict_evidence_faithfulness_rate", "value": strict_faithfulness_rate},
    {"metric": "any_unsupported_claim_rate", "value": any_unsupported_rate},
    {"metric": "major_unsupported_claim_rate", "value": major_unsupported_rate},
    {"metric": "abstention_true_positive", "value": tp},
    {"metric": "abstention_false_positive", "value": fp},
    {"metric": "abstention_false_negative", "value": fn},
    {"metric": "abstention_true_negative", "value": tn},
    {"metric": "abstention_precision", "value": precision},
    {"metric": "abstention_recall", "value": recall},
    {"metric": "abstention_f1", "value": f1},
])

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    summary.to_excel(writer, sheet_name="Summary_Metrics", index=False)
    correctness.to_excel(writer, sheet_name="Correctness", index=False)
    faithfulness.to_excel(writer, sheet_name="Faithfulness", index=False)
    unsupported.to_excel(writer, sheet_name="Unsupported_Claims", index=False)
    abstention.to_excel(writer, sheet_name="Abstention_Labels", index=False)
    age.to_excel(writer, sheet_name="Age_Appropriateness", index=False)
    class_metrics.to_excel(writer, sheet_name="By_Answerability", index=False)
    abst_df.to_excel(writer, sheet_name="Abstention_Details", index=False)

print("Evaluation complete.")
print("Questions:", len(df))
print("Strict correctness:", round(strict_correctness_rate, 3))
print("Strict faithfulness:", round(strict_faithfulness_rate, 3))
print("Any unsupported claims:", round(any_unsupported_rate, 3))
print("Abstention precision:", round(precision, 3) if pd.notna(precision) else "N/A")
print("Abstention recall:", round(recall, 3) if pd.notna(recall) else "N/A")
print("Saved:", OUTPUT_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import re
import numpy as np
import pandas as pd
import torch
from pathlib import Path
from transformers import AutoTokenizer, AutoModelForSequenceClassification

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

REVIEW_PATH = ROOT / "qwen_answer_human_evaluation_v1_annotated.xlsx"
OUTPUT_PATH = ROOT / "evidence_sufficiency_baseline_v1.xlsx"

df = pd.read_excel(REVIEW_PATH, sheet_name="Human_Review")
df.columns = df.columns.astype(str).str.strip()

required = [
    "question_id", "question", "retrieved_top5_evidence",
    "generated_answer", "abstention_judgement"
]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

df["question_id"] = df["question_id"].astype(str).str.strip()

df["human_sufficiency_label"] = (
    df["abstention_judgement"].astype(str).str.strip().str.upper()
    .map({
        "SHOULD_HAVE_ABSTAINED": "INSUFFICIENT",
        "APPROPRIATE": "INSUFFICIENT",
        "NOT_APPLICABLE": "UNVERIFIED"
    })
    .fillna("UNVERIFIED")
)

MODEL_NAME = "cross-encoder/nli-deberta-v3-base"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Reuse the loaded model if it is already the correct one.
if (
    "nli_model" in globals()
    and "nli_tokenizer" in globals()
    and getattr(nli_model.config, "_name_or_path", "") == MODEL_NAME
):
    print("Reusing loaded NLI model.")
else:
    print("Loading NLI model...")
    nli_tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    nli_model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)
    nli_model.to(DEVICE)
    nli_model.eval()

print("Label mapping:", nli_model.config.id2label)

label_map = {
    str(label).lower(): int(idx)
    for idx, label in nli_model.config.id2label.items()
}
entail_idx = next(
    (idx for label, idx in label_map.items() if "entail" in label),
    None
)
if entail_idx is None:
    raise ValueError(f"Could not identify entailment label: {label_map}")

def norm(s):
    return re.sub(r"\s+", " ", str(s)).strip()

def lexical_overlap(question, evidence):
    q = set(re.findall(r"\b[a-zA-Z]{3,}\b", norm(question).lower()))
    e = set(re.findall(r"\b[a-zA-Z]{3,}\b", norm(evidence).lower()))
    return len(q & e) / len(q) if q else 0.0

def score_nli(premise, hypothesis):
    inputs = nli_tokenizer(
        premise,
        hypothesis,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )
    inputs = {k: v.to(DEVICE) for k, v in inputs.items()}
    with torch.no_grad():
        probs = torch.softmax(nli_model(**inputs).logits[0], dim=-1)
    return float(probs[entail_idx].item())

rows = []
for _, row in df.iterrows():
    question = norm(row["question"])
    evidence = norm(row["retrieved_top5_evidence"])
    has_evidence = bool(evidence) and evidence.lower() != "nan"

    if has_evidence:
        # Diagnostic probes only; not a calibrated answerability probability.
        probes = [
            f"The supplied passage answers this question: {question}",
            f"The evidence contains the information needed to answer: {question}"
        ]
        scores = [score_nli(evidence[:5000], p) for p in probes]
    else:
        scores = []

    rows.append({
        "question_id": row["question_id"],
        "question": question,
        "human_sufficiency_label": row["human_sufficiency_label"],
        "human_abstention_label": row["abstention_judgement"],
        "nli_probe_max": max(scores) if scores else np.nan,
        "nli_probe_mean": float(np.mean(scores)) if scores else np.nan,
        "lexical_overlap": lexical_overlap(question, evidence) if has_evidence else 0.0,
        "evidence_present": has_evidence,
        "retrieved_evidence": evidence,
        "generated_answer": row["generated_answer"]
    })

results = pd.DataFrame(rows)

# Only explicit insufficient-evidence labels are used here.
# Unverified questions must not be treated as sufficient by default.
labelled = results[
    results["human_sufficiency_label"] == "INSUFFICIENT"
].copy()

sweep_rows = []
for threshold in np.round(np.arange(0.1, 1.0, 0.1), 2):
    pred = labelled["nli_probe_max"] < threshold
    gold = labelled["human_sufficiency_label"].eq("INSUFFICIENT")

    tp = int((pred & gold).sum())
    fp = int((pred & ~gold).sum())
    fn = int((~pred & gold).sum())
    tn = int((~pred & ~gold).sum())

    sweep_rows.append({
        "threshold": threshold,
        "labelled_rows": len(labelled),
        "tp_insufficient": tp,
        "fp_insufficient": fp,
        "fn_insufficient": fn,
        "tn_insufficient": tn,
        "recall_insufficient": tp / (tp + fn) if tp + fn else np.nan
    })

sweep = pd.DataFrame(sweep_rows)

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    results.to_excel(writer, sheet_name="Evidence_Scores", index=False)
    sweep.to_excel(writer, sheet_name="Exploratory_Thresholds", index=False)

print("\nDiagnostic complete.")
print("Questions:", len(results))
print("Explicit insufficient-evidence labels:", len(labelled))
print("Unverified labels:", int(
    (results["human_sufficiency_label"] == "UNVERIFIED").sum()
))
print("Saved:", OUTPUT_PATH)
print("\nNLI scores for questions marked insufficient:")
print(
    labelled[["question_id", "question", "nli_probe_max"]]
    .to_string(index=False)
)
```

```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "evaluation/child_benchmark_v1"
)

HUMAN_PATH = ROOT / "evidence_sufficiency_human_annotation_v1_REVIEWED.xlsx"
NLI_PATH = ROOT / "evidence_sufficiency_baseline_v1.xlsx"
OUTPUT_PATH = ROOT / "evidence_gate_evaluation_v1.xlsx"

human = pd.read_excel(HUMAN_PATH, sheet_name="Annotate_Evidence")
nli = pd.read_excel(NLI_PATH, sheet_name="Evidence_Scores")

human.columns = human.columns.astype(str).str.strip()
nli.columns = nli.columns.astype(str).str.strip()

human["question_id"] = human["question_id"].astype(str).str.strip()
nli["question_id"] = nli["question_id"].astype(str).str.strip()

human["evidence_sufficiency_label"] = (
    human["evidence_sufficiency_label"]
    .astype(str).str.strip().str.upper()
)

valid_labels = {"SUFFICIENT", "PARTIAL", "INSUFFICIENT", "CANNOT_JUDGE"}
bad_labels = set(human["evidence_sufficiency_label"]) - valid_labels
if bad_labels:
    raise ValueError(f"Unexpected evidence labels: {bad_labels}")

if human["question_id"].duplicated().any():
    raise ValueError("Duplicate question IDs in human annotations.")
if nli["question_id"].duplicated().any():
    raise ValueError("Duplicate question IDs in NLI scores.")

df = human.merge(
    nli[["question_id", "nli_probe_max", "nli_probe_mean", "lexical_overlap"]],
    on="question_id",
    how="left",
    validate="one_to_one"
)

if df["nli_probe_max"].isna().any():
    missing_ids = df.loc[df["nli_probe_max"].isna(), "question_id"].tolist()
    raise ValueError(f"Missing NLI scores for: {missing_ids}")

# Strict binary gate:
# SUFFICIENT = answer may proceed
# PARTIAL / INSUFFICIENT / CANNOT_JUDGE = do not generate an unrestricted answer
df["human_gate_label"] = np.where(
    df["evidence_sufficiency_label"].eq("SUFFICIENT"),
    "SUFFICIENT",
    "NOT_SUFFICIENT"
)

print("Human labels:")
print(df["evidence_sufficiency_label"].value_counts().to_string())
print("\nBinary gate labels:")
print(df["human_gate_label"].value_counts().to_string())

# Explore thresholds only; do not select a production threshold from this same set.
sweep_rows = []
for threshold in np.round(np.arange(0.01, 1.00, 0.01), 2):
    # Based on our diagnostic feature, low NLI scores predict insufficient evidence.
    predicted_not_sufficient = df["nli_probe_max"] < threshold
    actual_not_sufficient = df["human_gate_label"].eq("NOT_SUFFICIENT")

    tp = int((predicted_not_sufficient & actual_not_sufficient).sum())
    fp = int((predicted_not_sufficient & ~actual_not_sufficient).sum())
    fn = int((~predicted_not_sufficient & actual_not_sufficient).sum())
    tn = int((~predicted_not_sufficient & ~actual_not_sufficient).sum())

    precision = tp / (tp + fp) if tp + fp else np.nan
    recall = tp / (tp + fn) if tp + fn else np.nan
    specificity = tn / (tn + fp) if tn + fp else np.nan
    accuracy = (tp + tn) / len(df)

    sweep_rows.append({
        "threshold": threshold,
        "accuracy": accuracy,
        "precision_insufficient": precision,
        "recall_insufficient": recall,
        "specificity_sufficient": specificity,
        "TP_not_sufficient": tp,
        "FP_false_abstention": fp,
        "FN_missed_insufficiency": fn,
        "TN_sufficient": tn
    })

sweep = pd.DataFrame(sweep_rows)

# Show several thresholds rather than automatically picking a winner.
selected = sweep[sweep["threshold"].isin([0.05, 0.10, 0.20, 0.30, 0.50, 0.70, 0.90])]

# Use threshold 0.10 for a diagnostic error inspection only.
diagnostic_threshold = 0.10
df["diagnostic_prediction"] = np.where(
    df["nli_probe_max"] < diagnostic_threshold,
    "NOT_SUFFICIENT",
    "SUFFICIENT"
)
df["diagnostic_correct"] = (
    df["diagnostic_prediction"] == df["human_gate_label"]
)

errors = df[~df["diagnostic_correct"]].copy()

with pd.ExcelWriter(OUTPUT_PATH, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="All_50_Compared", index=False)
    selected.to_excel(writer, sheet_name="Threshold_Snapshot", index=False)
    sweep.to_excel(writer, sheet_name="Full_Threshold_Sweep", index=False)
    errors.to_excel(writer, sheet_name="Errors_at_010", index=False)

print("\nDiagnostic threshold:", diagnostic_threshold)
print("Accuracy at diagnostic threshold:", round(df["diagnostic_correct"].mean(), 3))
print("Errors at diagnostic threshold:", len(errors))
print("\nThreshold snapshot:")
print(selected.to_string(index=False))
print("\nSaved:", OUTPUT_PATH)
print("\nNote: this is exploratory evaluation on the same 50 questions used for annotation.")
print("Do not treat its best threshold as a validated production setting.")
```

```python

# CELL 1 — LOAD EXISTING NCERT RETRIEVAL PIPELINE
from google.colab import drive
drive.mount('/content/drive')

import os
import sys
import pickle
import subprocess
import importlib.util

# Install missing dependencies only
packages = {
    "faiss": "faiss-cpu",
    "rank_bm25": "rank-bm25",
    "sentence_transformers": "sentence-transformers",
}
for module, package in packages.items():
    if importlib.util.find_spec(module) is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package])

import numpy as np
import pandas as pd
import faiss
import torch
from sentence_transformers import SentenceTransformer, CrossEncoder

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
RET = f"{ROOT}/05_EXPERIMENTS/retrieval_evaluation/english_only_v1"

PATHS = {
    "chunks": f"{RET}/english_only_chunks.csv",
    "bm25": f"{RET}/indexes/bm25/bm25_index.pkl",
    "tokens": f"{RET}/indexes/bm25/tokenized_documents.pkl",
    "embeddings": f"{RET}/indexes/embeddings/chunk_embeddings.npy",
    "faiss": f"{RET}/indexes/faiss/index.faiss",
}

for name, path in PATHS.items():
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing {name} artifact: {path}")

# Load existing files; do not rebuild or overwrite anything.
chunks = pd.read_csv(PATHS["chunks"]).reset_index(drop=True)

with open(PATHS["bm25"], "rb") as f:
    bm25_index = pickle.load(f)

with open(PATHS["tokens"], "rb") as f:
    tokenized_documents = pickle.load(f)

embeddings = np.load(PATHS["embeddings"])
faiss_index = faiss.read_index(PATHS["faiss"])

print("=== ARTIFACT CHECK ===")
print("Chunks:", len(chunks))
print("Columns:", chunks.columns.tolist())
print("BM25 documents:", len(tokenized_documents))
print("Embeddings:", embeddings.shape)
print("FAISS vectors:", faiss_index.ntotal)
print("GPU available:", torch.cuda.is_available())

if not (
    len(chunks) == len(tokenized_documents)
    == embeddings.shape[0] == faiss_index.ntotal
):
    raise ValueError("Artifact counts do not match. Stop and inspect before proceeding.")

# Detect metadata columns without assuming exact spelling.
def find_column(candidates, required=True):
    mapping = {str(c).strip().lower(): c for c in chunks.columns}
    for name in candidates:
        if name.lower() in mapping:
            return mapping[name.lower()]
    if required:
        raise KeyError(
            f"Expected one of {candidates}; found {chunks.columns.tolist()}"
        )
    return None

TEXT_COL = find_column(["text", "passage", "chunk_text", "content"])
SOURCE_COL = find_column(
    ["source_file", "book", "source", "title"], required=False
)
PAGE_COL = find_column(
    ["pages", "page", "page_number", "page_no"], required=False
)
ID_COL = find_column(["chunk_id", "id"], required=False)

device = "cuda" if torch.cuda.is_available() else "cpu"

print("\nLoading embedding model and reranker...")
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2", device=device
)
reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2", device=device
)

def retrieve_passages(question, top_k=5, candidate_k=50, rrf_k=60):
    """
    Retrieve and rerank NCERT passages.
    Returns text, source/page metadata, and diagnostic scores.
    """
    question = str(question).strip()
    if not question:
        return []

    # 1. BM25 retrieval
    query_tokens = question.lower().split()
    bm25_scores = np.asarray(bm25_index.get_scores(query_tokens))
    bm25_ids = np.argsort(bm25_scores)[::-1][:candidate_k]

    # 2. Dense retrieval
    query_vector = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    _, dense_ids = faiss_index.search(
        query_vector, min(candidate_k, faiss_index.ntotal)
    )
    dense_ids = [int(i) for i in dense_ids[0] if i >= 0]

    # 3. Reciprocal Rank Fusion
    fusion_scores = {}

    for rank, idx in enumerate(bm25_ids, start=1):
        idx = int(idx)
        fusion_scores[idx] = fusion_scores.get(idx, 0.0) + 1 / (rrf_k + rank)

    for rank, idx in enumerate(dense_ids, start=1):
        fusion_scores[idx] = fusion_scores.get(idx, 0.0) + 1 / (rrf_k + rank)

    candidate_ids = sorted(
        fusion_scores, key=fusion_scores.get, reverse=True
    )[:candidate_k]

    # 4. CrossEncoder reranking
    pairs = [
        (question, str(chunks.iloc[idx][TEXT_COL]))
        for idx in candidate_ids
    ]
    scores = reranker.predict(pairs, show_progress_bar=False)

    ranked = sorted(
        zip(candidate_ids, scores),
        key=lambda pair: float(pair[1]),
        reverse=True
    )[:top_k]

    results = []
    for rank, (idx, score) in enumerate(ranked, start=1):
        row = chunks.iloc[int(idx)]

        result = {
            "rank": rank,
            "text": str(row[TEXT_COL]),
            "reranker_score": float(score),
            "rrf_score": float(fusion_scores[int(idx)]),
            "chunk_index": int(idx),
        }

        if SOURCE_COL:
            result["source_file"] = str(row[SOURCE_COL])
        if PAGE_COL:
            result["pages"] = str(row[PAGE_COL])
        if ID_COL:
            result["chunk_id"] = str(row[ID_COL])

        results.append(result)

    return results

# 5. Smoke test
question = "Why do traditional houses in Sikkim have sloping roofs?"
results = retrieve_passages(question, top_k=5)

print("\n=== RETRIEVAL SMOKE TEST ===")
print("Question:", question)
print("Results returned:", len(results))

for item in results:
    print(
        f"\n[{item['rank']}] reranker={item['reranker_score']:.4f} | "
        f"source={item.get('source_file', 'unknown')} | "
        f"pages={item.get('pages', 'unknown')}"
    )
    print(item["text"][:600].replace("\n", " "))
```

```python

# CELL 3 — CONNECT NCERT RETRIEVAL TO QWEN
# CPU-compatible diagnostic run. Saves progress after every question.

from google.colab import drive
drive.mount('/content/drive')

import os, time, json, gc
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
OUT_DIR = f"{ROOT}/05_EXPERIMENTS/child_benchmark_v1"
os.makedirs(OUT_DIR, exist_ok=True)

OUTPUT_PATH = f"{OUT_DIR}/qwen_integration_smoketest_v1.xlsx"
JSONL_PATH = f"{OUT_DIR}/qwen_integration_smoketest_v1_progress.jsonl"

# Reuse the already-loaded retriever and reranker from Cell 1.
required = ["retrieve_passages", "reranker", "embedding_model"]
missing = [name for name in required if name not in globals()]
if missing:
    raise RuntimeError(
        f"Missing {missing}. Run the retrieval setup cell successfully first."
    )

MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"
print("Device: CPU" if not torch.cuda.is_available() else "Device: GPU")
print("Loading Qwen; first load may take several minutes...")

if "qwen_model" not in globals() or "qwen_tokenizer" not in globals():
    qwen_tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    qwen_model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float32,
        low_cpu_mem_usage=True
    )
    qwen_model.to("cpu")
    qwen_model.eval()

print("Qwen loaded successfully.")

# Small diagnostic set: supported-style questions and likely abstentions.
TEST_QUESTIONS = [
    ("Q1", "What does the NCERT passage say about houses in Sikkim?"),
    ("Q2", "What materials are mentioned for making the homemade balance?"),
    ("Q3", "Why do traditional houses in Sikkim have sloping roofs?"),
    ("Q4", "What will the weather be like in Mumbai tomorrow?"),
    ("Q5", "Which birds are outside my house right now?"),
]

def build_evidence(question, top_k=3):
    passages = retrieve_passages(
        question, top_k=top_k, candidate_k=20
    )
    blocks = []
    for p in passages:
        source = p.get("source_file", "Unknown source")
        pages = p.get("pages", "Unknown page")
        blocks.append(
            f"[Source: {source}; pages: {pages}]\n{p['text']}"
        )
    return passages, "\n\n---\n\n".join(blocks)

def generate_grounded_answer(question, evidence):
    system_prompt = """
You are a kind AI tutor for a primary-school student.
Use ONLY the supplied NCERT evidence to answer.
Rules:
- Do not use outside knowledge or guess missing facts.
- A topic being mentioned is not proof that the passage explains it.
- If the passage does not directly answer the question, say:
  "I couldn't find enough information in the textbook passages I checked.
   Let's ask your teacher or look for another textbook passage."
- Never invent current weather, real-time observations, causes, or details.
- Keep the answer short, clear, and suitable for a child.
- Do not claim you verified something unless the evidence supports it.
"""
    user_prompt = f"""QUESTION:
{question}

NCERT EVIDENCE:
{evidence if evidence.strip() else "[No evidence retrieved]"}

Give a short answer grounded only in this evidence."""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ]

    prompt = qwen_tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = qwen_tokenizer(
        prompt, return_tensors="pt", truncation=True, max_length=3000
    )

    with torch.inference_mode():
        output = qwen_model.generate(
            **inputs,
            max_new_tokens=110,
            do_sample=False,
            repetition_penalty=1.08,
            pad_token_id=qwen_tokenizer.eos_token_id,
            use_cache=True
        )

    new_tokens = output[0][inputs["input_ids"].shape[1]:]
    return qwen_tokenizer.decode(
        new_tokens, skip_special_tokens=True
    ).strip()

# Resume completed questions if the cell is interrupted.
completed = {}
if os.path.exists(JSONL_PATH):
    with open(JSONL_PATH, "r", encoding="utf-8") as f:
        for line in f:
            try:
                item = json.loads(line)
                completed[item["question_id"]] = item
            except Exception:
                pass

for qid, question in TEST_QUESTIONS:
    if qid in completed:
        print(f"Skipping already completed {qid}")
        continue

    start = time.time()
    print(f"\nRunning {qid}: {question}")

    try:
        passages, evidence = build_evidence(question, top_k=3)
        answer = generate_grounded_answer(question, evidence)

        record = {
            "question_id": qid,
            "question": question,
            "generated_answer": answer,
            "evidence_passages": len(passages),
            "retrieved_sources": " | ".join(
                f"{p.get('source_file', 'Unknown')} "
                f"(pages {p.get('pages', 'Unknown')})"
                for p in passages
            ),
            "retrieved_evidence": evidence,
            "elapsed_seconds": round(time.time() - start, 1),
            "status": "OK"
        }

    except Exception as e:
        record = {
            "question_id": qid,
            "question": question,
            "generated_answer": "",
            "evidence_passages": 0,
            "retrieved_sources": "",
            "retrieved_evidence": "",
            "elapsed_seconds": round(time.time() - start, 1),
            "status": f"ERROR: {type(e).__name__}: {e}"
        }

    completed[qid] = record

    # Incremental Drive checkpoint so completed work isn't lost.
    with open(JSONL_PATH, "a", encoding="utf-8") as f:
        f.write(json.dumps(record, ensure_ascii=False) + "\n")

    print("Status:", record["status"])
    print("Time:", record["elapsed_seconds"], "seconds")
    print("Answer:", record["generated_answer"][:700])

    # Refresh the workbook after every question.
    pd.DataFrame(completed.values()).to_excel(OUTPUT_PATH, index=False)
    gc.collect()

print("\n=== SMOKE TEST COMPLETE ===")
print("Workbook:", OUTPUT_PATH)
print("Progress log:", JSONL_PATH)
print(pd.DataFrame(completed.values())[
    ["question_id", "status", "elapsed_seconds", "generated_answer"]
].to_string(index=False))
```

```python

# CELL 4 — CPU-FRIENDLY, EVIDENCE-FIRST TUTOR RESPONSE
from google.colab import drive
drive.mount('/content/drive')

import re
import os
import json
import pandas as pd

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
OUT_DIR = f"{ROOT}/05_EXPERIMENTS/child_benchmark_v1"
os.makedirs(OUT_DIR, exist_ok=True)

assert "retrieve_passages" in globals(), (
    "Run the retrieval setup cell first."
)

REALTIME_PATTERNS = [
    r"\bright now\b", r"\btoday\b", r"\btomorrow\b",
    r"\bcurrent(?:ly)?\b", r"\bweather\b",
    r"\boutside my\b", r"\bnear my house\b",
    r"\bwhat am i\b", r"\bwhere am i\b"
]

CAUSAL_MARKERS = [
    "because", "so that", "in order to", "to prevent",
    "this helps", "therefore", "as a result", "due to"
]

def split_sentences(text):
    text = re.sub(r"\s+", " ", str(text)).strip()
    return [
        s.strip() for s in re.split(r"(?<=[.!?])\s+", text)
        if len(s.strip()) > 25
    ]

def tutor_response(question, top_k=5):
    question = str(question).strip()

    if not question:
        return {
            "answer": "Ask me a question about your NCERT lessons!",
            "status": "EMPTY_QUESTION",
            "sources": [],
            "evidence": []
        }

    # Avoid pretending that static textbooks provide live information.
    if any(re.search(p, question, re.I) for p in REALTIME_PATTERNS):
        return {
            "answer": (
                "I can't know that from the NCERT textbook passages. "
                "Let's find a source that can answer this question."
            ),
            "status": "REAL_TIME_OR_PERSONAL_QUESTION",
            "sources": [],
            "evidence": []
        }

    passages = retrieve_passages(
        question, top_k=top_k, candidate_k=20
    )

    if not passages:
        return {
            "answer": "I couldn't find a relevant textbook passage. Let's look for another one.",
            "status": "NO_EVIDENCE",
            "sources": [],
            "evidence": []
        }

    # Rank sentences using word overlap as a transparent baseline.
    stopwords = {
        "what", "when", "where", "which", "who", "whom", "whose",
        "why", "how", "does", "do", "did", "is", "are", "was", "were",
        "the", "a", "an", "in", "on", "of", "for", "to", "and",
        "or", "with", "from", "that", "this", "have", "has", "it"
    }

    terms = {
        w.lower() for w in re.findall(r"[a-zA-Z]+", question)
        if len(w) > 2 and w.lower() not in stopwords
    }

    candidates = []
    for p in passages:
        for sentence in split_sentences(p["text"]):
            words = {
                w.lower() for w in re.findall(r"[a-zA-Z]+", sentence)
            }
            overlap = len(terms & words) / max(1, len(terms))
            if overlap > 0:
                candidates.append({
                    "sentence": sentence,
                    "overlap": overlap,
                    "source": p.get("source_file", "Unknown source"),
                    "pages": p.get("pages", "Unknown"),
                    "reranker_score": p.get("reranker_score")
                })

    candidates.sort(key=lambda x: x["overlap"], reverse=True)

    if not candidates or candidates[0]["overlap"] < 0.20:
        return {
            "answer": (
                "I couldn't find enough direct evidence in the passages "
                "I checked. Let's look for another part of the textbook."
            ),
            "status": "INSUFFICIENT_LEXICAL_EVIDENCE",
            "sources": [
                {"source": p.get("source_file"), "pages": p.get("pages")}
                for p in passages
            ],
            "evidence": []
        }

    # For causal questions, require an explicit causal cue in the
    # retrieved text before presenting any sentence as an explanation.
    if re.search(r"\bwhy\b|\bhow come\b|\bwhat causes\b", question, re.I):
        causal_candidates = [
            c for c in candidates
            if any(marker in c["sentence"].lower()
                   for marker in CAUSAL_MARKERS)
        ]
        if not causal_candidates:
            return {
                "answer": (
                    "The passages describe this topic, but I couldn't find "
                    "a clear explanation of why it happens. I don't want "
                    "to guess, so let's check another textbook passage."
                ),
                "status": "CAUSAL_EXPLANATION_NOT_FOUND",
                "sources": [
                    {"source": p.get("source_file"), "pages": p.get("pages")}
                    for p in passages
                ],
                "evidence": [c["sentence"] for c in candidates[:3]]
            }
        candidates = causal_candidates

    # Quote, rather than paraphrase, the selected source sentence.
    selected = []
    seen = set()
    for c in candidates:
        key = c["sentence"].lower()
        if key not in seen:
            selected.append(c)
            seen.add(key)
        if len(selected) == 2:
            break

    if not selected:
        return {
            "answer": "I couldn't find enough direct evidence to answer safely.",
            "status": "INSUFFICIENT_EVIDENCE",
            "sources": [],
            "evidence": []
        }

    answer = "Here is what the textbook passage says:\n" + "\n".join(
        f"• “{c['sentence']}”" for c in selected
    )

    return {
        "answer": answer,
        "status": "EXTRACTIVE_EVIDENCE_RESPONSE",
        "sources": [
            {
                "source": c["source"],
                "pages": c["pages"]
            } for c in selected
        ],
        "evidence": [c["sentence"] for c in selected]
    }

# Quick tests — retrieval only, no Qwen generation.
TESTS = [
    "What does the NCERT passage say about houses in Sikkim?",
    "What materials are mentioned for making the homemade balance?",
    "Why do traditional houses in Sikkim have sloping roofs?",
    "What will the weather be like in Mumbai tomorrow?",
    "Which birds are outside my house right now?"
]

rows = []
for q in TESTS:
    print("\n" + "=" * 70)
    print("QUESTION:", q)
    result = tutor_response(q)
    print("STATUS:", result["status"])
    print("ANSWER:", result["answer"])
    print("SOURCES:", result["sources"])

    rows.append({
        "question": q,
        "status": result["status"],
        "answer": result["answer"],
        "sources_json": json.dumps(result["sources"], ensure_ascii=False),
        "evidence_json": json.dumps(result["evidence"], ensure_ascii=False)
    })

out_path = f"{OUT_DIR}/evidence_first_response_smoketest_v1.xlsx"
pd.DataFrame(rows).to_excel(out_path, index=False)
print("\nSaved:", out_path)
```

```python

# CELL 6 — TARGETED RERANKING DIAGNOSTIC
from google.colab import drive
drive.mount('/content/drive')

import re
import pandas as pd

assert "retrieve_passages" in globals()

TESTS = [
    {
        "question": "What materials are mentioned for making the homemade balance?",
        "keywords": ["cup", "thread", "hanger", "balance"],
    },
    {
        "question": "What does the NCERT passage say about houses in Sikkim?",
        "keywords": ["wooden houses", "sloping roofs", "housing"],
    },
]

rows = []

for test in TESTS:
    question = test["question"]
    terms = test["keywords"]

    # Retrieve a broader candidate pool, without changing saved indexes.
    passages = retrieve_passages(
        question,
        top_k=10,
        candidate_k=50
    )

    scored = []
    for p in passages:
        text = p["text"].lower()
        hits = sum(1 for term in terms if term.lower() in text)
        score = p["reranker_score"]

        scored.append({
            **p,
            "keyword_hits": hits,
            "diagnostic_score": score + 0.75 * hits
        })

    # Show the existing ranking and diagnostic ranking side by side.
    baseline = sorted(scored, key=lambda x: x["reranker_score"], reverse=True)
    diagnostic = sorted(
        scored, key=lambda x: x["diagnostic_score"], reverse=True
    )

    print("\n" + "=" * 80)
    print("QUESTION:", question)

    for label, ranking in [
        ("BASELINE", baseline),
        ("DIAGNOSTIC", diagnostic)
    ]:
        print("\n", label)
        for rank, p in enumerate(ranking[:5], start=1):
            print(
                f"{rank}. keyword_hits={p['keyword_hits']} "
                f"reranker={p['reranker_score']:.3f} "
                f"source={p.get('source_file')} "
                f"pages={p.get('pages')}"
            )
            print(p["text"][:450].replace("\n", " "))

            rows.append({
                "question": question,
                "ranking_method": label,
                "rank": rank,
                "keyword_hits": p["keyword_hits"],
                "reranker_score": p["reranker_score"],
                "diagnostic_score": p["diagnostic_score"],
                "source_file": p.get("source_file", ""),
                "pages": p.get("pages", ""),
                "text": p["text"]
            })

out_path = (
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/child_benchmark_v1/"
    "targeted_reranking_diagnostic_v1.xlsx"
)
pd.DataFrame(rows).to_excel(out_path, index=False)
print("\nSaved:", out_path)
```

```python

# CELL 7 — LONG-RUN RETRIEVAL + QWEN GROUNDING EXPERIMENT
# CPU-compatible. Checkpoints every answer. Safe to rerun after interruption.

from google.colab import drive
drive.mount('/content/drive')

import os, re, json, time, gc
import pandas as pd
import torch

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
OUT_DIR = f"{ROOT}/05_EXPERIMENTS/child_benchmark_v1"
os.makedirs(OUT_DIR, exist_ok=True)

PROGRESS_PATH = f"{OUT_DIR}/long_qwen_grounding_progress_v1.jsonl"
XLSX_PATH = f"{OUT_DIR}/long_qwen_grounding_experiment_v1.xlsx"

assert "retrieve_passages" in globals(), "Run retrieval setup first."
assert "qwen_model" in globals() and "qwen_tokenizer" in globals(), (
    "Qwen must already be loaded. Run Cell 3 successfully first."
)

# Deliberately small diagnostic set: known answerable, difficult, and negative.
CASES = [
    {
        "id": "L01",
        "q": "What materials are mentioned for making the homemade balance?",
        "hints": ["cup", "thread", "hanger", "balance"],
        "kind": "answerable"
    },
    {
        "id": "L02",
        "q": "What does the NCERT passage say about houses in Sikkim?",
        "hints": ["wooden houses", "sloping roofs", "Sikkim", "housing"],
        "kind": "answerable"
    },
    {
        "id": "L03",
        "q": "Why do traditional houses in Sikkim have sloping roofs?",
        "hints": ["why", "traditional houses", "Sikkim", "sloping roofs"],
        "kind": "causal_unknown"
    },
    {
        "id": "L04",
        "q": "What will the weather be like in Mumbai tomorrow?",
        "hints": ["weather", "Mumbai", "tomorrow"],
        "kind": "realtime"
    },
    {
        "id": "L05",
        "q": "Which birds are outside my house right now?",
        "hints": ["birds", "outside", "right now"],
        "kind": "realtime"
    },
    {
        "id": "L06",
        "q": "How can children compare two objects to find which is heavier?",
        "hints": ["compare", "objects", "heavier", "cup", "thread", "hanger"],
        "kind": "answerable"
    },
    {
        "id": "L07",
        "q": "What are Loosong and Namsoong?",
        "hints": ["Loosong", "Namsoong", "festival", "Sikkim"],
        "kind": "answerable"
    },
    {
        "id": "L08",
        "q": "What is the exact angle in degrees of a traditional Sikkim roof?",
        "hints": ["exact angle", "degrees", "roof", "Sikkim"],
        "kind": "unknown_detail"
    }
]

# Resume using the progress file; each question/mode is saved independently.
completed = {}
if os.path.exists(PROGRESS_PATH):
    with open(PROGRESS_PATH, "r", encoding="utf-8") as f:
        for line in f:
            try:
                r = json.loads(line)
                completed[(r["case_id"], r["mode"])] = r
            except Exception:
                pass

def lexical_score(text, hints):
    text = text.lower()
    return sum(1 for term in hints if term.lower() in text)

def choose_evidence(passages, case, mode):
    if mode == "baseline":
        chosen = passages[:3]
    else:
        # Diagnostic only: compare normal reranking against manually
        # specified concepts for these eight known test questions.
        ranked = sorted(
            passages,
            key=lambda p: (
                lexical_score(p["text"], case["hints"]),
                p.get("reranker_score", -999)
            ),
            reverse=True
        )
        chosen = ranked[:3]

    blocks = []
    for p in chosen:
        blocks.append(
            f"[Source: {p.get('source_file', 'Unknown')}; "
            f"pages: {p.get('pages', 'Unknown')}]\n{p['text']}"
        )
    return chosen, "\n\n---\n\n".join(blocks)

def qwen_answer(question, evidence):
    system = """
You are a careful NCERT tutor for children in Classes 1–4.
Use only the supplied passages. Do not use outside knowledge.
Do not infer causes merely because two topics appear together.
Do not invent materials, events, observations, weather, or measurements.
If the passages do not explicitly answer the question, say so clearly.
Use simple language and keep the answer to at most 3 short sentences.
Cite the source/page when possible using the metadata supplied.
"""
    messages = [
        {"role": "system", "content": system},
        {"role": "user", "content":
            f"QUESTION:\n{question}\n\n"
            f"SUPPLIED NCERT PASSAGES:\n{evidence}\n\n"
            "Answer only if the passages support the answer."}
    ]
    prompt = qwen_tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = qwen_tokenizer(
        prompt, return_tensors="pt", truncation=True, max_length=2600
    )
    with torch.inference_mode():
        out = qwen_model.generate(
            **inputs,
            max_new_tokens=90,
            do_sample=False,
            repetition_penalty=1.08,
            pad_token_id=qwen_tokenizer.eos_token_id,
            use_cache=True
        )
    answer_tokens = out[0][inputs["input_ids"].shape[1]:]
    return qwen_tokenizer.decode(
        answer_tokens, skip_special_tokens=True
    ).strip()

def save_checkpoint():
    records = list(completed.values())
    pd.DataFrame(records).to_excel(XLSX_PATH, index=False)

start_all = time.time()
total_jobs = len(CASES) * 2

print("Device:", "GPU" if torch.cuda.is_available() else "CPU")
print("Cases:", len(CASES), "| maximum jobs:", total_jobs)
print("Already completed:", len(completed))
print("Checkpoint:", PROGRESS_PATH)
print("This can take a while on CPU; each result is saved incrementally.")

for i, case in enumerate(CASES, start=1):
    # Retrieve once per case, reuse candidates for both modes.
    try:
        passages = retrieve_passages(
            case["q"], top_k=10, candidate_k=50
        )
    except Exception as e:
        print(f"Retrieval failed for {case['id']}: {e}")
        continue

    for mode in ["baseline", "hint_rerank"]:
        key = (case["id"], mode)
        if key in completed:
            print("Already done:", key)
            continue

        t0 = time.time()
        selected, evidence = choose_evidence(passages, case, mode)

        # Deterministic rejection for real-time questions; don't waste
        # CPU generation on questions NCERT cannot answer.
        if case["kind"] == "realtime":
            answer = (
                "I can't determine real-time weather or what is outside "
                "your home from textbook passages."
            )
            status = "REALTIME_ABSTENTION"
        else:
            try:
                answer = qwen_answer(case["q"], evidence)
                status = "OK"
            except Exception as e:
                answer = ""
                status = f"ERROR: {type(e).__name__}: {e}"

        record = {
            "case_id": case["id"],
            "question": case["q"],
            "case_type": case["kind"],
            "mode": mode,
            "status": status,
            "answer": answer,
            "sources": " | ".join(
                f"{p.get('source_file', '')} pages={p.get('pages', '')}"
                for p in selected
            ),
            "selected_evidence": evidence,
            "elapsed_seconds": round(time.time() - t0, 1),
            "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
        }
        completed[key] = record

        # Append immediately; if Colab disconnects, completed jobs survive.
        with open(PROGRESS_PATH, "a", encoding="utf-8") as f:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

        save_checkpoint()
        gc.collect()

        print(
            f"\n[{len(completed)}/{total_jobs}] {case['id']} | {mode} | "
            f"{status} | {record['elapsed_seconds']}s"
        )
        print("Q:", case["q"])
        print("A:", answer[:450])

print("\n=== EXPERIMENT FINISHED ===")
print("Total elapsed minutes:", round((time.time() - start_all) / 60, 1))
print("Completed jobs:", len(completed), "/", total_jobs)
print("Workbook:", XLSX_PATH)
print("Progress log:", PROGRESS_PATH)

if completed:
    summary = pd.DataFrame(completed.values())
    print(summary[
        ["case_id", "mode", "status", "elapsed_seconds", "answer"]
    ].to_string(index=False))
```

```python

# CELL 8 — EXTENDED QWEN GROUNDING + INDEPENDENT CLAIM CHECK
# CPU-compatible, resumable, saves progress after each prompt variant.
# NLI scores are diagnostic, NOT a guarantee of factual correctness.

from google.colab import drive
drive.mount('/content/drive')

import os, re, json, time, gc
import pandas as pd
import numpy as np
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

ROOT = "/content/drive/MyDrive/NCERT_BOOKS"
OUT_DIR = f"{ROOT}/05_EXPERIMENTS/child_benchmark_v1"
os.makedirs(OUT_DIR, exist_ok=True)

PROGRESS = f"{OUT_DIR}/extended_grounding_progress_v1.jsonl"
RESULT_XLSX = f"{OUT_DIR}/extended_grounding_results_v1.xlsx"
CLAIMS_XLSX = f"{OUT_DIR}/extended_grounding_claim_checks_v1.xlsx"

assert "retrieve_passages" in globals(), "Run retrieval setup first."
assert "qwen_model" in globals() and "qwen_tokenizer" in globals(), (
    "Load Qwen using Cell 3 first."
)

# Use benchmark questions when available, supplemented with known diagnostics.
benchmark_path = (
    f"{ROOT}/05_EXPERIMENTS/retrieval_evaluation/english_only_v1/"
    "child_perspective_ncert_benchmark_v1.xlsx"
)

if os.path.exists(benchmark_path):
    bench = pd.read_excel(benchmark_path)
    print("Benchmark columns:", bench.columns.tolist())
else:
    bench = pd.DataFrame()

# Stable, targeted set so runs are comparable.
FALLBACK_CASES = [
    ("T01", "What materials are mentioned for making the homemade balance?", "answerable"),
    ("T02", "What does the NCERT passage say about houses in Sikkim?", "answerable"),
    ("T03", "Why do traditional houses in Sikkim have sloping roofs?", "unknown"),
    ("T04", "What will the weather be like in Mumbai tomorrow?", "realtime"),
    ("T05", "Which birds are outside my house right now?", "realtime"),
    ("T06", "What are Loosong and Namsoong?", "answerable"),
    ("T07", "What is the exact angle in degrees of a traditional Sikkim roof?", "unknown"),
    ("T08", "How can children compare two objects to find which is heavier?", "answerable"),
]

# If benchmark is readable, prefer its child-perspective questions.
def get_cases():
    if not bench.empty and {"question", "answerability"}.issubset(bench.columns):
        df = bench.copy()
        df["question"] = df["question"].astype(str)
        df = df[df["question"].str.strip().ne("")]

        ans_col = df["answerability"].astype(str).str.upper()
        answerable = df[ans_col.str.contains("ANSWERABLE", na=False)
                        & ~ans_col.str.contains("UNANSWERABLE", na=False)]
        unanswerable = df[ans_col.str.contains("UNANSWERABLE", na=False)]

        selected = []
        for i, (_, r) in enumerate(answerable.head(4).iterrows(), 1):
            selected.append((f"B-A{i}", r["question"], "answerable"))
        for i, (_, r) in enumerate(unanswerable.head(4).iterrows(), 1):
            selected.append((f"B-U{i}", r["question"], "unknown"))
        if len(selected) >= 6:
            return selected[:8]
    return FALLBACK_CASES

CASES = get_cases()
print("Questions selected:", len(CASES))
for x in CASES:
    print(x[0], x[2], x[1])

# Two prompt variants:
# A = conventional grounded generation.
# B = evidence-first, conservative answer-or-abstain.
PROMPTS = {
    "A_standard": (
        "You are a friendly NCERT tutor for Classes 1-4. "
        "Answer the question using the supplied textbook passages. "
        "Use simple language and cite the supplied source/page. "
        "Do not invent facts."
    ),
    "B_conservative": (
        "You are a strict evidence-only NCERT tutor for Classes 1-4. "
        "Every factual statement must be directly supported by the supplied "
        "passages. A question, heading, or mention of a topic does not prove "
        "its explanation. Never guess causes, measurements, current events, "
        "or personal observations. If evidence is insufficient, explicitly "
        "say you could not find the answer in these passages. Prefer a short "
        "answer over a detailed unsupported answer. Cite source/page."
    )
}

# Resume safely from previous completed cases.
completed = {}
if os.path.exists(PROGRESS):
    with open(PROGRESS, "r", encoding="utf-8") as f:
        for line in f:
            try:
                r = json.loads(line)
                completed[(r["case_id"], r["prompt_variant"])] = r
            except Exception:
                pass

# Load independent NLI model on CPU.
# Its scores are useful warning signals only; previous experiments showed
# both false acceptance and false rejection.
NLI_NAME = "cross-encoder/nli-deberta-v3-base"
print("\nLoading NLI diagnostic model on CPU...")
from sentence_transformers import CrossEncoder as NLICrossEncoder

if "nli_diagnostic_model" not in globals():
    nli_diagnostic_model = NLICrossEncoder(
        NLI_NAME, device="cpu", max_length=384
    )

nli_labels = nli_diagnostic_model.model.config.id2label
print("NLI label map:", nli_labels)

def make_evidence(question):
    passages = retrieve_passages(question, top_k=5, candidate_k=30)
    blocks = []
    for p in passages:
        blocks.append(
            f"[Source: {p.get('source_file', 'Unknown')}; "
            f"pages: {p.get('pages', 'Unknown')}]\n{p['text']}"
        )
    return passages, "\n\n---\n\n".join(blocks)

def answer_with_prompt(question, evidence, prompt_text):
    # Explicitly reject obvious real-time questions before generation.
    if re.search(
        r"\b(weather|tomorrow|right now|outside my house|currently)\b",
        question, re.I
    ):
        return (
            "I cannot determine real-time conditions from NCERT textbook "
            "passages. Please check a suitable current source."
        ), "RULE_BASED_ABSTENTION"

    messages = [
        {"role": "system", "content": prompt_text},
        {"role": "user", "content":
            f"QUESTION:\n{question}\n\n"
            f"NCERT PASSAGES:\n{evidence}\n\n"
            "Answer briefly using only the passages."}
    ]
    prompt = qwen_tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = qwen_tokenizer(
        prompt, return_tensors="pt", truncation=True, max_length=2500
    )
    with torch.inference_mode():
        output = qwen_model.generate(
            **inputs,
            max_new_tokens=110,
            do_sample=False,
            repetition_penalty=1.08,
            pad_token_id=qwen_tokenizer.eos_token_id,
            use_cache=True
        )
    new_tokens = output[0][inputs["input_ids"].shape[1]:]
    return qwen_tokenizer.decode(
        new_tokens, skip_special_tokens=True
    ).strip(), "GENERATED"

def split_claims(answer):
    # Lightweight sentence splitting; bullet/numbered lines are separated too.
    lines = re.split(r"[\n•]+", answer)
    claims = []
    for line in lines:
        line = re.sub(r"^\s*(?:[-*]|\d+[.)])\s*", "", line).strip()
        for sentence in re.split(r"(?<=[.!?])\s+", line):
            sentence = sentence.strip()
            if len(sentence.split()) >= 4:
                claims.append(sentence)
    return claims[:8]

def nli_check_claims(question_id, question, answer, passages):
    claims = split_claims(answer)
    output_rows = []

    if not claims or not passages:
        return output_rows

    # Compare each generated claim to each retrieved passage separately.
    pairs = []
    pair_meta = []
    for claim in claims:
        for p in passages[:5]:
            text = str(p["text"])[:1600]
            pairs.append((text, claim))
            pair_meta.append((claim, p))

    scores = nli_diagnostic_model.predict(
        pairs, batch_size=4, show_progress_bar=False
    )

    for (claim, p), raw in zip(pair_meta, scores):
        arr = np.asarray(raw, dtype=float).flatten()
        # Model labels are normally contradiction/entailment/neutral.
        entailment_idx = None
        contradiction_idx = None
        for idx, label in nli_labels.items():
            lab = str(label).lower()
            if "entail" in lab:
                entailment_idx = int(idx)
            elif "contrad" in lab:
                contradiction_idx = int(idx)

        # CrossEncoder may return logits; normalize to probabilities.
        shifted = arr - np.max(arr)
        probs = np.exp(shifted) / np.exp(shifted).sum()
        ent = float(probs[entailment_idx]) if entailment_idx is not None else None
        contra = float(probs[contradiction_idx]) if contradiction_idx is not None else None

        output_rows.append({
            "case_id": question_id,
            "question": question,
            "generated_claim": claim,
            "source_file": p.get("source_file", ""),
            "pages": p.get("pages", ""),
            "nli_entailment_probability": ent,
            "nli_contradiction_probability": contra,
            "diagnostic_flag": (
                "LOW_ENTAILMENT_REVIEW"
                if ent is not None and ent < 0.35
                else "POSSIBLE_SUPPORT_REVIEW"
            ),
            "passage": str(p["text"])[:1600]
        })
    return output_rows

all_claim_rows = []
total = len(CASES) * len(PROMPTS)
start_all = time.time()

for case_id, question, expected_type in CASES:
    try:
        passages, evidence = make_evidence(question)
    except Exception as e:
        print("Retrieval error:", case_id, repr(e))
        continue

    for variant, prompt_text in PROMPTS.items():
        key = (case_id, variant)
        if key in completed:
            print("Skipping completed:", key)
            continue

        t0 = time.time()
        try:
            answer, status = answer_with_prompt(
                question, evidence, prompt_text
            )
        except Exception as e:
            answer = ""
            status = f"ERROR: {type(e).__name__}: {e}"

        record = {
            "case_id": case_id,
            "question": question,
            "expected_type": expected_type,
            "prompt_variant": variant,
            "status": status,
            "answer": answer,
            "sources": " | ".join(
                f"{p.get('source_file', '')} pages={p.get('pages', '')}"
                for p in passages
            ),
            "evidence": evidence,
            "elapsed_seconds": round(time.time() - t0, 1)
        }

        completed[key] = record
        with open(PROGRESS, "a", encoding="utf-8") as f:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

        # Save generation results immediately.
        pd.DataFrame(completed.values()).to_excel(RESULT_XLSX, index=False)

        # Independently score answer claims against retrieved passages.
        if answer and status == "GENERATED":
            try:
                claim_rows = nli_check_claims(
                    case_id, question, answer, passages
                )
                all_claim_rows.extend(claim_rows)
                pd.DataFrame(all_claim_rows).to_excel(CLAIMS_XLSX, index=False)
            except Exception as e:
                print("NLI diagnostic failed:", repr(e))

        gc.collect()
        print(
            f"\n[{len(completed)}/{total}] {case_id} {variant} "
            f"| {status} | {record['elapsed_seconds']}s"
        )
        print("Q:", question)
        print("A:", answer[:500])

print("\n=== GROUNDING EXPERIMENT COMPLETE ===")
print("Elapsed minutes:", round((time.time() - start_all) / 60, 1))
print("Completed prompt runs:", len(completed), "/", total)
print("Results workbook:", RESULT_XLSX)
print("Claim diagnostics:", CLAIMS_XLSX)
print("Progress log:", PROGRESS)

if completed:
    print(pd.DataFrame(completed.values())[
        ["case_id", "prompt_variant", "status", "elapsed_seconds", "answer"]
    ].to_string(index=False))
```

```python

# ============================================================
# CELL 9 — ROBUST NCERT GROUNDING EXPERIMENT
# Self-loads Qwen and retrieval artifacts if needed.
# A: Direct generation
# B: Evidence-first generation
# NLI: Diagnostic claim checks on B answers only
#
# Checkpoints every strategy run to Google Drive.
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os, re, json, time, pickle, traceback
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import torch

ROOT = Path('/content/drive/MyDrive/NCERT_BOOKS')
RETR = ROOT / '05_EXPERIMENTS/retrieval_evaluation/english_only_v1'
OUT = ROOT / '05_EXPERIMENTS/child_benchmark_v1'
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now().strftime('%Y%m%d_%H%M%S')
CSV_PATH = OUT / f'grounding_v2_{RUN_ID}.csv'
LOG_PATH = OUT / f'grounding_v2_{RUN_ID}.jsonl'
XLSX_PATH = OUT / f'grounding_v2_{RUN_ID}.xlsx'
CONFIG_PATH = OUT / f'grounding_v2_{RUN_ID}_config.json'

print("RUN:", RUN_ID)
print("Torch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("Device:", "cuda" if torch.cuda.is_available() else "cpu")

# ------------------------------------------------------------
# 1. Load Qwen if the previous variables are unavailable
# ------------------------------------------------------------
def find_existing_model():
    # Reuse only if both objects exist and are compatible.
    for mn, tn in [
        ('model', 'tokenizer'),
        ('qwen_model', 'qwen_tokenizer'),
    ]:
        m = globals().get(mn)
        t = globals().get(tn)
        if m is not None and t is not None and hasattr(m, 'generate'):
            print(f"Reusing in-memory objects: {mn}, {tn}")
            return m, t
    return None, None

qwen_model, qwen_tokenizer = find_existing_model()

if qwen_model is None:
    print("\nLoading Qwen/Qwen2.5-1.5B-Instruct...")
    from transformers import AutoTokenizer, AutoModelForCausalLM

    MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"
    qwen_tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

    qwen_model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        torch_dtype=torch.float32 if not torch.cuda.is_available() else torch.float16,
        low_cpu_mem_usage=True
    )

    device = "cuda" if torch.cuda.is_available() else "cpu"
    qwen_model = qwen_model.to(device)
    qwen_model.eval()

    print("Qwen loaded on:", device)
else:
    qwen_model.eval()
    try:
        print("Reused Qwen device:", next(qwen_model.parameters()).device)
    except Exception:
        pass

# ------------------------------------------------------------
# 2. Load/rebuild retrieval stack from Drive
# ------------------------------------------------------------
# These are the exact expected artifacts from the earlier setup.
chunks_path = RETR / 'english_only_chunks.csv'
bm25_path = RETR / 'indexes/bm25/bm25_index.pkl'
tokens_path = RETR / 'indexes/bm25/tokenized_documents.pkl'
emb_path = RETR / 'indexes/embeddings/chunk_embeddings.npy'
faiss_path = RETR / 'indexes/faiss/index.faiss'

required_paths = [
    chunks_path, bm25_path, tokens_path, emb_path, faiss_path
]
missing = [str(p) for p in required_paths if not p.exists()]

if missing:
    print("\nExpected artifact paths not all found. Searching retrieval folder...")
    for p in RETR.rglob('*'):
        if p.is_file() and p.suffix.lower() in {'.csv', '.pkl', '.npy', '.faiss'}:
            print("FOUND:", p)
    raise FileNotFoundError(
        "Required retrieval artifact(s) missing. Check the printed paths and "
        "update the five artifact paths in Section 2."
    )

print("\nLoading retrieval artifacts...")
chunks_df = pd.read_csv(chunks_path)
with open(bm25_path, 'rb') as f:
    bm25_index = pickle.load(f)
with open(tokens_path, 'rb') as f:
    tokenized_documents = pickle.load(f)
embeddings = np.load(emb_path)

import faiss
faiss_index = faiss.read_index(str(faiss_path))

from sentence_transformers import SentenceTransformer, CrossEncoder

EMBED_MODEL_ID = 'sentence-transformers/all-MiniLM-L6-v2'
RERANK_MODEL_ID = 'cross-encoder/ms-marco-MiniLM-L-6-v2'

# Reuse an existing compatible model where possible.
embedder = globals().get('embedder')
if embedder is None or not hasattr(embedder, 'encode'):
    print("Loading embedding model...")
    embedder = SentenceTransformer(EMBED_MODEL_ID)

reranker = globals().get('reranker')
if reranker is None or not hasattr(reranker, 'predict'):
    print("Loading cross-encoder reranker...")
    reranker = CrossEncoder(RERANK_MODEL_ID)

# Detect the actual text column rather than silently assuming one.
text_candidates = ['text', 'chunk_text', 'content', 'passage', 'chunk']
TEXT_COL = next((c for c in text_candidates if c in chunks_df.columns), None)
if TEXT_COL is None:
    raise ValueError(f"No text column found in chunks CSV: {list(chunks_df.columns)}")

print("Chunk table:", chunks_df.shape)
print("Text column:", TEXT_COL)
print("BM25 documents:", len(tokenized_documents))
print("Embedding shape:", embeddings.shape)
print("FAISS vectors:", faiss_index.ntotal)

if len(chunks_df) != len(tokenized_documents) or len(chunks_df) != embeddings.shape[0] or len(chunks_df) != faiss_index.ntotal:
    raise ValueError(
        "Artifact alignment mismatch: chunks, BM25 docs, embeddings and FAISS "
        "must have the same row count. Stop rather than cite incorrect sources."
    )

# ------------------------------------------------------------
# 3. Robust hybrid retrieval + reranking
# ------------------------------------------------------------
def tokenize_query(s):
    return re.findall(r'\b\w+\b', str(s).lower())

def row_value(row, candidates, default=''):
    for c in candidates:
        if c in row.index:
            v = row[c]
            if pd.notna(v) and str(v).strip():
                return v
    return default

def retrieve_passages(question, top_k=5, candidate_k=50, rrf_k=60):
    query_tokens = tokenize_query(question)

    bm_scores = np.asarray(bm25_index.get_scores(query_tokens))
    bm_ids = np.argsort(bm_scores)[::-1][:candidate_k]

    q_emb = embedder.encode(
        [question], convert_to_numpy=True, normalize_embeddings=True
    ).astype('float32')
    dense_scores, dense_ids = faiss_index.search(
        q_emb, min(candidate_k, faiss_index.ntotal)
    )
    dense_ids = dense_ids[0].tolist()

    # Reciprocal Rank Fusion combines ranks, not incomparable raw scores.
    fused = {}
    for rank, idx in enumerate(bm_ids, start=1):
        fused[int(idx)] = fused.get(int(idx), 0) + 1.0 / (rrf_k + rank)
    for rank, idx in enumerate(dense_ids, start=1):
        if idx >= 0:
            fused[int(idx)] = fused.get(int(idx), 0) + 1.0 / (rrf_k + rank)

    candidate_ids = sorted(fused, key=fused.get, reverse=True)[:candidate_k]
    pairs = [(question, str(chunks_df.iloc[i][TEXT_COL])) for i in candidate_ids]
    ce_scores = reranker.predict(pairs, show_progress_bar=False)

    order = np.argsort(np.asarray(ce_scores))[::-1][:top_k]
    out = []
    for rank, pos in enumerate(order, start=1):
        idx = candidate_ids[int(pos)]
        row = chunks_df.iloc[idx]
        metadata = row.to_dict()

        out.append({
            'rank': rank,
            'text': str(row[TEXT_COL]),
            'chunk_index': int(idx),
            'chunk_id': str(row_value(row, ['chunk_id', 'id'], idx)),
            'book': str(row_value(row, ['book', 'book_title', 'source_file', 'title'])),
            'class': str(row_value(row, ['class', 'class_name', 'grade'])),
            'subject': str(row_value(row, ['subject'])),
            'page': str(row_value(row, ['page', 'page_number', 'pages'])),
            'source_file': str(row_value(row, ['source_file', 'filename', 'file'])),
            'reranker_score': float(ce_scores[int(pos)]),
            'rrf_score': float(fused[idx]),
            'metadata': metadata
        })
    return out

# Smoke-test retrieval before spending time generating answers.
smoke = retrieve_passages("What does the textbook say about plants?", top_k=2)
print("\nRetrieval smoke test:")
for s in smoke:
    print(
        s['rank'], s['book'], s['page'],
        s['text'][:180].replace('\n', ' ')
    )

# ------------------------------------------------------------
# 4. Load benchmark
# ------------------------------------------------------------
benchmark_candidates = [
    RETR / 'child_perspective_ncert_benchmark_v1.xlsx',
    OUT / 'child_perspective_ncert_benchmark_v1.xlsx'
]
benchmark_candidates += list(
    (ROOT / '05_EXPERIMENTS').rglob('child_perspective_ncert_benchmark_v1.xlsx')
)

benchmark_path = next((p for p in benchmark_candidates if p.exists()), None)
if benchmark_path is None:
    raise FileNotFoundError("Child-perspective benchmark workbook not found.")

bench = pd.read_excel(benchmark_path)
if not {'question', 'answerability'}.issubset(bench.columns):
    raise ValueError(f"Unexpected benchmark columns: {list(bench.columns)}")

bench['_group'] = bench['answerability'].astype(str).str.lower().map(
    lambda x: 'answerable'
    if ('answerable' in x or 'sufficient' in x)
    and 'unanswerable' not in x
    else 'unanswerable_or_uncertain'
)

# 12 questions total, balanced when possible; deterministic for reproducibility.
parts = []
for group, df in bench.groupby('_group', sort=True):
    n = min(6, len(df))
    if n:
        parts.append(df.sample(n=n, random_state=20261013))
sample = pd.concat(parts).drop_duplicates().sort_index().reset_index(drop=True)

if 'question_id' not in sample.columns:
    sample['question_id'] = [f'Q{i+1:03d}' for i in range(len(sample))]

print("\nBenchmark:", benchmark_path)
print("Questions selected:", len(sample))
print(sample[['question_id', 'answerability', 'question']].to_string(index=False))

# ------------------------------------------------------------
# 5. Generation prompts and deterministic missing-input handling
# ------------------------------------------------------------
REALTIME_RE = re.compile(
    r'\b(right now|at this moment|outside my house|outside my window|'
    r'tomorrow|weather tomorrow|will it rain|currently happening|'
    r'what can i hear|what am i looking at)\b', re.I
)
IMAGE_RE = re.compile(
    r'\b(picture i am looking at|image i am looking at|'
    r'photo i am looking at|this picture|this image)\b', re.I
)
ABSTAIN = "I couldn't find that answer in the textbook passages I have."

def forced_abstention(question):
    if REALTIME_RE.search(question):
        return (
            "I can't determine real-time information from NCERT textbook passages. "
            "Please use a suitable current source.",
            "real_time_or_external_information"
        )
    if IMAGE_RE.search(question):
        return (
            "I can't see the picture yet. Please upload it or type the question, "
            "and I'll help you solve it.",
            "missing_visual_input"
        )
    return None, None

def make_sources(passages):
    sources = []
    for i, p in enumerate(passages, start=1):
        sources.append({
            'source_id': f'S{i}',
            'book': p.get('book', ''),
            'class': p.get('class', ''),
            'subject': p.get('subject', ''),
            'page': p.get('page', ''),
            'source_file': p.get('source_file', ''),
            'chunk_id': p.get('chunk_id', ''),
            'chunk_index': p.get('chunk_index', ''),
            'reranker_score': p.get('reranker_score', ''),
            'excerpt': p.get('text', '')
        })
    return sources

def make_evidence(sources, max_chars=6500):
    blocks, count = [], 0
    for s in sources:
        block = (
            f"[{s['source_id']}] Book: {s['book']}; Class: {s['class']}; "
            f"Subject: {s['subject']}; Page: {s['page']}\n"
            f"Text: {s['excerpt']}\n"
        )
        if count + len(block) > max_chars:
            break
        blocks.append(block)
        count += len(block)
    return '\n'.join(blocks)

def generate(prompt, max_new_tokens=160):
    inputs = qwen_tokenizer(
        prompt, return_tensors='pt', truncation=True, max_length=3500
    )
    device = next(qwen_model.parameters()).device
    inputs = {k: v.to(device) for k, v in inputs.items()}
    n_input = inputs['input_ids'].shape[1]
    start = time.time()
    with torch.inference_mode():
        out = qwen_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.08,
            pad_token_id=qwen_tokenizer.eos_token_id,
            eos_token_id=qwen_tokenizer.eos_token_id
        )
    answer = qwen_tokenizer.decode(
        out[0][n_input:], skip_special_tokens=True
    ).strip()
    return answer, round(time.time() - start, 2)

def direct_prompt(q, evidence):
    return f"""You are a kind NCERT tutor for children in Classes 1–4.
Answer using the passages below. If the passages do not answer the question,
say so. Do not invent facts, numbers, causes, quotes or citations.
Use short, simple sentences.

PASSAGES:
{evidence}

QUESTION: {q}
ANSWER:"""

def evidence_prompt(q, evidence):
    return f"""You are an evidence-first NCERT tutor for children in Classes 1–4.
Use only facts explicitly stated in the passages. Do not use outside knowledge.
Do not infer causes from examples. Do not invent missing information.

Every factual sentence must end with the source ID that supports it, like [S1].
If no passage explicitly answers the question, say exactly:
"{ABSTAIN}"

Give at most 3 short, child-friendly sentences. Do not list extra examples
unless the passages explicitly contain them.

PASSAGES:
{evidence}

QUESTION: {q}
ANSWER:"""

def split_claims(text):
    return [
        s.strip() for s in re.split(r'(?<=[.!?])\s+|\n+', str(text))
        if len(s.strip()) > 12
    ]

# ------------------------------------------------------------
# 6. Optional NLI diagnostic: reuse existing model only
# ------------------------------------------------------------
nli_model = globals().get('nli_model', globals().get('nli', None))
nli_tokenizer = globals().get('nli_tokenizer', None)
nli_ready = nli_model is not None and nli_tokenizer is not None
print("\nNLI already loaded:", nli_ready)
print("NLI is diagnostic only; it does not determine whether to show an answer.")

def nli_check(claim, sources):
    if not nli_ready:
        return {'nli_status': 'NOT_LOADED', 'best_score': None, 'source_id': ''}
    try:
        pairs, ids = [], []
        for s in sources:
            if s.get('excerpt'):
                pairs.append((s['excerpt'][:1600], claim[:600]))
                ids.append(s['source_id'])
        if not pairs:
            return {'nli_status': 'NO_EVIDENCE', 'best_score': None, 'source_id': ''}

        enc = nli_tokenizer(
            [x[0] for x in pairs], [x[1] for x in pairs],
            padding=True, truncation=True, max_length=512,
            return_tensors='pt'
        )
        device = next(nli_model.parameters()).device
        enc = {k: v.to(device) for k, v in enc.items()}

        with torch.inference_mode():
            probs = torch.softmax(nli_model(**enc).logits, dim=-1).cpu().numpy()

        label_map = getattr(nli_model.config, 'id2label', {})
        ent_idx = next(
            (int(i) for i, lab in label_map.items()
             if 'entail' in str(lab).lower()),
            1
        )
        best_idx = int(np.argmax(probs[:, ent_idx]))
        score = float(probs[best_idx, ent_idx])
        return {
            'nli_status': 'DIAGNOSTIC_ONLY',
            'best_score': round(score, 4),
            'source_id': ids[best_idx]
        }
    except Exception as e:
        return {'nli_status': 'ERROR', 'best_score': None, 'source_id': str(e)[:100]}

# ------------------------------------------------------------
# 7. Run two generation strategies, checkpoint every answer
# ------------------------------------------------------------
records = []
config = {
    'run_id': RUN_ID,
    'benchmark': str(benchmark_path),
    'sample_size': len(sample),
    'strategies': ['A_direct', 'B_evidence_first'],
    'claim_nli': 'diagnostic only, existing model reused if available',
    'retrieval': 'BM25 + dense FAISS + RRF + cross-encoder',
    'limitations': [
        'Benchmark annotations are partly provisional.',
        'NLI is not a calibrated factuality detector.',
        'Citation correctness requires manual review.',
        'Automated output is not evidence of model safety.'
    ]
}
CONFIG_PATH.write_text(json.dumps(config, indent=2), encoding='utf-8')

for ix, row in sample.iterrows():
    qid = str(row.get('question_id', f'Q{ix+1:03d}'))
    question = str(row['question']).strip()
    label = str(row.get('answerability', ''))
    expected = str(row.get('expected_answer', ''))
    gold_pages = str(row.get('gold_pages', ''))

    print(f"\n[{ix+1}/{len(sample)}] {qid}: {question}")

    try:
        passages = retrieve_passages(question, top_k=5)
        sources = make_sources(passages)
        evidence = make_evidence(sources)
        ret_error = ''
    except Exception as e:
        passages, sources, evidence = [], [], ''
        ret_error = repr(e)

    forced, forced_reason = forced_abstention(question)

    for strategy, prompt_fn in [
        ('A_direct', direct_prompt),
        ('B_evidence_first', evidence_prompt)
    ]:
        start = time.time()
        gen_seconds = 0.0
        err = ''
        if forced is not None:
            answer, status = forced, 'RULE_BASED_ABSTENTION'
        elif ret_error:
            answer, status = ABSTAIN, 'RETRIEVAL_ERROR_ABSTENTION'
        elif not evidence.strip():
            answer, status = ABSTAIN, 'NO_EVIDENCE_ABSTENTION'
        else:
            try:
                answer, gen_seconds = generate(prompt_fn(question, evidence))
                if not answer:
                    answer, status = ABSTAIN, 'EMPTY_GENERATION_ABSTENTION'
                else:
                    status = 'GENERATED'
            except Exception as e:
                answer, status, err = ABSTAIN, 'GENERATION_ERROR_ABSTENTION', repr(e)

        # NLI diagnostics only for evidence-first generated answers.
        diagnostics = []
        if strategy == 'B_evidence_first' and status == 'GENERATED':
            for j, claim in enumerate(split_claims(answer), start=1):
                diagnostics.append({
                    'claim_index': j,
                    'claim': claim,
                    **nli_check(claim, sources)
                })

        rec = {
            'run_id': RUN_ID,
            'question_id': qid,
            'question': question,
            'answerability_label': label,
            'expected_answer': expected,
            'gold_pages': gold_pages,
            'strategy': strategy,
            'status': status,
            'abstention_reason': forced_reason or '',
            'answer': answer,
            'generation_seconds': gen_seconds,
            'elapsed_seconds': round(time.time() - start, 2),
            'retrieval_error': ret_error,
            'generation_error': err,
            'sources_json': json.dumps(sources, ensure_ascii=False),
            'evidence_text': evidence,
            'claim_diagnostics_json': json.dumps(diagnostics, ensure_ascii=False),
            'source_count': len(sources)
        }
        records.append(rec)

        # Durable checkpoint after every completed strategy.
        pd.DataFrame(records).to_csv(CSV_PATH, index=False)
        with open(LOG_PATH, 'a', encoding='utf-8') as f:
            f.write(json.dumps(rec, ensure_ascii=False) + '\n')

        print(f"  {strategy}: {status} ({gen_seconds:.1f}s)")
        print("  ", answer[:240].replace('\n', ' '))

# ------------------------------------------------------------
# 8. Export research workbook and run summary
# ------------------------------------------------------------
results = pd.DataFrame(records)
if results.empty:
    raise RuntimeError("No results generated. Check the progress log and errors.")

results['is_abstention'] = (
    results['status'].astype(str).str.contains('ABSTENTION')
    | results['answer'].astype(str).str.contains(
        r"couldn't find that answer|can't determine real-time|can't see the picture",
        case=False, regex=True
    )
)

summary = results.groupby('strategy').agg(
    questions=('question_id', 'nunique'),
    total_runs=('question_id', 'size'),
    abstentions=('is_abstention', 'sum'),
    mean_generation_seconds=('generation_seconds', 'mean'),
    median_generation_seconds=('generation_seconds', 'median'),
    runs_with_retrieved_sources=('source_count', lambda x: int((x > 0).sum()))
).reset_index()

source_rows = []
claim_rows = []
for _, r in results.iterrows():
    for s in json.loads(r['sources_json']):
        source_rows.append({
            'question_id': r['question_id'],
            'question': r['question'],
            'strategy': r['strategy'],
            'answer': r['answer'],
            **s
        })
    for c in json.loads(r['claim_diagnostics_json']):
        claim_rows.append({
            'question_id': r['question_id'],
            'question': r['question'],
            'strategy': r['strategy'],
            **c
        })

with pd.ExcelWriter(XLSX_PATH, engine='openpyxl') as writer:
    results.to_excel(writer, sheet_name='all_results', index=False)
    summary.to_excel(writer, sheet_name='strategy_summary', index=False)
    pd.DataFrame(source_rows).to_excel(writer, sheet_name='source_audit', index=False)
    pd.DataFrame(claim_rows).to_excel(writer, sheet_name='claim_diagnostics', index=False)

print("\n========== CELL 9 COMPLETE ==========")
print("Questions:", results['question_id'].nunique())
print("Generation runs:", len(results))
print("\nSummary:")
print(summary.to_string(index=False))
print("\nCSV:", CSV_PATH)
print("Workbook:", XLSX_PATH)
print("Progress log:", LOG_PATH)
print("Config:", CONFIG_PATH)
```

```python

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pandas as pd
import json, re, time

assert "retrieve_passages" in globals(), (
    "retrieve_passages is not in memory. Run the retrieval-loading cell first."
)

OUT = Path(
    "/content/drive/MyDrive/NCERT_BOOKS/"
    "05_EXPERIMENTS/child_benchmark_v1"
)
OUT.mkdir(parents=True, exist_ok=True)

# Small diagnostic set: known grounding failures + useful-answer cases.
TESTS = [
    ("E01", "Why did the star tell the friends to stop fighting?"),
    ("E02", "Can you help me find fruit names hidden in the word grid?"),
    ("E03", "Which wall made from stacked paper or clay bricks is sturdier?"),
    ("E04", "Which traditional game uses seven stones and a ball?"),
    ("E05", "How can I make a dance performance more fun using things I make myself?"),
    ("E06", "Why do ants visit different kinds of food?"),
    ("E07", "What is the exact name of every bird I can hear outside my house right now?"),
    ("E08", "Will it rain at my house tomorrow?"),
    ("E09", "Why do houses in the mountains sometimes have sloping roofs?"),
]

def safe_value(obj, *keys, default=""):
    for key in keys:
        if isinstance(obj, dict) and obj.get(key) is not None:
            return obj[key]
    return default

rows = []
for qid, question in TESTS:
    start = time.time()
    hits = retrieve_passages(question, top_k=5, candidate_k=50)
    elapsed = round(time.time() - start, 3)

    print(f"\n{'='*90}\n{qid} | {question}\nRetrieval time: {elapsed}s")

    for i, hit in enumerate(hits, 1):
        text = str(safe_value(hit, "text", default=""))
        book = safe_value(hit, "book", "source_file", default="Unknown")
        cls = safe_value(hit, "class", "class_name", default="Unknown")
        page = safe_value(hit, "page", "pages", default="Unknown")
        chunk_id = safe_value(hit, "chunk_id", default="")
        score = safe_value(hit, "reranker_score", default=None)

        print(f"\n[{i}] {book} | {cls} | page {page} | score={score}")
        print(text[:1000].replace("\n", " "))

        rows.append({
            "question_id": qid,
            "question": question,
            "rank": i,
            "book": str(book),
            "class": str(cls),
            "page": str(page),
            "chunk_id": str(chunk_id),
            "reranker_score": score,
            "rrf_score": safe_value(hit, "rrf_score", default=None),
            "chunk_index": safe_value(hit, "chunk_index", default=None),
            "text": text,
            "retrieval_seconds": elapsed,
        })

df = pd.DataFrame(rows)
stamp = time.strftime("%Y%m%d_%H%M%S")

csv_path = OUT / f"evidence_pack_{stamp}.csv"
xlsx_path = OUT / f"evidence_pack_{stamp}.xlsx"
json_path = OUT / f"evidence_pack_{stamp}.json"

df.to_csv(csv_path, index=False)
df.to_excel(xlsx_path, index=False)
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(rows, f, ensure_ascii=False, indent=2, default=str)

print("\n\nSAVED EVIDENCE PACK")
print(csv_path)
print(xlsx_path)
print(json_path)
print("\nRows:", len(df), "| Questions:", df["question_id"].nunique())
print("No Qwen inference was run.")
```
